> **Next cell.** Sets the figure style (IEEE journal: serif type, 7.16-in width, boxed axes, no text below 9 pt, colour-vision-safe palette) and three helpers: `card()` for results, `table()` for tables and `recorded()`, which loads the data of a recorded run from the public figures repository.
````python
#@title Figure and output style (formatting helpers; code hidden in Colab) { display-mode: "form" }
# ---------- figure and output style shared by every cell
# Figures follow the ISCAS 2027 paper (paper/iscas2027/scripts/fig_*.py): serif type (STIXGeneral, the Times-compatible
# face shipped with matplotlib, so Colab draws the same glyphs), boxed axes with inward ticks at the bottom and left, no
# grid, legends without a border, centred panel captions (a), (b), ... and the paper's palette; 7.16-in two-column or
# 3.5-in one-column widths and no text below 9 pt, so every label stays legible on screen. The two data colours are a colour-vision-safe pair; every figure
# also reads in greyscale (line styles, markers, hatching).
import matplotlib as mpl, matplotlib.pyplot as plt
from matplotlib.patches import Patch
from IPython.display import HTML, display
try:
    get_ipython().run_line_magic("config", "InlineBackend.figure_format = 'retina'")   # sharp figures in Colab
except Exception:
    pass
# the palette of the ISCAS 2027 paper's figures (paper/iscas2027/scripts/fig_*.py): ink, blue (the proposal, nominal),
# orange-red (the original S-DSM, a failure), dark red (limits and targets), green and gold (further series), grey bands
COL = dict(sig="#2a6fdb", ink="#1b1f24", key="#b03a2e", ok="#2a6fdb", bad="#e4572e", faint="#7c8594",
           aqua="#17a589", yellow="#c9a227", grid="#e3e7ec", band="#d5d8dc")
W2, W1 = 7.16, 3.5                       # IEEE two-column and one-column figure widths (inches)
plt.rcParams.update({
    "figure.dpi": 100, "savefig.dpi": 300, "figure.facecolor": "white", "axes.facecolor": "white",
    "font.family": "serif", "font.serif": ["STIXGeneral", "STIX Two Text", "Times New Roman", "DejaVu Serif"],
    "mathtext.fontset": "stix", "font.size": 10.5, "axes.labelsize": 10.5, "axes.titlesize": 10.5,
    "axes.titleweight": "normal", "axes.titlelocation": "center", "xtick.labelsize": 9.5, "ytick.labelsize": 9.5,
    "legend.fontsize": 9, "legend.frameon": True, "legend.framealpha": 0.92, "legend.edgecolor": "none",
    "legend.fancybox": False, "legend.borderpad": 0.4, "legend.handlelength": 1.8, "patch.linewidth": 0.6,
    "axes.linewidth": 0.8, "axes.edgecolor": "#1b1f24", "axes.labelcolor": "#1b1f24", "text.color": "#1b1f24",
    "axes.spines.top": True, "axes.spines.right": True, "axes.grid": False, "grid.color": "#e3e7ec",
    "grid.linewidth": 0.5, "grid.linestyle": ":", "axes.axisbelow": True,
    "xtick.direction": "in", "ytick.direction": "in", "xtick.top": False, "ytick.right": False,
    "xtick.major.size": 3.5, "ytick.major.size": 3.5, "xtick.minor.size": 2, "ytick.minor.size": 2,
    "xtick.major.width": 0.7, "ytick.major.width": 0.7, "xtick.color": "#1b1f24", "ytick.color": "#1b1f24",
    "lines.linewidth": 1.5, "lines.markersize": 5, "figure.constrained_layout.use": False,
})

_CENTER = ("<style>.output_png, .output_image, .jp-RenderedImage {text-align:center !important} "
           ".output_png img, .output_image img, .jp-RenderedImage img, img[src^='data:image/png']:not(.widget-image), "
           "img[src^='blob:']:not(.widget-image) {display:block !important; margin-left:auto !important; "
           "margin-right:auto !important; max-width:100% !important; height:auto !important}</style>")
DISPLAY_PX = 1000      # display width of every plot: at least GitHub's notebook column, so there (no styles) each viewer
                       # scales the centred canvas to the column; wider Jupyter and Colab columns keep this size and the
                       # style above centres it (a full-width 2000 px made every plot 1.6x larger on wide screens)
def _jcap(s):
    """Journal case of one label: an English first word is capitalised ('noise' -> 'Noise'); a Greek letter (σ, θ), a
    one-letter variable (q, k), math ($...$), a function (log, ln, exp), a net name or a process corner stays as
    written."""
    import re
    nets = {"dp", "dn", "outp", "outn", "qb", "clk", "tail", "vdd", "vss", "tt", "ss", "ff", "sf", "fs",
            "log", "ln", "exp"}                                        # math functions stay lower case
    m = re.match(r"([a-z][a-z]+)(?![_$^{a-zA-Z0-9])", s) if s else None
    return s[0].upper() + s[1:] if m and m.group(1) not in nets else s

def _journal_case(fig):
    """Journal style: axis labels and panel titles start with a capital letter ('(a) noise' -> '(a) Noise'), unless they
    start with a symbol, a number or math. Applied to every figure in show(), so each cell need not do it."""
    import re
    cap = _jcap
    # only the text is changed (Text.set_text): set_ylabel/set_title would also reset the alignment (a row label drawn with
    # rotation=0, ha="right" would be centred on the axis and overlap the plot) and the title's font size and weight
    for a in fig.axes:
        for lab in (a.xaxis.label, a.yaxis.label):
            lab.set_text(cap(lab.get_text()))
        for t in (a.title, a._left_title, a._right_title):
            s = t.get_text()
            m = re.match(r"^(\([a-z]\)\s+)(.*)$", s, re.S)
            new = (m.group(1) + cap(m.group(2))) if m else cap(s)
            if new != s:
                t.set_text(new)

def show(*args, **kwargs):
    """Replaces plt.show(): every open figure is displayed as one PNG, then closed. The figure is drawn in the middle of a
    white canvas of the two-column width (wider figures keep their width), so that it is centred in every viewer, and the
    PNG carries a display width (DISPLAY_PX) of about GitHub's notebook column: GitHub (which strips styles) scales it to
    the column, so the figure is centred there without a style, and in wider Jupyter and Colab columns the style printed
    before it centres it at this size; a plain PNG output is rendered by every notebook viewer (GitHub strips
    images embedded in HTML outputs). With show.capture set to a list, the PNGs are collected instead of displayed (static copies of widgets)."""
    import io
    from IPython.display import Image
    from PIL import Image as PILImage
    for num in plt.get_fignums():
        fig = plt.figure(num)
        _journal_case(fig)
        dpi = 110 if show.plain else 170
        buf = io.BytesIO(); fig.savefig(buf, format="png", dpi=dpi, facecolor="white")
        # every figure on one canvas of the two-column width, centred on it: the figure is centred in every viewer, also
        # where the centring style below is stripped (GitHub), and all figures share one display scale
        canvas_in = max(W2, fig.get_figwidth())
        im = PILImage.open(io.BytesIO(buf.getvalue()))
        wpx = int(round(canvas_in * dpi))
        if im.width < wpx:
            page = PILImage.new("RGB", (wpx, im.height), "white")
            page.paste(im.convert("RGB"), ((wpx - im.width) // 2, 0))
            buf = io.BytesIO(); page.save(buf, format="png", optimize=True)
        show.last = (max(wpx, im.width), im.height, im.width)   # canvas and drawn size: the interactive version takes them
        if show.capture is not None:
            show.capture.append(buf.getvalue())
        else:
            display(HTML(_CENTER))                       # centres the canvas in Jupyter and Colab
            show.handle = (display(Image(data=buf.getvalue(), width=DISPLAY_PX), display_id=True), buf.getvalue())
        plt.close(fig)
show.plain, show.capture, show.last, show.handle = False, None, None, None
plt.show = show

def display_widget(w, png=None):
    """Displays an ipywidget together with a static PNG of its initial view (png: bytes): a live notebook shows the widget,
    a viewer without a kernel (GitHub, a saved copy) shows the PNG instead of an empty output."""
    import base64
    bundle = w._repr_mimebundle_() if hasattr(w, "_repr_mimebundle_") else None
    if bundle is None or png is None:
        display(w); return
    import io
    from PIL import Image as PILImage
    im = PILImage.open(io.BytesIO(png))                 # centred on the canvas of show(), like every plot
    wpx = int(round(W2 * 170))
    if im.width < wpx:
        page = PILImage.new("RGB", (wpx, im.height), "white")
        page.paste(im.convert("RGB"), ((wpx - im.width) // 2, 0))
        buf = io.BytesIO(); page.save(buf, format="png", optimize=True); png = buf.getvalue()
    data = dict(bundle[0] if isinstance(bundle, tuple) else bundle)
    data["image/png"] = base64.b64encode(png).decode()
    display(HTML(_CENTER))
    display(data, raw=True, metadata={"image/png": {"width": DISPLAY_PX}})   # spans the column, like the plots of show()

# ---------- interactive companions of the key plots (Plotly): the static PNG above stays the figure of record (GitHub
# and the web copy show it, since they strip scripts); in Colab, Jupyter and nbviewer the interactive version below it adds
# hover values, zoom, pan and legend toggles. It keeps the static figure's journal style at the same display scale: the
# same face (STIX, loaded as a web font, else Times), the same sizes in points (9.5 pt ticks, 10.5 pt labels and panel
# captions, 8 pt legend), boxed axes with inward ticks, the palette, line widths and markers, and the static figure's
# aspect. plotly.js is loaded from its CDN, so each plot carries only its data.
_PT = DISPLAY_PX / (W2 * 72)                 # display pixels per point of the static figures (1000 px for 7.16 in: 1.94)
_PFONT = "'STIX Two Text', STIXGeneral, 'Times New Roman', Times, serif"
PLOTLY_LAYOUT = dict(
    template="simple_white", paper_bgcolor="white", plot_bgcolor="white", hovermode="closest",
    font=dict(family=_PFONT, size=round(9.5 * _PT), color=COL["ink"]),
    margin=dict(l=12, r=12, t=round(17 * _PT), b=12, autoexpand=True),
    legend=dict(bgcolor="rgba(255,255,255,0.92)", borderwidth=0, font=dict(size=round(8 * _PT)), tracegroupgap=0),
    hoverlabel=dict(font=dict(family=_PFONT, size=round(8.5 * _PT)), bgcolor="white", bordercolor=COL["faint"]),
    colorway=[COL["sig"], COL["bad"], COL["aqua"], COL["yellow"], COL["ink"], COL["faint"], COL["key"]])
_PAXIS = dict(showline=True, mirror=True, linecolor=COL["ink"], linewidth=0.8 * _PT, ticks="inside", ticklen=3.5 * _PT,
              tickwidth=0.7 * _PT, tickcolor=COL["ink"], showgrid=False, zeroline=False, automargin=True,
              title_font=dict(size=round(10.5 * _PT)), title_standoff=round(4 * _PT))
# plotly.js is loaded once per page and each plot is drawn when it is there: through require.js where the page has it
# (classic Notebook, nbviewer), otherwise through a script element (Colab, JupyterLab), whatever order the viewer runs an
# output's scripts in (a plain <script src> before Plotly.newPlot draws nothing in JupyterLab or under require.js). The
# web font is awaited for at most 1.5 s, so that the labels are measured in the face they are drawn in.
_PLOTLY_JS = ("(function(){var d=document.getElementById('%s');if(!d)return;var f=%s,c=%s;"
              "function hint(){var h=document.createElement('div');h.style.cssText='font:12px/1.4 sans-serif;color:#7c8594;"
              "text-align:center';h.textContent='Interactive figure: hover for values, drag to zoom, double-click to reset, "
              "click a legend entry to hide it';d.parentNode.insertBefore(h,d);}"
              "function draw(P){Promise.resolve(P.newPlot(d,f.data,f.layout,c)).then(function(){hint();});}"
              "function go(){if(window.Plotly&&window.Plotly.newPlot)return draw(window.Plotly);"
              "window.PlotlyConfig=window.PlotlyConfig||{MathJaxConfig:'local'};var u='https://cdn.plot.ly/plotly-%s.min';"
              "if(typeof define==='function'&&define.amd&&typeof require==='function'&&require.config){"
              "require.config({paths:{plotly:u}});require(['plotly'],draw);return;}"
              "var p=window.__nbPlotly||(window.__nbPlotly=new Promise(function(ok,no){var s=document.createElement('script');"
              "s.src=u+'.js';s.charset='utf-8';s.onload=function(){ok(window.Plotly);};s.onerror=no;document.head.appendChild(s);}));"
              "p.then(draw);}"
              "var w=new Promise(function(r){setTimeout(r,1500);});"
              "(document.fonts&&document.fonts.load?Promise.race([document.fonts.load('19px \"STIX Two Text\"'),w]):Promise.resolve())"
              ".then(go,go);})();")
_PFONT_CSS = ('<link rel="stylesheet" href="https://fonts.googleapis.com/css2?family=STIX+Two+Text:ital,wght@0,400;0,600;1,400'
              '&display=swap">')
def _ilog_lims(fig, axname):
    """(lowest, highest) decade (log10) of the positive data of the traces on one axis ('xaxis2' -> the traces on x2)."""
    letter, ref = axname[0], axname[0] + axname[5:]
    v = []
    for tr in fig.data:
        if (getattr(tr, letter + "axis", None) or letter) == ref and getattr(tr, letter, None) is not None:
            a = np.asarray([x for x in getattr(tr, letter) if x is not None], float)
            v.append(a[np.isfinite(a) & (a > 0)])
    v = np.concatenate(v) if v else np.array([])
    return (float(np.log10(v.min())), float(np.log10(v.max()))) if len(v) > 1 else (0.0, 1.0)

def _isig(v, n=6):
    """Float data rounded to n significant digits for the page (a plot's JSON carries 17 digits otherwise); other data as is."""
    a = np.asarray(v) if isinstance(v, (np.ndarray, list, tuple)) else None
    if a is None or a.dtype.kind != "f" or a.size == 0:
        return v
    with np.errstate(divide="ignore", invalid="ignore", over="ignore"):
        m = 10.0 ** (n - 1 - np.floor(np.log10(np.abs(a))))
        return np.where(np.isfinite(a) & (a != 0) & np.isfinite(m), np.round(a * m) / m, a)

def _ithin(x, y, tol):
    """Indices of the points of a waveform that keep it within tol of the full curve at every sample (Ramer-Douglas-
    Peucker on the vertical distance): the flat stretches of a transient lose their points, the edges keep theirs."""
    x, y = np.asarray(x, float), np.asarray(y, float); n = len(x)
    keep = np.zeros(n, bool); keep[[0, -1]] = True; todo = [(0, n - 1)] if n > 2 else []
    while todo:
        i, j = todo.pop()
        if j <= i + 1:
            continue
        t = (x[i + 1:j] - x[i]) / (x[j] - x[i]) if x[j] != x[i] else np.zeros(j - i - 1)
        d = np.abs(y[i + 1:j] - (y[i] + t * (y[j] - y[i]))); k = int(np.argmax(d))
        if d[k] > tol:
            keep[i + 1 + k] = True; todo += [(i, i + 1 + k), (i + 1 + k, j)]
    return np.flatnonzero(keep)

def _ijournal(fig):
    """The static figures' journal style on a plotly figure: axes, panel captions and in-plot labels in points at the
    display scale; trace line widths and marker sizes, which the builders give in points (lw, ms of the static figure)."""
    import plotly.graph_objects as go
    own = fig.layout.to_plotly_json(); own.pop("template", None)   # the builder's own settings win over the defaults
    fig.layout = go.Layout(**PLOTLY_LAYOUT).update(own)
    for k in [k for k in fig.layout if k.startswith(("xaxis", "yaxis"))]:   # the journal axes, under the builder's own
        a = fig.layout[k]; own_ax = a.to_plotly_json()                # settings (ticks off, a grid, smaller type)
        a.update(**_PAXIS); a.update(own_ax)
    for sh in fig.layout.shapes:                              # shapes opaque unless the builder says otherwise (the
        if sh.opacity is None:                                # template draws them at 0.3)
            sh.opacity = 1
    import re
    for a in fig.layout.annotations:                          # make_subplots' panel captions '(a) ...' at 10.5 pt; other
        if (a.text or "").startswith("(") and a.yref == "paper" and a.showarrow is False and a.font.size in (None, 16):
            a.font.size = round(10.5 * _PT)                   # (16: make_subplots' own size); labels without a size at 8 pt
            m = re.match(r"^(\([a-z]\)\s+)(.*)$", a.text, re.S)
            if m:
                a.text = m.group(1) + _jcap(m.group(2))
        elif a.font.size is None:
            a.font.size = round(8 * _PT)
        elif a.font.size < round(7.5 * _PT):
            a.font.size = round(7.5 * _PT)
    for ax in [k for k in fig.layout if k.startswith(("xaxis", "yaxis"))]:   # axis titles in journal case, as in show()
        a = fig.layout[ax]
        if a.title.text:
            a.title.text = _jcap(a.title.text)
        if a.type == "log":                                   # log axes as matplotlib draws them: labels 10^n at every decade up to
            if a.tickvals is None:                            # 7 decades, else every ceil(span/6) decades; unlabelled minor ticks
                lo, hi = a.range if a.range else _ilog_lims(fig, ax)
                d = a.dtick if isinstance(a.dtick, (int, float)) and a.dtick >= 1 else (1 if hi - lo <= 7 else int(np.ceil((hi - lo) / 6)))
                ks = [k for k in range(int(np.floor(lo)) - 2, int(np.ceil(hi)) + 3) if k % int(d) == 0]
                a.tickvals = [10.0 ** k for k in ks]
                a.ticktext = [f"10<sup>{k}</sup>".replace("-", "−") for k in ks]
            lo, hi = a.range if a.range else _ilog_lims(fig, ax)
            a.minor = (dict(ticks="inside", dtick="D1", ticklen=2 * _PT, tickwidth=0.6 * _PT, tickcolor=COL["ink"], showgrid=False)
                       if hi - lo <= 7 else dict(ticks=""))
    for tr in fig.data:                                       # the data at 6 significant digits (hover shows at most 4)
        for p_ in ("x", "y", "z", "customdata", "base"):
            if p_ in tr and tr[p_] is not None:
                tr[p_] = _isig(tr[p_])
    for tr in fig.data:                                       # line widths and marker sizes are given in points, as lw and
        line = getattr(tr, "line", None)                      # ms of the static figure: converted at the display scale
        if line is not None and "width" in line and line.width is not None and tr.type not in ("bar", "pie", "heatmap"):
            line.width = round(min(5.0, _PT * line.width), 2)
        mk = getattr(tr, "marker", None)
        if mk is not None and tr.type == "scatter" and isinstance(mk.size, (int, float)):
            mk.size = round(_PT * mk.size, 1)
    return fig

def _ilegend_rows(fig, width=DISPLAY_PX):
    """Rows of a horizontal legend across the plot width: entry widths from the names (8 pt type, symbol and gap)."""
    import math, re
    seen_g, total = set(), 0.0
    for t in fig.data:
        if t.showlegend is False or not t.name or (t.legendgroup and t.legendgroup in seen_g and t.showlegend is not True):
            continue
        seen_g.add(t.legendgroup)
        total += 0.46 * round(8 * _PT) * len(re.sub(r"<[^>]+>", "", t.name)) + 52
    return max(1, math.ceil(total / (width - 60))) if total else 0

def iplot(build, title=None, height=None):
    """Shows the figure above as an interactive figure: build() returns a plotly Figure, drawn from the same data and in
    its journal style (_ijournal). The height defaults to the static figure's aspect at the display width. A legend that the
    builder places below the panels becomes one compact row (or as many as its entries need) under the figure, which adds
    its own height; a legend placed inside the panels stays where the builder puts it. title is only kept as the plot's
    accessible name. The interactive figure replaces the static PNG of the same cell in one output that carries both
    (text/html and image/png), so that every viewer shows the figure once (owner, 2026-10-30): Colab and Jupyter draw the
    interactive version, and the PNG stays in the output as the plot of record. The plot's box has the figure's height in
    pixels (owner, R68): with a box of height 100 % of a parent that has none, Plotly's responsive resize, on a menu
    change or on the first draw in a live Colab run, read a height of zero and collapsed the plot into one line. A failure
    prints one line and leaves the run (and the static figure) intact."""
    try:
        import json, uuid
        from plotly.offline import get_plotlyjs_version
        fig = _ijournal(build())
        cw, ch, dw = show.last if show.last else (1, 0.44, 1)
        if height is None:
            height = round(DISPLAY_PX * ch / cw)
        width = round(DISPLAY_PX * min(1.0, (dw + 40) / cw))   # a figure narrower than the two-column width keeps its width
        lg = fig.layout.legend
        if fig.layout.showlegend is not False and (lg.y is None or lg.y < 0):
            rows = _ilegend_rows(fig, width)
            if rows:
                extra = 8 + rows * round(8 * _PT * 1.5)
                fig.update_layout(legend=dict(orientation="h", x=0.5, xanchor="center", yref="container", y=0, yanchor="bottom",
                                              entrywidthmode="pixels", entrywidth=0),
                                  margin=dict(b=fig.layout.margin.b + extra))
                height += extra
        fig.update_layout(height=height, title=None)
        div = "nbp-" + uuid.uuid4().hex[:12]
        js = _PLOTLY_JS % (div, fig.to_json().replace("</", "<\\/"),
                           json.dumps({"displaylogo": False, "responsive": True, "displayModeBar": True}),
                           get_plotlyjs_version())
        name = (title or "interactive figure").replace('"', "'")
        html = (_PFONT_CSS + f'<div class="nb-plotly" style="max-width:{width}px;margin:4px auto 6px">'
                f'<div id="{div}" role="img" aria-label="{name}" style="height:{height}px;width:100%"><p style="font:12px/1.4 '
                f'sans-serif;color:#7c8594;text-align:center">Interactive figure: shown in Colab or Jupyter; the static '
                f'figure is in the README.</p></div><script type="text/javascript">{js}</script></div>')
        if show.handle is not None:                          # one output: the interactive figure in place of the PNG
            import base64
            h, png = show.handle; show.handle = None
            h.update({"text/html": html, "image/png": base64.b64encode(png).decode()}, raw=True,
                     metadata={"image/png": {"width": DISPLAY_PX}})
        else:
            display(HTML(html))
    except Exception as e:                                   # e.g. plotly missing: the static figure above still holds
        print(f"(interactive version not shown: {type(e).__name__}: {e})")

def panel(ax, label, text=""):
    """Panel caption, '(a) text', centred above the axes, as in the ISCAS paper's figures."""
    ax.set_title(f"({label}) {text}".strip(), loc="center", fontsize=10.5, pad=5)

# ---------- result cards: the cells report their results as small tables instead of plain prints
_CARD = ("font-family:'STIX Two Text','Times New Roman',Times,serif;font-size:15px;line-height:1.45;"
         "border:1px solid rgba(128,128,128,.45);border-left:4px solid {c};padding:8px 14px;margin:8px 0;max-width:860px")
def card_html(title, rows=(), note=None, tone="ok"):
    """The HTML of a result card (see card). The values a cell is run for are written in <b>...</b>: they are drawn in the
    card's colour on a light tint, so that the reader sees at once what the cell produced (owner, 2026-10-30)."""
    c = {"ok": COL["ok"], "bad": COL["bad"], "info": COL["faint"]}[tone]
    hi = COL["key"] if tone == "info" else c
    mark = (f'<b style="color:{hi};background:{hi}1f;padding:0 3px;border-radius:3px">')
    body = "".join(f'<tr><td style="padding:2px 18px 2px 0;opacity:.8;vertical-align:top">{k}</td>'
                   f'<td style="padding:2px 0;font-variant-numeric:tabular-nums">{str(v).replace("<b>", mark)}</td></tr>'
                   for k, v in rows)
    return (f'<div style="{_CARD.format(c=c)}"><div style="font-weight:bold;margin-bottom:3px">{title}</div>'
            + (f'<table style="border-collapse:collapse;font-size:15px">{body}</table>' if rows else "")
            + (f'<div style="margin-top:4px;opacity:.85">{note}</div>' if note else "") + "</div>")

def card(title, rows=(), note=None, tone="ok"):
    """A compact result panel: a title line, then label | value rows, then an optional note.
    tone: 'ok' (blue: a check that passed / the proposal), 'bad' (orange: a weakness), 'info' (grey). The value that the
    cell is run for is written <b>...</b> and highlighted."""
    display(HTML(card_html(title, rows, note, tone)))

# ---------- precomputed explorers (owner, 2026-10-30): every view of an explorer is computed beforehand with the same
# model (dev/explore_precompute.py) and stored in the figures repository at a fixed commit, one image and one result card
# per setting; the menus switch between them in the browser, so they respond at once in Colab, Jupyter and nbviewer
# without running anything. The default view is also embedded as the plot of record (image/png). A new view replaces the
# old one only once its image has loaded (owner, R68): an image element with a new, not yet loaded source has no height,
# and Colab's output frame shrank to it.
EXPLORE = "https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/141d43fc1554cb80542d85ef4e42d058df6f4239/explore"
def explore(name, title):
    """Shows the precomputed explorer explore/<name>/ (index.json: controls and, per setting, the image and the card)."""
    import base64, json, uuid, html as _h, urllib.request
    get = lambda u: urllib.request.urlopen(u, timeout=60).read()
    base = f"{EXPLORE}/{name}/"
    idx = json.loads(get(base + "index.json").decode("utf-8"))
    keyed = [c for c in idx["controls"] if not c.get("zoom")]    # a zoom control scales the image in the browser
    key0 = "|".join(c["default"] for c in keyed)
    st0 = idx["states"][key0]
    eid = "nbx-" + uuid.uuid4().hex[:10]
    cards = list(dict.fromkeys(v["card"] for v in idx["states"].values()))    # many settings share one card: send each once
    states = {k: [v["img"], cards.index(v["card"])] for k, v in idx["states"].items()}
    sel = "".join(
        f'<label style="margin:0 14px 6px 0;display:inline-block">{_h.escape(c["label"])} '
        f'<select style="font:inherit;padding:1px 4px"{" data-zoom=1" if c.get("zoom") else ""}>' +
        "".join(f'<option value="{_h.escape(v)}"{" selected" if v == c["default"] else ""}>{_h.escape(l)}</option>'
                for l, v in c["options"]) + "</select></label>" for c in idx["controls"])
    js = ("(function(){var r=document.getElementById('@ID@'),S=@S@,C=@C@,b='@B@';if(!r)return;var s=r.querySelectorAll('select'),"
          "i=r.querySelector('img'),c=r.querySelector('.nbx-card');function u(){var k=[];s.forEach(function(e){"
          "if(e.dataset.zoom){i.style.width=(100*e.value)+'%';}else{k.push(e.value);}});var p=i.parentNode;"
          "setTimeout(function(){p.scrollLeft=(p.scrollWidth-p.clientWidth)/2;p.scrollTop=(p.scrollHeight-p.clientHeight)/2;},0);"
          "var v=S[k.join('|')];if(!v){c.innerHTML='<i>no view for this combination</i>';return;}var n=new Image();n.onload=function(){i.src=n.src;c.innerHTML=C[v[1]];};n.src=b+v[0];}"
          "s.forEach(function(e){e.addEventListener('change',u);});})();").replace("@ID@", eid).replace("@B@", base).replace(
          "@S@", json.dumps(states).replace("</", "<\\/")).replace("@C@", json.dumps(cards).replace("</", "<\\/"))
    page = (f'<div id="{eid}" style="font-family:\'STIX Two Text\',\'Times New Roman\',Times,serif;font-size:15px;'
            f'max-width:{DISPLAY_PX}px;margin:6px auto">'
            f'<div style="font:12px/1.4 sans-serif;color:{COL["faint"]};margin-bottom:4px">{_h.escape(title)}: '
            f'{_h.escape(idx.get("note", "choose any setting; the figure and the card change at once (precomputed with the model of this notebook, no run needed)."))}</div>'
            f'<div>{sel}</div><div style="overflow:auto;max-height:{idx.get("max_height", 2000)}px">'
            f'<img src="{base}{st0["img"]}" style="width:100%;max-width:none;display:block" alt="{_h.escape(title)}"></div>'
            f'<div class="nbx-card">{st0["card"]}</div><script type="text/javascript">{js}</script></div>')
    display({"text/html": page, "image/png": base64.b64encode(get(base + st0["img"])).decode()}, raw=True,
            metadata={"image/png": {"width": DISPLAY_PX}})

def table(header, rows, caption=None, align=None):
    """A booktabs-style table (top and bottom rules, no vertical lines), IEEE caption above it."""
    al = align or ["left"] + ["right"] * (len(header) - 1)
    th = "".join(f'<th style="text-align:{a};padding:4px 12px;border-bottom:1px solid currentColor;font-weight:bold">{h}</th>'
                 for h, a in zip(header, al))
    tr = "".join("<tr>" + "".join(f'<td style="text-align:{a};padding:3px 12px">{v}</td>' for v, a in zip(r, al)) + "</tr>" for r in rows)
    cap = (f'<div style="text-align:center;font-variant:small-caps;letter-spacing:.04em;margin-bottom:4px">{caption}</div>'
           if caption else "")
    display(HTML(f'<div style="font-family:\'STIX Two Text\',\'Times New Roman\',Times,serif;font-size:15px;margin:10px auto;max-width:900px;width:fit-content">{cap}'
                 f'<table style="border-collapse:collapse;margin:0 auto;border-top:1.6px solid currentColor;border-bottom:1.6px solid currentColor">'
                 f'<thead><tr>{th}</tr></thead><tbody>{tr}</tbody></table></div>'))

# icons, animations and the explorer page live in a small public companion repository (nothing secret)
ASSETS = "https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/main"
LAB_N = 30000                            # modulator clocks used by the attack experiments and sweeps

# the data of the recorded runs (Appendix B) lives in the same repository, one JSON file per record in data/, at a fixed
# commit, so that no code cell carries a record inline; recorded(<name>) loads one (once per session, a fresh copy per call)
RECORDS = "https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/a8da3912e14c70b6443e319681bc05b5489bc65e/data"
_RECORDED = {}
def recorded(name):
    """The record of a recorded run: data/<name>.json of the figures repository at the commit above (needs the internet)."""
    import json, time, urllib.request
    if name not in _RECORDED:
        url, err = f"{RECORDS}/{name}.json", None
        for attempt in range(3):
            try:
                with urllib.request.urlopen(url, timeout=60) as r:
                    _RECORDED[name] = r.read().decode("utf-8")
                break
            except Exception as e:                       # a network hiccup: two more tries
                err = e
                time.sleep(2 * (attempt + 1))
        else:
            raise RuntimeError(f"the recorded data {url} could not be loaded ({err}): the recorded runs are read from the "
                               "public figures repository, so this cell needs an internet connection")
    return json.loads(_RECORDED[name])

# cell timer: the start and end of every code cell run from here on, summed by section at the end of the notebook
import time as _time
CELL_TIMES = {}                          # first line of the cell -> [start, end] of its latest run
def _cell_start(info):
    CELL_TIMES[info.raw_cell.split("\n", 1)[0]] = [_time.time(), None]
    show.handle = None                   # an interactive figure replaces only a PNG of its own cell
def _cell_end(result):
    t = CELL_TIMES.get(result.info.raw_cell.split("\n", 1)[0])
    if t:
        t[1] = _time.time()
try:
    _ip = get_ipython()
    for _ev, _f in (("pre_run_cell", _cell_start), ("post_run_cell", _cell_end)):
        for _old in [g for g in _ip.events.callbacks[_ev] if getattr(g, "__name__", "") == _f.__name__]:
            _ip.events.unregister(_ev, _old)  # re-running this cell replaces the timer instead of adding a second one
        _ip.events.register(_ev, _f)
except NameError:                        # not in IPython
    pass
````

# **Abstract**
---
**In-modulator encryption XORs each output bit of a delta-sigma modulator with a keystream, so that a sensor encrypts its readings without a decimator. The secure delta-sigma modulator (S-DSM) of Yasar *et al.* [1] does so with a 16-bit key held on the chip, a linear keystream without a nonce and unauthenticated frames. This work keeps its front end and its bit-level XOR and adds four ideas new to in-modulator encryption: a 128-bit key that is never stored, regenerated at every power-up from static random-access memory (SRAM) cells used as a physically unclonable function (PUF), through a fuzzy extractor sized from the cell's simulated error rates and secret budget; the keystream and a 32-bit message authentication code of every frame from one Grain-128a core; a chip nonce, hashed from 64 power-up reads of the same cells, in every initialisation vector; and a one-time enrolment at −40 and 85 °C. Both designs are evaluated with open-source tools only. A bit-level model of the link reproduces the two-channel measurement of [1] (48.2 % and 50.0 % of bits differ, against 49.8 % and 50.05 % measured) and takes its circuit parameters from post-layout SKY130 simulations of the modulator's StrongARM comparator and of a six-transistor SRAM cell at 45 process, voltage and temperature (PVT) conditions. Three findings are new. First, analog differences between chips cannot replace a key: although the comparator holds the loop at 60.3–60.4 dB signal-to-noise-and-distortion ratio at every condition, a replica attack breaks all 360 simulated chip pairs, and comparator noise stops the attack only at 197× its largest simulated value. Second, key reliability and nonce freshness conflict: the cell noise that the fuzzy extractor corrects is too small to make one read a fresh nonce, so the nonce is hashed from 64 reads. Third, at the fast-nMOS corners some SRAM cells reverse their preferred state with temperature (27.8 % bit errors at 85 °C); only enrolment at both temperatures keeps the key failure rate below 10⁻⁶ at every corner, while retaining 284–352 secret bits. With an ageing law scaled from published NBTI measurements (the PDK has none), an array kept powered with its power-up values loses that target within 26 days, whereas writing back the inverse of each read makes the key more reliable with age, at the cost of the randomness the nonce draws from the same cells. In this model the S-DSM fails 11 of 15 attacks (three under stated assumptions); the proposal stops 14 at the working setting of the system simulations and all 15, 10 resisted and 5 detected, with the two-temperature enrolment. Power analysis lies outside the threat model, and a leakage test detects key-dependent leakage in both designs as built: in [1] at every clock of every frame, in the proposal only while the key is loaded and mixed with the initialisation vector at the start of a frame. In one SKY130 flow the proposal's keystream block takes 9,869 gate equivalents (GE) and 64.3 µW at 10 MHz, 3.6× the S-DSM, and has the best (lowest) figure of merit, energy per bit per security bit: 2.21, against 5.09 for Trivium and 14.1 for the S-DSM. At an oversampling ratio of 50, the proposed sensor sends 1.44–1.60× fewer bits per reading than the decimate-then-Trivium design of [1]. A key-generator compiler turns the reliability and secrecy specification into a 19,128 GE decoder, bit-identical to the model in 72 of 72 RTL reconstructions. A mixed-level simulation of the complete sensor regenerates the key from transistor-level cells bit for bit, and 3 of 2,000 transistor-level comparator decisions differ from the model, consistent with the comparator noise (2.4 expected). Every result is computed live in one Colab session, except a few long simulations, which are shown from recorded runs that a switch or a script reproduces.**

# **I. Introduction**
---
Wireless sensors increasingly feed control loops and therapies. Sensor capsules that float in a bioreactor report the glucose concentration and the cell density of a culture that produces pharmaceuticals or food ingredients, and a closed control loop acts on their readings [1]; a glucose sensor worn by a patient informs the next insulin dose. These readings are confidential, since they reveal a manufacturer's process or a person's health, and a forged, altered or replayed reading makes the loop act on false data, at the cost of a batch or of a patient's safety. **Every reading must therefore be encrypted and authenticated, at a power that the sensor can afford.**

The power available for this is small. The front end of the bioreactor sensor of [1], including its modulator, dissipates about 1.35 mW; an on-chip Trivium stream cipher [5] would add 220 µW, and a conventional design must in addition decimate the modulator output into words before encrypting them [1]. Yasar *et al.* [1] **avoid both costs by encrypting the bits of the delta-sigma modulator itself, for 24 µW**. This work examines whether such in-modulator encryption protects the readings against a realistic attacker, and we propose a design that does so without giving up its advantages. Subsection A describes the application, the threats on its radio link, the power budget that rules out a conventional cipher, the requirements that follow and the design we propose; Subsection B, the four research questions this notebook answers about the design of [1] and about the proposed alternative; Subsection C, the proposed design: the gaps that prior work leaves and what this work adds for each; and Subsection D, the method and the organisation of the notebook.

### A. Application and motivation: secure sensing for biomanufacturing

Biomanufacturing uses living cells in large bioreactors to produce active pharmaceutical ingredients, food ingredients and bio-based materials; the culture conditions determine the accuracy, efficiency and yield of the process [1]. Today the culture is monitored either by drawing samples for offline analysis or by large fixed probes; both risk contaminating the culture, and a fixed probe observes only one location. Yasar *et al.* [1] instead place a **multimodal electrochemical sensor in a capsule that floats inside the bioreactor**. The capsule reports its readings wirelessly, so that a closed control loop can observe the entire vessel in real time. The chip (65 nm CMOS, 1.2 V) is a potentiostat front end with three measurement modes:

* cyclic voltammetry (CV), to find the redox potential of a reaction;
* amperometry, to measure the glucose concentration (demonstrated from 1.5 to 5.6 mM, coefficient of determination R² = 0.9981), the nutrient the culture consumes;
* electrochemical impedance spectroscopy (EIS, up to 100 kHz), to measure the *E. coli* cell density (4.0 × 10⁷ to 2.4 × 10⁹ cells/mL), the growth of the culture.

A low-noise transimpedance amplifier (TIA) with four gain settings (3.30 kΩ to 3.06 MΩ) **provides 140.55 dB of dynamic range and 9.48 pA rms of input-referred noise over 100 kHz**, and a second-order 1-bit continuous-time delta-sigma modulator clocked at 10 MHz digitises the current.

The same three techniques underlie electrochemical sensing well beyond bioreactors:

* *medical monitoring:* blood-glucose meters and continuous glucose monitors (amperometric enzyme electrodes); lactate, oxygen and pH in blood and sweat; and neurotransmitters such as dopamine, measured by fast-scan cyclic voltammetry;
* *implants and wearables:* impedance monitoring of implanted electrodes and the surrounding tissue, and sweat-sensing patches;
* *diagnostics, food and water:* point-of-care biosensors for bacteria, viruses and proteins, and voltammetric detection of heavy metals and other contaminants;
* *energy and industry:* state-of-health estimation of batteries and fuel cells, and monitoring of corrosion and electroplating baths.

The secure link studied here applies in principle to any such sensor whose current is digitised by a delta-sigma modulator and transmitted wirelessly.

Because the readings drive the control loop, their **confidentiality and integrity** are essential [1]: an eavesdropper learns the state of the process, which the manufacturer treats as confidential, and a man-in-the-middle who alters packets makes the loop act on false data, with consequences from reduced yield to the loss of a batch or, in pharmaceutical production, a health hazard. In medical use the readings are personal health data and can set a therapy. Integrity includes freshness: a replayed frame is an old reading accepted as new, which hides any change in the culture or the patient since it was sent. Finally, a capsule in the attacker's hands should not reveal the key that it shares with the receiver.

Fig. 1 shows where these threats act: the readings cross an open radio link between the capsules and the receiver, on which **the attacker can listen and inject** forged, replayed or altered frames; the attacker can also choose what the sensor measures or obtain a sensor chip.

We propose a design that meets these needs. It keeps the front end of [1] and its encryption of each modulator bit by an exclusive-OR (XOR), with no decimator, and adds **four ideas new to in-modulator encryption**:

1. a 128-bit key that is never stored, regenerated at every power-up from the chip's own static random-access memory (SRAM) cells used as a physically unclonable function (PUF);
2. the keystream and a message authentication code (MAC) of every frame from one Grain-128a core;
3. a chip nonce, hashed from power-up reads of the same cells, in every initialisation vector;
4. a one-time enrolment at −40 and 85 °C.

Both designs are evaluated against the same 15 attacks, from transistor-level SKY130 circuits to the system, and the proposed design's cost is measured, all with open-source tools. Section II-E describes the in-modulator encryption of [1] on which the proposed design builds, and Section IV the proposed design in detail.

### B. Research questions

The evaluation in [1] covers an eavesdropper with a replica chip and two cryptanalyses (a zero input; the same key with the same message), on measured 65 nm silicon without process corners or temperature. **Four features are not tested there**: the 16-bit key is stored on the chip; the LFSR keystream is linear; it has no nonce, so it repeats after every power-up; and the frames are not authenticated, so forged or altered frames pass and, with no counter, replayed frames are accepted. This work addresses four questions, using open-source tools and the SkyWater SKY130 process design kit (PDK):

1. **How secure is the S-DSM against a realistic attacker?** Such an attacker may also search the key space, learn a few keystream bits, obtain a chip or transmit on the link. This needs a bit-level model of the link consistent with the measurements of [1] (Sections V and VI) and every attack run on it (Section IX).
2. **Can the stored key be eliminated and every frame authenticated, without changing the analog front end, and at what cost?** We propose a key regenerated at every power-up from the start-up state of SRAM cells, a physically unclonable function (PUF), corrected by a Bose-Chaudhuri-Hocquenghem (BCH) code and hashed; the Grain-128a stream cipher [6] in place of the LFSR; and Grain-128a's built-in message authentication code (MAC) with a frame counter and two nonces, a session nonce from the receiver and a chip nonce from the chip (Section II-C; design in Section IV, security in Section IX, cost in Section X).
3. **Do circuit non-idealities change these answers?** Comparator noise and offset, and SRAM-cell noise and temperature dependence, all vary with process, supply voltage and temperature (PVT). Do they degrade accuracy, protect the original design, or make the PUF key unreliable? This needs the circuits at block level (Section VII) and their effect at system level (Section VIII).
4. **Can the key generator be derived from a specification, and do the models hold together?** A key-generator compiler turns the reliability specification into RTL verified against the model (Section XI), and one mixed-level simulation of the complete sensor checks the abstractions behind the system-level results (Section XII).

Section XV answers the four questions quantitatively.

### C. Proposal: what prior work leaves open and what this work adds

The S-DSM of [1], and the prior work that Section III reviews, **leave a secure low-power sensor link open on five counts**. This work proposes an answer to each: the first four are ideas new to in-modulator encryption, and the fifth is an evaluation that no prior work of this kind provides.

1. **The key.** *Gap:* the S-DSM holds a 16-bit key on the chip, which exhaustive search finds and which an attacker who obtains a chip can read if it is stored in non-volatile memory, as this work assumes (Section II-E); silicon PUF key generators avoid a stored key but have not keyed a sensor link. *Proposal:* **a 128-bit key that is never stored, regenerated at every power-up from the chip's own SRAM array** through a BCH fuzzy extractor and a hash, and held only in volatile registers (Section IV).
2. **The keystream and the frame.** *Gap:* the S-DSM's 16-bit LFSR keystream is linear, so that Berlekamp-Massey recovers it from 32 keystream bits, and its frames are not authenticated, so that forged, altered or replayed frames are accepted. *Proposal:* **the keystream and a 32-bit MAC of every frame from one Grain-128a core** at the modulator's bit rate, with the front end, the modulator and the bit-level XOR of [1] unchanged.
3. **Freshness.** *Gap:* the S-DSM has no nonce, so that its keystream repeats after every power-up. *Proposal:* **a chip nonce, hashed from 64 power-up reads of the same SRAM cells**, in every initialisation vector beside the receiver's session nonce and a frame counter, so that a replayed session nonce repeats a keystream only on a 31-bit hash collision.
4. **Temperature.** *Gap* (**found in this work**): the key must hold over the sensor's temperature range, but at the fast-nMOS corners up to 26 % of the SRAM cells reverse their preferred state at 85 °C, so that an enrolment at 25 °C cannot meet the key-failure target at every corner (Sections VII-F and VIII-F). *Proposal:* **a one-time enrolment at −40 and 85 °C that keeps only the cells stable and equal at both temperatures**, sized from the simulated cells (a BCH code that corrects up to t = 17 flipped cells, t being its correction limit).
5. **The evaluation.** *Gap:* [1] evaluates an eavesdropper with a replica chip and two cryptanalyses on measured silicon at one condition, and no prior evaluation of such a link is open and executable. *Proposal:* **both designs evaluated from transistor to system in one reproducible notebook**, with open-source tools only: a model consistent with the measurements of [1] (48.2 % and 50.0 % of bits differ between two channels, against 49.8 % and 50.05 % measured; Section VI); circuit parameters from extracted SKY130 layouts of a StrongARM comparator at 45 PVT conditions and a 6T SRAM cell at 15 (Section VII); their effect on accuracy, power and security (Section VIII); 15 attacks, with **a leakage test** of where each design exposes its key to power analysis (Sections IX-D and X-H); RTL verified against published test vectors, synthesised, placed and routed in SKY130 in the same low-power form for every compared design and ranked by **a proposed figure of merit** that scales no number (Section X); **a key-generator compiler** from a reliability and secrecy specification to a decoder verified bit for bit against the model and placed and routed (Section XI); and **one simulation of the complete sensor across abstraction levels**, from transistor-level SRAM cells and comparators through the RTL decoder and cipher to the receiver's MAC check (Section XII).

**Summary of results.** The original design fails 11 of the 15 attacks; the proposed design resists 10 and detects 5 with the two-temperature enrolment (9 and 5 at the working setting t = 45, where the helper data leave too few secret bits, A11). **The analog circuits do not secure the original design**: comparator noise protects it only from 197× its worst PVT value, at 7.5 dB of signal-to-noise-and-distortion ratio (SNDR). The extracted SRAM cell has about one third of the assumed noise-to-mismatch ratio (0.0225 against 0.06), so that a 600-cell code enrolled at 25 °C needs t = 39 for the $10^{-6}$ target on the probability that the key fails to regenerate where the secrecy budget allows at most t = 35, and no searched code lets a 25 °C enrolment meet the target at all five corners; the two-temperature enrolment does, with t = 17.

**Every number is recomputed in one Colab session** (How to run this notebook, before the Abstract), except those of the recorded runs listed in the Appendix, which their switches or scripts re-run.

# **II. Background**
---
Section I traced the weaknesses of the S-DSM to **a stored short key, a linear keystream without a nonce and unauthenticated frames**. This section explains **why each of them matters and what removes it**, through the four concepts underlying the rest of the notebook. Subsection A explains how the sensor's delta-sigma (ΔΣ) modulator produces the raw bits that the S-DSM encrypts, and hence why encrypting them costs little power and why the bits may be hard to predict; B, what makes a keystream secure, how the S-DSM's 16-bit LFSR generates one, and the two attacks that recover it; C, why encryption does not prevent forgery, and what does; D, how a PUF with a fuzzy extractor derives a key from the chip's own transistors and reproduces it exactly at every power-up; and E describes the design of [1] that this work builds on, the secure delta-sigma modulator (S-DSM), with the assumptions this work makes where [1] is silent.

### A. Delta-sigma modulation: an oversampling 1-bit converter

In the sensor of [1], a second-order 1-bit continuous-time ΔΣ modulator digitises the current of the transimpedance amplifier, and its output bits are what the S-DSM encrypts and transmits. **The same modulator is the analog core of this work**: Section V models it bit for bit, Section VII simulates its StrongARM comparator at transistor level, Section VIII carries the comparator's noise and offset to the accuracy and security of the link, and Section XII runs the loop with the transistor-level comparator. This subsection gives the properties of the modulator on which these sections rely.

The modulator digitises a slowly varying signal with a **1-bit quantiser clocked much faster than the signal**: in [1], a clock rate $f_\mathrm{s}$ = 10 MHz against a signal bandwidth $f_\mathrm{B}$ = 100 kHz, an oversampling ratio $\mathrm{OSR} = f_\mathrm{s}/(2 f_\mathrm{B}) = 50$, or one output reading per 50 bits. Feedback makes the bits average to the input. An integrator accumulates the difference between the input $u$ and the fed-back output bit $v = \pm 1$ (the transmitted bit is $y \in \{0, 1\}$, with $v = 2y - 1$), and the comparator outputs +1 when the integrator is high, which pulls it back down, and −1 when it is low. The second-order loop of the model (Section V) is, at clock $k$, with integrator states $x_1$ and $x_2$,
$$x_1[k] = x_1[k-1] + \tfrac{1}{2}\,(u[k] - v[k-1]), \quad x_2[k] = x_2[k-1] + \tfrac{1}{2}\,(x_1[k] - v[k-1]), \quad v[k] = \begin{cases} +1 & x_2[k] \ge 0 \\ -1 & \text{otherwise,} \end{cases}$$
to which Section V adds leakage, noise, offset and dither. Because the integrators stay bounded, the long-run average of $v$ equals that of $u$: the **density of ones follows the input**. Fig. 3 shows the loop at work: below its block diagram, the integrator states $x_1$ and $x_2$ and the fed-back bits $v$ evolve clock by clock, while the mean of 50 consecutive output bits is compared with the input $u$.

The loop also shapes the quantisation noise. With the quantiser written as $v = \kappa x_2 + e$, where $\kappa$ is its gain and the large rounding error $e$ is treated as additive noise, the noise and signal transfer functions (NTF and STF) are, in the z-transform variable $z$ and with denominator $D(z)$,
$$\mathrm{NTF}(z) = \frac{(1 - z^{-1})^2}{D(z)}, \quad \mathrm{STF}(z) = \frac{\kappa/4}{D(z)}, \quad D(z) = 1 + \left(\tfrac{3\kappa}{4} - 2\right) z^{-1} + \left(1 - \tfrac{\kappa}{2}\right) z^{-2}.$$
At low frequency $D \approx D(1) = \kappa/4$, so that the input passes with unit gain and $\mathrm{NTF} \approx (4/\kappa)(1 - z^{-1})^2$, whose magnitude at frequency $f$, $(4/\kappa)\,(2 \sin(\pi f/f_\mathrm{s}))^2$, grows as $f^2$ (40 dB per decade) for any $\kappa$. **The quantisation noise is thus pushed above the band**, where the receiver's decimation filter removes it, leaving a high-resolution reading at $f_\mathrm{s}/\mathrm{OSR}$ = 200 kHz. The additive-noise model predicts the slope of the shaped noise but not its level, which scales with $1/\kappa$, since the gain of a 1-bit quantiser is not defined. Fig. 4 therefore checks the slope on the measured spectrum of the bit stream, against a dashed 40 dB/decade line and with the signal band shaded.

The additive-noise analysis above therefore predicts, and Fig. 3 and Fig. 4 test, a signal tone on a low floor inside 0–100 kHz, **noise rising at 40 dB per decade** (|NTF| ∝ f², power ∝ f⁴) above the band, and bits that toggle often while their 50-bit mean tracks the input.

**Key insights.** The modulator's bits carry the reading only in their average, and its quantisation noise lies outside the signal band: **the two properties that let the S-DSM encrypt the raw bits**.

* In Fig. 3 the output bits toggle every few clocks, yet their 50-bit mean follows the input. In Fig. 4 the noise rises along the +40 dB/decade line above the band, where the 20 kHz tone stands out from a low floor; the longer the FFT, the lower the in-band floor reads, because the noise power is spread over more bins.
* *Cost:* encrypting the raw bits one by one needs one XOR gate per bit and no decimation filter on the sensor, the low cost of the S-DSM.
* *Security:* each bit depends on the input history, on circuit noise and on every earlier decision, which [1] relies on against a replica chip (tested in Sections VIII and IX); **the bit density, however, still carries the input, so the bits must be encrypted**.

### B. Stream ciphers: XOR with a keystream

Subsection A showed that the modulator bits still carry the input and must be encrypted; the S-DSM does so with a stream cipher.

*1) Encryption by XOR:* The transmitter sends C = D ⊕ K, with D the data bit and K a keystream bit generated from a secret key k; the receiver computes C ⊕ K = D. Fig. 5 follows these operations bit by bit at the transmitter, at the receiver and at an attacker who XORs with a guessed keystream K̂. If K is unpredictable, C reveals nothing about D: **all the security rests on the keystream**.

*2) Two requirements:* (a) The keystream must be unpredictable without the key, even after some of it has been seen. (b) **It must never be reused**: two frames 1 and 2 under the same keystream give $C_1 \oplus C_2 = D_1 \oplus D_2$, whatever the key. A stream cipher therefore takes a public initialisation vector (IV), different for each frame, as well as the key.

*3) The linear feedback shift register (LFSR):* A shift register of length $L$ whose new bit is the XOR of a few of its bits (the *taps*); with well-chosen taps it cycles through all $2^L - 1$ non-zero states and outputs a long, balanced, random-looking sequence from a few gates. The S-DSM uses a 16-bit LFSR (taps 0, 2, 3 and 5) **whose initial state is the key**: $2^{16} - 1 = 65{,}535$ possible keys.

*4) Attack 1: exhaustive key search:* With a few known plaintext bits (a fixed field of the frame), **the attacker tests every key until the keystream matches**: at $R$ keys per second, $65{,}535/R$ seconds for a 16-bit key, $2^{128}/R$ for a 128-bit key. Fig. 7 sets the two times side by side on a logarithmic axis for the rate measured on one laptop core.

*5) Attack 2: Berlekamp-Massey:* The linear complexity L(n) of a sequence is the length of the shortest LFSR that generates its first n bits; the Berlekamp-Massey algorithm [3] finds that LFSR from the bits alone. For an LFSR of length L, **2L observed bits** recover the register: 32 bits for the S-DSM, after which every future bit is known. For a random sequence L(n) grows as about n/2 without bound. Fig. 6 contrasts the two cases: it draws the S-DSM's register, whose power-up state is the key, above the linear complexity of both keystreams against the number of bits observed.

*6) Nonlinear stream ciphers:* Trivium [5] (80-bit key) uses three shift registers with nonlinear (AND) feedback. Grain-128a [6], used in the proposed design, combines a 128-bit LFSR with a 128-bit *nonlinear* feedback shift register, a 128-bit key and a 96-bit IV, through a nonlinear output function, **so Berlekamp-Massey does not converge**; the published fast correlation attacks [18], [19] need far more keystream bits than one frame (Section III-E).

The analysis above predicts that, for the LFSR, L(n) stops at 16 and **a register rebuilt from 32 bits predicts every later bit**, whereas for Grain-128a L(n) follows n/2 and prediction is at chance (50 %). It also predicts that a 16-bit key falls to exhaustive search in a fraction of a millisecond on one core, whereas a 128-bit search is infeasible.

**Key insights.** The XOR hides the data from anyone without the keystream; **what fails in the S-DSM is the 16-bit LFSR that generates the keystream**, in both ways the analysis above predicts.

* With a guessed key, the attacker's decryption is wrong on 50.4 % of the 107,400 bits of a frame, while the receiver makes no errors (Fig. 5).
* In Fig. 6 the LFSR's linear complexity stops at 16, and Berlekamp-Massey rebuilds the register from 32 bits and predicts the next 1,000 bits with 100 % accuracy; over 160 bits, Grain-128a's L(n) tracks n/2, and the rebuilt register predicts the next 1,000 bits with 52 % accuracy.
* The exhaustive search (Fig. 7) reaches the model's key 0xC3A5 after 50,085 of 65,535 values, in 76 µs; the whole 16-bit space takes about 0.1 ms, whereas all 2¹²⁸ keys would take 5.1 × 10²⁹ s, about 1.2 × 10¹² times the age of the universe.
* The XOR itself is sound; **the weakness of the S-DSM lies in its keystream generator**, which a few known plaintext bits, provided by any fixed frame field, break in two ways (attacks A2 and A3 of Section IX). A 128-bit nonlinear cipher resists both, but encryption alone still lets an attacker alter or invent frames.

### C. Message authentication and freshness: encryption does not provide integrity

A forged reading is as harmful as a stolen one (Section I).

*1) Encryption is malleable:* for any bit pattern Δ, (C ⊕ Δ) ⊕ K = D ⊕ Δ: **inverting a bit of C inverts the same bit of D, without the key**. With the key or the keystream (subsection B), the attacker can encrypt a fabricated reading.

*2) Message authentication code (MAC):* A short tag $\tau$ computed from the message and the key; **the receiver recomputes it and rejects a frame whose tag differs**. Without the key, a $b$-bit tag is guessed with probability $2^{-b}$ per attempt, $2^{-32}$ in the proposed design. Grain-128a has a built-in MAC [6]: in its authenticated mode it generates two bits per data bit, one to encrypt and one to update the tag.

*3) Freshness: the frame counter:* A MAC does not stop replay of an old, valid frame, so each frame carries a 32-bit counter c, and **the receiver accepts only a counter larger than the last accepted one**.

*4) The nonce and the IV:* A **nonce** ("number used once") need not be secret, only never repeated under a key. Requirement (b) of subsection B forbids two frames with the same keystream, so the cipher starts from the key and an IV built from three fields:
   * **N, the session nonce**: a fresh random 32-bit value that the receiver sends in clear at the start of each session;
   * **c, the frame counter** of item 3, different for every frame of a session;
   * **$N_\mathrm{c}$, the chip nonce**: 31 bits that the chip hashes from 64 power-up reads of its own SRAM array, whose noisy cells differ at every power-up (one read is insufficiently noisy; Section VIII-F). Because $N$ travels in clear, an attacker can replay an old $N$; the counter restarts with each session, so without $N_\mathrm{c}$ the old keystream would repeat (attack A15). With it, a repeat needs a chance collision of its 31 bits (about 2⁻³¹ per pair of sessions), provided the 64 reads are independent and limited by thermal noise.

   The IV is $1 \,\Vert\, N \,\Vert\, c \,\Vert\, N_\mathrm{c}$ ($\Vert$: concatenation; the leading 1 selects the authenticated mode; 1 + 32 + 32 + 31 = 96 bits). Fig. 8(b) shows how these fields form the IV, Fig. 8(a) where the counter, the encrypted fields and the tag sit in the uplink frame, and Fig. 8(c) the downlink that delivers $N$ at session start. **The S-DSM of [1] has no IV**: its LFSR starts from the stored key alone.

The analysis above predicts that a frame with a changed payload, counter or encrypted chip ID fails the MAC check, except with probability $2^{-32}$, and that **a frame from an earlier session or with an old counter is rejected**.

(c) Downlink at session start: receiver nonce $N$ and helper data $h$.</p>

**Key insights.**

* Fig. 8 shows where each mechanism sits in the frame. They cost the link little: **96 bits per frame** (32-bit counter, 32-bit field for the 31-bit chip nonce, 32-bit tag, against a 107,400-bit payload) and, once per session on the downlink, 482 bits at the working setting t = 45 of the number of errors the BCH code corrects (subsection D; the 32-bit nonce and 450-bit helper data; the key generators of Sections VIII-F and XI need different helper data). The attacks that test them (spoofing, replay, bit-flipping) run in Section IX.
* The S-DSM has no MAC, counter or nonce, so **its receiver cannot tell a genuine frame from a forged or replayed one**. In the proposed design the protection costs one extra Grain-128a output bit per data bit; all of it depends on the key, whose origin subsection D explains.

### D. Physically unclonable functions and fuzzy extractors

A key held on the chip (assumed here to be non-volatile memory; Subsection E) can be read by an attacker who obtains the chip, and **copied into every chip of the design**.

*1) The SRAM cell at power-up:* A 6T SRAM cell is two cross-coupled inverters (nodes Q and Q̄) and two access transistors. When the supply $V_\mathrm{DD}$ ramps up with the access transistors off, **random mismatch (mainly threshold voltage) gives one inverter the advantage**, and the cell settles to 0 or 1 [13].

*2) Physically unclonable function (PUF):* Mismatch is random between chips, fixed for a chip and stored nowhere, so the power-up pattern of n cells is a fingerprint w: **different chips differ in about half of their cells, and the same chip nearly reproduces its pattern**. Fig. 9 illustrates both steps: ngspice transients of the storage nodes of three cells as $V_\mathrm{DD}$ ramps, and the fingerprints of two chips over repeated power-ups, with the cells that differ from the first power-up marked.

*3) Noise makes some cells unstable:* With mismatch spread $\sigma_\mathrm{m}$ and thermal-noise spread $\sigma_\mathrm{n}$ at the decision, **cells with small mismatch resolve differently between power-ups**; their fraction grows with $s = \sigma_\mathrm{n}/\sigma_\mathrm{m}$, and the cells that differ from the enrolled pattern grow with the distance from the enrolment temperature, partly through $s$ and partly through cells whose preferred state shifts with temperature.

*4) Fuzzy extractor [4]:* The key must be reproduced bit for bit, so a Bose-Chaudhuri-Hocquenghem (BCH) code over the Galois field GF(2¹⁰) that **corrects up to t errors removes the noise**.
   * *Enrolment* (once, at test): w is read; its syndrome h, a set of parity checks of w, is stored at the receiver as public helper data; the key is k = Hash(w), the first 128 bits of SHA-256 in the model.
   * *Reconstruction* (every power-up): the chip reads w′ = w ⊕ e, with e the pattern of flipped cells, and receives h; by linearity, syndrome(w′) combined with h is the syndrome of e alone, which the decoder finds if it has at most t ones, restoring w = w′ ⊕ e and k = Hash(w). Fig. 10 walks through these stages on a 600-cell grid, from enrolment to the decoding of a genuine power-up, and adds a counterfeit chip given the same helper data.

*5) Key failure:* **Reconstruction fails when more than $t$ cells flip**; for independent flips with probability $p$ per cell and $j$ flipped cells, $P(\text{key fails}) = \sum_{j>t} \binom{n}{j} p^j (1 - p)^{n-j}$, to be kept below $10^{-6}$ over the operating temperature range. Fig. 11 applies this sum to one chip enrolled at 25 °C: it shows the distribution of j relative to t as the temperature rises, and P(j > t) is compared with the $10^{-6}$ target from 0 to 85 °C.

*6) Helper-data leakage:* A BCH syndrome over $\mathrm{GF}(2^{10})$ **reveals 10 bits per correctable error**, $10t$ bits about $w$. For a nearly uniform 128-bit key, $n H_\infty - 10t \ge 128 + 2 \log_2(1/\epsilon) = 192$ bits for a key within statistical distance $\epsilon = 2^{-32}$ of uniform (Section VII-E), with $H_\infty$ the min-entropy (worst-case unpredictability) of one cell in bits.

The analysis above predicts that two chips differ in about 50 % of their cells; that a genuine power-up has a few to a few tens of flips, fewer than t, and recovers the key; that **a counterfeit given the same helper data has about n/2 errors and fails**; and that flips, and P(key fails), rise steeply with temperature.

**Key insights.** The SRAM array behaves as a key source must: **its pattern is unique to the chip, reproducible enough to be corrected, useless to a copy**, and least reliable far from its enrolment temperature, which sets the size of the code.

* *Unique to the chip.* The two modelled chips of Fig. 9 differ in 53 % of their cells, close to the 50 % expected from independent mismatch; the transients show how a cell settles (Section VII-C gives the statistics).
* *Reproducible within the code's reach.* The same chip differs from its first power-up in 4 to 12 cells at 25 °C and 8 to 14 at 85 °C, far fewer than the t = 45 the code corrects. In Fig. 10 a genuine power-up at 37 °C has 13 flipped cells, all corrected, and recovers the key (the reference frame Fig. 17 shows 27 flips at the same temperature because it uses the model's assumed, about 2.7 times noisier cell, $s_{25} = 0.06$, $t = 40$).
* *Useless to a copy.* The counterfeit chip given the same helper data has 318 errors (53 %), about n/2 and seven times what the code corrects, and derives a different key: the helper data reconstruct the key only on the chip that produced them.
* *Least reliable far from enrolment.* In Fig. 11 the mean number of flipped cells rises to 27.0 and $P(j > 45)$ to **1.5 × 10⁻⁵ at 85 °C** for this one chip, above the 10⁻⁶ target: the code must be sized for the worst temperature of the range, not for room temperature. This chip's temperature law, $s(T)$ at temperature $T$, anchored to the simulated post-layout error rates at −40, 25 and 85 °C, differs from the cell model's linear law, which underestimates the simulated 85 °C error rate (1.75 % against 3.14 % for the extracted cell). Section VII-E sizes the code over 1,000 modelled chips; $t = 45$ is the working setting of the system simulations.
* A PUF key is not held in non-volatile memory, so an unpowered chip holds no key to read out (provided the array and key registers have no read or scan path after test; A7), and a cloned design does not share it. The cost is the error-correcting code, whose size t depends on s and on temperature: hence the SKY130 simulation of s in Section VII, and the code sized for the worst process corner in Sections VIII and XI.

### E. The secure delta-sigma modulator of [1]

Yasar *et al.* [1] encrypt at the output of the data converter itself. The second-order 1-bit delta-sigma modulator of Subsection A outputs a high-rate bit stream that toggles frequently even for a constant input. In this secure delta-sigma modulator (S-DSM), each output bit is XORed with one bit of the keystream of a 16-bit linear-feedback shift register (LFSR), whose key is shared with the receiver; the receiver applies the same keystream and filters the recovered bits into readings. According to [1], dither injected at the comparator further reduces the predictability of the bitstream, and two channels that encrypt the same input with the same key disagree on about 50 % of their bits, so that an eavesdropper with a replica chip cannot predict the keystream. The security function costs **24 µW, 1.8 % of the chip power without security** (the overhead measure of [1]), and because the bits are sent without decimation, the wireless energy for the same information stays up to **2.67× below** that of its decimate-then-Trivium baseline at an oversampling ratio of 30 [1].

<p align="center" style="text-align:center"><img src="https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/dbb22aab44d606226b956ce604be2fc14db322b0/img/primer_sdsm_vs_proposal.png" width="760" alt="primer" style="display:inline-block"></p>

<p align="center" style="text-align:center"><b>Fig. 12.</b> Signal chain and key path of the two designs. (a) S-DSM [1]: a 16-bit key held on the chip seeds a 16-bit LFSR; no message authentication code (MAC) tag. (b) Proposal: no key in non-volatile memory (NVM); a 128-bit key regenerated from static random-access memory (SRAM) cells (600 in the system simulations) at every power-up keys Grain-128a, which adds a 32-bit MAC tag to every frame.</p>

Fig. 12 contrasts where the key lives. How the S-DSM stores its key is not described in [1]; this work assumes non-volatile on-chip memory (fuses, one-time-programmable memory or flash), on which only the key-extraction attack A7 depends. Nor does [1] state whether its LFSR restarts from the key in every frame; the model assumes that it does, and the verdicts of the multi-frame saturation attack (A6) and of keystream reuse across sessions (A15) rest on this assumption. **Three of the original design's 11 failures (A6, A7, A15) therefore depend on assumptions that [1] does not state.** If the key were instead loaded at power-up from a host, it would still sit in a register while the chip runs, be shared by every chip given it, and fall to brute force (A2) and Berlekamp-Massey (A3) from eavesdropping alone, so that the original design would fail 10 rather than 11 of the 15 attacks.

These concepts state the requirements precisely: a modulator bit that is cheap to encrypt, a keystream that is neither short nor linear nor repeated, frames that are authenticated and fresh, and **a key that is never stored**. Section III asks which prior approach meets them within a sensor's power budget, and what remains open.

# **III. Prior work**
---
The S-DSM of [1] is one of several answers to the problem of Section I. With the concepts of Section II in place, this section asks **what prior work has done to secure a low-power sensor link, and what it leaves open**. Subsections A to D review prior work in four groups, each for its approach, what it achieves and what it leaves open: cryptographic engines (A), physical-layer and analog-domain security (B), the in-modulator encryption of [1] (C), and silicon PUFs and key generators (D). Subsection E draws from published attacks the capabilities an attacker should be granted, and F states the gap that the rest of the notebook addresses and how far this work fills it.

Each line of work is **judged against the six requirements R1 to R6** of Section I-A, which reappear as the property rows of Table 36 (Section XIV).

### A. Cryptographic engines

Cryptographic engines realise a standard, well-analysed cipher or protocol as a dedicated low-power circuit. Intel's NanoAES [8] implements the Advanced Encryption Standard (AES-128) in 2,090 gates (22 nm), reaching 289 Gbps/W near threshold and 170 µW at 340 mV. SIMON 32/64 block-cipher engines [10] need as few as 548 gates. A full Datagram Transport Layer Security (DTLS) 1.3 engine with elliptic-curve key exchange [9] secures an Internet-of-Things (IoT) link for 44 µJ per handshake and 0.89 nJ/byte (65 nm). All three are fabricated and measured chips, in 22 nm [8], 14 nm [10] and 65 nm [9] CMOS, and [1] reports 220 µW for a Trivium stream cipher [5] next to its sensor, against 24 µW for the S-DSM. **The AES and DTLS engines meet R3 with 128-bit keys**, and the protocol engine also R5; SIMON 32/64, with a 64-bit key, falls short of R3.

As built, however, these engines encrypt *words*, **so the sensor decimates first (R2)**, forfeiting the wireless-energy advantage of sending the bits; run as keystream generators (a block cipher in counter mode) they could encrypt the bits directly. Each keeps its key, or the credentials it derives session keys from, inside the chip, to be provisioned and protected (R4). Next to the front end, a cipher such as the 220 µW Trivium of [1] is **a large overhead (R1)**.

### B. Physical-layer and analog-domain security

Physical-layer and analog-domain approaches draw security from the physics of the link or of the converter rather than from a digital cipher alone [2]. Bit-level frequency hopping counters selective jamming of Bluetooth Low Energy (BLE)-type links, demonstrated in a fabricated transmitter [11]. Physical-layer key generation derives a shared key from the reciprocal fading of the radio channel itself, demonstrated on commodity 802.11 radios [32]. Compressed-sensing encryption, reviewed in [2], uses the secret sensing matrix as the key and encrypts during acquisition at little added cost; [33] formalises it for low-complexity sensor nodes, concludes that it is not perfectly secure but gives some security at almost no cost, and applies it to electrocardiogram, image and speech signals. At the converter itself, S²ADC [12] shows that, without protection, the supply current of a successive-approximation analog-to-digital converter (ADC) **reveals its output bits to a neural-network attack with more than 99 % bit-wise accuracy** (except the least significant bit), and proposes a countermeasure.

Frequency hopping, however, **costs 10.9 mW (R1)** and still relies on encryption for confidentiality. Compressed-sensing encryption is a linear encoding and therefore not perfectly secret (R3) [2], [33], and a channel-based key accrues only as fast as the radio channel varies [32]. **None of these approaches authenticates frames (R5) or removes the stored key (R4)**.

### C. In-modulator encryption: the S-DSM [1]

The S-DSM carries the analog-domain line of B into the ΔΣ modulator: each output bit is XORed with a 16-bit LFSR keystream bit (Section II-A, B) and sent without decimation. It is demonstrated in silicon: [1] integrates it in a 65 nm multimodal electrochemical sensor chip for biomanufacturing and reports the two-channel measurements, the wireless-energy comparison and the chip's power breakdown that Sections VI and X reuse. It costs 24 µW, 1.8 % of the chip power (R1), and its wireless energy is up to 2.67× lower than that of decimate-then-encrypt at an oversampling ratio (OSR) of 30 (R2; Section II-E). **Among the works reviewed here, it is the only design that meets R1 and R2 together**, and this work therefore retains its architecture.

**It leaves the other requirements open**: a linear keystream recovered from 32 bits and a 16-bit key open to exhaustive search (R3); a stored key shared by every chip programmed with it (R4); no authentication or freshness (R5); and an evaluation of two cryptanalyses (R6).

### D. Silicon PUFs and key generators

PUFs and the key generators built on them **derive the key from random manufacturing variation, so that no secret is stored** (Section II-D). SRAM power-up states identify chips and supply randomness without added circuits [13]; fuzzy extractors [4] turn a noisy PUF into a key with an error-correcting code. PUFKY [14] combines a ring-oscillator PUF, a BCH syndrome secure sketch and an entropy accumulator into a complete 128-bit key generator with a failure rate below 10⁻⁹ (FPGA), and [26] implements a helper-data key extractor for SRAM PUFs with concatenated repetition and linear codes on FPGAs, the code structure that the compiler of Section XI generates. SRAM PUFs age, and writing the inverse of the power-up pattern back into the array counters it [30]. Circuit PUFs reach 0.19 pJ/b with stabilisation by burn-in and dark-bit masking [15], and 1.02 fJ/b with a 3.2 % native bit-error rate in a fully synthesisable cell [16]. Volatile PUF keys resist invasive probing better than fuse-based storage [15]. A wearable security engine combines the electrocardiogram (ECG), heart-rate variability and an SRAM PUF for authentication and key generation at 4 µW [17]. Sensor PUFs [20] make the response depend on both the sensed quantity and the chip's variation, which authenticates the measurement. The key generators [14], [17] meet R3, and **all these works address R4** (no stored secret and, under the PUF assumption, no clone).

None of these works, however, **encrypts or authenticates a sensor link (R2, R5)** or is evaluated against attacks on one (R6).

### E. Attacks that shape the threat model

Attacks on real devices and on the ciphers themselves show which capabilities an attacker should be granted. **Wireless medical devices have been attacked in practice**: software-radio attacks on an implantable cardioverter defibrillator replay captured commands that change its therapy [37], and attacks on a commercial glucose-monitoring and insulin-delivery system range from eavesdropping to forged commands [38] (R5). Keys held in non-volatile memory can be read out by an attacker who holds the chip, for example from the protected flash memory of a microcontroller by optical fault injection [34], and SRAM can retain data at low temperature [22] (R4); even an SRAM PUF has been cloned in the laboratory [21], and the public helper data of some key generators can be manipulated to learn the key [23], [24]. LFSR-based keystreams fall to correlation attacks [35], [36] and, when purely linear, to Berlekamp-Massey [3] (R3). Even Grain-128a has published full-round fast correlation attacks: that of [18] needs 2¹¹³·⁸ keystream bits and 2¹¹⁵·⁴ time, and the vectorial fast correlation attack of [19] lowers the data to 2¹⁰⁶·³ bits. Both are **far from practical and beyond one frame**, but below the nominal 2¹²⁸ level; [18] breaks only the mode without authentication, since in the authenticated mode only every second pre-output bit is observable as keystream.

For this work, **a realistic attacker listens, chooses the sensor's input, learns some keystream bits, obtains a chip and transmits**; Section IV turns these capabilities into the threat model.

### F. Research gap

Each line of work above solves part of the problem, and the parts do not combine by simple addition:

* *Cipher engines* (A) give a strong 128-bit cipher and, as a protocol engine, authenticated frames (R3, R5), but they work on decimated words and keep their key on the chip. Placing one on the sensor gives up the bit-level encryption that makes the S-DSM cheap (R2), and a cipher such as the Trivium of [1] costs about nine times the S-DSM's power (R1).
* *The S-DSM* (C) **is the only design reviewed that encrypts the modulator bits directly at a few per cent of the chip's power** (R1, R2), but its 16-bit LFSR, its stored key and its unauthenticated frames leave R3 to R5 open.
* *PUF key generators* (D) remove the stored key (R4), and some deliver 128-bit keys (R3), but none is connected to a sensor link, so none encrypts or authenticates frames (R2, R5).
* *Physical-layer and analog-domain techniques* (B) lower the cost of particular protections or block particular attacks, but give neither a strong cipher nor authenticated frames.
* *The published attacks* (E) show that an attacker who listens, chooses the input, holds a chip or transmits must be assumed, and no sensor-link evaluation among the works reviewed considers all of these at once, in a form that others can re-run (R6).

**The gap is therefore the combination**: a stored-key-free, authenticated, cryptographically strong link that still encrypts the modulator bits at microwatt cost, evaluated openly against that attacker. Table 2 maps each requirement to the works that meet it:

**Key insights.**

* **Among the works reviewed here, no design gives a sensor link with a security budget of tens of microwatts all of the following at once**: encryption of the modulator bitstream (no decimation on the sensor), a key that is never stored and that a cloned chip does not share (under the PUF assumption), a keystream that resists linear cryptanalysis, and authenticated and fresh frames (R1–R5). No prior evaluation of such a link is open and executable against a realistic attacker (R6).
* This work keeps the S-DSM's architecture [1] (R2), replaces the stored 16-bit key by an SRAM-PUF key regenerated through a BCH fuzzy extractor [4], [13], [14] (R4), replaces the LFSR by Grain-128a [6] (R3), and uses Grain-128a's authenticated mode with a frame counter, a session nonce and a chip nonce (R5). It evaluates both designs in one reproducible notebook grounded in SKY130 (R6).
* **This work meets R1 at a few per cent of the chip's power, above the S-DSM**: with its SKY130 gate-level power (Section X) added, unscaled, to the measured 65 nm power of the chip of [1], the keystream block takes 4.5 % of the chip's power against 1.7 % for the S-DSM.

Closing this gap begins with the attacker rather than the circuit, since **a security claim holds only against a stated attacker**. Section IV fixes what the attacker knows and can do, and then the architecture that answers each of the attacker's capabilities.

# **IV. Proposed work**
---
Section III ends with a gap: no prior design encrypts the modulator bits with an unstored key, a keystream that resists linear cryptanalysis and authenticated, fresh frames, at a security budget of tens of microwatts. Filling it raises two questions: **against which attacker must a delta-sigma sensor link be secure?** And **what must be added to the S-DSM of [1] to withstand that attacker?** As a security claim holds only against a stated attacker, this section fixes the attacker before any design is evaluated, then derives the proposed architecture from the attacker's capabilities and states what is new in it.

Subsection A states what the attacker knows and can do, so that every attack follows from a stated capability. Subsection B derives an architecture that **closes every attack** of Table 3 while keeping the S-DSM's unchanged front end and modulator and its one-XOR-per-bit encryption without decimation. Subsection C explains each of the four new ideas: where it fits in the chain, what it solves and how.

### A. Threat model

The threat model of [1] assumes an eavesdropper who knows the design, owns a replica chip and chooses its inputs, but has no physical access to the deployed sensor or its key; it excludes invasive attacks and denial of service such as jamming, which the controlled bioreactor would detect, and assumes that the symmetric key is shared securely in advance. We keep this adversary and **extend its capabilities step by step**, as a deployed product would encounter them, under five assumptions:

1. **The attacker knows the design** (Kerckhoffs's principle): the circuits, the LFSR taps, Grain-128a, the BCH code, the frame format and the chip ID, which is not secret. Only the key is secret: the 16-bit LFSR seed, or the 128-bit key $k$ derived from the SRAM fingerprint.
2. **The receiver, its key database and enrolment are trusted** (enrolment: Section II-D).
3. **Out of scope:** jamming and other denial of service; and, by assumption, invasive, fault-injection and side-channel attacks on a *powered* chip (probing the SRAM or the key register in use, power or electromagnetic analysis of the cipher and of the BCH decoder, whose run time depends on the number of errors, fault attacks on Grain), and the physical cloning [21] or remanence read-out [22] of an SRAM fingerprint. Reading the memory of an unpowered chip and building a counterfeit *are* in scope.
4. **No read-out path after test.** Enrolment reads the SRAM fingerprint $w$ out of the chip once; this path, and any read or scan path of the array and the key registers, is disabled before deployment, for example by a one-time-programmable fuse, which holds no secret. An attacker who re-opened it would obtain $w$ and hence $k$. The verdicts A7 and A9 rest on this assumption.
5. **Full power-down before every array read.** Every read of the SRAM array, at power-up and for each of the 64 reads of the chip nonce, starts from a supply discharged long and deep enough that no cell keeps its previous state; a short or shallow dip would reproduce the previous state, so that consecutive nonce reads would repeat each other. This is a requirement on the array's power switch, which is not simulated (Section VII-C), and control of the deployed chip's supply is not among the capabilities below.

Within these assumptions **the attacker has five capabilities** (Fig. 13), ordered roughly from weakest to strongest; an attacker at one level may use the levels before it.

1. **C1: listen to the radio.** The attacker records every transmitted frame and the public downlink (helper data and nonce). Unless stated otherwise, the attacker does not know the plaintext $D$.
2. **C2: choose the sensor input.** The attacker sets what the sensor measures (for example by spiking the culture), so the attacker knows the input but not the bits $D$ the modulator makes of it, which also depend on the circuit's noise and mismatch.
3. **C3: learn a few keystream bits.** The attacker learns a short run of keystream bits $K$ (64 in the model), for example from a known part of a frame.
4. **C4: hold the chip.** The attacker obtains a sensor chip: the attacker can read its non-volatile memory, or build a counterfeit that claims a genuine chip's identity.
5. **C5: transmit on the link.** The attacker is a man in the middle: the attacker can suppress, replace, replay or alter frames on the uplink, and the helper data and the session nonce on the downlink.

Table 3 derives the attacks from the capabilities. **An attack succeeds if the attacker reads the data** (confidentiality), makes the receiver accept chosen or altered data (authenticity, integrity) or an old frame (freshness), or learns about the data or the key (key secrecy). Of the attacks marked *[1]*, A4 and A5 are evaluated in [1], and A1 and A8 are named there without an evaluation.

**Key insights.** The four attacks of [1] (A1, A4, A5, A8) need only listening, a chosen input and forgery. The eleven added ones apply the same capabilities, plus holding the chip and learning keystream bits, to a short, stored key, a linear and repeated keystream and unauthenticated frames, and to the proposed design's own public data, the helper data (A10 to A12) and the session nonce (A15). ### B. The proposed architecture

Subsection A points to **three weaknesses of the original design**: a stored, short key, a linear keystream, and no frame authentication. The proposed design removes each with a standard building block of Section II (Fig. 14): a 128-bit key rebuilt from the chip's SRAM cells at each power-up, a Grain-128a keystream, and a counter, nonce and MAC tag on every frame. Notation:

* $D$: the modulator bits; $K$: the keystream; $C = D \oplus  K$: the transmitted bits;
* $w$: the chip's 600-bit SRAM fingerprint at enrolment; $w'$: the noisy fingerprint read at a later power-up; $e = w \oplus  w'$: the flipped cells;
* $t$: the number of flipped cells the BCH code corrects; $h$: the public helper data (the BCH syndrome of $w$, $10t$ bits);
* $k = \mathrm{hash}(w)$: the 128-bit key; $N$: the 32-bit session nonce; ctr: the 32-bit frame counter; IV: Grain-128a's 96-bit initialisation vector; tag: the 32-bit MAC.

<p align="center" style="text-align:center"><img src="https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/dbb22aab44d606226b956ce604be2fc14db322b0/img/architecture.png" width="860" alt="arch" style="display:inline-block"></p>

<p align="center" style="text-align:center"><b>Fig. 14.</b> Original design and proposal. (a) Original design [1]: front end, second-order continuous-time ΔΣ modulator, and LFSR-16 encryption with the stored 16-bit key (orange). (b) Proposal: the same front end and modulator (simplified), key generation, encryption and authentication (blue), and the receiver; Grain-128a drawn like the LFSR-16 of (a): its 128-bit NFSR with nonlinear feedback $g$ and its 128-bit LFSR with linear feedback $f$ (ten cells sketched each; the LFSR bit that also enters $g$ not drawn) and the output function (⊕; unlabelled, $h$ being the helper data), whose keystream $K$ drives the XOR and the MAC; $N_\mathrm{c}$: hash of 64 raw reads w′ and the session index; inset in the receiver: the frame format, field lengths in bits (tinted: encrypted). Dashed: public downlink of $N$ and $h$ at session start. S2D: single-ended-to-differential buffer; symbols as in the list above; tag′: tag recomputed by the receiver.</p>

The blocks are listed below **in the order a bit traverses them**; their parameters come from [1] (front end), the Grain-128a specification [6] and the SKY130 simulations of Section VII.

1. **Front end and ΔΣ modulator (unchanged from [1]).** The TIA converts the electrode current into a voltage, the single-ended-to-differential (S2D) buffer halves it, and the second-order 1-bit continuous-time modulator, clocked at 10 MHz, produces the bits $D$ (Fig. 14(a): class-AB counter-electrode driver, two-stage Miller TIA with feedback resistance $R_\mathrm{F}$ = 3.30 kΩ to 3.06 MΩ, StrongARM comparator, dither ladder and 1-bit digital-to-analog converter). Its accuracy is taken from [1] (65 nm) and checked with SKY130 parameters in Section VIII. *Defends against:* in the original design, A4 to A6 were to be stopped by the circuit noise and mismatch that make two chips' bits differ for the same input, and by the ×0.5 buffer, which keeps a railed TIA within the modulator's stable range; with the modelled dither, these defences fail against an attacker who scans every window, trims the replica and repeats frames (Section IX-A; the multi-frame attack also assumes the per-frame LFSR restart of Section II-E).
2. **Key generation at power-up: SRAM PUF, fuzzy extractor and hash.** At every power-up, 600 SRAM cells settle to the fingerprint $w'$; the BCH decoder, given the helper data $h$ from the receiver, restores $w$, and the hash turns it into $k$ (Section II-D). **No secret is held in on-chip non-volatile memory** [4], [13], [14]: the key exists only in volatile registers while the chip is powered. The system simulations use t = 45 with a 25 °C enrolment as a working setting, although with 600 cells no code then meets both the 10⁻⁶ key-failure target and the 192-bit secrecy budget (Section VII-E). The one-time enrolment at −40 and 85 °C, which brackets the 0–85 °C range and keeps only the cells stable and equal at both temperatures, meets both with t = 17 (935 cells, 600 kept; Section VIII-F). Section XI compiles it into a decoder verified in RTL simulation, placed and globally routed in this run (detailed routing recorded), with a SHA-256 core as the hash (a lightweight hash such as the SPONGENT-128 of [14] would be several times smaller), and finds a smaller code for an enrolment at 25 and 85 °C. The two-temperature enrolment adds to $h$ a public mask of the kept cells (935 bits); such pointer-type helper data is open to manipulation oracles [23], so the mask must be hashed with $w$ into $k$ [24], making any change yield a key that the MAC rejects (this binding is not modelled; the attacks of Section IX use the 25 °C enrolment, whose helper data is the syndrome alone). *Defends against:* A2 (128-bit key), A7 (no key in memory), A9 and A10 (a counterfeit's fingerprint or a corrupted $h$ gives a wrong key, which the ID and MAC checks detect), A11 (at t = 45 only 94 secret bits remain at the 95 % lower bound of the min-entropy, against 192, so A11 succeeds; the two-temperature enrolment, with 284–352 secret bits, resists it) and A12 (the syndrome construction and the hash make the accept/reject answers the same for any fingerprint with the same noise).
3. **Keystream: Grain-128a [6].** Its 128-bit LFSR and 128-bit nonlinear register are loaded with $k$ and the IV and clocked 256 times to initialise; in authenticated mode (first IV bit 1) the first 64 pre-output bits load the MAC's accumulator and shift register, then the even pre-output bits form $K$ and the odd bits drive the MAC (published test vectors reproduced in Section V-B). Grain-128a is preferred to the block-oriented NIST lightweight standard Ascon because its bit-serial datapath matches the modulator's one bit per clock without buffering; its successor Grain-128AEADv2 [25], with a 64-bit tag, would fit the same datapath (neither alternative is synthesised here). *Defends against:* A3 (its linear complexity grows with the output length; Section II-B) and A2.
4. **Counter, nonces, MAC and framer.** Each frame carries a 32-bit counter and the chip's 31-bit nonce $N_\mathrm{c}$ in clear, and the IV is $1 \,\Vert\, N \,\Vert\, \mathrm{ctr} \,\Vert\, N_\mathrm{c}$. Without non-volatile memory the chip keeps no counter across power cycles, so the uniqueness of the IV under a replayed $N$ rests on the cell noise: the array must be stable enough for the key and noisy enough for the nonce. One read is not enough (9.3 bits of collision entropy for the median chip at the typical corner tt, 25 °C, and 1.1 bits for the 1st-percentile chip at the fast-nMOS, slow-pMOS corner fs, −40 °C), so $N_\mathrm{c}$ is 31 bits of the hash of 64 raw reads $w'$ and a volatile session index, which give that chip 68 bits (Section VIII-F). A replayed $N$ thus repeats a keystream only on a 31-bit collision, provided that the reads are independent and limited by thermal noise, which assumption 5 and an on-chip health test (Section VIII-F) must ensure. The 32-bit MAC covers the counter, the encrypted chip ID and every payload bit of $C$ (encrypt-then-MAC); $N_\mathrm{c}$ enters through the IV. A forged tag passes with probability 2⁻³² per attempt; at about 93 frames per second a day of continuous attempts succeeds with probability about 2 × 10⁻³, so the receiver must limit the rejected frames per chip (or use a 64-bit tag, as Grain-128AEADv2 does). The counter wraps after 2³² frames, about 1.5 years of continuous operation, when the chip must be re-enrolled or the session renewed. The frame format is the inset of Fig. 14(b). *Defends against:* A8 and A14 (a valid tag needs $k$), A13 (old counter), A15 (chip-nonce collision only), and A4 to A6 (no two frames of a session share a keystream).
5. **Receiver.** For each frame it finds the sync word, checks that the counter is fresh, recomputes the tag, decrypts and checks the chip ID, and only if all three checks pass hands $D = C \oplus  K$ to the decimation filters (a cascaded integrator-comb (CIC) decimator and a compensating finite-impulse-response (FIR) filter). *Defends against:* every frame attack, by turning a forgery, replay, counterfeit or tampered key into a **rejected frame** (ATTACK DETECTED) instead of false data.

The session protocol runs in six steps:

1. **Enrolment** (one-time, at test, trusted): $w$ is read (at 25 °C in the system simulations; at −40 and 85 °C, with the cell mask added to $h$, in the final key generator); $h = \mathrm{syndrome}(w)$ and $k = \mathrm{hash}(w)$ are stored at the receiver with the chip ID. The chip keeps nothing.
2. **Power-up:** the chip reads its noisy fingerprint $w'$ (= $w'_1$), power-cycles the array 63 more times for the reads $w'_2$ to $w'_{64}$ and derives its chip nonce $N_\mathrm{c} = \mathrm{hash}(w'_1 \,\Vert \cdots \Vert\, w'_{64} \,\Vert\, \text{session index})$, truncated to 31 bits.
3. **Downlink:** the receiver sends $N$ and $h$, public and unauthenticated.
4. **Key reconstruction:** the chip decodes $e$ from $\mathrm{syndrome}(w') \oplus h$, restores $w = w' \oplus e$ and computes $k = \mathrm{hash}(w)$.
5. **Each frame:** ctr ← ctr + 1; Grain-128a is initialised with $k$ and $\mathrm{IV} = 1 \,\Vert\, N \,\Vert\, \mathrm{ctr} \,\Vert\, N_\mathrm{c}$; the chip ID and the bits $D$ are encrypted, the tag over (ctr, encrypted ID, $C$) appended, and the frame sent.
6. **Reception:** the receiver rebuilds the IV from its $N$ and the frame's ctr and $N_\mathrm{c}$ and runs the three checks of block 5.

Fig. 15 is a recorded snapshot of earlier settings: its correction limit (37) and signal-to-noise-and-distortion ratio (SNDR, 59.8 dB) precede the present t = 45 and SNDR of 60.0 dB (Section VIII). It shows the signal of every block at once: the key restored after correction, the frame verified, decrypted and filtered back into the sensor signal, and an attacker's keystream from a guessed key agreeing with the true one on **only about 50 %** of the bits.

**Key insights.** **Every one of the 15 attacks meets at least one block designed to stop it**, and most meet two (Section IX-A runs them all; four of the proposed design's verdicts hold by design, not by simulation). The costs, quantified later, are a new failure mode, a key that the PUF fails to regenerate when more than $t$ cells flip (Sections VIII-F, IX-B, XI); a digital block, whose area and gate-level power come from the SKY130 implementation (Section X), plus the key generator, whose area is synthesised in Section XI but whose power and energy are not costed; and a session-start downlink for $N$ and $h$ and 64 power-up reads for the chip nonce.

### C. The new ideas: where each fits, what it solves and how

Relative to the S-DSM of [1], the proposed design adds an SRAM array and **changes only the digital part of the chip**: the front end, the modulator and the encryption of each modulator bit by one XOR before any decimation, which give [1] its energy advantage, **are kept**. **Four ideas are new to in-modulator encryption** (Fig. 14(b)).

***1) A key that is never stored.*** *Where:* at power-up, between the SRAM array and Grain-128a (block 2). *What it solves:* **the extraction of a stored key from a captured chip (A7), a key shared by every chip of the design, and the exhaustive search (A2) of a 16-bit key.** *How:* the 600 SRAM cells power up to a fingerprint $w'$ that is **unique to the chip**; the BCH decoder, given the public helper data $h$ from the receiver, corrects the flipped cells to recover the enrolled fingerprint $w$, and a hash turns $w$ into **the 128-bit key $k$, which exists only in volatile registers while the chip is powered**.

***2) Keystream and authentication from one Grain-128a core.*** *Where:* in place of the 16-bit LFSR of [1], at the modulator's bit rate (blocks 3 and 4). *What it solves:* **the linear keystream that Berlekamp-Massey recovers (A3), and the unauthenticated frames** through which forged, altered and replayed frames reach the server (**spoofing A8, replay A13**). *How:* Grain-128a, keyed with $k$ and the initialisation vector IV, produces **both the keystream $K$, XORed with each modulator bit, and a 32-bit MAC tag** over the counter, the encrypted chip ID and every payload bit (encrypt-then-MAC); **the receiver discards any frame whose tag or counter fails**.

***3) A chip nonce from the same cells.*** *Where:* at power-up, from the SRAM array into every IV (block 4). *What it solves:* **the reuse of a keystream across sessions when a session nonce $N$ is replayed (A15)**, since without non-volatile memory the chip keeps no counter across power cycles. *How:* 64 power-up reads of the array and a volatile session index are hashed into **a 31-bit chip nonce $N_\mathrm{c}$**, so that the IV, $1 \,\Vert\, N \,\Vert\, \mathrm{ctr} \,\Vert\, N_\mathrm{c}$, **repeats only on a 31-bit collision**; **the cell noise that the key must tolerate is what supplies the nonce**.

***4) A one-time enrolment at two temperatures.*** *Where:* once, at test (session step 1). *What it solves:* **key failures at the fast-nMOS corners**, where cells reverse their preferred state with temperature, so that with a 25 °C enrolment no code meets both the $10^{-6}$ key-failure target and the secrecy budget. *How:* the fingerprint is **read at −40 and 85 °C and only the cells stable and equal at both temperatures are kept** (935 cells, 600 kept, the cell mask added to $h$); **a BCH code correcting $t$ = 17 flipped cells then meets the target with 284–352 secret bits**.

**The receiver sends only public data, $h$ and $N$, and no key is stored on the chip or sent.** Section I-C sets these ideas against the gaps of prior work, and Sections IX and X evaluate their security and their cost.

Section V turns both designs into a bit-level model of the link, so that every attack of Table 3 can be run on each.

# **V. A bit-level model of the link**
---
Section IV defined both designs and the 15 attacks; a verdict requires an executable model of them. This section asks **what each design transmits, bit by bit, and what an attacker can compute from it**. The remainder of the paper rests on this model: the consistency check against [1], the system studies and the 15 attacks run on this model, and the RTL of Section X is verified against it bit for bit. **The model must be exact to the bit** because an attack on a stream cipher can hinge on a single bit: Berlekamp-Massey recovers a 16-bit LFSR from 32 correct keystream bits and returns a wrong register when one of them is wrong. It therefore computes every bit of each block explicitly: the modulator's decisions, the LFSR and Grain-128a keystreams, the PUF fingerprint and its BCH correction, the frame and the receiver's checks. The two designs and the attacker are those of Section IV; the principles and notation are those of Section II. The model is behavioural and cycle-accurate: one step per modulator clock, every digital signal at its true bit width. The analog blocks are difference equations with Gaussian noise; the digital blocks are implemented exactly from their specifications. The bit-serial loops (modulator, LFSR, Grain-128a, both Berlekamp-Massey algorithms, BCH syndrome and decoder, the key-failure recursion) are compiled with `numba`; the rest uses `numpy`. After compilation one frame, including key reconstruction, keystream, MAC and receiver, takes about 0.02 s on one core (measured on the development machine). The circuit parameters start at the values of [1] or at stated assumptions, which Section VII replaces by SKY130 simulation.

Subsections A to H build the link one block per code cell, in the order in which the blocks are needed. A fixes the operating point of [1] and collects every setting of both designs in one place, **so that each later experiment changes a named setting and nothing else**. B implements both keystream generators bit for bit and verifies Grain-128a against its specification. C models the analog chain that produces the modulator's output bits $D$, with the non-idealities that make two chips or channels produce different bits (noise, offset, mismatch, dither), and the receiver's filters. D models how the proposed design regenerates its 128-bit key at every power-up, how often this fails and how much secrecy the public helper data leaves. E implements the attacker's tools once, for every attack of Subsection G and every study of Section VIII. F joins the blocks into one function, `simulate()`, which runs a complete frame from the sensor current to the server's reading, so that any attack can act at its true point in the link. G defines, for each attack of the threat model (Section IV), the attacker's actions and the test that decides success; Section IX runs them. H runs one frame of the proposed design without attack, as the first end-to-end test, and introduces the signal view of the later frame-level figures.

### A. Operating point, settings and seeds

The modulator runs at the clock frequency $f_\mathrm{s}$ = 10 MHz with the oversampling ratio OSR = 50. The receiver therefore delivers **one reading per 50 bits**, at $f_\mathrm{s}/\mathrm{OSR}$ = 200 kHz, for a signal bandwidth of 100 kHz. A frame holds $N_\mathrm{out} = 2148$ readings (a 2,048-point FFT plus 100 readings for the filters to settle), that is $N = 2148 \times 50 = 107{,}400$ modulator clocks. The analog quantities are normalised to the modulator's full scale (FS) of half the supply voltage $V_\mathrm{DD}$, $V_\mathrm{DD}/2$ = 0.6 V, so that the modulator input $u$ lies in $[-1, +1]$; the integrator states $x_1$, $x_2$ and the comparator's noise and offset are expressed in the same FS units ("$x_2$ units").

`DEFAULTS` holds **every setting of both designs**, listed below with its source; any of them can be overridden for a single call (for example `simulate(temp=85)`). The assumed circuit values are placeholders that Section VII replaces by SKY130 values. `ATK_DEF` holds the attacker's parameters (Subsection G).

Every random quantity comes from its own stream of `rng_for(seed)` (the PCG64 generator of `numpy`), so **equal settings give identical bits**. The master seed 2026 drives the modulator's noise; the other streams use fixed offsets (the PUF reading at power-up +17, the channel bits +5, the session nonce +71, the payload flips of a man in the middle +77, and each attack its own, Subsection G). The quantities that describe a chip rather than a run have chip seeds: the PUF's mismatch (`chip_seed` = 7), its enrolment reading (`chip_seed` × 7919 + 1) and the two dither LFSRs (0x1D2B and 0x7A31). A second chip or channel is simulated by changing its chip seeds, never by reusing a stream.

````python
#@title Constants and settings
# =====================================================================================
# Behavioural model of the secure sensor link
#   * original design (ESSERC'25 S-DSM): 2nd-order 1-bit delta-sigma modulator, bits XOR a 16-bit LFSR
#   * our proposal: the same modulator, keyed by an SRAM PUF + BCH fuzzy extractor, Grain-128a with its
#     MAC, a session nonce and a frame counter
# Bit-serial loops (modulator, LFSR, Grain-128a, Berlekamp-Massey, BCH) are compiled with numba; the rest is numpy.
# =====================================================================================
import hashlib, math, time
import numpy as np
from numba import njit
from scipy.special import ndtr, ndtri

FS, OSR = 10e6, 50                     # modulator clock and oversampling ratio (paper: 10 MHz, OSR 50)
FS_OUT = FS / OSR
VDD, VCM = 1.2, 0.6                    # the paper's 65 nm supply; the front end is modelled behaviourally
N_FFT = 2048
N_OUT = N_FFT + 100
N_DSM = N_OUT * OSR                    # modulator clocks per frame (107,400)
PUF_BITS, ID_BITS = 600, 16
NC_READS = 64                          # power-up reads of the array hashed into the chip nonce N_c (Section IV)
PREAMBLE = np.array([1, 0] * 16, np.uint8)
SYNC = np.array([int(c) for c in format(0xB5E3, "016b")], np.uint8)

DEFAULTS = dict(
    scheme="proposed", sig="sine", amp_uA=0.93, bin=205, dc_uA=0.5, force_uA=10,
    rtia=321.88e3, buffer=0.5, dither=True, dither_amp=0.022, comp_noise=1e-3, xlim=2.5,
    dither_at="comparator", int_noise=0.9e-5,           # dither placement; thermal noise added to x1 per clock (FS rms; 1.8e-5 at the input x g1)
    comp_offset=0.0,                                    # comparator input offset, x2 units (a per-chip constant)
    jitter_ps=0.0, flicker_fc=0.0, hd_a2=0.0, hd_a3=0.0,  # circuit impairments, off by default: feedback-DAC clock jitter
                                                        # (ps rms), 1/f corner of the input-referred noise (Hz), and the 2nd-
                                                        # and 3rd-order coefficients of the first integrator's input stage
    dither_seeds=(0x1D2B, 0x7A31), key16=0xC3A5, chip_id=0x0B77,
    chip_seed=7, temp=37, puf_sigma=0.06, puf_tc=0.008, bch_t=40,
    puf_bias=0.5, puf_years=0.0, puf_age_drift=0.03,    # P(cell = 1); years since enrolment; drift per sqrt(year)
    mac=True, frame_ctr=42, rx_last_ctr=41, mitm_p=0.0,
    chip_nonce=True,                                    # 31-bit chip nonce in the IV, from the raw power-up fingerprint (Section IV)
    nc_reads=NC_READS,                                  # power-up reads hashed into N_c (the array is power-cycled at session start)
    attack="none", ber=0.0, lead=200, rx_wrong_key=False, seed=2026,
)
ATK_DEF = dict(mismatch=0.1, ki_input="dc", flip_p=0.1, fake=0.3, bm_bits=64, clone=1, h_flip=1,
               ch_mismatch=0.1, manip_q=200,
               trim=(0.0, -0.03, 0.03, -0.1, 0.1, -0.2, 0.2),  # A4: replica input offsets the attacker tries (% FS)
               sat_frames=21, sat_uA=1.7,                     # A6: frames recorded, and the forced current (u = 0.45 at g = 0.5)
               reuse_sessions=11)                             # A15: sessions in which the attacker replays an old nonce

def rng_for(seed):
    # every random draw of the model comes from a generator seeded explicitly, so each run is repeatable
    return np.random.default_rng(int(seed) & 0xFFFFFFFF)
````

### B. Keystreams: the original 16-bit LFSR and Grain-128a

The model implements the two generators, the authenticated mode of Grain-128a and the IV of the proposed design as follows.

*1) LFSR of [1]:* A 16-bit Fibonacci register with the maximal-length feedback polynomial x¹⁶ + x¹⁴ + x¹³ + x¹¹ + 1 (taps at bits 0, 2, 3 and 5 of a right-shifting register), restarted from the stored key (an all-zero key is replaced by 1). Every keystream bit is a fixed linear function of the 16 key bits, so the keystream has linear complexity L = 16 and repeats after 2¹⁶ − 1 = 65,535 bits; **its first 16 output bits are the key itself**. The attacks do not depend on which primitive polynomial is used.

*2) Grain-128a [6]:* A 128-bit LFSR and a 128-bit nonlinear feedback shift register (NFSR) are loaded, the NFSR with the key $k$ and the LFSR with the 96-bit initialisation vector (IV) followed by 31 ones and one zero, and clocked 256 times with the pre-output bit fed back (initialisation). Afterwards each clock produces one pre-output bit $y$, a nonlinear function (with AND terms) of both registers, so the linear complexity grows with the length and **Berlekamp-Massey finds no short register**.

*3) Authenticated mode (first IV bit $\mathrm{IV}_0 = 1$):* The first 64 pre-output bits initialise a 32-bit accumulator $A$ ($y_0, \ldots, y_{31}$) and a 32-bit shift register $R$ ($y_{32}, \ldots, y_{63}$). Afterwards the even bits $y_{64}, y_{66}, \ldots$ form the keystream $K$ and the odd bits are shifted into $R$, one per message bit; for each message bit equal to 1, $A$ is XORed with $R$ before the shift. The 32-bit tag is $A$ after a final padding bit 1, that is $A \oplus R$.

*4) IV of the proposed design:* IV = 1 ‖ session nonce (32 bits, sent by the receiver) ‖ frame counter (32 bits) ‖ chip nonce $N_\mathrm{c}$ (31 bits, from the hash of 64 raw power-up reads, `nonce_reads`; Section IV-B). **Every frame of a session thus has its own keystream**, and a replayed session nonce repeats one only on a chance collision of $N_\mathrm{c}$ (about 2⁻³¹ per pair of sessions, Section VIII-F). The MAC covers the counter, the encrypted chip ID and every encrypted modulator bit (encrypt-then-MAC). With the counter and MAC switched off (a setting used in Section IX), the IV is all zero and every power-up reuses the same keystream.

The implementation is checked against the four published test vectors of Grain-128a (Table 3 and Appendix A of [6]), which give 22 values: the first 320 pre-output bits of all four; and, for the two vectors in authenticated mode, the initial accumulator and register, the first 128 keystream bits, the first 128 MAC-stream bits and the tags of five messages (of 0, 1, 1, 20 and 41 bits). Each is compared bit for bit, as a hexadecimal string, with the published value. **The pass criterion is 22 of 22**; any mismatch stops the notebook (`assert`).

> **Next cell.** 1. Defines the original 16-bit LFSR. 2. Defines Grain-128a: initialisation, pre-output, keystream and MAC stream, tag. 3. Defines the proposed design's IV (1 ‖ session nonce ‖ counter ‖ chip nonce $N_\mathrm{c}$) and the receiver's MAC check. 4. Defines `grain_self_test()`, which compares the implementation with the published test vectors. The cell only defines functions; no simulation is run.

````python
#@title Keystreams: LFSR-16 and Grain-128a
# ---------- the original design's keystream: a 16-bit LFSR, x^16 + x^14 + x^13 + x^11 + 1
@njit(cache=False)
def lfsr16_bits(seed, n):
    # Fibonacci form shifting right: the output is bit 0; the feedback bits 0, 2, 3, 5 are the taps x^16, x^14, x^13,
    # x^11 seen from the output end. An all-zero register would stay zero, so seed 0 is replaced by 1.
    s = seed & 0xFFFF
    if s == 0:
        s = 1
    out = np.empty(n, np.uint8)
    for i in range(n):
        fb = (s ^ (s >> 2) ^ (s >> 3) ^ (s >> 5)) & 1
        out[i] = s & 1
        s = (s >> 1) | (fb << 15)
    return out

# ---------- Grain-128a (Agren, Hell, Johansson, Meier 2011), bit-exact to the paper's test vectors
@njit(cache=False)
def grain_preoutput(key, iv, nout):
    """y_0 .. y_(nout-1) after the 256 initialisation clocks. key: 128 bits, iv: 96 bits (bit 0 = IV_0)."""
    T = 256 + nout
    s = np.zeros(128 + T, np.uint8)          # LFSR s_t .. ; the register at time t is s[t:t+128]
    b = np.zeros(128 + T, np.uint8)          # NFSR
    b[:128] = key
    s[:96] = iv
    s[96:127] = 1                            # s_96..s_126 = 1, s_127 = 0
    y_out = np.zeros(nout, np.uint8)
    # the registers are unrolled into arrays: each clock appends one new bit at index t + 128 instead of shifting
    for t in range(T):
        # pre-output y = h(b, s) + s_93 + seven NFSR taps
        b12 = b[t + 12]; b95 = b[t + 95]
        y = ((b12 & s[t + 8]) ^ (s[t + 13] & s[t + 20]) ^ (b95 & s[t + 42]) ^ (s[t + 60] & s[t + 79])
             ^ (b12 & b95 & s[t + 94]) ^ s[t + 93]
             ^ b[t + 2] ^ b[t + 15] ^ b[t + 36] ^ b[t + 45] ^ b[t + 64] ^ b[t + 73] ^ b[t + 89])
        # LFSR feedback f(s), and NFSR feedback s_0 + g(b): the LFSR output added to the nonlinear function g
        st = s[t]
        fs = st ^ s[t + 7] ^ s[t + 38] ^ s[t + 70] ^ s[t + 81] ^ s[t + 96]
        fb = (st ^ b[t] ^ b[t + 26] ^ b[t + 56] ^ b[t + 91] ^ b[t + 96]
              ^ (b[t + 3] & b[t + 67]) ^ (b[t + 11] & b[t + 13]) ^ (b[t + 17] & b[t + 18])
              ^ (b[t + 27] & b[t + 59]) ^ (b[t + 40] & b[t + 48]) ^ (b[t + 61] & b[t + 65])
              ^ (b[t + 68] & b[t + 84]) ^ (b[t + 88] & b[t + 92] & b[t + 93] & b95)
              ^ (b[t + 22] & b[t + 24] & b[t + 25]) ^ (b[t + 70] & b[t + 78] & b[t + 82]))
        if t < 256:                          # initialisation: the pre-output is fed back into both registers
            fs ^= y; fb ^= y
        else:
            y_out[t - 256] = y
        s[t + 128] = fs
        b[t + 128] = fb
    return y_out

def bits_to_int(bits):
    # most significant bit first
    v = 0
    for x in bits:
        v = (v << 1) | int(x)
    return v

def int_to_bits(v, n):
    return np.array([(v >> (n - 1 - i)) & 1 for i in range(n)], np.uint8)

def grain_keystream(key, iv, n_enc):
    """IV_0 = 0: enc = y_0, y_1, ...   IV_0 = 1 (authentication mode): enc = y_64, y_66, ..., the MAC uses
    y_65, y_67, ... and its 32-bit accumulator / shift register start as y_0..y_31 / y_32..y_63."""
    if iv[0] == 0:
        return dict(enc=grain_preoutput(key, iv, n_enc))
    # MAC-stream bits for a message of up to n_enc + 64 bits (the 32-bit frame counter is authenticated as well)
    n_auth = n_enc + 64
    y = grain_preoutput(key, iv, 64 + 2 * n_auth)
    return dict(enc=y[64:64 + 2 * n_enc:2].copy(), auth=y[65:65 + 2 * n_auth:2].copy(),
                A0=bits_to_int(y[:32]), R0=bits_to_int(y[32:64]))

@njit(cache=False)
def _grain_tag(msg, auth, A0, R0):
    # accumulator A and shift register R: a message bit 1 adds R to A; every message bit shifts the next MAC-stream
    # bit into R
    A = A0; R = R0
    for i in range(msg.shape[0]):
        if msg[i]:
            A ^= R
        R = ((R << 1) | auth[i]) & 0xFFFFFFFF
    return (A ^ R) & 0xFFFFFFFF                # the final padding bit m_L = 1

def grain_tag(msg, ks):
    return int(_grain_tag(np.asarray(msg, np.uint8), ks["auth"], np.int64(ks["A0"]), np.int64(ks["R0"])))

def hex_bits(h, n=None):
    # hex digits to bits, most significant bit of each digit first (the notation of the Grain-128a paper)
    h = h.replace(" ", "")
    bits = np.array([(int(c, 16) >> (3 - k)) & 1 for c in h for k in range(4)], np.uint8)
    return bits if n is None else bits[:n]

def bits_hex(bits):
    bits = np.concatenate([bits, np.zeros((-len(bits)) % 4, np.uint8)])
    return "".join(format(bits_to_int(bits[i:i + 4]), "x") for i in range(0, len(bits), 4))

# Table 3 and Appendix A of the Grain-128a paper, verbatim
GRAIN_TV = [
    dict(key="0000000000000000 0000000000000000", iv="0000000000000000 00000000",
         pre="c0207f221660650b 6a952ae26586136f a0904140c8621cfe 8660c0dec0969e94 36f4ace92cf1ebb7"),
    dict(key="0123456789abcdef 123456789abcdef0", iv="0123456789abcdef 12345678",
         pre="f88720c13f46e6a4 3c07eeed89161a4d d73bd6b8be8b6b11 6879714ebb630e0a 4c12f0399412982c"),
    dict(key="0000000000000000 0000000000000000", iv="8000000000000000 00000000",
         pre="564b362219bd90e3 01f259cf52bf5da9 deb1845be6993abd 2d3c77c4acb90e42 2640fbd6e8ae642a",
         acc="564b3622", reg="19bd90e3", ks="0d2b1f2ebc83da7e 6658ee3150f9ef47", mac="1cdbc7f1e52da547 36fa252828de82a0",
         tags=["4ff6a6c1", "653017e4", "7c8d8707", "522ab34f", "4b7821c9"]),
    dict(key="0123456789abcdef 123456789abcdef0", iv="8123456789abcdef 12345678",
         pre="7f2acdb7adfb701f 8d2083b3c32b43f1 962b3dcabf679378 db3536bfc25bed48 3008e6bcb395a156",
         acc="7f2acdb7", reg="adfb701f", ks="a49d971c976bf596 b45f93e242ded8c1", mac="3015919d61787b5c d7678db840a6571e",
         tags=["d2d1bda8", "24dc2d89", "89275d96", "379d2899", "9226b196"]),
]
GRAIN_MSGS = [("", 0), ("0", 1), ("8", 1), ("12340", 20), ("123456789e8", 41)]   # the test messages: (hex, bits)

def grain_self_test():
    # per test vector: the pre-output; in authentication mode (IV_0 = 1) also the MAC's initial accumulator and register,
    # the keystream, the MAC stream and the tags of the five test messages
    checks = []
    for n, tv in enumerate(GRAIN_TV):
        key, iv, pre = hex_bits(tv["key"], 128), hex_bits(tv["iv"], 96), tv["pre"].replace(" ", "")
        checks.append((f"TV{n+1} pre-output", bits_hex(grain_preoutput(key, iv, 4 * len(pre))) == pre))
        if iv[0] == 0:
            continue
        nenc = 4 * len(tv["ks"].replace(" ", ""))
        ks = grain_keystream(key, iv, nenc)
        checks.append((f"TV{n+1} accumulator", format(ks["A0"], "08x") == tv["acc"]))
        checks.append((f"TV{n+1} register", format(ks["R0"], "08x") == tv["reg"]))
        checks.append((f"TV{n+1} keystream", bits_hex(ks["enc"]) == tv["ks"].replace(" ", "")))
        checks.append((f"TV{n+1} macstream", bits_hex(ks["auth"][:nenc]) == tv["mac"].replace(" ", "")))
        for j, (h, nb) in enumerate(GRAIN_MSGS):
            checks.append((f"TV{n+1} tag m{j}", format(grain_tag(hex_bits(h, nb) if nb else np.zeros(0, np.uint8), ks), "08x") == tv["tags"][j]))
    return all(ok for _, ok in checks), checks

# The proposal's IV: IV_0 = 1 (authentication), the receiver's 32-bit session nonce N, the 32-bit frame counter and the
# chip's 31-bit nonce N_c (bits 65..95). N_c makes the keystream of a session unique even when an attacker replays an
# old N on the unauthenticated downlink (attack A15): it is taken from the raw power-up fingerprint w' (before error
# correction), whose noisy cells differ between power-ups, and from a volatile session index. One read holds too little
# noise (about 9 bits of collision entropy at tt, 25 °C; Section VIII-F), so the array is power-cycled NC_READS times at
# session start and all the raw reads are hashed.
# With the counter + MAC off the IV is all zero: no MAC and the same keystream every power-up.
def iv_bits(nonce, ctr, cn=0):
    iv = np.zeros(96, np.uint8)
    iv[0] = 1
    iv[1:33] = int_to_bits(nonce, 32)
    iv[33:65] = int_to_bits(ctr, 32)
    iv[65:96] = int_to_bits(cn, 31)
    return iv

def chip_nonce(w_raw, session=0):
    """N_c: 31 bits of SHA-256(w' || session index); w' is the raw power-up read (one read, or the stack of reads of
    nonce_reads, rows concatenated), before the BCH correction."""
    # the first 32 bits of the digest, shifted right to 31 bits
    d = hashlib.sha256(b"Nc" + np.packbits(np.asarray(w_raw, np.uint8).ravel()).tobytes() + bytes([session & 0xFF])).digest()
    return int.from_bytes(d[:4], "big") >> 1

def nonce_reads(puf, T, rng, age=0.0, first=None, n=NC_READS):
    """The raw reads hashed into N_c: n power-ups of the array at session start, each with fresh noise; `first` is the
    read the key is reconstructed from (it is reused, not read again)."""
    rest = [puf.read(T, rng, age) for _ in range(n - (first is not None))]
    return np.stack(([np.asarray(first, np.uint8)] if first is not None else []) + rest)

def keystream_nl(key, iv, n_enc):
    # iv None: the all-zero IV (no counter, no MAC: the same keystream at every power-up)
    return grain_keystream(key, np.zeros(96, np.uint8) if iv is None else iv, n_enc)

def mac_msg(ctr_bits, enc_id, C):
    # the tag covers the frame counter, the encrypted chip ID and the ciphertext
    return np.concatenate([ctr_bits, enc_id, C]).astype(np.uint8)

def mac_verify(key, nonce, ctr, enc_id, C, tag, cn=0):
    # the receiver regenerates the keystream of this frame's IV with its own key and recomputes the tag
    ks = keystream_nl(key, iv_bits(nonce, ctr, cn), ID_BITS + len(C))
    t = grain_tag(mac_msg(int_to_bits(ctr, 32), enc_id, C), ks)
    return t == tag, t, ks
````

### C. Front end, second-order 1-bit delta-sigma modulator and receiver filters

The chain is modelled clock by clock; at modulator clock $k$:

*1) Front end:* The sensor current $i[k]$ passes the transimpedance amplifier (TIA), whose output is $v_\mathrm{TIA} = \mathrm{clip}(V_\mathrm{CM} + i\,R_\mathrm{TIA},\ 0,\ V_\mathrm{DD})$, which saturates at the rails, and the single-ended-to-differential (S2D) buffer, $u = 0.5\,(v_\mathrm{TIA} - V_\mathrm{CM})/(V_\mathrm{DD}/2)$, so that $\lvert u\rvert \le 0.5$ **even when the TIA saturates**. An input-referred difference between two channels or chips (analog mismatch) is a constant δ added to the input $u$ of one of them.

*2) Loop:* **A discrete-time second-order loop of the same order as the continuous-time loop of [1]**. With the fed-back value $v[k] = 2y[k-1] - 1 = \pm 1$ ($y[k-1]$: the comparator's decision in the previous clock, item 4),
   $$x_1[k] = p\,x_1[k-1] + g_1\big(u[k] - v[k]\big) + \sigma_i\,n_i[k], \qquad x_2[k] = p\,x_2[k-1] + g_2\big(x_1[k] - v[k]\big),$$
   where $x_1$, $x_2$ are the integrator states (both start at 0, and $y[-1] = 1$); $g_1 = g_2 = 1/2$ the integrator gains (capacitor ratios), each multiplied by $(1 + e)$ for a per-chip relative error $e$; $p = A/(1 + A)$ the leak of an integrator whose amplifier has DC gain $A$ (1 by default, an ideal integrator); and $n_i$ unit Gaussian noise. When either state exceeds $\pm x_\mathrm{lim}$, both are clipped, and the fraction of clipped clocks is recorded.

*3) Dither:* Two 16-bit LFSRs of the key LFSR's polynomial, seeded per channel, give the bits $b_1$, $b_2$ and the dither $d = a_\mathrm{d}(b_1 - b_2) \in \{-0.022, 0, +0.022\}$. As described in [1] ("a slight chance for a bit flip at each sampling output without disrupting the underlying signal"), **it acts at the comparator**, whose input becomes $q = x_2(1 + d) + d$; the alternative `dither_at='feedback'` scales the fed-back value instead, $v = \pm(1 + d)$ with the previous clock's dither, and Section VI compares the two placements.

*4) Comparator:* $y[k] = 1$ if $q + V_\mathrm{os} + \sigma_\mathrm{c}\,n_\mathrm{c}[k] \ge 0$, else 0, with the offset $V_\mathrm{os}$ and input-referred noise $\sigma_\mathrm{c}$ in $x_2$ units and unit Gaussian noise $n_\mathrm{c}$; the output bit is $D[k] = y[k]$.

*5) Receiver filters and SNDR:* The receiver maps the bits to ±1 and decimates them by OSR = 50 with a third-order cascaded integrator-comb (CIC) filter (three running sums, one sample kept in 50, three first differences, division by 50³), **followed by a 31-tap FIR filter that compensates the CIC's droop** (frequency sampling of the inverse CIC response at 513 points, flat up to $0.4\,f_\mathrm{s}/\mathrm{OSR}$ = 80 kHz and zero above, Hamming window, unit DC gain). The SNDR is measured on the last 2,048 readings: mean removed, four-term Blackman-Harris window, power within ±4 bins of the tone over the power of all other bins from bin 4 to bin 1,023 (bins 0 to 3, including DC, excluded).

**Two channels differ only through the modelled effects**: noise and dither seeds, input mismatch δ, comparator offset and capacitor errors. *Limits:* excess loop delay is not modelled, and clock jitter, flicker noise and input-stage distortion are off by default (`jitter_ps`, `flicker_fc`, `hd_a2`, `hd_a3`; Section VIII switches them on once, for the spectra of the end-to-end frame).

> **Next cell.** 1. Defines the TIA and the buffer. 2. Defines the modulator `dsm2` (the loop of the equations above, compiled with numba, with seeded noise streams). 3. Defines the receiver's CIC and compensation filters (`decode`) and the SNDR measurement. 4. Defines the test signals (`sensor_current`) and `longest_run`, which the saturation attack uses. The cell only defines functions; no simulation is run.

````python
#@title Front end, modulator and receiver filters
# ---------- analog front end and the 2nd-order 1-bit delta-sigma modulator
def tia(i, rtia):                       # transimpedance stage, clipped at the rails
    return np.clip(VCM + i * rtia, 0.0, VDD)

def s2d(v, gain):                       # single-ended-to-differential buffer, normalised to the modulator full scale
    return gain * (v - VCM) / (VDD / 2)

@njit(cache=False)
def _dsm2(u, nc, ni, dither, amp, seed1, seed2, xlim, comp_noise, feedback, int_noise, offset, p1, p2, g1, g2, jit):
    n = u.shape[0]
    D = np.empty(n, np.uint8); X1 = np.empty(n, np.float32); X2 = np.empty(n, np.float32); DI = np.empty(n, np.float32)
    s1 = seed1 & 0xFFFF; s2 = seed2 & 0xFFFF
    if s1 == 0: s1 = 1
    if s2 == 0: s2 = 1
    x1 = 0.0; x2 = 0.0; y = 1; clip = 0; dprev = 0.0; vprev = 1.0
    for k in range(n):
        # feedback DAC level: +-1 from the previous bit, scaled by (1 + dither) when the dither sits in the feedback path
        v = (1.0 if y else -1.0) * ((1.0 + dprev) if feedback else 1.0)
        x1 = p1 * x1 + (g1 * (u[k] - v) + int_noise * ni[k])   # p < 1: leaky integrator (finite OTA gain); g: capacitor ratio
        x1 += g1 * (v - vprev) * jit[k]  # NRZ feedback DAC: an edge that moves by dt changes the charge by (step) x dt / Ts
        vprev = v
        x2 = p2 * x2 + g2 * (x1 - v)
        # the integrator outputs saturate at +-xlim (OTA swing); clip counts the clocks that saturate
        if abs(x1) > xlim or abs(x2) > xlim:
            clip += 1
            x1 = min(max(x1, -xlim), xlim); x2 = min(max(x2, -xlim), xlim)
        d = 0.0
        if dither:                       # two 16-bit LFSRs drive the dither ladder (paper: ratio 0.978..1.022)
            b1 = s1 & 1; f1 = (s1 ^ (s1 >> 2) ^ (s1 >> 3) ^ (s1 >> 5)) & 1; s1 = (s1 >> 1) | (f1 << 15)
            b2 = s2 & 1; f2 = (s2 ^ (s2 >> 2) ^ (s2 >> 3) ^ (s2 >> 5)) & 1; s2 = (s2 >> 1) | (f2 << 15)
            d = amp * (b1 - b2)
        # comparator: with the dither at its input the decision variable is x2 (1 + d) + d; the bit is the sign of that
        # plus the comparator's offset and noise
        q = x2 if feedback else x2 * (1.0 + d) + d
        y = 1 if q + offset + comp_noise * nc[k] >= 0.0 else 0
        dprev = d
        D[k] = y; X1[k] = x1; X2[k] = x2; DI[k] = d
    return D, X1, X2, DI, clip / n

def dsm2(u, P, seeds=None, noise_seed=0, offset=None, coef_err=None):
    """Modulator output bits D for input u. P: settings dict. The comparator and integrator noise are
    drawn from their own seeded streams so a run is repeatable. offset: the comparator's input offset in
    x2 units (a per-chip constant; default P['comp_offset']). Integrator non-idealities: P['ota_gain'], the OTA's
    DC gain (V/V; inf = ideal), makes each integrator leak by 1/A per clock (pole at A/(1+A)); coef_err = (e1, e2),
    the relative error of each integrator's capacitor ratio (a per-chip constant; default P['coef_err'])."""
    A = float(P.get("ota_gain", np.inf)); e1, e2 = P.get("coef_err", (0.0, 0.0)) if coef_err is None else coef_err
    p = 1.0 if not np.isfinite(A) else A / (1.0 + A)
    # one standard normal draw per clock for the comparator noise (nc) and for the first integrator's noise (ni)
    r = rng_for(noise_seed)
    nc = r.standard_normal(len(u)); ni = r.standard_normal(len(u))
    u = np.asarray(u, np.float64); jit = np.zeros(len(u))
    if P.get("jitter_ps", 0) or P.get("flicker_fc", 0) or P.get("hd_a2", 0) or P.get("hd_a3", 0):
        ri = rng_for(noise_seed + 0x5EED)                # own stream: the default noise draws above are unchanged
        u = u + P.get("hd_a2", 0.0) * u ** 2 + P.get("hd_a3", 0.0) * u ** 3   # input-stage distortion: HD2, HD3
        jit = P.get("jitter_ps", 0.0) * 1e-12 * FS * ri.standard_normal(len(u))   # edge jitter as a fraction of the clock period
        if P.get("flicker_fc", 0):                       # 1/f noise, input-referred, equal to the white thermal density at fc
            f = np.fft.rfftfreq(len(u), 1 / FS); sh = np.zeros_like(f); sh[1:] = np.sqrt(P["flicker_fc"] / f[1:])
            u = u + np.fft.irfft(np.fft.rfft(P["int_noise"] / 0.5 * ri.standard_normal(len(u))) * sh, len(u))
    # the loop: leak p per clock in both integrators, nominal coefficients 0.5 scaled by the capacitor-ratio errors
    s = seeds if seeds is not None else P["dither_seeds"]
    D, X1, X2, DI, clip = _dsm2(np.asarray(u, np.float64), nc, ni, bool(P["dither"]), float(P["dither_amp"]), int(s[0]), int(s[1]),
                                float(P["xlim"]), float(P["comp_noise"]), P["dither_at"] == "feedback", float(P["int_noise"]),
                                float(P.get("comp_offset", 0.0) if offset is None else offset), p, p,
                                0.5 * (1.0 + float(e1)), 0.5 * (1.0 + float(e2)), jit)
    return dict(D=D, X1=X1, X2=X2, DI=DI, clip=clip)

# ---------- receiver decimation: 3rd-order CIC + a 31-tap compensation FIR (flat to 0.4 fs_out)
def cic(D, R=OSR, order=3):
    # bits as +-1; `order` integrators (running sums) at the clock rate, decimation by R, `order` combs (differences) at
    # the output rate; the DC gain R^order is divided out
    x = 2.0 * np.asarray(D, np.float64) - 1.0
    for _ in range(order):
        x = np.cumsum(x)
    y = x[R - 1::R][: len(D) // R].copy()
    for _ in range(order):
        y = np.concatenate([y[:1], np.diff(y)])
    return y / R ** order

def _comp_taps(R=OSR, N=31, M=15, K=512):
    # frequency-sampling design: the inverse of the CIC's droop up to 0.4 fs_out (zero above it), sampled at K + 1
    # frequencies, transformed to N taps centred on M, Hamming-windowed and normalised to unity DC gain
    f = 0.5 * np.arange(K + 1) / K
    with np.errstate(divide="ignore", invalid="ignore"):
        G = np.where(f > 0.4, 0.0, 1.0 / np.abs(np.sin(np.pi * f) / (R * np.sin(np.pi * f / R))) ** 3)
    G[0] = 1.0
    w = np.ones(K + 1); w[0] = w[-1] = 0.5
    n = np.arange(N)
    h = np.array([2 * np.sum(w * G * np.cos(2 * np.pi * f * (k - M))) * (0.5 / K) for k in n])
    h *= 0.54 - 0.46 * np.cos(2 * np.pi * n / (N - 1))
    return h / h.sum()
COMP_TAPS = _comp_taps()

def decode(D):                           # what the receiver's filters make of a bit stream
    return np.convolve(cic(D), COMP_TAPS)[: len(D) // OSR]

def sndr(y, bin_):
    # SNDR of the last N_FFT output samples: 4-term Blackman-Harris window; signal = the bins within 4 of the tone,
    # noise and distortion = all other bins from 4 to N/2 (DC and its window leakage excluded)
    n = N_FFT
    seg = np.asarray(y[-n:], np.float64)
    i = np.arange(n); x = 2 * np.pi * i / (n - 1)
    w = 0.35875 - 0.48829 * np.cos(x) + 0.14128 * np.cos(2 * x) - 0.01168 * np.cos(3 * x)
    p = np.abs(np.fft.fft((seg - seg.mean()) * w)) ** 2
    k = np.arange(4, n // 2)
    sig = p[k[np.abs(k - bin_) <= 4]].sum(); tot = p[k].sum()
    return 10 * np.log10(sig / max(tot - sig, 1e-30))

def sensor_current(P, n):
    # test inputs: a sine on FFT bin `bin` (a whole number of periods in N_FFT output samples), a DC current, a large
    # forced current, or zero
    t = np.arange(n) / FS
    f = P["bin"] * FS_OUT / N_FFT
    return {"sine": P["amp_uA"] * 1e-6 * np.sin(2 * np.pi * f * t), "dc": np.full(n, P["dc_uA"] * 1e-6),
            "forced": np.full(n, P["force_uA"] * 1e-6), "zero": np.zeros(n)}[P["sig"]]

def longest_run(D):
    # the positions where the bit changes, and the longest stretch between two of them
    D = np.asarray(D, np.int8)
    edges = np.flatnonzero(np.diff(D))
    runs = np.diff(np.concatenate([[-1], edges, [len(D) - 1]]))
    return int(runs.max())
````

### D. SRAM PUF, BCH fuzzy extractor and key derivation

The key path is modelled in five steps:

*1) Cell model:* Each of the $n = 600$ cells has a fixed mismatch $m_i \sim \mathcal{N}(\mu, 1)$, drawn once per chip; $\mu = \Phi^{-1}(b)$ sets the bias $b$, the probability that a cell favours 1 ($\Phi$: the standard normal distribution function). At every power-up the cell reads 1 when $m_i + s(T)\,\nu_i > 0$, with fresh unit Gaussian noise $\nu_i$ and the noise-to-mismatch ratio $s(T) = s_{25}\,(1 + a\,\lvert T - 25\,{}^\circ\mathrm{C}\rvert)$ at temperature $T$ (its value $s_{25}$ at 25 °C and temperature coefficient $a$), so it reads differently from its enrolled bit $w_i$ with probability $p_i = \Phi(-m_i/s(T))$ if $w_i = 1$ and $\Phi(m_i/s(T))$ if $w_i = 0$. Ageing adds $0.03\sqrt{\mathrm{years}}\,r_i$ to $m_i$, with a fixed unit Gaussian drift $r_i$ per cell.

*2) Enrolment (once, at test, 25 °C):* The helper data $h$ is the syndrome of the fingerprint $w$ (the syndrome construction of [4]): the $t$ odd syndromes $S_1, S_3, \ldots, S_{2t-1}$, $S_j = \sum_i w_i\,\alpha^{ij}$ with $\alpha$ a primitive element of the field, of a binary BCH code over $\mathrm{GF}(2^{10})$ (primitive polynomial $x^{10} + x^3 + 1$), shortened from 1,023 to 600 bits, so $h$ has $10t$ bits. **The key is the first 128 bits of SHA-256**($w$).

*3) Reconstruction (every power-up):* Because the syndrome is linear, the noisy reading $w' = w \oplus e$, with the error pattern $e$ (one bit $e_i$ per cell, 1 where the cell flipped), gives $\mathrm{syndrome}(w') \oplus h = \mathrm{syndrome}(e)$. The decoder derives the even syndromes ($S_{2j} = S_j^2$), finds the error-locator polynomial with Berlekamp-Massey over $\mathrm{GF}(2^{10})$ and its roots with a Chien search over the 600 positions, and declares failure if the locator has degree L > t or fewer than L roots there. When $e$ has at most $t$ ones **it is found exactly**, and $w = w' \oplus e$ gives the same key.

*4) Key failure:* **The key fails when more than $t$ cells flip**. With independent $p_i$ the number of flips is Poisson-binomial; `fail_tail` computes $P(\sum_i e_i > t)$ exactly by a recursion over the 600 cells, accurate down to about $10^{-300}$, and `err_dist` returns the whole distribution, which the later sections use.

*5) Residual secrecy:* The public helper data reveals at most $10t$ bits of $w$. With the min-entropy $H_\infty = -\log_2 \max(b, 1 - b)$ per cell, at least $\lfloor 600\,H_\infty - 10t \rfloor$ bits remain secret (`puf_residual`) at the point estimate of the bias; the verdict of Section IX uses the 95 % lower bound of the min-entropy and **requires 192 bits** (the 128-bit key plus the hash margin of Section VII-E for a key within statistical distance ε = 2⁻³² of uniform).

The assumed cells of Subsection A (unbiased, $t = 40$) **leave 200 secret bits**; Section VII replaces $s_{25}$, $a$ and $b$ by fits to SKY130 Monte Carlo and sizes t.

> **Next cell.** 1. Builds GF(2¹⁰) and the BCH syndrome and decoder (Berlekamp-Massey and Chien search over GF(2¹⁰)). 2. Defines the hash and the `SRAMPUF` cell model. 3. Defines enrolment and key reconstruction. 4. Defines the exact key-failure probability (`fail_tail`) and the residual secrecy (`puf_residual`). The cell only defines functions; no simulation is run.

````python
#@title SRAM PUF, BCH fuzzy extractor, key
# ---------- SRAM PUF + BCH fuzzy extractor (syndrome construction over GF(2^10)) + hash
# GF(2^10) by its exponential and logarithm tables; EXP is stored twice over so that EXP[LOG[a] + LOG[b]] needs no modulo
GF_N = 1023
EXP = np.zeros(2 * GF_N, np.int64); LOG = np.zeros(GF_N + 1, np.int64)
_x = 1
for _i in range(GF_N):
    EXP[_i] = _x; LOG[_x] = _i
    _x <<= 1
    if _x & 1024:
        _x ^= 0x409                       # primitive polynomial x^10 + x^3 + 1
EXP[GF_N:] = EXP[:GF_N]

@njit(cache=False)
def _gmul(a, b, EXP, LOG):
    if a == 0 or b == 0:
        return 0
    return EXP[LOG[a] + LOG[b]]

@njit(cache=False)
def bch_syndrome_nb(bits, t, EXP):
    # the t odd syndromes S_(2j+1) = r(alpha^(2j+1)): the sum of alpha^(i(2j+1)) over the positions i of the ones (the
    # even syndromes follow as squares, so they need not be stored or published)
    S = np.zeros(t, np.int64)
    for i in range(bits.shape[0]):
        if bits[i]:
            for j in range(t):
                S[j] ^= EXP[(i * (2 * j + 1)) % 1023]
    return S

@njit(cache=False)
def bch_decode_nb(syn, t, n, EXP, LOG):
    """syn: odd syndromes of (noisy read) xor helper data. Returns (ok, error pattern, L)."""
    # all 2t syndromes from the odd ones: S_2i = S_i^2 in characteristic 2
    S = np.zeros(2 * t + 1, np.int64)
    for i in range(1, 2 * t + 1):
        if i % 2:
            S[i] = syn[(i - 1) // 2]
        else:
            S[i] = _gmul(S[i // 2], S[i // 2], EXP, LOG)
    e = np.zeros(n, np.uint8)
    nz = False
    for i in range(1, 2 * t + 1):
        if S[i]:
            nz = True
    if not nz:                            # all syndromes zero: no error
        return True, e, 0
    C = np.zeros(2 * t + 2, np.int64); B = np.zeros(2 * t + 2, np.int64); T = np.zeros(2 * t + 2, np.int64)
    C[0] = 1; B[0] = 1
    L = 0; m = 1; bb = 1
    # Berlekamp-Massey over GF(2^10): the shortest LFSR (error-locator polynomial C, length L) that generates the
    # syndromes; d is the discrepancy between the next syndrome and the LFSR's prediction, B the last C before a length
    # change, bb its discrepancy and m the clocks since then
    for k in range(2 * t):
        d = S[k + 1]
        for i in range(1, L + 1):
            d ^= _gmul(C[i], S[k + 1 - i], EXP, LOG)
        if d == 0:
            m += 1
            continue
        coef = EXP[(LOG[d] - LOG[bb] + 1023) % 1023]
        T[:] = C
        for i in range(2 * t + 2 - m):
            C[i + m] ^= _gmul(coef, B[i], EXP, LOG)
        if 2 * L <= k:
            L = k + 1 - L; B[:] = T; bb = d; m = 1
        else:
            m += 1
    if L > t:                             # more than t errors: beyond the code's correction capability
        return False, e, L
    roots = 0
    # Chien search: position pos is in error where C(alpha^-pos) = 0; decoding succeeds only if C has L roots among
    # the n code positions (otherwise the error pattern is not a valid correction)
    for pos in range(n):
        v = 0
        for k in range(L + 1):
            if C[k]:
                v ^= _gmul(C[k], EXP[(1023 - (pos * k) % 1023) % 1023], EXP, LOG)
        if v == 0:
            e[pos] = 1; roots += 1
    if roots == L:
        return True, e, L
    return False, np.zeros(n, np.uint8), L

def bch_syndrome(bits, t):
    return bch_syndrome_nb(np.asarray(bits, np.uint8), t, EXP)

def syn_to_bits(S):
    # helper data as bits: each 10-bit syndrome, most significant bit first
    return np.array([(int(v) >> (9 - b)) & 1 for v in S for b in range(10)], np.uint8)

def bits_to_syn(bits, t):
    return np.array([bits_to_int(bits[j * 10:(j + 1) * 10]) for j in range(t)], np.int64)

def hash_key(w):                          # key = first 128 bits of SHA-256(w); a lightweight hash in hardware
    d = hashlib.sha256(np.packbits(np.asarray(w, np.uint8)).tobytes()).digest()[:16]
    return np.unpackbits(np.frombuffer(d, np.uint8))

class SRAMPUF:
    """Cell i powers up to 1 iff m_i + s(T)·n > 0, m_i ~ N(mu, 1) fixed at fabrication, n ~ N(0, 1) per
    power-up, s(T) = s25·(1 + tc·|T − 25|). Bias sets mu so that P(1) = bias; ageing moves m_i along a fixed
    per-cell drift direction by drift·sqrt(years). The SKY130 Monte Carlo section fits s25, tc and mu."""
    def __init__(self, seed, n=PUF_BITS, sigma25=0.06, bias=0.5, tc=0.008):
        # separate streams for the cells' mismatch and their ageing directions; mu sets P(1) = Phi(mu) = bias
        r, rd = rng_for(seed), rng_for(seed * 31 + 7)
        mu = 0.0 if bias == 0.5 else float(ndtri(bias))
        self.mismatch = r.standard_normal(n) + mu
        self.drift = rd.standard_normal(n)
        self.sigma25, self.tc = sigma25, tc
    def sigma(self, T):
        return self.sigma25 * (1 + self.tc * abs(T - 25))
    def m(self, age=0.0):
        return self.mismatch + age * self.drift
    def read(self, T, rng, age=0.0):
        return (self.m(age) + self.sigma(T) * rng.standard_normal(len(self.mismatch)) > 0).astype(np.uint8)
    def p_err(self, w, T, age=0.0):       # chance each cell reads differently from its enrolled bit w
        q = ndtr(self.m(age) / self.sigma(T))
        return np.where(np.asarray(w) == 1, ndtr(-self.m(age) / self.sigma(T)), q)

@njit(cache=False)
def fail_tail(p, t):
    """Exact P(more than t of the cells flip): the Poisson-binomial tail, accurate down to ~1e-300."""
    # d[k] = P(exactly k flips among the cells so far), kept for k <= t only; tail collects the probability of moving
    # from t to t + 1 flips (once past t, the count stays past t)
    d = np.zeros(t + 1); d[0] = 1.0; tail = 0.0
    for i in range(p.shape[0]):
        q = p[i]; tail += d[t] * q
        for k in range(t, 0, -1):
            d[k] = d[k] * (1 - q) + d[k - 1] * q
        d[0] *= 1 - q
    return tail

@njit(cache=False)
def err_dist(p):
    # the whole distribution of the number of flips (Poisson binomial), by the same recursion
    d = np.zeros(p.shape[0] + 1); d[0] = 1.0
    for i in range(p.shape[0]):
        q = p[i]
        for k in range(i + 1, 0, -1):
            d[k] = d[k] * (1 - q) + d[k - 1] * q
        d[0] *= 1 - q
    return d

def min_ent_bit(b):
    # min-entropy of one bit with P(1) = b
    return -math.log2(max(b, 1 - b))

def puf_residual(P):                      # secret bits left after publishing 10t helper bits
    return int(math.floor(PUF_BITS * min_ent_bit(P["puf_bias"]) - 10 * P["bch_t"]))

def puf_age(P):
    return P["puf_age_drift"] * math.sqrt(max(0.0, P["puf_years"]))

def enroll(puf, t, rng):
    # one read w at 25 C, its helper data h (the BCH syndrome, public) and the key hash(w)
    w = puf.read(25, rng)
    return dict(w=w, h=bch_syndrome(w, t), key=hash_key(w))

def reconstruct(w_noisy, h, t):
    # syndrome(w') xor h = syndrome(w' xor w), the syndrome of the error pattern alone (the code is linear): decode it,
    # correct the read and hash
    sn = bch_syndrome(w_noisy, t)
    ok, e, L = bch_decode_nb(sn ^ h, t, len(w_noisy), EXP, LOG)
    w = w_noisy ^ e
    return dict(ok=ok, e=e, w=w, key=hash_key(w), syn=sn)
````

### E. Attack tools

The attacker has four tools:

*1) Exhaustive key search:* The 16-bit key of [1] has 65,535 non-zero values. The chip ID is a known plaintext encrypted by the first 16 keystream bits, so `brute_lfsr16` returns the first seed whose first 16 output bits match. **Against a 128-bit key the search is infeasible**.

*2) Berlekamp-Massey [3]:* The Berlekamp-Massey (BM) algorithm finds the shortest LFSR that generates a bit sequence; its length is the linear complexity L(n) of the first n bits (`bm_profile`). Any 2L consecutive correct bits of an L-bit LFSR determine the register, so **32 correct bits recover the 16-bit LFSR**, which then predicts every later bit (`lfsr_predict`) and, run backwards, every earlier one. For a random or nonlinear sequence L(n) grows as about n/2 and predictions are right only by chance; `bm_attack` returns the fraction of predicted bits that are correct.

*3) Windowed replica attack (`window_attack`):* The attacker forms the keystream estimate $\hat{K} = C \oplus \hat{D}$ from the transmitted (encrypted) bits $C$ and the bits $\hat{D}$ of a replica chip of the same design at the same input; $\hat{K}$ is correct wherever $\hat{D} = D$. BM succeeds with certainty on a window without error, so the attack draws 80 windows of 64 bits at random start positions, from clock 200 to clock N − 777 (N: the run length, capped at 30,000), and a window yields the key when its register predicts the next 512 true keystream bits with more than 99 % accuracy (a test with the true keystream, which can only favour the attacker). The attack succeeds when one window does; the earliest such window is extended over the frame. If the replica disagrees with the victim on a fraction ε of the bits, independently, a 64-bit window is clean with probability $(1 - \epsilon)^{64}$, and $80\,(1 - \epsilon)^{64}$ windows of 80 yield the key on average: about 29 at ε = 1/64, 3 at ε = 5 % and 0.1 at ε = 10 %. The analysis above therefore predicts that **the attack succeeds in about 95 % of frames at ε = 5 %** and in fewer than one in ten only beyond ε ≈ 10 %; bursty disagreements leave more clean windows, so this is a lower bound on the attacker's success.

*4) Helper-data manipulation oracle (`manip_oracle`):* The attacker adds to $h$ the syndrome of an error pattern of its choice and observes whether the chip still reconstructs its key; each query is a fresh power-up, with error weights cycling from 1 to t + 10. Because the syndrome is linear, each answer depends only on the chosen and the power-up error patterns, **not on $w$**: a *twin* chip with a random fingerprint, its own helper data and the same power-up errors gives the same answers by construction, which the 200 queries confirm numerically. (The answers reveal reliability information; in the cell model, which cells are unstable does not depend on their enrolled values.)

> **Next cell.** 1. Defines Berlekamp-Massey over GF(2), LFSR prediction and the linear-complexity profile. 2. Defines the 16-bit brute-force search. 3. Defines `window_attack`, the windowed replica attack. 4. Defines `manip_oracle`, the helper-data manipulation oracle with its twin chip. The cell only defines functions; no simulation is run.

````python
#@title Attack tools
# ---------- Berlekamp-Massey over GF(2): the attack that breaks any short LFSR
@njit(cache=False)
def bm2(s):
    # the shortest LFSR (connection polynomial c, length L) that generates s; d is the discrepancy between the next bit
    # and the current LFSR's prediction, b the polynomial before the last length change, which happened at bit m
    n = s.shape[0]
    c = np.zeros(n + 1, np.uint8); b = np.zeros(n + 1, np.uint8); tmp = np.zeros(n + 1, np.uint8)
    c[0] = 1; b[0] = 1
    L = 0; m = -1
    for i in range(n):
        d = s[i]
        for j in range(1, L + 1):
            d ^= c[j] & s[i - j]
        if d:
            tmp[:] = c
            sh = i - m
            for j in range(0, n + 1 - sh):
                c[j + sh] ^= b[j]
            if 2 * L <= i:
                L = i + 1 - L; m = i; b[:] = tmp
    return L, c[:L + 1].copy()

@njit(cache=False)
def lfsr_predict(c, L, known, n_out):
    # the last L known bits are the register; each new bit is the sum of c_j times the bit j places back
    buf = np.zeros(L + n_out, np.uint8)
    k0 = max(0, known.shape[0] - L)
    for i in range(L):
        buf[i] = known[k0 + i] if k0 + i < known.shape[0] else 0
    for k in range(n_out):
        nb = 0
        for j in range(1, L + 1):
            nb ^= c[j] & buf[L + k - j]
        buf[L + k] = nb
    return buf[L:].copy()

def bm_attack(known, future):
    L, c = bm2(np.asarray(known, np.uint8))
    g = lfsr_predict(c, L, np.asarray(known, np.uint8), len(future))
    return dict(acc=float(np.mean(g == future)), L=int(L), guess=g)

def bm_profile(s):                        # linear complexity after each bit
    return np.array([bm2(np.asarray(s[:i + 1], np.uint8))[0] for i in range(len(s))])

@njit(cache=False)
def brute_lfsr16(ks16):
    # every non-zero seed of the public LFSR-16, until one reproduces the known keystream bits (-1 if none does)
    for k in range(1, 0x10000):
        s = k; ok = True
        for i in range(ks16.shape[0]):
            if (s & 1) != ks16[i]:
                ok = False
                break
            fb = (s ^ (s >> 2) ^ (s >> 3) ^ (s >> 5)) & 1
            s = (s >> 1) | (fb << 15)
        if ok:
            return k
    return -1

def window_attack(kguess, ktrue, seed, tries=80):
    """BM on many 64-bit windows of a keystream guess. A window 'breaks' when its register predicts the next
    512 real keystream bits; the earliest such window is the one the attacker extends."""
    r = rng_for(seed); n = min(len(kguess), len(ktrue))
    broken, first = 0, -1
    # random window starts in the first 30,000 clocks, after the first 200 and leaving room for the 512 checked bits
    for _ in range(tries):
        s = 200 + int(r.integers(min(n, 30000) - 64 - 512 - 400))
        a = bm_attack(kguess[s:s + 64], ktrue[s + 64:s + 64 + 512])
        if a["acc"] > 0.99:
            broken += 1
            first = s if first < 0 or s < first else first
    return dict(broken=broken, tries=tries, first=first)

@njit(cache=False)
def _lfsr16_candidates(kg, M, thr):
    """Every position s of a keystream guess where the 16 bits kg[s:s+16], taken as the state of the public LFSR-16,
    predict the next M bits of the guess with agreement >= thr (a clean stretch of the guess). Returns positions."""
    n = kg.shape[0]; out = np.empty(n, np.int64); m = 0
    # the 16 bits at s0 as the register (bit j of the state = bit s0 + j of the guess), clocked M times against the guess
    for s0 in range(0, n - 16 - M):
        st = 0
        for j in range(16):
            st |= np.int64(kg[s0 + j]) << j
        agree = 0
        x = st
        for j in range(M):
            if (x & 1) == kg[s0 + j]:
                agree += 1
            fb = (x ^ (x >> 2) ^ (x >> 3) ^ (x >> 5)) & 1
            x = (x >> 1) | (fb << 15)
        if agree >= thr * M:
            out[m] = s0; m += 1
    return out[:m]

def _lfsr16_seed_at(state, steps):
    """Runs the LFSR-16 backwards: the seed from which `state` is reached after `steps` clocks."""
    # one step back: the bit that left the register is recovered from the feedback relation (the new bit 15 is the
    # XOR of old bits 0, 2, 3 and 5), and the register shifts left
    x = int(state)
    for _ in range(int(steps)):
        b0 = ((x >> 15) ^ (x >> 1) ^ (x >> 2) ^ (x >> 4)) & 1
        x = ((x << 1) & 0xFFFF) | b0
    return x

def lfsr16_scan(kguess, enc_id, id_bits, M=1024, thr=0.62):
    """The competent attacker on the original design: the LFSR-16 polynomial is public, so any 16 correct keystream bits
    are its state. Every position of the keystream guess is tried (no random sampling); a candidate state must predict
    the following M bits of the guess better than chance, and its seed, run back to the start of the frame, must decrypt
    the encrypted chip ID to the public ID. Returns the confirmed seed or -1, and the number of candidates tested."""
    kg = np.ascontiguousarray(kguess, np.uint8)
    cand = _lfsr16_candidates(kg, M, thr)
    # at most 400 candidates; each is run back to the frame start (the 16 ID bits precede the payload) and confirmed by
    # the encrypted chip ID
    tried = set()
    for s0 in cand[:400]:
        st = 0
        for j in range(16):
            st |= int(kg[s0 + j]) << j
        seed = _lfsr16_seed_at(st, int(s0) + ID_BITS)
        if seed in tried:
            continue
        tried.add(seed)
        if np.array_equal(lfsr16_bits(seed, ID_BITS) ^ enc_id, id_bits):
            return seed, len(cand)
    return -1, len(cand)

def manip_oracle(puf, en, P, Q, seed):
    """Helper-data manipulation with an accept/reject oracle: the attacker adds the syndrome of an error pattern of its
    choice to h. A twin chip with a different fingerprint but the same noise answers the same queries;
    identical answers carry no information about the fingerprint w."""
    t, r, age = P["bch_t"], rng_for(seed), puf_age(P)
    # the twin: a random fingerprint w2 with its own helper data and key
    w2 = r.integers(0, 2, PUF_BITS).astype(np.uint8); h2 = bch_syndrome(w2, t); key2 = hash_key(w2)
    same = acc = fails = 0; by_w = {}; last = None
    for q in range(Q):
        # query q: an error pattern of weight 1 to t + 10 (cycling), added to the helper data as its syndrome; the real
        # chip and the twin (same noise pattern wn xor w on its own fingerprint) each answer accept or reject
        wt = 1 + q % (t + 10)
        e = np.zeros(PUF_BITS, np.uint8); e[r.choice(PUF_BITS, wt, replace=False)] = 1
        se = bch_syndrome(e, t)
        wn = puf.read(P["temp"], rng_for(seed + 1000 + q), age)
        r1 = reconstruct(wn, en["h"] ^ se, t); a1 = bool(np.array_equal(r1["key"], en["key"]))
        r2 = reconstruct(w2 ^ wn ^ en["w"], h2 ^ se, t); a2 = bool(np.array_equal(r2["key"], key2))
        same += (a1 == a2 and r1["ok"] == r2["ok"]); acc += a1; fails += (not r1["ok"])
        bw = by_w.setdefault(wt, [0, 0]); bw[0] += 1; bw[1] += (not r1["ok"])
        last = se
    return dict(Q=Q, same=same, acc=acc, fails=fails, by_w=by_w, last_syn=last)
````

### F. One frame through the link, with an optional attack

`simulate()` runs the link block by block:

*1) Transmitter:* Sensor current → TIA → buffer → modulator (Subsection C), giving D.

*2) Key and keystream:* Original design: the stored 16-bit key seeds the LFSR. Proposal: the PUF, enrolled once at 25 °C, is read at the operating temperature, **corrected with the helper data and hashed into the key**; the hash of this reading and 63 further power-up reads gives the chip nonce $N_\mathrm{c}$; and Grain-128a runs with IV = 1 ‖ nonce ‖ counter ‖ $N_\mathrm{c}$. The flipped cells, $\sum_i p_i$, the key-failure probability and the residual secrecy are recorded. The first 16 keystream bits encrypt the chip ID and the following N bits encrypt D. For the original design, the encrypted chip-ID field and the restart of the LFSR from the stored key in every frame are assumptions that [1] does not state (Sections II-E and IX-A).

*3) Framing (Fig. 8):* Proposal: preamble (32 bits, alternating 1 and 0) ‖ sync word 0xB5E3 (16) ‖ counter in clear (32) ‖ chip nonce in clear (32, one bit unused) ‖ encrypted chip ID (16) ‖ C = D ⊕ K (107,400) ‖ MAC tag (32). **The original design's frame has no counter and no tag**.

*4) Channel:* 200 random bits precede the frame; an optional bit-error rate and an optional man in the middle who flips payload bits (header, counter and ID intact) can be set.

*5) Receiver:* It searches for the 48-bit preamble and sync word, recomputes the tag with its own key and nonce, checks that the counter exceeds the last one accepted, decrypts the chip ID and the payload, and filters $D_\mathrm{rx} = C_\mathrm{rx} \oplus K_\mathrm{rx}$ (received bits $C_\mathrm{rx}$, its own keystream $K_\mathrm{rx}$) into readings (Subsection C). A frame is accepted when the ID decrypts correctly and, with the MAC on, **the tag matches and the counter is fresh**; for a sine input, the SNDR of the readings is computed.

*6) Diagnostics:* The fraction of ones in D and its longest run of equal bits (used by the saturation attack), and BM on the first 64 keystream bits predicting the next 1,000 (a probe of the keystream's linearity).

An attack acts at the sensor input, on the chip (a counterfeit), on the helper-data downlink, on the radio channel, or only in the attacker's processing of the recorded frame; `simulate(attack=...)` applies it there and passes the frame to `run_attack` (Subsection G).

> **Next cell.** 1. Defines `build_frame` (preamble, sync, counter, encrypted ID, encrypted modulator bits, MAC tag). 2. Defines `simulate(**settings)`: transmitter, key, keystream, frame, channel and receiver, with the attack applied where the attacker acts; it returns every intermediate signal of the frame, so that any block can be plotted. The cell only defines functions; no simulation is run.

````python
#@title One frame through the link
# ---------- one frame through the whole link, with an optional attack hooked in where the attacker acts
def build_frame(D, ks, mac, ctr, id_bits, cn=0):
    # the first 16 keystream bits encrypt the chip ID, the next n the modulator bits
    n = len(D)
    K = ks["enc"][ID_BITS:ID_BITS + n]; C = D ^ K; enc_id = id_bits ^ ks["enc"][:ID_BITS]
    F = dict(K=K, C=C, enc_id=enc_id)
    if mac:                               # preamble | sync | counter, chip nonce (clear) | enc ID | C | tag
        F["ctr_bits"] = int_to_bits(ctr, 32)                # the chip nonce enters the tag through the IV
        F["tag"] = grain_tag(mac_msg(F["ctr_bits"], enc_id, C), ks)
        F["frame"] = np.concatenate([PREAMBLE, SYNC, F["ctr_bits"], int_to_bits(cn, 32), enc_id, C, int_to_bits(F["tag"], 32)])
    else:
        F["frame"] = np.concatenate([PREAMBLE, SYNC, enc_id, C])
    return F

def simulate(**over):
    """Run one frame. Keyword arguments override DEFAULTS; atk={...} overrides the attack knobs ATK_DEF."""
    P = {**DEFAULTS, **{k: v for k, v in over.items() if k != "atk"}}
    AK = {**ATK_DEF, **over.get("atk", {})}
    n, t, A, base = N_DSM, P["bch_t"], P["attack"], P["scheme"] == "baseline"
    S = dict(P=P, n=n)

    # transmitter analog chain (some attacks set the input the chip actually sees)
    i = sensor_current(P, n)
    if A == "knownin": i = np.full(n, 0.0 if AK["ki_input"] == "zero" else 0.5e-6)    # the attacker forces a known input
    elif A == "saturate": i = np.full(n, P["force_uA"] * 1e-6)                       # the attacker forces a large current
    elif A == "clone": i = np.full(n, AK["fake"] * 1e-6)                              # the counterfeit reports the attacker's value
    elif A == "replay": i = i * 0.4                                                   # the real reading has dropped
    S["i_sense"] = i; S["v_tia"] = tia(i, P["rtia"]); S["u"] = s2d(S["v_tia"], P["buffer"])
    S.update(dsm2(S["u"], P, noise_seed=P["seed"]))

    # key and keystream at the transmitter
    ctr_tx = P["frame_ctr"] + 1 if A == "replay" else P["frame_ctr"]
    rx = {}
    # original: the stored 16-bit key seeds the LFSR (no MAC). Proposal: enrol the PUF at test, read it again in the
    # field (or a clone's PUF), reconstruct the key through the possibly attacked helper data, then the session nonce,
    # the chip nonce and the keystream for this frame's IV
    if base:
        rx["key16"] = P["key16"]
        ks_tx = dict(enc=lfsr16_bits(P["key16"], ID_BITS + n)); S["mac"] = False
    else:
        puf = SRAMPUF(P["chip_seed"], PUF_BITS, P["puf_sigma"], P["puf_bias"], P["puf_tc"])
        en = enroll(puf, t, rng_for(P["chip_seed"] * 7919 + 1))                       # once, at test
        S["w_enroll"], S["key_enroll"] = en["w"], en["key"]; rx["key"], rx["h"] = en["key"], en["h"]
        puf_tx = SRAMPUF(P["chip_seed"] + 500 + AK["clone"], PUF_BITS, P["puf_sigma"], P["puf_bias"], P["puf_tc"]) if A == "clone" else puf
        h_bits = syn_to_bits(en["h"]); S["h_bits"] = h_bits.copy()
        if A == "htamper":                                                            # the attacker flips helper bits in transit
            idx = rng_for(P["seed"] + 51).choice(len(h_bits), AK["h_flip"], replace=False); h_bits[idx] ^= 1
        if A == "hmanip":                                                             # the attacker's oracle queries
            S["manip"] = manip_oracle(puf, en, P, AK["manip_q"], P["seed"] + 57); h_bits ^= syn_to_bits(S["manip"]["last_syn"])
        S["h_rx_bits"] = h_bits; S["h_changed"] = bool(np.any(h_bits != S["h_bits"]))
        S["w_noisy"] = puf_tx.read(P["temp"], rng_for(P["seed"] + 41 + AK["clone"] if A == "clone" else P["seed"] + 17), puf_age(P))
        S["p_err"] = puf_tx.p_err(S["w_enroll"], P["temp"], puf_age(P))
        S["key_fail"] = float(fail_tail(S["p_err"], t)); S["residual"] = puf_residual(P)
        S["e_true"] = S["w_noisy"] ^ S["w_enroll"]
        rc = reconstruct(S["w_noisy"], bits_to_syn(h_bits, t), t)
        S["e_hat"], S["key_tx"], S["fe_ok"] = rc["e"], rc["key"], rc["ok"]
        S["n_err_true"] = int(S["e_true"].sum()); S["key_match"] = bool(np.array_equal(S["key_tx"], S["key_enroll"]))
        S["mac"] = bool(P["mac"])
        if S["mac"]:                                                                   # fresh session nonce from the receiver
            S["nonce"] = int(rng_for(P["seed"] + 71).integers(0, 2 ** 32)) if over.get("nonce") is None else int(over["nonce"])
            rx["nonce"] = S["nonce"]
            S["cn"] = chip_nonce(nonce_reads(puf_tx, P["temp"], rng_for(P["seed"] + 19), puf_age(P), S["w_noisy"], P["nc_reads"]),
                                 P.get("session", 0)) if P["chip_nonce"] else 0
        ks_tx = keystream_nl(S["key_tx"], iv_bits(S["nonce"], ctr_tx, S.get("cn", 0)) if S["mac"] else None, ID_BITS + n)
    S["ks_all"] = ks_tx["enc"]
    id_bits = int_to_bits(P["chip_id"], ID_BITS)
    F = build_frame(S["D"], ks_tx, S["mac"], ctr_tx, id_bits, S.get("cn", 0))
    S.update(K=F["K"], C=F["C"], enc_id=F["enc_id"], frame=F["frame"], tag=F.get("tag"))
    body = 64 if S["mac"] else 0          # counter and chip nonce between the header and the encrypted ID
    if A == "reuse":                      # A15: this frame is recorded; later the attacker replays its session nonce N in
        R, sk = AK["reuse_sessions"], []  # R new sessions while forcing a large input, and takes the majority bit
        gf = 1                            # the forced current is stepped by 5 nA per session, so the sessions' bits differ
        for r in range(R):
            uf = s2d(tia(np.full(n, (AK["sat_uA"] - 0.005 * r) * 1e-6), P["rtia"]), P["buffer"])
            Dr_ = dsm2(uf, P, noise_seed=P["seed"] + 500 + r)["D"]
            if base:
                Kr = S["K"]                                       # the LFSR restarts from the stored key in every frame
            else:
                w_r = nonce_reads(puf, P["temp"], rng_for(P["seed"] + 600 + r), puf_age(P), None, P["nc_reads"])   # a new power-up
                cn_r = chip_nonce(w_r, 0) if P["chip_nonce"] else 0
                Kr = keystream_nl(S["key_tx"], iv_bits(S["nonce"], ctr_tx, cn_r) if S["mac"] else None, ID_BITS + n)["enc"][ID_BITS:ID_BITS + n]
                sk.append(cn_r)
            S.setdefault("_Cr", []).append(Dr_ ^ Kr)
        # majority of the R ciphertexts xor the known forced bit: the keystream guess if every session reused it
        kg = ((np.sum(S["_Cr"], axis=0) * 2 > R).astype(np.uint8)) ^ gf
        y_hat, y_true = decode(S["C"] ^ kg), decode(S["D"])
        r_ = float(np.corrcoef(y_hat[5:], y_true[5:])[0, 1])
        S["reuse"] = dict(sessions=R, kacc=float(np.mean(kg == S["K"])), corr=r_, cn_distinct=len(set(sk)) if sk else 0,
                          v="breached" if r_ > 0.9 else ("leak" if r_ > 0.5 else "safe"))
        del S["_Cr"]

    # what reaches the receiver: the chip's frame, or one the attacker substitutes
    on_air = S["frame"]
    if A == "spoof":
        Df = dsm2(np.full(n, s2d(tia(AK["fake"] * 1e-6, P["rtia"]), P["buffer"])), P, noise_seed=P["seed"] + 31)["D"]
        if base:
            S["bf_key"] = int(brute_lfsr16(S["ks_all"][:ID_BITS])); ks_e = dict(enc=lfsr16_bits(max(S["bf_key"], 1), ID_BITS + n))
        else:
            gk = rng_for(P["seed"] + 33).integers(0, 2, 128).astype(np.uint8)
            ks_e = keystream_nl(gk, iv_bits(S["nonce"], P["frame_ctr"] + 1, S.get("cn", 0)) if S["mac"] else None, ID_BITS + n)
        FF = build_frame(Df, ks_e, S["mac"], P["frame_ctr"] + 1, id_bits, S.get("cn", 0))
        S["fake"] = dict(D=Df, C=FF["C"]); on_air = FF["frame"]
    elif A == "replay":
        old = simulate(**{**over, "attack": "none"})
        S["old"] = dict(y=old["y"], ctr=P["frame_ctr"]); on_air = old["frame"]

    # channel: random bits before the frame (the receiver must find the header), then bit errors at rate ber
    rc_ = rng_for(P["seed"] + 5)
    rx_bits = np.concatenate([rc_.integers(0, 2, P["lead"]).astype(np.uint8), on_air])
    if P["ber"] > 0:
        rx_bits ^= (rc_.random(len(rx_bits)) < P["ber"]).astype(np.uint8)
    mp = AK["flip_p"] if A == "bitflip" else P["mitm_p"]
    S["mitm_flips"] = 0
    if mp > 0:                            # a man-in-the-middle flips payload bits (header and ID left intact)
        mask = (rng_for(P["seed"] + 77).random(n) < mp).astype(np.uint8)
        o = P["lead"] + 48 + body + ID_BITS
        rx_bits[o:o + n] ^= mask; S["mitm_mask"] = mask; S["mitm_flips"] = int(mask.sum())
    S["rx_bits"] = rx_bits

    # receiver: find the 48-bit preamble and sync word, regenerate the keystream with its own key, check the MAC and the
    # counter, decrypt and decode
    hdr = np.concatenate([PREAMBLE, SYNC]); pos = -1
    for k in range(len(rx_bits) - 48):
        if rx_bits[k] == hdr[0] and np.array_equal(rx_bits[k:k + 48], hdr):
            pos = k + 48; break
    if base:
        S["key_rx16"] = P["key16"] ^ 1 if P["rx_wrong_key"] else rx["key16"]
        ks_rx = lfsr16_bits(S["key_rx16"], ID_BITS + n)
    else:
        S["key_rx"] = rng_for(P["seed"] + 9).integers(0, 2, 128).astype(np.uint8) if P["rx_wrong_key"] else rx["key"]
        if not S["mac"]:
            ks_rx = keystream_nl(S["key_rx"], None, ID_BITS + n)["enc"]
    if pos >= 0:
        enc_id_rx = rx_bits[pos + body:pos + body + ID_BITS]; C_rx = rx_bits[pos + body + ID_BITS:pos + body + ID_BITS + n]
    else:
        enc_id_rx = np.zeros(ID_BITS, np.uint8); C_rx = np.zeros(n, np.uint8)
    S["rx_last"] = max(P["rx_last_ctr"], P["frame_ctr"]) if A == "replay" else P["rx_last_ctr"]
    if S["mac"]:                          # fresh counter + matching tag, recomputed with the receiver's key
        S["ctr_rx"] = bits_to_int(rx_bits[pos:pos + 32]) if pos >= 0 else 0
        S["cn_rx"] = bits_to_int(rx_bits[pos + 32:pos + 64]) if pos >= 0 else 0
        S["tag_rx"] = bits_to_int(rx_bits[pos + body + ID_BITS + n:pos + body + ID_BITS + n + 32]) if pos >= 0 else 0
        ok, S["tag_calc"], ks = mac_verify(S["key_rx"], rx["nonce"], S["ctr_rx"], enc_id_rx, C_rx, S["tag_rx"], S["cn_rx"])
        S["mac_ok"] = pos >= 0 and ok; S["ctr_fresh"] = S["ctr_rx"] > S["rx_last"]; ks_rx = ks["enc"]
    K_rx = ks_rx[ID_BITS:ID_BITS + n]
    S["id_ok"] = pos >= 0 and bool(np.array_equal(enc_id_rx ^ ks_rx[:ID_BITS], id_bits))
    S["accepted"] = S["id_ok"] and (not S["mac"] or (S["mac_ok"] and S["ctr_fresh"]))
    S.update(C_rx=C_rx, K_rx=K_rx, D_rx=C_rx ^ K_rx)
    S["y"] = decode(S["D_rx"])
    S["sndr"] = sndr(S["y"], P["bin"]) if P["sig"] == "sine" and A not in ("knownin", "saturate", "clone") else None
    # BM on the first 64 true keystream bits, predicting the next 1,000: how linear the keystream is
    S["eve_probe"] = bm_attack(S["K"][:64], S["K"][64:1064])
    S["ones"] = float(S["D"].mean()); S["run"] = longest_run(S["D"])
    S["atk"] = None if A == "none" else run_attack(S, P, A, AK, base)
    return S
````

### G. The 15 attacks and their verdicts

An attack succeeds with DATA BREACHED (the attacker's keystream estimate agrees with the true keystream on more than 99 % of the bits it is used for, so the data can be read), **FAKE DATA ACCEPTED** (the receiver accepts a frame that the attacker forged, cloned, replayed or altered), INFORMATION LEAKS (two ciphertexts or the oracle's answers reveal information about the data or the fingerprint) or KEY WEAKENED (fewer than 192 secret fingerprint bits remain, Subsection D). It fails with DATA SAFE, ATTACK DETECTED (the receiver rejects the frame) or NOT APPLICABLE (the design has no such block).

Table 5 lists what the attacker does in each attack, with **three rules**. (i) When a key-recovery attack ends with a register (Berlekamp-Massey, replica and saturation attacks), the register is extended forward from the end of its window and, for an LFSR, backward to the start of the frame; the keystream agreement is scored from the first bit the attacker did not already know. If no window yields the key, the attacker keeps the raw estimate, scored over the second half of the frame. (ii) Against the proposed design, BM cannot extend Grain-128a; a replica or constant guess that already agrees with the keystream on more than 97 % of the bits exposes only the frame in which it was made, and counts as a breach only when the counter and MAC are off, because only then is the same keystream reused in the next frame. (iii) Each attack draws its random quantities from its own seed offset (for example `seed` + 2 and + 3 for the replica's noise and windows, `seed` + 12 for the second channel, `seed` + 23 for the saturation windows, `seed` + 31 and + 33 for the forged frame and the guessed key), and the replica and second channel have their own dither seeds.

<p align="center" style="text-align:center"><b>Fig. 16.</b> Known-input replica attack on the original design with an ideal (top) and a real replica (bottom). Rows, for 32 bits of one frame under a known DC input: victim's bits $D$, replica's bits $D_\mathrm{r}$ (orange: differing bits) and keystream estimate $\hat{K} = C \oplus D_\mathrm{r}$. Right: linear complexity L(n) of K̂; dashed: n/2. Recorded animation (<code>python dev/assets/make_assets.py replica_attack</code>, SKY130-derived settings).</p>

Fig. 16 illustrates what the replica attack exploits (the verdicts are computed in Sections VIII and IX). With an ideal replica, $C \oplus D_\mathrm{r}$ is the keystream: L(n) stops at 16 and Berlekamp-Massey predicts the following bits 100 % right. A replica with a 0.1 % FS input offset and its own dither and noise disagrees with the victim on 49 % of the bits (recorded animation); in the window shown L(n) grows as n/2, and none of 80 random windows yields the key; but the two chips agree in runs, and **a scan of every window, with the replica's input trimmed, finds a clean stretch and the key** (Section IX-A, A4).

> **Next cell.** 1. Defines the verdicts. 2. Defines `run_attack`, the attacker's actions in each of the 15 attacks and the test that decides its verdict (Table 5). 3. Lists the attacks (`ATTACKS`) and defines `verdict_table()`, which runs every attack on both designs. The cell only defines functions; no simulation is run. Section IX runs them.

````python
#@title The 15 attacks and their verdicts
VERDICT = dict(breached="DATA BREACHED", spoofed="FAKE DATA ACCEPTED", leak="INFORMATION LEAKS", weak="KEY WEAKENED",
               safe="DATA SAFE", detected="ATTACK DETECTED", na="NOT APPLICABLE")

def run_attack(S, P, A, AK, base):
    """The attacker's side. Key-recovery attacks end in khat, the estimate of the keystream, which is used to decrypt
    the recorded ciphertext. Frame attacks are judged by what the receiver accepts."""
    n = S["n"]; E = dict(id=A, ytrue=decode(S["D"]))
    # the attacker's keystream without the key: the LFSR from a wrong seed, or Grain-128a under a random key
    def guess_ks():
        if base:
            return lfsr16_bits((P["key16"] ^ 0x5A5A) or 1, ID_BITS + n)[ID_BITS:]
        gk = rng_for(P["seed"] + 33).integers(0, 2, 128).astype(np.uint8)
        return keystream_nl(gk, None, ID_BITS + n)["enc"][ID_BITS:]
    # decrypt the recorded ciphertext with the estimate khat; breached if more than 99 % of its bits from frm on are right
    def recover(khat, frm=0):
        E.update(khat=khat, frm=frm, kacc=float(np.mean(khat[frm:] == S["K"][frm:])))
        E["dhat"] = S["C"] ^ khat; E["yhat"] = decode(E["dhat"])
        E["v"] = "breached" if E["kacc"] > 0.99 else "safe"
    # the replica attack: a guess of the modulator bits gives a keystream guess C xor D_guess; BM on random 64-bit
    # windows of it; the earliest window that predicts the true keystream is extended forwards and backwards
    def replica(dguess, seed):
        kg = S["C"] ^ dguess; w = window_attack(kg, S["K"], seed); E.update(win=w, kguess=kg)
        kh = kg.copy(); frm = n // 2
        if w["first"] >= 0:
            known = kg[w["first"]:w["first"] + 64]; L, c = bm2(known)
            kh[w["first"] + 64:] = lfsr_predict(c, L, known, n - w["first"] - 64); frm = w["first"] + 64
            if L > 0 and c[L]:            # an LFSR runs backwards too
                for k in range(w["first"] + L - 1, L - 1, -1):
                    v = kh[k]
                    for j in range(1, L):
                        v ^= c[j] & kh[k - j]
                    kh[k - L] = v
                frm = 0
        if not base:                      # BM cannot extend Grain-128a; a near-perfect guess of D still exposes this frame
            E["raw"] = float(np.mean(kg == S["K"]))
            if E["raw"] > 0.97:
                recover(kg, 0); E["v"] = "safe" if S["mac"] else "breached"; E["reuse"] = True; return
        recover(kh, frm)
        if not base:
            E["v"] = "safe"
    if A == "eaves":
        recover(guess_ks()); E["v"] = "safe"
    elif A == "brute":                    # the encrypted public chip ID gives 16 keystream bits; 2^16 seeds are searched
        if base:
            E["bf_key"] = int(brute_lfsr16(S["ks_all"][:ID_BITS])); recover(lfsr16_bits(max(E["bf_key"], 1), ID_BITS + n)[ID_BITS:])
        else:
            recover(guess_ks())
    elif A == "extract":                  # the original's key is read out of memory; the proposal stores no key
        recover(lfsr16_bits(P["key16"], ID_BITS + n)[ID_BITS:] if base else guess_ks())
    elif A == "bm":                       # BM on m known keystream bits, then the LFSR predicts the rest of the frame
        m = min(AK["bm_bits"], n - 1); known = S["K"][:m]; L, c = bm2(known)
        kh = np.concatenate([known, lfsr_predict(c, L, known, n - m)]); E.update(L=int(L), m=m); recover(kh, m)
    elif A == "knownin":                  # a replica chip with an input mismatch (% FS) and its own dither seeds
        Dr = dsm2(S["u"] + AK["mismatch"] / 100, P, seeds=(0x2C4D, 0x51E7), noise_seed=P["seed"] + 2)["D"]
        E["dguess"] = Dr; E["flip"] = float(np.mean(Dr != S["D"])); replica(Dr, P["seed"] + 3)
        if base and E["v"] != "breached":   # the competent attacker: trim the replica's input, scan every window
            idb = int_to_bits(P["chip_id"], ID_BITS)
            for tr in AK["trim"]:
                Dt = Dr if tr == 0 else dsm2(S["u"] + (AK["mismatch"] - tr) / 100, P, seeds=(0x2C4D, 0x51E7), noise_seed=P["seed"] + 2)["D"]
                seed16, nc = lfsr16_scan(S["C"] ^ Dt, S["enc_id"], idb)
                E.update(scan_trim=tr, scan_cand=nc)
                if seed16 >= 0:
                    E["scan_seed"] = seed16; recover(lfsr16_bits(seed16, ID_BITS + n)[ID_BITS:]); break
    elif A == "saturate":                 # the attacker forces a large current in F frames and takes the majority bit
        # first guess: every bit equal to the saturated level
        g = 1 if S["D"].mean() >= 0.5 else 0
        E["dguess"] = np.full(n, g, np.uint8); E["guess"] = g; replica(E["dguess"], P["seed"] + 23)
        if E["v"] != "breached":
            Cs = []                       # the forced current is stepped by 5 nA per frame, so the frames' bits differ
            for f in range(AK["sat_frames"]):
                uf = s2d(tia(np.full(n, (AK["sat_uA"] - 0.005 * f) * 1e-6), P["rtia"]), P["buffer"])
                Df = dsm2(uf, P, noise_seed=P["seed"] + 300 + f)["D"]
                Kf = S["K"] if (base or not S["mac"]) else keystream_nl(S["key_tx"], iv_bits(S["nonce"], P["frame_ctr"] + f, S.get("cn", 0)), ID_BITS + n)["enc"][ID_BITS:]
                Cs.append(Df ^ Kf)
            gf = 1 if uf[0] > 0 else 0
            kg = ((np.sum(Cs, axis=0) * 2 > len(Cs)).astype(np.uint8)) ^ gf
            E["mf_kacc"] = float(np.mean(kg == S["K"])); E["frames"] = len(Cs)
            if base:
                seed16, nc = lfsr16_scan(kg, S["enc_id"], int_to_bits(P["chip_id"], ID_BITS)); E["scan_cand"] = nc
                if seed16 >= 0:
                    E["scan_seed"] = seed16; recover(lfsr16_bits(seed16, ID_BITS + n)[ID_BITS:])
            elif E["mf_kacc"] > 0.99 and not S["mac"]:
                recover(kg, 0)
    elif A == "samekey":                  # a second frame of the same reading under the same key
        D2 = dsm2(S["u"] + AK["ch_mismatch"] / 100, P, seeds=(0x3A17, 0x6E29), noise_seed=P["seed"] + 12)["D"]
        K2 = keystream_nl(S["key_tx"], iv_bits(S["nonce"], P["frame_ctr"] + 1, S.get("cn", 0)), ID_BITS + n)["enc"][ID_BITS:] if (not base and S["mac"]) else S["K"]
        E["C2"] = D2 ^ K2; E["X"] = S["C"] ^ E["C2"]; E["xfrac"] = float(E["X"].mean()); E["dflip"] = float(np.mean(D2 != S["D"]))
        # leakage test: where both frames carry the same reading, P(D1 != D2) = (1 - u1*u2)/2, so the fraction of ones
        # in X gives the attacker the mean square of the reading: A_hat = sqrt(2*(1 - 2*xfrac)) estimates a sine's
        # amplitude. It leaks when it lands within 25 % of the true amplitude A = sqrt(2*mean(u^2)).
        A_true = float(np.sqrt(2 * np.mean(S["u"] ** 2)))
        E["a_hat"] = float(np.sqrt(max(0.0, 2 * (1 - 2 * E["xfrac"])))); E["a_true"] = A_true
        E["leak_err"] = abs(E["a_hat"] - A_true) / A_true if A_true > 0 else 1.0
        E["v"] = "leak" if (abs(E["xfrac"] - 0.5) > 0.03 or E["leak_err"] < 0.25) else "safe"
    elif A == "reuse":                    # A15: the attacker replays one old session nonce in several new sessions
        E["v"] = S["reuse"]["v"]; E.update(S["reuse"])
    elif A == "spoof":
        E["yfake"] = decode(S["fake"]["D"]); E["v"] = "spoofed" if S["accepted"] else "detected"
    elif A == "clone":
        E["v"] = "spoofed" if S["accepted"] else "detected"
    elif A == "replay":
        E["yold"] = S["old"]["y"]; E["v"] = "spoofed" if S["accepted"] else "detected"
    elif A == "bitflip":
        E["v"] = "safe" if not S["mitm_flips"] else ("spoofed" if S["accepted"] else "detected")
    elif A == "htamper":
        E["v"] = "na" if base else ("safe" if S["key_match"] else "detected")
    elif A == "hleak":
        # secret bits left at the point estimate of the cell bias and at the 95 % lower bound of the min-entropy
        # (P["puf_h_lo"], Section V-D); the verdict uses the lower bound against the secrecy requirement P["secret_req"]
        # (192 bits: the 128-bit key plus the leftover-hash margin 2 log2(1/eps), Section VII-E)
        E["residual"] = puf_residual(P)
        h_lo = P.get("puf_h_lo") or min_ent_bit(P["puf_bias"])
        E["residual_lo"] = int(math.floor(PUF_BITS * h_lo - 10 * P["bch_t"]))
        E["v"] = "na" if base else ("safe" if E["residual_lo"] >= P.get("secret_req", 128) else "weak")
    elif A == "hmanip":
        E["v"] = "na" if base else ("safe" if S["manip"]["same"] == S["manip"]["Q"] else "leak")
    return E

ATTACKS = [  # id, name, from the paper?
    ("eaves", "Passive eavesdropping", True), ("brute", "Brute-force key search", False),
    ("bm", "Berlekamp-Massey keystream attack", False), ("knownin", "Zero / DC input with a replica chip", True),
    ("samekey", "Same key, same message", True), ("saturate", "Forced-input saturation", False),
    ("extract", "Key extraction from memory", False), ("spoof", "Spoofing: forged frames (MITM)", True),
    ("clone", "Cloned / counterfeit sensor", False), ("htamper", "Helper-data tampering", False),
    ("hleak", "Helper-data leakage", False), ("hmanip", "Helper-data manipulation oracle", False),
    ("replay", "Replay of a recorded frame", False), ("bitflip", "Bit-flipping in transit", False),
    ("reuse", "Keystream reuse: replayed session nonce", False),
]

# what each verdict rests on, per design: simulated (the attack is run on the model), analytic (a formula, an identity or a
# design argument: for A4 and A6 on the proposal the estimate is simulated, but the verdict follows from the per-frame IV),
# by design (the proposal: the verdict follows from the design without an attack simulation: the link is encrypted, the
# key is 128 bits long, no key is stored, the key comes from the chip's own PUF), assumed (the original: a property of [1]
# taken as given: a key in non-volatile memory, which [1] does not state, or encryption against a passive eavesdropper)
EVIDENCE = dict(
    eaves=("assumed", "by design"), brute=("simulated", "by design"), bm=("simulated", "simulated"),
    knownin=("simulated", "analytic"), samekey=("simulated", "simulated"), saturate=("simulated", "analytic"),
    extract=("assumed", "by design"), spoof=("simulated", "simulated"), clone=("simulated", "by design"),
    htamper=("-", "simulated"), hleak=("-", "analytic"), hmanip=("-", "analytic"),
    replay=("simulated", "simulated"), bitflip=("simulated", "simulated"), reuse=("simulated", "simulated"))
PROPERTY = dict(eaves="confidentiality", brute="confidentiality", bm="confidentiality", knownin="confidentiality",
                samekey="confidentiality", saturate="confidentiality", reuse="confidentiality", extract="key secrecy",
                hleak="key secrecy", hmanip="key secrecy", htamper="availability", spoof="authenticity",
                clone="authenticity", bitflip="integrity", replay="freshness")

def verdict_table(**over):
    """Every attack against both designs with the given settings: {attack id: (original, proposal)}."""
    return {a: tuple(simulate(**{**over, "scheme": sc, "attack": a})["atk"]["v"] for sc in ("baseline", "proposed"))
            for a, _, _ in ATTACKS}
````

### H. A reference frame

**Without attack the link should be transparent**: the receiver finds the frame, the ID decrypts, the MAC verifies and the decrypted bits equal the modulator bits, so that the SNDR is the modulator's own. The expected number of PUF cells that differ at 37 °C from the 25 °C enrolment, $\sum_i p_i$, should lie below $t = 40$, so that the key is regenerated; with the assumed cell noise the margin is moderate, and a power-up with more than t flips has a non-zero probability (`fail_tail`).

The frame runs on the behavioural model with the `DEFAULTS` of Subsection A (assumed circuit parameters, before Section VII): proposed design, the test sine, one frame, 37 °C. Fig. 17 **tests the transparency in two ways**: panel (a) traces $D$, $K$, $C$, $D_\mathrm{rx}$ and $D \oplus D_\mathrm{rx}$ bit by bit, and panel (b) compares the server reading with the modulator input.

> **Next cell.** 1. Defines `plot_chain`, which draws one frame: every block's signal over 200 clocks (sensor current, second integrator $x_2$, modulator bits $D$, keystream $K$, transmitted bits $C$, received bits; during an attack also the attacker's rows) and, below, the receiver's reading. 2. Defines `plot_reference`, the view of a frame without attack: the bits with their scale, $D \oplus D_\mathrm{rx}$ over the whole frame, and the server's reading against the modulator input with its error. The cell only defines functions; no simulation is run.

````python
#@title Plotting helper: one frame, every block
import re, textwrap

def chain_rows(S):
    """The rows plot_chain draws for this frame: always the modulator, keystream and channel, plus the attacker's
    rows when an attack is running (the keystream estimate, the forged or flipped bits)."""
    return [r for r in chain_items(S) if r[0] != "hdr"]

def chain_items(S):
    """What plot_chain draws, top to bottom. Without an attack: the signal rows. Under an attack only the rows that show
    what the attack does are drawn, grouped by where they are observed (chip and channel, attacker, receiver, server); each
    group opens with a header line ('hdr', text, colour) that states what happens there in this frame. The chip's internal
    signals are the same under every attack and are drawn only where the attacker sets the input (A4, A6)."""
    # each row: (label, kind, signal, colour); kind 'a' an analog signal, 'd' bits, 'k' the attacker's keystream estimate
    # (wrong bits shaded), 'rx' bits at the receiver (changed bits shaded), 'chk' the receiver's checks as text
    E = S["atk"] or {}
    chip = [("sensor current", "a", S["i_sense"] / max(1e-12, np.abs(S["i_sense"][:4000]).max()), COL["sig"]),
            ("integrator 2, $x_2$", "a", S["X2"] / S["P"]["xlim"], COL["sig"]),
            ("modulator bits $D$", "d", S["D"], COL["ink"]), ("keystream $K$", "d", S["K"], COL["key"]),
            ("$C$ on the air", "d", S["C"], COL["ink"])]
    used = COL["bad"] if E.get("v") == "spoofed" else COL["ink"]  # orange: the receiver passes the attacker's bits on
    d_rx = ("receiver bits $D_{\\rm rx}$", "d", S["D_rx"], used if S["accepted"] else COL["grid"])
    if not E:
        return chip + [d_rx]
    A = E["id"]
    chip = ([chip[0], chip[2]] if A in ("knownin", "saturate") else []) + [chip[4]]
    atk = []
    if "fake" in S: atk.append(("attacker's forged $C$", "rx", S["fake"]["C"], COL["key"]))   # shaded where it differs from C
    if "khat" in E: atk.append(("attacker's estimate $\\hat{K}$", "k", E["khat"], COL["key"]))
    if "X" in E: atk.append(("attacker: $C_1 \\oplus C_2$", "d", E["X"], COL["key"]))
    rx = []
    if "old" in S or "mitm_mask" in S: rx.append(("$C$ at the receiver", "rx", S["C_rx"], COL["ink"]))
    rx.append(("receiver checks", "chk", None, COL["ink"]))
    if "fake" in S or "old" in S or "mitm_mask" in S or A == "clone": rx.append(d_rx)   # the frame is not the chip's own
    h = story(S)
    return ([("hdr",) + h["chip"]] + chip + [("hdr",) + h["attacker"]] + atk + [("hdr",) + h["receiver"]] + rx
            + [("hdr",) + h["server"]])

def receiver_checks(S):
    """(name, state) of each check of the receiver on this frame; state is 'pass', 'fail' or 'none' (no such check)."""
    st = lambda ok: "pass" if ok else "fail"
    return [("ID", st(S["id_ok"])), ("counter", st(S["ctr_fresh"]) if S["mac"] else "none"),
            ("MAC", st(S["mac_ok"]) if S["mac"] else "none")]

def story(S):
    """One line per group of rows: what happens at the chip, at the attacker, at the receiver and at the server in this
    frame, with the colour of the outcome (orange: the attack gains something; blue: it is stopped or learns nothing)."""
    E, P, base = S["atk"], S["P"], S["P"]["scheme"] == "baseline"
    A, v = E["id"], E["v"]
    bad, ok, ink, grey = COL["bad"], COL["ok"], COL["ink"], COL["faint"]
    chip = ("Chip and channel: alike in both designs; $C = D \\oplus K$ looks random", grey)
    if A in ("knownin", "saturate"):        # the only attacks whose chip rows differ: the attacker sets the input
        chip = (f"Chip and channel: the attacker forces {(0 if ATK_DEF.get('ki_input') == 'zero' else 0.5) if A == 'knownin' else P['force_uA']:g} µA into the "
                "electrode; $C = D \\oplus K$ still looks random", grey)
    frame = "fake" in S or "old" in S or "mitm_mask" in S or A == "clone"   # the receiver gets a frame not sent by the chip
    pct = lambda x: f"{100 * x:.0f}\u00a0%"                  # no-break space: "100 %" stays on one line
    kacc = E.get("kacc")
    if A == "spoof":
        a = (f"Attacker: encrypts a fake {ATK_DEF['fake']} µA reading with the 16-bit key found by brute force"
             if base else f"Attacker: cannot find the 128-bit key; encrypts a fake {ATK_DEF['fake']} µA reading with a guess")
        a += "; shaded: changed bits"
    elif A == "brute":
        a = (f"Attacker: tries all 65,535 keys; its estimate $\\hat{{K}}$ equals $K$ on {pct(kacc)} of the bits" if base else
             f"Attacker: one guess of the 128-bit key; $\\hat{{K}}$ is wrong on {pct(1 - kacc)} of the bits (shaded)")
    elif A == "replay":
        a = "Attacker: resends a frame recorded when the reading was larger"
    elif A == "bitflip":
        a = f"Attacker: flips {S['mitm_flips']:,} payload bits in transit"
    elif A == "samekey":
        a = f"Attacker: XORs two frames; $C_1 \\oplus C_2$ has {100 * E['xfrac']:.1f} % ones (50 %: nothing leaks)"
    elif A == "clone":
        a = "Attacker: a counterfeit chip, with its own fingerprint, sends the frames"
    elif A == "htamper":
        a = "Attacker: inverts one bit of the helper data $h$ on the downlink"
    elif A == "hleak":
        a = "Attacker: reads the public helper data $h$"
    elif A == "hmanip":
        a = "Attacker: powers the chip up with altered helper data and watches accept or reject"
    elif A == "extract":
        a = ("Attacker: reads the 16-bit key from the chip's memory" if base else
             "Attacker: finds no stored key (it exists only while the PUF is powered); guesses")
    elif kacc is not None:
        how = dict(eaves="records $C$ and guesses the keystream", bm="runs Berlekamp-Massey on known keystream bits",
                   knownin="forces a known input and predicts $D$", saturate="saturates the input and predicts $D$").get(A, "attacks the keystream")
        a = f"Attacker: {how}; $\\hat{{K}}$ equals $K$ on {pct(kacc)} of the bits"
    else:
        a = "Attacker: " + dict((x, n) for x, n, _ in ATTACKS)[A]
    gain = v in ("spoofed", "breached", "leak", "weak")
    attacker = (a, bad if gain else ok if v in ("safe", "detected") else grey)
    chk = receiver_checks(S)
    if S["accepted"]:
        passed = [n for n, s in chk if s == "pass"]; absent = [n for n, s in chk if s == "none"]
        r = ((f"Receiver: its only check, {passed[0]}, passes" if absent and len(passed) == 1 else
              f"Receiver: {', '.join(passed[:-1])} and {passed[-1]} pass" if len(passed) > 1 else f"Receiver: {passed[0]} passes")
             + ", so the frame is accepted" + ("; $D_{\\rm rx}$ (orange) is the attacker's" if v == "spoofed" and frame else ""))
        receiver = (r, bad if v == "spoofed" else ink)
    else:
        failed = [n for n, s in chk if s == "fail"]
        receiver = (f"Receiver: {' and '.join(failed)} {'fails' if len(failed) == 1 else 'fail'}, so the frame is discarded"
                    + ("; $D_{\\rm rx}$ in grey" if frame else ""), ok)
    if "old" in S or "mitm_mask" in S:
        receiver = (receiver[0] + (". Shaded: changed bits" if "old" in S else ". Shaded: the flipped bits"),
                    receiver[1])
    if v == "spoofed":
        server = ("Server: shows the attacker's data in place of the true reading", bad)
    elif not S["accepted"]:
        server = ("Server: shows nothing for this frame; the attack is detected", ok)
    elif v == "breached":
        server = ("Server: true reading; the attacker's decryption (dashed) equals it", bad)
    elif v == "na":
        server = ("Server: true reading; this design has no helper data to attack", grey)
    elif gain:
        server = ("Server: true reading; but the attacker learns about the data or the key", bad)
    elif "yhat" in E:
        server = ("Server: true reading; the attacker decrypts only noise (dashed)", ok)
    else:
        server = ("Server: true reading; the attack has no effect", ok)
    return dict(chip=chip, attacker=attacker, receiver=receiver, server=server)

HDR_WRAP = 66                            # characters per header line (one panel of a two-column figure)

def wrap_tex(text, width=HDR_WRAP):
    """Lines of text wrapped at width characters, never inside a $...$ math span (counted at its rendered length)."""
    # split at the $ signs: the odd parts are math, kept whole and counted by their rendered length (commands, braces,
    # sub- and superscript marks removed); a math span written against a word (e.g. "$D$;") is glued to it
    parts = text.split("$")
    words = []
    for i, p in enumerate(parts):
        new = [("$" + p + "$", len(re.sub(r"\\\w+|[{}_^]", "", p)) or 1)] if i % 2 else [(w, len(w)) for w in p.split(" ") if w]
        glued = parts[i - 1][-1:] not in ("", " ") if i % 2 else p[:1] not in ("", " ")      # no space between them in the text
        if i and glued and words and new:                                                     # e.g. "$D$;": no space before ";"
            (w0, k0), (pw, pk) = new.pop(0), words.pop(); words.append((pw + w0, pk + k0))
        words += new
    # greedy filling: a word that would pass the width starts a new line
    lines, cur, n = [], [], 0
    for w, k in words:
        if cur and n + 1 + k > width:
            lines.append(" ".join(cur)); cur, n = [], 0
        cur.append(w); n += k + (1 if n else 0)
    return lines + [" ".join(cur)] if cur else lines

def chain_ratios(S):
    """Height ratios of the axes plot_chain needs for this frame (header lines, signal rows, the reading)."""
    r = []
    for it in chain_items(S):
        r.append(0.62 * len(wrap_tex(it[1])) + 0.25 if it[0] == "hdr" else 1.0)
    return r + [3.4]

def plot_reading(ax, S, j=None, small=False):
    """The reading panel of plot_chain: the true reading (grey), the server's output (black; orange when it is the
    attacker's data) and what the attacker decrypts, forges or replays (red, dashed: the attacker's colour, not an
    outcome colour); a rejected frame is marked. small: the thin lines and short labels of the attack atlas."""
    E = S["atk"] or {}
    j = np.arange(60, 180) if j is None else j
    k = 0.6 if small else 1.0
    if E:
        yt = E.get("ytrue", S["y"])
        ax.plot(j, yt[j], color=COL["faint"], lw=1.0 * k, label="true reading")
    else:                                  # no attack: the true reading is the modulator input (delayed and scaled by the
        _, _, uin, err = reading_error(S)  # receiver filters), drawn wide and light under the server output
        ax.plot(j, uin[j], color=COL["faint"], lw=3.2, alpha=0.55, label="true reading (the input)")
        lo, hi = min(uin[j].min(), S["y"][j].min()), max(uin[j].max(), S["y"][j].max())
        ax.set_ylim(lo - 0.06 * (hi - lo), hi + 0.42 * (hi - lo))
        nd = int(np.count_nonzero(S["D_rx"] != S["D"]))
        rms = 100 * np.sqrt(np.mean(err[60:len(err) - 10] ** 2))
        ax.text(0.5, 0.97, f"$D_{{\\rm rx}}$ = $D$: {nd} of {len(S['D']):,} bits differ;  reading − input: {rms:.3f} % FS rms",
                transform=ax.transAxes, ha="center", va="top", fontsize=8.5, color=COL["ok"] if nd == 0 else COL["bad"],
                bbox=dict(boxstyle="square,pad=0.15", fc="white", ec="none"), zorder=6)     # the grid stays behind the text
    if S["accepted"]:
        fooled = E.get("v") == "spoofed"
        ax.plot(j, S["y"][j], color=COL["bad"] if fooled else COL["ink"], lw=(2.6 if fooled else 1.4) * k, alpha=0.85,
                label="server output (attacker's data)" if fooled else "server output")
    for key, lab in (("yhat", "what the attacker decrypts"), ("yfake", "attacker's forged reading"), ("yold", "old reading replayed")):
        if key in E:
            ax.plot(j, E[key][j], color=COL["key"], lw=1.2 * k, ls=(0, (3, 1.6)), label=lab)
    if not S["accepted"]:
        ax.text(0.5, 0.5, "rejected" if small else "frame rejected:\nthe server shows nothing", transform=ax.transAxes,
                ha="center", va="center", fontsize=7.5 if small else 8.5, color=COL["ok"],
                bbox=dict(boxstyle="square,pad=0.2", fc="white", ec="none"), zorder=5)
    ax.set_xlim(j[0], j[-1])
    return j

def plot_chain(S, axs=None, k0=2100, W=200, title=None, labels=True, legend=True):
    """Every block's signal for one frame: bits over W clocks, then the readings. Under an attack, each group of rows
    opens with a line that says what happens there, and the bits the attacker changed are shaded."""
    E = S["atk"] or {}
    items = chain_items(S)
    if axs is None:
        fig, axs = plt.subplots(len(items) + 1, 1, figsize=(W1 + 1.4, 0.36 * len(items) + 1.9), sharex=False,
                                gridspec_kw=dict(height_ratios=chain_ratios(S), hspace=0.18))
        fig.subplots_adjust(left=0.30, right=0.97, top=0.92, bottom=0.17)
    x = np.arange(W)
    for ax, it in zip(axs, items):
        if it[0] == "hdr":                 # a group header across the whole panel, from its left edge (ink: the outcome is
                                           # the verdict line under the panel title)
            _, text, c = it
            ax.axis("off")
            fr = ax.figure
            tr = mpl.transforms.blended_transform_factory(getattr(fr, "transSubfigure", fr.transFigure), ax.transAxes)
            body = "\n".join(wrap_tex(text))
            ax.text(0.02, 0.0, body, transform=tr, ha="left", va="bottom", fontsize=8.5, color=COL["ink"], linespacing=1.15)
            continue
        name, kind, sig, c = it
        if kind == "chk":                  # the receiver's checks as text: ✓ passes (blue), × fails (orange), grey: none
            for i, (nm, state) in enumerate(receiver_checks(S)):
                tc = dict(pass_=COL["ok"], fail=COL["bad"], none=COL["faint"])["pass_" if state == "pass" else state]
                lab = f"no {nm}" if state == "none" else f"✓ {nm}" if state == "pass" else f"× {nm}"
                ax.text((i + 0.5) / 3, 0.5, lab, transform=ax.transAxes, ha="center", va="center", fontsize=8.5, color=tc)
            ax.set_ylim(0, 1)
        elif kind == "a":
            v = sig[k0:k0 + W]
            ax.plot(x, v, color=c, lw=1.1); ax.set_ylim(-1.2, 1.2)
        else:
            v = sig[k0:k0 + W]
            ax.step(x, v, where="post", color=c, lw=0.8); ax.set_ylim(-0.3, 1.3)
            if kind == "k":                # mark the bits the attacker got wrong (the header above states their share)
                wrong = v != S["K"][k0:k0 + W]
                ax.fill_between(x, -0.3, 1.3, where=wrong, step="post", color=COL["bad"], alpha=0.65, lw=0)
            if kind == "rx":               # shade the bits that differ from those the chip sent
                diff = v != S["C"][k0:k0 + W]
                ax.fill_between(x, -0.3, 1.3, where=diff, step="post", color=COL["bad"], alpha=0.30, lw=0)
        ax.set_yticks([]); ax.set_xticks([]); ax.set_xlim(0, W); ax.grid(False)
        ax.tick_params(top=False, right=False)
        for sp in ("top", "right", "left"): ax.spines[sp].set_visible(False)
        ax.spines["bottom"].set_color(COL["grid"])
        if labels:
            ax.set_ylabel(name, rotation=0, ha="right", va="center", fontsize=9)
    ax = axs[-1]
    j = plot_reading(ax, S)
    ax.set_xlim(j[0], j[-1]); ax.set_xticks([80, 120, 160])
    ax.set_xlabel("Output sample"); ax.set_yticks([]); ax.tick_params(right=False)
    if labels:
        ax.set_ylabel("reading", rotation=0, ha="right", va="center", fontsize=9)
    if legend:
        ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.36), fontsize=9, ncol=2, frameon=False)
    if title:
        axs[0].set_title(title, loc="center", fontsize=10, pad=6)
    return axs

def reading_error(S, j0=60):
    """The server's reading against the modulator input u. The input is delayed by the receiver filters' group delay and
    scaled by their gain at the tone (both fitted by least squares; the gain also carries the front end's sign).
    Returns (delay in clocks, gain, input at each output sample, error = reading - input)."""
    y, u = np.asarray(S["y"], float), np.asarray(S["u"], float)
    k, t = np.arange(len(y)) * OSR, np.arange(len(u))
    seg = slice(j0, len(y) - 10)
    def fit(lag):
        uu = np.interp(k - lag, t, u)
        g = np.dot(y[seg], uu[seg]) / np.dot(uu[seg], uu[seg])
        return np.mean((y[seg] - g * uu[seg]) ** 2), g, g * uu
    # the lag with the smallest residual: whole clocks over two output periods, then steps of 0.02 clock around the best
    lag0 = min(range(0, 2 * OSR), key=lambda l: fit(l)[0])
    lag = min(np.arange(lag0 - 1, lag0 + 1.001, 0.02), key=lambda l: fit(l)[0])
    _, g, uin = fit(lag)
    return lag, g, uin, y - uin

def plot_reference(S, k0=2100, W=48, j=(60, 180)):
    """One frame without attack, with scales: (a) the bits over W clocks and D xor D_rx, which must be 0 everywhere;
    (b) the server's reading against the delayed input, and (c) their difference in % of full scale."""
    fig = plt.figure(figsize=(W2, 3.3))
    gs = fig.add_gridspec(2, 2, width_ratios=[1.0, 1.15], height_ratios=[1.25, 1.0], left=0.105, right=0.985,
                          bottom=0.14, top=0.92, wspace=0.30, hspace=0.12)
    aa = fig.add_subplot(gs[:, 0]); ab = fig.add_subplot(gs[0, 1]); ac = fig.add_subplot(gs[1, 1], sharex=ab)
    # (a) five bit rows stacked 1.6 apart: D, K, C, D_rx and D xor D_rx, which must stay 0 everywhere
    x = np.arange(k0, k0 + W + 1)
    zero = S["D"] ^ S["D_rx"]
    rows = [("$D$", S["D"], COL["ink"]), ("$K$", S["K"], COL["key"]), ("$C = D \\oplus K$", S["C"], COL["ink"]),
            ("$D_{\\rm rx} = C \\oplus K$", S["D_rx"], COL["ink"]), ("$D \\oplus D_{\\rm rx}$", zero, COL["ok"])]
    for i, (name, b, c) in enumerate(rows):
        off = 1.6 * (len(rows) - 1 - i)
        v = np.append(b[k0:k0 + W], b[k0 + W - 1]).astype(float)
        aa.step(x, off + v, where="post", color=c, lw=1.6 if i == len(rows) - 1 else 0.9)
        aa.axhline(off, color=COL["grid"], lw=0.5, zorder=0)
    aa.set_yticks([1.6 * (len(rows) - 1 - i) + 0.5 for i in range(len(rows))], [r[0] for r in rows])
    aa.tick_params(axis="y", length=0, right=False); aa.grid(False)
    aa.set_ylim(-0.9, 1.6 * (len(rows) - 1) + 1.3); aa.set_xlim(k0, k0 + W)
    aa.set_xlabel("clock cycle")
    n_bad = int(zero.sum())
    aa.text(k0 + W / 2, -0.25, f"{'0' if n_bad == 0 else n_bad} of {len(zero):,} bits differ", ha="center", va="top",
            fontsize=9, color=COL["ok"] if n_bad == 0 else COL["bad"], fontweight="bold")
    panel(aa, "a", "bits, each row 0 (lower) or 1")
    lag, gain, uin, err = reading_error(S)
    jj = np.arange(*j)
    ab.plot(jj, uin[jj], color=COL["faint"], lw=3.0, alpha=0.6, label="input $u$, delayed and scaled")
    ab.plot(jj, S["y"][jj], color=COL["ink"], lw=1.0, label="server reading")
    ab.set_ylabel("value (FS)"); ab.tick_params(labelbottom=False)
    lo, hi = min(uin[jj].min(), S["y"][jj].min()), max(uin[jj].max(), S["y"][jj].max())
    ab.set_ylim(lo - 0.08, hi + 0.42 * (hi - lo))                       # the legend above the sine
    ab.legend(loc="upper center", ncol=2, fontsize=8.5, borderpad=0.3, handlelength=1.5, columnspacing=1.0)
    panel(ab, "b", "reading after decryption and decimation")
    seg = slice(60, len(err) - 10)
    rms = 100 * np.sqrt(np.mean(err[seg] ** 2))
    ac.plot(jj, 100 * err[jj], color=COL["sig"], lw=0.9)
    ac.axhline(0, color=COL["grid"], lw=0.8, zorder=0)
    # (c) the error axis spans 3.2 times the largest error, with ticks at half of it rounded to one digit
    m = 3.2 * np.abs(100 * err[jj]).max()
    tk = float(f"{m / 2:.1g}")
    ac.set_ylim(-m, m); ac.set_yticks([-tk, 0, tk]); ac.set_ylabel("error (% FS)")
    ac.set_xlabel("output sample"); ac.set_xlim(jj[0], jj[-1]); ac.set_xticks(range(jj[0], jj[-1] + 1, 30))
    ac.text(0.02, 0.96, f"rms {rms:.3f} % FS, SNDR {S['sndr']:.1f} dB\ninput delayed {lag:.1f} clocks, "
            f"×{abs(gain):.4f}", transform=ac.transAxes, ha="left", va="top", fontsize=8, color=COL["sig"])
    fig.align_ylabels([ab, ac])
    return fig

def verdict_line(S):
    # the frame's verdict and whether the receiver accepted the frame
    E = S["atk"]
    if E is None:
        return "no attack"
    rx = "receiver accepts" if S["accepted"] else "receiver rejects"
    return f"{VERDICT[E['v']]}  ({rx})"
````

# **VI. Consistency check against the measurements of [1]**
---
The model of Section V computes every bit of both designs, but it has been built from specifications, not from the chip. Before any attack is evaluated on it, this section asks **whether it reproduces what the chip of [1] was measured to do**. **Every later verdict is computed on this model**. [1] argues the security of the original design from the measured disagreement between two channels of its chip, and the energy comparison of the designs rests on the energy table of [1]; if the model did not reproduce the channel disagreement, the attacks of Section IX that rest on analog decorrelation could not be trusted.

The check uses the modulator and receiver of Section V-C with the `DEFAULTS` of Section V-A, before any SKY130 parameter. It identifies the modelled effect that makes two model channels disagree, **not the effect at work in the chip of [1]**. Subsection A reproduces the measurement of [1] in which two channels encrypt the same input with the same key (Fig. 4(c) of [1]) and identifies which modelled effect makes their outputs differ; Subsection B checks what the energy table of [1] (Fig. 4(b)) rests on, without using any radio number; Subsection C runs both experiments and compares them with [1].

### A. Two channels, same key, same input (Fig. 4(c) of [1])

The chip of [1] has two nominally identical channels; the only deliberate difference between them is their dither sequence. With the same key, their ciphertext bits differ **in 50.05 % of positions for a zero input and in 49.8 % for the same message**. [1] relies on the dither and on the activity of the bitstream to conceal the keystream. Five relations determine what the model can show about this measurement:

*1) The measured quantity:* With the same keystream $K$ and the ciphertext bits $C_1$, $C_2$ and modulator bits $D_1$, $D_2$ of channels 1 and 2, $C_1 \oplus C_2 = (D_1 \oplus K) \oplus (D_2 \oplus K) = D_1 \oplus D_2$: the fraction of differing ciphertext bits is that of the modulator bits, **0 % for identical modulators and 50 % for independent ones**.

*2) Perturbations at the quantiser are suppressed:* Dither and comparator noise act at the comparator: they change a decision only when the second integrator's state $x_2$ lies within the perturbation of the threshold, and **the loop corrects it as it corrects quantisation error** (noise shaping).

*3) Perturbations ahead of the integrators are not:* An input-referred difference δ between the channels (analog mismatch, added to the modulator input $u$ of channel 2) and the first integrator's thermal noise $\sigma_\mathrm{i}$ **are integrated twice before the quantiser and are not shaped**. A constant δ acts at every clock: the trajectories diverge until the decision sequences are shifted and the bits uncorrelated.

*4) A zero input does not discriminate:* With $u = 0$ the loop idles in a short repetitive pattern whose phase is set by the first decisions and the smallest perturbation, so **two channels disagree on about half of their bits whatever the cause**.

*5) The independence value:* Two fully decorrelated channels give independent bits, each a one with probability $(1 + u)/2$; they differ with probability $(1 - u^2)/2$ per clock, $(1 - A^2/2)/2$ on average for a sine of amplitude $A$: about 48.4 % for the test signal ($A$ ≈ 0.25 FS) and 50 % for a zero input. **This is an expected value for independent channels, not a bound**: anticorrelated channels can exceed it.

The analysis above therefore predicts the following fractions of differing bits. (i) Zero input: **about 50 % for every combination of effects**. (ii) Same message: at most a few per cent from dither and comparator noise alone; **at the independence value of item 5 once mismatch is present**, for any mismatch above a small threshold; an intermediate value from integrator noise, whose level this first-order argument does not predict.

The runs use the behavioural model of Section V-C (`dsm2`) with `DEFAULTS` and dither at the comparator, comparator noise $\sigma_\mathrm{c} = 10^{-3}$ on in every run, and two inputs: zero ($u = 0$) and the same message (the test sine through the TIA and buffer). Table 6 lists the other settings.

### B. Transmitted bits behind the energy table (Fig. 4(b) of [1])

The energy table of [1] compares the S-DSM with a conventional design, a decimating ADC followed by the Trivium cipher [5], for sending the same information. Both designs send 20 Nyquist samples. The number of bits each sends per reading (Nyquist sample) is **set by the architecture alone**:
* *S-DSM:* one encrypted bit per modulator clock, OSR (the oversampling ratio) bits per reading.
* *ADC + Trivium:* 16-bit words at SR/NR times the Nyquist rate (SR: the decimated output rate; NR: the Nyquist rate), 16 × SR/NR bits per reading.

If the energies of [1] are set by the bits transmitted, as [1] states, **the ratio of its two energies equals the ratio of the bits**, 16 × (SR/NR) / OSR, whatever the radio. **The bit ratio should therefore equal the energy ratio of [1] in all four configurations**, to the rounding of the table of [1].

The check uses the four configurations of Fig. 4(b) of [1], OSR 50 and 30, each with SR/NR = 4.5 and 5; **its metric is the largest relative difference between the bit ratio and the energy ratio reported in [1]**, over the four configurations. Radio and security power do not enter the comparison, since no radio is designed in this work.

### C. Runs, results and interpretation

The first code cell below defines both experiments; the second runs them, compares them with [1] and draws Fig. 18.

> **Next cell.** 1. Defines `channel_pair_flips`: one input through two modulators that differ in dither seeds, noise streams and an optional input offset, and the fraction (and running fraction) of bits that differ. 2. Defines `channel_experiments()`, which runs it for the zero input and the same message, for the 8 combinations of dither, integrator noise and mismatch and for dither in the feedback, and measures the SNDR for each dither placement. 3. Enters the table of Fig. 4(b) of [1]. The cell only defines functions; no simulation is run.

````python
#@title Fig. 4(b) and 4(c) of [1]: experiment definitions
# ---------- Fig. 4(c) of the paper: two channels of one chip, same key, same input, different dither
PAPER_FIG4C = dict(zero=0.5005, same=0.498)

def channel_pair_flips(P, u, mismatch_fs, seeds2=(0x3A17, 0x6E29)):
    # two channels of one chip: the second has its own dither seeds and noise draws and an input offset mismatch_fs (FS);
    # returns the fraction of differing bits and its running mean
    D1 = dsm2(u, P, noise_seed=P["seed"] + 11)["D"]
    D2 = dsm2(u + mismatch_fs, P, seeds=seeds2, noise_seed=P["seed"] + 12)["D"]
    d = (D1 != D2)
    return float(d.mean()), np.cumsum(d) / np.arange(1, len(d) + 1)

def channel_experiments(P=None, mismatch_pct=0.1, n=LAB_N):
    """Bit-flip probability between two channels, split by what makes them differ."""
    P = {**DEFAULTS, **(P or {})}
    noise = P["int_noise"] if P["int_noise"] > 0 else 0.9e-5
    u_zero = np.full(n, s2d(tia(0.0, P["rtia"]), P["buffer"]))
    u_same = s2d(tia(sensor_current({**P, "sig": "sine"}, n), P["rtia"]), P["buffer"])
    # every combination of dither, integrator noise and input mismatch (each on or off), for a zero input and for the
    # test sine; then all on with the dither moved into the feedback path
    rows = []
    for dith in (True, False):
        for nz in (noise, 0.0):
            for mm in (mismatch_pct / 100, 0.0):
                Q = {**P, "dither": dith, "int_noise": nz, "dither_at": "comparator"}
                z, _ = channel_pair_flips(Q, u_zero, mm); s, tr = channel_pair_flips(Q, u_same, mm)
                rows.append(dict(dither=dith, noise=nz, mismatch=mm * 100, at="comparator", zero=z, same=s, trace=tr))
    Q = {**P, "dither": True, "int_noise": noise, "dither_at": "feedback"}
    z, _ = channel_pair_flips(Q, u_zero, mismatch_pct / 100); s, tr = channel_pair_flips(Q, u_same, mismatch_pct / 100)
    rows.append(dict(dither=True, noise=noise, mismatch=mismatch_pct, at="feedback", zero=z, same=s, trace=tr))
    # the SNDR of a full frame with the dither at the comparator and in the feedback path
    def sndr_at(at):
        Q = {**P, "int_noise": noise, "dither_at": at, "sig": "sine"}
        u = s2d(tia(sensor_current(Q, N_DSM), Q["rtia"]), Q["buffer"])
        return sndr(decode(dsm2(u, Q, noise_seed=Q["seed"])["D"]), Q["bin"])
    # independence value: two channels whose bits are independent, each with P(1) = (1 + u)/2, differ with
    # probability mean((1 - u^2)/2), i.e. (1 - A^2/2)/2 for a sine of amplitude A
    indep = dict(zero=float(np.mean((1 - u_zero ** 2) / 2)), same=float(np.mean((1 - u_same ** 2) / 2)),
                 amp=float(np.abs(u_same).max()))
    # mismatch sweep of the same-message test, all effects on, four independent seed sets per value
    Q = {**P, "dither": True, "int_noise": noise, "dither_at": "comparator"}
    dsweep = []
    for d in (0.0, 0.003, 0.01, 0.03, 0.1, 0.3):
        v = [float(np.mean(dsm2(u_same, Q, noise_seed=Q["seed"] + 11 + 100 * k)["D"]
                           != dsm2(u_same + d / 100, Q, seeds=(0x3A17 + 97 * k, 0x6E29 + 89 * k),
                                   noise_seed=Q["seed"] + 12 + 100 * k)["D"])) for k in range(4)]
        dsweep.append((d, v))
    return dict(rows=rows, noise=noise, mismatch=mismatch_pct, indep=indep, dsweep=dsweep,
                sndr=dict(comparator=sndr_at("comparator"), feedback=sndr_at("feedback")))

# ---------- Fig. 4(b): energy for the same information (radio 3.1 mW at 200 kb/s, 20 Nyquist samples)
PAPER_FIG4A_UW = dict(fe=321, tia=271, s2d=250, dsm=510, sdsm=24, trivium=220)
# Fig. 4(b) of [1], per row: OSR, OSR / (SR/NR), SR/NR, and the energies [1] reports for the decimate-then-Trivium design
# and for the S-DSM; the energy of a design: the radio bits it sends for NYQ readings at RADIO_J_PER_BIT, plus its security
# block's power over the information time T_INFO
PAPER_FIG4B = [(50, 11.1, 4.5, 22.35, 15.5), (50, 10, 5, 24.82, 15.5), (30, 6.67, 4.5, 22.36, 9.3), (30, 6, 5, 24.84, 9.3)]
RADIO_J_PER_BIT, T_INFO, NYQ = 3.1e-3 / 200e3, 100e-6, 20

def energy_sdsm(osr, p_sec_uw, extra_bits=0.0):        # the S-DSM sends OSR bits per Nyquist sample
    return (osr * NYQ + extra_bits) * RADIO_J_PER_BIT + p_sec_uw * 1e-6 * T_INFO

def energy_words(srnr, p_sec_uw, bits=16):             # a decimating ADC + cipher sends 16-bit words
    return bits * srnr * NYQ * RADIO_J_PER_BIT + p_sec_uw * 1e-6 * T_INFO
````

**Key insights.**

* *Agreement with Fig. 4(c) of [1] (first result panel, Table 7).* With all effects on, the model gives 50.0 % of bits differing for the zero input (measured 50.05 %) and 48.2 % for the same message (measured 49.8 %): 1.6 percentage points below the measurement for the same message. For the zero input, all nine rows of Table 7 give 47.9–54.6 %, within 5 percentage points of 50 %, including the row with no dither, no integrator noise and no mismatch (49.9 %): prediction (i) holds, and the zero-input test cannot distinguish the effects.
* *Same message, decomposed (second result panel, Fig. 18(a)).* Mismatch alone, without dither and integrator noise, gives 48.2 %; adding noise or dither changes this by at most 0.3 percentage points (48.2–48.5 %). Integrator thermal noise without mismatch gives 26.2 % (26.6 % with dither), dither alone 4.8 % and comparator noise alone 0.3 %. Over time, the two runs with mismatch rise to about 47–48 % within the first thousand clocks and end at 48.2 % and 48.5 %; the run with integrator noise and no mismatch rises more slowly, to about 21 % at the thousandth clock, stays between 21 % and 30 % thereafter and ends at 26.6 %; the run with dither alone stays below about 16 % after the first 100 clocks and settles near 5 %. Prediction (ii) holds for the effects at the quantiser and for mismatch; integrator noise alone reaches about half the independence value.
* *Mismatch sweep (third result panel).* The independence value for the test sine (amplitude 0.249 FS) is 48.4 %. Without mismatch the four seed sets give 24.7 % on average (23.9–26.6 %); 0.003 % FS gives 45.7 %; from 0.01 % FS upwards every value lies between 47.4 % and 48.9 %, with means of 48.3–48.5 %, at the independence value whatever δ is. As prediction (ii) states, above about 0.01 % FS the result does not depend on the assumed mismatch: **the 48.2 % of the first result panel is the independence value, not a fitted value**. The measured 49.8 % would be the independence value for a sine of amplitude 0.089 FS.
* *Transmitted bits (fourth result panel, Table 8, Fig. 18(b)).* At OSR 50 the S-DSM sends 50 bits per reading and the ADC with Trivium 72 or 80; at OSR 30, 30 against 72 or 80. The bit ratios, 1.44, 1.60, 2.40 and 2.67, agree with the energy ratios of [1], 1.44, 1.60, 2.40 and 2.67, within 0.18 %, inside the rounding of the table of [1], as predicted. This confirms the bit counts behind the energy table of [1], not a measurement: Section X compares the proposed design by its transmitted bits, which the architecture sets, and claims no radio energy of its own (Section X-E uses the radio of [1] only as a labelled example).

* *A consistency check, not a validation.* No parameter is fitted (δ is assumed, and the result does not depend on it), but neither comparison discriminates between models: the zero input gives about 50 % for any effect, and the same-message value saturates at the independence value, so it confirms that the channels are decorrelated, not how strongly. **The 1.6-point gap could be explained by a smaller test amplitude in [1]**, which [1] does not state; a deficiency of the model cannot be excluded.
* *With the dither as modelled, the model needs a further effect to reach the measured disagreement.* [1] attributes the disagreement of its channels to the dither, and its chip measurements are the stronger evidence. The model's dither is simplified (a discrete-time loop without excess loop delay, the ladder ratio of 0.978 to 1.022 read as ±0.022 of full scale at the comparator input), and the loop corrects most of the decisions it changes. **This is a statement about the model, not about the chip of [1]:** the security of [1] is not attributed to mismatch rather than to dither, and the channel offset serves only as the model's means of reproducing the measured decorrelation. Section IX-C sweeps the dither amplitude: in the model, only a dither amplitude $a_\mathrm{d} \ge 0.2$ protects at every replica offset (one seed per grid point). Dither in the feedback (first result panel, last row of Table 7), with all effects on, gives 47.9 % (zero input) and 47.2 % (same message), no more than mismatch already gives, but the modulator's SNDR decreases from 60.3 dB (dither at the comparator) to 39.5 dB; this 21 dB loss is consistent with the choice of [1] to dither at the comparator.
* *Consequence for the security evaluation.* Full decorrelation does not protect the original design against the same-key test: the fraction of differing bits settles at $(1 - A^2/2)/2$ and reveals the amplitude of the reading (A5, Section IX-A). The resistance to the replica attack depends on how strongly the victim's and the replica's bits decorrelate, through dither and analog differences together, which Section VIII sweeps; key search, Berlekamp-Massey and frame forgery do not depend on them.

**The check runs on assumed circuit parameters**. The two circuits whose non-idealities decide the security, the modulator's comparator and the SRAM cell of the PUF, are simulated at transistor level in Section VII, and their values replace the assumptions.

# **VII. Circuit-level simulation in SKY130**
---
The model is consistent with the measurements of [1] (Section VI), but some of its settings describe circuits rather than algorithms: the comparator's input-referred noise and offset (`comp_noise` and a per-chip offset) and the SRAM cell's noise-to-mismatch ratio, temperature coefficient and bias (`puf_sigma`, `puf_tc`, `puf_bias`), so far illustrative, with a placeholder code strength t = 40. This section answers which values these settings take in a real process. It determines them with SKY130 transistor-level models, as schematics, as extracted layouts and across process, voltage and temperature (PVT), because the security of both designs rests on them: analog differences between two chips ahead of the quantiser decide whether their bits agree (Section VI), and the PUF's noise-to-mismatch ratio sets how many fingerprint bits flip at each power-up, hence the code strength, the key-failure probability and the remaining secret bits. All circuit numbers of the abstract are obtained here.

Each circuit is simulated in ngspice with the PDK's device, noise and mismatch models, **first as a schematic, then as an extracted layout**. Subsection A installs one fixed, open-source tool set, so that every circuit number traces to the same tools and versions. B measures the StrongARM comparator's input-referred noise $\sigma_\mathrm{n}$, offset spread $\sigma_\mathrm{os}$ and energy per decision and expresses noise and offset in the model's units, after a transient that shows the decision process node by node and verifies that the comparator resolves small and large inputs of both signs, one decision per clock cycle. C determines how reliably a SKY130 6T SRAM cell powers up to the same value at three temperatures and expresses the result in the three settings of the cell model of Section V-D: the noise-to-mismatch ratio $s_{25}$ at 25 °C, its temperature coefficient $a$ and the bias $P(1)$, the probability that a cell powers up to 1. D draws, verifies and extracts both cells and simulates them again, to find whether the layout changes the values of B and C and to obtain the values the model is fitted to; it then compares the transient behaviour of both cells before and after extraction on identical stimuli, to show where the layout parasitics act and that they do not change any decision, and shows that the transistor-level results do not depend on the simulator's time step or tolerance. E replaces the model's illustrative circuit settings with the post-layout values and sizes the BCH code from the simulated PUF parameters. F measures, on recorded grids, how the extracted comparator's noise, offset and energy and the extracted SRAM cell's power-up statistics change across PVT, and which mechanisms cause the changes: it separates the dynamic and leakage components of the comparator's energy across the 45 PVT conditions, separates the SRAM-PUF errors into noise and reproducible flips with a closed-form law tested against the recorded Monte Carlo of all five corners, tests the rotation of the latch's mismatch at the ff and fs corners against a competing explanation, and checks whether the error rates depend on the 1 µs supply ramp of every recorded run. G estimates, from a recorded run with an assumed BTI law because the PDK has no ageing model, how ten years of NBTI on the pull-ups change the extracted cell's error rate, with the power-up value held in the array and with its inverse written back.

> **Role of the circuits.** The StrongARM latch and the 6T SRAM cell are standard topologies, simulated here at block level in SKY130; Section VIII carries their values to the complete sensor.

### B. The StrongARM comparator

The quantiser, clocked at 10 MHz [1], is a StrongARM latch [7] (Fig. 19). While the clock φ is low, the reset devices M8–M11 precharge the internal nodes $d_\mathrm{p}$, $d_\mathrm{n}$ and the outputs $\mathrm{out}_\mathrm{p}$, $\mathrm{out}_\mathrm{n}$ to the supply voltage $V_\mathrm{DD}$. When φ rises, the tail M7 turns on and the input pair M1/M2 discharges $d_\mathrm{p}$ and $d_\mathrm{n}$, of capacitance $C_\mathrm{P}$, with currents $I_\mathrm{D} \pm g_m \Delta V_\mathrm{in}/2$ ($\Delta V_\mathrm{in}$: the differential input; $I_\mathrm{D}$, $g_m$: drain current and transconductance of each input device) for an integration time $T_\mathrm{int}$, until the nodes have fallen by about a threshold voltage $V_\mathrm{TH}$ and the cross-coupled latch M3–M6 turns on and regenerates the difference to the rails, as $v_0\,e^{t/\tau}$ with $\tau \approx C_\mathrm{out}/g_{m,\mathrm{latch}}$ ($v_0$: the difference when regeneration starts; $t$: the time since then; $\tau$: the regeneration time constant; $C_\mathrm{out}$: the output-node capacitance; $g_{m,\mathrm{latch}}$: the transconductance of a latch device). For these device sizes $\tau$ = 36 ps (the slope of the simulated decision delay against $\ln \Delta V_\mathrm{in}$, `dev/spice_comparator/RESULTS.md`), so even a 10 µV input reaches half-swing 0.62 ns after the clock edge, far inside the 50 ns half-period: **metastability does not limit this comparator, and its accuracy is set by noise and offset alone.** Two inverters load both outputs equally ahead of the NOR SR latch, which holds the decision q through the reset phase.

During integration each input device adds channel noise of one-sided power spectral density $S_i = 4kT\gamma g_m$ to its node ($k$: Boltzmann's constant, $T$: absolute temperature, $\gamma$: channel-noise factor, 2/3 for a long-channel device). The signal deposits a charge $g_m \Delta V_\mathrm{in} T_\mathrm{int}$ and the noise a charge whose variance grows with $T_\mathrm{int}$; referred to the input with $T_\mathrm{int} \approx C_\mathrm{P} V_\mathrm{TH}/I_\mathrm{D}$, this gives the first-order result [7]

$$\sigma_\mathrm{n}^2 \approx \frac{4kT\gamma}{g_m T_\mathrm{int}} = \frac{kT}{C_\mathrm{P}} \cdot \frac{4\gamma\,(I_\mathrm{D}/g_m)}{V_\mathrm{TH}},$$

the kT/C noise of the integrating node scaled by a bias factor, which rises as $\sqrt{T}$. With $C_\mathrm{P}$ of a few femtofarads ($\sqrt{kT/C_\mathrm{P}}$ = 0.91 mV at 5 fF and 300 K), $I_\mathrm{D}/g_m$ ≈ 0.1 V and $V_\mathrm{TH}$ ≈ 0.5 V, the factor is about 0.5. The latch adds noise only after the integration gain $g_m T_\mathrm{int}/C_\mathrm{P}$, so its share is small. The analysis above therefore predicts **$\sigma_\mathrm{n}$ between a few tenths of a millivolt and about 1 mV, almost all of it from the input pair**. For these devices the PDK's model gives $\gamma$ near 2 rather than 2/3, and $V_\mathrm{TH}$ with body effect is about 0.9 V. Gaussian noise makes the probability of deciding 1 a probit curve, $P(q = 1) = \Phi((\Delta V_\mathrm{in} - \mu)/\sigma_\mathrm{n})$ ($\Phi$: standard normal distribution function, $\mu$: systematic offset), whose fit to many decisions measures $\sigma_\mathrm{n}$ and $\mu$. Only white channel noise is included: the input pair's 1/f noise (corner near 22 MHz; about 95 µV rms over the 10 Hz–100 kHz band, `dev/spice_comparator/RESULTS.md`) changes little within a clock period and acts as a slowly varying offset of about a sixth of $\sigma_\mathrm{n}$ and 2 % of $\sigma_\mathrm{os}$.

By Pelgrom's law a device's threshold varies with $\sigma(V_\mathrm{T}) = A/\sqrt{WL}$, a pair's difference with $\sqrt{2}\,A/\sqrt{WL}$ ($A$: process constant; $W$, $L$: width and length). The input pair's mismatch appears directly at the input and the latch's is divided by the integration gain, so the input devices are drawn long (W/L = 4/0.5 µm, WL = 2 µm²): with the PDK's nMOS coefficient $A$ = 3.356 mV·µm, $\sigma(V_\mathrm{T})$ = 2.4 mV per device and $\sqrt{2} \times 2.4$ = 3.4 mV for the pair. The PDK also varies the subthreshold offset and oxide thickness, and the latch adds its share, so the analysis above predicts **$\sigma_\mathrm{os}$ of the order of 4–5 mV**, an order of magnitude above the noise: a constant of each chip, Gaussian over chips, to first order independent of temperature. The energy per decision is $C_\mathrm{sw} V_\mathrm{DD}^2$ ($C_\mathrm{sw}$: the capacitance charged per cycle). *Model units:* the model decides $y = 1$ (its comparator output bit, with $x_2$ the second integrator's state) when $x_2 + \text{offset} + \texttt{comp\_noise}\cdot\mathcal{N}(0, 1) \ge 0$ (Section V-C), with $x_2 = 1$ mapped to $V_\mathrm{unit} = V_\mathrm{DD}/4$ = 0.45 V differential at 1.8 V, which keeps the second integrator within the swing of a 1.8 V OTA when $x_2$ clips at $\pm 2.5$ (±0.56 V on each output around 0.9 V). Hence `comp_noise` = $\sigma_\mathrm{n}/V_\mathrm{unit}$, and the per-chip offset has standard deviation $\sigma_\mathrm{os}/V_\mathrm{unit}$.

The comparator is simulated live at transistor level, as the schematic netlist of Fig. 19: SKY130 tt, 27 °C, $V_\mathrm{DD}$ = 1.8 V, input common mode 0.9 V, an ideal 10 MHz clock with 100 ps edges. Device sizes (W/L in µm): input pair M1/M2 4/0.5, tail M7 2/0.15, latch nMOS M3/M4 1/0.15 and pMOS M5/M6 2/0.15, reset devices M8–M11 1/0.15, followed by inverter buffers and the NOR SR latch on a separate supply, measured apart from the core. Every transistor carries the junction areas and perimeters of a single-finger layout with 0.29 µm of diffusion (without them the models contain no junction capacitance); the schematic assumes 1 fF of wiring on $d_\mathrm{p}$, $d_\mathrm{n}$, $\mathrm{out}_\mathrm{p}$ and $\mathrm{out}_\mathrm{n}$ (Section VII-D replaces it with the extracted values) and a 5 fF load on each output of the SR latch. Breakpoints hold the time step at 20 ps around every rising clock edge, **because a larger step during regeneration can place the latch on the wrong side**. The measurements are set up as follows:

* *Transient.* Six decisions without mismatch or noise (Fig. 20).
* *Noise.* (Fig. 22) ngspice's transient analysis contains no device noise, so it is injected in three steps: (i) a noiseless transient at $\Delta V_\mathrm{in}$ = +0.5 mV records the terminal voltages of the 11 core transistors on a 20 ps grid around the third rising edge (the first cycle is not in periodic steady state); (ii) at each grid point, a copy of each device biased at those voltages gives the PDK's channel thermal-noise PSD from one `.noise` analysis at 1 GHz, where the 1/f noise is a few per cent of the thermal noise, averaged over the two devices of each symmetric pair; (iii) each core transistor receives an independent drain–source noise current with samples of variance $S/(2 \cdot 20\ \mathrm{ps})$ ($S$: that device's noise PSD), linearly interpolated, from 0.2 ns before to 0.5 ns after each rising edge. On a test RC circuit driven by its resistor's $4kT/R$ noise (resistance $R$, capacitance $C$) through the same sources, the rms node voltage is 0.874 ± 0.044 of $\sqrt{kT/C}$ for $RC$ = 35 ps and 0.950 ± 0.044 for $RC$ = 150 ps at this grid (0.939 and 0.977 at a 10 ps grid; `dev/spice_comparator/RESULTS.md`): if the decision nodes behaved like these, **$\sigma_\mathrm{n}$ would be underestimated by about 5–13 %**; in an earlier standalone characterisation, refining the grid to 10 and 5 ps raised $\sigma_\mathrm{n}$ by 1 and 2 %, within its standard error (both checks at tt and 27 °C). A pilot (9 inputs within ±2 mV, 20 decisions each) estimates $\sigma_\mathrm{n}$; the main run applies eleven inputs over ±2.5 times that estimate, 100 decisions each (1,100 in all, independent noise per decision, one comparator per netlist, inputs shuffled), and the probit curve is fitted by maximum likelihood, with standard errors from the likelihood's curvature. Two further runs inject the noise of the input pair only (40 decisions per input) and of the latch, tail and reset devices only (eleven inputs within ±0.4 mV, 30 decisions each); independent sources add in power, so their variances should sum to $\sigma_\mathrm{n}^2$. The first-order formula takes $g_m$, $I_\mathrm{D}$, $C_\mathrm{P}$ (from the discharge slope of $d_\mathrm{p}$) and $T_\mathrm{int}$ (until the latch nMOS turns on) of the input device from a noiseless transient, with $\gamma$ = 2/3 and with the PDK's $\gamma$.
* *Offset.* (Fig. 24) 96 Monte Carlo comparators with the PDK's mismatch on the threshold voltage, oxide thickness and subthreshold offset of every device (4 per netlist, one mismatch seed per netlist, noise off). Each finds its threshold with a successive-approximation search of ideal behavioural elements inside the transient: 16 steps, the first 16 mV and each 1.8 times smaller than the last (the final step 2.4 µV); a ratio below 2 makes the search redundant, so a wrong early decision is corrected, and limits it to ±36 mV. Fig. 25(b) traces this search clock by clock for four instances, and Fig. 25(a) collects the thresholds of all 96 into the offset distribution with its fitted Gaussian.
* *Energy.* With the transient testbench of Fig. 20, the supply charge over one clock period with an input of ±1 mV alternating, so that the output toggles in every cycle (the worst case), for the core, the buffers and SR latch, and the clock; the first cycle is discarded.
* *Statistics.* All seeds are fixed. The quick statistics used live give $\sigma_\mathrm{n}$ to about ±5 % (one standard error) and $\sigma_\mathrm{os}$ to −12 %/+17 % (95 % interval of a standard deviation from 96 samples); `quick=False` in the noise cell takes about four times as many decisions and Monte Carlo instances.

The module builds the SKY130 netlists, runs ngspice, injects and fits the noise, searches the offsets, measures the energy, maps the results onto the model and plots them; it also holds the verification functions cited in the text.

````python
%%writefile comparator_spice.py
"""
comparator_spice.py -- StrongARM latch comparator in SkyWater SKY130 (ngspice),
characterised for the behavioural model of a secure 2nd-order 1-bit CT delta-sigma
sensor (ESSERC'25-style: the StrongARM samples x2 at 10 MHz, bits XORed with a keystream).

Everything a notebook cell needs is in this one file: netlist generation, running
ngspice in batch mode, and the measurements

  * transient of a few decisions             -> run_transient_demo(), plot_transient() (clock and q, one decision in detail)
  * input-referred noise sigma_n (probit)     -> measure_noise(),      plot_noise()
  * offset sigma_os (Monte Carlo mismatch)    -> measure_offset(),     plot_offset()
  * energy / decision and delay vs dVin       -> measure_energy(), measure_delay(), plot_delay()
  * mapping to the behavioural model          -> map_to_model()

Only the standard library + numpy + scipy + matplotlib are used.  No paths are
hard-coded: the SKY130 model library (".../libs.tech/ngspice/sky130.lib.spice") and
the ngspice binary are passed in through a config dict (see default_config()).

ngspice compatibility: plain SPICE + .control blocks in batch mode (ngspice -b);
tested with ngspice-41 (conda-forge, what Colab gets) and ngspice-46.  The work
directory gets a .spiceinit with
    set ngbehavior=hsa      (needed to read the SKY130 libraries)
    set ng_nomodcheck       (skip model-parameter checks, faster)
    set num_threads=1       (one OpenMP thread per ngspice; we run processes in parallel)
ngspice reads .spiceinit from the current directory, so every run uses cwd=workdir.
"""

import contextlib
import json
import math
import os
import re
import shutil
import subprocess
import threading
import time
from concurrent.futures import ThreadPoolExecutor

import numpy as np
from scipy import optimize, stats

KB = 1.380649e-23          # Boltzmann constant [J/K]
Q_E = 1.602176634e-19      # elementary charge [C]


# ----------------------------------------------------------------------------------
# Configuration
# ----------------------------------------------------------------------------------

# Device sizes in um (SKY130 1.8 V devices).  Justification (see RESULTS.md):
#   M1/M2  input pair      4/0.5  : L > Lmin cuts Vth mismatch (Pelgrom ~ 1/sqrt(WL)) and
#                                   1/f noise; W sets gm.  Load on the integrator ~ 5 fF.
#   M7     tail            2/0.15 : switch, sets the peak integration current (~0.3 mA).
#   M3/M4  latch NMOS      1/0.15 : minimum-ish, fast regeneration, small load on dp/dn.
#   M5/M6  latch PMOS      2/0.15 : ~2x NMOS for the mobility ratio.
#   M8-M11 reset PMOS      1/0.15 : precharge outp/outn/dp/dn to VDD in < 1 ns.
#   buffers / NOR SR latch  : small digital gates on a separate supply (vddd).
DEFAULT_SIZES = {
    "M1": (4.0, 0.5), "M2": (4.0, 0.5),        # input pair (nfet)
    "M3": (1.0, 0.15), "M4": (1.0, 0.15),      # latch nfet
    "M5": (2.0, 0.15), "M6": (2.0, 0.15),      # latch pfet
    "M7": (2.0, 0.15),                         # tail nfet (clocked)
    "M8": (1.0, 0.15), "M9": (1.0, 0.15),      # reset pfet on outn / outp
    "M10": (1.0, 0.15), "M11": (1.0, 0.15),    # reset pfet on dp / dn
    "INVN": (0.5, 0.15), "INVP": (1.0, 0.15),  # output buffers (outn->sn, outp->sp)
    "NORN": (0.5, 0.15), "NORP": (2.0, 0.15),  # NOR SR latch
}


def default_config(lib_path, ngspice_bin="ngspice", workdir="spice_work", quick=True):
    """Return the config dict used by every function.

    lib_path    : path to SKY130 ".../libs.tech/ngspice/sky130.lib.spice"
    ngspice_bin : ngspice executable
    workdir     : directory for netlists / outputs (created)
    quick       : True = 'quick' statistics (default, ~1-2 min), False = 'full'
    """
    cfg = dict(
        lib_path=lib_path, ngspice_bin=ngspice_bin, workdir=workdir,
        corner="tt",            # process corner: tt / ff / ss / sf / fs
        model_mode="slim",      # 'slim' = include only nfet/pfet_01v8 files (fast load), 'full' = .lib
        temp_c=27.0,            # simulation temperature (ngspice default)
        vdd=1.8, vcm=0.9,       # supply and input common mode [V]
        f_clk=10e6,             # comparator clock [Hz]
        t_edge=100e-12,         # clock rise / fall time [s]
        t_first=20e-9,          # first rising clock edge [s]
        c_par=1e-15,            # wiring capacitance added on dp, dn, outp, outn [F]
        c_load=5e-15,           # load on q, qb (next gate, e.g. the keystream XOR) [F]
        sizes=dict(DEFAULT_SIZES),
        # --- noise ---
        noise_nt=20e-12,        # spacing of the injected noise samples [s] (5/10/20 ps agree)
        noise_win=(-0.2e-9, 0.5e-9),   # noise window around each rising edge [s] (vs +1 ns: same)
        noise_psd_freq=1e9,     # frequency at which the PDK channel-noise PSD is read [Hz]
        # --- time-step control around the decisions (see make_netlist) ---
        fine_win=(-0.1e-9, 1.0e-9),
        fine_dt=20e-12,         # 5 / 10 / 20 ps give identical thresholds (to 0.1 uV)
        trtol=7,                # ngspice default truncation-error factor
        # --- run control ---
        n_jobs=max(1, min(4, len(os.sched_getaffinity(0)) or 1)),
        seed=1,
        timeout=1800,
    )
    if quick:
        cfg.update(noise_levels=11, noise_per_level=100, noise_pilot_per_level=20,
                   offset_n=96)
    else:
        cfg.update(noise_levels=15, noise_per_level=300, noise_pilot_per_level=40,
                   offset_n=400,
                   noise_nt=10e-12, noise_win=(-0.3e-9, 0.7e-9))   # finer noise grid
    return cfg


# ----------------------------------------------------------------------------------
# ngspice plumbing
# ----------------------------------------------------------------------------------

SPICEINIT = """* written by comparator_spice.py
set ngbehavior=hsa
set ng_nomodcheck
set num_threads=1
"""


def write_spiceinit(workdir):
    """Write the .spiceinit that ngspice reads from its working directory."""
    os.makedirs(workdir, exist_ok=True)
    with open(os.path.join(workdir, ".spiceinit"), "w") as f:
        f.write(SPICEINIT)


def ngspice_version(ngspice_bin):
    out = subprocess.run([ngspice_bin, "--version"], capture_output=True, text=True).stdout
    m = re.search(r"ngspice-([\w.+-]+)", out)
    return m.group(0) if m else out.strip()



# One core per ngspice process. Several single-threaded ngspice processes that share a set of cores ran 3.5 times slower
# on a two-core VM than the same processes pinned one per core (the scheduler kept moving them between cores); each run
# therefore takes a free core from this pool for its duration (taskset), which is harmless where the effect is absent.
_CORE_LOCK = threading.Lock()
_FREE_CORES = sorted(os.sched_getaffinity(0))


def _pinned(cmd):
    """Context manager: (command prefixed with taskset to one free core, or unchanged if none is free)."""
    @contextlib.contextmanager
    def cm():
        with _CORE_LOCK:
            core = _FREE_CORES.pop(0) if _FREE_CORES and shutil.which("taskset") else None
        try:
            yield (["taskset", "-c", str(core)] + cmd) if core is not None else cmd
        finally:
            if core is not None:
                with _CORE_LOCK:
                    _FREE_CORES.append(core)
    return cm()


def run_ngspice(ngspice_bin, netlist, workdir, name="run", timeout=1800):
    """Write `netlist` to workdir/<name>.cir and run `ngspice -b` in workdir.

    Returns the log text.  Raises RuntimeError if ngspice fails or prints an error.
    """
    write_spiceinit(workdir)
    cir = os.path.join(workdir, name + ".cir")
    with open(cir, "w") as f:
        f.write(netlist)
    env = dict(os.environ, OMP_NUM_THREADS="1")
    with _pinned([ngspice_bin, "-b", name + ".cir"]) as cmd:
        p = subprocess.run(cmd, cwd=workdir, env=env, capture_output=True, text=True, timeout=timeout)
    log = p.stdout + p.stderr
    with open(os.path.join(workdir, name + ".log"), "w") as f:
        f.write(log)
    bad = [l for l in log.splitlines()
           if re.search(r"^\s*(Error|ERROR)|fatal|aborted|singular matrix|Timestep too small|no such vector|Could not find", l)]
    if p.returncode != 0 or bad:
        raise RuntimeError(f"ngspice run '{name}' failed (rc={p.returncode}):\n"
                           + "\n".join(bad[:20]) + "\n...see " + name + ".log")
    return log


def run_many(cfg, jobs):
    """Run several (netlist, name) jobs in parallel ngspice processes (cfg['n_jobs'])."""
    def one(job):
        net, name = job
        return run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, cfg["timeout"])
    with ThreadPoolExecutor(max_workers=cfg["n_jobs"]) as ex:
        return list(ex.map(one, jobs))


def read_wrdata(path):
    """Read a file written by `wrdata` with `set wr_singlescale` and `set wr_vecnames`."""
    with open(path) as f:
        names = f.readline().split()
    data = np.loadtxt(path, skiprows=1, ndmin=2)
    if data.shape[0] < 1 or data.shape[1] != len(names):
        raise RuntimeError(f"{path}: {data.shape[0]} rows of {data.shape[1]} columns for {len(names)} vectors (no or truncated data)")
    return {n: data[:, i] for i, n in enumerate(names)}


def read_ascii_raw(path):
    """Minimal reader for ngspice ASCII rawfiles (set filetype=ascii; write f.raw).
    Returns a list of (plotname, {vector: real ndarray})."""
    lines = open(path).read().splitlines()
    plots, i = [], 0
    while i < len(lines):
        if not lines[i].startswith("Title:"):
            i += 1
            continue
        hdr = {}
        while not lines[i].startswith("Variables:"):
            k, _, v = lines[i].partition(":")
            hdr[k.strip()] = v.strip()
            i += 1
        nv, npt = int(hdr["No. Variables"]), int(hdr["No. Points"])
        names = [lines[i + 1 + j].split()[1] for j in range(nv)]
        i += 1 + nv + 1                       # skip "Values:"
        toks = []
        while len(toks) < npt * (nv + 1) and i < len(lines):
            toks += lines[i].split()
            i += 1
        vals = [float(t.split(",")[0]) for t in toks]      # complex "re,im" -> re
        arr = np.array(vals).reshape(npt, nv + 1)[:, 1:]
        plots.append((hdr.get("Plotname", ""), dict(zip(names, arr.T))))
    return plots


# ----------------------------------------------------------------------------------
# SKY130 model include
# ----------------------------------------------------------------------------------

def model_include(cfg, mismatch=False):
    """Netlist lines that load the SKY130 nfet_01v8 / pfet_01v8 models.

    'full' mode:  .lib "<sky130.lib.spice>" tt   (or tt_mm with mismatch)  -- parses the
                  whole PDK (~8 s per ngspice start).
    'slim' mode:  reads corners/<corner>.spice of the same PDK and includes only the
                  nfet_01v8 / pfet_01v8 model + mismatch files it lists, plus
                  parameters/lod.spice and '.option scale=1.0u' -- exactly the files the
                  full .lib uses for these two devices, but loads in ~0.05 s.
                  demo.py checks that both modes give identical device currents.
    Mismatch: SKY130 puts AGAUSS(0,1,1)*slope/sqrt(W*L*mult) on vth0, toxe and voff of
    every model bin, gated by MC_MM_SWITCH; ngspice draws it per device instance.
    """
    corner = cfg["corner"]
    lib = os.path.abspath(cfg["lib_path"])
    if cfg.get("model_mode", "slim") == "full":
        return [f'.lib "{lib}" {corner}_mm' if mismatch else f'.lib "{lib}" {corner}']
    # slim: from the corner file, only the .include lines of the nfet/pfet_01v8 model and mismatch files (at least four:
    # model and mismatch for each device), plus the option and parameter file the full library sets for them
    base = os.path.dirname(lib)
    corner_file = os.path.join(base, "corners", corner + ".spice")
    incs = []
    for line in open(corner_file):
        m = re.match(r'\s*\.include\s+"?([^"\s]+)"?', line, re.I)
        if m and re.search(r"sky130_fd_pr__[np]fet_01v8__", m.group(1)):
            incs.append(os.path.normpath(os.path.join(os.path.dirname(corner_file), m.group(1))))
    if len(incs) < 4:
        raise RuntimeError(f"could not find nfet/pfet_01v8 includes in {corner_file}")
    for p in incs:
        if not os.path.exists(p):
            raise FileNotFoundError(f"{p} missing - is libs.ref/sky130_fd_pr installed?")
    out = ["* SKY130 nfet_01v8/pfet_01v8 only (slim include of corner '%s')" % corner,
           ".option scale=1.0u",
           ".param mc_mm_switch=%d mc_pr_switch=0" % (1 if mismatch else 0),
           f'.include "{os.path.join(base, "parameters", "lod.spice")}"']
    out += [f'.include "{p}"' for p in incs]
    return out


# ----------------------------------------------------------------------------------
# Netlist generation
# ----------------------------------------------------------------------------------

# The StrongARM core.  Node names are local; flat_comparator() prefixes them per instance.
#   name   type  drain  gate   source  bulk
CORE = [
    ("M1", "n", "dp", "inp", "tail", "0"),       # input pair: inp > inn discharges dp faster
    ("M2", "n", "dn", "inn", "tail", "0"),
    ("M3", "n", "outn", "outp", "dp", "0"),      # cross-coupled NMOS
    ("M4", "n", "outp", "outn", "dn", "0"),
    ("M5", "p", "outn", "outp", "vdd", "vdd"),   # cross-coupled PMOS
    ("M6", "p", "outp", "outn", "vdd", "vdd"),
    ("M7", "n", "tail", "clk", "0", "0"),        # clocked tail
    ("M8", "p", "outn", "clk", "vdd", "vdd"),    # reset / precharge to VDD while clk = 0
    ("M9", "p", "outp", "clk", "vdd", "vdd"),
    ("M10", "p", "dp", "clk", "vdd", "vdd"),
    ("M11", "p", "dn", "clk", "vdd", "vdd"),
]
# Output stage on its own supply (vddd): buffers give outp/outn identical loads whatever
# the SR-latch state (an SR latch driven directly from outp/outn showed ~1 mV hysteresis),
#   sn = not(outn), sp = not(outp);  q = nor(sp, qb), qb = nor(sn, q)
#   -> q = 1 when V(inp) > V(inn).
DIGITAL = [
    ("MI1P", "p", "sn", "outn", "vddd", "vddd", "INVP"), ("MI1N", "n", "sn", "outn", "0", "0", "INVN"),
    ("MI2P", "p", "sp", "outp", "vddd", "vddd", "INVP"), ("MI2N", "n", "sp", "outp", "0", "0", "INVN"),
    # q = nor(sp, qb)
    ("MR1P", "p", "r1m", "sp", "vddd", "vddd", "NORP"), ("MR2P", "p", "q", "qb", "r1m", "vddd", "NORP"),
    ("MR1N", "n", "q", "sp", "0", "0", "NORN"), ("MR2N", "n", "q", "qb", "0", "0", "NORN"),
    # qb = nor(sn, q)
    ("MR3P", "p", "r2m", "sn", "vddd", "vddd", "NORP"), ("MR4P", "p", "qb", "q", "r2m", "vddd", "NORP"),
    ("MR3N", "n", "qb", "sn", "0", "0", "NORN"), ("MR4N", "n", "qb", "q", "0", "0", "NORN"),
]
GLOBAL_NODES = {"0", "vdd", "vddd", "clk"}
NOISY_DEVICES = [d[0] for d in CORE]     # devices that get injected channel noise


def mos_line(name, typ, d, g, s, b, w, l):
    """One SKY130 transistor (subcircuit call, W/L in um).  Drain/source junction area
    and perimeter follow the usual single-finger layout (0.29 um diffusion extension),
    otherwise the models have no junction capacitance at all."""
    model = "sky130_fd_pr__nfet_01v8" if typ == "n" else "sky130_fd_pr__pfet_01v8"
    ad = w * 0.29
    pd = 2 * (w + 0.29)
    return (f"X{name} {d} {g} {s} {b} {model} W={w:g} L={l:g} nf=1 "
            f"ad={ad:.4g} as={ad:.4g} pd={pd:.4g} ps={pd:.4g} nrd={0.29 / w:.4g} nrs={0.29 / w:.4g}")


def flat_comparator(cfg, p, with_latch=True):
    """Flattened netlist text of one comparator instance with node/device prefix `p`.
    Its ports are p_inp, p_inn (inputs), p_q, p_qb (outputs) and the global nodes.
    with_latch=False drops the NOR SR latch (the statistics runs read the decision from
    outp/outn at the end of the evaluation phase; the buffers that load outp/outn stay)."""
    sz = cfg["sizes"]
    n = lambda x: x if x in GLOBAL_NODES else f"{p}_{x}"
    out = [f"* ---- comparator {p}"]
    lay = cfg.get("layout")              # comparator_layout.postlayout(): the extracted core
    if lay:
        # extracted core: drawn source/drain geometry, and the extracted wiring capacitances in place of c_par;
        # same device names and order as the schematic, so the mismatch draws and the noise sources are unchanged
        for d in lay["devices"]:
            model = f"sky130_fd_pr__{d['kind']}_01v8"
            out.append(f"X{d['name']}_{p} {n(d['d'])} {n(d['g'])} {n(d['s'])} {n(d['b'])} {model} W={d['w']:g} L={d['l']:g} "
                       f"nf=1 ad={d['ad']} as={d['as_']} pd={d['pd']} ps={d['ps']} nrd={0.29 / d['w']:.4g} nrs={0.29 / d['w']:.4g}")
    for name, typ, d, g, s, b in ([] if lay else CORE):
        w, l = sz[name]
        out.append(mos_line(f"{name}_{p}", typ, n(d), n(g), n(s), n(b), w, l))
    for name, typ, d, g, s, b, key in DIGITAL:
        if not with_latch and name.startswith("MR"):
            continue
        w, l = sz[key]
        out.append(mos_line(f"{name}_{p}", typ, n(d), n(g), n(s), n(b), w, l))
    if lay:
        out += [f"Cx{i}_{p} {n(a)} {n(b)} {v:.6g}" for i, (a, b, v) in enumerate(lay["caps"]) if a != b]
    else:
        for node in ("dp", "dn", "outp", "outn"):
            out.append(f"Cw_{node}_{p} {n(node)} 0 {cfg['c_par']:.4g}")
    if with_latch:
        out.append(f"Cl_q_{p} {n('q')} 0 {cfg['c_load']:.4g}")
        out.append(f"Cl_qb_{p} {n('qb')} 0 {cfg['c_load']:.4g}")
    return out


def device_path(cfg, p, name):
    """ngspice name of the BSIM4 device inside the SKY130 subcircuit of transistor name_p."""
    typ = dict((d[0], d[1]) for d in CORE)[name]
    model = "sky130_fd_pr__nfet_01v8" if typ == "n" else "sky130_fd_pr__pfet_01v8"
    return f"m.x{name.lower()}_{p}.m{model}"


def make_netlist(cfg, instances, t_stop, title="StrongARM comparator", mismatch=False,
                 seed=None, saves=(), wrfile="out.dat", tmax=None, extra=(), inputs=None,
                 with_latch=True, fine_edges=None):
    """Assemble a complete ngspice netlist.

    instances : list of instance prefixes, e.g. ["c0", "c1"]
    inputs    : {prefix: list-of-lines} that drive p_inp / p_inn (default: DC at vcm)
    extra     : additional element lines (noise sources, SAR logic, ...)
    saves     : vectors to save and write with wrdata to `wrfile`
    fine_edges: rising-edge times around which the time step is forced small (breakpoints
                every cfg['fine_dt'] in cfg['fine_win']).  Needed: with a large tmax, ngspice's
                truncation-error control lets the step grow to ~100 ps during regeneration
                and the latch can land on the wrong side (seen in the ramp check).
    """
    T = 1.0 / cfg["f_clk"]
    tr = cfg["t_edge"]
    L = [f"* {title}"]
    L += model_include(cfg, mismatch)
    if seed is not None:
        L.append(f".option seed={int(seed)}")      # reproducible AGAUSS mismatch draws
    L.append(f".temp {cfg['temp_c']}")
    L.append(f".option trtol={cfg['trtol']}")
    L.append(f"Vdd vdd 0 {cfg['vdd']}")
    L.append(f"Vddd vddd 0 {cfg['vdd']}")
    # 10 MHz clock, 50 % duty: eval (clk = 1) for T/2 starting at t_first + k*T
    L.append(f"Vclk clk 0 PULSE(0 {cfg['vdd']} {cfg['t_first']:.6g} {tr:.4g} {tr:.4g} "
             f"{T / 2 - tr:.6g} {T:.6g})")
    for p in instances:
        if inputs and p in inputs:
            L += inputs[p]
        else:
            L.append(f"V{p}p {p}_inp 0 {cfg['vcm']}")
            L.append(f"V{p}n {p}_inn 0 {cfg['vcm']}")
        L += flat_comparator(cfg, p, with_latch)
    L += list(extra)
    if fine_edges is not None and len(fine_edges):
        L += fine_step_source(cfg, fine_edges)
    if tmax is None:
        tmax = T / 50
    if cfg.get("spice_options"):
        L.append(".options " + cfg["spice_options"])  # the convergence check of Section VII-D tightens the tolerances
    L += [".control",
          "set wr_singlescale", "set wr_vecnames", "option numdgt=12",
          "save " + " ".join(saves),
          f"tran {tmax:.4g} {t_stop:.6g} 0 {tmax:.4g}",
          f"wrdata {wrfile} " + " ".join(saves),
          ".endc", ".end"]
    return "\n".join(L) + "\n"


def fine_step_source(cfg, edges, name="fine"):
    """Dummy PWL source whose corners (breakpoints) force small time steps around edges."""
    w0, w1 = cfg["fine_win"]
    dt = cfg["fine_dt"]
    tg = w0 + dt * np.arange(int(round((w1 - w0) / dt)) + 1)
    tt = (np.asarray(edges)[:, None] + tg[None, :]).ravel()
    return [f"V{name}brk {name}brk 0 PWL(0 0 " + " ".join(f"{x:.12e} 0" for x in tt) + ")",
            f"R{name}brk {name}brk 0 1e3"]


def edge_times(cfg, n_cycles):
    """Start times of the rising clock edges."""
    return cfg["t_first"] + np.arange(n_cycles) / cfg["f_clk"]


def sample_at(t, y, ts):
    """y at the times ts, linearly interpolated; a time outside the simulated waveform is an error, not its end value."""
    ts = np.asarray(ts, float)
    if ts.size and (ts.min() < t[0] - 1e-15 or ts.max() > t[-1] + 1e-15):
        raise RuntimeError(f"sample time {ts.max():.6g} s outside the simulated waveform ({t[0]:.6g} to {t[-1]:.6g} s)")
    return np.interp(ts, t, y)


# ----------------------------------------------------------------------------------
# 1) Transient of a few decisions
# ----------------------------------------------------------------------------------

def run_transient_demo(cfg, dvins=(1e-3, -1e-3, 0.2e-3, -0.2e-3, 20e-3, -20e-3), name="tran_demo"):
    """Transient of len(dvins) decisions at f_clk with a PWL differential input that changes
    in the middle of each reset phase.  Returns the waveforms and the decisions."""
    T = 1 / cfg["f_clk"]
    n = len(dvins)
    te = edge_times(cfg, n)
    # inp = vcm + dv/2, inn = vcm - dv/2; dv_k is applied from T/4 before edge k
    # (middle of the reset phase) with a 1 ns ramp
    pts_p = [(0.0, cfg["vcm"] + dvins[0] / 2)]
    pts_n = [(0.0, cfg["vcm"] - dvins[0] / 2)]
    for k in range(1, n):
        ta = te[k] - T / 4
        pts_p += [(ta, cfg["vcm"] + dvins[k - 1] / 2), (ta + 1e-9, cfg["vcm"] + dvins[k] / 2)]
        pts_n += [(ta, cfg["vcm"] - dvins[k - 1] / 2), (ta + 1e-9, cfg["vcm"] - dvins[k] / 2)]
    pwl = lambda pts: "PWL(" + " ".join(f"{a:.6g} {b:.9g}" for a, b in pts) + ")"
    inputs = {"c0": [f"Vc0p c0_inp 0 {pwl(pts_p)}", f"Vc0n c0_inn 0 {pwl(pts_n)}"]}
    saves = ["v(clk)", "v(c0_inp)", "v(c0_inn)", "v(c0_dp)", "v(c0_dn)", "v(c0_tail)",
             "v(c0_outp)", "v(c0_outn)", "v(c0_q)", "v(c0_qb)", "i(Vdd)", "i(Vddd)", "i(Vclk)"]
    t_stop = te[-1] + T
    c = dict(cfg, fine_dt=10e-12, fine_win=(-0.3e-9, 1.6e-9))     # fine steps for the zoom plot
    net = make_netlist(c, ["c0"], t_stop, "transient demo", saves=saves,
                       wrfile=name + ".dat", tmax=0.5e-9, inputs=inputs, fine_edges=te)
    run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, cfg["timeout"])
    w = read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
    q = sample_at(w["time"], w["v(c0_q)"], te + T * 0.45)
    return dict(wave=w, edges=te, dvins=np.array(dvins), decisions=(q > cfg["vdd"] / 2).astype(int))


# ----------------------------------------------------------------------------------
# 2) Input-referred noise
# ----------------------------------------------------------------------------------
#
# Method (ngspice's transient analysis has no device noise):
#   a) Run a noiseless transient (dVin = +0.5 mV, 3 cycles) and record the terminal
#      voltages of the 11 StrongARM transistors on a grid t_j (spacing NT = 10 ps) in a
#      window from 0.3 ns before to 0.7 ns after the rising clock edge of the 3rd cycle
#      (steady-state periodic operation; the 1st cycle starts from the DC operating point
#      and is different).
#   b) For every device and every t_j, bias an identical test transistor at those DC
#      terminal voltages and let ngspice's .noise analysis return the PDK's own channel
#      thermal-noise PSD S_id(t_j) [A^2/Hz] (BSIM4 tnoimod=1 as in the SKY130 cards, read at
#      1 GHz where 1/f noise is negligible).  This is the "frozen operating point" picture
#      of cyclostationary noise: S_id follows the device from triode (reset) through
#      saturation (integration) into regeneration.  For symmetric pairs both devices get
#      the pair average so that the injected noise is unbiased.
#   c) In the noisy runs every one of the 11 transistors gets a current source from drain
#      to source carrying i_n(t_j) = sqrt(S_id(t_j) / (2 NT)) * N(0,1), independent
#      per device, per decision and per comparator instance, linearly interpolated (PWL)
#      and zero outside the window.  A sequence of independent samples of variance s^2
#      spaced NT has a one-sided PSD of 2 s^2 NT at f << 1/NT, so i_n has PSD S_id.
#      check_noise_injection() verifies this on an RC (kT/C).  Outside the window the
#      decision is already latched (or the comparator is being reset), so noise there
#      cannot change the decision; restricting it keeps the run at the real 10 MHz clock
#      without forcing 10 ps steps through the whole 100 ns period.
#   d) Many identical instances, each with its own dVin, run in one netlist; the decision
#      of each cycle is read from q.  P(q = 1) vs dVin is fitted with a Gaussian CDF
#      (probit, maximum likelihood): P = Phi((dVin - mu) / sigma_n).
#   Not included: 1/f noise (shows up as a slowly varying offset, see RESULTS.md),
#   gate-induced noise, the body-resistance noise of the BSIM4 rbody network (a few %),
#   and noise of the buffers / SR latch (they only see rail-to-rail signals).

PAIRS = [("M1", "M2"), ("M3", "M4"), ("M5", "M6"), ("M8", "M9"), ("M10", "M11")]
CORE_NODES = ["inp", "inn", "dp", "dn", "outp", "outn", "tail"]


def noise_grid(cfg):
    nt = cfg["noise_nt"]
    w0, w1 = cfg["noise_win"]
    n = int(round((w1 - w0) / nt))
    return w0 + nt * np.arange(n + 1)            # times relative to the rising edge


def reference_trajectory(cfg, dv=0.5e-3, name="ref_traj"):
    """Noiseless transient; returns terminal voltages of every core device on the noise
    grid of the 3rd decision: {dev: dict(vg, vd, vs, vb)} and the grid."""
    T = 1 / cfg["f_clk"]
    te = edge_times(cfg, 3)
    inputs = {"c0": [f"Vc0p c0_inp 0 {cfg['vcm'] + dv / 2:.9g}",
                     f"Vc0n c0_inn 0 {cfg['vcm'] - dv / 2:.9g}"]}
    saves = ["v(clk)"] + [f"v(c0_{x})" for x in CORE_NODES]
    net = make_netlist(cfg, ["c0"], te[-1] + T / 2, "reference trajectory", saves=saves,
                       wrfile=name + ".dat", tmax=cfg["noise_nt"] / 2, inputs=inputs)
    run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, cfg["timeout"])
    w = read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
    tg = noise_grid(cfg)
    ts = te[-1] + tg
    volt = {"0": np.zeros_like(ts), "vdd": np.full_like(ts, cfg["vdd"]),
            "clk": sample_at(w["time"], w["v(clk)"], ts)}
    for x in CORE_NODES:
        volt[x] = sample_at(w["time"], w[f"v(c0_{x})"], ts)
    bias = {}
    for name_, typ, d, g, s, b in CORE:
        bias[name_] = dict(vg=volt[g], vd=volt[d], vs=volt[s], vb=volt[b], typ=typ)
    return dict(t=tg, bias=bias, wave=w, edge=te[-1])


def channel_noise_psd(cfg, traj, name="psd_lookup"):
    """PDK channel thermal-noise PSD S_id(t) [A^2/Hz] of every core device along the
    trajectory, from one ngspice .noise analysis on biased copies of the devices."""
    sz = cfg["sizes"]
    L = ["* channel noise PSD lookup"] + model_include(cfg, False) + [f".temp {cfg['temp_c']}"]
    keys, prev = [], "0"
    k = 0
    for dev, bz in traj["bias"].items():
        w, l = sz[dev]
        for j in range(len(traj["t"])):
            tag = f"{dev.lower()}_{j}"
            ac = " ac 1" if k == 0 else ""
            L.append(f"VTg_{tag} g_{tag} 0 {bz['vg'][j]:.9g}{ac}")
            L.append(f"VTd_{tag} d_{tag} 0 {bz['vd'][j]:.9g}")
            L.append(f"VTs_{tag} s_{tag} 0 {bz['vs'][j]:.9g}")
            L.append(f"VTb_{tag} b_{tag} 0 {bz['vb'][j]:.9g}")
            L.append(mos_line(f"T_{tag}", bz["typ"], f"d_{tag}", f"g_{tag}", f"s_{tag}",
                              f"b_{tag}", w, l))
            # transresistance of 1 ohm: V(out) = sum of all drain currents, so every
            # device's channel noise reaches the output with unity gain
            L.append(f"H_{tag} o_{tag} {prev} VTd_{tag} 1")
            prev = f"o_{tag}"
            keys.append((dev, j, tag))
            k += 1
    first = keys[0][2]
    f = cfg["noise_psd_freq"]
    L += [".control", "set filetype=ascii",
          # the trailing "1" (pts_per_summary) makes ngspice keep per-device spectra
          f"noise v({prev}) VTg_{first} lin 1 {f:.4g} {f:.4g} 1",
          "setplot noise1", f"write {name}.raw", ".endc", ".end"]
    run_ngspice(cfg["ngspice_bin"], "\n".join(L) + "\n", cfg["workdir"], name, cfg["timeout"])
    plots = read_ascii_raw(os.path.join(cfg["workdir"], name + ".raw"))
    vec = [d for p, d in plots if "Noise Spectral" in p][0]
    S = {dev: np.zeros(len(traj["t"])) for dev in traj["bias"]}
    for dev, j, tag in keys:
        model = "nfet" if traj["bias"][dev]["typ"] == "n" else "pfet"
        S[dev][j] = vec[f"onoise.m.xt_{tag}.msky130_fd_pr__{model}_01v8.id"][0] ** 2
    for a, b in PAIRS:                     # symmetric pairs: use the pair average
        m = 0.5 * (S[a] + S[b])
        S[a], S[b] = m.copy(), m.copy()
    return S


def noise_sources(cfg, p, S, edges, rng):
    """Drain-source noise currents of instance p for the given rising edges.

    Each source is a B-source  I = pwl(time, t0, i0, t1, i1, ...)  (ngspice evaluates the
    B-source pwl() by bisection, a PWL current source scans its point list linearly, which
    makes long runs quadratic).  B-sources do not set breakpoints, so one dummy PWL voltage
    source with the same time grid forces the simulator onto every noise sample."""
    tg = noise_grid(cfg)
    nt = cfg["noise_nt"]
    nodes = {d[0]: (d[2], d[4]) for d in CORE}
    n = lambda x: x if x in GLOBAL_NODES else f"{p}_{x}"
    tt = (np.asarray(edges)[:, None] + tg[None, :]).ravel()
    tstr = [f"{x:.12e}" for x in tt]          # 12 digits: ps resolution at us times
    out = []
    for dev in NOISY_DEVICES:
        amp = np.sqrt(S[dev] / (2 * nt))
        amp[0] = amp[-1] = 0.0                       # zero outside the windows
        vals = (amp[None, :] * rng.standard_normal((len(edges), len(tg)))).ravel()
        body = ", ".join(f"{a}, {v:.4e}" for a, v in zip(tstr, vals))
        d, s_ = nodes[dev]
        # a second zero point after the last window: ngspice's B-source pwl() extrapolates linearly beyond its last point,
        # and the slope into the final zero would otherwise grow into a large current that overturns the last decision
        out.append(f"Bn_{dev}_{p} {n(d)} {n(s_)} I = pwl(time, 0, 0, {body}, {tt[-1] + 1e-9:.12e}, 0)")
    return out, tstr


def run_noise_job(cfg, dv_seq, S, seed, name, noise=True):
    """Netlist of ONE nominal comparator (no mismatch, no SR latch) for len(dv_seq)+1 clock
    cycles; cycle 0 is a dummy (it starts from the DC operating point), cycle k >= 1 sees
    dVin = dv_seq[k-1] (the input steps in the middle of the preceding reset phase).
    One instance per netlist on purpose: instances in one netlist share the time grid, and
    each instance's regeneration forces small steps on all of them."""
    # differential input as two PWL sources around vcm: the level of cycle k is applied a quarter period before its
    # rising edge (mid reset phase), with a 1 ns step
    T = 1 / cfg["f_clk"]
    n_cyc = len(dv_seq) + 1
    te = edge_times(cfg, n_cyc)
    dv_all = np.concatenate([[dv_seq[0]], dv_seq])
    pts_p, pts_n = [(0.0, cfg["vcm"] + dv_all[0] / 2)], [(0.0, cfg["vcm"] - dv_all[0] / 2)]
    for k in range(1, n_cyc):
        ta = te[k] - T / 4
        pts_p += [(ta, cfg["vcm"] + dv_all[k - 1] / 2), (ta + 1e-9, cfg["vcm"] + dv_all[k] / 2)]
        pts_n += [(ta, cfg["vcm"] - dv_all[k - 1] / 2), (ta + 1e-9, cfg["vcm"] - dv_all[k] / 2)]
    pwl = lambda pts: "PWL(" + " ".join(f"{a:.12e} {b:.12g}" for a, b in pts) + ")"
    inputs = {"c0": [f"Vc0p c0_inp 0 {pwl(pts_p)}", f"Vc0n c0_inn 0 {pwl(pts_n)}"]}
    extra = []
    if noise:
        rng = np.random.default_rng(seed)
        # the noise currents, and a dummy PWL voltage source whose corners put a breakpoint on every noise sample
        src, tstr = noise_sources(cfg, "c0", S, te[1:], rng)
        extra = src + ["Vbrk brk 0 PWL(0 0 " + " ".join(f"{x} 0" for x in tstr) + ")",
                       "Rbrk brk 0 1e3"]
    saves = ["v(c0_outp)", "v(c0_outn)"]
    net = make_netlist(cfg, ["c0"], te[-1] + T / 2, "noise job", saves=saves,
                       wrfile=name + ".dat", tmax=cfg.get("noise_tmax", 5e-9), inputs=inputs, extra=extra,
                       with_latch=False, fine_edges=te)
    return net, te


def read_decisions(cfg, name, te, prefix="c0"):
    """Decision of every cycle: 1 if outn fell (V(inp) > V(inn)), read at 0.45 T."""
    T = 1 / cfg["f_clk"]
    w = read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
    ts = te + 0.45 * T
    vp = sample_at(w["time"], w[f"v({prefix}_outp)"], ts)
    vn = sample_at(w["time"], w[f"v({prefix}_outn)"], ts)
    return (vp > vn).astype(int)


def noise_campaign(cfg, S, dvs, n_per_level, seed, tag, cycles_per_job=50):
    """n_per_level decisions at every level in dvs, spread (shuffled) over parallel jobs."""
    # every level n_per_level times, in random order, cut into jobs of cycles_per_job decisions run in parallel
    rng = np.random.default_rng(seed)
    seq = np.repeat(np.arange(len(dvs)), n_per_level)
    rng.shuffle(seq)
    jobs, meta = [], []
    for j, i0 in enumerate(range(0, len(seq), cycles_per_job)):
        idx = seq[i0:i0 + cycles_per_job]
        nm = f"noise_{tag}_{j}"
        net, te = run_noise_job(cfg, dvs[idx], S, seed * 7919 + j, nm)
        jobs.append((net, nm))
        meta.append((nm, te, idx))
    run_many(cfg, jobs)
    # ones and trials per level, gathered from all jobs
    ones = np.zeros(len(dvs))
    tot = np.zeros(len(dvs))
    for nm, te, idx in meta:
        dec = read_decisions(cfg, nm, te)[1:]              # drop dummy cycle 0
        np.add.at(ones, idx, dec)
        np.add.at(tot, idx, 1)
    return ones, tot


def probit_fit(dv, k, n):
    """Maximum-likelihood fit of P(1) = Phi((dv - mu)/sigma) to k ones out of n trials.
    Returns mu, sigma and their standard errors (from the observed Fisher information)."""
    dv, k, n = map(np.asarray, (dv, k, n))

    def nll(par):
        mu, ls = par
        p = stats.norm.cdf((dv - mu) / np.exp(ls))
        p = np.clip(p, 1e-12, 1 - 1e-12)
        return -np.sum(k * np.log(p) + (n - k) * np.log(1 - p))

    # start: the level closest to P = 0.5 for mu and a quarter of the span for sigma (fitted as ln sigma, kept positive)
    pk = k / n
    s0 = max(np.ptp(dv) / 4, 1e-6)
    mid = dv[np.argmin(np.abs(pk - 0.5))]
    r = optimize.minimize(nll, [mid, np.log(s0)], method="Nelder-Mead",
                          options=dict(xatol=1e-9, fatol=1e-9, maxiter=4000))
    mu, ls = r.x
    # numerical Hessian in (mu, sigma)
    sig = np.exp(ls)
    f = lambda m, s: nll([m, np.log(s)])
    h = np.array([sig * 1e-2, sig * 1e-2])
    x0 = np.array([mu, sig])
    H = np.zeros((2, 2))
    for i in range(2):
        for j in range(2):
            ei, ej = np.eye(2)[i] * h[i], np.eye(2)[j] * h[j]
            H[i, j] = (f(*(x0 + ei + ej)) - f(*(x0 + ei - ej)) - f(*(x0 - ei + ej))
                       + f(*(x0 - ei - ej))) / (4 * h[i] * h[j])
    # standard errors: the inverse of the observed Fisher information (the Hessian of the negative log-likelihood)
    try:
        cov = np.linalg.inv(H)
        se = np.sqrt(np.diag(cov))
    except np.linalg.LinAlgError:
        se = np.array([np.nan, np.nan])
    return dict(mu=mu, sigma=sig, mu_se=se[0], sigma_se=se[1])


def measure_noise(cfg, sigma_guess=None, levels=None, n_per_level=None, seed=None, log=print):
    """Input-referred noise of the nominal comparator (no mismatch).

    A short pilot (9 levels over +-2 mV) gives sigma_guess unless it is passed in; the
    main run uses `levels` dVin values spanning +-2.5 sigma_guess with `n_per_level`
    decisions each.  Returns dict with sigma (= sigma_n [V]), mu (systematic offset of
    the nominal, symmetric comparator: should be ~0), standard errors and raw counts."""
    t0 = time.time()
    seed = cfg["seed"] if seed is None else seed
    levels = levels or cfg["noise_levels"]
    n_per_level = n_per_level or cfg["noise_per_level"]
    # the channel-noise PSD of every device along the nominal decision trajectory: the amplitude of the injected currents
    traj = reference_trajectory(cfg)
    S = channel_noise_psd(cfg, traj)
    log(f"  channel-noise PSD lookup along the trajectory done ({time.time() - t0:.1f} s)")
    if sigma_guess is None:
        pil = np.linspace(-2e-3, 2e-3, 9)
        k1, n1 = noise_campaign(cfg, S, pil, cfg["noise_pilot_per_level"], seed + 1, "pilot")
        sigma_guess = probit_fit(pil, k1, n1)["sigma"]
        log(f"  pilot: sigma ~ {sigma_guess * 1e3:.3f} mV ({time.time() - t0:.1f} s)")
    dvs = np.linspace(-2.5, 2.5, levels) * sigma_guess
    k, n = noise_campaign(cfg, S, dvs, n_per_level, seed + 2, "main")
    fit = probit_fit(dvs, k, n)
    fit.update(dvin=dvs, ones=k, trials=n, psd=S, traj_t=traj["t"],
               runtime_s=time.time() - t0, sigma_guess=sigma_guess)
    log(f"  sigma_n = {fit['sigma'] * 1e3:.3f} +- {fit['sigma_se'] * 1e3:.3f} mV, "
        f"mu = {fit['mu'] * 1e3:+.3f} +- {fit['mu_se'] * 1e3:.3f} mV, "
        f"{int(n.sum())} decisions ({fit['runtime_s']:.1f} s)")
    return fit


def measure_noise_part(cfg, S, devices, dvs, n_per_level, seed, tag):
    """Input-referred noise when only `devices` carry their channel noise (the PSD of the others set to zero, the same
    netlists otherwise): independent noise sources add in power, so the parts' variances should sum to sigma_n^2.
    Returns the probit fit with its levels and counts, as measure_noise."""
    Sx = {d: (np.asarray(S[d]) if d in devices else np.zeros_like(np.asarray(S[d]))) for d in S}
    k, n = noise_campaign(cfg, Sx, np.asarray(dvs), n_per_level, seed, tag)
    fit = probit_fit(dvs, k, n)
    fit.update(dvin=np.asarray(dvs), ones=k, trials=n, devices=tuple(devices))
    return fit


def check_noise_injection(cfg, r=10e3, c=5e-15, n_cycles=200, name="ktc_check"):
    """Sanity check of the PWL noise injection: a resistor's thermal current noise
    (S = 4kT/R) injected with the same windowed PWL scheme into R||C must give
    var(v) = kT/C at the end of each window.  Returns (measured, expected) rms [V]."""
    kT = KB * (273.15 + cfg["temp_c"])
    tg = noise_grid(cfg)
    S = np.full(len(tg), 4 * kT / r)
    nt = cfg["noise_nt"]
    rng = np.random.default_rng(cfg["seed"] + 77)
    T = 1 / cfg["f_clk"]
    te = edge_times(cfg, n_cycles)
    vals = []
    # the expected value includes the reduction from the linear interpolation:
    #   var = (1/C^2) int h(t)^2 ... evaluated numerically below for the sampled process
    jobs = []
    names = []
    for part in range(0, n_cycles, 20):
        ed = te[:min(20, n_cycles - part)]          # every run starts again at t = 0
        amp = np.sqrt(S / (2 * nt))
        amp[0] = amp[-1] = 0
        z = rng.standard_normal((len(ed), len(tg)))
        pts = " ".join(f"{t_:.12e} {v:.4e}" for k, e in enumerate(ed)
                       for t_, v in zip(e + tg, amp * z[k]))
        nm = f"{name}_{part}"
        net = "\n".join([
            "* kT/C check", f"In 0 x PWL(0 0 {pts})", f"R1 x 0 {r:g}", f"C1 x 0 {c:g}",
            ".control", "set wr_singlescale", "set wr_vecnames", "option numdgt=12",
            "save v(x)", f"tran 1n {ed[-1] + T / 2:.6g} 0 1n", f"wrdata {nm}.dat v(x)",
            ".endc", ".end"]) + "\n"
        jobs.append((net, nm))
        names.append((nm, ed))
    run_many(cfg, jobs)
    t_s = tg[-2]                    # just before the window closes
    for nm, ed in names:
        w = read_wrdata(os.path.join(cfg["workdir"], nm + ".dat"))
        vals += list(sample_at(w["time"], w["v(x)"], ed + t_s))
    vals = np.array(vals)
    return float(np.std(vals)), float(np.sqrt(kT / c)), len(vals)


# ----------------------------------------------------------------------------------
# 3) Offset: Monte Carlo over device mismatch
# ----------------------------------------------------------------------------------
#
# Every comparator instance in a netlist gets its own mismatch draw (SKY130 AGAUSS in the
# model cards, evaluated per device instance when ngspice expands the subcircuits);
# `.option seed=N` makes the draws reproducible (verify_mismatch() checks both claims).
# Each instance finds its own switching threshold with a successive-approximation loop
# built from ideal behavioural elements, all inside the same transient:
#     dV_{k+1} = dV_k - d_k * step_k,  d_k = +1 if the comparator said "inp > inn" else -1
#     step_k = A * r^-(k-1)  (r = 1.8 < 2: redundant, a wrong early decision is corrected)
# implemented as a 1 F "integrator" node p_vd charged by a B-source current during a gate
# pulse late in the evaluation phase (the decision is latched on outp/outn by then; the
# input moving by a few mV cannot flip a regenerated latch), and B-sources that put
# vcm +- V(p_vd)/2 on the comparator inputs.  After n_steps updates V(p_vd) is the
# threshold to within ~A r^-n_steps (< 10 uV).  Noise is off: sigma_os is mismatch only.

def sar_elements(cfg, insts, n_cycles, a0, ratio):
    """Shared gate/step sources and per-instance SAR loop elements."""
    # gate: a pulse 10-40 ns after each rising edge whose area is 1 s, so each clock moves the integrator by exactly one
    # step; step: the size A r^-(k-1) of update k, set 2-3 ns into the preceding reset phase
    T = 1 / cfg["f_clk"]
    te = edge_times(cfg, n_cycles)
    g0, g1, tr = 10e-9, 40e-9, 1e-9              # update window inside the eval phase
    amp = 1.0 / (g1 - g0 - tr)                  # integral of the gate pulse = 1 s
    gp, sp = ["0 0"], ["0 0"]
    for k in range(1, n_cycles):                # cycle 0: dummy, no update
        t = te[k]
        gp += [f"{t + g0:.12e} 0", f"{t + g0 + tr:.12e} {amp:.8g}",
               f"{t + g1 - tr:.12e} {amp:.8g}", f"{t + g1:.12e} 0"]
        step = a0 * ratio ** (-(k - 1))
        sp += [f"{t - T / 2 + 2e-9:.12e} {sp[-1].split()[1]}", f"{t - T / 2 + 3e-9:.12e} {step:.8g}"]
    L = [f"Vgate sar_gate 0 PWL({' '.join(gp)})", f"Vstep sar_step 0 PWL({' '.join(sp)})"]
    inputs = {}
    # per instance: the 1 F integrator node vd, charged by -gate x step x sign(outp - outn) (tanh: a smooth sign), and the
    # comparator inputs at vcm +- vd/2
    for p in insts:
        L += [f"Cint_{p} {p}_vd 0 1", f"Rint_{p} {p}_vd 0 1e15",
              f"Bint_{p} 0 {p}_vd I = -V(sar_gate)*V(sar_step)*tanh(20*(V({p}_outp)-V({p}_outn)))"]
        inputs[p] = [f"B{p}p {p}_inp 0 V = {cfg['vcm']} + V({p}_vd)/2",
                     f"B{p}n {p}_inn 0 V = {cfg['vcm']} - V({p}_vd)/2"]
    return L, inputs, te


def run_offset_job(cfg, n_inst, seed, name, n_steps=16, a0=16e-3, ratio=1.8):
    """Netlist with n_inst mismatched comparators (seeded), each running its own SAR."""
    T = 1 / cfg["f_clk"]
    insts = [f"m{i}" for i in range(n_inst)]
    n_cycles = n_steps + 1
    extra, inputs, te = sar_elements(cfg, insts, n_cycles, a0, ratio)
    saves = [f"v({p}_vd)" for p in insts]
    net = make_netlist(cfg, insts, te[-1] + T / 2, "offset Monte Carlo (SAR)", mismatch=True,
                       seed=seed, saves=saves, wrfile=name + ".dat", tmax=5e-9,
                       inputs=inputs, extra=extra, with_latch=False, fine_edges=te)
    return net, te


def read_offsets(cfg, name, n_inst, te):
    """Threshold of each instance = V(p_vd) after the last update, plus its history."""
    T = 1 / cfg["f_clk"]
    w = read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
    hist = np.array([sample_at(w["time"], w[f"v(m{i}_vd)"], te + 0.47 * T) for i in range(n_inst)])
    return hist[:, -1], hist


def measure_offset(cfg, n_mc=None, inst_per_job=4, seed=None, n_steps=16, log=print):
    """sigma_os from n_mc Monte Carlo instances (mismatch only, corner cfg['corner']).
    Job j uses seed (seed*1000 + j) -> fully reproducible."""
    t0 = time.time()
    seed = cfg["seed"] if seed is None else seed
    n_mc = n_mc or cfg["offset_n"]
    jobs, meta = [], []
    for j, i0 in enumerate(range(0, n_mc, inst_per_job)):
        ni = min(inst_per_job, n_mc - i0)
        nm = f"offset_{j}"
        net, te = run_offset_job(cfg, ni, seed * 1000 + j, nm, n_steps)
        jobs.append((net, nm))
        meta.append((nm, ni, te))
    run_many(cfg, jobs)
    vos, hist = [], []
    for nm, ni, te in meta:
        v, h = read_offsets(cfg, nm, ni, te)
        vos += list(v)
        hist += list(h)
    vos = np.array(vos)
    hist = np.array(hist)
    # the farthest the search can reach is the sum of all steps, a0 r / (r - 1); an offset within 2 % of it is clipped
    a0, ratio = 16e-3, 1.8
    vmax = a0 * ratio / (ratio - 1)
    clipped = int(np.sum(np.abs(vos) > 0.98 * vmax))
    # sigma with its chi-square 95 % interval; Shapiro-Wilk p-value as a check of normality
    sd = float(np.std(vos, ddof=1))
    res = dict(vos=vos, history=hist, sigma=sd, mean=float(np.mean(vos)),
               sigma_ci95=(sd * math.sqrt((len(vos) - 1) / stats.chi2.ppf(0.975, len(vos) - 1)),
                           sd * math.sqrt((len(vos) - 1) / stats.chi2.ppf(0.025, len(vos) - 1))),
               last_step=a0 * ratio ** (-(n_steps - 1)), n=len(vos), clipped=clipped,
               shapiro_p=float(stats.shapiro(vos).pvalue) if len(vos) >= 3 else float("nan"),
               runtime_s=time.time() - t0)
    log(f"  sigma_os = {sd * 1e3:.3f} mV (95% CI {res['sigma_ci95'][0] * 1e3:.2f}-"
        f"{res['sigma_ci95'][1] * 1e3:.2f}), mean {res['mean'] * 1e3:+.3f} mV, N = {len(vos)}, "
        f"{clipped} at the search limit ({res['runtime_s']:.1f} s)")
    return res


def ramp_threshold(cfg, n_inst, seed, name, v_lo=-20e-3, v_hi=20e-3, n_cycles=201):
    """Cross-check of the SAR: the same seeded instances (identical netlist text for the
    comparators, so identical mismatch draws) driven by a slow staircase ramp of dVin, one
    step per clock cycle; threshold = midpoint of the last-0 / first-1 levels."""
    T = 1 / cfg["f_clk"]
    insts = [f"m{i}" for i in range(n_inst)]
    te = edge_times(cfg, n_cycles + 1)
    levels = np.linspace(v_lo, v_hi, n_cycles)
    dv_all = np.concatenate([[levels[0]], levels])
    pts = [(0.0, dv_all[0])]
    for k in range(1, len(te)):
        ta = te[k] - T / 4
        pts += [(ta, dv_all[k - 1]), (ta + 1e-9, dv_all[k])]
    pwl = "PWL(" + " ".join(f"{a:.12e} {b:.12g}" for a, b in pts) + ")"
    # same element types/order as run_offset_job so the AGAUSS sequence is identical:
    # the SAR elements are present but the gate never opens (step = 0)
    extra, _, _ = sar_elements(cfg, insts, 1, 0.0, 1.8)
    extra = [l for l in extra if not l.startswith(("Bint", "Cint", "Rint"))]
    extra += ["Vramp ramp 0 " + pwl]
    inputs = {p: [f"B{p}p {p}_inp 0 V = {cfg['vcm']} + V(ramp)/2",
                  f"B{p}n {p}_inn 0 V = {cfg['vcm']} - V(ramp)/2"] for p in insts}
    saves = [f"v({p}_outp)" for p in insts] + [f"v({p}_outn)" for p in insts]
    net = make_netlist(cfg, insts, te[-1] + T / 2, "offset ramp check", mismatch=True, seed=seed,
                       saves=saves, wrfile=name + ".dat", tmax=5e-9, inputs=inputs,
                       extra=extra, with_latch=False, fine_edges=te)
    run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, cfg["timeout"])
    th = []
    for i in range(n_inst):
        d = read_decisions(cfg, name, te, prefix=f"m{i}")[1:]
        if not d.any() or d.all():
            raise RuntimeError(f"ramp check: instance {i} does not switch within the ramp")
        first1 = int(np.argmax(d > 0))
        if not np.all(d[first1:] == 1):
            raise RuntimeError(f"ramp check: instance {i} switches non-monotonically (no noise is injected)")
        th.append(0.5 * (levels[max(first1 - 1, 0)] + levels[first1]))
    return np.array(th), levels[1] - levels[0]


def verify_mismatch(cfg, n=40, seed=5, name="mm_check"):
    """Checks that (a) every device instance in one netlist gets its own mismatch draw and
    (b) the draws are reproducible with .option seed: n identical nfets (W/L of the input
    pair) at the same bias, run twice with `seed` and once with `seed+1`.
    Returns the drain currents and the derived Vth spread estimate."""
    w, l = cfg["sizes"]["M1"]

    def net(sd):
        L = ["* mismatch check"] + model_include(cfg, True) + [f".option seed={sd}",
                                                                f".temp {cfg['temp_c']}",
                                                                "Vg g 0 0.75", "Vd d 0 0.9"]
        # n identical nfets at the same bias, each with a 0 V source in its drain to measure its current
        for i in range(n):
            L += [f"Vs{i} d s{i} 0", mos_line(f"T{i}", "n", f"s{i}", "g", "0", "0", w, l)]
        L += [".control", "op", "set wr_singlescale", "set wr_vecnames", "option numdgt=12",
              "let gm0 = @m.xt0.msky130_fd_pr__nfet_01v8[gm]",
              f"wrdata {name}_{sd}.dat " + " ".join(f"i(Vs{i})" for i in range(n)) + " gm0",
              ".endc", ".end"]
        return "\n".join(L) + "\n"

    out = {}
    for tag, sd in (("a", seed), ("b", seed), ("c", seed + 1)):
        run_ngspice(cfg["ngspice_bin"], net(sd), cfg["workdir"], f"{name}_{tag}", cfg["timeout"])
        d = read_wrdata(os.path.join(cfg["workdir"], f"{name}_{sd}.dat"))
        out[tag] = np.array([d[f"i(vs{i})"][0] if f"i(vs{i})" in d else d[f"i(Vs{i})"][0]
                             for i in range(n)])
        out["gm"] = float(d["gm0"][0])
    ia, ib, ic = out["a"], out["b"], out["c"]
    # with gm/Id known, sigma(Id)/Id ~ (gm/Id) * sigma(Vth_eff) for small mismatch
    sig_vth_eff = float(np.std(ia / np.mean(ia), ddof=1) / (out["gm"] / np.mean(ia)))
    return dict(n=n, all_distinct=bool(len(np.unique(np.round(ia, 15))) == n),
                same_seed_identical=bool(np.array_equal(ia, ib)),
                other_seed_differs=bool(not np.allclose(ia, ic)),
                rel_spread_id=float(np.std(ia, ddof=1) / np.mean(ia)),
                sigma_vth_eff_mV=sig_vth_eff * 1e3, currents_a=ia, currents_c=ic)


# ----------------------------------------------------------------------------------
# 4) Energy per decision and delay vs dVin
# ----------------------------------------------------------------------------------

def measure_energy(cfg, n_cycles=8, name="energy"):
    """Energy per decision at f_clk for the full comparator (core + buffers + SR latch).

    Two input patterns: 'toggle' (dVin = +-1 mV alternating: q toggles every cycle, worst
    case for the latch) and 'constant' (+1 mV: q never toggles).  Per cycle we report
      core    : -int VDD i(Vdd)  over one period (StrongARM core)
      digital : -int VDD i(Vddd) (buffers + NOR SR latch + 5 fF loads on q, qb)
      clock   : VDD * (charge the ideal clock source delivers while clk is high), i.e. what
                a CMOS clock buffer would draw from VDD to drive the comparator's clock pins.
    Cycle 0 (DC start) is skipped."""
    T = 1 / cfg["f_clk"]
    out = {}
    for pat in ("toggle", "constant"):
        dvs = [1e-3 * ((-1) ** k if pat == "toggle" else 1) for k in range(n_cycles)]
        te = edge_times(cfg, n_cycles)
        pts_p = [(0.0, cfg["vcm"] + dvs[0] / 2)]
        pts_n = [(0.0, cfg["vcm"] - dvs[0] / 2)]
        for k in range(1, n_cycles):
            ta = te[k] - T / 4
            pts_p += [(ta, cfg["vcm"] + dvs[k - 1] / 2), (ta + 1e-9, cfg["vcm"] + dvs[k] / 2)]
            pts_n += [(ta, cfg["vcm"] - dvs[k - 1] / 2), (ta + 1e-9, cfg["vcm"] - dvs[k] / 2)]
        pwl = lambda pts: "PWL(" + " ".join(f"{a:.12e} {b:.12g}" for a, b in pts) + ")"
        inputs = {"c0": [f"Vc0p c0_inp 0 {pwl(pts_p)}", f"Vc0n c0_inn 0 {pwl(pts_n)}"]}
        saves = ["v(clk)", "i(Vdd)", "i(Vddd)", "i(Vclk)", "v(c0_q)"]
        nm = f"{name}_{pat}"
        net = make_netlist(cfg, ["c0"], te[-1] + T - 2e-9, "energy", saves=saves,
                           wrfile=nm + ".dat", tmax=1e-9, inputs=inputs, fine_edges=te)
        run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], nm, cfg["timeout"])
        w = read_wrdata(os.path.join(cfg["workdir"], nm + ".dat"))
        t = w["time"]
        # per cycle after the first: the energy drawn from each supply over one period, and VDD times the charge the
        # clock source delivers while the clock is high
        ec, ed, ek = [], [], []
        for k in range(1, n_cycles):
            m = (t >= te[k] - 1e-9) & (t <= te[k] + T - 2e-9)
            ec.append(-np.trapezoid(cfg["vdd"] * w["i(Vdd)"][m], t[m]))
            ed.append(-np.trapezoid(cfg["vdd"] * w["i(Vddd)"][m], t[m]))
            mh = (t >= te[k] - 1e-12) & (t <= te[k] + T / 2 - 1e-12)
            q_clk = -np.trapezoid(w["i(Vclk)"][mh], t[mh])
            ek.append(cfg["vdd"] * q_clk)
        out[pat] = dict(core=float(np.mean(ec)), digital=float(np.mean(ed)),
                        clock=float(np.mean(ek)),
                        total=float(np.mean(ec) + np.mean(ed) + np.mean(ek)))
    out["power_at_fclk_W"] = out["toggle"]["total"] * cfg["f_clk"]
    return out


def measure_delay(cfg, dvins=None, name="delay"):
    """Clock-to-decision delay vs dVin (noiseless, nominal comparator with SR latch).

    t_latch : clock 50 % -> |V(outp) - V(outn)| = VDD/2 (StrongARM regenerated)
    t_q     : clock 50 % -> q crosses VDD/2 (SR-latch output valid)
    The regeneration time constant tau is the slope of t_latch vs -ln(dVin) for small dVin.
    One netlist per dVin (run in parallel), 3 cycles with dVin = +dv, -dv, +dv so that q
    toggles in the measured 3rd cycle; 5 ps breakpoints around the edges."""
    if dvins is None:
        dvins = np.logspace(-5, -1, 13)          # 10 uV .. 100 mV
    dvins = np.asarray(dvins)
    T = 1 / cfg["f_clk"]
    te = edge_times(cfg, 3)
    c = dict(cfg, fine_dt=5e-12, fine_win=(-0.1e-9, 2.0e-9))

    def pwl(dv, sgn):
        v = lambda k: cfg["vcm"] + sgn * dv / 2 * (1 if k % 2 == 0 else -1)
        pts = [(0.0, v(0))]
        for k in (1, 2):
            ta = te[k] - T / 4
            pts += [(ta, v(k - 1)), (ta + 1e-9, v(k))]
        return "PWL(" + " ".join(f"{a:.12e} {b:.12g}" for a, b in pts) + ")"

    jobs = []
    for i, dv in enumerate(dvins):
        inputs = {"c0": [f"Vc0p c0_inp 0 {pwl(dv, 1)}", f"Vc0n c0_inn 0 {pwl(dv, -1)}"]}
        saves = ["v(clk)", "v(c0_outp)", "v(c0_outn)", "v(c0_q)"]
        nm = f"{name}_{i}"
        jobs.append((make_netlist(c, ["c0"], te[-1] + T / 2, "delay", saves=saves,
                                  wrfile=nm + ".dat", tmax=1e-9, inputs=inputs, fine_edges=te), nm))
    run_many(cfg, jobs)
    half = cfg["vdd"] / 2

    # time of the first upward crossing of lev, linearly interpolated; no crossing is an error, not a number
    def cross(tt, y, lev):
        s_ = y > lev
        i = int(np.argmax(s_))
        if not s_.any() or i == 0:
            raise RuntimeError(f"delay measurement: no upward crossing of {lev:.3g} V in the measured cycle")
        return tt[i - 1] + (lev - y[i - 1]) * (tt[i] - tt[i - 1]) / (y[i] - y[i - 1])

    # in the third cycle: the clock's 50 % point, then the latch outputs' VDD/2 split and q's VDD/2 crossing
    tl, tq = [], []
    for _, nm in jobs:
        w = read_wrdata(os.path.join(cfg["workdir"], nm + ".dat"))
        t = w["time"]
        m = (t >= te[-1] - 0.2e-9) & (t <= te[-1] + 10e-9)
        tt = t[m]
        t_clk = cross(tt, w["v(clk)"][m], half)
        tl.append(cross(tt, w["v(c0_outp)"][m] - w["v(c0_outn)"][m], half) - t_clk)
        tq.append(cross(tt, w["v(c0_q)"][m], half) - t_clk)
    tl, tq = np.array(tl), np.array(tq)
    # tau: minus the slope of t_latch against ln(dVin) over the small inputs (<= 1 mV), where regeneration dominates
    small = (dvins <= 1e-3) & np.isfinite(tl)
    slope = np.polyfit(np.log(dvins[small]), tl[small], 1)[0] if small.sum() >= 2 else np.nan
    return dict(dvin=dvins, t_latch=tl, t_q=tq, tau_regen=float(-slope))


# ----------------------------------------------------------------------------------
# kT/C-style cross-check of sigma_n
# ----------------------------------------------------------------------------------

def noise_estimate(cfg, psd=None, name="est_traj"):
    """First-order analytical estimate of the input-referred noise from the integration
    phase only (Razavi, SSC Magazine 2015; Nuzzo et al., TCAS-I 2008):

        during integration each input device dumps i_n (PSD S = 4kT gamma gm) on its drain
        node C_P for T_int, the signal dumps gm dVin T_int, so
            sigma^2 ~ 4 kT gamma / (gm T_int)   with T_int ~ C_P V_thn / I_D
                   = (kT / C_P) * 4 gamma (I_D/gm) / V_thn
    gm, I_D come from the device at the start of integration (saved in a noiseless
    transient), C_P = I_D / |dV_dp/dt|, T_int = time from the clock edge until dp has
    fallen by V_th of the latch NMOS (when M3 turns on and regeneration starts).
    gamma: 2/3 (long-channel textbook) and gamma_PDK = S_id / (4 kT gm) from the PDK noise
    model at the same instant.  The estimate ignores the latch devices and reset noise,
    so it should come out somewhat below the simulated sigma_n."""
    T = 1 / cfg["f_clk"]
    te = edge_times(cfg, 3)
    dev1 = device_path(cfg, "c0", "M1")
    dev3 = device_path(cfg, "c0", "M3")
    inputs = {"c0": [f"Vc0p c0_inp 0 {cfg['vcm']}", f"Vc0n c0_inn 0 {cfg['vcm']}"]}
    saves = ["v(clk)", "v(c0_dp)", "v(c0_tail)", f"@{dev1}[gm]", f"@{dev1}[id]", f"@{dev3}[vth]"]
    # 2 ps steps only where the waveforms are read (the integration after the third edge), the usual fine steps around
    # the other edges: the same values as a 2 ps limit throughout (within 1 %), in about a twentieth of the time
    fine = fine_step_source(dict(cfg, fine_win=(-0.05e-9, 1.5e-9), fine_dt=2e-12), [te[-1]], name="est")
    net = make_netlist(cfg, ["c0"], te[-1] + 3e-9, "noise estimate", saves=saves,
                       wrfile=name + ".dat", inputs=inputs, extra=fine, fine_edges=te)
    run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, cfg["timeout"])
    w = read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
    t = w["time"]
    m = (t >= te[-1]) & (t <= te[-1] + 2e-9)
    tt, vdp = t[m], w["v(c0_dp)"][m]
    gm = w[f"@{dev1}[gm]"][m]
    idd = np.abs(w[f"@{dev1}[id]"][m])
    vth3_t = np.abs(w[f"@{dev3}[vth]"][m])        # Vth of M3 incl. body effect, vs time
    vdd = cfg["vdd"]
    # instant where dp has dropped by 50 mV: integration well under way
    i1 = np.argmax(vdp < vdd - 0.05)
    i2 = np.argmax(vdp < vdd - 0.10)
    slope = (vdp[i2] - vdp[i1]) / (tt[i2] - tt[i1])
    c_p = float(np.mean(idd[i1:i2 + 1]) / abs(slope))
    g = float(np.mean(gm[i1:i2 + 1]))
    i_d = float(np.mean(idd[i1:i2 + 1]))
    t_clk = te[-1] + cfg["t_edge"] / 2
    i_on = int(np.argmax(vdd - vdp > vth3_t))        # M3 turns on: Vgs3 = VDD - V(dp) > Vth3
    t_int = float(tt[i_on] - t_clk)
    vth3 = float(vth3_t[i_on])
    kT = KB * (273.15 + cfg["temp_c"])
    out = dict(C_P=c_p, gm=g, I_D=i_d, gm_over_id=g / i_d, Vth_latch=vth3, T_int=t_int,
               ktc_rms=math.sqrt(kT / c_p))
    gam = {"gamma=2/3": 2 / 3}
    if psd is not None:
        # PDK gamma at the same instant: S_id(M1) / (4 kT gm), psd = (t_grid, S dict) from
        # measure_noise (same 3rd-cycle trajectory, dVin = 0.5 mV instead of 0)
        tg = psd[0] + te[-1]
        s1 = np.interp(tt[i1:i2 + 1], tg, psd[1]["M1"]).mean()
        gam["gamma_PDK"] = float(s1 / (4 * kT * g))
    for k, gm_ in gam.items():
        out[f"sigma_{k}"] = math.sqrt(4 * kT * gm_ / (g * t_int))
        out[f"value_{k}"] = gm_
    return out


def flicker_estimate(cfg, traj, f_lo=10.0, f_hi=100e3, name="flicker"):
    """Rough size of the 1/f noise that the per-decision model leaves out.

    The input-pair device M1 is biased at its mid-integration operating point (from the
    reference trajectory) and ngspice's .noise gives its gate-referred noise density
    (thermal + 1/f) from 10 Hz to 1 GHz; the pair is sqrt(2) x one device.  Integrated over
    f_lo..f_hi (default: the sensor's 100 kHz signal band at OSR 50) it is a slowly varying
    input offset, not white per-decision noise, and it is upper-bounded here (the device is
    on only half the time; the delta-sigma loop further shapes it like the offset)."""
    tg = traj["t"]
    j = int(np.argmin(np.abs(tg - 0.15e-9)))           # ~mid integration
    bz = traj["bias"]["M1"]
    w, l = cfg["sizes"]["M1"]
    L = ["* flicker estimate"] + model_include(cfg, False) + [f".temp {cfg['temp_c']}",
         f"Vg g 0 {bz['vg'][j]:.9g} ac 1", f"Vd d 0 {bz['vd'][j]:.9g}",
         f"Vs s 0 {bz['vs'][j]:.9g}", mos_line("F1", "n", "d", "g", "s", "0", w, l),
         "H1 out 0 Vd 1",
         ".control", "set filetype=ascii",
         f"noise v(out) Vg dec 20 1 1e9", "setplot noise1", f"write {name}.raw", ".endc", ".end"]
    run_ngspice(cfg["ngspice_bin"], "\n".join(L) + "\n", cfg["workdir"], name, cfg["timeout"])
    vec = [d for p, d in read_ascii_raw(os.path.join(cfg["workdir"], name + ".raw"))
           if "Noise Spectral" in p][0]
    f = vec["frequency"]
    sv = 2 * vec["inoise_spectrum"] ** 2            # pair, V^2/Hz, gate (= input) referred
    th = np.median(sv[f > 1e8])                     # thermal floor
    m = (f >= f_lo) & (f <= f_hi)
    v_band = float(np.sqrt(np.trapezoid(np.maximum(sv[m] - th, 0), f[m])))
    above = np.where(sv > 2 * th)[0]
    f_corner = float(f[above[-1]]) if len(above) else float("nan")
    return dict(f=f, s_vin=sv, s_thermal=float(th), f_corner=f_corner,
                v_flicker_band_rms=v_band, band=(f_lo, f_hi),
                s_vin_1k=float(np.interp(1e3, f, sv)))


# ----------------------------------------------------------------------------------
# 5) Mapping to the behavioural model
# ----------------------------------------------------------------------------------

def map_to_model(sigma_n, sigma_os, vdd=1.8, unit_fraction=0.25, xlim=2.5):
    """Behavioural model: y = 1 if x2 + comp_noise * N(0,1) >= 0, x2 normalised so that
    +-1 is the feedback reference.  The comparator sees dVin = V_unit * x2 (differential).
    Choice: V_unit = unit_fraction * VDD (default VDD/4 = 0.45 V differential).  With the
    model's clip xlim = 2.5 the second integrator then swings up to 2.5 * 0.45 = 1.125 V
    differential = +-0.56 V per output around 0.9 V (0.34 .. 1.46 V), about what a
    two-stage SKY130 OTA can deliver at 1.8 V, so x2 = 1 <-> VDD/4 uses the swing without
    clipping the OTA first.  Offset: a per-chip constant comp_offset ~ N(0, sigma_os/V_unit)
    (decision y = 1 if x2 + comp_offset + comp_noise*N(0,1) >= 0)."""
    v_unit = unit_fraction * vdd
    alt = {}
    for fr in (1 / 8, 1 / 4, 1 / 2):
        alt[f"VDD*{fr:g}"] = dict(V_unit=fr * vdd, comp_noise=sigma_n / (fr * vdd),
                                  comp_offset_sigma=sigma_os / (fr * vdd))
    return dict(V_unit=v_unit, comp_noise=sigma_n / v_unit, comp_offset_sigma=sigma_os / v_unit,
                x2_swing_diff_at_xlim=xlim * v_unit, per_output_swing=xlim * v_unit / 2,
                alternatives=alt)


# ----------------------------------------------------------------------------------
# Model-include check (slim vs full PDK library)
# ----------------------------------------------------------------------------------

def check_model_modes(cfg, name="libcheck"):
    """Same tiny circuit with the 'slim' and the 'full' (.lib sky130.lib.spice tt) include:
    returns the drain currents of an nfet and a pfet in both modes (should be identical)."""
    res = {}
    for mode in ("slim", "full"):
        c = dict(cfg, model_mode=mode)
        L = ["* model include check"] + model_include(c, False) + [
            "Vdd vdd 0 1.8", "Vg g 0 0.8", "Vd d 0 0.9", "Vgp gp 0 1.0", "Vdp dp 0 0.9",
            mos_line("N1", "n", "d", "g", "0", "0", 4.0, 0.5),
            mos_line("P1", "p", "dp", "gp", "vdd", "vdd", 2.0, 0.15),
            ".control", "op", "set wr_singlescale", "set wr_vecnames",
            f"wrdata {name}_{mode}.dat i(Vd) i(Vdp)", ".endc", ".end"]
        t0 = time.time()
        run_ngspice(cfg["ngspice_bin"], "\n".join(L) + "\n", cfg["workdir"], f"{name}_{mode}",
                    cfg["timeout"])
        d = read_wrdata(os.path.join(cfg["workdir"], f"{name}_{mode}.dat"))
        lk = {k.lower(): v for k, v in d.items()}
        res[mode] = dict(id_n=float(lk["i(vd)"][0]), id_p=float(lk["i(vdp)"][0]),
                         load_s=time.time() - t0)
    res["identical"] = bool(np.isclose(res["slim"]["id_n"], res["full"]["id_n"], rtol=1e-9)
                            and np.isclose(res["slim"]["id_p"], res["full"]["id_p"], rtol=1e-9))
    return res


# ----------------------------------------------------------------------------------
# Plots (matplotlib; each returns the Figure).  Sized for an IEEE two-column page
# (7.16 in) or a single column (~3.5-4.8 in); fonts, ticks and grid come from the
# caller's rcParams (the notebook sets an IEEE style), so no font sizes are set here.
# ----------------------------------------------------------------------------------

C1, C2, C3, C4 = "#2a78d6", "#eb6834", "#1b9e77", "#d39200"   # blue, orange (+ 2 spares)
INK, GREY = "#131c28", "#7c8594"
INK2 = GREY


def _style(ax):
    """Kept for compatibility: the look comes from rcParams."""
    ax.set_axisbelow(True)


def _mv(x, fmt="{:+g}"):
    """Format a value in mV with a typographic minus sign."""
    return fmt.format(x).replace("-", "−")


def _save(fig, path):
    if path:
        fig.savefig(path, dpi=300, bbox_inches="tight")


def plot_transient(res, cfg, zoom_cycle=2, path=None):
    """(a) the clock and (b) the latched decision q, stacked on one time axis, with each cycle's input and decision;
    (c)-(f) one decision zoomed (-0.2 .. +1.5 ns): clock and tail, integration nodes, latch outputs, SR-latch outputs."""
    import matplotlib.pyplot as plt
    w, te, dv = res["wave"], res["edges"], res["dvins"]
    t = w["time"] * 1e9
    T = 1 / cfg["f_clk"]
    vdd = cfg["vdd"]
    fig = plt.figure(figsize=(7.16, 6.4), layout="constrained")
    gs = fig.add_gridspec(4, 2, height_ratios=[0.55, 0.9, 1, 1])
    ac = fig.add_subplot(gs[0, :])
    aq = fig.add_subplot(gs[1, :], sharex=ac)
    ac.plot(t, w["v(clk)"], color=GREY, lw=1.3)
    aq.plot(t, w["v(c0_q)"], color=C1, lw=1.6)
    for k, (e, d) in enumerate(zip(te, dv)):
        aq.text((e + T / 2) * 1e9, vdd + 0.25, f"{_mv(d * 1e3)} mV $\\rightarrow$ {res['decisions'][k]}",
                ha="center", va="bottom", color=INK, fontsize=9)
        for a in (ac, aq):
            a.axvline(e * 1e9, color=GREY, lw=0.6, ls=":")
    for a, lab in ((ac, r"$\phi$ (V)"), (aq, "$q$ (V)")):      # node names as in the schematic: clock phi, output q
        a.set_ylabel(lab)
        a.set_yticks([0, vdd / 2, vdd])
    ac.set_ylim(-0.2, vdd + 0.2)
    aq.set_ylim(-0.2, vdd + 0.85)
    ac.tick_params(labelbottom=False)
    ac.set_title(r"(a) Clock $\phi$")
    aq.set_title(f"(b) Latched decision $q$, one per clock cycle at {cfg['f_clk'] / 1e6:g} MHz "
                 r"($\Delta V_\mathrm{in} \rightarrow q$)")
    aq.set_xlim(0, t[-1])
    aq.set_xlabel("Time (ns)")

    e = te[zoom_cycle]
    m = (w["time"] > e - 0.2e-9) & (w["time"] < e + 1.5e-9)
    tz = (w["time"][m] - e) * 1e9
    sub = [fig.add_subplot(gs[2 + i // 2, i % 2]) for i in range(4)]
    for a in sub[1:]:
        a.sharex(sub[0])
        a.sharey(sub[0])
    zl = f"{_mv(dv[zoom_cycle] * 1e3)} mV decision"
    rows = [(f"(c) Clock and tail node ({zl})", [("v(clk)", r"clock $\phi$", GREY, "-"), ("v(c0_tail)", "tail", C1, "-")],
             "center right"),
            ("(d) Integration nodes", [("v(c0_dp)", "$d_p$", C1, "-"), ("v(c0_dn)", "$d_n$", C2, "--")],
             "upper right"),
            ("(e) Latch outputs", [("v(c0_outp)", r"$\mathrm{out}_p$", C1, "-"), ("v(c0_outn)", r"$\mathrm{out}_n$", C2, "--")],
             "center right"),
            ("(f) SR-latch outputs", [("v(c0_q)", "$q$", C1, "-"), ("v(c0_qb)", r"$\bar{q}$", C2, "--")],
             "center left")]
    for i, (a, (title, sig, loc)) in enumerate(zip(sub, rows)):
        for key, lab, col, ls in sig:
            a.plot(tz, w[key][m], color=col, ls=ls, lw=1.5, label=lab)
        a.set_title(title)
        a.legend(loc=loc)
        if i % 2 == 0:
            a.set_ylabel("Voltage (V)")
        else:
            a.tick_params(labelleft=False)
        if i >= 2:
            a.set_xlabel("Time after the rising clock edge (ns)")
        else:
            a.tick_params(labelbottom=False)
    sub[0].set_xlim(-0.2, 1.5)
    sub[0].set_xticks([0, 0.5, 1.0, 1.5])
    sub[0].set_ylim(-0.2, vdd + 0.4)
    sub[0].set_yticks(np.arange(0, vdd + 0.01, 0.6))
    _save(fig, path)
    return fig


def plot_noise(fit, path=None, parts=()):
    """P(q = 1) vs dVin with binomial (Wilson) 95 % intervals and the probit fit; `parts`: (label, fit) of runs in which
    only some devices are noisy (measure_noise_part), drawn with their own markers and line styles."""
    import matplotlib.pyplot as plt
    from matplotlib.legend_handler import HandlerTuple
    def wilson(k, n):
        lo, hi = [], []
        for kk, nn in zip(k, n):              # Wilson interval
            z = 1.96
            c = (kk + z * z / 2) / (nn + z * z)
            h = z * math.sqrt(kk * (nn - kk) / nn + z * z / 4) / (nn + z * z)
            lo.append(c - h)
            hi.append(c + h)
        return np.array(lo), np.array(hi)
    dv, k, n = np.asarray(fit["dvin"]), np.asarray(fit["ones"]), np.asarray(fit["trials"])
    p = k / n
    lo, hi = wilson(k, n)
    fig, ax = plt.subplots(figsize=(4.4, 3.8 if parts else 3.2), layout="constrained")
    x = np.linspace(dv.min() * 1.1, dv.max() * 1.1, 300)
    pts = ax.errorbar(dv * 1e3, p, yerr=[p - lo, hi - p], fmt="o", ms=5,
                      color=C2, mec=C2, mfc="white", mew=1.2, ecolor=C2, elinewidth=1.0, capsize=2.5, zorder=3)
    fl, = ax.plot(x * 1e3, stats.norm.cdf((x - fit["mu"]) / fit["sigma"]), color=C1, lw=1.5, zorder=2)
    sig = lambda f: rf"{f['sigma'] * 1e3:.3f} $\pm$ {f['sigma_se'] * 1e3:.3f} mV"
    if not parts:
        ax.legend([pts, fl], [f"SPICE, {int(n[0])} decisions/level", r"Probit fit, $\sigma_\mathrm{n}$ = " + sig(fit)],
                  loc="upper left")
    else:
        # one entry per run (its points and its fit), below the axes, where no curve passes
        hs, ls_ = [(pts, fl)], [rf"all devices ({int(n[0])} per level): $\sigma_\mathrm{{n}}$ = " + sig(fit)]
        for (lab, f), mk, ls, col in zip(parts, ("s", "^"), ("--", ":"), (C3, INK)):
            dp_, kp, npp = np.asarray(f["dvin"]), np.asarray(f["ones"]), np.asarray(f["trials"])
            m_, = ax.plot(dp_ * 1e3, kp / npp, mk, ms=4, mfc="white", mec=col, mew=1.0, ls="none", zorder=3)
            l_, = ax.plot(x * 1e3, stats.norm.cdf((x - f["mu"]) / f["sigma"]), color=col, lw=1.2, ls=ls, zorder=2)
            hs.append((m_, l_)); ls_.append(f"{lab} only ({int(npp[0])} per level): " + sig(f))
        fig.legend(hs, ls_, loc="outside lower center", ncol=1, handler_map={tuple: HandlerTuple(ndivide=None)},
                   handlelength=3.0)
    ax.set_xlim(x.min() * 1e3, x.max() * 1e3)
    ax.set_ylim(-0.04, 1.04)
    ax.set_xlabel(r"Differential input $\Delta V_{\mathrm{in}}$ (mV)")
    ax.set_ylabel(r"Probability $P(q = 1)$")
    _save(fig, path)
    return fig


def plot_noise_psd(fit, path=None):
    """Injected channel-noise PSD S_id(t) of each device group along the decision."""
    import matplotlib.pyplot as plt
    t = np.asarray(fit.get("traj_t", fit.get("psd_t"))) * 1e12
    S = fit["psd"]
    groups = [("M1", "Input pair M1/M2", C1, "-"), ("M3", "Latch NMOS M3/M4", C2, "--"),
              ("M5", "Latch PMOS M5/M6", C3, "-."), ("M7", "Tail M7", INK, ":"),
              ("M8", r"Reset $\mathrm{out}_p$/$\mathrm{out}_n$ M8/M9", C4, "-"),
              ("M10", r"Reset $d_p$/$d_n$ M10/M11", GREY, "--")]
    fig, ax = plt.subplots(figsize=(4.8, 3.6), layout="constrained")
    for dev, lab, col, ls in groups:
        ax.semilogy(t, np.maximum(np.asarray(S[dev]), 1e-27), color=col, ls=ls, lw=1.5, label=lab)
    ax.set_ylim(1e-26, 1e-21)
    ax.set_xlim(t.min(), t.max())
    ax.set_xlabel("Time after the rising clock edge (ps)")
    ax.set_ylabel(r"Channel-noise PSD $S_{id}$ (A$^2$/Hz)")
    fig.legend(loc="outside lower center", ncol=2)
    _save(fig, path)
    return fig


def plot_offset(off, path=None):
    """(a) histogram of the Monte Carlo offsets; (b) a few SAR convergence traces."""
    import matplotlib.pyplot as plt
    v = np.asarray(off["vos"]) * 1e3
    fig, axs = plt.subplots(1, 2, figsize=(7.16, 2.9), layout="constrained")
    ax = axs[0]
    nb = max(8, int(np.sqrt(len(v)) * 1.5))
    ax.hist(v, bins=nb, color=C1, alpha=0.85, edgecolor="white", lw=0.6, density=True,
            label=f"SPICE MC, $N$ = {len(v)}")
    x = np.linspace(v.min() - 2, v.max() + 2, 300)
    ax.plot(x, stats.norm.pdf(x, off["mean"] * 1e3, off["sigma"] * 1e3), color=C2, lw=1.5,
            label=rf"Gaussian, $\sigma_{{\mathrm{{os}}}}$ = {off['sigma'] * 1e3:.2f} mV")
    ax.set_xlim(x.min(), x.max())
    ax.set_ylim(0, ax.get_ylim()[1] * 1.3)
    ax.set_xlabel(r"Input-referred offset $V_{\mathrm{os}}$ (mV)")
    ax.set_ylabel(r"Probability density (mV$^{-1}$)")
    ax.set_title("(a) Offset distribution")
    ax.legend(loc="upper left")
    ax = axs[1]
    h = np.asarray(off["history"]) * 1e3
    sty = [(C1, "-", "o"), (C2, "--", "s"), (C3, "-.", "^"), (C4, ":", "D")]
    for i, (col, ls, mk) in zip(range(min(4, len(h))), sty):
        ax.plot(np.arange(h.shape[1]), h[i], ls=ls, marker=mk, ms=4, lw=1.2, color=col,
                label=f"Instance {i + 1}")
    ax.set_xlim(-0.5, h.shape[1] - 0.5)
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
    ax.set_xlabel("Clock cycle")
    ax.set_ylabel(r"Applied $\Delta V_{\mathrm{in}}$ (mV)")
    ax.set_title("(b) Successive-approximation search")
    ax.legend(loc="lower right", ncol=2)
    _save(fig, path)
    return fig


def plot_delay(dl, path=None):
    """Clock-to-output delay vs |dVin| with the regeneration-time-constant fit."""
    import matplotlib.pyplot as plt
    dvin, t_latch, t_q = (np.asarray(dl[k]) for k in ("dvin", "t_latch", "t_q"))
    fig, ax = plt.subplots(figsize=(4.4, 3.2), layout="constrained")
    x = dvin * 1e3
    ax.semilogx(x, t_q * 1e12, "s-", color=C2, lw=1.2, ms=4.5, mfc="white", label=r"Clock $\phi$ to $q$")
    ax.semilogx(x, t_latch * 1e12, "o-", color=C1, lw=1.2, ms=4.5,
                label=r"Clock $\phi$ to latch")
    small = dvin <= 1e-3
    if small.sum() >= 2:
        c = np.polyfit(np.log(dvin[small]), t_latch[small], 1)
        xx = np.logspace(np.log10(dvin.min()), -3, 50)
        ax.semilogx(xx * 1e3, np.polyval(c, np.log(xx)) * 1e12, "--", color=INK, lw=1.0,
                    label=rf"Fit, $\tau_{{\mathrm{{regen}}}}$ = {dl['tau_regen'] * 1e12:.1f} ps")
    ax.set_xlabel(r"Input magnitude $|\Delta V_{\mathrm{in}}|$ (mV)")
    ax.set_ylabel("Delay (ps)")
    lo = min(t_latch.min(), t_q.min()) * 1e12
    hi = max(t_latch.max(), t_q.max()) * 1e12
    ax.set_ylim(lo - 0.45 * (hi - lo), hi + 0.08 * (hi - lo))
    ax.legend(loc="lower left")
    _save(fig, path)
    return fig
````

````python
#@title SKY130 StrongARM comparator: transient
# Block level, comparator: a transistor-level StrongARM comparator in SKY130 (ngspice, corner tt, 27 C, VDD 1.8 V,
# 10 MHz clock) makes six decisions on +-1, +-0.2 and +-20 mV differential inputs. Live SPICE run.
# Inputs: SKY130_LIB, NGSPICE, WORK (setup cell); dev/spice_comparator/comparator_spice.py.
# Outputs: cfg and t0 (used by the noise and offset cells), a card with the decisions, the transient figure.
import importlib, comparator_spice as cs
# reload: picks up edits to the module without restarting the kernel
importlib.reload(cs)
cfg = cs.default_config(str(SKY130_LIB), str(NGSPICE), str(WORK / "spice_cmp"), quick=True)   # quick=False: 4x the statistics
cfg["n_jobs"] = max(2, min(4, os.cpu_count() or 2))         # parallel ngspice processes: 2 to 4
t0 = time.time()
# one transient run; the decision q is read after each rising clock edge (default dVin set of run_transient_demo)
tr = cs.run_transient_demo(cfg)
card("StrongARM comparator: six decisions in SKY130 (ngspice, tt, 27 °C, 1.8 V)",
     # a decision is correct when q = 1 exactly for dVin > 0
     [("ΔV<sub>in</sub> → q", ", ".join(f"{d * 1e3:+g} mV → {q}" for d, q in zip(tr["dvins"], tr["decisions"]))),
      ("all decisions correct", "<b>yes</b>" if all((d > 0) == bool(q) for d, q in zip(tr["dvins"], tr["decisions"])) else "<b>no</b>"),
      ("run time", f"{time.time() - t0:.1f} s")])
cs.plot_transient(tr, cfg); plt.show()
# interactive companion: (a) the clock and (b) q over the six decisions (hover: time and voltage); (c)-(f) one decision
# after its clock edge, as above. The menu picks the decision shown in (c)-(f): all six are in the data, and the menu
# opens at the static figure's (+0.2 mV). Waveforms within 2 mV of the ngspice output (_ithin).
def _itran():
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    w, te, dv, dec = tr["wave"], tr["edges"], tr["dvins"], tr["decisions"]
    t, vdd, T, zc = 1e9 * np.asarray(w["time"]), cfg["vdd"], 1 / cfg["f_clk"], 2
    mv = lambda d: cs._mv(d * 1e3)
    ttl_c = lambda k: f"(c) Clock and tail node ({mv(dv[k])} mV decision)"
    f = make_subplots(4, 2, specs=[[{"colspan": 2}, None], [{"colspan": 2}, None], [{}, {}], [{}, {}]], row_heights=[0.55, 0.9, 1, 1],
                      vertical_spacing=0.1, horizontal_spacing=0.05,
                      subplot_titles=("(a) Clock ϕ", f"(b) Latched decision <i>q</i>, one per clock cycle at {cfg['f_clk'] / 1e6:g} MHz "
                                      "(Δ<i>V</i><sub>in</sub> → <i>q</i>)", ttl_c(zc), "(d) Integration nodes", "(e) Latch outputs",
                                      "(f) SR-latch outputs"))
    for node, col, wd, r, nm in (("v(clk)", cs.GREY, 1.3, 1, "ϕ"), ("v(c0_q)", cs.C1, 1.6, 2, "<i>q</i>")):
        v = np.asarray(w[node]); i_ = _ithin(t, v, 0.002)
        f.add_trace(go.Scatter(x=t[i_], y=v[i_], mode="lines", line=dict(color=col, width=wd), showlegend=False,
                               hovertemplate=nm + ": %{y:.3f} V at %{x:.2f} ns<extra></extra>"), r, 1)
    for k, e in enumerate(te):
        for r in (1, 2):
            f.add_vline(x=e * 1e9, line=dict(color=cs.GREY, width=0.6, dash="dot"), row=r, col=1)
        f.add_annotation(x=(e + T / 2) * 1e9, y=vdd + 0.25, text=f"{mv(dv[k])} mV → {dec[k]}", yanchor="bottom", showarrow=False,
                         font=dict(size=round(9 * _PT)), row=2, col=1)
    # (c)-(f): the nodes shown zoomed after a clock edge, one panel and legend each: (row, column, legend, [(node, label,
    # colour, dash)]); all six decisions are added, only the menu's one is visible
    rows_z = [(3, 1, "legend", [("v(clk)", "clock ϕ", cs.GREY, "solid"), ("v(c0_tail)", "tail", cs.C1, "solid")]),
              (3, 2, "legend2", [("v(c0_dp)", "<i>d</i><sub>p</sub>", cs.C1, "solid"), ("v(c0_dn)", "<i>d</i><sub>n</sub>", cs.C2, "dash")]),
              (4, 1, "legend3", [("v(c0_outp)", "out<sub>p</sub>", cs.C1, "solid"), ("v(c0_outn)", "out<sub>n</sub>", cs.C2, "dash")]),
              (4, 2, "legend4", [("v(c0_q)", "<i>q</i>", cs.C1, "solid"), ("v(c0_qb)", "<i>q̄</i>", cs.C2, "dash")])]
    n0, owner = len(f.data), []                                     # zoom traces: one set per decision, the menu shows one set
    for k, e in enumerate(te):
        m = (w["time"] > e - 0.2e-9) & (w["time"] < e + 1.5e-9); tz = (np.asarray(w["time"])[m] - e) * 1e9
        for r, c, lg, sig in rows_z:
            for key, lab, col, dash in sig:
                v = np.asarray(w[key])[m]; i_ = _ithin(tz, v, 0.002)
                f.add_trace(go.Scatter(x=tz[i_], y=v[i_], mode="lines", line=dict(color=col, width=1.5, dash=dash), name=lab, legend=lg,
                                       visible=k == zc, hovertemplate=lab + ": %{y:.3f} V, %{x:.3f} ns after the edge<extra>"
                                       + f"{mv(dv[k])} mV</extra>"), r, c)
                owner.append(k)
    vis = lambda k: [True] * n0 + [o == k for o in owner]
    f.update_layout(updatemenus=[dict(type="dropdown", active=zc, x=0.0, xanchor="left", y=1.0, yanchor="bottom", pad=dict(t=0, b=34),
                                      bgcolor="white", bordercolor=COL["faint"], font=dict(size=round(8 * _PT)), showactive=True,
                                      buttons=[dict(label=f"(c)–(f): {mv(dv[k])} mV → {dec[k]}", method="update",
                                                    args=[{"visible": vis(k)}, {"annotations[2].text": ttl_c(k)}]) for k in range(len(te))])])
    for r in (1, 2):
        f.update_xaxes(range=[0, float(t[-1])], row=r, col=1)
    f.update_xaxes(showticklabels=False, row=1, col=1); f.update_xaxes(title_text="Time (ns)", row=2, col=1)
    f.update_yaxes(title_text="ϕ (V)", tickvals=[0, vdd / 2, vdd], range=[-0.2, vdd + 0.2], row=1, col=1)
    f.update_yaxes(title_text="<i>q</i> (V)", tickvals=[0, vdd / 2, vdd], range=[-0.2, vdd + 0.85], row=2, col=1)
    for r, c, lg, _ in rows_z:
        f.update_xaxes(range=[-0.2, 1.5], tickvals=[0, 0.5, 1.0, 1.5], showticklabels=r == 4,
                       title_text="Time after the rising clock edge (ns)" if r == 4 else None, row=r, col=c)
        f.update_yaxes(range=[-0.2, vdd + 0.4], tickvals=list(np.arange(0, vdd + 0.01, 0.6)), showticklabels=c == 1,
                       title_text="Voltage (V)" if c == 1 else None, row=r, col=c)
    pos = {"legend": ("right", "middle"), "legend2": ("right", "top"), "legend3": ("right", "middle"), "legend4": ("left", "middle")}
    for (r, c, lg, _), ax_ in zip(rows_z, ("3", "4", "5", "6")):
        xd, yd = f.layout["xaxis" + ax_].domain, f.layout["yaxis" + ax_].domain; xa, ya = pos[lg]
        f.layout[lg] = dict(x=xd[1] - 0.01 if xa == "right" else xd[0] + 0.01, xanchor=xa, yanchor=ya,
                            y=yd[1] - 0.01 if ya == "top" else (yd[0] + yd[1]) / 2, font=dict(size=round(8 * _PT)))
    f.update_layout(margin=dict(t=round(17 * _PT) + 42))
    return f
iplot(_itran, "Fig.: StrongARM comparator transient, six decisions and one zoomed (interactive)")
````

**All six decisions are correct**, including the ±0.2 mV inputs, less than half the input-referred noise simulated below (Fig. 21; result panel). Each decision appears at q within a nanosecond of the rising clock edge and is held through the reset phase, so the feedback DAC sees a stable bit. In the detailed cycle (Fig. 21(c)–(f)), the tail node discharges when the clock rises, $d_\mathrm{p}$ and $d_\mathrm{n}$ discharge at rates set by the input, and the latch outputs separate once they have fallen by about one threshold voltage. Integration, regeneration and latching thus complete well within the 50 ns evaluation phase, so **the accuracy is limited by noise and offset**, which the next two cells measure.

> **Next cell.** 1. Measures the energy per decision from the supply current, with the output toggling every cycle. 2. Runs 1,100 decisions against eleven small differential inputs with the PDK's channel noise injected. 3. Fits the probability of deciding 1 with a Gaussian (probit) curve, which gives $\sigma_\mathrm{n}$. 4. Repeats the measurement with the noise of the input pair only and of the other devices only, and evaluates the first-order formula at the simulated operating point. Run time: about 40 s on two cores.

````python
#@title SKY130 StrongARM comparator: noise and energy
# Block level, comparator: input-referred noise and energy per decision of the StrongARM comparator (SKY130, ngspice,
# tt, 27 C, 1.8 V, 10 MHz). Live SPICE run with the quick statistics.
# Noise method (measure_noise): the PDK channel-noise PSD is read along the comparator's reference trajectory and
# injected as noise sources; at each of a set of dVin levels around 0 the fraction of decisions q = 1 is counted, and a
# probit fit P(q = 1) = Phi((dVin - mu) / sigma_n) gives sigma_n (a pilot run over +-2 mV sets the level range).
# Where the noise comes from: the first-order kT/C estimate of the Theory with this run's operating point
# (noise_estimate: gm, I_D, C_P and T_int of the input device from a noiseless transient), and two further runs in which
# only the input pair, or only the latch, tail and reset devices, carry their channel noise (measure_noise_part).
# Energy (measure_energy): supply charge per clock period with q toggling every cycle (worst case for the latch),
# core + digital (buffers, SR latch, 5 fF loads) + clock-driver share.
# Inputs: SKY130_LIB, NGSPICE, WORK (setup cell). Outputs: cfg, t0, en, nf (nf['sigma'] is used by the offset cell),
# est, nf_pair, nf_rest.
import importlib, comparator_spice as cs
importlib.reload(cs)
cfg = cs.default_config(str(SKY130_LIB), str(NGSPICE), str(WORK / "spice_cmp"), quick=True)   # quick=False: 4x the statistics
cfg["n_jobs"] = max(2, min(4, len(os.sched_getaffinity(0)) or 2))   # parallel ngspice jobs: the cores this process may use, 2..4
t0 = time.time()
en = cs.measure_energy(cfg)
nf = cs.measure_noise(cfg, log=lambda *a: None)
est = cs.noise_estimate(cfg, (nf["traj_t"], nf["psd"]))
PAIR = ("M1", "M2")
nf_pair = cs.measure_noise_part(cfg, nf["psd"], PAIR, nf["dvin"], 40, cfg["seed"] + 3, "pair")
nf_rest = cs.measure_noise_part(cfg, nf["psd"], [d for d in cs.NOISY_DEVICES if d not in PAIR],
                                np.linspace(-0.4e-3, 0.4e-3, 11), 30, cfg["seed"] + 4, "rest")   # a steeper curve: narrower levels
q_sum = np.hypot(nf_pair["sigma"], nf_rest["sigma"])            # independent sources add in power
mv = lambda f: f"{f['sigma'] * 1e3:.3f} ± {f['sigma_se'] * 1e3:.3f} mV"
card("StrongARM comparator: noise and energy",
     [("input-referred noise σ<sub>n</sub>", f"<b>{nf['sigma'] * 1e3:.3f} mV</b>"),
      ("noise of the input pair M1/M2 alone", f"<b>{mv(nf_pair)}</b>, {100 * nf_pair['sigma'] ** 2 / q_sum ** 2:.0f} % of the noise power"),
      ("noise of the latch, tail and reset devices alone", f"{mv(nf_rest)}; both parts in quadrature {q_sum * 1e3:.3f} mV"),
      ("first-order kT/C estimate, this run's operating point", f"{est['sigma_gamma=2/3'] * 1e3:.2f} mV with γ = 2/3, "
       f"{est['sigma_gamma_PDK'] * 1e3:.2f} mV with the PDK's γ = {est['value_gamma_PDK']:.2f} "
       f"(g<sub>m</sub> = {est['gm'] * 1e3:.2f} mS, T<sub>int</sub> = {est['T_int'] * 1e12:.0f} ps, C<sub>P</sub> = {est['C_P'] * 1e15:.1f} fF)"),
      # power = E * f_clk: J * 10e6 Hz = W, then x1e6 for uW
      ("energy per decision", f"{en['toggle']['total'] * 1e15:.0f} fJ = {en['toggle']['total'] * 10e6 * 1e6:.2f} µW at 10 MHz")])
cs.plot_noise(nf, parts=[("input pair M1/M2", nf_pair), ("latch, tail, reset", nf_rest)]); plt.show()
# interactive companion: the three noise runs as above, each level's P(q = 1) with its Wilson 95 % interval (hover: the
# decisions counted) and its probit fit; a legend entry hides a run.
def _inoise():
    import plotly.graph_objects as go
    from scipy import stats as _st
    f = go.Figure()
    runs = [("all devices", nf, cs.C2, "circle-open", "solid"), ("input pair M1/M2 only", nf_pair, cs.C3, "square-open", "dash"),
            ("latch, tail, reset only", nf_rest, cs.INK, "triangle-up-open", "dot")]
    x = np.linspace(min(np.min(r[1]["dvin"]) for r in runs) * 1.1, max(np.max(r[1]["dvin"]) for r in runs) * 1.1, 300)
    for lab, r, col, mk, dash in runs:
        dv, k, n = np.asarray(r["dvin"]), np.asarray(r["ones"]), np.asarray(r["trials"]); p = k / n; z = 1.96
        c = (k + z * z / 2) / (n + z * z); h = z * np.sqrt(k * (n - k) / n + z * z / 4) / (n + z * z)
        nm = f"{lab}: σ<sub>n</sub> = {r['sigma'] * 1e3:.3f} ± {r['sigma_se'] * 1e3:.3f} mV"
        f.add_trace(go.Scatter(x=dv * 1e3, y=p, mode="markers", marker=dict(symbol=mk, size=7, color=col, line=dict(width=1.2)),
                               error_y=dict(type="data", symmetric=False, array=(c + h) - p, arrayminus=p - (c - h), color=col,
                                            thickness=1, width=3),
                               customdata=np.c_[k, n], name=nm, legendgroup=lab,
                               hovertemplate=lab + ": P(q = 1) = %{y:.3f} (%{customdata[0]} of %{customdata[1]}) at "
                                             "%{x:.3f} mV<extra></extra>"))
        f.add_trace(go.Scatter(x=x * 1e3, y=_st.norm.cdf((x - r["mu"]) / r["sigma"]), mode="lines", line=dict(color=col, width=1.5,
                               dash=dash), legendgroup=lab, showlegend=False, hovertemplate=lab + " fit: %{y:.3f} at %{x:.3f} mV<extra></extra>"))
    f.update_xaxes(title_text="Differential input Δ<i>V</i><sub>in</sub> (mV)", range=[x.min() * 1e3, x.max() * 1e3])
    f.update_yaxes(title_text="Probability <i>P</i>(<i>q</i> = 1)", range=[-0.04, 1.04])
    f.update_layout(legend=dict(y=-0.25))
    return f
iplot(_inoise, "Fig.: StrongARM comparator input-referred noise, three runs (interactive)")
````

The simulated decisions follow the Gaussian curve across the whole input range (Fig. 23); the probit fit gives **$\sigma_\mathrm{n}$ = 0.560 mV**, inside the predicted range and below the $\sqrt{kT/C_\mathrm{P}}$ of a 5 fF node. **The noise comes from the input pair:** with only M1/M2 noisy, $\sigma$ = 0.516 ± 0.043 mV, 94 % of the noise power; with only the latch, tail and reset devices noisy, 0.128 ± 0.013 mV, a curve four times steeper. In quadrature the two give 0.532 mV, within one standard error of $\sigma_\mathrm{n}$, as predicted. At the simulated operating point ($g_m$ = 0.46 mS, $T_\mathrm{int}$ = 144 ps, $C_\mathrm{P}$ = 4.8 fF), the first-order formula gives 0.41 mV with $\gamma$ = 2/3 and 0.70 mV with the PDK's $\gamma$ = 1.95, which bracket the input pair's 0.516 mV. The comparator requires 186 fJ per decision, 1.86 µW at 10 MHz, or $C_\mathrm{sw} = E/V_\mathrm{DD}^2$ ≈ 57 fF ($E$: the energy per decision) charged per cycle.

**The comparator's noise is thus the kT/C noise of its integrating nodes**, set by the input pair's channel noise during the integration; the latch adds a few per cent. The first-order formula, with a constant $g_m$ and a sharp end of the integration at $T_\mathrm{int}$, places it within a factor of 1.4 either way; lowering it would take a larger node capacitance or a longer integration, at the cost of energy and speed. Inside a second-order loop, noise at the quantiser is shaped like quantisation noise, so a value of this size is not expected to limit the modulator's SNDR; Section VIII-A tests this, and Section VIII-C evaluates the comparator's share of the power.

### C. The SRAM-PUF cell

The three settings of the cell model of Section V-D, $s_{25}$, $a$ and the bias $P(1)$, from which Section VII-E sizes the code, follow from a race between the two inverters at power-up. With the word line at 0, both storage nodes q and q̄ start at 0 V and rise with $V_\mathrm{DD}$ (Fig. 26). Once the loop gain of the cross-coupled inverters exceeds one, any difference between q and q̄ grows exponentially and the cell falls to one side [13], decided by a mismatch voltage $\Delta V_{\mathrm{m},i}$, fixed for cell $i$ (the threshold mismatch of its two pull-down and two pull-up transistors), plus a noise voltage $v_\mathrm{n}$, new at every power-up (the thermal noise of the same devices at the moment of decision). The cell powers up to 1 when $\Delta V_{\mathrm{m},i} + v_\mathrm{n} > 0$; dividing by the mismatch spread $\sigma_\mathrm{m}$ over cells gives the cell model of Section V-D, $m_i + s\,n > 0$, with the normalised mismatch $m_i \sim \mathcal{N}(\mu, 1)$ of mean $\mu$ (the bias), fresh $n \sim \mathcal{N}(0, 1)$ and the noise-to-mismatch ratio $s = \sigma_\mathrm{v}/\sigma_\mathrm{m}$ ($\sigma_\mathrm{v}$: rms noise voltage). A cell returns 1 with probability $p_i = \Phi(m_i/s)$, so **only the few cells with $\lvert m_i\rvert \lesssim s$ are unstable**.

A cell disagrees with the enrolment power-up with probability $2p_i(1 - p_i)$; averaged over $m \sim \mathcal{N}(0, 1)$ for $s \ll 1$, the bit error rate (BER) is proportional to the noise-to-mismatch ratio,

$$\mathrm{BER} \approx \frac{\sqrt{2}}{\pi}\, s \approx 0.45\, s.$$

With the PDK's mismatch coefficients the pull-down nMOS (W/L = 0.64/0.15 µm) has $\sigma(V_\mathrm{T})$ = 10.8 mV and the minimum-size pull-up pMOS about 23 mV, so $\sigma_\mathrm{m}$ is a few tens of millivolts, while the noise on a storage node of capacitance $C$ of about 2 fF is of the order of $\sqrt{kT/C}$ = 1.4 mV at 25 °C. **The analysis above therefore predicts $s$ of the order of a few hundredths**, well below the illustrative $s_{25} = 0.06$, a BER of about 1 % at 25 °C, and $P(1) = \Phi(\mu) = 0.5$ for a symmetric cell. If only the thermal noise changed with temperature, $s$ would grow as $\sqrt{T}$, by only 10 % from 25 to 85 °C (about 0.0016 per °C), and the BER would fall at −40 °C. Temperature also shifts the threshold voltages and the supply at which the cell resolves, and with them the relative weight of the nMOS and pMOS mismatch, so a cell's preferred value can change in either direction; the model represents both effects by $s(T) = s_{25}\,(1 + a\,\lvert T - 25\,{}^\circ\mathrm{C}\rvert)$, symmetric about the 25 °C enrolment. The second prediction, tested by the fit below, is conditional: if the BER rises on both sides of 25 °C and the fitted $a$ is much larger than 0.0016 per °C, the temperature behaviour is governed by the change of the mismatch weights rather than by kT; Section VII-F analyses this mechanism.

The simulations are set up as follows:

* *Level and circuit.* Transistor level, the schematic netlist of Fig. 26, SKY130 1.8 V devices at tt with the PDK's mismatch models. Pull-down W/L = 0.64/0.15 µm, pull-up and access 0.42/0.15 µm: a cell ratio of 1.5 for read stability and a minimum-width pull-up for writeability, as in a memory cell; minimum sizes maximise the mismatch, the source of a PUF's identity. Junction areas as in Section VII-B.
* *Stimulus.* (Fig. 27(a)) $V_\mathrm{DD}$ ramps linearly from 0 to 1.8 V in 1 µs with every node at 0 V; the word line stays at 0, and the bit lines and the n-well follow $V_\mathrm{DD}$. The power-up value is the sign of $q - \bar{q}$ when the transient stops (0.8, 0.6 and 0.5 µs at −40, 25 and 85 °C); the smallest $\lvert q - \bar{q}\rvert$ over all runs checks that no cell is read unresolved (99.9 % of the supply in the recorded post-layout run). The maximum time step is 0.5 ns below 25 °C, 1 ns from 25 °C up to 85 °C and 2 ns at 85 °C; on 200 noiseless cells these steps give the same bits as a 0.1 ns reference, whereas larger steps reversed up to 9 % of the noiseless power-ups.
* *Mismatch.* One netlist holds 64 cells (a chunk), each transistor with its own draw from the PDK's mismatch model; chunk c uses the mismatch seed 1 + c, so the cell population depends only on the number of cells, not on the number of processor cores, and every cell is reproducible (checked by one noiseless power-up per chunk at 25 °C).
* *Noise.* ngspice transients contain no device noise, so each storage node receives a white noise current (Fig. 27(b); envelope testbench in Fig. 27(c)): a unit Gaussian sample-and-hold source with a new sample every 5 ns (white up to 100 MHz, at least six times the storage node's bandwidth at the decision), scaled by the PDK's channel-noise PSD of the nominal cell held at its metastable point (q shorted to q̄; half of it to each node), from `.noise` analyses at 10 MHz, where the spectrum is white, at every supply from 0.05 to 1.8 V in 50 mV steps and at each temperature. This noise envelope is interpolated against $V_\mathrm{DD}$ inside the netlist, so the noise follows the ramp. The power-ups of a chunk run in one ngspice session, re-seeded before each, with the temperature changed inside the session, so the mismatch is identical at every power-up and temperature. Only thermal noise is modelled (no 1/f, random-telegraph or supply noise), so the simulated $s$ is likely to underestimate that of silicon; the PDK's mismatch in its denominator is not verified against silicon either.
* *Temperatures and metric.* 25 °C (the enrolment temperature), 85 °C (the upper end of the 0–85 °C range) and −40 °C (a cold extreme below it; 0 °C is interpolated piecewise-linearly between −40 and 25 °C where needed, Sections VII-E and VIII-F). The enrolment is the first power-up at 25 °C; the BER is the fraction of all later power-ups that differ from it; a cell is unstable if its value changes between power-ups at that temperature.
* *Fits.* (i) A maximum-likelihood fit of $s_{25}$, $a$ and $\mu$ to the counts of ones of all cells at all temperatures, each cell's latent mismatch integrated out numerically (Fig. 29). (ii) A BER-matched fit, used for the model because the error rate is what the fuzzy extractor sees: $s_{25}$ and $a$ by weighted least squares of the model's $\mathrm{BER}(T)$, computed exactly from the bivariate normal distribution of the enrolment and a later power-up, against the simulated $\mathrm{BER}(T)$, weighted by its bootstrap standard error, and μ from the bias at 25 °C; intervals by bootstrap over cells (200 resamples).
* *Recorded and live runs.* The power-up waveforms are simulated live. A Monte Carlo that resolves error rates of about 1 % takes about 40 minutes on two cores, so the runs of the schematic and the extracted cell (1,024 cells × 8 power-ups × 3 temperatures each; `dev/spice_puf/demo.py`, `dev/spice_puf/demo_postlayout.py`, same code and seeds) are recorded; `PUF_MODE = "full"` re-runs both live and fits the model to the live run, and `PEX_LIVE = True` re-runs the first 128 cells × 5 power-ups of the extracted cell (about 105 s), which are exactly the first 128 cells of the record, checked bit for bit. The recorded runs also contain power-ups at 125 °C, outside the specified range; they are not used (`dev/spice_puf/refit_0_85.py` repeats the fits without them, from the same bits).

> **Next cell.** 1. Writes the SRAM-PUF module (about 1,000 lines) to `puf_spice.py`; no simulation is run. 2. The cells below import it. The module holds the 6T cell netlist and the power-up transient, the noise envelope (the PDK's channel noise at each supply and temperature), the Monte Carlo over mismatch, noise and temperature, and the fits that map the result onto the cell model of Section V-D.

````python
%%writefile puf_spice.py
"""SKY130 6T-SRAM PUF characterisation with ngspice (Code-a-Chip, ISSCC 2027).

What this module does
---------------------
1. Builds a 6T SRAM cell from the standard SKY130 1.8 V devices
   (sky130_fd_pr__nfet_01v8 / sky130_fd_pr__pfet_01v8), word line held at 0,
   bit lines precharged (tied to the supply), and powers it up by ramping VDD
   0 -> 1.8 V.  The cell settles to 0 or 1 depending on device mismatch.
2. Monte Carlo: one netlist holds many cells; the PDK's own mismatch model
   (the "<corner>_mm" library section, MC_MM_SWITCH=1) gives every transistor
   instance its own draw.  `.option seed=<n>` makes the draws reproducible.
3. Repeatability: ngspice transients contain no device noise, so every
   storage node gets a white noise current with the shot-noise PSD of the two
   transistors that drive it, evaluated at the nominal cell's metastable bias
   (see `noise_envelope`).  The noise uses `trrandom` sources re-seeded with
   `setseed` before every power-up, so repeats share the SAME mismatch (one
   parsed circuit) but get fresh, reproducible noise.
4. Temperature sweep inside the same ngspice session (`option temp=`), so the
   mismatch is identical at every temperature.
5. Fits the behavioural model  bit_i = [m_i + s(T) n > 0],  m_i ~ N(mu, 1),
   n ~ N(0, 1),  s(T) = s25 (1 + a |T - 25|)  by maximum likelihood.
6. Matplotlib figures.

Only the standard library, numpy, scipy and matplotlib are used.  Every path
is passed in through a config dict (see `make_config`).
"""

import contextlib
import json
import math
import os
import re
import shutil
import subprocess
import threading
import time
from concurrent.futures import ThreadPoolExecutor

import numpy as np
from scipy import optimize, special, stats

Q_E = 1.602176634e-19      # elementary charge [C]
K_B = 1.380649e-23         # Boltzmann constant [J/K]

# ----------------------------------------------------------------------------
# Configuration
# ----------------------------------------------------------------------------

# Cell geometry (um).  All devices use the minimum SKY130 1.8 V length 0.15 um.
#   pull-down  W=0.64: cell ratio PD/PG = 1.5 (read stability of a real 6T cell)
#   pull-up    W=0.42: minimum width, weakest device (write-ability)
#   pass gate  W=0.42: minimum width; held off (WL = 0) during power-up
# 0.42 um is the smallest width the SKY130 model bins cover (wmin = 0.42 um).
# Small devices maximise the mismatch, which is what a PUF wants.
CELL = dict(w_pd=0.64, w_pu=0.42, w_pg=0.42, l=0.15,
            diff_ext=0.29)   # source/drain diffusion length for AD/AS/PD/PS (um)

# Power-up stimulus and transient settings.
POWERUP = dict(
    vdd=1.8,            # final supply [V]
    t_ramp=1e-6,        # 0 -> vdd linear ramp time [s]
    v_noise_on=0.0,     # noise sources start when VDD reaches this [V] (0: from t = 0;
                        # at 125 C the nodes start to separate at VDD ~ 0.1 V)
    nt=5e-9,            # noise sample interval [s] -> white up to 1/(2 nt) = 100 MHz,
                        # >= 6x the storage-node bandwidth at the decision (<= ~16 MHz)
    # Time-step limit and stop time per temperature (interpolated / stepped in T).
    # Measured, 200 cells, noise off: default tolerances with tmax <= 0.5 ns at
    # -40 C, <= 1 ns at 25 C, <= 2 ns at 85-125 C give bit-identical results to a
    # 0.1 ns gear reference and to a run with abstol=1e-15 chgtol=1e-19 reltol=1e-4;
    # larger steps flip up to 9 % (-40 C) / 8 % (25 C) of the cells.  The slowest
    # cell resolves (|q - qb| > VDD/2) at 0.68 / 0.51 / 0.42 / 0.31 us at
    # -40 / 25 / 85 / 125 C.
    T_pts=[-40.0, 25.0, 85.0, 125.0],
    t_stop_pts=[0.8e-6, 0.6e-6, 0.5e-6, 0.4e-6],
)


def tran_settings(T, pu=POWERUP):
    """(t_stop, t_max) for a power-up at temperature T [C]."""
    # stop time interpolated between the measured points (np.interp holds the end values outside -40 to 125 C);
    # below -40 C the cells resolve later still, so the -40 C value is extended by 25 %
    t_stop = float(np.interp(T, pu["T_pts"], pu["t_stop_pts"]))
    if T < -40:
        t_stop = pu["t_stop_pts"][0] * 1.25
    # time-step limit: the largest step that was bit-identical to the reference at the next colder measured
    # temperature (see POWERUP)
    t_max = 0.5e-9 if T < 25 else (1e-9 if T < 85 else 2e-9)
    return t_stop, t_max * pu.get("tmax_scale", 1.0)       # tmax_scale < 1: the convergence check of Section VII-D


def make_config(model_lib, ngspice="ngspice", workdir="puf_work", corner="tt",
                jobs=None, model_mode="lib"):
    """Collect paths and run options in one dict.

    model_lib : path to the SKY130 ngspice library, normally
                <PDK_ROOT>/sky130A/libs.tech/ngspice/sky130.lib.spice
                (needs <PDK_ROOT>/sky130A/libs.ref/sky130_fd_pr/ next to it).
    corner    : MOSFET corner; '<corner>_mm' (mismatch) and '<corner>' sections are used.
    jobs      : parallel ngspice processes (default: number of CPUs, max 8).
    """
    if jobs is None:
        # one ngspice process per CPU this process may use (sched_getaffinity respects taskset), at most 8
        jobs = max(1, min(8, len(os.sched_getaffinity(0)) or 1))
    return dict(model_lib=os.path.abspath(model_lib), ngspice=ngspice,
                workdir=os.path.abspath(workdir), corner=corner, jobs=jobs,
                model_mode=model_mode)


def write_spiceinit(workdir):
    """ngspice reads `.spiceinit` from the CURRENT directory, and only if that is
    missing from $HOME.  We always run ngspice with cwd=workdir and put this
    file there, so a user's ~/.spiceinit (e.g. one set up for another PDK) is
    ignored.  ngbehavior=hsa is required to read the SKY130 libraries
    (HSPICE-style parameter handling); ng_nomodcheck only speeds up loading."""
    os.makedirs(workdir, exist_ok=True)
    with open(os.path.join(workdir, ".spiceinit"), "w") as f:
        f.write("set ngbehavior=hsa\nset ng_nomodcheck\n")


def ngspice_version(ngspice_bin):
    """The version printed by `ngspice -v` (for example '44.2'), recorded with the results."""
    out = subprocess.run([ngspice_bin, "-v"], capture_output=True, text=True).stdout
    m = re.search(r"ngspice-(\S+)", out)
    return m.group(1) if m else out.strip()



# One core per ngspice process. Several single-threaded ngspice processes that share a set of cores ran 3.5 times slower
# on a two-core VM than the same processes pinned one per core (the scheduler kept moving them between cores); each run
# therefore takes a free core from this pool for its duration (taskset), which is harmless where the effect is absent.
_CORE_LOCK = threading.Lock()
_FREE_CORES = sorted(os.sched_getaffinity(0))


def _pinned(cmd):
    """Context manager: (command prefixed with taskset to one free core, or unchanged if none is free)."""
    @contextlib.contextmanager
    def cm():
        with _CORE_LOCK:
            core = _FREE_CORES.pop(0) if _FREE_CORES and shutil.which("taskset") else None
        try:
            yield (["taskset", "-c", str(core)] + cmd) if core is not None else cmd
        finally:
            if core is not None:
                with _CORE_LOCK:
                    _FREE_CORES.append(core)
    return cm()


def run_ngspice(ngspice_bin, netlist, workdir, name="run", timeout=3600):
    """Write `netlist` to <workdir>/<name>.cir and run `ngspice -b` on it with
    cwd=workdir.  Returns (log text, wall seconds).  Raises on fatal errors."""
    write_spiceinit(workdir)
    path = os.path.join(workdir, name + ".cir")
    with open(path, "w") as f:
        f.write(netlist)
    t0 = time.time()
    with _pinned([ngspice_bin, "-b", name + ".cir"]) as cmd:
        p = subprocess.run(cmd, cwd=workdir, capture_output=True, text=True, timeout=timeout)
    dt = time.time() - t0
    log = p.stdout + p.stderr
    with open(os.path.join(workdir, name + ".log"), "w") as f:
        f.write(log)
    # ngspice can finish with status 0 although a command of the control block failed (the error is only printed),
    # so the log is also scanned for the messages that mean missing or invalid results
    bad = [l for l in log.splitlines()
           if re.search(r"fatal|Timestep too small|aborted|no such vector|Could not find",
                        l, re.I)]
    if p.returncode != 0 or bad:
        raise RuntimeError(f"ngspice failed ({name}, rc={p.returncode}):\n" + "\n".join(bad[:10])
                           + f"\n(see {os.path.join(workdir, name + '.log')})")
    return log, dt


def read_wrdata(path):
    """Read a file written by `wrdata` with wr_singlescale + wr_vecnames."""
    # first line: the vector names (wr_vecnames); then one row per time point: the time (one shared column,
    # wr_singlescale), then each vector in the order of the wrdata command
    with open(path) as f:
        names = f.readline().split()
    data = np.loadtxt(path, skiprows=1, ndmin=2)
    if data.shape[0] < 1 or data.shape[1] != len(names):
        raise RuntimeError(f"{path}: {data.shape[0]} rows of {data.shape[1]} columns for {len(names)} vectors (no or truncated data)")
    return names, data


# ----------------------------------------------------------------------------
# Netlists
# ----------------------------------------------------------------------------

def _mos(name, d, g, s, b, kind, w, l, ext):
    """One SKY130 1.8 V transistor (a subcircuit in the PDK, W/L in um).
    Diffusion area/perimeter as in a single-finger layout with `ext` um of
    source/drain diffusion: gives the storage nodes realistic junction C."""
    return (f"{name} {d} {g} {s} {b} sky130_fd_pr__{kind}_01v8 W={w:g} L={l:g} "
            f"ad={w*ext:.4g} as={w*ext:.4g} pd={2*(w+ext):.4g} ps={2*(w+ext):.4g} "
            f"nrd={ext/w:.4g} nrs={ext/w:.4g}")


def _mos_layout(d, ext):
    """One transistor of the extracted layout: drawn W/L and the extracted source/drain areas and perimeters
    (nrd/nrs as in the schematic, so the two cells differ only by what the layout changes)."""
    return (f"X{d['name']} {d['d']} {d['g']} {d['s']} {d['b']} sky130_fd_pr__{d['kind']}_01v8 W={d['w']:g} L={d['l']:g} "
            f"ad={d['ad']} as={d['as_']} pd={d['pd']} ps={d['ps']} nrd={ext / d['w']:.4g} nrs={ext / d['w']:.4g}")


def cell_subckt(cell=CELL, noise=True, nt=1e-9, td=0.0):
    """6T cell.  Ports: vdd bl blb wl q qb env.
    Inverter A (XPUa/XPDa) drives q from qb, inverter B drives qb from q.
    With noise=True each storage node gets   I_n = v(env) * v(nz)   where nz is a
    unit-variance Gaussian sample-and-hold source (`trrandom` type 2, new sample
    every nt) and env is the shared amplitude node (see noise_envelope).
    If cell carries 'layout' (puf_layout.postlayout_cell), the transistors take the extracted geometry and the
    extracted parasitic capacitors are added; instance names and order are those of the schematic, so the PDK's
    per-instance mismatch draws are the same as in the schematic cell."""
    c = cell
    L = [".subckt sram6t vdd bl blb wl q qb env"]
    if c.get("layout"):
        # post-layout cell: the extracted device geometry, then the extracted capacitors (one whose two terminals
        # are on the same net has no effect and is dropped)
        L += [_mos_layout(d, c["diff_ext"]) for d in c["layout"]["devices"]]
        L += [f"Cpex{i} {a} {b} {v:.6g}" for i, (a, b, v) in enumerate(c["layout"]["caps"]) if a != b]
    else:
        # schematic cell: inverter A (pull-up, pull-down) drives q from qb, inverter B drives qb from q, and the two
        # pass gates connect q and qb to the bit lines when the word line is high
        L += [_mos("XPUa", "q", "qb", "vdd", "vdd", "pfet", c["w_pu"], c["l"], c["diff_ext"]),
              _mos("XPDa", "q", "qb", "0", "0", "nfet", c["w_pd"], c["l"], c["diff_ext"]),
              _mos("XPUb", "qb", "q", "vdd", "vdd", "pfet", c["w_pu"], c["l"], c["diff_ext"]),
              _mos("XPDb", "qb", "q", "0", "0", "nfet", c["w_pd"], c["l"], c["diff_ext"]),
              _mos("XPGa", "bl", "wl", "q", "0", "nfet", c["w_pg"], c["l"], c["diff_ext"]),
              _mos("XPGb", "blb", "wl", "qb", "0", "nfet", c["w_pg"], c["l"], c["diff_ext"])]
    if noise:
        # trrandom(2 ts td sigma mean): Gaussian (type 2), a new sample every ts after the delay td, held between
        # samples; each B source injects env x sample as a current into its storage node (independent per node)
        L += [f"Vnza nza 0 dc 0 trrandom(2 {nt:g} {td:g} 1 0)",
              f"Vnzb nzb 0 dc 0 trrandom(2 {nt:g} {td:g} 1 0)",
              "Bna 0 q I=v(env)*v(nza)",
              "Bnb 0 qb I=v(env)*v(nzb)"]
    else:
        L += ["Rdum env 0 1e12"]          # keep the port connected
    L.append(".ends sram6t")
    return "\n".join(L)


def model_include(cfg, mismatch):
    """`.lib` line for the PDK.  mismatch=True uses the '<corner>_mm' section,
    which sets MC_MM_SWITCH=1: every transistor instance then draws its own
    AGAUSS() for vth0, voff and toxe (nfet) / vth0, voff, nfactor and toxe (pfet),
    each with sigma = slope / sqrt(W L mult)  (e.g. nfet vth0: 3.356 mV um)."""
    sec = cfg["corner"] + ("_mm" if mismatch else "")
    return f'.lib "{cfg["model_lib"]}" {sec}'


def make_envelope_netlist(cfg, temps, vgrid, cell=CELL, f_noise=1e7):
    """Netlist for the noise envelope.  The NOMINAL cell (no mismatch) with q
    and qb shorted (node m) sits at its metastable point.  For each temperature
    and each VDD in vgrid:  op (supply current, device gm/id), ac (|Z| of node m
    for a 1 A test current) and noise (BSIM4 output noise at m), all at the
    single frequency f_noise = 10 MHz.  There the current noise is white:
    flicker noise is <2 % of it and the rise above ~30 MHz caused by the
    BSIM4 body-resistor network is <6 % (spectra checked 100 Hz - 1 GHz at
    -40/25/125 C); 10 MHz is also within the node bandwidth at the decision."""
    # internal names of the BSIM4 devices inside the PDK subcircuits of inverter A of cell X0, to read gm and id
    nd, pd = "@m.x0.xpda.msky130_fd_pr__nfet_01v8", "@m.x0.xpua.msky130_fd_pr__pfet_01v8"
    L = ["* SKY130 6T SRAM: metastable-point noise vs VDD and T (noise envelope)",
         model_include(cfg, mismatch=False),
         cell_subckt(cell, noise=False),
         f"Vdd vdd 0 {POWERUP['vdd']}",
         "Venv env 0 0",
         # ports vdd bl blb wl q qb env: bit lines at VDD, word line off, q and qb both on node m
         "X0 vdd vdd vdd 0 m m env sram6t",
         "Iac 0 m dc 0 ac 1",
         ".control"]
    # per (T, VDD): op (bias point), ac (|Z| of node m), noise (output noise at m), then one ENV line in the log
    # ($&op1.x prints vector x of plot op1); destroy all frees the plots so the next point is again op1, ac1, noise1
    for k, T in enumerate(temps):
        L.append(f"option temp={T:g}")
        for j, v in enumerate(vgrid):
            L += [f"alter vdd dc={v:.4g}", "op",
                  f"let itot=abs(i(vdd))", f"let gmn=abs({nd}[gm])", f"let idn=abs({nd}[id])",
                  f"let gmp=abs({pd}[gm])", f"let idp=abs({pd}[id])", "let vm=v(m)",
                  f"ac lin 1 {f_noise:g} {f_noise:g}", "let zm=abs(v(m))",
                  f"noise v(m) Iac lin 1 {f_noise:g} {f_noise:g}", "setplot noise1",
                  "let on=onoise_spectrum",
                  f"echo ENV {k} {j} $&op1.itot $&op1.vm $&op1.gmn $&op1.idn $&op1.gmp "
                  "$&op1.idp $&ac1.zm $&on",
                  "destroy all"]
    L += [".endc", ".end", ""]
    return "\n".join(L)


def noise_envelope(cfg, temps, cell=CELL, nt=POWERUP["nt"], noise_scale=1.0,
                   vgrid=None, name="envelope"):
    """Noise model and its magnitude (returns per-temperature tables).

    Physics: before the cell decides, both storage nodes sit at the metastable
    point V_M(VDD) of the cross-coupled pair; at the VDD where the cell resolves
    (~0.5-0.8 V) the transistors are in weak / moderate inversion, carrying
    pA-nA.  Their channel (thermal) noise is the only fundamental noise left in
    an otherwise deterministic power-up.  We take its PSD from the PDK's own
    BSIM4 noise model: for the nominal cell with q and qb shorted (node m), the
    white current-noise PSD injected into m is

        S_m(VDD, T) = onoise(m)^2 / |Z_m|^2     at f = 10 MHz,

    and each storage node gets half of it, S_node = S_m / 2 (node m collects
    both halves of the cell).  For comparison the weak-inversion shot-noise
    estimate is S_node = 2 q I_VDD (each node is driven by one PMOS and one
    NMOS carrying the same current); the two agree within ~0.6-1.4x in the
    VDD window where cells decide (see RESULTS.md).  Integrated
    on the ~2 fF node capacitance this is a kT/C-type noise of order 1 mV rms.

    Implementation: the amplitude env = sqrt(noise_scale * S_node / (2 nt))
    multiplies a unit Gaussian sample-and-hold source with sample time nt; its
    one-sided PSD at low frequency is 2 nt (verified with an RC filter:
    output variance = nt / (2 R C)), so the node sees PSD = noise_scale * S_node
    up to ~1/(2 nt).  The envelope is tabulated vs VDD (log-interpolated in the
    netlist) per temperature.

    Approximations: the PSD is that of the nominal cell at its metastable bias
    (each real cell's currents differ by its mismatch: a +-10-20 % effect on
    the noise rms); after the decision the injected noise no longer matters;
    1/f noise and RTN are not included (use noise_scale for sensitivity)."""
    if vgrid is None:
        # 0.05 to 1.8 V in 50 mV steps
        vgrid = np.round(np.arange(0.05, POWERUP["vdd"] + 1e-9, 0.05), 3)
    wd = os.path.join(cfg["workdir"], name)
    os.makedirs(wd, exist_ok=True)
    log, dt = run_ngspice(cfg["ngspice"], make_envelope_netlist(cfg, temps, vgrid, cell), wd, name)
    # ENV lines: indices (k, j), then itot, vm, gmn, idn, gmp, idp, |Z_m| and the output noise at m
    rows = {}
    for line in log.splitlines():
        if line.startswith("ENV "):
            f = line.split()
            rows[(int(f[1]), int(f[2]))] = [float(x) for x in f[3:]]
    out = dict(temps=list(temps), nt=nt, noise_scale=noise_scale, runtime_s=dt, tables=[])
    for k, T in enumerate(temps):
        a = np.array([rows[(k, j)] for j in range(len(vgrid))])
        itot, vm, gmn, idn, gmp, idp, zm, on = a.T
        s_node = 0.5 * (on / zm) ** 2               # BSIM4 white channel noise per storage node
        s_shot = 2 * Q_E * itot                     # weak-inversion shot-noise estimate
        amp = np.sqrt(noise_scale * s_node / (2 * nt))
        # gm UT / id is about 1/n (0.6 to 0.8) in weak inversion and smaller in strong inversion: it records the
        # operating region of each device at the metastable point
        ut = K_B * (T + 273.15) / Q_E
        out["tables"].append(dict(
            T=T, vdd=list(map(float, vgrid)), vm=vm.tolist(), i_tot=itot.tolist(),
            s_node=s_node.tolist(), s_shot=s_shot.tolist(), amp=amp.tolist(),
            gm_ut_over_id_n=(gmn * ut / np.maximum(idn, 1e-30)).tolist(),
            gm_ut_over_id_p=(gmp * ut / np.maximum(idp, 1e-30)).tolist()))
    return out


def _env_pwl(table, n_pts=19, vmin=0.05):
    """B-source expression exp(pwl(v(vdd), V_k, ln amp_k, ...)) on a coarse grid."""
    vdd = np.asarray(table["vdd"])
    amp = np.maximum(np.asarray(table["amp"]), 1e-30)
    # the amplitude spans decades over VDD, so it is interpolated in ln(amp) and exponentiated in the netlist; the
    # coarse grid keeps the B-source expression short (it is evaluated at every time step)
    grid = np.linspace(vmin, vdd[-1], n_pts)
    la = np.interp(grid, vdd, np.log(amp))
    pts = ", ".join(f"{v:.4g},{a:.5g}" for v, a in zip(grid, la))
    return f"exp(pwl(v(vdd), {pts}))"


def make_mc_netlist(cfg, n_cells, temps, repeats, mismatch_seed, noise_seed0,
                    envelope=None, cell=CELL, pu=POWERUP, out_prefix="mc",
                    reference=True):
    """Monte Carlo netlist: n_cells independent cells in one circuit.

    The .control block runs, for each temperature (index k) and repeat r:
        option temp=T ; setseed S(k, r) ; tran ... ; wrdata <final q, qb>
    All runs reuse the circuit parsed once, so the mismatch draws are fixed;
    only the noise sources (trrandom, re-seeded by setseed) change.
    With reference=True an extra noise-off run at 25 C is done first (all
    envelope selectors at 0): its bits are deterministic and are used to check
    mismatch reproducibility between processes / ngspice versions.
    envelope=None -> noiseless netlist (no noise sources at all)."""
    noise = envelope is not None
    nt = pu["nt"]
    td = pu["t_ramp"] * pu["v_noise_on"] / pu["vdd"]      # noise starts at VDD = v_noise_on
    # .option seed fixes the PDK's mismatch draws (one AGAUSS per device parameter per instance); the supply ramps
    # linearly from 0 V and the word line stays at 0 V
    L = [f"* SKY130 6T SRAM PUF Monte Carlo: {n_cells} cells, seed {mismatch_seed}",
         f".option seed={int(mismatch_seed)}",
         model_include(cfg, mismatch=True),
         cell_subckt(cell, noise=noise, nt=nt, td=td),
         f"Vdd vdd 0 PWL(0 0 {pu['t_ramp']:g} {pu['vdd']:g})",
         "Vwl wl 0 0"]
    if noise:
        # one envelope per temperature in a single B source, env = sum_k v(sel_k) amp_k(VDD): the control block sets
        # selector k to 1 and the others to 0 with `alter`, which switches the noise level in the circuit parsed
        # once, so the mismatch draws stay the same
        tabs = {t["T"]: t for t in envelope["tables"]}
        terms = []
        for k, T in enumerate(temps):
            L.append(f"Vsel{k} sel{k} 0 dc 0")
            terms.append(f"v(sel{k})*{_env_pwl(tabs[T])}")
        L.append("Benv env 0 V=" + " + ".join(terms))
    else:
        L.append("Venv env 0 0")
    for i in range(n_cells):
        # bit lines precharged: BL = BLB = VDD (tied to the ramping supply)
        L.append(f"X{i} vdd vdd vdd wl q{i} qb{i} env sram6t")
    # only the storage nodes are saved (keeps the output small)
    vecs = " ".join(f"v(q{i}) v(qb{i})" for i in range(n_cells))
    L.append(".save " + vecs)

    def tran(T):
        t_stop, t_max = tran_settings(T, pu)
        # only the last few ns are stored (tstart); uic: start from all nodes at 0 V
        return f"tran 1n {t_stop:g} {t_stop - 3e-9:g} {t_max:g} uic"

    if pu.get("options"):
        L.append(".options " + pu["options"])          # tightened tolerances (convergence check); none by default
    L += [".control", "set wr_singlescale", "set wr_vecnames"]
    # wrdata format: a header line of names and one shared time column (read_wrdata)
    if reference:
        L += ["option temp=25"]
        if noise:
            L += [f"alter vsel{j} dc=0" for j in range(len(temps))]
        L += ["setseed 1", tran(25), f"wrdata {out_prefix}_ref.txt {vecs}", "destroy all"]
    for k, T in enumerate(temps):
        L.append(f"option temp={T:g}")
        if noise:
            L += [f"alter vsel{j} dc={1 if j == k else 0}" for j in range(len(temps))]
        # one power-up per repeat, each with its own noise seed (up to 1000 repeats per temperature), so every read
        # gets fresh but reproducible noise; destroy all frees the stored vectors of the previous run
        for r in range(repeats):
            L += [f"setseed {int(noise_seed0 + 1000 * k + r + 1)}", tran(T),
                  f"wrdata {out_prefix}_t{k}_r{r}.txt {vecs}", "destroy all"]
    L += [".endc", ".end", ""]
    return "\n".join(L)


def _final_bits(path, n_cells, vdd_final, t_stop=None):
    # last time point; after the time column the vectors alternate q_i, qb_i (the order of the wrdata list)
    names, d = read_wrdata(path)
    if t_stop is not None and d[-1, 0] < t_stop * (1 - 1e-6):
        raise RuntimeError(f"{path}: the power-up ends at {d[-1, 0]:.4g} s, before its stop time {t_stop:.4g} s")
    last = d[-1, 1:]
    q, qb = last[0::2], last[1::2]
    assert q.size == n_cells
    # the bit a cell settles to: 1 if q ends above qb
    bits = (q > qb).astype(np.int8)
    # resolution margin: |q - qb| relative to the supply at t_stop
    margin = np.abs(q - qb) / vdd_final
    return bits, margin, q - qb


def run_chunk(cfg, n_cells, temps, repeats, mismatch_seed, noise_seed0, envelope,
              name, cell=CELL, pu=POWERUP, reference=True):
    """Run one Monte Carlo netlist; return bits[T, R, N] and diagnostics."""
    wd = os.path.join(cfg["workdir"], name)
    os.makedirs(wd, exist_ok=True)
    net = make_mc_netlist(cfg, n_cells, temps, repeats, mismatch_seed, noise_seed0,
                          envelope, cell, pu, out_prefix="mc", reference=reference)
    _, dt = run_ngspice(cfg["ngspice"], net, wd, name)
    # the supply at the stop time (the ramp may not have reached its final value), the scale of the margin
    vend = lambda T: pu["vdd"] * min(1.0, tran_settings(T, pu)[0] / pu["t_ramp"])
    # bits of every (temperature, repeat) run and the smallest |q - qb| / VDD of all reads (a small margin would
    # mean a cell not yet resolved at the stop time)
    bits = np.zeros((len(temps), repeats, n_cells), np.int8)
    min_margin = np.inf
    for k, T in enumerate(temps):
        for r in range(repeats):
            b, m, _ = _final_bits(os.path.join(wd, f"mc_t{k}_r{r}.txt"), n_cells, vend(T), tran_settings(T, pu)[0])
            bits[k, r] = b
            min_margin = min(min_margin, m.min())
    res = dict(bits=bits, runtime_s=dt, min_margin=float(min_margin), seed=mismatch_seed)
    if reference:
        b, m, dq = _final_bits(os.path.join(wd, "mc_ref.txt"), n_cells, vend(25), tran_settings(25, pu)[0])
        res.update(ref_bits=b, ref_dq=dq)
    return res


def make_wave_netlist(cfg, n_cells, temps, mismatch_seed, envelope=None, cell=CELL,
                      pu=POWERUP, noise_seed=1, t_stop=None, out_prefix="wave", repeats=1):
    """Full-waveform power-up of a few cells (for the figure and for finding the
    decision window).  Same cell, same `.option seed` convention as the MC."""
    noise = envelope is not None
    nt = pu["nt"]
    td = pu["t_ramp"] * pu["v_noise_on"] / pu["vdd"]
    t_stop = t_stop or max(tran_settings(T, pu)[0] for T in temps)
    L = [f"* SKY130 6T SRAM power-up waveforms: {n_cells} cells, seed {mismatch_seed}",
         f".option seed={int(mismatch_seed)}",
         model_include(cfg, mismatch=True),
         cell_subckt(cell, noise=noise, nt=nt, td=td),
         f"Vdd vdd 0 PWL(0 0 {pu['t_ramp']:g} {pu['vdd']:g})",
         "Vwl wl 0 0"]
    if noise:
        # the same per-temperature envelope selectors as in make_mc_netlist
        tabs = {t["T"]: t for t in envelope["tables"]}
        terms = []
        for k, T in enumerate(temps):
            L.append(f"Vsel{k} sel{k} 0 dc 0")
            terms.append(f"v(sel{k})*{_env_pwl(tabs[T])}")
        L.append("Benv env 0 V=" + " + ".join(terms))
    else:
        L.append("Venv env 0 0")
    for i in range(n_cells):
        L.append(f"X{i} vdd vdd vdd wl q{i} qb{i} env sram6t")
    vecs = "v(vdd) " + " ".join(f"v(q{i}) v(qb{i})" for i in range(n_cells))
    L.append(".save " + vecs)
    L += [".control", "set wr_singlescale", "set wr_vecnames"]
    for k, T in enumerate(temps):
        tmax = tran_settings(T, pu)[1]
        L.append(f"option temp={T:g}")
        if noise:
            L += [f"alter vsel{j} dc={1 if j == k else 0}" for j in range(len(temps))]
        # the whole waveform from t = 0 is stored (the Monte Carlo runs keep only the last 3 ns)
        for r in range(repeats):
            L += [f"setseed {int(noise_seed + r)}", f"tran 1n {t_stop:g} 0 {tmax:g} uic",
                  f"wrdata {out_prefix}_t{k}_r{r}.txt {vecs}", "destroy all"]
    L += [".endc", ".end", ""]
    return "\n".join(L)


def run_waveforms(cfg, n_cells, temps, mismatch_seed, envelope=None, name="waves",
                  cell=CELL, pu=POWERUP, noise_seed=1, t_stop=None, repeats=1):
    """Returns (list over temperatures of list over repeats of
    dict(T, t, vdd, q[N, t], qb[N, t]), runtime)."""
    wd = os.path.join(cfg["workdir"], name)
    os.makedirs(wd, exist_ok=True)
    net = make_wave_netlist(cfg, n_cells, temps, mismatch_seed, envelope, cell, pu,
                            noise_seed, t_stop, repeats=repeats)
    _, dt = run_ngspice(cfg["ngspice"], net, wd, name)
    out = []
    for k, T in enumerate(temps):
        reps = []
        for r in range(repeats):
            names, d = read_wrdata(os.path.join(wd, f"wave_t{k}_r{r}.txt"))
            # columns: time, v(vdd), then q_i and qb_i alternating
            reps.append(dict(T=T, t=d[:, 0], vdd=d[:, 1], q=d[:, 2::2].T, qb=d[:, 3::2].T))
        out.append(reps)
    return out, dt


def decision_vdd(w, frac=0.5):
    """VDD at which |q - qb| first exceeds frac * VDD, per cell; a cell that never resolves is an error, not NaN."""
    dq = np.abs(w["q"] - w["qb"])
    ok = dq > frac * np.maximum(w["vdd"], 1e-3)
    if not ok.any(axis=1).all():
        raise RuntimeError(f"cells {np.flatnonzero(~ok.any(axis=1)).tolist()} never resolve within the simulated power-up")
    return w["vdd"][ok.argmax(axis=1)]           # argmax of a boolean row: the index of its first True


# ----------------------------------------------------------------------------
# Monte Carlo driver
# ----------------------------------------------------------------------------

def run_montecarlo(cfg, n_cells, temps, repeats, envelope, chunk=64, mismatch_seed=1,
                   noise_seed=1, cell=CELL, pu=POWERUP, name="mc", log=print):
    """Power up n_cells cells `repeats` times at each temperature.

    The cells are split into chunks of `chunk` cells; chunk c is one netlist /
    one ngspice process with `.option seed=mismatch_seed + c` (so the cell
    population depends only on n_cells, chunk and mismatch_seed, not on the
    number of CPUs).  Chunks run in parallel, cfg['jobs'] at a time.
    Returns dict(bits[T, R, N] int8, temps, repeats, ...)."""
    # full chunks and a shorter last one
    sizes = [min(chunk, n_cells - s) for s in range(0, n_cells, chunk)]
    t0 = time.time()

    # chunk c: mismatch seed mismatch_seed + c; its noise seeds are offset by 100000 c, disjoint from the other
    # chunks' (each chunk uses fewer than 100000 seeds)
    def one(c):
        return run_chunk(cfg, sizes[c], temps, repeats, mismatch_seed + c,
                         noise_seed + 100000 * c, envelope, f"{name}_{c:03d}", cell, pu)

    # the threads only wait on their ngspice processes, so cfg['jobs'] chunks run at once; map keeps the chunk order
    with ThreadPoolExecutor(max_workers=cfg["jobs"]) as ex:
        res = list(ex.map(one, range(len(sizes))))
    wall = time.time() - t0
    # the chunks joined along the cell axis
    bits = np.concatenate([r["bits"] for r in res], axis=2)
    out = dict(bits=bits, temps=list(temps), repeats=repeats, n_cells=n_cells, chunk=chunk,
               mismatch_seed=mismatch_seed, noise_seed=noise_seed, wall_s=wall,
               cpu_s=float(sum(r["runtime_s"] for r in res)),
               min_margin=float(min(r["min_margin"] for r in res)),
               ref_bits=np.concatenate([r["ref_bits"] for r in res]),
               ref_dq=np.concatenate([r["ref_dq"] for r in res]))
    log(f"  MC: {n_cells} cells x {repeats} repeats x {len(temps)} temps in {len(sizes)} "
        f"chunks: {wall:.1f} s wall, {out['cpu_s']:.1f} s ngspice CPU, "
        f"min resolution margin {out['min_margin']:.3f}")
    return out


# ----------------------------------------------------------------------------
# Statistics of the SPICE results
# ----------------------------------------------------------------------------

def analyse(bits, temps, t_enrol=25, n_boot=1000, seed=0):
    """PUF metrics per temperature from bits[T, R, N].

    Enrolment = the FIRST power-up at t_enrol (like enroll() in the behavioural
    model, which reads the PUF once at 25 C).  BER(T) = fraction of later reads
    at T that differ from the enrolled bit (at t_enrol the enrolment read itself
    is excluded).  Also given vs a majority-vote ("golden") 25 C reference.
    Confidence intervals: bootstrap over cells (reads of one cell are correlated)."""
    bits = np.asarray(bits)
    nT, R, N = bits.shape
    # enrolled bits: the first read at t_enrol; golden reference: the majority of all reads at t_enrol
    k0 = list(temps).index(t_enrol)
    w = bits[k0, 0]
    golden = (bits[k0].mean(0) > 0.5).astype(np.int8)
    # n_boot resamples of the cells (with replacement), shared by all temperatures
    rng = np.random.default_rng(seed)
    boot_idx = rng.integers(0, N, size=(n_boot, N))
    rows = []
    for k, T in enumerate(temps):
        reads = bits[k, 1:] if k == k0 else bits[k]
        err = (reads != w).mean(0)             # per-cell error rate vs enrolment
        err_g = (bits[k] != golden).mean(0)
        p1 = bits[k].mean(0)
        b_err = err[boot_idx].mean(1)
        rows.append(dict(
            T=T, bias=float(p1.mean()),
            bias_se=float(p1.std(ddof=1) / math.sqrt(N)),
            ber=float(err.mean()), ber_ci=[float(np.percentile(b_err, 2.5)),
                                           float(np.percentile(b_err, 97.5))],
            ber_golden=float(err_g.mean()),
            unstable=float(((p1 > 0) & (p1 < 1)).mean()),     # flipped at least once in R reads
            n_unstable=int(((p1 > 0) & (p1 < 1)).sum()),
            flipped_vs_enrol=float((np.round(p1) != w).mean()), # majority at T differs from enrolment
            p1=p1.tolist()))
    return dict(n_cells=N, repeats=R, temps=list(temps), t_enrol=t_enrol, rows=rows)


# ----------------------------------------------------------------------------
# Behavioural model and its fit
# ----------------------------------------------------------------------------
#   bit_i(T) = [ m_i + s(T) * n > 0 ],   m_i ~ N(mu, 1) fixed per cell,
#   n ~ N(0, 1) fresh at every power-up,  s(T) = s25 * (1 + a |T - 25|),
#   so P(bit_i = 1 | m_i) = Phi(m_i / s(T)) and the population bias is
#   P(1) = Phi(mu / sqrt(1 + s^2)) ~ Phi(mu).

def s_of_T(T, s25, a):
    return s25 * (1 + a * np.abs(np.asarray(T, float) - 25.0))


def _patterns(bits):
    """Collapse cells with identical per-temperature counts of ones."""
    counts = np.asarray(bits).sum(axis=1).T          # [N, T]
    pats, mult = np.unique(counts, axis=0, return_counts=True)
    return pats, mult


def _cell_loglik(s_T, mu, pats, R):
    """log P(pattern | mu, s_T) for each count pattern: the integral over the
    latent m of N(m; mu, 1) * prod_T Binom(k_T; R, Phi(m / s_T)).
    Outside |m| < c = 9 max(s_T) every read is deterministic (Phi = 0 or 1 to
    < 1e-19), so only the all-zero / all-one patterns get the analytic tails
    P(m < -c), P(m > c).  Inside, the trapezoid grid is the union of grids
    with step s_T/12 over |m| < 9 s_T for every temperature (so it resolves
    every Phi(m/s_T), whatever the spread of s_T), at most ~1000 points."""
    # integration grid in m: 217 points over |m| < 9 s_T for every temperature (step s_T / 12) and 217 over
    # |m| < c, merged and sorted
    s_T = np.asarray(s_T, float)
    c = 9.0 * float(s_T.max())
    u = np.linspace(-9.0, 9.0, 217)
    m = np.unique(np.concatenate([u * st for st in s_T] + [np.linspace(-c, c, 217)]))
    # trapezoid weights of the non-uniform grid, times the prior N(m; mu, 1), in logs
    dm = np.diff(m)
    wts = np.zeros_like(m)
    wts[:-1] += dm / 2
    wts[1:] += dm / 2
    logw = stats.norm.logpdf(m, mu, 1.0) + np.log(wts)
    # ln prod_T Phi(m/s_T)^k_T (1 - Phi(m/s_T))^(R - k_T) for every pattern and grid point (log_ndtr stays
    # accurate far in the tails, where Phi itself rounds to 0 or 1)
    acc = np.zeros((pats.shape[0], m.size))
    for t in range(len(s_T)):
        z = m / s_T[t]
        l1, l0 = special.log_ndtr(z), special.log_ndtr(-z)
        k = pats[:, t:t + 1]
        acc += k * l1[None, :] + (R - k) * l0[None, :]
    # sum over the grid with the largest term factored out (log-sum-exp)
    a = acc + logw[None, :]
    amax = a.max(axis=1, keepdims=True)
    lik = np.exp(amax[:, 0]) * np.exp(a - amax).sum(axis=1)
    # the analytic tails beyond |m| = c, where every read is 0 (m < -c) or 1 (m > c)
    all0 = (pats == 0).all(axis=1)
    all1 = (pats == R).all(axis=1)
    lik = lik + all0 * stats.norm.cdf(-c - mu) + all1 * stats.norm.sf(c - mu)
    ll = np.log(np.maximum(lik, 1e-300))
    # binomial coefficients (constant in the parameters, kept for a proper likelihood)
    ll += (special.gammaln(R + 1) - special.gammaln(pats + 1)
           - special.gammaln(R - pats + 1)).sum(axis=1)
    return ll


def _nll_linear(theta, temps, pats, mult, R):
    # theta = (mu, ln s25, b) with s(T) = s25 + b |T - 25|  (b = s25 * a; this
    # parameterisation stays well conditioned when s25 is small)
    mu, log_s25, b = theta
    s_T = math.exp(log_s25) + b * np.abs(np.asarray(temps, float) - 25.0)
    if np.any(s_T <= 1e-4):
        return 1e12
    return -float((mult * _cell_loglik(s_T, mu, pats, R)).sum())


def _nll_free(theta, temps, pats, mult, R):
    # theta = (mu, ln s_T1, ..., ln s_Tn): a free noise level at every temperature (the log keeps it positive)
    mu, log_s = theta[0], np.asarray(theta[1:])
    return -float((mult * _cell_loglik(np.exp(log_s), mu, pats, R)).sum())


def _hessian(f, x, rel=1e-3):
    # numerical Hessian of f at x by central differences, step rel |x_i| (at least 1e-4) in coordinate i
    x = np.asarray(x, float)
    n = x.size
    hstep = np.maximum(np.abs(x) * rel, 1e-4)
    H = np.zeros((n, n))
    f0 = f(x)
    for i in range(n):
        for j in range(i, n):
            ei = np.zeros(n); ej = np.zeros(n)
            ei[i] = hstep[i]; ej[j] = hstep[j]
            # diagonal: the second difference; off the diagonal: the four-point mixed difference
            if i == j:
                H[i, i] = (f(x + ei) - 2 * f0 + f(x - ei)) / hstep[i] ** 2
            else:
                H[i, j] = H[j, i] = (f(x + ei + ej) - f(x + ei - ej) - f(x - ei + ej)
                                     + f(x - ei - ej)) / (4 * hstep[i] * hstep[j])
    return H


def fit_model(bits, temps, n_boot=100, seed=0, x0=(0.0, math.log(0.03), 3e-4), log=print):
    """Maximum-likelihood fit of (mu, s25, a) to bits[T, R, N].

    Every cell keeps ONE latent m_i across all temperatures and repeats (as in
    the behavioural model, where the mismatch is fixed and only s(T) changes),
    so cells that are stable at 25 C but flip at another temperature are what
    determine `a`.  Uncertainty: (i) inverse Hessian of the negative
    log-likelihood (1 sigma, delta method for s25), (ii) bootstrap over cells
    (multinomial re-weighting of the count patterns) -> 16/84 % and 2.5/97.5 %.
    Also returns a fit with a free s_T at every temperature (shared m_i) to
    show whether the linear-in-|T-25| form is adequate."""
    bits = np.asarray(bits)
    R = bits.shape[1]
    temps = np.asarray(temps, float)
    # a cell enters the likelihood only through its count of ones at each temperature, so every distinct count
    # pattern is evaluated once and weighted by the number of cells that show it
    pats, mult = _patterns(bits)

    # Nelder-Mead on the negative log-likelihood for given pattern weights (the bootstrap passes re-drawn weights)
    def fit(weights, start, tol=1e-5):
        f = lambda th: _nll_linear(th, temps, pats, weights, R)
        r = optimize.minimize(f, start, method="Nelder-Mead",
                              options=dict(xatol=tol, fatol=tol, maxiter=4000))
        return r.x.copy(), r.fun

    th, nll = fit(mult, np.array(x0, float))
    # covariance of (mu, ln s25, b): the inverse Hessian of the negative log-likelihood at the optimum
    f = lambda t: _nll_linear(t, temps, pats, mult, R)
    try:
        cov = np.linalg.inv(_hessian(f, th))
        se = np.sqrt(np.maximum(np.diag(cov), 0))
    except np.linalg.LinAlgError:
        cov, se = None, np.full(3, np.nan)
    mu, s25, b = float(th[0]), math.exp(th[1]), float(th[2])
    a = b / s25
    # delta method for a = b / s25 (cov of (ln s25, b))
    a_se = float("nan")
    if cov is not None:
        g = np.array([-a, 1.0 / s25])
        a_se = float(math.sqrt(max(g @ cov[1:, 1:] @ g, 0)))
    # p1: the population's P(bit = 1); s25_se by the delta method, d s25 = s25 d(ln s25)
    out = dict(mu=mu, s25=s25, a=a, b=b, p1=float(stats.norm.cdf(mu / math.sqrt(1 + s25 ** 2))),
               mu_se=float(se[0]), s25_se=float(s25 * se[1]), b_se=float(se[2]), a_se=a_se,
               nll=nll, n_cells=int(mult.sum()), n_patterns=int(len(mult)), repeats=R,
               temps=temps.tolist())
    # bootstrap over cells: re-draw the number of cells of every count pattern (multinomial), refit from the optimum
    rng = np.random.default_rng(seed)
    bs = []
    for _ in range(n_boot):
        wts = rng.multinomial(int(mult.sum()), mult / mult.sum())
        tb, _ = fit(wts, th, tol=1e-3)
        bs.append([tb[0], math.exp(tb[1]), tb[2], tb[2] / math.exp(tb[1])])
    if bs:
        bs = np.array(bs)
        for i, k in enumerate(["mu", "s25", "b", "a"]):
            out[k + "_boot68"] = np.percentile(bs[:, i], [16, 84]).tolist()
            out[k + "_boot95"] = np.percentile(bs[:, i], [2.5, 97.5]).tolist()
    # free s_T per temperature (shared latent m_i), started from the linear fit
    g = lambda t: _nll_free(t, temps, pats, mult, R)
    x0f = np.r_[th[0], np.log(np.maximum(s25 + b * np.abs(temps - 25), 1e-3))]
    rf = optimize.minimize(g, x0f, method="Nelder-Mead",
                           options=dict(xatol=1e-5, fatol=1e-6, maxiter=8000))
    try:
        sef = np.sqrt(np.maximum(np.diag(np.linalg.inv(_hessian(g, rf.x))), 0))
    except np.linalg.LinAlgError:
        sef = np.full(rf.x.size, np.nan)
    sT = np.exp(rf.x[1:])
    out["free"] = dict(mu=float(rf.x[0]), s_T=sT.tolist(), s_T_se=(sT * sef[1:]).tolist(),
                       nll=float(rf.fun))
    # likelihood-ratio test: linear s(T) vs free s_T (df = nT - 2)
    lr = 2 * (nll - rf.fun)
    out["lr_linear_vs_free"] = dict(stat=float(lr), df=int(len(temps) - 2),
                                    p=float(stats.chi2.sf(max(lr, 0), max(len(temps) - 2, 1))))
    log(f"  ML fit: mu={mu:+.3f}+-{out['mu_se']:.3f}  s25={s25:.4f}+-{out['s25_se']:.4f}  "
        f"b={b:.2e}+-{out['b_se']:.1e}/C  a=b/s25={a:.4f}  (P(1)={out['p1']:.3f}); "
        f"LR test linear vs free s(T): p={out['lr_linear_vs_free']['p']:.3g}")
    return out


def _bvn_lower(h, k, rho):
    """P(Z1 < h, Z2 < k) for a standard bivariate normal with correlation rho
    (Owen's T formulation, exact)."""
    # h = k = 0: the orthant probability (Sheppard's formula)
    if abs(h) < 1e-12 and abs(k) < 1e-12:
        return 0.25 + math.asin(rho) / (2 * math.pi)
    # Owen (1956): P = Phi(h)/2 + Phi(k)/2 - T(h, a_h) - T(k, a_k) - beta, with a_h = (k - rho h) / (h sqrt(1 - rho^2)),
    # a_k likewise, and beta = 1/2 if h k < 0, or if one of them is 0 and h + k < 0 (else 0)
    r = math.sqrt(max(1 - rho * rho, 1e-300))
    def tt(x, y):
        return 0.0 if x == 0 else float(special.owens_t(x, (y - rho * x) / (x * r)))
    beta = 0.0 if (h * k > 0 or (h * k == 0 and h + k >= 0)) else 0.5
    return 0.5 * stats.norm.cdf(h) + 0.5 * stats.norm.cdf(k) - tt(h, k) - tt(k, h) - beta


def model_ber(temps, s25, a, mu, t_enrol=25):
    """Expected BER of the behavioural model vs a single enrolment read at t_enrol:
    X = m + s(t_enrol) n0 (enrolment), Y = m + s(T) n1 (later read); m ~ N(mu, 1).
    BER = P(X<0) + P(Y<0) - 2 P(X<0, Y<0)  (bivariate normal, corr 1/(sx sy))."""
    out = []
    s0 = float(s_of_T(t_enrol, s25, a))
    for T in np.atleast_1d(temps):
        s1 = float(s_of_T(T, s25, a))
        # X and Y share m: var X = 1 + s0^2, var Y = 1 + s1^2, cov(X, Y) = var m = 1; standardised, X < 0 and Y < 0
        # become Z1 < -mu / sx and Z2 < -mu / sy with correlation 1 / (sx sy)
        sx, sy = math.sqrt(1 + s0 ** 2), math.sqrt(1 + s1 ** 2)
        h, k = -mu / sx, -mu / sy
        pxy = _bvn_lower(h, k, 1.0 / (sx * sy))
        out.append(float(stats.norm.cdf(h) + stats.norm.cdf(k) - 2 * pxy))
    return np.array(out)


def fit_ber_matched(bits, temps, t_enrol=25, n_boot=200, seed=0, log=print):
    """Second fit: choose (s25, a) so that the model's BER(T) vs a single 25 C
    enrolment matches the SPICE BER(T) (weighted least squares, weights from
    the bootstrap s.e. of each BER); mu from the 25 C bias, mu = Phi^-1(P(1)).
    This reproduces what matters for the fuzzy extractor (the error rate) even
    when the full count statistics are not noise-like (see fit_model's LR test).
    Also returns s_eff(T): the s that alone reproduces BER(T) at each T
    (s at 25 C from the 25 C repeat BER).  Uncertainty: bootstrap over cells."""
    bits = np.asarray(bits)
    nT, R, N = bits.shape
    temps = list(temps)
    k0 = temps.index(t_enrol)

    # BER(T) against the first read at t_enrol, and the bias at t_enrol, of the cells idx (the bootstrap passes
    # resampled indices)
    def ber_bias(idx):
        b = bits[:, :, idx]
        w = b[k0, 0]
        ber = np.array([((b[k, 1:] if k == k0 else b[k]) != w).mean() for k in range(nT)])
        return ber, b[k0].mean()

    # s_eff(T): the single noise level that, used for both reads (a = 0), gives the SPICE BER at T; solved in ln s
    # with brentq on 1e-5 to 5 (NaN if there is no root in that range)
    def s_eff(ber, mu):
        out = []
        for k, T in enumerate(temps):
            if k == k0:   # both reads at 25 C: BER(s, s) = ber
                f = lambda ls: model_ber([T], math.exp(ls), 0.0, mu)[0] - ber[k]
            else:
                f = lambda ls: model_ber([T], math.exp(ls), 0.0, mu, t_enrol=T)[0] - ber[k]
            try:
                out.append(math.exp(optimize.brentq(f, math.log(1e-5), math.log(5.0))))
            except ValueError:
                out.append(float("nan"))
        return np.array(out)

    def fit2(ber, mu, sd, start):
        # th = (ln s25, ln b): s(T) = s25 + b |T - 25|, a = b / s25; the cost is the weighted sum of squares (chi^2)
        # of model minus SPICE BER
        def f(th):
            s25, b = math.exp(th[0]), math.exp(th[1])
            return float((((model_ber(temps, s25, b / s25, mu) - ber) / sd) ** 2).sum())
        r = optimize.minimize(f, start, method="Nelder-Mead",
                              options=dict(xatol=1e-5, fatol=1e-8, maxiter=2000))
        return r.x, r.fun

    rng = np.random.default_rng(seed)
    allidx = np.arange(N)
    ber0, bias0 = ber_bias(allidx)
    boots = [ber_bias(rng.integers(0, N, N)) for _ in range(n_boot)]
    # weights: the bootstrap s.e. of each BER, floored at half a count, 0.5 / (N R), so that a BER with no spread
    # does not get an infinite weight
    sd = np.maximum(np.std([b[0] for b in boots], axis=0, ddof=1), 0.5 / (N * R))
    # mu from the bias at t_enrol: P(1) = Phi(mu / sqrt(1 + s^2)), close to Phi(mu) for small s
    mu0 = float(stats.norm.ppf(np.clip(bias0, 1e-3, 1 - 1e-3)))
    th0, chi2 = fit2(ber0, mu0, sd, np.array([math.log(0.03), math.log(3e-4)]))
    # every resample refitted (with mu from its own bias), started from the full fit
    bs = []
    for ber_b, bias_b in boots:
        mub = float(stats.norm.ppf(np.clip(bias_b, 1e-3, 1 - 1e-3)))
        tb, _ = fit2(ber_b, mub, sd, th0)
        bs.append([mub, math.exp(tb[0]), math.exp(tb[1]), math.exp(tb[1] - tb[0])])
    bs = np.array(bs)
    # s_eff and its spread from at most 50 resamples (one root solve per temperature each)
    seff = s_eff(ber0, mu0)
    seff_b = np.array([s_eff(b[0], float(stats.norm.ppf(np.clip(b[1], 1e-3, 1 - 1e-3))))
                       for b in boots[:min(n_boot, 50)]])
    out = dict(mu=mu0, s25=float(math.exp(th0[0])), b=float(math.exp(th0[1])),
               a=float(math.exp(th0[1] - th0[0])), chi2=float(chi2),
               dof=int(nT - 2), ber=ber0.tolist(), ber_se=sd.tolist(),
               s_eff=seff.tolist(),
               s_eff_boot68=np.nanpercentile(seff_b, [16, 84], axis=0).T.tolist())
    for i, k in enumerate(["mu", "s25", "b", "a"]):
        out[k + "_se"] = float(np.std(bs[:, i], ddof=1))
        out[k + "_boot68"] = np.percentile(bs[:, i], [16, 84]).tolist()
        out[k + "_boot95"] = np.percentile(bs[:, i], [2.5, 97.5]).tolist()
    log(f"  BER-matched fit: mu={out['mu']:+.3f}+-{out['mu_se']:.3f}  s25={out['s25']:.4f} "
        f"[{out['s25_boot68'][0]:.4f}, {out['s25_boot68'][1]:.4f}]  b={out['b']:.2e}/C "
        f"[{out['b_boot68'][0]:.1e}, {out['b_boot68'][1]:.1e}]  a={out['a']:.4f}  chi2={chi2:.1f}/{nT-2}")
    log("  s_eff(T) reproducing BER(T): " + ", ".join(f"{T:g} C: {v:.4f}" for T, v in zip(temps, seff)))
    return out


def model_unstable(temps, s25, a, mu, R):
    """Model probability that a cell flips at least once in R reads at T."""
    # average over m ~ N(mu, 1) on a fine grid of +-8 sigma: a cell with P(1) = p is unstable unless all R reads
    # agree, 1 - p^R - (1 - p)^R
    m = np.linspace(mu - 8, mu + 8, 40001)
    w = stats.norm.pdf(m, mu, 1) * (m[1] - m[0])
    out = []
    for T in np.atleast_1d(temps):
        p = stats.norm.cdf(m / float(s_of_T(T, s25, a)))
        out.append(float((w * (1 - p ** R - (1 - p) ** R)).sum()))
    return np.array(out)


def model_count_dist(T, s25, a, mu, R):
    """Model distribution of k = number of ones in R reads of one cell at T."""
    # mixture of binomials over m ~ N(mu, 1) on |m| < 9 s; beyond it every read is 0 or 1, so the two tails add
    # to k = 0 and k = R
    s = float(s_of_T(T, s25, a))
    c = 9 * s
    m = np.linspace(-c, c, 4001)
    w = stats.norm.pdf(m, mu, 1) * (m[1] - m[0])
    p = stats.norm.cdf(m / s)
    k = np.arange(R + 1)
    pk = (w[None, :] * stats.binom.pmf(k[:, None], R, p[None, :])).sum(1)
    pk[0] += stats.norm.cdf(-c - mu)
    pk[R] += stats.norm.sf(c - mu)
    return pk


# ----------------------------------------------------------------------------
# Figures (matplotlib).  Sized for an IEEE two-column page (7.16 in) or one
# column (~3.5-4.8 in); fonts, ticks and grid come from the caller's rcParams
# (the notebook sets an IEEE style), so no font sizes are set here.
# ----------------------------------------------------------------------------
C1, C2, C3 = "#2a78d6", "#eb6834", "#1b9e77"     # blue, orange (+ 1 spare)
INK, CGRAY = "#131c28", "#7c8594"
INK2 = CGRAY


def _style(ax):
    """Kept for compatibility: the look comes from rcParams."""
    ax.set_axisbelow(True)


def _save(fig, path):
    # 300 dpi when a path is given (the format follows its extension)
    if path:
        fig.savefig(path, dpi=300, bbox_inches="tight")


def _minus(s):
    # the typographic minus sign (U+2212) for negative temperatures in titles
    return s.replace("-", "−")


def plot_waveforms(waves, cells=None, T_index=0, path=None):
    """Power-up waveforms: q and qb of a few cells diverge as VDD ramps."""
    import matplotlib.pyplot as plt
    # first repeat at the chosen temperature
    w = waves[T_index][0]
    n = w["q"].shape[0]
    cells = list(range(min(4, n))) if cells is None else cells
    fig, axs = plt.subplots(1, len(cells), figsize=(min(7.16, 1.85 * len(cells) + 0.3), 2.5),
                            sharex=True, sharey=True, layout="constrained")
    axs = np.atleast_1d(axs)
    t_us = w["t"] * 1e6
    for i, (ax, c) in enumerate(zip(axs, cells)):
        ax.plot(t_us, w["vdd"], color=CGRAY, lw=1.0, label=r"$V_{\mathrm{DD}}$")
        ax.plot(t_us, w["q"][c], color=C1, lw=1.5, label="q")
        ax.plot(t_us, w["qb"][c], color=C2, lw=1.5, ls="--", label="qb")
        # the bit the cell settles to (1 if q ends above qb)
        bit = int(w["q"][c, -1] > w["qb"][c, -1])
        ax.set_title(f"({chr(97 + i)}) Cell {c + 1}, bit {bit}")
        ax.set_xlabel("Time (µs)")
    axs[0].set_xlim(0, t_us[-1])
    axs[0].set_ylim(-0.05, max(w["vdd"].max(), w["q"].max(), w["qb"].max()) * 1.08)
    axs[0].set_ylabel("Voltage (V)")
    axs[0].legend(loc="upper left")
    _save(fig, path)
    return fig


def plot_p1_hist(bits, temps, fit=None, path=None):
    """Distribution of k = number of ones in R power-ups per cell, per temperature
    (bars: SPICE; dots: fitted behavioural model)."""
    import matplotlib.pyplot as plt
    bits = np.asarray(bits)
    nT, R, N = bits.shape
    fig, axs = plt.subplots(1, nT, figsize=(7.16, 2.85), sharey=True, layout="constrained")
    axs = np.atleast_1d(axs)
    k = np.arange(R + 1)
    y_lo = 0.5 / N
    for i, (ax, T) in enumerate(zip(axs, temps)):
        # fraction of cells with k ones in R reads; only the non-zero bars are drawn (log axis)
        cnt = np.bincount(bits[i].sum(0), minlength=R + 1) / N
        ax.bar(k[cnt > 0], cnt[cnt > 0], width=0.7, color=C1, alpha=0.85, label="SPICE MC")
        if fit is not None:
            pk = model_count_dist(T, fit["s25"], fit["a"], fit["mu"], R)
            y_lo = min(y_lo, 0.5 * pk.min())
            ax.plot(k, pk, "o", color=C2, ms=5, mec=INK, mew=0.6, label="ML-fitted model")
        ax.set_yscale("log")
        ax.set_xticks(k)
        ax.set_xlim(-0.6, R + 0.6)
        ax.grid(False, axis="x")
        # unstable cells: 0 < k < R
        unst = int(round(cnt[1:R].sum() * N))
        ax.set_title(_minus(f"({chr(97 + i)}) {T:g} °C"))
        ax.text(0.5, 0.95, f"{unst} of {N}\nunstable", transform=ax.transAxes, ha="center",
                va="top", color=INK)
    axs[0].set_ylim(y_lo, 1.0)
    axs[0].set_ylabel("Fraction of cells")
    fig.supxlabel(f"Number of ones $k$ in {R} power-ups", fontsize=plt.rcParams["axes.labelsize"])
    h, l = axs[0].get_legend_handles_labels()
    fig.legend(h[::-1], l[::-1], loc="outside upper right", ncol=2)
    _save(fig, path)
    return fig


def plot_ber(stats_rows, fit, temps, current=(0.06, 0.008, 0.0), fit_ber=None, spec=None, path=None):
    """BER vs temperature (enrolment = one power-up at 25 C): SPICE points with
    95 % bootstrap intervals, the fitted model and the model's current defaults.
    spec: (T_min, T_max) of the specified operating range; temperatures outside it are shaded."""
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(7.16, 2.8), layout="constrained")   # the legend to the right of the axes
    T = np.array([r["T"] for r in stats_rows])
    ber = np.array([r["ber"] for r in stats_rows])
    lo = np.array([r["ber_ci"][0] for r in stats_rows])
    hi = np.array([r["ber_ci"][1] for r in stats_rows])
    Tg = np.linspace(min(temps), max(temps), 121)
    par = lambda s25, a: rf"$s_{{25}}$ = {s25:.3g}, $a$ = {a:.3g} °C$^{{-1}}$"
    ax.errorbar(T, 100 * ber, yerr=[100 * (ber - lo), 100 * (hi - ber)], fmt="o", ms=5.5,
                color=INK, ecolor=INK, elinewidth=1.0, capsize=3, zorder=4,
                label="SPICE MC, 95% bootstrap CI")
    ax.plot(Tg, 100 * model_ber(Tg, fit["s25"], fit["a"], fit["mu"]), color=C2, lw=1.5,
            label="ML fit, " + par(fit["s25"], fit["a"]))
    if fit_ber is not None:
        ax.plot(Tg, 100 * model_ber(Tg, fit_ber["s25"], fit_ber["a"], fit_ber["mu"]), color=C1,
                lw=1.5, ls="--", label="BER-matched fit, " + par(fit_ber["s25"], fit_ber["a"]))
    ax.plot(Tg, 100 * model_ber(Tg, *current), color=CGRAY, lw=1.5, ls="-.",
            label="Prior model, " + par(current[0], current[1]))
    ax.set_xticks(sorted(set(int(t) for t in temps)))
    ax.set_xlim(min(temps) - 10, max(temps) + 10)
    if spec is not None:                 # outside the specified range: shown, but not what the code is sized for
        lo_x, hi_x = ax.get_xlim()
        for a0, a1 in ((lo_x, spec[0]), (spec[1], hi_x)):
            if a1 > a0:
                ax.axvspan(a0, a1, color=CGRAY, alpha=0.12, lw=0, zorder=0,
                           label=f"outside the specified {spec[0]}–{spec[1]} °C range" if a0 == lo_x else None)
    ax.set_xlabel("Temperature (°C)")
    ax.set_ylabel("Bit-error rate (%)")
    ax.set_ylim(bottom=0)
    h, l = ax.get_legend_handles_labels()
    # the SPICE entry first in the legend, then the curves in drawing order
    order = [l.index(x) for x in l if x.startswith("SPICE")] + [i for i, x in enumerate(l)
                                                                 if not x.startswith("SPICE")]
    fig.legend([h[i] for i in order], [l[i] for i in order], loc="outside right center", ncol=1)
    _save(fig, path)
    return fig
````

**Limitations: ageing (Subsection G) and data remanence (not simulated).** *Ageing:* negative-bias temperature instability (NBTI) shifts the threshold of a pMOS while it is on; in a cell that holds its power-up value, the stressed pMOS pulls up the preferred node, weakening the cell's own mismatch, so the power-up error rate grows over the lifetime ([30], who also propose writing the inverted power-up value into the array to reverse the stress). The PDK has no ageing model: Subsection G imposes the shift with an assumed BTI law in a recorded run, and Section VIII-F carries it to the key; the key-failure probabilities of Sections VII-E and XI are those of a fresh chip. *Data remanence:* an SRAM cell keeps its content for a while after the supply is removed [13], whereas every simulated power-up starts with all nodes at 0 V. **This is an assumption about the power-down:** before each read, at power-up and between the 64 reads of the chip nonce (Section VIII-F), the array's supply must stay off long enough and fall low enough for every cell to lose its state (requirements that grow at low temperature [22] and are not simulated); otherwise consecutive nonce reads repeat each other. A product must enforce a minimum off-time or actively discharge the array's supply before every read.

**Key insights.**

* The temperature coefficient is about 14 times the 0.0016 per °C of the kT dependence alone, and the BER rises below 25 °C, so both conditions of the prediction hold: the change of the mismatch weights, not kT, appears to govern the temperature behaviour. The noise-to-mismatch ratio is **2.9 times smaller** than the illustrative $s_{25} = 0.06$.
* The law $s(T)$ is symmetric about 25 °C, whereas the simulated cell flips less often at −40 °C than at 85 °C, so a noise-only law with one coefficient **cannot follow both −40 and 85 °C** and underestimates the end of the range that sizes the code. The excess at 85 °C is mainly *reproducible flips*, cells whose mismatch changes sign with temperature and which then power up to the opposite value every time (counted in Section VII-F). For the extracted cell the underestimate lies outside the 95 % interval of the simulation (2.20–4.11 %), so Section VII-E sizes the code from the simulated error rates, not from the law.
* For the 600-cell fingerprint, s ≈ 0.02 gives about 5 flipped bits per power-up at 25 °C and about 14 at 85 °C, which a code with t of a few tens corrects. The standard error of $s_{25}$ is 22 % of its value and the simulated noise is thermal only, so Section VII-E also checks the code against $s_{25} = 0.06$.

### D. Post-layout extraction

The values of Sections VII-B and C come from schematics. The layout adds wiring capacitance to every node, coupling between nodes, and, wherever the two halves of a differential circuit differ, **an asymmetry that acts in the same direction on every chip**, with the following expected effects:

* *Comparator noise and offset* are set during the integration phase, through the capacitance $C_\mathrm{P}$ of $d_\mathrm{p}$ and $d_\mathrm{n}$ ($\sigma_\mathrm{n}^2 \propto kT/C_\mathrm{P}$, and $T_\mathrm{int} \propto C_\mathrm{P}$). If the extracted $C_\mathrm{P}$ matches the 1 fF of wiring the schematic assumed, neither should change beyond its statistical error.
* *Systematic offset.* A capacitance difference $\Delta C$ between the two sides of the latch makes one side regenerate faster and shifts the switching point of every chip by the same amount. In a mirror-symmetric layout the only asymmetry left is where the cross-coupling wires cross the axis, so **a systematic offset far below** $\sigma_\mathrm{os}$ is expected.
* *Energy.* Every added wiring capacitance $C_\mathrm{w}$ that switches each cycle costs $C_\mathrm{w} V_\mathrm{DD}^2$ = 3.2 fJ per fF at 1.8 V; the schematic had no clock wiring, so the clock energy should rise.
* *SRAM-PUF bias.* If q and q̄ carry different capacitances, the lighter node rises faster and every cell leans towards the same value: P(1) ≠ 0.5, a lower min-entropy per cell and more similar fingerprints. A point-symmetric cell, whose two halves are the same shapes rotated by 180°, should give equal capacitances and P(1) = 0.5 within the sampling error.
* *SRAM-PUF noise and mismatch.* The extra node capacitance changes the noise and the race slightly; $s_{25}$ should stay within its statistical interval, while the few cells whose mismatch is smaller than the change may take a different noiseless value.

Each cell passes through five steps:

*1) Drawing:* The cells are drawn as code with the KLayout Python API, rectangle by rectangle on the SKY130 layers (diffusion, poly, contacts, li1, met1, implants, n-well), with the design rule of each coordinate noted in the source and the schematics' device sizes. **The SRAM cell is point-symmetric**: half A (inverter A, pass gate A, node q) is rotated by 180° to form half B, and each inverter's gate poly runs out to a contact under the opposite storage node, so the cross-couple needs no crossing wires (logic design rules, not the foundry's denser SRAM rules). The comparator core (M1–M11) is mirror-symmetric about its vertical axis, with the tail M7 on the axis and the clock routed up both sides; only the cross-couple cannot be mirrored, so its two li1 routes cross the axis at different heights.

*2) Design-rule check (DRC):* Magic checks the layout against the PDK's full rule deck.

*3) Extraction:* Magic extracts the devices with their drawn junctions and every parasitic capacitance (area, fringe and coupling, no lower threshold). Wire resistance is not extracted: the longest internal route, the comparator's li1 cross-couple, is about 10 squares (about 130 Ω at 12.8 Ω/□, plus contacts of the order of 100 Ω each), and with the about 1.3 fF it drives its RC is below 1 ps, **far below τ = 36 ps**.

*4) Layout versus schematic (LVS):* netgen compares the extracted netlist with the schematic, with the PDK's setup file.

*5) Mapping onto the schematic:* The extracted devices and capacitors are renamed onto the schematic's transistors and nets (labelled nets from the pins, unlabelled ones such as the comparator's tail from the terminals of transistors whose gate net is known), so the PDK draws the same mismatch and the testbenches inject the same noise: **every comparison is paired over an identical population** (individual noisy decisions can still differ, because the noise acts on a changed circuit). The comparator's buffers and SR latch remain at schematic level.

The post-layout simulations use the settings and seeds of Sections VII-B and C. The SRAM Monte Carlo and the comparator's noise, offset (the same 96 instances) and energy are shown from recorded runs; `PEX_LIVE = True` re-runs the comparator live (about 50 s) and the SRAM check as described in Section VII-C. Drawing, DRC, extraction, LVS and the waveform comparison of Fig. 35 **always run live**.

> **Next cell.** 1. Writes the layout flow shared by both circuits to `sky130_layout.py`; no tool is run. The flow covers GDS out of KLayout, DRC and extraction with Magic, LVS with netgen, and the mapping of an extraction back onto the schematic's transistors and nets, which finds unlabelled internal nets (such as the comparator's tail) from the transistors' terminals.

````python
%%writefile sky130_layout.py
"""SKY130 layout flow shared by the SRAM-PUF cell (puf_layout.py) and the StrongARM comparator (comparator_layout.py).

  write_gds      boxes and pin labels -> GDS, with the KLayout Python API (pip package `klayout`, headless in Colab)
  render         the same boxes plotted with matplotlib, coloured like KLayout's SKY130 layer set
  drc            Magic with the PDK's full SKY130 rule deck
  extract        Magic extraction: devices with drawn source/drain areas and perimeters, and every parasitic
                 capacitance (area, fringe and coupling; cthresh 0).  Wire resistance is not extracted: the li1 and
                 met1 runs here are a few squares, ohms against transistors of kilo-ohms and more.
  lvs            netgen with the PDK's setup file
  map_extraction extracted devices and capacitors renamed onto a schematic's transistors and nets, found from the
                 labelled nets and the transistors' terminals (internal nets such as a tail node are inferred)

Shapes are (layer, x0, y0, x1, y1) in nm; pins are (pin layer, text, (x0, y0, x1, y1)).  Text on a pin layer
(li1 67/16, met1 68/16) becomes a subcircuit port in Magic.
"""
import os
import re
import subprocess

# GDS (layer, datatype) of the SKY130 layers used here
GDS = dict(nwell=(64, 20), diff=(65, 20), tap=(65, 44), poly=(66, 20), licon=(66, 44), npc=(95, 20),
           li=(67, 20), li_pin=(67, 16), mcon=(67, 44), met1=(68, 20), met1_pin=(68, 16),
           nsdm=(93, 44), psdm=(94, 20))

# Display colours for `render` (close to the KLayout SKY130 layer properties), drawn bottom to top
STYLE = [("nwell", "#7f7f7f", 0.12), ("psdm", "#d9a400", 0.10), ("nsdm", "#2a9d8f", 0.10), ("npc", "#9b5de5", 0.10),
         ("diff", "#2e8b57", 0.55), ("tap", "#1f6f43", 0.55), ("poly", "#d62828", 0.55), ("licon", "#111111", 0.9),
         ("li", "#457b9d", 0.40), ("mcon", "#6a4c93", 0.9), ("met1", "#1d3557", 0.25)]


def write_gds(shapes, pins, path, name):
    """Write boxes and pins to GDS with KLayout (database unit 1 nm).  Returns dict(path, name, bbox_um)."""
    import klayout.db as kdb
    # one top cell; each box goes on its layer, each pin is a box plus a text label at its centre (Magic turns the label
    # into a port); coordinates are integers in database units of 1 nm
    ly = kdb.Layout()
    ly.dbu = 0.001
    top = ly.create_cell(name)
    idx = {k: ly.layer(*v) for k, v in GDS.items()}
    for layer, x0, y0, x1, y1 in shapes:
        top.shapes(idx[layer]).insert(kdb.Box(x0, y0, x1, y1))
    for layer, text, (x0, y0, x1, y1) in pins:
        top.shapes(idx[layer]).insert(kdb.Box(x0, y0, x1, y1))
        top.shapes(idx[layer]).insert(kdb.Text(text, kdb.Trans((x0 + x1) // 2, (y0 + y1) // 2)))
    os.makedirs(os.path.dirname(os.path.abspath(path)), exist_ok=True)
    ly.write(path)
    bb = top.bbox()
    return dict(path=path, name=name, bbox_um=(bb.left / 1000, bb.bottom / 1000, bb.right / 1000, bb.top / 1000))


def extent(shapes, layers=("diff", "tap", "poly", "li", "met1")):
    """Width, height and area (um) of the drawn shapes on the given layers (implant and well margins excluded)."""
    core = [s for s in shapes if s[0] in layers]
    w = (max(s[3] for s in core) - min(s[1] for s in core)) / 1000
    h = (max(s[4] for s in core) - min(s[2] for s in core)) / 1000
    return dict(width_um=w, height_um=h, area_um2=w * h)


def render(shapes, pins, ax=None, path=None, title=None, figsize=(3.3, 6.2), legend=True, pin_size=7.5):
    """Plot the layout with matplotlib, one patch per drawn box."""
    import matplotlib.pyplot as plt
    from matplotlib.patches import Rectangle, Patch
    own = ax is None
    if own:
        fig, ax = plt.subplots(figsize=figsize)
    # layers in STYLE order (wells and implants first, metal last), each box a translucent rectangle; pins as dark labels
    for layer, color, alpha in STYLE:
        for s in shapes:
            if s[0] == layer:
                x0, y0, x1, y1 = (v / 1000 for v in s[1:])
                ax.add_patch(Rectangle((x0, y0), x1 - x0, y1 - y0, fc=color, ec=color, lw=0.4, alpha=alpha))
    for _, text, (x0, y0, x1, y1) in pins:
        ax.text((x0 + x1) / 2000, (y0 + y1) / 2000, text, ha="center", va="center", fontsize=pin_size,
                color="white", fontweight="bold", bbox=dict(boxstyle="round,pad=0.15", fc="#131c28", ec="none", alpha=0.75))
    ax.set_aspect("equal")
    ax.autoscale_view()
    ax.set_xlabel("x (µm)")
    ax.set_ylabel("y (µm)")
    ax.grid(False)
    if title:
        ax.set_title(title)
    if legend:
        handles = [Patch(fc=c, ec=c, alpha=min(1, a + 0.25), label=n) for n, c, a in STYLE]
        ax.legend(handles=handles, loc="upper left", bbox_to_anchor=(1.02, 1.0), fontsize=7.5, frameon=False)
    if own and path:
        fig.savefig(path, dpi=200, bbox_inches="tight")
    return ax


# ----------------------------------------------------------------------------
# Magic and netgen
# ----------------------------------------------------------------------------

def tool_config(pdk_root, magic="magic", netgen="netgen", workdir="layout_work"):
    # tool paths and the PDK's setup files: Magic's sky130A.magicrc (technology, rule decks) and netgen's setup Tcl
    tech = os.path.join(pdk_root, "sky130A", "libs.tech")
    return dict(pdk_root=os.path.abspath(pdk_root), magic=magic, netgen=netgen, workdir=os.path.abspath(workdir),
                magicrc=os.path.join(tech, "magic", "sky130A.magicrc"),
                netgen_setup=os.path.join(tech, "netgen", "sky130A_setup.tcl"))


def _run_magic(tcfg, script, name):
    # run a Tcl script in batch Magic (no display, no console) with the PDK's rc file; the log is kept next to the script
    wd = tcfg["workdir"]
    os.makedirs(wd, exist_ok=True)
    tcl = os.path.join(wd, name + ".tcl")
    with open(tcl, "w") as f:
        f.write(script + "\nquit -noprompt\n")
    env = dict(os.environ, PDK_ROOT=tcfg["pdk_root"])
    p = subprocess.run([tcfg["magic"], "-dnull", "-noconsole", "-rcfile", tcfg["magicrc"], tcl],
                       cwd=wd, capture_output=True, text=True, timeout=900, env=env)
    log = p.stdout + p.stderr
    with open(os.path.join(wd, name + ".log"), "w") as f:
        f.write(log)
    if p.returncode != 0:
        raise RuntimeError(f"magic failed ({name}, rc={p.returncode}); see {os.path.join(wd, name + '.log')}")
    return log


def _load(gds, name):
    return f"gds read {os.path.abspath(gds)}\nload {name}\nselect top cell\n"


def drc(tcfg, gds, name):
    """Magic DRC with the full SKY130 deck.  Returns dict(count, errors=[(rule, n_boxes)])."""
    # full rule deck, check the whole cell, wait for completion, then print the total and the boxes per rule for parsing
    log = _run_magic(tcfg, _load(gds, name) + "drc style drc(full)\ndrc check\ndrc catchup\n"
                     "set n [drc list count total]\nputs \"DRCCOUNT $n\"\n"
                     "foreach {why boxes} [drc listall why] { puts \"DRCERR [llength $boxes] $why\" }\n", f"drc_{name}")
    m = re.search(r"DRCCOUNT (\d+)", log)
    if not m:
        raise RuntimeError(f"magic DRC gave no count; see drc_{name}.log")
    errors = [(l.split(" ", 2)[2], int(l.split(" ", 2)[1])) for l in log.splitlines() if l.startswith("DRCERR ")]
    return dict(count=int(m.group(1)), errors=errors)


def extract(tcfg, gds, name):
    """Magic extraction.  Writes <name>_lvs.spice (devices only, for LVS) and <name>_pex.spice (devices and all
    parasitic capacitances) in the work directory and returns their paths."""
    wd = tcfg["workdir"]
    # extract devices with capacitance and coupling; then two netlists: devices only for LVS, and devices with every
    # parasitic capacitor (cthresh 0: no capacitor dropped) for simulation
    base = _load(gds, name) + "extract do local\nextract do capacitance\nextract do coupling\nextract all\n"
    _run_magic(tcfg, base + f"ext2spice lvs\next2spice -o {name}_lvs.spice\n", f"extract_lvs_{name}")
    _run_magic(tcfg, base + "ext2spice format ngspice\next2spice cthresh 0\next2spice rthresh 0\n"
               f"ext2spice -o {name}_pex.spice\n", f"extract_pex_{name}")
    return dict(lvs=os.path.join(wd, f"{name}_lvs.spice"), pex=os.path.join(wd, f"{name}_pex.spice"))


def write_schematic(devices, ports, path, name):
    """A reference netlist for LVS: devices as (name, kind 'nfet'/'pfet', d, g, s, b, w, l)."""
    lines = [f".subckt {name} " + " ".join(ports)]
    lines += [f"X{n} {d} {g} {s} {b} sky130_fd_pr__{k}_01v8 w={w:g} l={l:g}" for n, k, d, g, s, b, w, l in devices]
    lines += [f".ends {name}", ""]
    with open(path, "w") as fh:
        fh.write("\n".join(lines))
    return path


def lvs(tcfg, layout_spice, schematic_spice, name):
    """netgen LVS with the PDK setup.  Returns dict(match, report)."""
    wd = tcfg["workdir"]
    rpt = os.path.join(wd, f"lvs_{name}.rpt")
    cmd = [tcfg["netgen"], "-batch", "lvs", f"{layout_spice} {name}", f"{schematic_spice} {name}", tcfg["netgen_setup"], rpt]
    p = subprocess.run(cmd, cwd=wd, capture_output=True, text=True, timeout=900)
    with open(os.path.join(wd, f"lvs_{name}.log"), "w") as f:
        f.write(p.stdout + p.stderr)
    text = open(rpt).read() if os.path.exists(rpt) else p.stdout
    return dict(match="Circuits match uniquely" in text and "Netlists do not match" not in text, report=rpt)


# ----------------------------------------------------------------------------
# From the extraction to a simulation netlist
# ----------------------------------------------------------------------------

def _um(v):
    # a SPICE length (with u/n/p suffix, or in metres) in µm
    v = v.lower()
    for suf, m in (("u", 1.0), ("n", 1e-3), ("p", 1e-6)):
        if v.endswith(suf):
            return float(v[:-1]) * m
    x = float(v)
    return x * 1e6 if x < 1e-3 else x


def parse_pex(path):
    """Devices and capacitors of a Magic ngspice-format extraction: (ports, devices, caps), devices as
    dict(name, d, g, s, b, kind, w, l, params) and caps as (node1, node2, farads)."""
    # join continuation lines, then read the subcircuit ports, the sky130 transistors (X lines: d g s b model w= l= ...)
    # and the capacitors (C lines: node node value with a SPICE suffix)
    text = open(path).read().replace("\n+", " ")
    ports, devices, caps = None, [], []
    for line in text.splitlines():
        t = line.split()
        if not t or t[0].startswith("*"):
            continue
        low = t[0].lower()
        if low == ".subckt":
            ports = t[2:]
        elif low == ".option" and any(x.lower().startswith("scale=") for x in t):
            raise RuntimeError(f"unexpected {line!r} in {path}; the code assumes ext2spice scale off")
        elif low.startswith("x") and len(t) >= 6 and "sky130_fd_pr__" in t[5]:
            params = {k.lower(): v for k, v in (x.split("=", 1) for x in t[6:] if "=" in x)}
            kind = "pfet" if "pfet" in t[5] else "nfet"
            devices.append(dict(name=t[0], d=t[1], g=t[2], s=t[3], b=t[4], kind=kind, w=_um(params["w"]),
                                l=_um(params["l"]), params=params))
        elif low.startswith("c") and len(t) >= 4:
            v = t[3].lower()
            mult = {"f": 1e-15, "a": 1e-18, "p": 1e-12, "n": 1e-9}
            caps.append((t[1], t[2], float(v[:-1]) * mult[v[-1]] if v[-1] in mult else float(v)))
    return ports, devices, caps


def map_extraction(pex_path, schematic, labels):
    """Rename an extraction onto a schematic.

    schematic : [(name, kind, d, g, s, b, w, l)] in the order the simulation netlist uses
    labels    : {extracted net name: schematic net name} for the labelled nets (several may map to one net)
    Unlabelled internal nets are inferred from the terminals of transistors whose gate is already known.
    Returns dict(devices=[dict(name, kind, d, g, s, b, w, l, ad, as_, pd, ps)] in schematic order,
                 caps=[(net1, net2, farads)], node_cap={net: farads}, netmap)."""
    _, devs, caps = parse_pex(pex_path)
    netmap = dict(labels)
    taken = {}                                   # schematic net -> extracted net, for inferred nets only
    out = {}
    remaining = list(devs)
    # Match extracted devices to schematic devices in rounds: a candidate must have the same type and size and a gate on
    # the expected (already known) net; its drain and source may be swapped. A schematic device is matched only when one
    # candidate is best (most terminals on known nets); its still unknown nets then become known, which lets the next
    # round match transistors further inside the circuit (for example those on the tail node). Stop when a round makes no
    # progress.
    while len(out) < len(schematic):
        progress = False
        for name, kind, d, g, s, b, w, l in schematic:
            if name in out:
                continue
            hits = []
            for ed in remaining:
                if ed["kind"] != kind or abs(ed["w"] - w) > 1e-6 or abs(ed["l"] - l) > 1e-6 or netmap.get(ed["g"]) != g:
                    continue
                for flip in (False, True):
                    ed_d, ed_s = (ed["s"], ed["d"]) if flip else (ed["d"], ed["s"])
                    pairs = [(ed_d, d), (ed_s, s), (ed["b"], b)]
                    if all(netmap.get(e, sn) == sn and taken.get(sn, e) == e for e, sn in pairs):
                        hits.append((sum(e in netmap for e, _ in pairs), flip, ed))
            if not hits:
                continue
            best = max(h[0] for h in hits)
            top = [h for h in hits if h[0] == best]
            if len(top) != 1:
                continue
            _, flip, ed = top[0]
            ed_d, ed_s = (ed["s"], ed["d"]) if flip else (ed["d"], ed["s"])
            for e, sn in ((ed_d, d), (ed_s, s), (ed["b"], b)):
                if e not in netmap:
                    netmap[e], taken[sn] = sn, e
            # the drawn junction areas and perimeters, swapped with drain and source when the device was matched flipped
            p = ed["params"]
            ad, as_ = (p.get("as"), p.get("ad")) if flip else (p.get("ad"), p.get("as"))
            pd_, ps = (p.get("ps"), p.get("pd")) if flip else (p.get("pd"), p.get("ps"))
            out[name] = dict(name=name, kind=kind, d=d, g=g, s=s, b=b, w=w, l=l, ad=ad, as_=as_, pd=pd_, ps=ps)
            remaining.remove(ed)
            progress = True
        if not progress:
            missing = [n for n, *_ in schematic if n not in out]
            raise RuntimeError(f"cannot match extracted devices to {missing}; unmatched: {[d['name'] for d in remaining]}")
    if remaining:
        raise RuntimeError(f"extracted devices left over: {[d['name'] for d in remaining]}")
    unknown = {n for a, b_, _ in caps for n in (a, b_) if n not in netmap}
    if unknown:
        raise RuntimeError(f"capacitors on nets that are not in the schematic: {sorted(unknown)}")
    # every capacitor on schematic nets, and each node's total capacitance (coupling capacitors counted on both nodes)
    caps = [(netmap[a], netmap[b_], v) for a, b_, v in caps]
    node_cap = {}
    for a, b_, v in caps:
        if a != b_:
            for n in (a, b_):
                node_cap[n] = node_cap.get(n, 0.0) + v
    return dict(devices=[out[n] for n, *_ in schematic], caps=caps, node_cap=node_cap, netmap=netmap)
````

> **Next cell.** 1. Writes `puf_layout.py`, which draws the point-symmetric 6T cell, and 2. `comparator_layout.py`, which draws the mirror-symmetric StrongARM core, rectangle by rectangle, with the design rule each coordinate respects in the comments.

````python
%%writefile puf_layout.py
"""SKY130 6T SRAM-PUF cell: layout, DRC, extraction and LVS, for the post-layout Monte Carlo of puf_spice.py.

The cell is drawn with the KLayout Python API (pip package `klayout`, headless in Colab); the devices are those of
puf_spice.CELL (pull-down W = 0.64, pull-up and pass gate W = 0.42, all L = 0.15 um).  It is point-symmetric, like a
production bit cell: one half (inverter A, pass gate A, node q) is drawn and the other half is the same shapes rotated
by 180 degrees, so q and qb see identical wiring and the layout adds no systematic skew to the PUF.  The cross-couple
needs no crossing wires: each inverter's gate poly runs out as a branch to a contact under the opposite storage
node's li1 strap.  It is drawn with the logic (periphery) design rules, not the foundry's SRAM rules, so it is larger
than the foundry bit cell.  DRC, extraction and LVS (Magic, netgen) are in sky130_layout.py; `postlayout_cell` turns
the extraction into the `sram6t` subcircuit of puf_spice.cell_subckt, with the schematic's transistor names and order
so that the PDK's per-instance mismatch draws are unchanged.
"""
import sky130_layout as sl

NAME = "sram6t"
# Conventions: coordinates are integers in nm, shapes are (layer, x0, y0, x1, y1); half A lies mostly at x < 0 and
# half B is half A rotated by 180 degrees about the origin (the cell centre, where vdd and the n-well tap sit).
# the schematic of puf_spice.cell_subckt, in its device order: (name, kind, d, g, s, b, w, l)
def schematic(cell):
    """cell = puf_spice.CELL (widths and length in um).  Returns the six devices as (name, kind, d, g, s, b, w, l) in
    the order of puf_spice.cell_subckt; map_extraction renames the extracted devices onto this list."""
    c = cell
    return [("PUa", "pfet", "q", "qb", "vdd", "vdd", c["w_pu"], c["l"]), ("PDa", "nfet", "q", "qb", "0", "0", c["w_pd"], c["l"]),
            ("PUb", "pfet", "qb", "q", "vdd", "vdd", c["w_pu"], c["l"]), ("PDb", "nfet", "qb", "q", "0", "0", c["w_pd"], c["l"]),
            ("PGa", "nfet", "bl", "wl", "q", "0", c["w_pg"], c["l"]), ("PGb", "nfet", "blb", "wl", "qb", "0", c["w_pg"], c["l"])]

# labelled nets of the layout -> nets of the simulation subcircuit (the two drawn word-line pads both map onto wl,
# because the simulation drives a single word line)
LABELS = {"q": "q", "qb": "qb", "vdd": "vdd", "vss": "0", "bl": "bl", "blb": "blb", "wla": "wl", "wlb": "wl"}


def _half(cell):
    """Shapes of half A (inverter A drives q, pass gate A connects q to bl), in nm.  Returns a list of
    (layer, x0, y0, x1, y1).  Coordinates follow the design-rule budget in the comments; Magic's DRC is the check."""
    W = lambda k: int(round(cell[k] * 1000))
    wpd, wpu, wpg, L = W("w_pd"), W("w_pu"), W("w_pg"), W("l")
    assert L == 150 and wpd >= wpg, "geometry is laid out for L = 0.15 um and W_PD >= W_PG"
    yN, yP = -2170, 475                    # bottom of the nmos row (A), bottom of the pmos row (A)
    xa = -335                              # gate A (poly) left edge; its right edge is xa + L
    # x layout of the nmos row, left to right: bit-line contact, PGa gate (xpg), node q, PDa gate (xa), vss contact.
    # Gate A is one vertical poly line through both rows, so it is the common gate of PDa and PUa (inverter A).
    S = []
    add = lambda layer, *b: S.append((layer,) + tuple(int(v) for v in b))
    # nmos row: bl | PGa | q | PDa | vss; the PD part is wider and is reached 250 nm before the PDa gate
    xj = xa - 255                          # diffusion jog (>= 250 from PDa, >= 75 from the PGa poly)
    xpg = xj - 75 - L                      # PGa gate left edge
    add("diff", xpg - 290, yN, xj, yN + wpg)
    add("diff", xj, yN, xa + L + 290, yN + wpd)
    add("nsdm", xpg - 290 - 125, yN - 125, xa + L + 290 + 125, yN + wpd + 125)
    add("poly", xpg, yN - 450, xpg + L, yN + wpg + 130)                     # PGa gate, down to its word-line pad
    add("poly", xa, yN - 130, xa + L, yP + wpu + 130)                       # gate A: PDa and PUa
    # pmos row: q | PUa | vdd (drawn above the cell centre; after rotation half B's pmos row sits below it)
    add("diff", xa - 300, yP, xa + L + 290, yP + wpu)
    add("psdm", xa - 300 - 125, yP - 125, xa + L + 290 + 125, yP + wpu + 125)
    # contacts on diffusion (licon 0.17, >= 0.055 from the gate, >= 0.06 inside the diffusion)
    add("licon", xpg - 230, yN + 125, xpg - 60, yN + 295)                  # bl
    add("licon", -570, yN + 125, -400, yN + 295)                            # q (nmos)
    add("licon", -125, yN + 235, 45, yN + 405)                              # vss
    add("licon", -570, yP + 125, -400, yP + 295)                            # q (pmos)
    add("licon", -125, yP + 125, 45, yP + 295)                              # vdd
    # q: one li1 strap joins the nmos and pmos drains and reaches the contact on gate B's branch
    ylic = 1290                            # y of the q contact on gate B's poly branch (above the pmos row)
    add("li", -570, yN + 45, -400, ylic + 250)
    add("poly", -650, ylic - 50, xa + L, ylic + 220)                        # pad and branch of gate B ...
    add("poly", -650, ylic + 10, -xa, ylic + 160)                           # ... running to gate B at x = -xa - L .. -xa
    add("licon", -570, ylic, -400, ylic + 170)
    add("npc", -670, ylic - 100, -300, ylic + 270)
    # word line: pad on the PGa poly below the nmos row
    yw = yN - 400                          # li1 pad >= 0.17 below the bit-line pad
    add("poly", xpg - 185, yw - 50, xpg + L, yw + 220)
    add("licon", xpg - 105, yw, xpg + 65, yw + 170)
    add("npc", xpg - 205, yw - 100, xpg + L + 20, yw + 270)
    add("li", xpg - 105, yw - 85, xpg + 65, yw + 255)
    # bit line pad
    add("li", xpg - 230, yN + 40, xpg - 60, yN + 380)
    # vss: li1 strap from the source down to a substrate tap
    add("li", -125, yN - 580, 45, yN + 485)
    add("tap", -245, yN - 560, 165, yN - 270)
    add("licon", -125, yN - 500, 45, yN - 330)
    add("psdm", -370, yN - 685, 290, yN - 125)
    # vdd: li1 pad, mcon and the met1 strap through the cell centre (shared with half B)
    add("li", -125, yP + 40, 45, yP + 380)
    add("mcon", -125, yP + 125, 45, yP + 295)
    add("met1", -185, -(yP + 355), 185, yP + 355)                           # symmetric in y: drawn twice, identical
    # n-well and its tap at the centre (both point-symmetric), tied to vdd
    add("nwell", xa - 300 - 180, -(yP + wpu + 180), -(xa - 300 - 180), yP + wpu + 180)
    add("tap", -125, -205, 125, 205)
    add("licon", -85, -85, 85, 85)
    add("nsdm", -250, -(yP - 125), 250, yP - 125)
    add("li", -85, -170, 85, 170)
    add("mcon", -85, -85, 85, 85)
    # pin boxes (labels for extraction / LVS); geometry() adds their rotated copies for half B
    pins = [("li_pin", "q", (-570, -85, -400, 85)),
            ("li_pin", "bl", (xpg - 230, yN + 125, xpg - 60, yN + 295)),
            ("li_pin", "wla", (xpg - 105, yw, xpg + 65, yw + 170)),
            ("li_pin", "vss", (-125, yN - 500, 45, yN - 330))]
    return S, pins


def geometry(cell):
    """All shapes of the full cell and its pins: half A plus half A rotated by 180 degrees about the origin."""
    half, pins = _half(cell)
    # rotation by 180 degrees: (x0, y0, x1, y1) -> (-x1, -y1, -x0, -y0), keeping x0 < x1 and y0 < y1.  The shapes on
    # the centre (vdd strap, n-well, n-well tap) are already point-symmetric and are simply drawn twice.
    rot = lambda b: (-b[2], -b[3], -b[0], -b[1])
    shapes = half + [(s[0],) + rot(s[1:]) for s in half]
    other = {"q": "qb", "bl": "blb", "wla": "wlb", "vss": "vss"}          # net of each pin's rotated copy
    pins = pins + [(layer, other[name], rot(b)) for layer, name, b in pins]
    pins.append(("met1_pin", "vdd", (-185, -85, 185, 85)))
    return shapes, pins


def draw(path, cell):
    """Write the cell layout to the GDS file `path` (cell NAME)."""
    return sl.write_gds(*geometry(cell), path, NAME)


def cell_report(cell):
    """Bounding box of the drawn shapes (sky130_layout.extent)."""
    return sl.extent(geometry(cell)[0])


def render(cell, **kw):
    """Matplotlib drawing of the cell for the notebook figure; keyword arguments go to sky130_layout.render."""
    return sl.render(*geometry(cell), **kw)


def drc(tcfg, gds):
    """Magic DRC of the GDS (tool configuration tcfg); see sky130_layout.drc."""
    return sl.drc(tcfg, gds, NAME)


def extract(tcfg, gds):
    """Magic extraction of devices and parasitic capacitances to a SPICE file; see sky130_layout.extract."""
    return sl.extract(tcfg, gds, NAME)


def schematic_netlist(cell, path):
    """LVS reference with the layout's port names (two word lines and vss, as drawn)."""
    lvs_names = {"0": "vss"}               # simulator ground "0" -> the layout's vss label
    devs = []
    for n, k, d, g, s, b, w, l in schematic(cell):
        g = {"PGa": "wla", "PGb": "wlb"}.get(n, g)          # each pass gate has its own drawn word-line pad
        devs.append((n, k, lvs_names.get(d, d), g, lvs_names.get(s, s), lvs_names.get(b, b), w, l))
    return sl.write_schematic(devs, ["vdd", "vss", "bl", "blb", "wla", "wlb", "q", "qb"], path, NAME)


def lvs(tcfg, layout_spice, schematic_spice):
    """netgen LVS of the extracted cell against the schematic reference; see sky130_layout.lvs."""
    return sl.lvs(tcfg, layout_spice, schematic_spice, NAME)


def postlayout_cell(pex_path, cell):
    """The extraction renamed onto puf_spice's sram6t: dict(devices, caps, node_cap, netmap)."""
    return sl.map_extraction(pex_path, schematic(cell), LABELS)
````

````python
%%writefile comparator_layout.py
"""SKY130 StrongARM comparator core (M1-M11 of comparator_spice.CORE): layout, DRC, extraction and LVS.

Drawn with the KLayout Python API, mirror-symmetric about x = 0 in the textbook arrangement, from the bottom up:
substrate taps under the ground rail; the clocked tail M7 on the axis; the input pair M1/M2 (4/0.5 um); the
latch nMOS M3/M4 above them, sharing the dp/dn wires; the n-well with the latch pMOS M5/M6 and the reset devices
M8-M11, its tap and the vdd rail at the top.  All gates are horizontal (W along x).  The only part that cannot be
mirrored is the cross-couple: outp has to reach the left gates and outn the right gates, so the two li1 routes
cross the axis at different heights (over the gate polys, which they do not contact).  The extraction shows what
that one asymmetry costs as a systematic offset.  The output buffers and the SR latch are not drawn: they stay at
schematic level in the simulation (their gate loads are in their device models).
"""
import sky130_layout as sl

# Conventions: all coordinates are integers in nm (1 um = 1000); shapes are tuples (layer, x0, y0, x1, y1).
# Only the left half (x < 0) and the shapes on the axis are written out; geometry() mirrors the left half about x = 0.
# The layer names are those of sky130_layout (diff, poly, licon, li, mcon, met1, nsdm/psdm implants, npc, nwell, tap).
NAME = "strongarm"
# Labelled nets of the layout -> nets of comparator_spice.CORE (the layout's "vss" is the simulator's ground "0").
LABELS = {"inp": "inp", "inn": "inn", "clk": "clk", "vdd": "vdd", "vss": "0", "outp": "outp", "outn": "outn"}
PORTS = ["inp", "inn", "clk", "vdd", "vss", "outp", "outn"]      # port order of the LVS reference subcircuit


def schematic(core, sizes):
    """comparator_spice.CORE as (name, kind, d, g, s, b, w, l), in its order.

    core  : comparator_spice.CORE, rows (name, "n"/"p", drain, gate, source, bulk)
    sizes : {name: (W, L)} in um (comparator_spice.DEFAULT_SIZES)
    The order matters: map_extraction renames the extracted devices onto this list, so the post-layout netlist keeps
    the schematic's instance names and order (and hence the PDK's per-instance mismatch draws)."""
    return [(n, "nfet" if t == "n" else "pfet", d, g, s, b, *sizes[n]) for n, t, d, g, s, b in core]


def _device(S, x0, x1, y0, L, nsd=290):
    """A transistor with a horizontal gate: diffusion x0..x1 (W = x1 - x0) by nsd | L | nsd, contacts (licon, pitch
    0.4 um) in both source/drain rows and an li1 strip over each row.  Returns the y of the gate's bottom and top.

    S : shape list to append to;  x0, x1, y0, L, nsd : nm.  The diffusion runs vertically: source/drain row (nsd
    = 0.29 um, the same extension comparator_spice.mos_line assumes for AD/AS/PD/PS), the gate (L), the second row."""
    add = lambda layer, *b: S.append((layer,) + tuple(int(v) for v in b))
    yg0, yg1 = y0 + nsd, y0 + nsd + L
    add("diff", x0, y0, x1, yg1 + nsd)
    add("poly", x0 - 130, yg0, x1 + 130, yg1)              # poly end cap of 0.13 um beyond the diffusion on both sides
    # 0.17 um licon squares, 0.06 um inside the diffusion edge, on a 0.4 um pitch along W
    xs = list(range(x0 + 60, x1 - 60 - 170 + 1, 400))
    for row in (y0 + 60, yg1 + 60):                        # lower and upper source/drain row
        for x in xs:
            add("licon", x, row, x + 170, row + 170)
        add("li", x0 - 20, row - 20, x1 + 20, row + 190)
    return yg0, yg1


def geometry(sizes):
    """All shapes and pins of the comparator core: (shapes, pins), shapes as (layer, x0, y0, x1, y1) in nm and pins
    as (pin layer, net label, box).  The coordinates are hand-placed for DEFAULT_SIZES (asserted below); `left` holds
    the left half (M1, M3, M5, M8, M10 and their wiring), `centre` the shapes on or across the axis (taps, rails, M7,
    the cross-couple), and the right half (M2, M4, M6, M9, M11) is the mirror image of `left`."""
    W = lambda n: (int(round(sizes[n][0] * 1000)), int(round(sizes[n][1] * 1000)))
    expect = {"M1": (4000, 500), "M3": (1000, 150), "M5": (2000, 150), "M7": (2000, 150), "M8": (1000, 150), "M10": (1000, 150)}
    assert all(W(k) == v for k, v in expect.items()), "geometry is laid out for comparator_spice.DEFAULT_SIZES"
    left, centre = [], []                                  # left half (mirrored later) and shapes drawn once
    L = lambda layer, *b: left.append((layer,) + tuple(int(v) for v in b))
    C = lambda layer, *b: centre.append((layer,) + tuple(int(v) for v in b))

    # substrate taps under the ground rail (met1), and the clock rail
    C("tap", -4000, 60, 4000, 340)
    C("psdm", -4125, -65, 4125, 485)
    C("li", -4020, 95, 4020, 305)
    for x in range(-3880, 3880 - 170 + 1, 400):
        C("licon", x, 115, x + 170, 285)
        C("mcon", x, 115, x + 170, 285)
    C("met1", -5900, 0, 5900, 400)                         # vss
    C("met1", -5790, 830, 5790, 1120)                      # clk
    # tail M7 on the axis, its gate contacted at both ends onto the clock rail
    _device(centre, -1000, 1000, 610, 150)
    C("nsdm", -1125, 485, 1125, 1485)
    C("li", -85, 305, 85, 650)                             # source to the taps
    C("poly", -1460, 900, 1460, 1050)
    L("poly", -1460, 840, -1130, 1110)
    L("licon", -1380, 890, -1210, 1060)
    L("npc", -1480, 790, -1110, 1160)
    L("li", -1380, 810, -1210, 1140)
    L("mcon", -1380, 890, -1210, 1060)
    # input pair M1 (tail -> dp), gate pad on the left
    _device(left, -4420, -420, 1610, 500)
    L("li", -700, 1090, -530, 1860)                        # tail: M7 drain to M1 source
    L("poly", -4880, 1930, -4550, 2200)
    L("licon", -4800, 1980, -4630, 2150)
    L("npc", -4900, 1880, -4530, 2250)
    L("li", -4800, 1900, -4630, 2230)
    # latch nMOS M3 (dp -> outn) on top of M1; its gate runs to the left vertical gate poly
    _device(left, -1420, -420, 2960, 150)
    L("poly", -1550, 3250, -180, 3400)
    L("li", -1440, 2440, -400, 3210)                       # dp: M1 drain and M3 source
    L("li", -5260, 2440, -400, 2650)                       # dp, out to the reset device M10
    L("nsdm", -4545, 1485, -295, 3815)
    L("poly", -330, 3250, -180, 5200)                      # gate of M3 and M5 (net outp)
    # n-well row: M10 (dp reset), M8 (outn reset), M5 (latch pMOS)
    for x0, x1 in ((-5320, -4320), (-4020, -3020), (-2420, -420)):
        _device(left, x0, x1, 4760, 150)
    L("poly", -2550, 5050, -180, 5200)                     # M5 gate to the vertical gate poly
    L("poly", -5480, 5050, -2890, 5200)                    # M10 + M8 gate (clk) ...
    L("poly", -5810, 4990, -5480, 5260)                    # ... its pad (>= 0.235 from the p-diffusion) ...
    L("licon", -5730, 5040, -5560, 5210)
    L("npc", -5830, 4940, -5460, 5310)
    L("li", -5730, 4960, -5560, 5290)
    L("mcon", -5730, 5040, -5560, 5210)
    L("met1", -5790, 830, -5500, 5270)                     # ... and the clock up the side
    L("li", -4040, 4800, -400, 5010)                       # outn: M8 and M5 drains
    L("li", -1340, 3440, -1170, 5010)                      # outn: M3 drain up to the pMOS drains
    L("li", -5260, 2440, -5090, 5010)                      # dp up to M10
    L("li", -5340, 5240, -400, 5450)                       # vdd: pMOS sources
    for x0, x1 in ((-5320, -4320), (-4020, -3020), (-2420, -420)):
        for x in range(x0 + 60, x1 - 60 - 170 + 1, 400):
            L("mcon", x, 5260, x + 170, 5430)
    C("psdm", -5445, 4635, 5445, 5635)
    C("tap", -5260, 5760, 5260, 6050)                      # n-well tap
    C("nsdm", -5385, 5635, 5385, 6175)
    C("li", -5280, 5800, 5280, 6010)
    for x in range(-5140, 5140 - 170 + 1, 400):
        C("licon", x, 5820, x + 170, 5990)
        C("mcon", x, 5820, x + 170, 5990)
    C("nwell", -5500, 4580, 5500, 6230)
    C("met1", -5320, 5200, 5320, 6050)                     # vdd
    # cross-couple: A takes outp to the left gate poly, B takes outn to the right gate poly
    C("poly", -390, 3800, -120, 4130)
    C("licon", -340, 3880, -170, 4050)
    C("li", -420, 3880, 1340, 4050)
    C("poly", 120, 4270, 390, 4600)
    C("licon", 170, 4350, 340, 4520)
    C("li", -1340, 4350, 420, 4520)
    C("npc", -440, 3780, 440, 4660)

    # mirror about x = 0: (x0, y0, x1, y1) -> (-x1, y0, -x0, y1), keeping x0 < x1
    mirror = lambda b: (-b[2], b[1], -b[0], b[3])
    shapes = centre + left + [(s[0],) + mirror(s[1:]) for s in left]
    # pins of the left half; their mirror images carry the complementary net names (inn, outp)
    pins = [("li_pin", "inp", (-4800, 1980, -4630, 2150)), ("li_pin", "outn", (-1340, 3700, -1170, 3870))]
    other = {"inp": "inn", "outn": "outp"}
    pins += [(layer, other[t], mirror(b)) for layer, t, b in pins]
    pins += [("met1_pin", "vss", (-200, 0, 200, 400)), ("met1_pin", "clk", (-200, 830, 200, 1120)),
             ("met1_pin", "vdd", (-200, 5500, 200, 5700))]
    return shapes, pins


def draw(path, sizes):
    """Write the layout to the GDS file `path` (cell NAME)."""
    return sl.write_gds(*geometry(sizes), path, NAME)


def cell_report(sizes):
    """Bounding box of the drawn shapes (sky130_layout.extent)."""
    return sl.extent(geometry(sizes)[0])


def render(sizes, **kw):
    """Matplotlib drawing of the layout for the notebook figure; keyword arguments go to sky130_layout.render."""
    kw.setdefault("figsize", (7.2, 4.4))
    return sl.render(*geometry(sizes), **kw)


def drc(tcfg, gds):
    """Magic DRC of the GDS (tool configuration tcfg); see sky130_layout.drc."""
    return sl.drc(tcfg, gds, NAME)


def extract(tcfg, gds):
    """Magic extraction of devices and parasitic capacitances to a SPICE file; see sky130_layout.extract."""
    return sl.extract(tcfg, gds, NAME)


def schematic_netlist(core, sizes, path):
    """LVS reference: the schematic core with ground renamed "0" -> "vss" to match the layout's port names."""
    devs = [(n, k, *("vss" if x == "0" else x for x in (d, g, s, b)), w, l) for n, k, d, g, s, b, w, l in schematic(core, sizes)]
    return sl.write_schematic(devs, PORTS, path, NAME)


def lvs(tcfg, layout_spice, schematic_spice):
    """netgen LVS of the extracted layout against the schematic reference; see sky130_layout.lvs."""
    return sl.lvs(tcfg, layout_spice, schematic_spice, NAME)


def postlayout(pex_path, core, sizes):
    """The extraction renamed onto comparator_spice's core: dict(devices, caps, node_cap, netmap)."""
    return sl.map_extraction(pex_path, schematic(core, sizes), LABELS)
````

````python
#@title SKY130 6T SRAM-PUF: layout, DRC, LVS and extraction
# The SRAM-PUF cell from drawing to post-layout netlist, with open-source tools only. sky130_layout holds the shared
# layout primitives and tool wrappers, puf_layout the cell itself (both written by the cells above).
import importlib, klayout, sky130_layout as sl, puf_layout as pl
importlib.reload(sl); importlib.reload(pl)
from matplotlib.patches import Patch
t0 = time.time()
LD = WORK / "layout" / "sram6t"
TL_PUF = sl.tool_config(str(PDK_ROOT), magic=str(MAGIC), netgen=str(NETGEN), workdir=str(LD))
gds_puf = pl.draw(str(LD / "sram6t.gds"), ps.CELL)                     # 1. KLayout: rectangles from Python -> GDS
drc_puf = pl.drc(TL_PUF, gds_puf["path"])                              # 2. Magic: full SKY130 rule deck
ext_puf = pl.extract(TL_PUF, gds_puf["path"])                          # 3. Magic: devices + parasitic capacitances
lvs_puf = pl.lvs(TL_PUF, ext_puf["lvs"], pl.schematic_netlist(ps.CELL, str(LD / "sram6t_sch.spice")))   # 4. netgen
LAY_PUF = pl.postlayout_cell(ext_puf["pex"], ps.CELL)                  # the extraction as puf_spice's sram6t subcircuit
# cell size and the extracted capacitance of each internal node; cap(a, b) sums the coupling capacitors between a and b
size, nc = pl.cell_report(ps.CELL), LAY_PUF["node_cap"]
cap = lambda a, b: sum(v for x, y, v in LAY_PUF["caps"] if {x, y} == {a, b})
card("SRAM-PUF cell in SKY130: layout, DRC, LVS and extraction",
     [(f"layout (KLayout {klayout.__version__}, Python API)", f"{size['width_um']:.2f} × {size['height_um']:.2f} µm, 6 transistors, point-symmetric"),
      ("DRC (Magic, full SKY130 rule deck)", f"<b>{drc_puf['count']} violations</b>"),
      ("LVS (netgen, against the schematic)", "<b>circuits match uniquely</b>: 6 devices, 8 nets" if lvs_puf["match"] else "<b>netlists do not match</b>"),
      ("extracted wiring capacitance on q / qb", f"<b>{1e15 * nc['q']:.3f} / {1e15 * nc['qb']:.3f} fF</b>, of which {1e15 * cap('q', 'qb'):.3f} fF couples q to qb"),
      ("run time", f"{time.time() - t0:.0f} s")], tone="ok" if drc_puf["count"] == 0 and lvs_puf["match"] else "bad")
# the tools' own words: netgen's match lines and the transistors Magic extracted (device, nodes, W/L)
print("netgen LVS report:", *[l.strip() for l in open(lvs_puf["report"]) if "match" in l.lower()][-2:], sep="\n  ")
print("Magic extraction, transistors:", *[l.rstrip() for l in open(ext_puf["pex"]) if l.startswith("X")], sep="\n  ", end="")

# the layout drawings label the pins with the names of the schematics (the GDS keeps the netlist's names)
NET = {"q": "$q$", "qb": r"$\bar{q}$", "bl": "BL", "blb": r"$\overline{\mathrm{BL}}$", "wla": "WL", "wlb": "WL", "vdd": r"$V_\mathrm{DD}$",
       "vss": "GND", "inp": r"$V_\mathrm{in+}$", "inn": r"$V_\mathrm{in-}$", "outp": r"$\mathrm{out}_p$", "outn": r"$\mathrm{out}_n$",
       "dp": "$d_p$", "dn": "$d_n$", "clk": r"$\phi$"}
def name_pins(ax):
    for t in ax.texts:
        t.set_text(NET.get(t.get_text(), t.get_text()))
    ax.set_xlabel("$x$ (µm)"); ax.set_ylabel("$y$ (µm)")
# (a) the drawn layout; (b) where the wiring capacitance of the storage nodes q and qb goes (to ground, the supply, the
# other node, the bit lines and the word line), from the extracted coupling capacitors
fig, (a0, a1) = plt.subplots(1, 2, figsize=(W2, 4.8), gridspec_kw=dict(width_ratios=[1.0, 1.3]))
pl.render(ps.CELL, ax=a0, legend=False, pin_size=7); name_pins(a0)
panel(a0, "a", "layout (2 halves, rotated 180°)")
groups = [("ground", lambda n: cap(n, "0")), (NET["vdd"], lambda n: cap(n, "vdd")), ("other node", lambda n: cap("q", "qb")),
          ("bit lines", lambda n: cap(n, "bl") + cap(n, "blb")), ("word line", lambda n: cap(n, "wl"))]
yb = np.arange(len(groups))
for j, (node, c, h) in enumerate((("q", COL["ok"], ""), ("qb", COL["bad"], "////"))):
    a1.barh(yb + (j - 0.5) * 0.36, [1e15 * f(node) for _, f in groups], height=0.34, color=c, hatch=h, ec=COL["ink"], lw=0.5, label=NET[node])
a1.set_yticks(yb, [g for g, _ in groups]); a1.invert_yaxis()
a1.set_xlabel("Extracted capacitance (fF)"); a1.legend(loc="lower right"); a1.grid(axis="y", visible=False)
panel(a1, "b", r"where the wiring capacitance of $q$ and $\bar{q}$ goes")
fig.legend(handles=[Patch(fc=c, ec=c, alpha=min(1, a + 0.25), label=n) for n, c, a in sl.STYLE], loc="lower center", ncol=6,
           fontsize=8, frameon=False, bbox_to_anchor=(0.5, -0.02))
fig.tight_layout(rect=(0, 0.07, 1, 1)); plt.show()
````

#### Numerical convergence of the transistor-level results

A transient simulation has converged when a smaller time step and a tighter error tolerance leave its result unchanged. Both cells turn analog trajectories into binary outcomes, a comparator decision and the value a cell powers up to, so the test is exact: with the same noise and mismatch seeds, only an outcome within the numerical error of the threshold could change, and **the analysis above predicts that no decision and no bit changes**. The settings of every run were chosen in development as the largest steps that agree with finer references (for the SRAM cell, a 0.1 ns step and tolerances down to `reltol` = $10^{-4}$; for the comparator, 5, 10 and 20 ps steps around the clock edge); this cell repeats the check live on the extracted netlists.

**The check runs live at transistor level on the extracted netlists** of this subsection, at tt and 1.8 V. *Comparator:* 300 decisions, 60 at each of five inputs over ±2 $\sigma_\mathrm{n}$, with the seeded channel noise of Section VII-B, at the standard settings (a 5 ns step limit between clock edges, 20 ps steps around each edge, ngspice's default tolerances) and with both steps halved and `reltol` = $10^{-4}$. *SRAM cell:* 64 cells, two noisy power-ups each at 25 °C with the cell's noise envelope, at the standard step limit of 1 ns and with 0.5 ns and `reltol` = $10^{-4}$.

> **Next cell.** 1. Simulates the 300 comparator decisions at both settings. 2. Simulates the 128 power-ups at both settings. 3. Compares them decision by decision and bit by bit, fits $\sigma_\mathrm{n}$ to both sets of decisions, and prints the result panel.

````python
#@title Numerical convergence: half the time steps, tighter tolerance
# Numerical convergence of the transistor-level results: the post-layout comparator's noise decisions and the post-layout
# SRAM cell's power-ups, simulated twice with the same noise and mismatch seeds, at the settings of every run and with half
# the time steps (comparator: the step limit between clock edges and the forced step around them; SRAM cell: the step
# limit) and a ten times tighter relative tolerance (reltol = 1e-4). Converged numerics give the same decisions and the
# same bits; the scale of the comparison is the noise itself (the probit's standard error; the cells that differ between
# two noisy power-ups). Inputs: cs, ps, cfg_post, nf_post, pcfg, POST, env_w (cells above). Outputs: CONV.
from concurrent.futures import ThreadPoolExecutor
t0 = time.time()
FINE_CMP = dict(noise_tmax=2.5e-9, fine_dt=cfg_post["fine_dt"] / 2, spice_options="reltol=1e-4")
FINE_PUF = dict(tmax_scale=0.5, options="reltol=1e-4")
# comparator: 60 decisions at each of 5 inputs over +-2 sigma_n, in jobs of 50 clocks with seeded channel-noise sources
traj_c = cs.reference_trajectory(cfg_post); S_c = cs.channel_noise_psd(cfg_post, traj_c)
dv_c = np.linspace(-2, 2, 5) * nf_post["sigma"]
seq_c = np.repeat(np.arange(len(dv_c)), 60); np.random.default_rng(2041).shuffle(seq_c)
def cmp_decisions(c, tag):
    """Every decision of the campaign, in campaign order, for the comparator config c."""
    jobs, meta = [], []
    for j, i0 in enumerate(range(0, len(seq_c), 50)):
        nm = f"conv_{tag}_{j}"
        net, te = cs.run_noise_job(c, dv_c[seq_c[i0:i0 + 50]], S_c, 2041 * 7919 + j, nm)
        jobs.append((net, nm)); meta.append((nm, te))
    cs.run_many(c, jobs)
    return np.concatenate([cs.read_decisions(c, nm, te)[1:] for nm, te in meta])       # cycle 0 of each job is a dummy
dec_d, dec_f = cmp_decisions(cfg_post, "default"), cmp_decisions(dict(cfg_post, **FINE_CMP), "fine")
fit = lambda d: cs.probit_fit(dv_c, np.bincount(seq_c, weights=d, minlength=len(dv_c)), np.bincount(seq_c, minlength=len(dv_c)))
fit_d, fit_f = fit(dec_d), fit(dec_f)
# SRAM cell: 64 extracted cells, two noisy power-ups at 25 °C with the cell's noise envelope, both settings in parallel
def puf_bits(pu, tag):
    return ps.run_chunk(pcfg, 64, [25], 2, 7, 900, env_w, f"conv_puf_{tag}", cell=POST, pu=pu, reference=False)["bits"][0]
with ThreadPoolExecutor(2) as ex:
    bits_d, bits_f = ex.map(lambda a: puf_bits(*a), ((ps.POWERUP, "default"), (dict(ps.POWERUP, **FINE_PUF), "fine")))
CONV = dict(cmp_diff=int((dec_d != dec_f).sum()), cmp_n=dec_d.size, puf_diff=int((bits_d != bits_f).sum()), puf_n=bits_d.size,
            puf_noise=int((bits_d[0] != bits_d[1]).sum()), fit_d=fit_d, fit_f=fit_f, t=time.time() - t0)
mv = lambda f: f"{1e3 * f['sigma']:.3f} ± {1e3 * f['sigma_se']:.3f} mV"
card("Numerical convergence: half the time steps, ten times tighter relative tolerance",
     [("comparator, post-layout: decisions that change", f"<b>{CONV['cmp_diff']} of {CONV['cmp_n']}</b> (60 at each of 5 inputs over "
                                                          "±2 σ<sub>n</sub>, the same noise)"),
      ("σ<sub>n</sub> of these decisions (probit)", f"{mv(fit_d)} at the run's settings; {mv(fit_f)} with the finer settings"),
      ("SRAM cell, post-layout: power-up bits that change", f"<b>{CONV['puf_diff']} of {CONV['puf_n']}</b> (64 cells × 2 power-ups, "
                                                            "25 °C, the same noise and mismatch)"),
      ("for scale: cells that the noise changes between the two power-ups", f"{CONV['puf_noise']} of 64"),
      ("run time", f"{CONV['t']:.0f} s")], tone="ok" if CONV["cmp_diff"] == 0 and CONV["puf_diff"] == 0 else "bad")
````

**Key insights.** As predicted, **no decision and no bit changes**: the 300 comparator decisions and the 128 power-up bits are identical at both settings, so the probit gives the same $\sigma_\mathrm{n}$ at both, 0.549 ± 0.051 mV (consistent with the 0.575 mV of the full measurement), while the noise alone changes 1 of the 64 cells between two power-ups of the same chip. The outcomes reported in this section are therefore set by the circuit's noise and mismatch, not by the simulator's settings. The check covers tt and 1.8 V, the comparator at 27 °C and the SRAM cell at 25 °C; the step limits of the other temperatures were set by the same comparison in development (0.5 ns at −40 °C, 2 ns from 85 °C).

> **Next cell.** 1. Opens both GDS files with KLayout's renderer, which draws without a window, and applies the PDK's layer table `sky130A.lyp`. 2. Renders each cell at a fixed height with 2× oversampling. 3. Draws both with a common legend of the layers used.

````python
#@title Fig.: the layouts in KLayout
# KLayout's own renderer (klayout.lay draws without a window) with the PDK's KLayout layer table, sky130A.lyp: the colours and
# stipples of the KLayout window, on a white background. The same views drive the interactive viewer of the next cell.
import io, klayout.lay as klay, klayout.db as kdb
from matplotlib.patches import Patch
LYP = PDK_ROOT / "sky130A" / "libs.tech" / "klayout" / "tech" / "sky130A.lyp"
def kl_view(gds):
    """The GDS on a white background, without KLayout's grid (its ruler is part of the grid: the figures draw their own scale).
    The fill colours are the PDK's; the outlines of its very light layers (li1, licon1, mcon) are darkened to stay visible."""
    v = klay.LayoutView()
    v.load_layout(str(gds), True); v.load_layer_props(str(LYP))
    v.set_config("background-color", "#ffffff"); v.set_config("grid-visible", "false"); v.max_hier(); v.zoom_fit()
    it = v.begin_layers()
    while not it.at_end():
        n = it.current(); rgb = [(n.fill_color >> s) & 255 for s in (16, 8, 0)]
        if 0.2126 * rgb[0] + 0.7152 * rgb[1] + 0.0722 * rgb[2] > 0.78 * 255:      # luminance of the fill: a light layer
            n.frame_color = 0xFF000000 | sum(int(0.55 * c) << s for c, s in zip(rgb, (16, 8, 0)))
        it.next()
    return v
def kl_layers(v):
    """The layers the layout uses: (name, fill colour, an iterator to the view's layer entry)."""
    out, it = [], v.begin_layers()
    while not it.at_end():
        n = it.current()
        if not n.bbox().empty():
            out.append((n.name.split(" - ")[0], f"#{n.fill_color & 0xFFFFFF:06x}", it.dup()))
        it.next()
    return out
def kl_size(v, w, h):
    v.resize(w, h); v.zoom_fit()                                               # the window size, then the whole cell
kl_png = lambda v: v.get_pixels_with_options(v.viewport_width(), v.viewport_height(), 0, 2).to_png_data()   # 2x oversampled
def kl_fit(v, h=700, foot=0.1):
    """The window (h pixels high) and the view: the whole cell with a small margin and a free strip below it, `foot` times the
    cell's height, for the scale bar. Returns the image's aspect ratio (width / height)."""
    b = v.active_cellview().cell.dbbox(); m = 0.03 * max(b.width(), b.height())
    box = kdb.DBox(b.left - m, b.bottom - m - foot * b.height(), b.right + m, b.top + m)
    v.resize(round(h * box.width() / box.height()), h); v.zoom_box(box)
    return box.width() / box.height()
def kl_scalebar(ax, v, img, um):
    """A scale bar of `um` µm at the lower left of a rendered view, its length written above it."""
    px = img.shape[1] / v.box().width()                                        # pixels per µm
    x0, y0 = 0.04 * img.shape[1], 0.965 * img.shape[0]
    ax.plot([x0, x0 + um * px], [y0, y0], color=COL["ink"], lw=2.5, solid_capstyle="butt")
    ax.text(x0 + um * px / 2, y0 - 0.012 * img.shape[0], f"{um:g} µm", ha="center", va="bottom", fontsize=8.5, color=COL["ink"])
# Render both cells: one view per layout, fitted with a strip for the scale bar, as two panels whose widths follow the
# cells' aspect ratios; below them, a legend of the layers that the two layouts use, in KLayout's colours
KL = {"6T SRAM-PUF cell": kl_view(gds_puf["path"]), "StrongARM comparator": kl_view(gds_cmp["path"])}

asp = [kl_fit(v) for v in KL.values()]
fig, axs = plt.subplots(1, 2, figsize=(W2, W2 / sum(asp) + 0.8), gridspec_kw=dict(width_ratios=asp))
for ax, v, lab, title in zip(axs, KL.values(), "ab", ("6T cell", "StrongARM comparator")):
    img = plt.imread(io.BytesIO(kl_png(v)), format="png")
    ax.imshow(img); ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
    kl_scalebar(ax, v, img, 2)
    panel(ax, lab, title)
fig.tight_layout()
for ax in axs:
    ax.apply_aspect()                                                          # the images' final size, for the legend below
used = {n: (c, f"#{it.current().frame_color & 0xFFFFFF:06x}") for v in KL.values() for n, c, it in kl_layers(v)}
fig.legend(handles=[Patch(fc=c, ec=e, lw=0.8, label=n.replace(".drawing", "")) for n, (c, e) in used.items()], loc="upper center",
           bbox_to_anchor=(0.5, min(ax.get_position().y0 for ax in axs) - 0.01), ncol=7, fontsize=7.5, frameon=False,
           handlelength=1.0, columnspacing=0.9)
plt.show()
````

### E. The model with SKY130 parameters

**Five of the model's illustrative settings are replaced**. `comp_noise` = $\sigma_\mathrm{n} / V_\mathrm{unit}$ (Section VII-B); `puf_sigma`, `puf_tc` and `puf_bias` are $s_{25}$, a and P(1) of the cell model (Section VII-C). The fifth, `int_noise`, the front-end noise lumped at the first integrator, is not simulated; as in Section V-A, the input-referred 9.48 pA rms measured in [1] in a 100 kHz band is taken through the transimpedance gain $R_\mathrm{TIA}$ and the buffer, normalised to full scale, spread as white noise over the 5 MHz Nyquist band of one clock (a factor √(5 MHz/100 kHz)) and multiplied by the first integrator's gain $g_1 = 0.5$, because the model adds it at that integrator's output.

**The BCH code is sized from the simulated PUF parameters**. The key is reconstructed if at most t of the n = 600 fingerprint cells flip at a power-up, t being the number of errors the BCH code corrects. At temperature T, cell $j$ of a chip disagrees with its enrolled value with probability $p_j(T)$; the number of flipped cells $N$ is Poisson-binomial, and the key-failure probability $P_\mathrm{fail}(t, T) = P(N > t)$ is computed exactly. The design rule is $P_\mathrm{fail} < 10^{-6}$ at every temperature from 0 to 85 °C.

The linear law $s(T) = s_{25}\,(1 + a\,\lvert T - 25\,{}^\circ\mathrm{C}\rvert)$ of Section VII-C underestimates the 85 °C error rate of the extracted cell (1.75 % against 3.14 % simulated). A noise-only model anchored to each simulated error rate removes that bias but misses that part of the 85 °C errors are reproducible flips, the same cells wrong at every hot power-up (Section VII-F): a chip with F fixed flips fails almost surely when F > t, so failures concentrate on the chips with many flips, which one chip with one enrolment read does not show. The code is therefore sized with a **two-component cell model over a chip population**. Cell $j$ has power-up noise of rms $s(T)$ and the mismatch $m_j(T) = \mu + m_{0,j} \cos\theta(T) + m_{\perp,j} \sin\theta(T)$, with $m_0$, $m_\perp$ independent standard normal and $\mu$ the bias, so that a rotation θ reverses the sign of a fixed fraction θ/π of the cells. $s(T)$ follows from the disagreement between repeated power-ups at $T$, $\theta(T)$ from the error rate against the 25 °C enrolment, with opposite signs below and above 25 °C (a monotonic change of the nMOS/pMOS mismatch weight); both are piecewise linear with a knot at 25 °C, so the worst case over 0–85 °C lies at 0, 25 or 85 °C. Each simulated chip draws its own cells and its own enrolment read; the result is the fraction of chips that meet 10⁻⁶ and the 99th percentile of the t they require.

The helper data is $10t$ public bits, so $n H_\infty - 10t$ bits remain secret, $H_\infty$ being the min-entropy per cell (Section V-D, with the bias $b = P(1)$). Because $P(1)$ is estimated from a finite number of cells, $H_{\infty}$ is bounded from below by the one-sided 95 % Clopper–Pearson upper bound on max(P(1), 1 − P(1)), that is, 0.5 by the symmetry of the cell plus a margin for the observed bias and its sampling error; the budget is reported from this bound to the point estimate. By the leftover-hash lemma, a 128-bit key within statistical distance ε of uniform requires $128 + 2\log_2(1/\epsilon)$ bits of min-entropy from a universal hash family: with ε = 2⁻³², **192 secret bits**, not 128. SHA-256 is not a universal family, so applying the margin to it is a heuristic (random-oracle) assumption, as in common PUF practice; a provable bound would need a seeded universal hash before SHA-256. Spatial correlation between cells, which would lower the entropy, is not modelled.

With P(1) = 0.493 from 1,024 cells, the analysis above predicts a lower bound on $H_{\infty}$ near 0.91 bit (point estimate 0.979), so that 600 cells hold about 545–588 bits and the 192-bit requirement allows t ≈ 35 at the lower bound. At 85 °C the simulated error rate of 3.14 % gives 18.8 flipped cells on average; if about half of them are fixed flips, the spread between chips is that of the flip count, and the 99th-percentile chip should need t ≈ 40, **more than the secrecy bound allows**.

The sizing runs the behavioural model with the post-layout values of Section VII-D (for the PUF, the recorded 1,024-cell run), schematic values alongside. The population is 1,000 Monte Carlo chips of 600 cells, each enrolled by one simulated noisy read at 25 °C (seeded from `chip_seed`); $s(T)$ comes from the mean disagreement between the 28 pairs of power-ups at each of −40, 25 and 85 °C of the recorded run, θ(T) from its error rates (1.37 %, 0.99 % and 3.14 %). The 85 °C error rate carries a 95 % bootstrap interval over cells (2.20–4.11 %), at whose ends the population is also evaluated, **so the required t is an interval**. For comparison, the earlier method (a noise-only model of this chip, one enrolment read, anchored to the same error rates) and the linear law are sized by the same 10⁻⁶ criterion, and the schematic values by the earlier method; a robustness check repeats the sizing with the law and $s_{25} = 0.06$, because the simulated noise is thermal only. The later sections keep t = 45, the largest t that leaves 128 secret bits at the point estimate, so that their results and the RTL of Section XII stay comparable.

````python
#@title Set the model's circuit parameters and size the BCH code
# Everything the rest of the notebook uses, now from SKY130 simulation of the extracted layouts or the paper's measurements
# 1. modulator thermal noise from the paper's 9.48 pA rms (100 kHz) at the TIA input, referred to the modulator
#    input (TIA gain, x0.5 buffer, full scale VDD/2) and spread over the 5 MHz Nyquist band of one clock; the model adds
#    int_noise to the first integrator's output, where an input-referred noise appears multiplied by g1 = 0.5
v_tia = 9.48e-12 * DEFAULTS["rtia"]
int_noise_in = v_tia * DEFAULTS["buffer"] / (VDD / 2) * np.sqrt((FS / 2) / 100e3)    # referred to the modulator input
int_noise = 0.5 * int_noise_in                                                       # at the first integrator's output
cal_from = lambda comp, fit: dict(comp_noise=float(comp["comp_noise"]), int_noise=float(int_noise), puf_sigma=float(fit["s25"]),
                                  puf_tc=float(fit["tc"]), puf_bias=float(fit["bias"]))
CAL_SCH, CAL = cal_from(COMP, PUF_FIT), cal_from(COMP_POST, PUF_FIT_POST)      # schematic, and post-layout (used)
before = {k: DEFAULTS[k] for k in CAL}

# 2. BCH strength, first by the earlier method (kept for comparison): one chip, a noise-only cell model whose noise-to-mismatch
#    ratio s(T) reproduces the simulated bit-error rate at each temperature (interpolated linearly in between; anchor_s and
#    anchored() also give the cell model of each process corner in Section VII-F). The linear law s25 (1 + a|T - 25 °C|)
#    is symmetric about 25 °C and underestimates the 85 °C rate; it is sized as a second comparison.
from scipy.optimize import brentq
_mref, _nref = rng_for(99).standard_normal(8192), rng_for(100).standard_normal(8192)
def anchor_s(mu, ber, s_enrol=None):
    """s at which the model's mean error rate (cells N(mu, 1) enrolled by one read with noise s_enrol) equals ber."""
    m = _mref + mu
    err = lambda s0, s: np.where(m + s0 * _nref > 0, ndtr(-m / s), ndtr(m / s)).mean()
    f = (lambda s: err(s, s) - ber) if s_enrol is None else (lambda s: err(s_enrol, s) - ber)
    lo, hi = 1e-4, (4.0 if s_enrol is None else 20.0)
    if f(lo) >= 0:                    # below the floor that the noisy enrolment read sets: the closest the model can get
        return lo
    return brentq(f, lo, hi) if f(hi) > 0 else hi
def anchored(mu, ber_T):
    """s(T), T in deg C, from the measured error rates {T: ber} against a 25 °C enrolment."""
    s25 = anchor_s(mu, ber_T[25]); sT = {T: (s25 if T == 25 else anchor_s(mu, b, s25)) for T, b in ber_T.items()}
    return lambda T: float(np.interp(T, sorted(sT), [sT[x] for x in sorted(sT)]))
def size_bch(mu, ber, bias):
    """(t that meets 1e-6 over 0-85 °C or None, largest t leaving 128 secret bits, P(more than t errors) for t = 0..120)."""
    s_of = anchored(mu, dict(zip(TEMPS, ber)))
    puf = SRAMPUF(DEFAULTS["chip_seed"], PUF_BITS, s_of(25), bias, 0.0)
    w_ = puf.read(25, rng_for(DEFAULTS["chip_seed"] * 7919 + 1)); worst_ = np.zeros(121)
    for T in range(0, 86, 5):
        s = s_of(T); pe = np.where(w_ == 1, ndtr(-puf.mismatch / s), ndtr(puf.mismatch / s))
        worst_ = np.maximum(worst_, np.cumsum(err_dist(pe)[::-1])[::-1][1:122])
    t_sec = max(t for t in range(121) if puf_residual({**DEFAULTS, "puf_bias": bias, "bch_t": t}) >= 128)
    return next((t for t in range(121) if worst_[t] < 1e-6), None), t_sec, worst_
def size_law(cal):                         # the same criterion with the linear law of Section VII-C (comparison only)
    Pl = {**DEFAULTS, **cal}
    puf = SRAMPUF(Pl["chip_seed"], PUF_BITS, Pl["puf_sigma"], Pl["puf_bias"], Pl["puf_tc"])
    wl = enroll(puf, 40, rng_for(Pl["chip_seed"] * 7919 + 1))["w"]
    dists = {T: err_dist(puf.p_err(wl, T)) for T in range(0, 86, 5)}
    return next((t for t in range(5, 121) if max(d[t + 1:].sum() for d in dists.values()) < 1e-6), None)
from scipy.special import ndtri
from scipy.stats import beta as _beta
BER_POST = [r["ber"] for r in an_post["rows"]] if PUF_MODE == "full" else PUF_POST_FULL["ber"]   # 1,024 cells x 8 power-ups
BER_SCH = [r["ber"] for r in an["rows"]]
t_meet, t_secret, worst = size_bch(float(ndtri(CAL["puf_bias"])), BER_POST, CAL["puf_bias"])
t_meet_sch, t_secret_sch, _ = size_bch(float(ndtri(CAL_SCH["puf_bias"])), BER_SCH, CAL_SCH["puf_bias"])
t_law = size_law(CAL)
i85 = TEMPS.index(85); ber85_law = float(ps.model_ber([85], CAL["puf_sigma"], CAL["puf_tc"], float(ndtri(CAL["puf_bias"])))[0])

# 3. The secret budget as an interval (review M2). The min-entropy per cell is bounded from below by the one-sided 95 %
#    Clopper-Pearson upper bound on max(P(1), 1 - P(1)), i.e. P(1) = 0.5 plus a bias margin of the observed bias and its
#    sampling error; the upper end of the interval is the point estimate. The 128-bit key is hashed from the fingerprint, so by
#    the leftover-hash lemma the secret must exceed 128 + 2 log2(1/eps) bits for a key within statistical distance eps of
#    uniform; eps = 2^-32 (SHA-256 is treated as a universal hash). Spatial correlation between cells is not modelled (the
#    SPICE draws are independent).
EPS_HASH = 2.0 ** -32
SECRET_REQ = 128 + int(round(2 * np.log2(1 / EPS_HASH)))                       # 192 bits
def h_bounds(ones, n, conf=0.95):
    """Min-entropy per cell: (lower confidence bound, point estimate), from `ones` of n cells powering up to 1."""
    k = max(int(ones), n - int(ones))
    return float(-np.log2(_beta.ppf(conf, k + 1, n - k))), float(-np.log2(k / n))
budget = lambda h, t, n=PUF_BITS: int(np.floor(n * h - 10 * t))                 # secret bits left after 10t helper bits
N_POST = PUF_POST_FULL["n_cells"]
H_TT = (h_bounds(round(CAL["puf_bias"] * N_POST), N_POST)[0], min_ent_bit(CAL["puf_bias"]))   # 1,024-cell post-layout run (tt)
t_bound = max(t for t in range(121) if budget(H_TT[0], t) >= SECRET_REQ)     # largest t meeting the requirement at the bound
DEFAULTS.update(puf_h_lo=H_TT[0], secret_req=SECRET_REQ)   # the helper-data leakage verdict (A11) uses the lower bound and 192 bits

# 4. Flip-aware key reliability over a chip population (review M3). Two-component cell model: cell i has the mismatch
#    m_i(T) = mu + m0_i cos(theta(T)) + mp_i sin(theta(T)) (m0, mp independent standard normal; theta, the rotation of
#    Section VII-F, makes a reproducible set of cells change sign) and power-up noise of rms s(T). s(T) comes from the
#    disagreement between repeated power-ups at T, theta(T) from the bit-error rate against the 25 °C enrolment; theta has the
#    opposite sign below 25 °C (a monotonic change of the nMOS/pMOS weight). Each chip is enrolled by its own noisy read(s);
#    P(fail) is exact per chip (Poisson-binomial). s(T) and theta(T) are piecewise linear with a knot at 25 °C, so the worst
#    case over 0-85 °C lies at 0, 25 or 85 °C.
_m0, _mp, _ne = rng_for(201).standard_normal(16384), rng_for(202).standard_normal(16384), rng_for(203).standard_normal(16384)
def s_from_q(mu, q):
    """Noise-to-mismatch ratio at which two power-ups of cells N(mu, 1) disagree with probability q."""
    m = _m0 + mu
    return brentq(lambda s: (2 * ndtr(m / s) * ndtr(-m / s)).mean() - q, 1e-5, 5.0)
def theta_from_ber(mu, s_e, s_T, ber):
    """Rotation (rad) at which a read with noise s_T disagrees with a one-read enrolment (noise s_e) with probability ber."""
    w = mu + _m0 + s_e * _ne > 0
    def f(th):
        m = mu + _m0 * np.cos(th) + _mp * np.sin(th)
        return np.where(w, ndtr(-m / s_T), ndtr(m / s_T)).mean() - ber
    return 0.0 if f(0.0) >= 0 else brentq(f, 0.0, np.pi / 2)
def two_component(mu, ber, q):
    """{T: s(T)}, {T: signed theta(T)} from the bit-error rates {T: ber} (25 °C enrolment) and power-up disagreements {T: q}."""
    sT = {T: s_from_q(mu, v) for T, v in q.items()}
    th = {T: (0.0 if T == 25 else np.sign(T - 25) * theta_from_ber(mu, sT[25], sT[T], ber[T])) for T in q}
    return sT, th
from numba import prange
@njit(cache=False, parallel=True)
def fail_tails(p, tmax):
    """P(more than t of the cells err), t = 0..tmax, for each row of p: exact, the probability mass leaving 'at most t'."""
    C, n = p.shape
    out = np.zeros((C, tmax + 1))
    for c in prange(C):
        d = np.zeros(tmax + 1); d[0] = 1.0
        for i in range(n):
            q = p[c, i]
            for k in range(tmax, 0, -1):
                out[c, k] += d[k] * q
                d[k] = d[k] * (1 - q) + d[k - 1] * q
            out[c, 0] += d[0] * q
            d[0] *= 1 - q
    return out
def chip_population(mu, sT, th, enrol=(25,), n_chips=1000, n_arr=PUF_BITS, n_key=PUF_BITS, reads=3, temps=(0, 25, 85),
                    tmax=120, seed=0):
    """Worst P(fail) over temps for t = 0..tmax, per chip, and the t each chip requires for 1e-6 (tmax + 1: none).
    enrol = (25,): one read at 25 °C; (Ta, Tb): `reads` reads at each, the cells stable and equal at both kept (enrolled
    value: the first read at Ta), the first n_key kept cells of the n_arr-cell array used (fewer kept: the chip fails)."""
    r = rng_for(seed); Ts = sorted(sT)
    s_of = lambda T: float(np.interp(T, Ts, [sT[x] for x in Ts])); th_of = lambda T: float(np.interp(T, Ts, [th[x] for x in Ts]))
    m0, mp = r.standard_normal((n_chips, n_arr)), r.standard_normal((n_chips, n_arr))
    mT = lambda T: mu + m0 * np.cos(th_of(T)) + mp * np.sin(th_of(T))
    read = lambda T: mT(T) + s_of(T) * r.standard_normal((n_chips, n_arr)) > 0
    if len(enrol) == 1:
        w, idx, kept = read(enrol[0]), None, None
    else:
        Ra, Rb = (np.stack([read(T) for _ in range(reads)]) for T in enrol)
        keep = (Ra.all(0) | ~Ra.any(0)) & (Rb.all(0) | ~Rb.any(0)) & (Ra[0] == Rb[0])
        w, kept = Ra[0], keep.sum(1)
        idx = np.argsort(~keep, axis=1, kind="stable")[:, :n_key]             # the first n_key kept cells
    worst = np.zeros((n_chips, tmax + 1))
    for T in temps:
        p = np.where(w, ndtr(-mT(T) / s_of(T)), ndtr(mT(T) / s_of(T)))
        worst = np.maximum(worst, fail_tails(p if idx is None else np.take_along_axis(p, idx, 1), tmax))
    if kept is not None:
        worst[kept < n_key] = 1.0
    ok = worst <= 1e-6
    return dict(worst=worst, kept=kept, treq=np.where(ok.any(1), ok.argmax(1), tmax + 1))
# the recorded 1,024-cell post-layout run: mean disagreement between the 28 pairs of its 8 power-ups at each temperature and the
# reproducible flips at 85 °C (dev/spice_puf/results/full_ng41_postlayout/bits.npy; recomputed from the live bits in full mode)
Q_POST = {-40: 0.012137, 25: 0.010917, 85: 0.016009}; FLIP85_POST = 0.0146
if PUF_MODE == "full" and mc_post["bits"].shape[2] >= 1024:
    _b = mc_post["bits"]; _R = _b.shape[1]
    Q_POST = {T: float(np.mean([(_b[i, a] != _b[i, c]).mean() for a in range(_R) for c in range(a + 1, _R)])) for i, T in enumerate(TEMPS)}
    _st = _b[i85].min(0) == _b[i85].max(0); FLIP85_POST = float((_st & (_b[i85, 0] != _b[TEMPS.index(25), 0])).mean())
mu_post = float(ndtri(CAL["puf_bias"]))
BER85_CI = PUF_POST_FULL["ber_ci"][i85]                                     # 95 % bootstrap interval over cells
POP = {}
for lab, b85 in (("lo", BER85_CI[0]), ("mid", BER_POST[i85]), ("hi", BER85_CI[1])):
    sT_, th_ = two_component(mu_post, {**dict(zip(TEMPS, BER_POST)), 85: b85}, Q_POST)
    POP[lab] = dict(chip_population(mu_post, sT_, th_, seed=DEFAULTS["chip_seed"]), sT=sT_, th=th_)
t99 = {k: int(np.percentile(v["treq"], 99, method="higher")) for k, v in POP.items()}
t50 = int(np.median(POP["mid"]["treq"]))

# 5. The code of the behavioural simulations. Sections VIII to XII (and the RTL of Section XII) keep
#    the code of the earlier sizing, t = the largest value leaving 128 secret bits at the point estimate of P(1); the card
#    states whether it meets the reliability target (population, upper end of the 85 °C interval) and the secrecy requirement
#    (lower bound of the budget). The schematic column uses the same rule.
t_star, t_sch = t_secret, t_secret_sch
REL_MET = t99["hi"] <= t_star
SEC_MET = budget(H_TT[0], t_star) >= SECRET_REQ
TARGET_MET = REL_MET and SEC_MET
CAL["bch_t"], CAL_SCH["bch_t"], before["bch_t"] = t_star, t_sch, DEFAULTS["bch_t"]
P = {**DEFAULTS, **CAL}
w = enroll(SRAMPUF(P["chip_seed"], PUF_BITS, P["puf_sigma"], P["puf_bias"], P["puf_tc"]), 40, rng_for(P["chip_seed"] * 7919 + 1))["w"]

src = dict(comp_noise="StrongARM noise of the extracted layout (Section VII-D), σn / (VDD/4)",
           int_noise="paper's 9.48 pA, referred to the modulator input (1.80e-5 FS) and multiplied by g1 = 0.5 at the first integrator's output",
           puf_sigma="SRAM-PUF Monte Carlo of the extracted cell (Section VII-D), noise / mismatch at 25 °C",
           puf_tc="the same Monte Carlo, temperature fit", puf_bias="the same Monte Carlo, fraction of cells powering up to 1",
           bch_t="the code of the behavioural simulations: the largest t that leaves 128 secret bits at the point estimate of P(1) "
                 "(the card below gives its reliability and its secret budget)")
table(["Setting", "Before", "Schematic", "Post-layout (used)", "Source"],
      [(f"<code>{k}</code>", f"{before[k]:.4g}", f"{CAL_SCH[k]:.4g}", f"<b>{CAL[k]:.4g}</b>", src[k]) for k in CAL],
      caption="Table 12: model settings replaced by SKY130 simulation (post-layout) and the measurements of [1]",
      align=["left", "right", "right", "right", "left"])
pct = lambda v: f"{100 * v:.1f} %"
card(f"BCH code and secret budget (post-layout, {PUF_BITS} cells, 25 °C enrolment, {len(POP['mid']['treq']):,} simulated chips): t = {t_star}",
     [("simulated bit-error rate, " + ", ".join(f"{T} °C" for T in TEMPS).replace("-", "−"), ", ".join(f"{100 * b:.2f} %" for b in BER_POST)
       + f" (85 °C: 95 % interval {100 * BER85_CI[0]:.2f}–{100 * BER85_CI[1]:.2f} %)"),
      ("two-component cell model at 85 °C", f"noise s = {POP['mid']['sT'][85]:.3f} (from repeated power-ups), rotation θ = "
       f"{np.degrees(POP['mid']['th'][85]):.1f}°: θ/π = {pct(POP['mid']['th'][85] / np.pi)} of the cells change sign "
       f"(reproducible flips counted in the recorded bits: {pct(FLIP85_POST)})"),
      ("t required for P(key fails) &lt; 10<sup>−6</sup> over 0–85 °C: median, 99th percentile of the chips",
       f"{t50}, <b>{t99['mid']}</b> (99th percentile {t99['lo']}–{t99['hi']} over the 85 °C interval)"),
      (f"chips meeting 10<sup>−6</sup> at t = {t_star}", f"<b>{pct(np.mean(POP['mid']['treq'] <= t_star))}</b> "
       f"({pct(min(np.mean(v['treq'] <= t_star) for v in POP.values()))} to {pct(max(np.mean(v['treq'] <= t_star) for v in POP.values()))} over the 85 °C interval)"),
      ("for comparison: noise-only model of one chip (the earlier method); linear law",
       f"t = {t_meet if t_meet is not None else '&gt; 120'}; t = {t_law} (the law gives {100 * ber85_law:.2f} % at 85 °C, against "
       f"{100 * BER_POST[i85]:.2f} % simulated)"),
      (f"min-entropy per cell, P(1) = {CAL['puf_bias']:.3f} from {N_POST:,} cells: 95 % lower bound – point estimate",
       f"{H_TT[0]:.3f}–{H_TT[1]:.3f} bit"),
      (f"secret bits left after publishing h at t = {t_star}", f"<b>{budget(H_TT[0], t_star)}–{budget(H_TT[1], t_star)}</b> "
       f"(need ≥ {SECRET_REQ} = 128 + 2 log<sub>2</sub>(1/ε), ε = 2<sup>−32</sup>)"),
      (f"largest t that leaves {SECRET_REQ} secret bits at the lower bound", f"t = {t_bound}"),
      ("the same sizing from the schematic values (noise-only, one chip)", f"t = {t_meet_sch} required; 128 secret bits at the point "
       f"estimate up to t = {t_secret_sch}"),
      ("verdict", "<b>both limits met</b>" if TARGET_MET else
       ("<b>reliable, not secret enough</b>" if REL_MET else "<b>neither limit met over the 85 °C interval</b>" if not SEC_MET else "<b>secret, not reliable</b>")
       + f": with {PUF_BITS} cells and a 25 °C enrolment, reliability needs t ≥ {t99['hi']} and secrecy allows t ≤ {t_bound}; "
       f"t = {t_star} is kept for the behavioural simulations, and Sections VIII-F and XI size the final key generator")],
     tone="ok" if TARGET_MET else "bad")
DEFAULTS.update(CAL)                       # every section below runs on these values (the explorer link of Section XIII too)

# robustness: the SPICE cell has thermal noise only (no 1/f, RTN or supply noise), so s25 likely underestimates silicon noise.
# With the conservative s25 = 0.06 (the value assumed before the SKY130 simulations), what would the code need?
pc = SRAMPUF(P["chip_seed"], PUF_BITS, 0.06, P["puf_bias"], P["puf_tc"])
dc = {T: err_dist(pc.p_err(w, T)) for T in range(0, 86, 5)}
t_cons = next((t for t in range(5, 61) if max(d[t + 1:].sum() for d in dc.values()) < 1e-6), None)
card(f"Robustness check with the conservative noise s<sub>25</sub> = 0.06 ({0.06 / CAL['puf_sigma']:.1f}× the SPICE value)",
     [("BCH strength needed (linear law, this chip)", (f"<b>t = {t_cons}</b>, leaving <b>{max(0, budget(H_TT[0], t_cons))}–{budget(H_TT[1], t_cons)} secret bits</b>" if budget(H_TT[1], t_cons) > 0
       else f"t = {t_cons}, which leaves no secret bits") + f" (need ≥ {SECRET_REQ})" if t_cons else "no t up to 60 meets 10<sup>−6</sup> with 600 cells")],
     note="The SPICE cell has thermal noise only, so real silicon may need more cells or a concatenated code (Section XV).", tone="bad")
````

**Key insights.** In Table 12, the comparator noise becomes `comp_noise` = 1.279 × 10⁻³ (from 10⁻³), the PUF's noise-to-mismatch ratio 0.02253 (from 0.06), its temperature coefficient 0.02051 (from 0.008) and its bias 0.4928 (from 0.5); `int_noise` = 0.899 × 10⁻⁵ at the first integrator's output (1.80 × 10⁻⁵ referred to the modulator input) **is essentially the value already used** (0.9 × 10⁻⁵).

* *Cell model at 85 °C.* The lower noise-to-mismatch ratio of the SKY130 cell does not carry over to the hot error rate, which is 3.2 times that at 25 °C. The repeated power-ups give a noise of s = 0.035; the rest of the error rate is a rotation θ = 4.9°, so θ/π = 2.7 % of the cells change sign. The recorded bits count 1.5 % reproducible flips (cells wrong in all eight hot power-ups), about half of the 3.14 %, as the prediction assumed; the remaining sign changes are expected to lie near the threshold, where they appear as unstable cells (not counted separately).
* *Reliability over the population.* The t required for 10⁻⁶ has a median of 29 and a 99th percentile of 39 (33 to 46 across the 85 °C interval), close to the predicted t ≈ 40. At t = 45, all chips meet the target at the point estimate, and 98.9 % at the upper end of the interval. The earlier noise-only sizing of the simulated chip requires t = 48, the linear law t = 32, and the schematic values, by the earlier method, t = 41: the flip-aware population model needs a weaker code than the noise-only model of one chip, because fixed flips vary less from power-up to power-up than the equivalent noise; the bounded entropy budget, below, is the larger change.
* *Secrecy.* The min-entropy per cell lies between 0.908 bit (95 % lower bound) and 0.979 bit (point estimate), as predicted. At t = 45, **94–137 secret bits** remain after the helper data is published, against the 192 required.
* *Verdict.* With 600 cells and a 25 °C enrolment, reliability over the 85 °C interval needs t ≥ 46 and secrecy (192 bits at the lower bound) allows t ≤ 35: neither the code of the simulations (t = 45) nor any other meets both. With the conservative $s_{25} = 0.06$ (2.7× the SPICE value), the law requires t = 60, which leaves no secret bits. **t = 45 is therefore a working setting for the system simulations, not the final key generator.** More entropy per corrected error comes from the two-temperature enrolment of Section VIII-F, which removes the flipping cells, and the larger arrays of Section XI.

### F. Process, voltage and temperature corners

The comparator's noise, offset and energy, measured as in Section VII-B, each have a predictable first-order dependence on process, supply and temperature:

* *Noise.* $\sigma_\mathrm{n}^2$ ≈ (kT/$C_\mathrm{P}$)·4γ($I_\mathrm{D} / g_m$)/$V_\mathrm{TH}$. The kT term alone predicts a rise by √(398.15 K/233.15 K) = 1.31 from −40 to 125 °C. The integration ends when the output nodes have fallen by about one threshold voltage, so a corner or a temperature that lowers the threshold shortens the integration and adds noise beyond kT. A higher supply raises the input common mode ($V_\mathrm{DD}/2$) and with it the overdrive $I_\mathrm{D}/g_m$ of the input pair, which also adds noise.
* *Offset.* Pelgrom's law depends on device area, not on the process corner, which shifts the mean thresholds rather than their mismatch; $\sigma_\mathrm{os}$ **should therefore be nearly the same at all corners**. It may rise when the integration gain decreases (hot), because the latch's mismatch is then referred to the input with less attenuation.
* *Energy.* The dynamic part, $C_\mathrm{sw} V_\mathrm{DD}^2$, should rise by $(1.98/1.62)^2 = 1.49$ from the lowest to the highest supply and not depend on temperature. Subthreshold leakage, which grows exponentially with temperature and is largest where the threshold is lowest, adds to it at hot, fast corners.

For the SRAM-PUF cell, the power-up race is decided while $V_\mathrm{DD}$ ramps through about 0.5–0.9 V, close to the threshold voltages. A process corner moves the nMOS and pMOS thresholds, in the same direction (ss, ff) or in opposite directions (sf: slow nMOS, fast pMOS; fs: fast nMOS, slow pMOS), which changes the supply at which a cell resolves and, above all, the relative weight of the nMOS and pMOS mismatch in its preference. **Two kinds of error are therefore expected against a 25 °C enrolment**: noise errors from unstable cells, and reproducible flips from cells whose preference reverses with temperature. Masking the cells unstable at 25 °C removes only the first; keeping only the cells stable and equal at two enrolment temperatures also removes the flips between them (Section VIII-F shows that the two must bracket the operating range). The supply corner is not simulated for this cell, because it decides long before $V_\mathrm{DD}$ reaches its final value.

Both extracted cells of Section VII-D are simulated at the five process corners of the PDK: tt, ss, ff, sf and fs.

* *Comparator:* 45 conditions, the 5 corners × 1.62, 1.8 and 1.98 V (±10 %) × −40, 27 and 125 °C, each with the quick noise, offset (the same 96 instances) and energy measurements and seeds of Section VII-B on the extracted core, the input common mode at $V_\mathrm{DD}/2$ (the noise-injection grid and window verified at tt and 27 °C only); the systematic offset is the 50 % point μ of the noise fit of the mismatch-free comparator, which the layout sets.
* *SRAM-PUF cell:* the 5 corners × −40, 25 and 85 °C, 512 cells × 5 power-ups per temperature, with the Monte Carlo and seeds of Section VII-C on the extracted cell and the noise envelope recomputed at each corner (power-ups at 125 °C not used, as in Sections VII-C and D). The linear temperature law cannot follow corners whose errors rise abruptly above 25 °C, so each corner receives a cell model anchored to its own simulated error rates (the s(T) that reproduces the simulated BER at each temperature, interpolated in between), which the compiler of Section XI searches with and which Section VIII-F extends with the mismatch rotation θ(T) derived below to evaluate the key. The 85 °C errors are split into reproducible flips and unstable cells, and the error rate after each masking is cross-validated: the cells to keep are chosen from power-ups 1–3, the error rate is measured on power-ups 4–5.
* *Recorded runs:* both grids are recorded runs of the same code and seeds, about 20 minutes (comparator) and 45 minutes (SRAM-PUF) on four cores; each script extracts the layout once and stops unless it is DRC- and LVS-clean, and its command (`python dev/spice_comparator/demo_pvt.py --pdk-root <pdk>`, `python dev/spice_puf/demo_pvt.py --pdk-root <pdk>`) is repeated in the result panel. The tables, the figure and the two analyses that follow are computed live from these records.

Fig. 38 sets these predictions against the grids: panels (a) and (b) give the comparator's noise and offset spread at every corner, supply and temperature, with the mean at each temperature dotted, **so that the effects of temperature and process can be told apart**, and panel (c) gives the SRAM-PUF cell's bit-error rate against temperature for each process corner.

#### Comparator energy across PVT: the dynamic and the leakage term

A StrongARM comparator draws no static current, so its energy per decision is

$$E \approx C_{sw}\,V_{DD}^2 + I_{leak}\,V_{DD}\,T_{clk},$$

where $C_{sw}$ is the switched capacitance (set by the devices and the layout), $I_{leak}$ the leakage current and $T_{clk}$ = 100 ns the clock period. The ratio $E/V_{DD}^2$ is the effective switched capacitance. As in Section VII-F, the analysis above predicts that the dynamic term makes $E/V_{DD}^2$ constant, independent of the supply and nearly independent of the temperature, and that **leakage raises it at 125 °C at the fast-nMOS corners (ff, fs)**.

The analysis uses the energy per decision of the recorded grid above at the 45 PVT conditions (output toggling every cycle); its metrics are $E$ and $E/V_{DD}^2$, grouped by process corner. Fig. 39 plots $E/V_{DD}^2$ at every condition against the grey band of its range at 27 °C, so that **a constant dynamic term appears as points inside that band and a leakage term as points above it**.

> **Next cell.** 1. Reads the energy per decision at the 45 conditions. 2. Computes the effective switched capacitance $E/V_{DD}^2$ and its range at 27 °C. 3. Finds the condition with the largest $E/V_{DD}^2$. 4. Prints the result panel and draws Fig. 39.

#### The SRAM-PUF error law across corners: noise and mismatch rotation

A closed-form law separates the SRAM-PUF errors of Table 14 and Table 15 into noise and reproducible flips:

*1) The power-up decision:* Cell $i$ settles to $b_i = [\,m_i + \sigma_n\,n > 0\,]$, where $m_i \sim \mathcal{N}(\mu, \sigma_m^2)$ is its mismatch, fixed at fabrication, $\mu$ the bias of the mismatch, $\sigma_m$ its spread, and $n \sim \mathcal{N}(0,1)$ the noise, drawn anew at every power-up with rms $\sigma_n$. **Only the noise-to-mismatch ratio $s = \sigma_n/\sigma_m$ and the bias matter**.

*2) The error rate in closed form:* A read disagrees with the enrolment when the decision variables at enrolment and at temperature $T$ have different signs, which for two jointly Gaussian variables with correlation $\rho$ happens with probability $\arccos(\rho)/\pi$ (Sheppard's formula). Two effects lower the correlation: noise, $s_e$ at enrolment and $s_T$ at $T$, and a **rotation** $\theta(T)$ of the mismatch itself: the relative weight of the nMOS and the pMOS mismatch changes with temperature, so that the mismatch at $T$ is $m' = m\cos\theta + m_\perp\sin\theta$, with $m_\perp$ an independent mismatch component of the same spread. With $\mu = 0$, the BER is
   $$\mathrm{BER}(T) = \frac{1}{\pi}\arccos\frac{\cos\theta(T)}{\sqrt{(1+s_e^2)(1+s_T^2)}} \;\approx\; \frac{1}{\pi}\sqrt{s_e^2 + s_T^2 + \theta^2}.$$
   Noise and reproducible flips add in quadrature. Two power-ups at the same temperature disagree with probability $\arccos\!\big(1/(1+s_T^2)\big)/\pi$, so the Monte Carlo gives $s_T$ directly, and $\theta$ then follows from the BER. θ is therefore not fitted and not an independent prediction: it is a reparametrisation of the BER in excess of the noise-only value. What can be tested is its physical reading: a rotation θ reverses the preferred value of a fraction θ/π of the cells, so θ/π should match the fraction of recorded cells wrong in every hot power-up. The analysis above predicts that, at the corners with a large excess, θ/π approaches the counted reproducible flips from above (the count misses the sign-changed cells that are also unstable).

The law is tested on the recorded Monte Carlo of all five corners (the SRAM-PUF grid above), enrolled on the first power-up at 25 °C. The noise-only BER ($\theta = 0$) uses $s_e = s_{25}$ and $s_T$, and $\theta(T)$ makes the closed form equal the simulated BER. **The independent check counts the reproducible flips at 85 °C**: cells with the same value in all five hot power-ups, opposite to the enrolment. The metrics are the BER, $\theta$, and θ/π against the counted flips.

> **Next cell.** 1. For each corner and temperature, measures the BER against the 25 °C enrolment and $s_T$ from repeated power-ups. 2. Computes the noise-only BER. 3. Extracts the rotation $\theta(T)$ from the excess BER. 4. Counts the reproducible flips at 85 °C and compares them with θ/π. 5. Prints the result panel and draws Fig. 40.

````python
#@title Fig.: SRAM-PUF error law across corners
# The SRAM-PUF error law against the transistor-level Monte Carlo of the five corners. BER vs a 25 °C
# enrolment (first power-up), s_T from the disagreement between repeated power-ups at T, the noise-only prediction
# arccos(1/sqrt((1+s_e^2)(1+s_T^2)))/pi, and the mismatch rotation theta(T) that explains the rest
i25 = TEMPS.index(25)
TH = {}
for j, c in enumerate(PVT["puf"]["corners"]):
    b = BITS[j].astype(np.int8); w0 = b[i25, 0]; R = b.shape[1]
    rows = []
    for i, T in enumerate(TEMPS):
        reads = b[i, 1:] if T == 25 else b[i]
        ber = float((reads != w0).mean())
        q = float(np.mean([(b[i, a] != b[i, a2]).mean() for a in range(R) for a2 in range(a + 1, R)]))   # same-T disagreement
        # two power-ups of a cell with mismatch m ~ N(0, 1) and noise s disagree with probability arccos(1/(1 + s^2))/pi
        # on average over the cells; inverting it gives the noise s_T from the measured same-temperature disagreement q
        rows.append(dict(T=T, ber=ber, sT=float(np.sqrt(max(1 / np.cos(np.pi * q) - 1, 0)))))
    se = rows[i25]["sT"]
    for r in rows:
        r["noise_only"] = float(np.arccos(1 / np.sqrt((1 + se ** 2) * (1 + r["sT"] ** 2))) / np.pi)
        # with a mismatch that rotates by theta between enrolment and T, the error rate becomes
        # arccos(cos(theta)/sqrt((1 + s_e^2)(1 + s_T^2)))/pi; solving the measured BER for theta gives the rotation
        r["theta"] = float(np.degrees(np.arccos(min(1.0, np.cos(np.pi * r["ber"]) * np.sqrt((1 + se ** 2) * (1 + r["sT"] ** 2))))))
    st85 = b[TEMPS.index(85)].min(axis=0) == b[TEMPS.index(85)].max(axis=0)                      # stable in every hot power-up
    TH[c["corner"]] = dict(rows=rows, flips85=float((st85 & (b[TEMPS.index(85), 0] != w0)).mean()))   # counted reproducible flips
FL = {k: (v["rows"][TEMPS.index(85)]["theta"] / 180, v["flips85"]) for k, v in TH.items()}   # theta/pi and the counted flips, 85 °C
TH = {k: v["rows"] for k, v in TH.items()}
worst = max(((k, r) for k, v in TH.items() for r in v if r["T"] == 85), key=lambda kr: kr[1]["theta"])
card("SRAM-PUF error law against the transistor-level Monte Carlo of the five corners",
     [("noise s<sub>T</sub> from repeated power-ups, " + ", ".join(f"{T} °C" for T in TEMPS).replace("-", "−"),
       ", ".join(f"{k} " + "/".join(f"{r['sT']:.3f}" for r in v) for k, v in TH.items())),
      ("BER at 85 °C: measured | noise-only theory", ", ".join(f"{k} {100 * v[TEMPS.index(85)]['ber']:.1f} | {100 * v[TEMPS.index(85)]['noise_only']:.1f} %"
                                                       for k, v in TH.items())),
      ("rotation θ at 85 °C (the excess over noise, reparametrised)", ", ".join(f"{k} {v[TEMPS.index(85)]['theta']:.0f}°" for k, v in TH.items())),
      ("θ/π against the counted reproducible flips at 85 °C (independent count)", ", ".join(f"{k} {100 * a_:.1f} | {100 * f_:.1f} %" for k, (a_, f_) in FL.items())
       + f": largest at {worst[0]}, θ = <b>{worst[1]['theta']:.0f}°</b>")], tone="ok")
pcol = {"tt": COL["ink"], "ss": COL["ok"], "ff": COL["bad"], "sf": COL["aqua"], "fs": COL["yellow"]}
pmk = {"tt": "o", "ss": "s", "ff": "^", "sf": "D", "fs": "v"}          # marker shape by corner as well (ff and fs differ in shape)
fig, axs = plt.subplots(1, 3, figsize=(W2, 2.9))
ax = axs[0]                                               # theta/pi from the BER against the counted flips (85 °C)
mx = 100 * max(max(v) for v in FL.values()) * 1.15
ax.plot([0, mx], [0, mx], ":", color=COL["faint"], lw=1)
for k, (a_, f_) in FL.items():
    ax.plot(100 * f_, 100 * a_, pmk[k], color=pcol[k], ms=6, mec=COL["ink"], mew=0.4, label=k)
ax.set_xlim(0, mx); ax.set_ylim(0, mx); ax.set_xlabel("Counted reproducible flips (%)"); ax.set_ylabel(r"$\theta/\pi$ from the BER (%)")
ax.legend(fontsize=7, ncol=2, loc="upper left"); panel(ax, "a", r"85 °C: $\theta/\pi$ vs flip count")
ax = axs[1]                                               # 85 °C: measured against noise only, per corner
i85 = TEMPS.index(85); xb = np.arange(len(TH))
meas = [100 * v[i85]["ber"] for v in TH.values()]; noise = [100 * v[i85]["noise_only"] for v in TH.values()]
ax.bar(xb - 0.19, meas, 0.36, color=[pcol[k] for k in TH], ec=COL["ink"], lw=0.5)
ax.bar(xb + 0.19, noise, 0.36, color="white", ec=COL["ink"], lw=0.5, hatch="////")
for x, y in zip(xb - 0.19, meas):
    ax.annotate(f"{y:.1f}", (x, y), xytext=(0, 2), textcoords="offset points", ha="center", va="bottom", fontsize=7.5)
ax.set_xticks(xb, list(TH)); ax.set_ylabel("BER at 85 °C vs 25 °C enrolment (%)"); ax.grid(axis="x", visible=False)
from matplotlib.patches import Patch
ax.set_ylim(0, 1.32 * max(meas))
ax.legend(handles=[Patch(fc=COL["faint"], ec=COL["ink"], lw=0.5, label="measured (corner colour)"),
                   Patch(fc="white", ec=COL["ink"], lw=0.5, hatch="////", label=r"noise only ($\theta = 0$)")], fontsize=7, loc="upper left")
panel(ax, "b", "85 °C: noise vs flips")
ax = axs[2]
for k, v in TH.items():
    ax.plot(TEMPS, [r["theta"] for r in v], "-", marker=pmk[k], color=pcol[k], ms=4, mec=COL["ink"], mew=0.3, lw=1.4, label=k)
ax.set_xticks(TEMPS); ax.set_xlabel("Temperature (°C)"); ax.set_ylabel(r"Mismatch rotation $\theta$ (°)")
ax.legend(fontsize=7, ncol=2, loc="upper left"); panel(ax, "c", r"the rotation $\theta(T)$")
fig.tight_layout(); plt.show()
````

**Key insights.**

* *Noise.* The repeated power-ups give $s_T$ between 0.009 (fs, −40 °C) and 0.064 (sf, 25 °C); at tt, 0.026, 0.020 and 0.023 at −40, 25 and 85 °C.
* *At 85 °C.* The simulated BER of the PVT grid exceeds the noise-only value at every corner, most at the fast-nMOS corners: 10.0 against 2.1 % at ff and 27.8 against 1.8 % at fs (tt 2.7 against 1.0 %, ss 2.0 against 1.4 %, sf 5.0 against 2.7 %; Fig. 40(b)).
* *The rotation.* The excess corresponds to a rotation of the mismatch of up to θ = 50° at fs and 85 °C (Fig. 40(c)); ff follows with 18°, the other corners stay below 10°.
* *The independent check.* θ/π and the counted reproducible flips at 85 °C are 27.8 and 25.6 % at fs, 9.8 and 8.4 % at ff, 4.2 and 2.5 % at sf, 2.5 and 2.0 % at tt and 1.4 and 1.0 % at ss (Fig. 40(a)). At every corner θ/π lies above the count, by 0.4 to 2.2 percentage points, as predicted: the independent flip count confirms that θ, solved from the BER in excess of the measured noise, **describes a fixed set of cells that reverse their preferred value, not extra noise**.

**In the SKY130 Monte Carlo models, the mismatch itself rotates with temperature at the fast-nMOS corners, and the cells reverse their preferred state**. Noise-based reasoning at 25 °C cannot reveal it: fs, the corner with the most flips at 85 °C, has the lowest noise of all corners at 25 °C. Section VIII-F builds the key-failure model on these two components.

#### The fast-nMOS flips: mechanism check and supply ramp

The rotation θ of Fig. 40, largest at ff and fs, **admits a competing explanation**, and every recorded run powers the cell up with the same 1 µs supply ramp:

*1) Two candidate mechanisms:* The error law reads the excess BER at 85 °C as a rotation of the latch's mismatch with temperature, a property of the cell. Every recorded run, however, powers the cell up with both bit lines tied to the ramping supply and the word line at 0 V; the pass gates are then off but leak, most when hot and at fast nMOS, injecting a current that carries their own mismatch. That would also make the preferred value depend on temperature, but **as a property of the testbench, not of a cell in an array**, whose bit lines are precharged or floating at power-up.

*2) Discriminating test:* Flips from the latch do not change when the bit lines float or are grounded, nor when the pass gates' mismatch is switched off; flips from pass-gate leakage fall in all three cases (no leakage path from the supply, or equal leakage on both sides). If the rotation is the cause, **the BER and the reproducible flips at 85 °C should agree between the four testbenches** within the counting scatter. Fig. 41 shows the floating-bit-line testbench with the series source of point 3.

*3) Direct measurement of the rotation:* A voltage source in series with the gate of one inverter, bisected without noise to the value at which the cell's decision changes, gives each cell's input-referred mismatch $\Delta V_\mathrm{m}$; the correlation ρ of $\Delta V_\mathrm{m}$ between 25 and 85 °C gives the rotation directly, θ = arccos ρ, independently of the BER from which Fig. 40 derives it. **The two values of θ should agree**.

*4) Supply ramp:* The cell resolves at a supply voltage, not a time, so **a slower ramp should change little**, whereas a ramp faster than the cell's time constant can raise the resolve voltage and change marginal cells. From 1 to 100 µs, the BER against the 1 µs enrolment should therefore stay near the 25 °C repeatability; at 0.1 µs it may rise.

Both tests are recorded transistor-level runs of the extracted 6T cell of Section VII-D: 128 cells with the mismatch draws and noise sequences of the first 128 cells of the PVT grid, so that **the testbench with tied bit lines reproduces the recorded grid**.

* *Mechanism:* corners ff and fs, 25 and 85 °C, five noisy power-ups, one noiseless power-up and the bisection of $\Delta V_\mathrm{m}$ (nine steps over ±0.2 V) per temperature, in four testbenches: bit lines tied to $V_\mathrm{DD}$ (as recorded), floating (20 fF to ground each), grounded, and tied with the pass-gate mismatch switched off (the latch transistors keep theirs).
* *Ramp:* typical corner, 25 °C, bit lines tied, ramps of 0.1, 1, 10 and 100 µs, five noisy power-ups each.
* *Metrics:* the BER at 85 °C against the 25 °C enrolment, the reproducible flips, θ from $\Delta V_\mathrm{m}$ and from the BER, and the BER of each ramp against the 1 µs enrolment and against its own first power-up.

The run (about 44 minutes of ngspice time, 22 minutes on two cores) is recorded; the cell reads its compact summary. Fig. 42 **answers the three questions in turn**: panel (a) compares the BER of the four testbenches at 85 °C, split into reproducible flips and other errors, panel (b) plots each cell's $\Delta V_\mathrm{m}$ at 25 °C against 85 °C with the reversed signs marked, and panel (c) gives the BER against ramp time.

### G. Ageing of the SRAM-PUF cell (recorded, assumed BTI law)

Subsections C to F characterise a fresh cell, but the key generator must regenerate the same key throughout the chip's life, and transistors age. In a 6T cell that holds a value in a process with a silicon-dioxide gate, **the dominant mechanism is negative-bias temperature instability (NBTI)**: the pull-up pMOS that is on (gate at 0 V, source and body at $V_\mathrm{DD}$) slowly loses drive as charge builds up in and at its gate oxide, which raises its threshold magnitude $|V_\mathrm{th}|$ [27], [28]. Under constant stress the shift follows a power law of the stress time $\tau$,

$$\Delta V_\mathrm{th} = A \left(\frac{\tau}{1\ \mathrm{year}}\right)^{n},$$

with the time exponent $n$ between about 1/6 and 1/4 and the one-year shift $A$ thermally activated and growing with the gate-oxide field; part of the shift recovers when the stress is removed. **The nMOS are not aged**: positive-bias instability is assumed small, as usual with a silicon-dioxide gate, and hot-carrier degradation needs the current of a switching transistor [27], which a cell holding its value does not draw.

*Emulated ageing* (Fig. 43). The SKY130 PDK has no ageing model (neither BTI nor hot-carrier degradation), so ngspice cannot age the cell by itself, and this subsection emulates the ageing instead. The law above is taken from the model of [27] and the measurements it was fitted to [28], [29], scaled to the SKY130 oxide, supply and temperature; it gives each pull-up's threshold shift from the time that pull-up is on, which depends on the value the cell holds. The shift is imposed on the extracted cell as a voltage source in series with the pull-up's gate, which acts as the same rise of $|V_\mathrm{th}|$, and the recorded post-layout Monte Carlo of Subsection D is re-run with it at 1, 3 and 10 years and at −40, 25 and 85 °C, with the same mismatch and noise seeds; Section VIII-F carries the result to the key. **$A$ and $n$ are therefore assumptions, and every number of this subsection is conditional on them.**

*Dependence on the held value.* A cell that holds $q = 1$ keeps pull-up PUa (which drives $q$) on; one that holds 0 keeps PUb on. A dedicated PUF array that is never written holds its power-up value, so the stressed pull-up is that of the cell's preferred side, and its weakening moves the cell towards the threshold. If instead the inverse of every read is written back into the array [30], **the other pull-up ages and the cell moves away from the threshold**. Referred to the gate of PUa, cell $i$ has a mismatch $u_i$, the gate shift at which its noiseless power-up turns from 1 to 0, and noise of rms $\sigma_\mathrm{n}$; a shift $\Delta_i = \Delta V_{\mathrm{th},a} - \Delta V_{\mathrm{th},b}$ of the two pull-ups acts like a shift of that gate, so that $P(\text{read } 1) = \Phi\big((u_i - \Delta_i)/\sigma_\mathrm{n}\big)$, with $\Phi$ the standard normal distribution function.

*Prediction.* With the held value every stable cell moves towards the threshold by $A(\tau/1\,\mathrm{y})^n$, and the cells whose $|u_i|$ is smaller reach it. There a cell starts to read its other value and so to stress its other pull-up, and because the shift grows as $\tau^n$ with $n < 1$, the first stress of that pull-up counts most: the cell is held just on its preferred side, where noise flips some of its reads. The error rate against the enrolment therefore grows with age, by less than the $0.4\,\Delta/\sigma_u$ that cells reading at random would add (half the fraction of a Gaussian spread $\sigma_u$ within $\Delta$ of the threshold). With the write-back every cell moves away from the threshold, the error rate falls below its fresh value, and the unstable cells become stable. Quantitatively, the stress of each pull-up follows from the cell's own probability of reading 1, so the shifts are integrated on a fine time grid with $P(\text{read } 1) = \Phi\big((u_i - \Delta_i)/\sigma_\mathrm{n}\big)$, and the same formula, with $u_i$ and $\sigma_\mathrm{n}$ measured on the fresh cell, predicts each aged error rate that the transistor-level runs then measure. Fig. 44 tests the prediction in both directions: panels (a) and (b) give the error rate and the unstable cells against years of operation for the held value and the inverted write-back, with the per-cell prediction as crosses, and panel (c) shows how the mismatch distribution moves relative to the noise band around the threshold. Because $\Delta$ grows as $\tau^n$ with $n$ near 1/6, **most of the 10-year shift is reached within the first year**. Powering the array down between reads helps in proportion: the shift depends only on the time each pull-up is on, so an array powered 1 % of the time reaches every state a hundred times later, although its shift after ten years is still $0.01^{1/6} \approx 46$ % of the full-time value.

The study is a recorded run of the same flow (`python dev/spice_puf/demo_ageing.py` in three stages, summarised by `ageing_report.py`), with the following settings.

* *Cell and Monte Carlo:* the extracted cell and the recorded post-layout Monte Carlo of Subsection D, 1,024 cells × 8 power-ups at −40, 25 and 85 °C, with the same mismatch and noise seeds, so that an aged read differs from the fresh read of the same cell only by the ageing. With no shift, the aged netlist reproduces all 1,536 recorded reads of the first 64 cells.
* *Shift:* nominal law $A$ = 2.83 mV, $n$ = 1/6 (the long-time exponent of hydrogen diffusion [27]), scaled from 10 mV after $1.5 \times 10^4$ s at a gate-source voltage $|V_\mathrm{GS}|$ = 2.9 V and 100 °C on a 0.18-µm pMOS [28] (Fig. 2 of [27]) with the oxide-field and temperature terms of [27] (field constant $E_0$ = 2.0 MV/cm, activation energy $E_\mathrm{a}$ = 0.13 eV) to $|V_\mathrm{GS}|$ = 1.8 V, the 4.23 nm oxide of the SKY130 pMOS, $|V_\mathrm{th}|$ = 0.45 V and 37 °C, the capsule's temperature; the anchor's oxide is taken as 4 nm, the thickest of [28], and a thinner one or a larger $|V_\mathrm{th}|$ gives a smaller shift (0.65 V: 28 % smaller). Pessimistic law $A$ = 7.88 mV, $n$ = 1/4, from 18 mV after $5 \times 10^3$ s at 2.9 V and 150 °C on a 3.8 nm oxide [29], scaled the same way: 14.0 mV after 10 years, more than the nominal law gives at 85 °C (8.0 mV). No recovery is modelled and the array is powered all the time (the worst case); the mobility loss that accompanies NBTI is not modelled. The aged states are 1, 3 and 10 years with the nominal law, and 10 years with the pessimistic law as a second test of the prediction.
* *Stress:* each pull-up's stress time is integrated on a fine logarithmic time grid at the rate of the cell's probability, from the calibration at 25 °C, of holding the value that turns it on (held value) or the opposite (write-back); one-year steps with the reads of the previous state cannot follow a cell near the threshold, which they drive past it.
* *Calibration:* $u_i$ at each temperature by a noiseless bisection of the PUa gate shift (10 steps over ±250 mV, 0.5 mV resolution); $\sigma_\mathrm{n}$ from the disagreement between the fresh power-ups.
* *Metrics:* the bit-error rate (BER) against the fresh 25 °C enrolment (its first power-up), the unstable cells (not all eight reads equal) and the predicted BER.

````python
#@title Fig.: SRAM-PUF cell ageing (recorded, assumed BTI law)
# Ageing of the extracted SRAM-PUF cell by NBTI (recorded; ASSUMED BTI parameters: the PDK has no ageing model).
# Recorded transistor-level runs (dev/spice_puf/demo_ageing.py, ngspice 41; summary by dev/spice_puf/ageing_report.py),
# loaded from the figures repository (data/puf_ageing.json): the fresh and aged reads of the 1,024 cells of Section VII-D against the fresh 25 °C enrolment, the
# calibration (each cell's mismatch referred to the gate of pull-up PUa) and the per-cell prediction from it.
AGE = recorded("puf_ageing")

def _puf_ageing():                                        # local names only
    from scipy.special import ndtr
    st, cal, fr = AGE["settings"], {r["T"]: r for r in AGE["calib"]}, AGE["fresh"]
    yrs, Ts = [0] + st["years"], st["temps"]
    row = lambda tag, T: next(r for r in (fr if tag is None else AGE["aged"][tag]["rows"]) if r["T"] == T)
    tag = lambda sc, y: f"{sc}_{y:g}y"
    pred = lambda sc, y: AGE["pred_fresh"] if y == 0 else AGE["aged"][tag(sc, y)]["pred"]
    # differences between the simulated and the predicted BER over every recorded aged state of one law
    errs = lambda pess: [abs(r["ber"] - p) for k, v in AGE["aged"].items() if k.startswith("pess_") == pess
                         for r, p in zip(v["rows"], v["pred"])]
    u25 = np.array([np.nan if v is None else v for v in AGE["u25_mV"]])
    pc = lambda v: f"{100 * v:.2f} %"
    nfmt = lambda n: {round(1 / 6, 6): "1/6", 0.25: "1/4"}.get(round(n, 6), f"{n:g}")
    laws = st["laws"]
    bers = lambda sc, T: " | ".join(pc(row(None if y == 0 else tag(sc, y), T)["ber"]) for y in yrs)
    # cells pulled to the threshold that read at random would add half the fraction of the spread within the shift
    s10 = {sc: np.array(AGE["shift10_mV"][sc]) for sc in ("hold", "invert")}
    sh10 = AGE["aged"][tag("hold", 10)]["shift_stable_mV"]
    bound = 0.5 * (2 * ndtr(sh10 / cal[25]["sigma_mV"]) - 1)
    near = lambda d: int(np.sum(np.abs(d) < 3 * cal[25]["noise_mV"]))   # cells inside the grey band of panel (c)
    rows = [("assumed laws (the PDK has no ageing model)",
             "; ".join(f"{k}: ΔV<sub>th</sub> = {1e3 * a:.2f} mV × (τ / 1 year)<sup>{nfmt(n)}</sup>" for k, (a, n) in laws.items())
             + f" on the pull-up that is on ({st['law_note']})"),
            ("zero shift reproduces the recorded fresh reads (chunk 0)",
             f"{100 * AGE['check']['identical_fraction']:.1f} % of {AGE['check']['reads']:,} reads"),
            ("mismatch referred to the PUa gate: spread σ<sub>u</sub> at −40 | 25 | 85 °C",
             " | ".join(f"{cal[T]['sigma_mV']:.1f}" for T in Ts) + " mV; noise referred to it: "
             + " | ".join(f"{cal[T]['noise_mV']:.2f}" for T in Ts) + " mV"),
            ("rotation from the per-cell correlation, 25 → 85 °C | from the error rate (Section VII-E)",
             f"{np.degrees(np.arccos(AGE['rho']['25_85'])):.1f}° | {np.degrees(abs(AGE['model']['th']['85'])):.1f}°"),
            ("shift of a stable cell after 1 | 3 | 10 years, nominal law (dV<sub>a</sub> − dV<sub>b</sub>, median)",
             " | ".join(f"{AGE['aged'][tag('hold', y)]['shift_stable_mV']:.2f}" for y in st["years"]) + " mV"),
            ("BER against the fresh 25 °C enrolment, held value: fresh | 1 | 3 | 10 years, 25 °C",
             bers("hold", 25)),
            ("the same at 85 °C", bers("hold", 85)),
            ("inverted write-back: fresh | 1 | 3 | 10 years, 25 °C", bers("invert", 25)),
            ("the same at 85 °C", bers("invert", 85)),
            ("bound: cells within the 10-year shift reading at random would add, at 25 °C",
             f"{100 * bound:.2f} percentage points"),
            ("unstable cells at 25 °C (8 reads not all equal): fresh | 10 years held | inverted",
             f"{pc(row(None, 25)['unstable'])} | {pc(row(tag('hold', 10), 25)['unstable'])} | {pc(row(tag('invert', 10), 25)['unstable'])}"),
            ("cells within three times the noise of the threshold at 25 °C (grey band in c): fresh | 10 years held | inverted; "
             "held cells that crossed to their other side",
             f"{near(u25)} | {near(u25 - s10['hold'])} | {near(u25 - s10['invert'])}; {int(np.sum((u25 - s10['hold']) * u25 < 0))}")]
    if "pess_hold_10y" in AGE["aged"]:
        pr = lambda sc: " | ".join(f"{pc(r['ber'])} ({pc(p)})" for r, p in zip(AGE["aged"][f"pess_{sc}_10y"]["rows"],
                                                                              AGE["aged"][f"pess_{sc}_10y"]["pred"]))
        rows.append(("pessimistic law, 10 years: shift of a stable cell; BER (per-cell prediction) at −40 | 25 | 85 °C, "
                     "held value; inverted write-back",
                     f"{AGE['aged']['pess_hold_10y']['shift_stable_mV']:.2f} mV; {pr('hold')}; {pr('invert')}"))
        pi = AGE["aged"]["pess_invert_10y"]
        rows.append(("pessimistic law, 10 years, inverted: cells with all 24 reads equal (8 power-ups × 3 temperatures) | "
                     "cells read wrong at least once | of these, cells whose enrolment read differs from their noiseless value",
                     f"{pi['cells_stable_all']:,} of {st['cells']:,} | {pi['cells_wrong']} | {pi['cells_wrong_minority']} "
                     f"(such enrolment reads: {AGE['enrol_minority']} cells)"))
    rows += [("largest difference between a simulated aged BER and its per-cell prediction from the calibration: "
              "nominal | pessimistic law",
              " | ".join(f"<b>{100 * max(e):.2f} percentage points</b> over {len(e)} conditions" for e in (errs(False), errs(True)) if e)),
             ("recorded run", f"{st['cells']:,} cells × {st['repeats']} power-ups × 3 temperatures × {len(AGE['aged'])} aged states, "
              f"ngspice {st['ngspice']}, {AGE['cpu_s_total'] / 7200:.1f} h on two cores: "
              "<code>python dev/spice_puf/demo_ageing.py --stage age</code>")]
    card("SRAM-PUF cell ageing by NBTI (extracted cell, recorded; assumed BTI law)", rows, tone="info")

    tc = {-40: COL["aqua"], 25: COL["ink"], 85: COL["bad"]}
    fig, axs = plt.subplots(1, 3, figsize=(W2, 3.2))
    for ax, key, lab, title in ((axs[0], "ber", "a", "BER against the fresh enrolment"),
                                (axs[1], "unstable", "b", "Unstable cells")):
        for T in Ts:
            for sc, ls, mk in (("hold", "-", "o"), ("invert", "--", "s")):
                v = [100 * row(None if y == 0 else tag(sc, y), T)[key] for y in yrs]
                ax.plot(yrs, v, ls=ls, marker=mk, ms=3.5, color=tc[T], mfc=tc[T] if sc == "hold" else "white", lw=1.2)
                if key == "ber":                          # 95 % interval over cells, and the per-cell prediction
                    ci = [row(None if y == 0 else tag(sc, y), T)["ber_ci"] for y in yrs]
                    ax.vlines(yrs, [100 * c[0] for c in ci], [100 * c[1] for c in ci], color=tc[T], lw=0.7)
                    ax.plot(yrs, [100 * pred(sc, y)[Ts.index(T)] for y in yrs], "x", ms=4, color=tc[T], mew=0.8)
        ax.set_xlabel("Years of operation"); ax.set_ylabel(("BER" if key == "ber" else "Unstable cells") + " (%)")
        ax.set_xticks(yrs); ax.set_ylim(0, None); panel(ax, lab, title)
    from matplotlib.lines import Line2D                       # one legend row under the figure for (a) and (b)
    fig.legend([Line2D([], [], color=tc[T], lw=1.2) for T in Ts] +
               [Line2D([], [], color=COL["ink"], ls="-", marker="o", ms=3.5),
                Line2D([], [], color=COL["ink"], ls="--", marker="s", ms=3.5, mfc="white"),
                Line2D([], [], color=COL["ink"], ls="", marker="x", ms=4)],
               [f"{T} °C".replace("-", "−") for T in Ts] + ["held value", "inverted write-back", "prediction"],
               loc="lower center", bbox_to_anchor=(0.5, 0.0), ncol=6, frameon=False, fontsize=8, handlelength=2.0,
               columnspacing=1.4)
    ax = axs[2]                                               # mismatch at 25 °C referred to the PUa gate, fresh and aged
    edges = np.linspace(-150, 150, 61)
    ax.axvspan(-3 * cal[25]["noise_mV"], 3 * cal[25]["noise_mV"], color=COL["band"], alpha=0.8, lw=0)
    top = max(ax.hist(u25, edges, histtype="stepfilled", color=COL["faint"], alpha=0.6, label="fresh")[0].max(),
              ax.hist(u25 - s10["hold"], edges, histtype="step", color=COL["bad"], lw=1.3, label="held")[0].max(),
              ax.hist(u25 - s10["invert"], edges, histtype="step", color=COL["ok"], lw=1.3, ls="--", label="inverted")[0].max())
    ax.set_xlim(-150, 150); ax.set_ylim(0, 1.12 * top)       # just above the tallest bar; the legend sits over the low tail
    ax.set_xlabel(r"$u_i - \Delta_i$ at 25 °C (mV)"); ax.set_ylabel("Cells per 5 mV")
    ax.legend(fontsize=7.5, loc="upper left", handlelength=1.6); panel(ax, "c", "Mismatch at the PUa gate")
    fig.tight_layout(rect=(0, 0.07, 1, 1)); show()

_puf_ageing()
````

**Key insights.**

* *Calibration.* Referred to the gate of PUa, the cells' mismatch has a spread $\sigma_u$ of 55.6 mV at 25 °C (56.0 and 53.2 mV at −40 and 85 °C) and the noise an rms of 1.28 mV; the per-cell correlation of the mismatch between 25 and 85 °C gives a rotation of 4.5°, against 4.9° from the error rate in Section VII-E, an independent confirmation of the cell model at tt.
* *Held value.* The shift of a stable cell grows from 2.83 mV after one year to 4.16 mV after ten (nominal law), small against $\sigma_u$, but it moves every cell towards the threshold: the BER at 25 °C rises from 0.99 % fresh to 1.18 % after one year and 1.53 % after ten, at 85 °C from 3.14 % to 4.68 % (Fig. 44(a)). The rise is well below the 2.98 percentage points that cells reading at random would add, as predicted: the cells pulled to the threshold stay on their preferred side (none of the 1,024 crosses it) and gather near it (58 cells within three times the noise of the threshold fresh, 100 after ten years; Fig. 44(c)), and the unstable cells rise from 2.83 % to 5.57 %. With the pessimistic law (14.01 mV after ten years) the BER reaches 4.87 % at 25 °C and 9.84 % at 85 °C.
* *Inverted write-back.* The same shifts push every cell away from the threshold: the BER at 25 °C falls to 0.78 % after one year and 0.72 % after ten (85 °C: 1.35 %), the unstable cells to 1.17 %, and the cells within three times the noise of the threshold to 24, the dip in Fig. 44(c). With the pessimistic law every cell is made stable: all 1,024 cells read the same value in all 24 reads, and the 0.59 % that remains at every temperature are the 6 cells whose single enrolment read differs from their noiseless value, a residue of the enrolment, not of the ageing.
* *Prediction.* From the fresh calibration alone, the per-cell formula predicts every aged BER of the nominal law within 0.20 percentage points over its 18 conditions (Fig. 44(a), crosses). With the pessimistic law the largest difference is 1.08 points, held at 25 °C (3.79 % predicted, 4.87 % simulated): the stress follows the read probability at 25 °C and so pulls the cells onto the 25 °C threshold, where a read depends most on the noise model, and the stronger law pulls more of them there. **At transistor level, an NBTI shift of the pull-ups acts on the cell as a shift of its mismatch**, which allows Study F, part 3 (Section VIII-F) to carry the ageing to the key with the cell model.
* *Implications.* Which way the error rate drifts is set by how the array is operated, not by the circuit: a held array ages towards more errors, a written-back one towards fewer. The drift is small at block level, but the code of the key generator is sized with little margin, so Section VIII-F evaluates its effect on the key.

# **VIII. From transistor to system: accuracy, power and security**
---
Section VII hands over two circuits characterised at transistor level across process, supply voltage and temperature (PVT), the StrongARM comparator of the modulator and the 6T SRAM-PUF cell, and a BCH code sized for the cell. Block-level figures of merit do not establish whether the sensor works or is secure; this section therefore determines the effect of the circuit values on the system: the accuracy of the reading, the power and, above all, the **security** of the original design of [1] and of the proposed design. It takes the bit-exact model and its attack tools (Section V, consistent with [1], Section VI), the circuit values and BCH code (Section VII-E) and the recorded PVT grids (Section VII-F).

Three parts prepare the studies: a table traces each circuit quantity along its path to the system metric it affects and states the PVT coverage and its limits; the simulation setup fixes the level of abstraction, the common settings and the replica attack with its security criterion; and one frame of the complete proposed design, at the nominal condition and at a worst condition, checks that the blocks operate together and serves as the reference for studies A to G. Study A finds how much the comparator's noise and offset degrade the accuracy at the 45 PVT conditions, and how large each must become before the accuracy is lost. **The original design's only defence against an attacker who owns a second chip is the analog difference between the chips** (Section IV): study B asks whether the comparator provides it at any PVT condition, first through its offset alone and then through noise and offset, and, if not, how large its noise or offset must be and at what cost in accuracy; an interactive explorer then sets both and shows what the original design gains in security and loses in accuracy. Study C finds the comparator's share of the sensor's power at the 45 PVT conditions, and whether it affects the comparison of designs in Section X. Study D finds what the modulator requires for accuracy (the lumped front-end noise of [1] and the OTA DC gain), and whether this chip-to-chip noise could protect the original design where the comparator could not; study E asks whether the chip-to-chip mismatch of the integrator coefficients protects it, and at what mismatch. Study F derives what the noise and temperature behaviour of the SRAM-PUF cell do to the reliability and secrecy of the proposed design's key over a population of chips: whether the BCH code, which corrects up to t = 45 flipped cells, makes the key reliable at every corner, which enrolment removes the reproducible flips, and which code then meets both requirements. A second explorer sizes the key generator for other choices of corner, enrolment, array, BCH correction and key-failure target, and the last part of F measures how random one power-up read of the same cells is at every corner and temperature, to size the number of reads hashed into the chip nonce, which keeps a session's keystream fresh when an attacker replays an old session nonce (A15, Section IX). A third part carries the ageing of Section VII-G to the key: for how many years each code keeps its target with the held value and with the inverted write-back, and what the write-back costs the chip nonce. G collects the results of studies A to F in one table, stating which circuits impose a requirement and of what kind.

### Simulation setup

**Abstraction level: behavioural, with circuit parameters from SPICE.** The studies run the bit-exact model of Section V (a discrete-time loop, one step per clock, without excess loop delay) with the SKY130-derived settings of Section VII-E. In each run only the quantity under study changes. The comparator enters as a decision threshold: at clock $k$ it decides $v[k] = \operatorname{sgn}(x_2[k] + o + n[k])$, where $x_2$ is the second integrator's output, $o$ the chip's offset (one draw from $\mathcal{N}(0, \sigma_{os}^2)$ per chip) and $n[k]$ its noise (a new draw from $\mathcal{N}(0, \sigma_n^2)$ at every decision). The SRAM cell enters as its SPICE noise-to-mismatch ratio, temperature behaviour and bias. This makes thousands of system runs possible (one 107,400-clock frame at transistor level would take hours); Section XII tests the abstraction decision by decision.

**Model units.** $x_2 = 1$ corresponds to $V_{DD}/4$ at the comparator input, with $V_{DD}$ the supply voltage (Section VII-B), 450 mV at 1.8 V, so the post-layout $\sigma_\mathrm{n}$ = 0.575 mV is $1.28 \times 10^{-3}$ model units.

**The replica attack** (the security metric of studies B, D and E). The original design transmits $C = D \oplus K$, the modulator bits $D$ XORed with the keystream $K$ of a 16-bit LFSR. The attacker owns a second chip, a *replica*, applies the same known input and reads its unencrypted modulator bits $\hat{D}$. The keystream estimate is

$$\hat{K} = C \oplus \hat{D} = K \oplus (D \oplus \hat{D}),$$

which is correct wherever the two chips made the same decisions. A 64-bit window of $\hat{K}$ ***yields the key*** when the LFSR that Berlekamp-Massey finds in it predicts the next 512 keystream bits with more than 99 % accuracy; the attacker can confirm a candidate by decrypting the chip ID. Two attackers are simulated on the same pairs.

* *The full-scan, multi-frame attacker (the security criterion)* records up to four frames (each with a new noise realisation of both chips) and examines **every** 64-bit window from clock 200 on (after the common reset state), about 29,700 windows per frame. Berlekamp-Massey tolerates a wrong bit only at the start of a window (LFSR length L = 16): in the testbench's check (20 trials per position), a wrong bit among the first 16 never prevented the key from being found, one at bit 17 did in 19 of 20 trials, and one at bit 18 or later always. The scan therefore runs on every window whose last 40 bits are clean and stops at the first that yields the key; a pair is *broken* when any window of its frames does.
* *The attacker of [1] (secondary)* runs `window_attack` (Section V-E), Berlekamp-Massey on 80 random windows of the first frame; a criterion based on it would count a design as protected when it compromised fewer than 10 % of the pairs. It serves for comparison and for the theory.

**Security criterion.** A design is secure against the replica attack when the full-scan attacker's success probability is negligible. A simulation can only bound this probability, so a setting is called **protected** when that attacker breaks none of at least 100 independent pairs: the 95 % Clopper-Pearson upper bound on its success probability per pair is then 3.6 %, and, if the frames of a pair are treated as independent (they share the chips' offsets and dither seeds, so this bound is optimistic), 0.92 % or less per frame (at least 400 frames). Every broken fraction is given with its 95 % Clopper-Pearson interval (error bars in the figures).

**Pairs and statistics.** Every pair draws its own seeds from the pair index (offsets, capacitor-ratio errors of study E, dither and noise), so all pairs, also at different swept points, are independent. Every swept point starts with 24 pairs; each point in the transition (some but not all pairs broken), and the first point above it at which no pair is broken, is raised to 100 pairs, the minimum of the criterion. Each PVT condition has 8 pairs (360 in all). For the theory, the testbench also measures in the first frame of each pair the fraction p of differing bits and the fraction $f_{64}$ of 64-bit windows in which no bit differs (a *clean* window).

**Bare loop:** the loop without dither, noise or clipping, for the exact identities of studies A and E.

**Recorded and live.** All system runs execute live, in at most about 20 s per study on two cores (run times in the result panels and in How to run); only the circuit inputs at PVT come from the recorded grids of Section VII-F.

> **Next cell.** 1. Builds the SKY130-derived settings `P_sys`: the model's defaults with the post-layout circuit values. 2. Generates the two inputs (the sine for accuracy, the constant 0.5 µA for the attack) and the original design's keystream. 3. Defines the helpers `sndr_at`, `replica_pairs` (both attackers on n pairs), `rp_summary`, `replica_sweep`, `protection`, `bm_tolerance`, `bare_loop` and `crossing`. 4. Sets the sweep ranges and prints the testbench result panel.

````python
#@title System-level testbench
# The system-level testbench shared by studies A to F: the link of Section V with the post-layout
# circuit values, its two inputs, and the two system metrics: the decoded SNDR (accuracy) and the success of the replica
# attack (security of the original design). Only the quantity under study is changed from run to run.
t0 = time.time()
from scipy.stats import beta as beta_dist
P_sys = {**DEFAULTS, **CAL}
VU = lambda vdd: 0.25 * vdd                         # model unit of the comparator input (cs.map_to_model): x2 = 1 <-> VDD/4
u_sine = s2d(tia(sensor_current({**P_sys, "sig": "sine"}, N_DSM), P_sys["rtia"]), P_sys["buffer"])   # accuracy input
u_dc = np.full(LAB_N, s2d(tia(0.5e-6, P_sys["rtia"]), P_sys["buffer"]))                             # attack input
K16 = lfsr16_bits(DEFAULTS["key16"], ID_BITS + LAB_N)[ID_BITS:]           # the original design's keystream
REPLICA_DITHER = (0x2C4D, 0x51E7)                                         # the replica chip's own dither seeds
Qn = lambda sn_v, vdd=1.8: {**P_sys, "comp_noise": sn_v / VU(vdd)}       # settings with a comparator noise in volts

def sndr_at(P, seed=0, **kw):
    """SNDR of the decoded full-scale sine (2,048-point FFT of the receiver's output)."""
    return sndr(decode(dsm2(u_sine, P, noise_seed=seed, **kw)["D"]), P["bin"])

N_FRAMES = 4                                      # frames of 30,000 clocks per pair that the full-scan attacker records
SCAN0 = 200                                       # first window start (after the start-up, as in window_attack)
K16X = lfsr16_bits(DEFAULTS["key16"], ID_BITS + LAB_N + 600)[ID_BITS:]   # the keystream, with 576 bits past the capture
TAIL = 40                                         # a window yields the key only if its last 40 bits are clean (BM check below)

def cp95(k, n):
    """Clopper-Pearson 95 % interval of a binomial fraction k/n."""
    return (0.0 if k == 0 else float(beta_dist.ppf(0.025, k, n - k + 1)),
            1.0 if k == n else float(beta_dist.ppf(0.975, k + 1, n - k)))

def full_scan(D1, D2):
    """The full-scan attacker on one frame: every 64-bit window of the keystream guess from clock SCAN0 on. Windows whose
    last TAIL bits are clean are run through Berlekamp-Massey (the others cannot yield the key); a window yields the key
    when its register predicts the next 512 keystream bits with more than 99 % accuracy. Returns (key found, number of
    windows scanned, fraction of windows with no differing bit, f64)."""
    # e marks the clocks on which the two chips disagree; kg = C xor D_replica is the attacker's keystream guess, wrong
    # exactly where e = 1. With the prefix sum c of e, a window [s, s + 64) is clean when c[s + 64] == c[s], and its last
    # TAIL bits are clean when c[s + 64] == c[s + 64 - TAIL]; only those windows are tried
    e = (D1 ^ D2).astype(np.int64); kg = D1 ^ K16X[:len(D1)] ^ D2
    c = np.concatenate([[0], np.cumsum(e)]); s = np.arange(SCAN0, len(e) - 63)
    f64 = float(np.mean(c[s + 64] == c[s]))
    for s0 in s[c[s + 64] == c[s + 64 - TAIL]]:
        if bm_attack(kg[s0:s0 + 64], K16X[s0 + 64:s0 + 576])["acc"] > 0.99:
            return True, len(s), f64
    return False, len(s), f64

def replica_pairs(P, sos_v, vdd, n, seed, coef_sigma=0.0, k0=0, frames=N_FRAMES):
    """The replica attack on n victim/replica pairs (pair indices k0 ... k0 + n - 1). Every pair has its own seeds, drawn
    from (seed, k): the two chips' comparator offsets (N(0, sigma_os)), capacitor-ratio errors (N(0, coef_sigma)),
    dither seeds and noise seeds (a new noise realisation in every frame). Two attackers: the full-scan attacker scans
    every window of up to `frames` frames and stops at the first that yields the key (the security criterion); the
    attacker of [1] runs Berlekamp-Massey on 80 random 64-bit windows of the first frame (window_attack). Returns per-pair
    arrays: broken (full scan), frames scanned, broken80, f64 and p of the first frame."""
    out = {k: [] for k in ("broken", "frames", "broken80", "f64", "p")}
    for k in range(k0, k0 + n):
        # every random choice of pair k comes from the generator seeded with (seed, k): reproducible pair by pair
        r = np.random.default_rng((int(seed), k))
        o1, o2 = r.normal(0, sos_v / VU(vdd), 2)
        c1, c2 = (r.normal(0, coef_sigma, 2), r.normal(0, coef_sigma, 2)) if coef_sigma else ((0.0, 0.0), (0.0, 0.0))
        dv, dr = tuple(int(x) for x in r.integers(1, 1 << 16, 2)), tuple(int(x) for x in r.integers(1, 1 << 16, 2))
        nv, nr = (int(x) for x in r.integers(1, 1 << 30, 2))
        hit = False
        for f in range(frames):
            D1 = dsm2(u_dc, P, seeds=dv, noise_seed=nv + 7919 * f, offset=o1, coef_err=c1)["D"]
            D2 = dsm2(u_dc, P, seeds=dr, noise_seed=nr + 7919 * f, offset=o2, coef_err=c2)["D"]
            hit, _, f64 = full_scan(D1, D2)
            if f == 0:
                x = (D1 ^ D2)[SCAN0:].astype(np.int64)
                out["broken80"].append(window_attack(D1 ^ K16 ^ D2, K16, (int(seed) * 100003 + k) & 0xFFFFFFF)["broken"] > 0)
                out["f64"].append(f64); out["p"].append(float(x.mean()))
            if hit:
                break
        out["broken"].append(hit); out["frames"].append(f + 1)
    return {k: np.array(v) for k, v in out.items()}

def rp_merge(a, b):                               # more pairs of the same setting
    return {k: np.concatenate([a[k], b[k]]) for k in a}

def rp_summary(R):
    """Success of both attackers with Clopper-Pearson 95 % intervals, and the theory's ingredients."""
    n, k, k80 = len(R["broken"]), int(R["broken"].sum()), int(R["broken80"].sum())
    nf = int(R["frames"][~R["broken"]].sum())         # frames scanned without success (all frames of the unbroken pairs)
    return dict(n=n, k=k, success=k / n, ci=cp95(k, n), k80=k80, success80=k80 / n, ci80=cp95(k80, n),
                f64=float(R["f64"].mean()), p=float(R["p"].mean()), bound=float(np.mean(1 - (1 - R["f64"]) ** 80)),
                frames_clean=nf, frame_hi=cp95(0, nf)[1] if (k == 0 and nf) else None)

def replica_sweep(run, xs, n0=24, n_top=100):
    """run(x, n, k0) -> per-pair arrays. A coarse pass with n0 pairs at every point; then every point in the transition
    (some but not all pairs broken) and the first point above it with no pair broken get n_top pairs in total, repeated
    until the first unbroken point above the last broken one has n_top pairs."""
    R = [run(x, n0, 0) for x in xs]
    while True:
        ks = [int(r["broken"].sum()) for r in R]; ns = [len(r["broken"]) for r in R]
        brk = [i for i, k in enumerate(ks) if k > 0]
        want = {i for i, (k, m) in enumerate(zip(ks, ns)) if 0 < k < m}
        if brk and brk[-1] + 1 < len(xs):
            want.add(brk[-1] + 1)
        todo = [i for i in sorted(want) if ns[i] < n_top]
        if not todo:
            return [rp_summary(r) for r in R]
        for i in todo:
            R[i] = rp_merge(R[i], run(xs[i], n_top - ns[i], ns[i]))

def protection(xs, S):
    """The protection point of a sweep: the first swept value from which on no pair is broken by the full-scan attacker
    (with at least 100 pairs there), and the last value at which a pair is broken. None if the last point is broken."""
    brk = [i for i, s in enumerate(S) if s["k"] > 0]
    i = brk[-1] + 1 if brk else 0
    return (float(xs[i]) if i < len(xs) else None, float(xs[brk[-1]]) if brk else None, S[i] if i < len(xs) else None)

def bm_tolerance(trials=20):
    """How Berlekamp-Massey tolerates one wrong bit in a 64-bit window, by its position: the share of trials that still
    yield the key (random positions of the original design's keystream)."""
    r = np.random.default_rng(64); ok = np.zeros(64)
    for j in range(64):
        for _ in range(trials):
            s0 = int(r.integers(SCAN0, LAB_N - 64)); w = K16X[s0:s0 + 64].copy(); w[j] ^= 1
            ok[j] += bm_attack(w, K16X[s0 + 64:s0 + 576])["acc"] > 0.99
    return ok / trials
BM_TOL = bm_tolerance()

def bare_loop(u, offset=0.0, x2_0=0.0, g1=0.5, g2=0.5):
    """The modulator's loop without dither, noise or clipping, for the exact identities of the theory."""
    x1, x2, y, out = 0.0, x2_0, 1, np.zeros(len(u), np.uint8)
    for k in range(len(u)):
        v = 1.0 if y else -1.0
        x1 += g1 * (u[k] - v); x2 += g2 * (x1 - v)
        y = 1 if x2 + offset >= 0 else 0; out[k] = y
    return out

def crossing(xs, ys, level):                        # first x (log-interpolated) at which ys falls below level
    for (x0, y0), (x1, y1) in zip(zip(xs, ys), zip(xs[1:], ys[1:])):
        if y0 >= level > y1:
            return float(np.exp(np.log(x0) + (np.log(x1) - np.log(x0)) * (y0 - level) / (y0 - y1)))
    return None

sn0, os0 = COMP_POST["comp_noise"] * VU(1.8), COMP_POST["comp_offset_sigma"] * VU(1.8)   # the post-layout comparator
sn_max, os_max = max(r["sigma_n"] for r in PVT["cmp"]), max(r["sigma_os"] for r in PVT["cmp"])
SN_SWEEP = np.array([0.1, 0.3, 1, 3, 10, 20, 30, 50, 100, 300, 1000]) * 1e-3            # comparator noise sweep (V)
OS_SWEEP = np.array([1, 3, 10, 30, 100, 300, 600, 1000, 2000, 3000]) * 1e-3             # comparator offset sweep (V)
amp_in = (u_sine.max() - u_sine.min()) / 2
card("System-level testbench",
     [("circuit values (post-layout, tt, 27 °C, 1.8 V)", f"comparator σ<sub>n</sub> = {1e3 * sn0:.3f} mV, σ<sub>os</sub> = {1e3 * os0:.2f} mV; "
       f"integrator noise {P_sys['int_noise']:.2e} FS rms per clock; SRAM-PUF s<sub>25</sub> = {CAL['puf_sigma']:.4f}"),
      ("PVT grid (Section VII-F)", f"{len(PVT['cmp'])} comparator conditions; σ<sub>n</sub> up to {1e3 * sn_max:.3f} mV, "
       f"σ<sub>os</sub> up to {1e3 * os_max:.2f} mV; {len(PVT['puf']['corners'])} SRAM-PUF corners × {len(TEMPS)} temperatures"),
      ("accuracy input", f"full-scale sine through the TIA and buffer: amplitude {amp_in:.3f} of the modulator's full scale, "
       f"{N_DSM:,} clocks, OSR {OSR}"),
      ("attack input", f"constant sensor current 0.5 µA (modulator input {u_dc[0]:.3f}), {LAB_N:,} clocks per chip; "
       f"each pair with its own seeds; full-scan attacker: every 64-bit window of up to {N_FRAMES} frames; attacker of [1]: 80 "
       "random windows of one frame"),
      ("Berlekamp-Massey with one wrong bit in a 64-bit window (20 trials per position)",
       f"key still recovered: positions 1–16 in {100 * BM_TOL[:16].mean():.0f} %, position 17 in {100 * BM_TOL[16]:.0f} %, "
       f"positions 18–64 in {100 * BM_TOL[17:].mean():.0f} %; the scan therefore tests every window whose last {TAIL} bits agree"),
      ("model units", f"x<sub>2</sub> = 1 ↔ V<sub>DD</sub>/4 = {1e3 * VU(1.8):.0f} mV at the comparator input (1.8 V)")])
````

### The complete sensor, end to end

One frame of the complete proposed design is simulated through every block with the SKY130 circuit values, at the nominal condition and at a worst condition with every block at one corner and temperature. The frame follows the signal chain of Section V: the TIA and buffer, which give the modulator input $u$; the second-order loop, with the noise transfer function $|\mathrm{NTF}|^2 \propto (2\sin \pi f/f_s)^4$ at frequency $f$ and clock rate $f_s$; the key from 600 SRAM cells, BCH-corrected up to t = 45 and hashed to 128 bits (Section V-D); Grain-128a with an initialisation vector (IV) from the session nonce and the frame counter, $C = D \oplus K$, and a 32-bit tag; and the receiver, which verifies the tag and the counter, decrypts $D_{rx} = C \oplus K$ and decimates with the CIC and compensation filters to 200 kS/s. Fig. 45 follows one nominal frame along this chain, from the sensor current and $x_2$ through $D$, $K$ and $C$ to $D_\mathrm{rx}$, and compares the server output with the input as the receiver's filters delay and scale it, **so that each block can be checked against its neighbours**.

At the nominal condition the differing SRAM cells should stay well below t, so that the key is recovered and $D_{rx} = D$ bit for bit. At the worst condition (sf and 85 °C; the key generator's worst corners are ff and fs, study F) the recorded 5.0 % error rate at 85 °C (Section VII-F), about 30 of 600 cells plus the noise of the 25 °C enrolment read, comes close to t = 45: **the key should be recovered in this frame, but without the $10^{-6}$ reliability**. The comparator noise, below 1 mV, is two orders of magnitude below the level at which it matters (study A), so the SNDR should agree at both conditions to within the spread between noise realisations.

The bit-stream spectrum should have two regions: below about 10 kHz the unshaped front-end noise sets a flat floor; above it, the shaped noise rises by 40 dB per decade and dominates the band. Toward half the clock rate the shaping levels off: $|\mathrm{NTF}|^2$ peaks at $f_s/2$, where the 40 dB-per-decade asymptote overstates it by $40\log_{10}(\pi/2)$ = 7.8 dB. A 1-bit loop also forms a limit-cycle tone at $(1-|u|)\,f_s/2$; the sine of amplitude $A$ sweeps it down to $(1-A)\,f_s/2$ and dwells longest there, so the spectrum should peak close to $(1-A)\,f_s/2$, about 3.8 MHz for $A = 0.249$, **nearly 40 times the band edge**.

Since $D_{rx} = D$, the decoded output carries exactly the modulator's noise; **the third-order CIC and compensation FIR remove the noise above the band but not the shaped noise inside it**. Its floor should therefore be flat only at the lowest frequencies, rise by 40 dB per decade across the band, 14.7 dB from 30 to 70 kHz, and fall steeply in the transition band of the compensation filter, centred on 80 kHz. Fig. 46 tests these predictions on the nominal, worst and impaired frames: panel (a) the regions of the bit-stream spectrum, panel (b) the floor of the decoded output.

A third frame, of an impaired chip, adds three impairments that the reference model lacks. (i) *Clock jitter* on the non-return-to-zero feedback DAC of the continuous-time loop: an edge that moves by δt changes the fed-back charge by ($v_k - v_{k-1}$) δt/$T_\mathrm{s}$, with $T_\mathrm{s}$ the clock period; with ±1 levels and a transition at about half the clocks, an rms jitter $\sigma_\mathrm{t}$ = 10 ps at $T_\mathrm{s}$ = 100 ns gives a white error of about $2 \cdot \sqrt{0.5} \cdot 10^{-4} \approx 1.4 \times 10^{-4}$ of full scale per clock, eight times the integrator noise, so the floor should rise by about 18 dB. (ii) *1/f noise* at the integrator input, with a corner $f_\mathrm{c}$ at which it equals the white thermal density: the floor rises by $10 \log_{10}(1 + f_\mathrm{c}/f)$, about 10 dB at 5 kHz for $f_\mathrm{c}$ = 50 kHz. (iii) *Distortion* of the input stage, $u + a_2 u^2 + a_3 u^3$ with second- and third-order coefficients $a_2$ and $a_3$, gives harmonics at 40 and 60 kHz, inside the band, which the loop does not suppress. Since the shaped quantisation noise dominates the in-band noise, **the SNDR should drop mainly through the third harmonic**.

Computes the spectra of the bit streams (2 to 8 kHz floor, in-band noise from 1 kHz) and decoded outputs. 3. Reports floors, in-band noise, harmonics, the limit-cycle tone, filter gains and SNDRs. 4. Draws the spectra.

````python
#@title Fig.: the complete sensor, end to end: spectra
# The same two frames in the frequency domain, plus a third frame with circuit impairments that the reference model leaves
# out: (a) the spectrum of the modulator's bitstream, which shows the noise shaping of the second-order loop up to half
# the clock rate; (b) the spectrum of the receiver's decoded output in the signal band, from which the SNDR is computed
from scipy.signal import welch
# assumed impairments (no values in [1]): 10 ps rms jitter on the NRZ feedback-DAC edges, a 50 kHz 1/f corner of the
# input-referred noise, and an input stage with HD2 = -80 dBc and HD3 = -70 dBc at the test amplitude
IMPAIR = dict(jitter_ps=10.0, flicker_fc=50e3, hd2_dbc=-80.0, hd3_dbc=-70.0)
IMP_KW = dict(jitter=dict(jitter_ps=IMPAIR["jitter_ps"]), flicker=dict(flicker_fc=IMPAIR["flicker_fc"]),
              distortion=dict(hd_a2=2 * 10 ** (IMPAIR["hd2_dbc"] / 20) / amp_in, hd_a3=4 * 10 ** (IMPAIR["hd3_dbc"] / 20) / amp_in ** 2))
_nom = {**P_sys, **FRAME_CASES["nominal"]["over"]}
FRAMES_SPEC = {**FRAMES, "impaired": simulate(**{**_nom, **{k: v for kw in IMP_KW.values() for k, v in kw.items()}})}
FRAMES_ALONE = {k: simulate(**{**_nom, **kw}) for k, kw in IMP_KW.items()}   # each impairment on its own, for the breakdown
SNDR_SPREAD = np.array([simulate(**{**_nom, "seed": _nom["seed"] + 100 + i})["sndr"] for i in range(8)])   # noise realisations
F_SIG = DEFAULTS["bin"] * FS_OUT / N_FFT                          # the test tone (Hz)
F_BAND = FS / (2 * OSR)                                           # the signal band of the receiver (Hz)
def bh(n):                                                        # four-term Blackman-Harris window, as in sndr()
    x = 2 * np.pi * np.arange(n) / (n - 1)
    return 0.35875 - 0.48829 * np.cos(x) + 0.14128 * np.cos(2 * x) - 0.01168 * np.cos(3 * x)
NW_Y = 1024                                                       # Welch segments of the decoded output (195 Hz bins)
SPEC = {}
for k, S in {**FRAMES_SPEC, **FRAMES_ALONE}.items():
    f, p = welch(2.0 * S["D"] - 1.0, fs=FS, window=bh(16384), nperseg=16384, scaling="spectrum")   # bitstream, ±1
    y = np.asarray(S["y"][-N_FFT:], float); w = bh(N_FFT)
    py = np.abs(np.fft.rfft((y - y.mean()) * w)) ** 2 / (w.sum() / 2) ** 2                     # decoded output; full-scale sine = 0 dBFS
    fd, pd = welch(2.0 * S["D"] - 1.0, fs=FS, window=bh(16384), nperseg=16384, scaling="density")  # for noise integration
    fw, pw = welch(np.asarray(S["y"][60:], float), fs=FS_OUT, window=bh(NW_Y), nperseg=NW_Y, scaling="spectrum")  # decoded, averaged
    SPEC[k] = dict(f=f, p=p, pd=pd, fy=np.fft.rfftfreq(N_FFT, 1 / FS_OUT), py=py, fw=fw, pw=pw / 0.5)
# the in-band noise of the bitstream, integrated from 1 kHz to the band edge, excluding the tone: one number per case
def inband_noise(s):                                              # noise power / full-scale sine power, in dB
    m = (s["f"] > 1e3) & (s["f"] < F_BAND) & (np.abs(s["f"] - F_SIG) > 4 * FS / 16384)
    return 10 * np.log10(s["pd"][m].sum() * (s["f"][1] - s["f"][0]) / 0.5)
def floor_db(s, f0=2e3, f1=8e3):                                  # the low-frequency floor of the bitstream (dBFS per bin)
    m = (s["f"] >= f0) & (s["f"] <= f1)
    return 10 * np.log10(np.median(s["p"][m]) / 0.5)
# what shapes the two panels: (a) above 1 MHz, the strongest component of the bitstream and the share of its power there; the
# limit-cycle tone of a 1-bit loop lies at (1 - |u|) fs/2 for a slowly varying input u, so the sine of amplitude A sweeps it
# down to (1 - A) fs/2, where it dwells longest; (b) the receiver's filters (3rd-order CIC at 10 MHz x the 31-tap compensation
# FIR at 200 kS/s) and the rise of the decoded noise floor across the band against the 40 dB per decade of the shaped noise
from scipy.signal import freqz
def hf_peak(s):                                                   # (frequency, dBFS per bin, share of power above 1 MHz)
    m = s["f"] > 1e6; j = int(np.argmax(s["p"][m]))
    return s["f"][m][j], 10 * np.log10(s["p"][m][j] / 0.5), s["p"][m].sum() / s["p"].sum()
def rx_gain_db(f):                                                # receiver filter gain at f (Hz), relative to DC
    cic3 = np.abs(np.sin(np.pi * f * OSR / FS) / (OSR * np.sin(np.pi * f / FS))) ** 3
    return 20 * np.log10(cic3 * np.abs(freqz(COMP_TAPS, worN=[f], fs=FS_OUT)[1][0]))
def out_floor(s, f0, f1):                                         # median decoded-output floor between f0 and f1 (dBFS per
    m = (s["fw"] >= f0) & (s["fw"] <= f1)                         # 195 Hz bin of the Welch average)
    return 10 * np.log10(np.median(s["pw"][m]))
HF = hf_peak(SPEC["nominal"])
def sndr_to(S, f_hi):                                             # SNDR as sndr() computes it, with the noise integrated to f_hi
    y = np.asarray(S["y"][-N_FFT:], float); w = bh(N_FFT); p = np.abs(np.fft.fft((y - y.mean()) * w)) ** 2
    k = np.arange(4, int(round(f_hi / (FS_OUT / N_FFT))) + 1); k = k[k < N_FFT // 2]; b0 = DEFAULTS["bin"]
    sig = p[k[np.abs(k - b0) <= 4]].sum()
    return 10 * np.log10(sig / (p[k].sum() - sig))
def hd_dbc(s, h):                                                 # harmonic h of the decoded output relative to the tone (dBc)
    b0 = DEFAULTS["bin"]; pw = lambda b: s["py"][b - 3:b + 4].sum()
    return 10 * np.log10(pw(h * b0) / pw(b0))
card("The complete sensor, end to end: spectra",
     [("test tone", f"{F_SIG / 1e3:.2f} kHz, amplitude {amp_in:.3f} of full scale ({20 * np.log10(amp_in):.1f} dBFS); signal band 0–{F_BAND / 1e3:.0f} kHz "
       f"(clock {FS / 1e6:.0f} MHz, OSR {OSR})"),
      ("noise in the reference model (nominal, worst)", "integrator thermal noise (white, 1.8 × 10<sup>−5</sup> FS rms per clock, from [1]); "
       "comparator noise and offset (SKY130); dither; no distortion, 1/f noise, clock jitter, excess loop delay or supply noise"),
      ("assumed impairments (impaired)", f"DAC clock jitter {IMPAIR['jitter_ps']:.0f} ps rms; 1/f corner {IMPAIR['flicker_fc'] / 1e3:.0f} kHz; "
       f"input stage HD2 {IMPAIR['hd2_dbc']:.0f} dBc, HD3 {IMPAIR['hd3_dbc']:.0f} dBc; otherwise the nominal condition"),
      ("low-frequency floor of the bitstream (2–8 kHz)", ", ".join(f"{k}: {floor_db(SPEC[k]):.1f} dBFS/bin" for k in FRAMES_SPEC)),
      ("in-band quantisation and circuit noise of the bitstream (1 kHz to the band edge)",
       ", ".join(f"{k}: {inband_noise(SPEC[k]):.1f} dBFS" for k in FRAMES_SPEC)),
      ("HD2, HD3 of the decoded output", ", ".join(f"{k}: {hd_dbc(SPEC[k], 2):.1f}, {hd_dbc(SPEC[k], 3):.1f} dBc" for k in FRAMES_SPEC)),
      ("each impairment alone (floor 2–8 kHz; HD3; SNDR)", "; ".join(
          f"{k}: {floor_db(SPEC[k]):.1f} dBFS/bin, {hd_dbc(SPEC[k], 3):.1f} dBc, {S['sndr']:.1f} dB" for k, S in FRAMES_ALONE.items())),
      ("bitstream above 1 MHz (nominal): strongest component; share of the bitstream power",
       f"{HF[0] / 1e6:.2f} MHz at {HF[1]:.1f} dBFS/bin (the 1-bit loop's limit-cycle tone, swept by the sine down to (1 − A)·f<sub>s</sub>/2 = "
       f"{(1 - amp_in) * FS / 2e6:.2f} MHz); {100 * HF[2]:.0f} % of the power lies above 1 MHz"),
      ("receiver filter (CIC × compensation) gain at 20, 80, 90, 100 kHz",
       ", ".join(f"{rx_gain_db(f):.1f}" for f in (20e3, 80e3, 90e3, 100e3)) + " dB (the transition band of the 31-tap compensation FIR is centred on 0.4 f<sub>out</sub> = 80 kHz)"),
      (f"decoded-output floor, 25–35 kHz and 65–75 kHz (nominal; Welch average, {NW_Y}-point segments, {FS_OUT / NW_Y:.0f} Hz bins)",
       f"{out_floor(SPEC['nominal'], 25e3, 35e3):.1f} and {out_floor(SPEC['nominal'], 65e3, 75e3):.1f} dBFS/bin: a rise of "
       f"{out_floor(SPEC['nominal'], 65e3, 75e3) - out_floor(SPEC['nominal'], 25e3, 35e3):.1f} dB, against "
       f"{40 * np.log10(70 / 30):.1f} dB for noise rising at 40 dB per decade"),
      ("SNDR spread of single nominal frames (8 noise realisations)", f"mean {SNDR_SPREAD.mean():.1f} dB, σ {SNDR_SPREAD.std():.1f} dB, "
       f"range {SNDR_SPREAD.min():.1f} to {SNDR_SPREAD.max():.1f} dB"),
      ("SNDR bandwidth", f"sndr() integrates the noise from bin 4 ({4 * FS_OUT / N_FFT:.0f} Hz) to {FS_OUT / 2e3:.0f} kHz, the Nyquist frequency "
       f"of the decoded output; integrated only to the 80 kHz edge of the compensation filter instead: nominal "
       f"{sndr_to(FRAMES_SPEC['nominal'], 80e3):.1f} dB (to {FS_OUT / 2e3:.0f} kHz: {sndr_to(FRAMES_SPEC['nominal'], FS_OUT / 2):.1f} dB)"),
      ("SNDR of the decoded output (2,048-point Blackman-Harris FFT, noise to 100 kHz)", ", ".join(f"{k}: <b>{S['sndr']:.1f} dB</b>" for k, S in FRAMES_SPEC.items()))])
fig, axs = plt.subplots(1, 2, figsize=(W2, 3.1))
# as Fig. 2(b), (c) of the ISCAS paper: the nominal frame dashed and on top, so the worst frame under it stays visible;
# small triangles mark the labelled peaks (the signal in both panels, HD3 of the impaired frame in the decoded output)
sty = {"impaired": dict(color=COL["aqua"], lw=0.8, zorder=2), "nominal": dict(color=COL["ok"], lw=1.0, ls=(0, (2.5, 1.2)), zorder=5),
       "worst": dict(color=COL["bad"], lw=0.8, zorder=3)}
dB_ = lambda v: f"{v:.0f} dB".replace("-", "\u2212")
def peak_(fx, px, f0, w):                                          # level of the largest bin within w of f0
    m = np.abs(fx - f0) <= w; return float(px[m].max())
mark_ = lambda ax, x, y, up: ax.plot([x], [y + up], "v", ms=4.5, color=COL["ink"], mew=0, zorder=8, clip_on=False)
ax = axs[0]
for k in ("impaired", "nominal", "worst"):
    s = SPEC[k]
    ax.semilogx(s["f"][1:], 10 * np.log10(s["p"][1:] / 0.5 + 1e-30), label=f"{k} ({FRAMES_SPEC[k]['sndr']:.1f} dB)", **sty[k])
fr = np.logspace(np.log10(2e4), np.log10(1e6), 50)
ref = 10 * np.log10(SPEC["nominal"]["p"][np.searchsorted(SPEC["nominal"]["f"], 1e6)] / 0.5)
ax.semilogx(fr, ref + 40 * np.log10(fr / 1e6), "--", color=COL["ink"], lw=0.9, label="40 dB/decade")
fl = floor_db(SPEC["nominal"])
ax.hlines(fl, 1.2e3, 1.5e4, color=COL["ink"], lw=0.9, ls=":")
ax.text(1.3e3, fl - 4, "white floor: integrator\nthermal noise", fontsize=7.5, color=COL["ink"], va="top")
ax.annotate("floor raised\nby DAC\nclock jitter", (6e3, floor_db(SPEC["impaired"])), xytext=(2.7e3, -80), ha="left", va="center",
            fontsize=7.5, color=COL["aqua"], arrowprops=dict(arrowstyle="-", lw=0.6, color=COL["aqua"], shrinkB=1))
_fq = 4e5; _pq = 10 * np.log10(np.median(SPEC["nominal"]["p"][np.abs(SPEC["nominal"]["f"] - _fq) < 2e4]) / 0.5)
ax.annotate("shaped\nquantisation and\ncomparator noise", (_fq, _pq + 4), xytext=(1.35e5, -20), ha="left", va="center",
            fontsize=7.5, color=COL["ink"], arrowprops=dict(arrowstyle="-", lw=0.6, color=COL["ink"], shrinkB=1))
ax.annotate("limit-cycle tone,\n$(1-A)\\,f_s/2$", (HF[0], HF[1]), xytext=(2.2e6, -72), ha="center", va="center",
            fontsize=7.5, color=COL["ink"], arrowprops=dict(arrowstyle="->", lw=0.7, color=COL["ink"]))
ax.axvline(F_BAND, color=COL["faint"], ls=":", lw=1)
ax.text(F_BAND * 0.9, 0.97, "band edge", fontsize=7, color=COL["faint"], rotation=90, va="top", ha="right", transform=ax.get_xaxis_transform())
ax.set_ylim(-152, 0); ax.set_xlim(FS / 16384, FS / 2); ax.set_xlabel("Frequency (Hz)"); ax.set_ylabel("Power (dBFS per bin)")
_sa = peak_(SPEC["nominal"]["f"], 10 * np.log10(SPEC["nominal"]["p"] / 0.5 + 1e-30), F_SIG, 2e3)
mark_(ax, F_SIG, _sa, 9)
ax.text(F_SIG / 1.35, _sa - 2, f"{F_SIG / 1e3:.0f} kHz\n{dB_(_sa)}", fontsize=7, ha="right", va="top", linespacing=0.95)
ax.legend(fontsize=7, loc="lower right"); panel(ax, "a", "modulator bitstream")
ax = axs[1]
for k in ("impaired", "nominal", "worst"):
    s = SPEC[k]
    ax.plot(s["fw"] / 1e3, 10 * np.log10(s["pw"] + 1e-30), **{**sty[k], "lw": 1.2}, label=k)
si = SPEC["impaired"]; _ph = peak_(si["fw"], 10 * np.log10(si["pw"] + 1e-30), 3 * F_SIG, 1e3)   # HD3 of the impaired frame
mark_(ax, 3 * F_SIG / 1e3, _ph, 7)
ax.text(3 * F_SIG / 1e3 - 2.5, _ph + 6, f"HD3, {3 * F_SIG / 1e3:.0f} kHz\n{dB_(_ph)} (impaired)", fontsize=7, ha="right", va="bottom",
        color=COL["aqua"], linespacing=0.95)
_sb = peak_(SPEC["nominal"]["fw"], 10 * np.log10(SPEC["nominal"]["pw"] + 1e-30), F_SIG, 2e3)
mark_(ax, F_SIG / 1e3, _sb, 7)
ax.text(F_SIG / 1e3 + 2.5, _sb - 3, f"{F_SIG / 1e3:.0f} kHz\n{dB_(_sb)}", fontsize=7, ha="left", va="top", linespacing=0.95)
_j = int(np.searchsorted(SPEC["nominal"]["fw"], 88e3))
ax.annotate("receiver filter\nroll-off, 80–100 kHz", (88, 10 * np.log10(SPEC["nominal"]["pw"][_j])), xytext=(80, -25), ha="center",
            va="center", fontsize=7.5, color=COL["ink"], arrowprops=dict(arrowstyle="-", lw=0.6, color=COL["ink"], shrinkB=1))
ax.set_xlim(0, F_BAND / 1e3); ax.set_xlabel("Frequency (kHz)"); ax.set_ylabel("Power (dBFS per bin)")
ax.legend(fontsize=7, loc="upper left", bbox_to_anchor=(0.32, 0.88)); panel(ax, "b", "decoded output, signal band")   # the empty band
fig.tight_layout(); plt.show()
# interactive companion: the same spectra (bitstream decimated on a log-spaced grid that keeps the largest bins; the decoded
# output in full); one legend entry per frame toggles it in both panels; hover gives the frequency, the level and the frame
def _ilogpick(f, p, n=2000, n_peak=40):                           # log-spaced bin indices plus the n_peak largest bins
    i = np.unique(np.round(np.geomspace(1, len(f) - 1, n)).astype(int))
    return np.unique(np.concatenate([i, 1 + np.argsort(p[1:])[-n_peak:]]))
def _ispec():
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    f = make_subplots(rows=1, cols=2, horizontal_spacing=0.09,
                      subplot_titles=("(a) Modulator bitstream", "(b) Decoded output, signal band"))
    _icond = {k: FRAME_CASES[k]["label"] for k in FRAME_CASES if k in SPEC}
    _icond["impaired"] = (f"impaired: nominal with {IMPAIR['jitter_ps']:.0f} ps DAC clock jitter, {IMPAIR['flicker_fc'] / 1e3:.0f} kHz "
                          f"1/f corner, HD2 {IMPAIR['hd2_dbc']:.0f} dBc, HD3 {IMPAIR['hd3_dbc']:.0f} dBc").replace("-", "−")
    _ist = {"impaired": dict(color=COL["aqua"], width=0.8), "nominal": dict(color=COL["ok"], width=1.0, dash="dash"),
            "worst": dict(color=COL["bad"], width=0.8)}
    for rank, k in ((1, "impaired"), (3, "worst"), (2, "nominal")):  # nominal drawn last (on top), as in the static figure
        s = SPEC[k]; i = _ilogpick(s["f"], s["p"]); nm = f"{k} ({FRAMES_SPEC[k]['sndr']:.1f} dB)"
        f.add_trace(go.Scatter(x=s["f"][i], y=10 * np.log10(s["p"][i] / 0.5 + 1e-30), mode="lines", line=_ist[k], name=nm,
                                 legendgroup=k, legendrank=rank,
                                 hovertemplate="%{x:,.0f} Hz<br>%{y:.1f} dBFS per bin<extra>" + _icond.get(k, k) + "</extra>"), row=1, col=1)
        f.add_trace(go.Scatter(x=s["fw"] / 1e3, y=10 * np.log10(s["pw"] + 1e-30), mode="lines", line={**_ist[k], "width": 1.2},
                               name=nm, legendgroup=k, showlegend=False,
                               hovertemplate="%{x:.2f} kHz<br>%{y:.1f} dBFS per bin<extra>" + _icond.get(k, k) + "</extra>"), row=1, col=2)
    fr = np.logspace(np.log10(2e4), np.log10(1e6), 50)
    ref = 10 * np.log10(SPEC["nominal"]["p"][np.searchsorted(SPEC["nominal"]["f"], 1e6)] / 0.5)
    f.add_trace(go.Scatter(x=fr, y=ref + 40 * np.log10(fr / 1e6), mode="lines", line=dict(color=COL["ink"], width=0.9, dash="dash"),
                           name="40 dB/decade", legendrank=4, hoverinfo="skip"), row=1, col=1)
    f.add_trace(go.Scatter(x=[1.2e3, 1.5e4], y=[fl, fl], mode="lines", line=dict(color=COL["ink"], width=0.9, dash="dot"),
                           showlegend=False, hoverinfo="skip"), row=1, col=1)
    f.add_vline(x=F_BAND, line=dict(color=COL["faint"], width=1, dash="dot"), row=1, col=1)
    L, ink = np.log10, COL["ink"]
    note = lambda x, y, text, tc=ink, **kw: f.add_annotation(x=x, y=y, text=text, showarrow=False, font=dict(color=tc), **kw)
    arrow = lambda x, y, tx, ty, text, tc=ink, head=0, **kw: f.add_annotation(
        x=x, y=y, ax=tx, ay=ty, text=text, showarrow=True, arrowhead=head, arrowwidth=1.2, arrowcolor=tc, font=dict(color=tc), **kw)
    # (a): the static figure's labels, at its positions (log-frequency positions as log10, as plotly takes them)
    note(L(1.3e3), fl - 4, "white floor: integrator<br>thermal noise", xanchor="left", yanchor="top", align="left", row=1, col=1)
    arrow(L(6e3), floor_db(SPEC["impaired"]), L(2.7e3), -80, "floor raised<br>by DAC<br>clock jitter", COL["aqua"], axref="x", ayref="y",
          xanchor="left", align="left", row=1, col=1)
    arrow(L(_fq), _pq + 4, L(1.35e5), -20, "shaped<br>quantisation and<br>comparator noise", axref="x", ayref="y", xanchor="left",
          align="left", row=1, col=1)
    arrow(L(HF[0]), HF[1], L(2.2e6), -72, "limit-cycle tone,<br>(1 − <i>A</i>) <i>f</i><sub>s</sub>/2", head=2, axref="x", ayref="y",
          row=1, col=1)
    note(L(F_BAND * 0.9), -3, "band edge", COL["faint"], textangle=-90, xanchor="right", yanchor="top", row=1, col=1)
    f.add_trace(go.Scatter(x=[F_SIG], y=[_sa + 9], mode="markers", marker=dict(symbol="triangle-down", size=4.5, color=ink),
                           showlegend=False, hoverinfo="skip"), row=1, col=1)
    note(L(F_SIG / 1.35), _sa - 2, f"{F_SIG / 1e3:.0f} kHz<br>{dB_(_sa)}", xanchor="right", yanchor="top", align="right", row=1, col=1)
    # (b): the marked signal and HD3, and the receiver filter's roll-off
    f.add_trace(go.Scatter(x=[3 * F_SIG / 1e3, F_SIG / 1e3], y=[_ph + 7, _sb + 7], mode="markers", showlegend=False, hoverinfo="skip",
                           marker=dict(symbol="triangle-down", size=4.5, color=ink)), row=1, col=2)
    note(3 * F_SIG / 1e3 - 2.5, _ph + 6, f"HD3, {3 * F_SIG / 1e3:.0f} kHz<br>{dB_(_ph)} (impaired)", COL["aqua"], xanchor="right",
         yanchor="bottom", align="right", row=1, col=2)
    note(F_SIG / 1e3 + 2.5, _sb - 3, f"{F_SIG / 1e3:.0f} kHz<br>{dB_(_sb)}", xanchor="left", yanchor="top", align="left", row=1, col=2)
    arrow(88, 10 * np.log10(SPEC["nominal"]["pw"][_j]), 80, -25, "receiver filter<br>roll-off, 80–100 kHz", axref="x2", ayref="y2",
          row=1, col=2)
    f.update_xaxes(type="log", range=[np.log10(FS / 16384), np.log10(FS / 2)], title_text="Frequency (Hz)", row=1, col=1)
    f.update_yaxes(range=[-152, 0], title_text="Power (dBFS per bin)", row=1, col=1)
    f.update_xaxes(range=[0, F_BAND / 1e3], title_text="Frequency (kHz)", row=1, col=2)
    f.update_yaxes(title_text="Power (dBFS per bin)", row=1, col=2)
    f.update_layout(legend=dict(x=f.layout.xaxis.domain[1] - 0.005, y=0.03, xanchor="right", yanchor="bottom"))
    return f
iplot(_ispec, "Fig.: spectra of the bitstream and of the decoded output, three frames (interactive)")
````

<p align="center" style="text-align:center"><b>Fig. 46.</b> Spectra of the complete sensor (0 dBFS: full-scale sine). Blue: nominal (tt; SRAM array at 37 °C); orange: every block at sf (the corner of the noisiest comparator condition) and 85 °C; green: nominal with assumed impairments (10 ps rms DAC clock jitter, 50 kHz 1/f corner, HD2 −80 dBc, HD3 −70 dBc). **Key insights.**

* *The link, end to end.* At the nominal condition, 13 of the 600 SRAM cells differ (expected 11.2); every check passes and 0 of 107,400 decrypted bits differ from the modulator bits. At the worst condition (sf, 85 °C; comparator $\sigma_\mathrm{n}$ = 0.732 mV and $\sigma_\mathrm{os}$ = 4.85 mV, interpolated), 40 cells differ (expected 40.1), close to t = 45 as predicted. The key is recovered and every check passes in this frame, but the sf cell model gives a probability of $1.3 \times 10^{-1}$ that a power-up at 85 °C yields a wrong key, far from the $10^{-6}$ target. In both frames the fraction of ones is 0.500, no run exceeds 3 clocks, and the integrators never clip; in the nominal frame (Fig. 45) the server's reading follows the input to within 0.018 % of full scale rms. **In the behavioural model, the complete link operates end to end at both conditions.** The key generator tolerates the nominal read with a margin of 32 cells; at sf and 85 °C the margin shrinks to 5 cells and the key fails in about one power-up in eight, which is why study F sizes it per corner.
* *Spectra.* In Fig. 46, the nominal and worst spectra coincide; only the impaired frame departs from them. The bit-stream spectrum (Fig. 46(a)) has the two predicted regions: between 2 and 8 kHz the lumped front-end noise sets a flat floor at −126.8 dBFS per bin at both conditions (below 2.4 kHz the window's main lobe around DC dominates); above it, the spectrum rises at 40 dB per decade. The in-band noise is −66.5 dBFS at both, so the shaped quantisation noise, not the circuit noise, sets the SNDR. Above about 1 MHz the spectrum levels off near −35 dBFS per bin; its strongest component, at 3.82 MHz (−11.1 dBFS per bin), lies next to the predicted 3.75 MHz, and 96 % of the bit-stream power lies above 1 MHz. The decoded output (Fig. 46(b)) shows the 20.02 kHz tone at −12.1 dBFS. Its floor rises from −107.6 (25–35 kHz) to −90.1 dBFS per bin (65–75 kHz), by 17.5 dB against the predicted 14.7 dB (folded noise and the compensation filter's passband are possible contributors to the 2.8 dB excess), and falls above 80 kHz, where the filters attenuate by 6.7 dB at 80 kHz, 42.8 dB at 90 kHz and 59.5 dB at 100 kHz.
* *Accuracy.* The SNDR is 60.0 dB at the nominal and 60.1 dB at the worst condition: as predicted, the accuracy does not change between them. Integrated only to the 80 kHz edge of the compensation filter, the nominal SNDR is 60.3 dB, so the attenuated noise between 80 and 100 kHz costs 0.3 dB. Even without circuit distortion the decoded output has a third harmonic at −76.3 dBc (nominal), from the input-correlated quantisation error that the dither does not fully remove.
* *An impaired chip.* With the assumed impairments, the low-frequency floor rises to −107.4 dBFS per bin, 19.4 dB above the nominal. The jitter alone raises it to −108.2 dBFS per bin (+18.6 dB, against about 18 dB predicted), and the 1/f noise alone to −117.1 dBFS per bin (+9.7 dB, against about 10 dB): the clock jitter sets the impaired frame's floor and masks the 1/f noise. The distortion appears as a third harmonic at −66.1 dBc in the impaired frame and −65.3 dBc alone, above the applied −70 dBc because it adds to the loop's own harmonic; the second harmonic, −79.4 dBc, stays close to the applied −80 dBc. The in-band noise is unchanged (−66.8 dBFS against −66.5 dBFS); the SNDR of the impaired frame, 59.5 dB, lies about 1 dB below the mean of single nominal frames (60.6 dB, σ 0.3 dB), consistent with the third harmonic. Over eight realisations, single nominal frames range from 60.2 to 61.1 dB, so the jitter-only (60.9 dB) and 1/f-only (60.3 dB) frames do not differ measurably from the nominal; the distortion-only frame (59.4 dB) is below the range. The 60.0 dB of the nominal frame and the 60.3–60.4 dB of study A (other seeds, no offset) differ by about this spread, not by a circuit effect. **The 60 dB of the reference model is optimistic for a chip**: its clean low-frequency floor, set by the model's only unshaped noise source, is an idealisation; the assumed impairments cost about 0.5–1 dB at the nominal condition, the studies below omit them, and the clock jitter and input-stage linearity need a specification in the chip design.

### A. Accuracy: comparator noise and offset

In model units, the second-order modulator of Section V-C is

$$x_1[k] = x_1[k-1] + g_1\big(u[k] - v[k-1]\big),\qquad x_2[k] = x_2[k-1] + g_2\big(x_1[k] - v[k-1]\big),\qquad v[k] = \operatorname{sgn}\big(x_2[k] + o + n[k]\big),$$

with the input $u[k]$ (full scale ±1), the integrator outputs $x_1$, $x_2$, the fed-back decision $v[k] = \pm 1$, the integrator coefficients $g_1 = g_2 = g = 1/2$, and the comparator's offset $o$ and noise $n[k]$ (dither omitted for clarity).

* *Noise is shaped like quantisation error, until it lowers the quantiser's gain.* If the input $q = x_2 + n$ of the 1-bit quantiser is Gaussian with standard deviation $\sigma_q$, the quantiser behaves on average like a gain plus an uncorrelated error (its stochastic describing function),
  $$v = k\,q + e,\qquad k = \sqrt{2/\pi}\,/\,\sigma_q,\qquad \sigma_e^2 = 1 - 2/\pi,\qquad \sigma_q^2 = \sigma_{x_2}^2 + \sigma_n^2 ,$$
  with $\sigma_{x_2}$ the rms of $x_2$, so the loop gives $V(z) = \mathrm{STF}(z)\,U(z) + \mathrm{NTF}(z)\,\big(E(z) + k\,N(z)\big)$ ($V$, $U$, $E$, $N$: the $z$-transforms of $v$, $u$, $e$, $n$; STF: the signal transfer function), with $\mathrm{NTF}(z) = (1 - z^{-1})^2 / D(z)$ and $D(z) = (1 - z^{-1})^2 + k g z^{-1}\,(1 + g - z^{-1})$: the comparator noise is shaped by the same NTF as the quantisation error, and only weakens the shaping by widening $\sigma_q$ and so lowering $k$. The in-band noise power is
  $$P_{\text{in}} = \big(\sigma_e^2 + k^2\sigma_n^2\big)\,\frac{1}{\pi}\int_{\text{band}} \big|\mathrm{NTF}(e^{j\omega})\big|^2\,d\omega ,$$
  with $\omega$ the angular frequency in radians per clock and $k$ solved self-consistently from $g$, the input amplitude and the band alone (no simulated or fitted value); the argument holds while the loop's own signal dominates $\sigma_q$. The analysis above therefore predicts that the SNDR does not change until $\sigma_\mathrm{n}$ reaches a sizeable fraction of the swing of $x_2$, so that its −3 dB point lies near 100 mV, two orders of magnitude above the PVT values.
* *An offset is a change of initial state.* Substituting $\tilde{x}_2 = x_2 + o$ turns the loop with offset $o$ into the offset-free loop started from $\tilde{x}_2[0] = o$, so without dither, noise and clipping the two produce **identical bits**, and because the bit density of a stable loop converges to its input whatever its initial state, the offset causes **no DC error**. The identity holds while the integrator stays inside its range, $|o| < x_{\lim} - \max|\tilde{x}_2|$ with $x_{\lim} = 2.5$ and $\max|\tilde{x}_2|$ the peak swing of the offset-free loop (a conservative limit: peak swing in either direction, any contact with the rail counted as failure); beyond it the second integrator clips and the reading is lost. The analysis above predicts identical bits and a DC error at the noise floor up to the integrator range.

Fig. 47 **sets both predictions against the simulation**: panel (a) the SNDR against $\sigma_\mathrm{n}$ with the theory curve and the 45 PVT conditions, panel (b) the DC error of the reading against the offset, below and beyond the theory's integrator range.

The study runs the behavioural model with the testbench settings:

* *Noise at PVT:* each condition's $\sigma_\mathrm{n}$ in model units at its own supply; no offset; one noise seed.
* *Noise sweep:* $\sigma_\mathrm{n}$ from 0.1 mV to 1 V (11 points) at tt, 27 °C, 1.8 V, three noise seeds per point. The theory curve uses 120 points over the same range.
* *Offset identity:* the bare loop on the first 20,000 clocks of the sine, for o = 0.01, 0.3 and −1 model units.
* *DC error:* the constant attack input for 26,850 clocks, with offsets of 1 mV to 3 V (15 points, both signs) and of 3× the largest PVT $\sigma_\mathrm{os}$; the error is the larger change of the decoded reading in ppm of full scale (drawn at $10^{-3}$ ppm when smaller), lost above 0.1 %.

* *Noise.* The SNDR is 60.3–60.4 dB at all 45 PVT conditions (Fig. 47(a)). It begins to fall only above about 20 mV, some 25× the worst PVT value, and falls by 3 dB at $\sigma_\mathrm{n}$ ≈ 72 mV, 88× the largest PVT value: the noise is shaped like quantisation error.
* *Theory against simulation.* The describing function predicts the −3 dB point at about 99 mV, 1.4× the simulated 72 mV. * *Offset.* The bare loop with an offset and the offset-free loop started from the shifted state give identical bits for all three offsets tested (`True`). The offset is thus absorbed up to an integrator range more than 100× the largest PVT $\sigma_\mathrm{os}$.
* *Design consequence.* **For accuracy, the comparator can be sized for energy, not for noise or offset.** Study B asks whether the same two quantities can make two chips different enough to protect the original design.

### B. Security of the original design: comparator noise and offset

The original design's only defence against an attacker who owns a second chip (Section IV) is the analog difference between the chips. With $\hat{K} = K \oplus (D \oplus \hat{D})$ (simulation setup), any 64-bit window whose last 48 bits agree between victim and replica yields the LFSR, and the whole keystream follows; **a clean window always qualifies**.

* *The full-scan attacker* **succeeds as soon as one such stretch occurs anywhere in its capture**.
* *The 80-window attacker of [1]* examines $W = 80$ random windows of one frame. If a fraction $f_{64}$ of all 64-bit windows is clean and the windows are drawn independently, it succeeds on a chip pair with probability
$$P_{\text{succ}} \;\ge\; 1 - \left(1 - f_{64}\right)^{W},$$
a lower bound, because a window whose only errors lie in its first 16 bits also yields the key.

Two properties determine $f_{64}$:

* *Disagreements come in bursts.* Two loops driven by the same input follow nearly the same trajectory, differ only where $x_2$ passes close to the threshold, and resynchronise, because the loop bounds the state difference. If disagreements were independent with rate $p$, then $f_{64} = (1-p)^{64}$, negligible at $p$ of a few tens of per cent. **Because they cluster, long clean stretches remain, and the attack needs only one**.
* *An offset does not change the dynamics* (study A), so it cannot decorrelate two chips until it pushes the loop into clipping beyond the integrator range (700 mV). Noise, in contrast, randomises every decision with $|x_2 + o| \lesssim \sigma_n$: as $\sigma_\mathrm{n}$ grows it eliminates the clean windows, and $f_{64}$ and $P_{\text{succ}}$ fall steeply.

The theory makes three predictions. (i) **Two chips that differ only in their comparator offsets agree on most bits, and the attack always succeeds**. (ii) At the PVT noise, the 80-window attack succeeds whenever $f_{64}$ is a few per cent: $f_{64} = 5\,\%$ already gives $P_{\text{succ}} \ge 98\,\%$; it falls below 10 % of pairs only for $f_{64}$ below about 0.13 %. The full-scan attacker needs one qualifying stretch in about 119,000 windows (four frames), so protection against it needs $f_{64}$ well below $10^{-5}$, far more noise. (iii) An offset protects only through clipping: the fraction of chips with $|o| > 700$ mV, $2\,[1 - \Phi(700\,\text{mV}/\sigma_{os})]$ with Φ the standard normal distribution function, is 2 % at $\sigma_\mathrm{os}$ = 300 mV and 24 % at 600 mV. The success rate should begin to fall slightly below the integrator range and, since clipped loops share long runs of equal bits, not reach zero.

#### Part 1: the comparator offset alone

This part tests prediction (i) with the behavioural model. Victim and replica differ only in their comparators: independent offsets from the schematic SKY130 distribution (Section VII-B) and own noise sequences; the front-end noise lumped at the first integrator is zero. Both chips receive the testbench's attack input. The run simulates 24 chip pairs without dither and 24 with dither (the replica with its own dither seeds), from a fixed seed, and measures the fraction of bits that differ between the chips and the number of pairs the replica attack compromises. Fig. 48 tests prediction (i) by the distribution of that fraction over the pairs, without and with dither: **a small fraction leaves the long agreeing stretches that the attack needs**.

> **Next cell.** 1. Draws 24 offset pairs. 2. Runs both modulators without and with dither. 3. Counts differing bits and runs the 80-window attacker (`window_attack`; where it breaks every pair, the full scan does too). 4. Prints the result panel.

````python
#@title Fig.: study B, part 1: the comparator offset alone
# Part 1 of the offset study: can the comparator offset alone make a replica chip useless to the attacker? Victim and
# replica see the same input and differ only in their comparator offsets, drawn from the SKY130 Monte Carlo spread; the
# attacker XORs the ciphertext with the replica's bits (an estimate of the keystream) and runs the windowed
# Berlekamp-Massey attack on it.
# Fixed seed; a constant 0.5 µA input through the TIA and the S2D buffer, held for LAB_N clocks:
r = rng_for(4242); u = np.full(LAB_N, s2d(tia(0.5e-6, DEFAULTS["rtia"]), DEFAULTS["buffer"]))
# model settings: the SKY130 comparator noise, no integrator noise (so that only the offsets distinguish the chips)
P0 = {**DEFAULTS, "comp_noise": COMP["comp_noise"], "int_noise": 0.0}
# the original design's keystream (16-bit LFSR); its first ID_BITS bits encrypt the chip ID, the rest the payload
K = lfsr16_bits(DEFAULTS["key16"], ID_BITS + LAB_N)[ID_BITS:]
res = {}
# without and with the dither of [1]: 24 victim/replica pairs, each with its own two offsets
for dith in (False, True):
    flips, broken = [], []
    for k in range(24):
        o1, o2 = r.normal(0, COMP["comp_offset_sigma"], 2)
        P = {**P0, "dither": dith}
        # victim (noise seed 1, offset o1) and replica (other dither seeds, noise seed 2, offset o2): two different chips
        D1 = dsm2(u, P, noise_seed=1, offset=o1)["D"]; D2 = dsm2(u, P, seeds=(0x2C4D, 0x51E7), noise_seed=2, offset=o2)["D"]
        flips.append(np.mean(D1 != D2))                  # fraction of clocks on which the two chips decide differently
        broken.append(window_attack(D1 ^ K ^ D2, K, 100 + k)["broken"] > 0)      # Berlekamp-Massey on C xor replica
    res[dith] = (np.array(flips), np.array(broken))
# card: median disagreement and broken pairs, with and without dither; then the histogram of the disagreement
card("Victim and replica differ only in their SKY130 comparator offsets",
     [(f"{'with' if d else 'without'} dither", f"median {np.median(res[d][0]) * 100:.1f} % of bits differ; Berlekamp-Massey breaks the original design "
       f"in <b>{res[d][1].sum()} of {len(res[d][1])}</b> chip pairs") for d in (False, True)],
     note="The loop absorbs the offset as a shift of its state (study A), so it gives no protection.", tone="bad")
fig, ax = plt.subplots(figsize=(W1 + 1.0, 2.8))
ax.hist([res[False][0] * 100, res[True][0] * 100], bins=np.linspace(0, 12, 25), color=["white", COL["ok"]], ec=COL["ink"], lw=0.6,
        hatch="", label=["no dither", "dither"])
ax.set_xlabel("Bits that differ, victim vs replica (%)"); ax.set_ylabel("Chip pairs"); ax.grid(axis="x", visible=False)
ax.set_xlim(0, 12); ax.legend(loc="upper right"); fig.tight_layout(); plt.show()
OFFREP = res                                     # part 1, also marked in panel (b) of the figure of part 2
````

Panel (c): one pair at both noises, 600 clocks around the longest clean stretch of the post-layout capture, with each capture's disagreement rate and clean-window fraction over its whole length.

<p align="center" style="text-align:center"><b>Fig. 49.</b> Study B, part 2: security of the original design (behavioural loop, post-layout comparator). Pairs broken by the replica attack against (a) the comparator noise $\sigma_\mathrm{n}$ and (b) its offset $\sigma_\mathrm{os}$; orange: S-DSM, full scan (95 % intervals); square: proposal, by construction; markers: PVT conditions, as in Fig. 47; grey: PVT range; dotted vertical: protection point (a), integrator range (b). (c) One pair's disagreements (ticks) and clean 64-bit windows (shading), at the post-layout and the protecting noise (top, bottom).</p>

**Key insights.**

* *PVT.* The full-scan attacker breaks 360 of the 360 simulated chip pairs at the 45 PVT conditions (95 % interval 99–100 %; Fig. 49(a), (b), markers), the 80-window attacker of [1] 356. The two chips disagree on 25 % of their bits, yet 4.9 % of all 64-bit windows remain clean (mean over the PVT pairs): the disagreements come in bursts, separated by clean stretches such as the one in Fig. 49(c). Independent errors at that rate would leave $(1 - p)^{64} = 1 \times 10^{-8}$; at this $f_{64}$ the bound gives close to 100 % even for 80 windows (prediction ii). The chips disagree far more than in part 1, mainly because of the lumped front-end noise absent there (and the post-layout comparator values); no run isolates its effect on p, and study D examines its effect on the attack.
* *Protective noise level.* The full-scan attacker breaks all 24 pairs up to $\sigma_\mathrm{n}$ = 50 mV, 87 of 100 at 70 mV, 14 of 100 at 100 mV and 3 of 100 at 130 mV (95 % interval 1–9 %), and **none of 100 from $\sigma_\mathrm{n}$ = 160 mV** (none of 24 at 200 and 300 mV). This noise (bounds as in the simulation setup) is 197× the largest PVT value and costs 7.5 dB of SNDR (interpolated from study A), more than the 3 dB at which the accuracy counts as lost (72 mV). Even at 160 mV the 95 % upper bound is 3.6 % per pair, not a negligible probability, and an attacker who records more frames gains further opportunities. The 80-window attacker of [1] falls below 10 % of the pairs from $\sigma_\mathrm{n}$ ≈ 60 mV and breaks none of 100 at 70 mV, where the full scan still breaks 87: protection defined by it understates the required noise by more than 2×.
* *Offset.* No offset tested protects the original design. The full-scan attacker breaks every pair up to $\sigma_\mathrm{os}$ = 300 mV, 75 of 100 at 600 mV, where a quarter of the chips already exceed the integrator range (prediction iii), 41 of 100 at 1 V and still 24 of 100 (95 % interval 16–34 %) at 3 V, beyond the supply (the result panel lists every point). Beyond the integrator range the loops clip, their bits become nearly constant, and the clean windows do not disappear.
* *Theory against simulation.* Over the noise sweep, the bound $1 - (1 - f_{64})^{80}$ differs from the 80-window attacker's success by 0.05 on average (as a fraction of pairs), and lies below it, as a lower bound should.

**In the model, with its simplified dither, the comparator cannot give the original design its security at any of the 45 PVT conditions**: analog "protection" and accuracy are in direct conflict. The proposed design therefore derives its security from a key.

#### Explore: comparator noise against accuracy and security

### D. The modulator's own non-idealities: integrator noise and OTA gain

The integrators' non-idealities, the front-end noise of [1] lumped at the first integrator and the OTA DC gain, are swept as model parameters (Section V-C).

* *Integrator noise is not shaped.* Noise of rms $\sigma_i$ (full scale, per clock) added at the first integrator's output appears at the modulator input as $\sigma_i/g_1$, white and ahead of any noise shaping, so only $1/\mathrm{OSR}$ of its power falls in the signal band, where it competes with the in-band quantisation noise $P_q = P_{sig}\,10^{-\mathrm{SNDR}_0/10}$ ($P_{sig} = 0.249^2/2$; $\mathrm{SNDR}_0$: the simulated low-noise SNDR of study A):
  $$\mathrm{SNDR}(\sigma_i) \approx -10\log_{10}\!\left(10^{-\mathrm{SNDR}_0/10} + \frac{\sigma_i^2}{g_1^2\,\mathrm{OSR}\,P_{sig}}\right), \qquad \sigma_{i,\,-3\,\mathrm{dB}} = g_1\sqrt{\mathrm{OSR}\,P_q}.$$
  where $\sigma_{i,\,-3\,\mathrm{dB}}$ is the integrator noise at which the SNDR has fallen by 3 dB. The analysis above predicts that the SNDR falls by 3 dB at about 66× the integrator noise derived from [1].
* *Finite OTA gain makes the integrators leak.* With an OTA DC gain $A$, each integrator leaks with the pole $p = A/(1+A)$ (Section V-C), and the zeros of the NTF move from $z = 1$ to $z = p$. In the signal band $|1 - p\,e^{-j\omega}|^2 \approx \varepsilon^2 + \omega^2$ with $\varepsilon = 1/(1+A)$, and integrating $|\mathrm{NTF}|^2 \propto (\varepsilon^2 + \omega^2)^2$ over $0 \le \omega \le B = \pi/\mathrm{OSR}$ gives the rise of the in-band quantisation noise,
  $$R(A) = 1 + \frac{10}{3}\left(\frac{\varepsilon}{B}\right)^{2} + 5\left(\frac{\varepsilon}{B}\right)^{4}, \qquad \mathrm{SNDR}(A) = \mathrm{SNDR}_0 - 10\log_{10}R(A),$$
  for small leakage ($\varepsilon \ll 1$) and an otherwise ideal loop. The loss should therefore stay below 1 dB for $\varepsilon/B \lesssim 0.27$, that is $A \gtrsim 1.2\,\mathrm{OSR}$ (about 36 dB at OSR = 50): the classic rule that the OTA gain must exceed the oversampling ratio.
* *Integrator noise decorrelates two chips faster than comparator noise, but not before the accuracy is lost.* Unlike comparator noise, which acts only near the threshold, integrator noise enters the state at every clock and accumulates, so it removes clean windows more effectively. The 80-window attacker's success should therefore fall with the integrator noise, but stretches on which the two chips agree, and with them the full-scan attack, should persist as long as the SNDR is intact.

The simulations are behavioural, at the nominal corner only:

* *Accuracy:* the integrator noise from 0.1 to 300× the value derived from [1] (8 points), and the OTA gain from 20 to 100 dB (9 points), three noise seeds each. The two theory curves use the simulated $\mathrm{SNDR}_0$ and no other simulated or fitted value.
* *Security:* the replica attack of study B (both attackers, post-layout offsets) at 1, 3, 10, 30, 100, 300 and 1,000× the integrator noise of [1]; pairs as in study B.

Fig. 51 compares each accuracy sweep with its theory curve, (a) and (b), and draws the attackers' success against the integrator noise, (c), with the settings that lose over 3 dB of SNDR marked, **so that accuracy and security can be read on the same noise axis**.

> **Next cell.** 1. Simulates the SNDR over both sweeps. 2. Evaluates the two theory curves. 3. Runs the replica attack against the integrator noise. 4. Prints the result panel and draws Fig. 51.

````python
#@title Fig.: study D, modulator non-idealities
# Study D: accuracy against the modulator's own non-idealities (behavioural in this work): the first integrator's thermal
# noise and the OTA's DC gain, with the two theory curves (unshaped input noise; the leaky integrators' NTF), and the
# replica attack against the integrator noise (full-scan attacker with Clopper-Pearson intervals, and the 80-window attacker)
t0 = time.time()
NI0 = P_sys["int_noise"]
NI_SWEEP = np.array([0.1, 0.3, 1, 3, 10, 30, 100, 300])                     # multiples of the value derived from [1]
A_SWEEP = np.array([20, 25, 30, 35, 40, 50, 60, 80, 100])                   # OTA DC gain (dB)
# SNDR against the first-integrator noise (multiples of the value derived from [1]) and against the OTA DC gain
sndr_ni = np.array([np.mean([sndr_at({**P_sys, "int_noise": NI0 * m}, k) for k in range(3)]) for m in NI_SWEEP])
sndr_A = np.array([np.mean([sndr_at({**P_sys, "ota_gain": 10 ** (a / 20)}, k) for k in range(3)]) for a in A_SWEEP])
s_ref = sndr_sweep[0]                                                      # the low-noise SNDR of study A
P_sig, g1 = amp_in ** 2 / 2, 0.5
ni_th = np.logspace(-1, np.log10(300), 100)
# theory 1: the integrator noise enters at the input, unshaped; referred through g1 and filtered to the signal band it
# adds (ni NI0)^2 / (g1^2 OSR) to the quantisation-limited noise floor 10^(-s_ref/10) P_sig; -3 dB where they are equal
sndr_ni_th = -10 * np.log10(10 ** (-s_ref / 10) + (ni_th * NI0) ** 2 / (g1 ** 2 * OSR * P_sig))
ni_3db_th = g1 * np.sqrt(OSR * P_sig * 10 ** (-s_ref / 10)) / NI0
# theory 2: a finite DC gain A makes each integrator leaky (pole at 1 - 1/(1 + A)); the NTF's zeros leave DC, which
# raises the in-band quantisation noise by 1 + 10/3 x^2 + 5 x^4, x = leak / (pi / OSR)
A_th = np.linspace(18, 100, 200); x_ = (1 / (1 + 10 ** (A_th / 20))) / (np.pi / OSR)
sndr_A_th = s_ref - 10 * np.log10(1 + 10 / 3 * x_ ** 2 + 5 * x_ ** 4)
A_1db_th = float(A_th[np.argmax(sndr_A_th >= s_ref - 1)])
NI_SEC = np.array([1, 3, 10, 30, 100, 300, 1000])                          # integrator noise for the attack (x value of [1])
rep_ni = replica_sweep(lambda m, n, k0: replica_pairs({**P_sys, "int_noise": NI0 * m}, os0, 1.8, n, 5, k0=k0), NI_SEC)
ni_p, ni_b, ni_S = protection(NI_SEC, rep_ni)
SYS.update(ni_prot=ni_p, ni_3db=crossing(NI_SWEEP, sndr_ni, s_ref - 3), ni_3db_th=ni_3db_th, A_1db_th=A_1db_th,
           A_1db=next((a for a, s in zip(A_SWEEP, sndr_A) if all(x >= s_ref - 1 for x in sndr_A[list(A_SWEEP).index(a):])), None))
card("Study D, the modulator's own non-idealities: accuracy and security",
     [("integrator noise for −3 dB: simulation | theory g<sub>1</sub>√(OSR·P<sub>q</sub>)",
       f"<b>{SYS['ni_3db']:.0f}×</b> | {ni_3db_th:.0f}× the value derived from [1]: the SNDR is limited by quantisation, not by thermal noise"),
      ("OTA DC gain for less than 1 dB loss: simulation | theory R(A)", f"<b>≥ {SYS['A_1db']:.0f} dB</b> (sweep grid) | {A_1db_th:.0f} dB; "
       f"at 30 dB the SNDR decreases by {s_ref - np.interp(30, A_SWEEP, sndr_A):.1f} dB (theory {s_ref - np.interp(30, A_th, sndr_A_th):.1f} dB)"),
      (f"full-scan attacker ({N_FRAMES} frames) against the integrator noise: pairs broken [95 % interval]", ", ".join(
       f"{m}×: {x['k']}/{x['n']} [{100 * x['ci'][0]:.0f}–{100 * x['ci'][1]:.0f} %]" for m, x in zip(NI_SEC, rep_ni))
       + ("; <b>no protection in the sweep</b>" if ni_p is None else f"; protects from {ni_p:g}×")),
      ("the 80-window attacker of [1] (secondary)", ", ".join(f"{m}×: {x['k80']}/{x['n']}" for m, x in zip(NI_SEC, rep_ni))
       + "; f<sub>64</sub>: " + ", ".join(f"{100 * x['f64']:.2g} %" for x in rep_ni)),
      ("run time", f"{time.time() - t0:.0f} s")], tone="ok")
fig, axs = plt.subplots(1, 3, figsize=(W2, 3.2))
ax = axs[0]
ax.semilogx(NI_SWEEP, sndr_ni, "o-", color=COL["ink"], ms=3.5, lw=1.5, label="simulation")
ax.semilogx(ni_th, sndr_ni_th, "--", color=COL["key"], lw=1.2, label="theory")
ax.axvline(1, color=COL["ok"], lw=1.0, ls=":")
ax.text(1.15, 0.5, "from [1]", transform=ax.get_xaxis_transform(), fontsize=7.5, color=COL["ok"])
ax.set_ylim(min(np.min(sndr_ni), np.min(sndr_ni_th)) - 0.4 * (np.max(sndr_ni) - min(np.min(sndr_ni), np.min(sndr_ni_th))),
            np.max(sndr_ni) + 1.5)                                       # every point inside the frame, the key below them
ax.set_xlabel("Integrator noise (× value of [1])"); ax.set_ylabel("Modulator SNDR (dB)")
ax.legend(fontsize=6.5, loc="lower right", handlelength=1.3, borderaxespad=0.3)   # right of the dotted line
panel(ax, "a", "SNDR vs integrator noise")
ax = axs[1]
ax.plot(A_SWEEP, sndr_A, "o-", color=COL["ink"], ms=3.5, lw=1.5, label="simulation")
ax.plot(A_th, sndr_A_th, "--", color=COL["key"], lw=1.2, label="theory $R(A)$")
ax.axvline(20 * np.log10(OSR), color=COL["faint"], lw=0.9, ls=":")
ax.text(20 * np.log10(OSR) + 1.5, 0.55, "$A$ = OSR", transform=ax.get_xaxis_transform(), fontsize=7.5, color=COL["faint"])
ax.set_ylim(sndr_A.min() - 2, s_ref + 3)
ax.set_xlabel("OTA DC gain $A$ (dB)"); ax.set_ylabel("Modulator SNDR (dB)"); ax.legend(fontsize=7, loc="lower right")
panel(ax, "b", "SNDR vs OTA gain")
ax = axs[2]
if SYS["ni_3db"]:
    ax.axvspan(SYS["ni_3db"], NI_SEC[-1] * 1.25, color=COL["bad"], alpha=0.10, lw=0)
    ax.text(SYS["ni_3db"] * 1.15, 72, "SNDR\nbelow\n−3 dB", fontsize=7, ha="left", va="center", color=COL["bad"])
ax.set_xscale("log")
err_curve(ax, NI_SEC, rep_ni, fmt="o-", color=COL["ink"], ms=3.5, lw=1.5, label=f"full scan, {N_FRAMES} frames")
ax.plot(NI_SEC, [100 * x["success80"] for x in rep_ni], "s:", color=COL["faint"], ms=2.5, lw=1.0, label="80 windows [1]")
ax.plot(NI_SEC, [100 * x["bound"] for x in rep_ni], "--", color=COL["key"], lw=1.2, label="bound (80 windows)")
ax.set_ylim(-5, 165); ax.set_yticks(range(0, 101, 20)); ax.set_xlim(0.7, NI_SEC[-1] * 1.25)   # headroom: the legend above the data
ax.set_xlabel("Integrator noise (× value of [1])"); ax.set_ylabel("Pairs broken (%)")
ax2 = ax.twinx()                                                            # f64 on a log scale
ax2.semilogy(NI_SEC, [max(100 * x["f64"], 1e-3) for x in rep_ni], "s:", color=COL["ok"], ms=3.5, lw=1.0, label="clean windows $f_{64}$ (right)")
ax2.set_ylim(5e-4, 3e3); ax2.set_yticks([1e-3, 1e-2, 1e-1, 1, 10]); ax2.set_ylabel("Clean 64-bit windows $f_{64}$ (%)", color=COL["ok"]); ax2.tick_params(axis="y", colors=COL["ok"])
ax2.grid(False)
hs = [a.get_legend_handles_labels() for a in (ax, ax2)]
ax.legend(hs[0][0] + hs[1][0], hs[0][1] + hs[1][1], fontsize=6.5, loc="upper left")
panel(ax, "c", "security vs integrator noise")
fig.tight_layout(); plt.show()
````

### E. Security of the original design: integrator capacitor mismatch

The chip-to-chip mismatch of the coefficients $g_1$ and $g_2$ (capacitor ratios) is **an analog difference a replica cannot copy**. Let each chip's coefficients be $g_1(1+\epsilon_1)$ and $g_2(1+\epsilon_2)$, with $\epsilon_1$, $\epsilon_2$ the relative errors of the two capacitor ratios.

* *Invariance to $g_2$.* The 1-bit quantiser sees only the sign of $x_2$. Given the decisions, $x_2$ is linear in $g_2$: multiplying $g_2$ by $(1 + \epsilon_2)$ multiplies $x_2$ by $(1 + \epsilon_2)$ at every clock. This leaves every sign, and so every decision, unchanged. The analysis above therefore predicts that, without offset, dither and clipping, the bitstream is **exactly independent of $g_2$**, and with them only weakly dependent.
* *Relevant spread.* In a switched-capacitor loop, $g_1$ and $g_2$ are capacitor ratios, which match chip to chip to about 0.1–1 %. The modulator of [1] is a continuous-time loop (Section V): its coefficients are $RC$ or $g_m/C$ products, set by absolute resistor, transconductance and capacitor values, which, untrimmed, spread by ±10–30 % from chip to chip (a typical value, assumed here; [1] reports no spread). The relevant range is therefore 10–30 %, not the 1 % of a capacitor ratio; the sweep covers both. A spread common to both integrators of a chip moves $g_1$ and $g_2$ together, and since only $g_1$ matters, it acts like the independent errors simulated here.
* *Sensitivity to $g_1$.* The first coefficient matters because it scales $x_1$ but not the feedback $v$ that enters $x_2$, which changes the trajectory. Even so, two chips with different $g_1$ agree over long stretches (study B). An error in $g_1$ should therefore change a fraction of the decisions, and the full-scan attacker should fail only once the mismatch removes every stretch of about 48 bits on which the two chips agree, in every frame.

The simulations are behavioural, at the nominal corner only:

* *Scale-invariance check:* the bare loop on the first 20,000 clocks of the sine, with a relative error of 0.01 % to 10 % (5 points) in $g_1$ alone or in $g_2$ alone. The metric is the fraction of decisions unchanged from the error-free loop.
* *Replica attack:* each chip draws independent errors of both coefficients from $\mathcal{N}(0, \sigma^2)$, with the chip-to-chip mismatch σ = 0.01 % to 50 % (9 points); attackers, pairs and offsets as in study B.

Fig. 52(a) tests the invariance to $g_2$ and the sensitivity to $g_1$ by the fraction of decisions unchanged, and Fig. 52(b) shows whether any mismatch of the sweep, from a matched capacitor ratio to the spread of a continuous-time loop, **stops the full-scan attacker**.

> **Next cell.** 1. Runs the bare loop with an error in $g_1$ or $g_2$ alone. 2. Runs the replica attack over the mismatch sweep. 3. Finds the protection point, if any. 4. Prints the result panel and draws Fig. 52.

````python
#@title Fig.: study E, capacitor mismatch
# Study E: security of the original design against the integrators' capacitor-ratio mismatch: the scale-invariance check
# on the bare loop (g2 cannot change a decision, g1 can), then the replica attack with mismatched chips
t0 = time.time()
E_SWEEP = np.array([1e-4, 1e-3, 1e-2, 3e-2, 0.1])                         # relative error of one coefficient
# Theory check on the bare loop (no noise, dither or offset): the decision is the sign of x2, and scaling g2 scales x2
# by a positive factor at every clock, so an error in g2 alone cannot change a decision; an error in g1 changes the
# path of x1 and therefore can. ref_bits are the decisions with the nominal g1 = g2 = 1/2.
ref_bits = bare_loop(u_id)
# fraction of decisions unchanged when only g1 (then only g2) carries the relative error e
agree_g1 = np.array([np.mean(bare_loop(u_id, g1=0.5 * (1 + e)) == ref_bits) for e in E_SWEEP])
agree_g2 = np.array([np.mean(bare_loop(u_id, g2=0.5 * (1 + e)) == ref_bits) for e in E_SWEEP])
CM_SWEEP = np.array([0.01, 0.03, 0.1, 0.3, 1, 3, 10, 30, 50]) * 1e-2       # sigma of each capacitor ratio, chip to chip
# Replica attack with chip-to-chip capacitor-ratio mismatch: each chip draws its own coefficient errors with standard
# deviation x; replica_pairs runs victim/replica pairs at the post-layout offset spread os0 and a 1.8 µA input, and
# replica_sweep repeats it over CM_SWEEP, adding pairs where the outcome is uncertain (k0: seed offset per point)
rep_cm = replica_sweep(lambda x, n, k0: replica_pairs(P_sys, os0, 1.8, n, 6, coef_sigma=x, k0=k0), CM_SWEEP)
# success rate per point; the smallest mismatch from which no pair is broken (protection); the result at sigma = 1 %
s_cm = g(rep_cm, "success"); cm_p, cm_b, cm_S = protection(CM_SWEEP, rep_cm); cm_1 = rep_cm[list(CM_SWEEP).index(1e-2)]
# results kept for the circuit-to-system map (study G) and the findings
SYS.update(cm_prot=cm_p, cm_1pc=cm_1["success"], cm_1pc_ci=cm_1["ci"], cm_10_80=crossing(CM_SWEEP, g(rep_cm, "success80"), 0.1),
           g2_exact=bool((agree_g2 == 1).all()))
card("Study E, security of the original design: integrator capacitor mismatch",
     [("theory check: g<sub>2</sub> error alone (bare loop)", f"decisions unchanged at every error up to 10 %: <b>{SYS['g2_exact']}</b>"),
      ("theory check: g<sub>1</sub> error alone", ", ".join(f"{100 * e:g} %: {100 * a:.1f} %" for e, a in zip(E_SWEEP, agree_g1))
       + " of decisions unchanged"),
      (f"full-scan attacker ({N_FRAMES} frames), capacitor-ratio mismatch between chips: pairs broken [95 % interval]",
       ", ".join(f"{100 * x:g} %: {S['k']}/{S['n']} [{100 * S['ci'][0]:.0f}–{100 * S['ci'][1]:.0f} %]" for x, S in zip(CM_SWEEP, rep_cm))),
      ("protection by capacitor mismatch", f"at σ = 1 %: <b>{cm_1['k']} of {cm_1['n']}</b> pairs broken; " + (
       f"no pair broken from σ = {100 * cm_p:g} % ({cm_S['k']} of {cm_S['n']})" if cm_p else
       f"<b>none in the sweep</b>: still {cm_S['k'] if cm_S else rep_cm[-1]['k']} of {rep_cm[-1]['n']} pairs at σ = {100 * CM_SWEEP[-1]:.0f} %")),
      ("the 80-window attacker of [1] (secondary)", ", ".join(f"{100 * x:g} %: {S['k80']}/{S['n']}" for x, S in zip(CM_SWEEP, rep_cm))),
      ("run time", f"{time.time() - t0:.0f} s")], tone="bad")
# (a) the scale-invariance check; (b) pairs broken against the mismatch: full-scan attacker with 95 % intervals, the
# 80-window attacker of [1], and the bound from the measured clean-window fraction f_64
fig, axs = plt.subplots(1, 2, figsize=(W2, 2.9))
ax = axs[0]
ax.semilogx(100 * E_SWEEP, 100 * agree_g2, "s-", color=COL["ok"], ms=4, lw=1.5, label="error in $g_2$ only (theory: 100 %)")
ax.semilogx(100 * E_SWEEP, 100 * agree_g1, "o-", color=COL["bad"], ms=4, lw=1.5, label="error in $g_1$ only")
ax.set_ylim(0, 105); ax.set_xlabel("Relative coefficient error (%)"); ax.set_ylabel("Decisions unchanged (%)")
ax.legend(fontsize=7, loc="lower left"); panel(ax, "a", "scale invariance of the 1-bit loop")
ax = axs[1]
ax.set_xscale("log")
err_curve(ax, 100 * CM_SWEEP, rep_cm, fmt="o-", color=COL["ink"], ms=3.5, lw=1.5, label=f"full scan, {N_FRAMES} frames")
ax.plot(100 * CM_SWEEP, 100 * g(rep_cm, "success80"), "s:", color=COL["faint"], ms=2.5, lw=1.0, label="80 windows, one frame [1]")
ax.plot(100 * CM_SWEEP, 100 * g(rep_cm, "bound"), "--", color=COL["key"], lw=1.2, label=r"bound $1 - (1 - f_{64})^{80}$")
ax.axvline(1, color=COL["bad"], lw=0.9, ls=":")                            # 1 %: the mismatch of a matched capacitor ratio
ax.plot(1, 100 * SYS["cm_1pc"], "o", ms=7, mfc="none", mec=COL["bad"], mew=1.2)
ax.annotate(f"matched ratio, 1 %:\n{cm_1['k']} of {cm_1['n']} pairs\nbroken", (1, 100 * SYS["cm_1pc"]), xytext=(0.7, 72),
            textcoords="data", ha="right", va="center", fontsize=7.5, color=COL["bad"],
            arrowprops=dict(arrowstyle="-", lw=0.6, color=COL["bad"]))
ax.set_ylim(-5, 108); ax.set_xlabel(r"Capacitor-ratio mismatch $\sigma$ (%)"); ax.set_ylabel("Pairs broken (%)")
ax.legend(fontsize=6.5, loc="lower left", handlelength=1.6); panel(ax, "b", "security vs capacitor mismatch")
fig.tight_layout(); plt.show()
````

**Key insights.**

* *Scale invariance.* The bare loop confirms both predictions (Fig. 52(a)). An error in $g_2$ alone leaves every decision unchanged up to 10 % (`True`). An error in $g_1$ alone changes decisions: 99.7 % stay unchanged at 0.01 %, 96.8 % at 0.1 %, and 76–82 % from 1 % to 10 %.
* *Replica attack.* The full-scan attacker (Fig. 52(b)) breaks all 24 pairs at every mismatch up to 3 %, including the 1 % of a matched capacitor ratio; 78 of 100 pairs at 10 % (95 % interval 69–86 %), 31 of 100 at 30 % (22–41 %) and still 16 of 100 at 50 % (9–25 %). The 80-window attacker of [1] breaks fewer: 44 of 100 at 10 %, 13 at 30 % and 6 at 50 %. No mismatch in the sweep protects the design: **the integrators give the original design no security either**, neither at the 1 % of a matched capacitor ratio nor at the ±10–30 % spread of a continuous-time loop such as that of [1]. In the model, with its simplified dither, no analog block of the original design protects it; the proposed design rests its security on the SRAM-PUF cell, which study F carries to the key.

### F. Security of the proposed design: from SRAM-PUF cell to key

**The proposed design's security rests on one transistor-level quantity**: the power-up behaviour of the SRAM cells that regenerate the key, whose errors the BCH code of Section VII-E (t = 45) corrects. The cell's error law (Section VII-F) is noise of noise-to-mismatch ratio $s(T)$ at temperature $T$ plus, at the fast-nMOS corners and high temperature, a rotation θ(T) of the mismatch that makes a fixed set of cells change sign. Three relations carry this law to the key over a population of chips.

*1) Key failure over a population (Section VII-E):* With per-cell error probabilities $p_i$, the number of errors $E$ in the 600-bit fingerprint is Poisson-binomial, and the key fails when $E$ exceeds the correction capability $t$: $P_{\text{fail}} = P(E > t)$, **computed exactly for each chip and compared with the target** $10^{-6}$.

*2) Secrecy:* The BCH code over the Galois field $\mathrm{GF}(2^{10})$ publishes 10 helper bits per correctable error, so with the min-entropy $H_\infty = -\log_2\max(P_1, 1 - P_1)$ of a cell that powers up to 1 with probability $P_1$, $n H_\infty - 10t$ of the $n$ cells' bits remain secret, against the 192 bits required at the 95 % lower bound on $H_\infty$ (Section VII-E). A two-temperature enrolment also publishes a keep-mask, so the entropy is that of the kept cells, estimated from them alone (the mask, pointer-type helper data open to manipulation [23], must be hashed with the enrolled fingerprint $w$ into the key $k$ [24], so that any change yields a key the receiver rejects; not modelled). The analysis above predicts that with 512 cells per corner the lower bound lies near 0.8 bit, **so 600 cells allow at most t ≈ 27**.

*3) Two-temperature enrolment and choice of temperatures:* If the mismatch rotates monotonically with temperature, a cell's sign changes at most once over any range of θ shorter than π, so a cell with the same sign at both enrolment temperatures keeps it between them, but not outside. An enrolment at 25 and 85 °C thus says nothing about 0 °C, the bottom of the specified range, where θ has the opposite sign; **one at −40 and 85 °C, the ends of the simulated range, covers 0–85 °C**. The cost is the fraction of cells whose sign changes, about θ(85 °C)/π, plus the unstable ones. The analysis above therefore predicts that with the −40 and 85 °C enrolment the kept cells' held-out error rate is low at all three temperatures, whereas the 25 and 85 °C enrolment leaves errors at −40 °C wherever θ(−40 °C) is large, and that the key-failure probability no longer depends on the flips at any corner. Fig. 53 tests relations 1 and 3 together against the secrecy limit of relation 2: panel (b) gives the t each corner requires with the 25 °C enrolment, panel (c) with the enrolment at −40 and 85 °C, and the grey region marks the codes that leave too few secret bits.

The study runs at the behavioural level, with the two-component cell model of Section VII-E **anchored at each corner to the recorded transistor-level Monte Carlo** of Section VII-F (Table 14: 512 cells × 5 power-ups at −40, 25 and 85 °C per corner). s(T) and θ(T) are extracted at each temperature as in Section VII-E and interpolated linearly. The other settings are as follows.

* *Population:* 1,000 chips per corner and enrolment, seeded from `chip_seed`. For the two-temperature enrolment each chip has 1,000 cells, keeps those stable and equal over three reads at each temperature, and uses the first 600 kept (a chip with fewer fails).
* *Enrolments:* one read at 25 °C, or three reads at −40 and at 85 °C; the 25 and 85 °C enrolment of Table 15 is compared on the recorded bits.
* *Cross-validation on the recorded bits:* for each two-temperature enrolment, the mask is chosen from power-ups 1–3, and the kept cells' error rate against the enrolled value is evaluated on the held-out power-ups 4–5 at −40, 25 and 85 °C, with a 95 % Clopper–Pearson interval (reads of a cell treated as independent, so the intervals are somewhat narrow). The model, not fitted to it, also predicts the kept fraction.
* *Code and ranges:* t from 0 to 120; worst case over 0, 25 and 85 °C, where the piecewise-linear s(T) and θ(T) put the maximum over 0–85 °C. Panel (a) follows this chip (`chip_seed`) from −40 to 85 °C in 5 °C steps; the figure draws probabilities below $10^{-16}$ at $10^{-16}$.
* *Secrecy:* $H_\infty$ per corner (95 % lower bound and point estimate) from all 512 cells or the kept cells; the corner with the lowest bound sets the limit.
* *Metrics (per corner):* chips meeting $10^{-6}$ at t = 45; 99th percentile of the required t; held-out error rates; secret bits; the array size for which 99.9 % of chips keep 600 cells.

> **Next cell.** 1. Anchors the model per corner and simulates 1,000 chips per enrolment. 2. Cross-validates both two-temperature enrolments on the recorded bits. 3. Bounds the min-entropy and derives the secrecy limits. 4. Evaluates this chip against temperature. 5. Prints the result panel and draws Fig. 53.

````python
#@title Fig.: study F, SRAM-PUF to key
# Study F: from SRAM-PUF cell to key, at the five corners. Each corner gets the two-component cell model of Section
# VII-E (chip_population: noise s(T) from repeated power-ups, rotation theta(T) from the error rate), anchored to its
# recorded PVT Monte Carlo (512 cells x 5 power-ups at -40, 25 and 85 °C). Key reliability over 1,000 simulated chips for a
# 25 °C enrolment and for the enrolment at -40 and 85 °C; both two-temperature enrolments (25 + 85 and -40 + 85 °C) are
# also cross-validated on the recorded bits (mask from power-ups 1-3, error rates on the held-out power-ups 4-5, with 95 %
# Clopper-Pearson intervals); the secret budget uses the lower confidence bound on the min-entropy of the kept cells.
t_used = CAL["bch_t"]
E_RD, H_RD, N_ARR = [0, 1, 2], [3, 4], 1000              # screening reads, held-out reads, simulated array (two-temperature)
ENROL = {"25": (25,), "−40+85": (-40, 85)}
cp95 = lambda k, n: (float(_beta.ppf(0.025, k, n - k + 1)) if k else 0.0, float(_beta.ppf(0.975, k + 1, n - k)) if k < n else 1.0)
PUF_SYS = {}
for j, c in enumerate(PVT["puf"]["corners"]):
    # b[temperature, power-up, cell]: the corner's recorded bits; w0, the 25 °C enrolment (first power-up at 25 °C);
    # ber[T], the error rate against w0; q[T], the mean disagreement between two power-ups at T (noise alone); from these
    # two_component gives the noise s(T) and the mismatch rotation theta(T) of the corner
    k, b, mu = c["corner"], BITS[j].astype(np.int8), c["fit"]["mu"]; w0 = b[i25, 0]; R = b.shape[1]
    ber = {T: float(((b[i, 1:] if T == 25 else b[i]) != w0).mean()) for i, T in enumerate(TEMPS)}
    q = {T: float(np.mean([(b[i, x] != b[i, y]).mean() for x in range(R) for y in range(x + 1, R)])) for i, T in enumerate(TEMPS)}
    sT, th = two_component(mu, ber, q)
    # cross-validation of a two-temperature enrolment on the recorded bits: keep the cells that are stable over the three
    # screening reads at both temperatures and agree between them; count their errors on the two held-out power-ups
    st3 = lambda i: b[i][E_RD].min(axis=0) == b[i][E_RD].max(axis=0)
    xv = {}
    for name, (Ta, Tb) in (("25+85", (25, 85)), ("−40+85", (-40, 85))):          # recorded-bit cross-validation
        ia, ib = TEMPS.index(Ta), TEMPS.index(Tb)
        keep = st3(ia) & st3(ib) & (b[ia, E_RD[0]] == b[ib, E_RD[0]]); w = b[ia, E_RD[0]]
        err = {T: (int((b[i][H_RD][:, keep] != w[keep]).sum()), int(keep.sum()) * len(H_RD)) for i, T in enumerate(TEMPS)}
        xv[name] = dict(keep=float(keep.mean()), err=err, h=h_bounds(int(w[keep].sum()), int(keep.sum())))
    # 1,000 modelled chips per enrolment: for each chip the worst key-failure probability over 0-85 °C at every t, the t it
    # requires, and (two-temperature enrolment) the number of cells kept
    pop = {name: chip_population(mu, sT, th, enrol=e, n_arr=PUF_BITS if len(e) == 1 else N_ARR, tmax=120 if len(e) == 1 else 60,
                                 seed=DEFAULTS["chip_seed"] * 100 + j) for name, e in ENROL.items()}
    PUF_SYS[k] = dict(PUF_CELL[k], sT=sT, th=th, xv=xv, pop=pop, h_all=h_bounds(int(w0.sum()), len(w0)),
                      t99={n: int(np.percentile(p["treq"], 99, method="higher")) for n, p in pop.items()},
                      frac={n: float(np.mean(p["treq"] <= t_used)) for n, p in pop.items()},
                      keep_model=float(np.mean(pop["−40+85"]["kept"]) / N_ARR))
# this chip (chip_seed) at every temperature, 25 °C enrolment, t = t_used (one chip of the population model per corner)
TT, t_grid = np.arange(-40, 86, 5), np.arange(0, 121)
key_TT = {k: np.array([chip_population(q["mu"], q["sT"], q["th"], n_chips=1, temps=(int(T),), seed=DEFAULTS["chip_seed"])["worst"][0]
                       for T in TT]) for k, q in PUF_SYS.items()}                # [temperature, t], every t of the code
key_T = {k: v[:, t_used] for k, v in key_TT.items()}
p99 = lambda wst: np.percentile(wst, 99, axis=0)                                # the 99th-percentile chip, for each t
# the key generator these models support: -40 + 85 °C enrolment, 600 kept cells, the largest 99th-percentile t of the corners
t_fin = max(q["t99"]["−40+85"] for q in PUF_SYS.values())
h_keep = {k: q["xv"]["−40+85"]["h"] for k, q in PUF_SYS.items()}
k_h = min(h_keep, key=lambda x: h_keep[x][0])                                  # corner with the lowest entropy bound
lim = lambda h: int(np.floor((PUF_BITS * h - SECRET_REQ) / 10))                 # largest t leaving SECRET_REQ secret bits
t_lim25 = lim(min(q["h_all"][0] for q in PUF_SYS.values())); t_lim2 = lim(h_keep[k_h][0])
kf = {k: np.quantile(q["pop"]["−40+85"]["kept"] / N_ARR, 0.001) for k, q in PUF_SYS.items()}
k_kf = min(kf, key=kf.get); n_cells_fin = int(np.ceil(PUF_BITS / kf[k_kf]))      # 99.9 % of chips keep 600 cells
xmax = lambda name, k: max(PUF_SYS[k]["xv"][name]["err"].items(), key=lambda e: e[1][0] / e[1][1])
ci = lambda e: "{:.2f} % ({:.2f}–{:.2f})".format(100 * e[0] / e[1], *(100 * x for x in cp95(*e)))
pc_ = lambda v: f"{100 * v:.1f} %"
card(f"Study F, security of the proposal: from SRAM-PUF cell to key (two-component cell model, 1,000 chips per corner)",
     [(f"this chip, t = {t_used}, 25 °C enrolment: first temperature from −40 °C up with P(key fails) above 10<sup>−6</sup>",
       ", ".join(f"{k} " + next((f"{T} °C".replace("-", "−") for T, v in zip(TT, key_T[k]) if v > 1e-6), "none") for k in PUF_SYS)),
      (f"chips meeting 10<sup>−6</sup> over 0–85 °C at t = {t_used}, 25 °C enrolment", ", ".join(f"{k} {pc_(q['frac']['25'])}" for k, q in PUF_SYS.items())),
      ("t required for 10<sup>−6</sup>, 99th percentile of the chips, 25 °C enrolment",
       ", ".join(f"{k} {q['t99']['25'] if q['t99']['25'] <= 120 else '&gt; 120'}" for k, q in PUF_SYS.items())),
      ("recorded bits, enrolment at 25 and 85 °C: held-out error rate of the kept cells at −40 °C (95 % interval)",
       ", ".join(f"{k} {ci(q['xv']['25+85']['err'][-40])}" for k, q in PUF_SYS.items())),
      ("recorded bits, enrolment at −40 and 85 °C: highest held-out error rate of the kept cells over −40, 25, 85 °C",
       ", ".join(f"{k} {ci(xmax('−40+85', k)[1])} at {xmax('−40+85', k)[0]} °C".replace("-4", "−4") for k in PUF_SYS)),
      ("cells kept by the −40 + 85 °C enrolment: recorded | model (not fitted to it)",
       ", ".join(f"{k} {pc_(q['xv']['−40+85']['keep'])} | {pc_(q['keep_model'])}" for k, q in PUF_SYS.items())),
      ("t required for 10<sup>−6</sup>, 99th percentile, enrolment at −40 and 85 °C",
       ", ".join(f"{k} {q['t99']['−40+85']}" for k, q in PUF_SYS.items()) + f"; chips meeting 10<sup>−6</sup> at t = {t_fin}: at least "
       f"{pc_(min(np.mean(q['pop']['−40+85']['treq'] <= t_fin) for q in PUF_SYS.values()))} at every corner"),
      ("min-entropy per cell, 95 % lower bound – point estimate: all cells | cells kept at −40 and 85 °C",
       ", ".join(f"{k} {q['h_all'][0]:.3f}–{q['h_all'][1]:.3f} | {h_keep[k][0]:.3f}–{h_keep[k][1]:.3f}" for k, q in PUF_SYS.items())),
      (f"secrecy limit (≥ {SECRET_REQ} secret bits at the lower bound of the worst corner)",
       f"t ≤ {t_lim25} with {PUF_BITS} cells; t ≤ {t_lim2} with {PUF_BITS} cells kept at −40 and 85 °C"),
      (f"key generator: enrolment at −40 and 85 °C, {PUF_BITS} kept cells, t = {t_fin}",
       f"<b>{int(np.floor(PUF_BITS * h_keep[k_h][0] - 10 * t_fin))}–{int(np.floor(PUF_BITS * h_keep[k_h][1] - 10 * t_fin))} secret bits</b> "
       f"at {k_h}; <b>{n_cells_fin} cells</b> so that 99.9 % of the chips keep {PUF_BITS} at {k_kf}")],
     tone="ok" if t_fin <= t_lim2 else "bad")
pcol = {"tt": COL["ink"], "ss": COL["ok"], "ff": COL["bad"], "sf": COL["aqua"], "fs": COL["yellow"]}
lsty = lambda k: dict(ls="--", lw=1.8, zorder=5) if k == "tt" else dict(ls="-", lw=1.4, zorder=3)   # tt on top, dashed
fig, axs = plt.subplots(1, 3, figsize=(W2, 3.4), gridspec_kw=dict(wspace=0.12))
ax = axs[0]
ax.axvspan(0, 85, color=COL["faint"], alpha=0.2, lw=0)
for k in PUF_SYS:
    ax.semilogy(TT, np.maximum(key_T[k], 1e-16), color=pcol[k], label=k, **lsty(k))
ax.axhline(1e-6, color=COL["key"], lw=0.9, ls=":"); ax.set_ylim(1e-17, 30); ax.set_xticks([-40, 0, 25, 85])
ax.set_xlabel("Temperature (°C)"); ax.set_ylabel("P(key fails)")
ax.legend(ncol=2, fontsize=7, loc="upper left", handlelength=1.8, columnspacing=0.8)
panel(ax, "a", f"This chip, t = {t_used}")
# panels (b), (c) as Fig. 3 of the ISCAS paper: grey band where fewer than SECRET_REQ bits stay secret, a legend of the t
# each corner requires for 1e-6 (99th-percentile chip), circles at that t, the 1e-6 target dotted; (c) zoomed to t <= 52
for ax, name, lab, title, tl, xmax in ((axs[1], "25", "b", "25 °C enrolment", t_lim25, t_grid[-1]),
                                       (axs[2], "−40+85", "c", "−40 and 85 °C enrolment", t_lim2, 52)):
    ax.axvspan(tl + 0.5, xmax, color=COL["band"], alpha=0.6, lw=0, zorder=0)
    if lab == "b":                                   # right of the curves (b), left of the dashed t (c): clear of every line
        ax.text(0.985 * xmax, 1e-3, f"fewer than\n{SECRET_REQ} secret\nbits", fontsize=8, color=COL["ink"], ha="right", va="center")
    else:
        ax.text(t_used - 1.5, 1e-3, f"fewer than {SECRET_REQ}\nsecret bits", fontsize=8, color=COL["ink"], ha="right", va="center")
    for k, q in PUF_SYS.items():
        y = np.pad(p99(q["pop"][name]["worst"]), (0, len(t_grid)), mode="edge")[:len(t_grid)]   # beyond tmax: below 1e-16
        t_k = q["t99"][name]; ok_k = t_k < q["pop"][name]["worst"].shape[1]
        ax.semilogy(t_grid, np.maximum(y, 1e-16), color=pcol[k], label=f"{k}: {t_k}" if ok_k else f"{k}: > {t_k - 1}", **lsty(k))
        if ok_k:                                                                                        # the t this corner requires for 1e-6
            ax.plot(t_k, 1e-6, "o", ms=4, color=pcol[k], mec="white", mew=0.5, zorder=6)
    ax.axhline(1e-6, color=COL["key"], lw=0.9, ls=":")
    if t_used <= xmax:
        ax.axvline(t_used, color=COL["ink"], lw=0.8, ls="--")
    ax.set_ylim(1e-17, 30); ax.set_xlim(0, xmax); ax.set_xlabel("BCH correction t")
    ax.tick_params(labelleft=False)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.2), fontsize=7.5, handlelength=1.4, labelspacing=0.15, ncol=3,
              columnspacing=0.9, title="required $t$ (99th-percentile chip)", title_fontsize=7.5, frameon=False)   # off the curves
    panel(ax, lab, title)
axs[1].text(2, 2e-6, "$10^{-6}$", fontsize=8, color=COL["key"], va="bottom")
axs[1].text(t_used + 2, 1e-12, f"t = {t_used}", fontsize=7.5, color=COL["ink"], ha="left")
fig.subplots_adjust(left=0.085, right=0.99, bottom=0.265, top=0.91)
plt.show()
# interactive companion: the same curves per corner (one legend entry per corner toggles it in all three panels); hover
# gives the corner, the temperature or t, the probability and the t that corner requires; panel (c) opens at t <= 52 like
# the static figure and holds the whole t range for zooming out. The slider below the panels sets t: panel (a) shows this
# chip at that t (key_TT holds every t) and the dashed line in (b) and (c) moves to it
def _ipuf():
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    f = make_subplots(rows=1, cols=3, horizontal_spacing=0.05, shared_yaxes=True,
                      subplot_titles=(f"(a) This chip, t = {t_used}", "(b) 25 °C enrolment", "(c) −40 and 85 °C enrolment"))
    _idash = lambda k: dict(color=pcol.get(k, COL["ink"]), width=2.2 if k == "tt" else 1.6, dash="dash" if k == "tt" else "solid")
    for k in PUF_SYS:
        f.add_trace(go.Scatter(x=TT, y=np.maximum(key_T[k], 1e-16), mode="lines", line=_idash(k), name=k, legendgroup=k,
                               hovertemplate=f"corner {k}<br>" + "%{x:.0f} °C<br>P(key fails) = %{y:.2e}<extra></extra>"), row=1, col=1)
    for col, name, tl, x_hi, lab in ((2, "25", t_lim25, float(t_grid[-1]), "25 °C enrolment"), (3, "−40+85", t_lim2, 52.0, "−40 and 85 °C enrolment")):
        f.add_vrect(x0=tl + 0.5, x1=float(t_grid[-1]), fillcolor=COL["band"], opacity=0.6, line_width=0, layer="below",
                    exclude_empty_subplots=False, row=1, col=col)
        for k, q in PUF_SYS.items():
            yk = np.pad(p99(q["pop"][name]["worst"]), (0, len(t_grid)), mode="edge")[:len(t_grid)]
            t_k = q["t99"][name]; ok_k = t_k < q["pop"][name]["worst"].shape[1]
            req = f"{t_k}" if ok_k else f"> {t_k - 1}"
            f.add_trace(go.Scatter(x=t_grid, y=np.maximum(yk, 1e-16), mode="lines", line=_idash(k), name=k, legendgroup=k, showlegend=False,
                                   customdata=yk, hovertemplate=f"corner {k}, {lab}<br>" + "t = %{x}<br>P(key fails), 99th-percentile chip = "
                                   "%{customdata:.2e}<br>" + f"required t for 10<sup>−6</sup>: {req}<extra></extra>"), row=1, col=col)
            if ok_k:
                f.add_trace(go.Scatter(x=[t_k], y=[1e-6], mode="markers", marker=dict(size=8, color=pcol.get(k, COL["ink"]),
                                       line=dict(color="white", width=0.8)), name=k, legendgroup=k, showlegend=False,
                                       hovertemplate=f"corner {k}, {lab}: required t = {t_k}<extra></extra>"), row=1, col=col)
        f.update_xaxes(range=[0, x_hi], title_text="BCH correction t", row=1, col=col)
    iv = []                                                             # indices of the t-line shapes
    for col in (1, 2, 3):
        x_line = [-40, 85] if col == 1 else [0, float(t_grid[-1])]
        f.add_trace(go.Scatter(x=x_line, y=[1e-6, 1e-6], mode="lines", line=dict(color=COL["key"], width=1, dash="dot"),
                               name="10<sup>−6</sup> target", legendgroup="target", showlegend=col == 1, hoverinfo="skip"), row=1, col=col)
        if col > 1:                                                     # the t line: a shape, which the slider moves
            iv.append(len(f.layout.shapes))
            f.add_vline(x=t_used, line=dict(color=COL["ink"], width=1, dash="dash"), row=1, col=col)
            if col == 2:
                f.add_trace(go.Scatter(x=[None], y=[None], mode="lines", line=dict(color=COL["ink"], width=1, dash="dash"),
                                       name="t (slider)", hoverinfo="skip"), row=1, col=col)
    f.add_vrect(x0=0, x1=85, fillcolor=COL["faint"], opacity=0.2, line_width=0, layer="below", exclude_empty_subplots=False, row=1, col=1)
    f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", name=f"fewer than {SECRET_REQ} secret bits (grey band)", hoverinfo="skip",
                           marker=dict(symbol="square", size=12, color=COL["band"])), row=1, col=2)
    f.update_xaxes(range=[-40, 85], tickvals=[-40, 0, 25, 85], title_text="Temperature (°C)", row=1, col=1)
    f.update_yaxes(type="log", range=[-17, np.log10(30)], exponentformat="power", row=1, col=1)
    f.update_yaxes(type="log", range=[-17, np.log10(30)], row=1, col=2)
    f.update_yaxes(type="log", range=[-17, np.log10(30)], row=1, col=3)
    f.update_yaxes(title_text="P(key fails)", row=1, col=1)
    ia = [i for i, tr in enumerate(f.data) if tr.xaxis in (None, "x") and tr.legendgroup in PUF_SYS]       # (a): this chip
    steps = [dict(method="update", label=str(t),                    # (a) at this t (4 significant digits), the t lines
                  args=[{"y": [_isig(np.maximum(key_TT[k][:, t], 1e-16), 4) for k in PUF_SYS]},
                        {"annotations[0].text": f"(a) This chip, t = {t}", **{f"shapes[{i}].{e}": t for i in iv for e in ("x0", "x1")}}, ia])
             for t in list(range(0, 61)) + list(range(65, 121, 5))]
    f.update_layout(sliders=[dict(steps=steps, active=[s_["label"] for s_ in steps].index(str(t_used)), x=0.06, len=0.9, y=0,
                                  yanchor="top", pad=dict(t=round(30 * _PT), b=0), currentvalue=dict(prefix="BCH correction t = ",
                                  font=dict(size=round(8 * _PT))), font=dict(size=round(7 * _PT)), tickcolor=COL["faint"],
                                  bordercolor=COL["faint"], bgcolor="white", activebgcolor=COL["faint"])],
                    margin=dict(b=round(30 * _PT) + 70))
    return f
iplot(_ipuf, "Fig.: SRAM-PUF key failure vs temperature and vs BCH t, per corner (interactive)",
      height=round(DISPLAY_PX * show.last[1] / show.last[0]) + 70)
````

<p align="center" style="text-align:center"><b>Fig. 53.</b> Study F: SRAM-PUF to key (cell model per corner). (a) Key-failure probability of the model's reference chip against temperature; t = 45, 25 °C enrolment; grey: 0–85 °C. (b), (c) Worst case over 0–85 °C of the 99th percentile of 1,000 modelled chips against t; enrolment at 25 °C (b), −40 and 85 °C (c, t up to 52); dotted: $10^{-6}$ target; circles and the key below each panel: t required; dashed vertical: t = 45; grey: under 192 secret bits.</p>

**Key insights.**

* *25 °C enrolment.* For one simulated chip at t = 45 (Fig. 53(a)) the key-failure probability stays below $10^{-6}$ from −40 to 85 °C at tt and ss; it exceeds the target from 55 °C at ff and from 40 °C at fs, where θ grows, and already at −40 °C at sf, whose cell is the noisiest. Over the population, all chips meet the target at tt and ss, 72.9 % at sf and none at ff and fs. The t each corner requires (99th percentile; dots in Fig. 53(b)) is 33 at tt, 29 at ss, 55 at sf, 89 at ff and more than 120 at fs; Section VII-E, anchored to the 1,024-cell post-layout run instead of 512 recorded cells per corner, finds t = 39 at tt. The secrecy limit with 600 cells, set by the lowest bound (0.772 bit at ff), is t ≤ 27, as item 2 predicts: **with a 25 °C enrolment no corner, not even tt, meets both requirements.** The BCH code is sized for noise; the reproducible flips at ff and fs, and the higher noise at sf, would need codes far beyond the secrecy limit.
* *Choice of enrolment temperatures (recorded bits).* With the enrolment at 25 and 85 °C, the kept cells' held-out error rate at −40 °C is 1.72 % (95 % interval 1.01–2.75 %) at ss and 3.09 % (2.08–4.40 %) at sf, as predicted by item 3, and 0.31–0.98 % at the other corners. With the enrolment at −40 and 85 °C, the highest held-out error rate over the three temperatures is at most 1.01 % (0.46–1.91 %, ff at 25 °C) at every corner; at sf it is 0.11 %. This enrolment keeps 95.3, 94.5, 86.9, 88.5 and 68.9 % of the recorded cells at tt, ss, ff, sf and fs; the model predicts 94.7, 94.9, 85.6, 88.4 and 68.7 %, within 1.3 percentage points at every corner. **The two-temperature enrolment excludes the flipping cells from the key, provided its two temperatures bracket the operating range** (at tt, ff and fs two held-out reads do not distinguish the two enrolments at −40 °C).
* *Enrolment at −40 and 85 °C.* The 99th percentile of the required t is 9 at tt, 10 at ss, 13 at ff, 16 at sf and 17 at fs (Fig. 53(c)); at t = 17, at least 99.7 % of the chips meet $10^{-6}$ at every corner. The min-entropy of the kept cells lies between 0.758 bit (95 % lower bound, fs) and 0.941 bit (point estimate, sf); the mask removes little entropy (at fs, 0.758 against 0.777 bit for all cells). With 600 kept cells the limit is t ≤ 26. The key generator with this enrolment, 600 kept cells and t = 17 leaves 284–352 secret bits at fs, more than the 192 required, and needs 935 cells so that 99.9 % of the simulated chips keep 600 at fs. fs, the best corner at 25 °C (Section VII-F), thus sets every limit of the key generator and must be screened hot.
* *Limits.* These results rest on the two-component model, independent cells, thermal-only cell noise (Section VII-C) and, for the held-out error rates, two reads per cell. At the conservative $s_{25} = 0.06$, 2.7 times the simulated value, a tt chip enrolled at 25 °C needs t = 60 and keeps no secret bits (Section VII-E), so the cell noise must be measured in silicon before the code is fixed.

Section XII regenerates the key from transistor-level cells as an independent check of the cell model's error rate at one condition per run; **it does not test the two-temperature enrolment or the t = 17 code**. Section XI sizes the decoder and the array of the complete key generator.

#### Explore: size the key generator

The explorer works at the behavioural level, as study F: each setting simulates 1,000 chips of the chosen corner with `chip_population` and the seeds of study F, and follows the reference chip (`chip_seed`) over temperature. Secrecy is the budget ⌊600 h − 10 t⌋ at the 95 % lower bound of the corner's min-entropy per cell h (all 600 cells for the 25 °C enrolment, the kept cells for the enrolment at −40 and 85 °C), against the 192 bits required. **A design point meets both requirements when 99 % of the chips meet the target and the budget holds**. In Fig. 54, panel (b) places the chosen t against the t the 99th-percentile chip requires and against the secret-bit limit (grey), and panel (c) shows how that requirement spreads over the chips, so that one view shows whether a design point meets both requirements. The two-temperature enrolment uses the array of study F, the 25 °C enrolment its 600 cells. The menus offer the five corners, both enrolments, 16 values of t from 0 to 120 and both targets; every combination was simulated beforehand, so a choice shows its figure and result panel at once, without running the cell.

> **Next cell.** 1. Defines `keygen_view`, which simulates the population for one design point, draws Fig. 54 and returns the result panel (it computed every view of the menus). 2. Shows the explorer, opening on the key generator of study F (enrolment at −40 and 85 °C, its t and its array) at the corner with the fewest kept cells.

````python
#@title Explore: key-generator sizing
# Explore: size the key generator for any corner, enrolment, BCH correction t and key-failure target. Each view
# re-simulates 1,000 chips with the cell model and the seeds of study F (chip_population) in keygen_view below and draws
# the three panels and the card; the two-temperature enrolment uses the study's array of n_cells_fin cells, the 25 °C
# enrolment its 600 cells. Every setting of the menus was computed beforehand with this function
# (dev/explore_precompute.py) and is stored in the figures repository, so the menus respond at once, without a kernel;
# the default view, the key generator of study F, is also embedded as the plot of record.
ENR_KG = {"25": (25,), "−40+85": (-40, 85)}
def keygen_view(corner, enrol, t, cells, target):
    """One design point: draws P(key fails) of the reference chip against temperature at t, the worst case over 0-85 °C of
    the median and the 99th-percentile chip against t, and the t each chip requires; returns the card's rows and verdict."""
    q, j, e = PUF_SYS[corner], list(PUF_SYS).index(corner), ENR_KG[enrol]
    n_arr = PUF_BITS if len(e) == 1 else int(cells)
    # 1,000 chips of this corner: for each, the worst key-failure probability over 0, 25 and 85 °C at every t (pop['worst'],
    # chips x t); the reference chip separately, over the full temperature axis TT at the chosen t
    pop = chip_population(q["mu"], q["sT"], q["th"], enrol=e, n_arr=n_arr, tmax=120, seed=DEFAULTS["chip_seed"] * 100 + j)
    ref = np.array([chip_population(q["mu"], q["sT"], q["th"], enrol=e, n_arr=n_arr, n_chips=1, temps=(int(T),), tmax=120,
                                    seed=DEFAULTS["chip_seed"])["worst"][0, t] for T in TT])       # the reference chip (chip_seed)
    # the smallest t at which each chip meets the target (argmax of the first True), its 99th percentile over the chips,
    # and the fraction of chips the chosen t serves; secrecy: secret bits at t and the largest t that keeps SECRET_REQ
    wst = pop["worst"]; ok = wst <= target
    treq = np.where(ok.any(1), ok.argmax(1), wst.shape[1])                       # the t each chip requires (121: none up to 120)
    t99 = int(np.percentile(treq, 99, method="higher")); meet = float(np.mean(treq <= t))
    h = q["h_all"] if len(e) == 1 else q["xv"]["−40+85"]["h"]                    # min-entropy per cell: (95 % lower bound, estimate)
    sec = budget(h[0], t); t_lim = max([x for x in range(121) if budget(h[0], x) >= SECRET_REQ], default=-1)
    tg = np.arange(wst.shape[1]); tl = f"10<sup>{int(round(np.log10(target)))}</sup>".replace("-", "−")
    fig, axs = plt.subplots(1, 3, figsize=(W2, 3.1), gridspec_kw=dict(wspace=0.12))
    ax = axs[0]
    ax.axvspan(0, 85, color=COL["faint"], alpha=0.2, lw=0)
    ax.semilogy(TT, np.maximum(ref, 1e-16), color=COL["ink"], lw=1.4)
    ax.axhline(target, color=COL["key"], lw=0.9, ls=":"); ax.set_ylim(1e-17, 30); ax.set_xticks([-40, 0, 25, 85])
    ax.set_xlabel("Temperature (°C)"); ax.set_ylabel("P(key fails)")
    panel(ax, "a", f"Reference chip, t = {t}")
    ax = axs[1]
    if t_lim < 120:
        ax.axvspan(t_lim + 0.5, 120, color=COL["band"], alpha=0.6, lw=0, zorder=0)
        ax.text(118, 1e-10, f"fewer than {SECRET_REQ}\nsecret bits", fontsize=8, color=COL["ink"], ha="right", va="center")
    ax.semilogy(tg, np.maximum(np.median(wst, axis=0), 1e-16), color=COL["ok"], lw=1.2, ls="--", label="median chip")
    ax.semilogy(tg, np.maximum(np.percentile(wst, 99, axis=0), 1e-16), color=COL["ok"], lw=1.6, label="99th percentile")
    if t99 <= 120:
        ax.plot(t99, target, "o", ms=4.5, color=COL["ok"], mec="white", mew=0.5, zorder=6)
    ax.axhline(target, color=COL["key"], lw=0.9, ls=":"); ax.axvline(t, color=COL["ink"], lw=0.8, ls="--")
    ax.set_ylim(1e-17, 30); ax.set_xlim(0, 120); ax.set_xlabel("BCH correction t"); ax.tick_params(labelleft=False)
    ax.legend(loc="upper right", fontsize=7.5, handlelength=1.8)
    panel(ax, "b", f"1,000 chips, {sec} secret bits")
    ax = axs[2]
    ax.hist(np.minimum(treq, 121), bins=np.arange(-0.5, 122.5, 1), color=COL["ok"], alpha=0.75, lw=0)
    ax.axvline(t, color=COL["ink"], lw=0.8, ls="--")
    if (treq > 120).any():
        ax.text(0.97, 0.95, f"{int((treq > 120).sum()):,} chips:\nnone up to 120", transform=ax.transAxes, ha="right", va="top", fontsize=7.5)
    ax.set_xlim(-1, min(122, 1.4 * max(int(treq.max()), t) + 6)); ax.set_xlabel("Required t"); ax.set_ylabel("Chips")
    ax.yaxis.set_label_position("right"); ax.yaxis.tick_right()
    panel(ax, "c", f"{100 * meet:.1f} % meet the target")
    fig.subplots_adjust(left=0.085, right=0.93, bottom=0.17, top=0.9)
    show()
    rows = [("design point", f"corner {corner}; enrolment " + ("one read at 25 °C" if len(e) == 1 else "three reads at −40 and at 85 °C")
             + f"; {n_arr:,} cells; t = {t}; target {tl}"),
            (f"chips meeting {tl} at t = {t} (worst case over 0, 25, 85 °C)", f"<b>{100 * meet:.1f} %</b>; the 99th-percentile chip "
             f"requires t {'= ' + str(t99) if t99 <= 120 else '&gt; 120'}"),
            (f"secret bits at the 95 % lower bound of the min-entropy ({h[0]:.3f} bit per cell)",
             f"<b>{sec}</b> (at least {SECRET_REQ} required; t ≤ {t_lim} meets it)")]
    if len(e) == 2:
        kf = pop["kept"] / n_arr
        rows.append((f"cells kept by the enrolment (of {n_arr:,})", f"median {int(np.median(pop['kept'])):,}; "
                     f"{100 * np.mean(pop['kept'] >= PUF_BITS):.1f} % of the chips keep {PUF_BITS}; "
                     f"{int(np.ceil(PUF_BITS / np.quantile(kf, 0.001))):,} cells let 99.9 % keep {PUF_BITS}"))
    good = meet >= 0.99 and sec >= SECRET_REQ
    rows.append(("verdict", "meets both requirements (99 % of the chips, secret budget)" if good else
                 ("fails: " + ", ".join(x for x, b in (("reliability", meet < 0.99), ("secrecy", sec < SECRET_REQ)) if b))))
    return rows, good
explore("keygen", "Key-generator sizing explorer")
````

#### The chip nonce from the same cells

The chip nonce $N_\mathrm{c}$ keeps the keystream of a session fresh when an attacker replays an old session nonce $N$ on the unauthenticated downlink (A15, Section IX). It is hashed from raw power-up reads of the SRAM array, so **it is only as fresh as those reads are random**: two sessions whose reads are identical get the same $N_\mathrm{c}$, the same IV and the same keystream.

Cell $i$ powers up to 1 with probability $p_i = \Phi(m_i / s(T))$, independently at every power-up, where $m_i$ is its fixed mismatch and $s(T)$ the noise-to-mismatch ratio of the cell model above. Two reads of the same chip are identical with probability $\prod_i \left(p_i^2 + (1 - p_i)^2\right) = 2^{-H_2}$, where $H_2$ is the collision (Rényi order-2) entropy of one read. Stable cells contribute almost nothing, a cell with $p_i = 0.5$ one bit, and **hashing cannot add entropy**. With $n$ independent reads hashed together, two sessions share $N_\mathrm{c}$ with probability at most $2^{-nH_2} + 2^{-31}$, the second term being a collision of the 31-bit hash between different reads. The chip therefore power-cycles the array $n$ times at session start (the key is reconstructed from the first read) and hashes all $n$ raw reads with a volatile session index: $N_\mathrm{c} = \mathrm{hash}(w'_1 \,\Vert \cdots \Vert\, w'_n \,\Vert\, \text{session index})$, with $w'_j$ the $j$-th raw read and $\Vert$ concatenation, truncated to 31 bits.

**The evaluation runs at the behavioural level**, with the cell models of the five corners from the study above, the same 1,000 simulated chips per corner, and the 600-cell array of the working setting. Its three measurements and its assumptions are as follows.

* *Exact entropy:* $H_2$ of one read for every chip, corner and temperature; reported for the median and the 1st-percentile chip (99 % of the chips have more).
* *Monte Carlo check of the formula:* on the 60 least stable cells of the worst condition's 1st-percentile chip, where identical reads are frequent, the share of identical pairs among 100,000 pairs of simulated power-ups.
* *Direct count:* the number of distinct $N_\mathrm{c}$ in 4,000 sessions of that chip, each with one read or with $n = 64$ reads of the full array, hashed by the model's `chip_nonce`.
* *Assumptions:* successive power-ups are independent, which requires the array to be discharged between them (otherwise residual charge biases the next read towards the last one); the noise is the thermal noise of the cell model only. The power switch and the read sequencer are not simulated at transistor level.

> **Next cell.** 1. Computes $H_2$ per chip, corner and temperature. 2. Checks the formula by Monte Carlo. 3. Counts the distinct chip nonces. 4. Prints the result panel.

````python
#@title Study F, part 2: entropy of the chip nonce
# Study F, part 2: the chip nonce N_c. Two power-ups of one chip give the same N_c if their raw reads are identical
# (the hash then repeats) or by a collision of the 31-bit hash (2^-31). Cell i reads 1 with probability
# p_i = Phi(m_i / s(T)), independently per power-up, so two reads are identical with probability
# prod_i (p_i^2 + (1 - p_i)^2) = 2^-H2, H2 the collision (Renyi-2) entropy of one read; n reads give n H2 bits.
# Inputs: PUF_SYS (mu, s(T) per corner, study F), TEMPS, NC_READS, chip_nonce, rng_for, card.
# Exact H2 for 1,000 simulated chips per corner and temperature; a Monte Carlo count of identical reads checks the
# formula, and a count of N_c values over many power-ups of the worst chip compares one read with NC_READS reads.
from scipy.special import ndtr
N_NC_CHIPS, N_NC_PU = 1000, 4000
h2 = lambda p: -np.log2(p ** 2 + (1 - p) ** 2).sum(axis=-1)          # collision entropy of one read, bits
NC = {}
for j, (k, q) in enumerate(PUF_SYS.items()):
    m = q["mu"] + rng_for(DEFAULTS["chip_seed"] * 100 + j).standard_normal((N_NC_CHIPS, PUF_BITS))   # the chips of study F
    for T in TEMPS:
        p = ndtr(m / q["sT"][T])
        NC[k, T] = dict(m=m, s=q["sT"][T], h2=h2(p), diff=(2 * p * (1 - p)).sum(axis=1))
pct1 = {kT: float(np.percentile(v["h2"], 1)) for kT, v in NC.items()}   # 1st-percentile chip (99 % of chips have more)
kw = min(pct1, key=pct1.get); vw = NC[kw]
iw = int(np.argsort(vw["h2"])[N_NC_CHIPS // 100])                      # the worst condition's 1st-percentile chip
mw, sw = vw["m"][iw], vw["s"]
# Monte Carlo check of the formula on this chip's least stable cells, where identical reads are frequent enough to count
rng_nc = rng_for(DEFAULTS["seed"] + 900)
cells = np.argsort(np.abs(mw))[:60]; pc = ndtr(mw[cells] / sw)
rd = mw[cells] + sw * rng_nc.standard_normal((200_000, 60)) > 0
mc_same, mc_pred = float(np.mean(np.all(rd[0::2] == rd[1::2], axis=1))), float(np.prod(pc ** 2 + (1 - pc) ** 2))
# distinct N_c over N_NC_PU sessions of this chip, each session one read or NC_READS reads of the 600-cell array
def nc_distinct(n_reads):
    return len({chip_nonce(mw + sw * rng_nc.standard_normal((n_reads, PUF_BITS)) > 0, 0) for _ in range(N_NC_PU)})
d1, dN = nc_distinct(1), nc_distinct(NC_READS)
hc = lambda k_, T_: f"{k_} {T_} °C".replace("-", "−")
kb = max(pct1, key=pct1.get)
card(f"Study F, part 2: entropy of the chip nonce (collision entropy H<sub>2</sub> of the 600-cell read, {N_NC_CHIPS:,} chips per corner and temperature)",
     [("cells that differ between two power-ups, median chip, tt at −40 | 25 | 85 °C",
       " | ".join(f"{np.median(NC['tt', T]['diff']):.1f}" for T in TEMPS)),
      ("H<sub>2</sub> of one read, median chip: tt at −40 | 25 | 85 °C",
       " | ".join(f"{np.median(NC['tt', T]['h2']):.1f} bits" for T in TEMPS)),
      ("H<sub>2</sub> of one read, 1st-percentile chip: lowest | highest condition",
       f"{pct1[kw]:.1f} bits ({hc(*kw)}) | {pct1[kb]:.1f} bits ({hc(*kb)})"),
      (f"Monte Carlo check, {hc(*kw)}, 60 least stable cells of the 1st-percentile chip: identical reads counted | predicted",
       f"{mc_same:.2e} | {mc_pred:.2e}"),
      (f"distinct N<sub>c</sub> in {N_NC_PU:,} sessions of that chip: one read | {NC_READS} reads per session",
       f"{d1:,} | <b>{dN:,}</b>"),
      (f"{NC_READS} reads, 1st-percentile chip at {hc(*kw)}: H<sub>2</sub> and P(same N<sub>c</sub>) per pair of sessions",
       f"<b>{NC_READS * pct1[kw]:.0f} bits</b>; P ≤ 2<sup>−{NC_READS * pct1[kw]:.0f}</sup> + 2<sup>−31</sup> ≈ 2<sup>−31</sup>")],
     tone="ok" if NC_READS * pct1[kw] >= 31 and dN == N_NC_PU else "bad")
````

**Key insights.**

* *Entropy of a single read.* At tt, about 5 to 7 of the 600 cells differ between two power-ups of the median chip, and one read has $H_2 = 9.3$ bits at 25 °C (12.0 at −40 °C, 10.9 at 85 °C): two sessions would share $N_\mathrm{c}$ about once in 600. For the 1st-percentile chip $H_2$ ranges from 1.1 bits at fs, −40 °C, where the noise is lowest, to 20.1 bits at sf, 25 °C. The Monte Carlo count agrees with the formula ($4.82 \times 10^{-1}$ identical pairs counted against $4.80 \times 10^{-1}$ predicted). Key reliability and nonce freshness thus conflict: the fewer cells flip, the better the key and the worse the nonce. A one-read nonce would leave A15 open in practice (a repeated keystream after a few hundred replayed sessions at tt, 25 °C, after a few at fs, −40 °C).
* *64 reads per session.* Over 4,000 sessions of the worst chip, one read per session yields only 41 distinct chip nonces, whereas 64 reads yield 4,000, one per session. **With 64 reads the 1st-percentile chip at the worst condition has 68 bits of collision entropy, so $N_\mathrm{c}$ repeats only through the 31-bit hash, with probability about $2^{-31}$ per pair of sessions.** Power-cycling the array 64 times at session start closes A15 at every corner and temperature with the cells already on the chip: the cost is a power switch for the array, a 6-bit read counter, and 64 reads of 600 bits through the SHA-256 core that already hashes the key (76 blocks of 65 clocks, about 0.5 ms at 10 MHz), once per session. The behavioural model uses these 64 reads throughout; Section XII hashes its one simulated read.
* *Assumptions.* The bound assumes independent reads and thermal noise alone: other noise sources of real cells (random telegraph noise, supply ramp) would only add entropy, while incomplete discharge between power-ups would remove some, so the discharge time is a design requirement of the power switch. An on-chip health test (not designed here) must ensure at every session start that the reads are independent and as noisy as assumed, also against an attacker who cools the chip to lower its noise.

#### The key over the chip's lifetime

Section VII-G found how NBTI moves each cell of the extracted array, with an assumed BTI law; whether the key generator keeps its target over the chip's life depends on how those moves add up in the 600 cells of each chip. The cell model of Section VII-E carries them: a shift $\Delta_i$ of the gate of pull-up PUa moves the mismatch of cell $i$, in units of its spread, by $\Delta_i / \sigma_u(T)$, with $\sigma_u(T)$ the calibrated spread of Section VII-G. With the held value the stressed pull-up is that of the preferred side, so every cell drifts towards the threshold and the key-failure probability rises with age until the code is exceeded; with the inverted write-back every cell drifts away and the probability falls. The analysis above therefore predicts **a finite lifetime for both codes with the held value and none with the write-back**, and, since the write-back also stabilises the unstable cells, a loss of the randomness that the chip nonce draws from the same cells. Fig. 55 tests these predictions: panel (a) follows the key-failure probability of the 99th-percentile chip over the years for both codes, with the held value and with the write-back, and panel (b) the entropy of one read.

The study runs at the behavioural level and is recorded with the ageing (`python dev/spice_puf/ageing_report.py`); the settings are as follows.

* *Chips:* the 1,000 modelled chips of Section VII-E (tt, anchored to the 1,024-cell post-layout run), enrolled fresh: one read at 25 °C with t = 45, or three reads at −40 and at 85 °C with t = 17 (1,000-cell array, the first 600 kept cells), as in study F.
* *Ageing:* the law of Section VII-G, nominal and pessimistic; the stress time of each pull-up accumulates on a fine time grid with the cell's probability of reading the value that turns it on (held value) or the opposite (inverted write-back); the array powered all the time.
* *Metrics:* the key-failure probability per power-up of the 99th-percentile chip, worst case over 0, 25 and 85 °C, against years; the lifetime, the age at which it exceeds $10^{-6}$; the collision entropy $H_2$ of one 600-cell read at 25 °C (median and 1st-percentile chip), the quantity part 2 sizes the chip nonce with.

> **Next cell.** 1. Reads the key-level curves of the recorded ageing. 2. Prints the lifetimes and the entropy of one read. 3. Draws Fig. 55.

````python
#@title Fig.: study F, part 3: the key over the chip's lifetime (recorded ageing)
# Study F, part 3: the key over the chip's lifetime (recorded; ASSUMED BTI law). The ageing of Section VII-G carried to
# the key: the cell model of Section VII-E (anchored to the 1,024-cell post-layout run, the same 1,000 chips as its
# population) with each cell's mismatch shifted by (dV_a - dV_b) / sigma_u(T); computed once by
# dev/spice_puf/ageing_report.py and loaded as AGE (Section VII-G).
def _system_ageing():                                     # local names only
    st, K = AGE["settings"], AGE["key"]
    ages = np.array(st["ages"]); laws = list(st["laws"])
    life = lambda k: K[k]["life_years"]
    fy = lambda v: ("fails fresh" if v == 0 else "&gt; 100 years" if v is None else f"{8766 * v:.0f} hours" if v < 1 / 365.25
                    else f"{365.25 * v:.0f} days" if v < 1 else f"{v:.1f} years")
    at = lambda k, y: float(np.interp(y, ages, K[k]["p99"]))
    sci = lambda v: "0" if v <= 0 else (lambda m, e: f"{m:.1f} × 10<sup>{e}</sup>".replace("-", "−"))(*(lambda e: (v / 10 ** e, e))(int(np.floor(np.log10(v)))))
    h2 = lambda k, y, q="h2_median": float(np.interp(y, ages, K[k][q]))
    nfmt = lambda n: {round(1 / 6, 6): "1/6", 0.25: "1/4"}.get(round(n, 6), f"{n:g}")
    law_txt = {l: f"{1e3 * st['laws'][l][0]:.2f} mV after one year, n = {nfmt(st['laws'][l][1])}" for l in laws}
    rows = []
    for en, lab in (("25", "t = 45, 25 °C enrolment"), ("-40+85", "t = 17, enrolment at −40 and 85 °C")):
        for l in laws:
            rows.append((f"{lab}, {l} law: lifetime of 10<sup>−6</sup>, held value | inverted write-back",
                         f"<b>{fy(life(f'{l}|{en}|hold'))} | {fy(life(f'{l}|{en}|invert'))}</b>"))
        rows.append((f"{lab}, nominal law: P(key fails) of the 99th-percentile chip at 1 | 3 | 10 years, held value",
                     " | ".join(sci(at(f'nominal|{en}|hold', y)) for y in (1, 3, 10))))
    # the model counts only the time the array is powered: an array powered a fraction d of the time reaches each state
    # 1/d times later (no recovery assumed), so a held array's lifetime scales as 1/d
    lh = life("nominal|25|hold")
    if lh:
        rows.append(("array powered only while it is read, nominal law, t = 45: lifetime of the held value at a duty of 1 | 0.1 %",
                     f"{fy(lh / 0.01)} | {fy(lh / 0.001)}"))
    rows.append(("t = 45, nominal law, inverted write-back: P(key fails) of the 99th-percentile chip fresh | 1 | 10 years",
                 " | ".join(sci(at("nominal|25|invert", y)) for y in (0, 1, 10))))
    rows.append(("collision entropy of one 600-cell read at 25 °C, median chip: fresh | 10 years held | inverted",
                 f"{h2('nominal|25|hold', 0):.1f} | {h2('nominal|25|hold', 10):.1f} | {h2('nominal|25|invert', 10):.1f} bits"))
    rows.append(("the same, 1st-percentile chip",
                 f"{h2('nominal|25|hold', 0, 'h2_p1'):.1f} | {h2('nominal|25|hold', 10, 'h2_p1'):.1f} | "
                 f"{h2('nominal|25|invert', 10, 'h2_p1'):.1f} bits"))
    rows.append(("fresh t required (99th-percentile chip, worst case over 0–85 °C)",
                 f"{AGE['t99_fresh']['25']} with the 25 °C enrolment, {AGE['t99_fresh']['-40+85']} with the enrolment at −40 and 85 °C"))
    rows.append(("laws", "; ".join(f"{l}: {law_txt[l]}" for l in laws)))
    card("Study F, part 3: the key over the chip's lifetime (cell model with the recorded ageing, 1,000 chips, tt)", rows, tone="info")

    ecol = {"25": COL["ink"], "-40+85": COL["aqua"]}
    from matplotlib.lines import Line2D
    fig, axs = plt.subplots(1, 2, figsize=(W2, 3.0))
    yr = ages <= 10                                       # the first ten years, on a linear axis (age 0: fresh)
    ax = axs[0]
    for en in ("25", "-40+85"):
        for sc, ls in (("hold", "-"), ("invert", "--")):
            ax.semilogy(ages[yr], np.maximum(np.array(K[f"nominal|{en}|{sc}"]["p99"])[yr], 1e-40), ls=ls, color=ecol[en], lw=1.5)
            if "pessimistic" in laws:
                ax.semilogy(ages[yr], np.maximum(np.array(K[f"pessimistic|{en}|{sc}"]["p99"])[yr], 1e-40), ls=ls,
                            color=ecol[en], lw=0.7, alpha=0.6)
    ax.axhline(1e-6, color=COL["key"], lw=0.9, ls=":")
    ax.set_ylim(3e-41, 3); ax.set_xlim(0, 10); ax.set_yticks([1e0, 1e-10, 1e-20, 1e-30, 1e-40])
    ax.set_xlabel("Years of operation"); ax.set_ylabel("P(key fails), 99th-percentile chip")
    ax.legend([Line2D([], [], color=ecol[en], ls=ls, lw=1.5) for en in ("25", "-40+85") for ls in ("-", "--")] +
              [Line2D([], [], color=COL["faint"], lw=0.7)],
              [f"t = {t}, {sc}" for t in (45, 17) for sc in ("held", "inverted")] + ["pessimistic law"],
              fontsize=7.5, loc="center right", bbox_to_anchor=(1.0, 0.6), ncol=2, handlelength=2.2, columnspacing=1.0)
    panel(ax, "a", "Key failure against age")
    ax = axs[1]
    for sc, ls in (("hold", "-"), ("invert", "--")):
        lab = "held" if sc == "hold" else "inverted"
        ax.plot(ages[yr], np.array(K[f"nominal|25|{sc}"]["h2_median"])[yr], ls=ls, color=COL["ink"], lw=1.5, label=f"median, {lab}")
        ax.plot(ages[yr], np.array(K[f"nominal|25|{sc}"]["h2_p1"])[yr], ls=ls, color=COL["bad"], lw=1.2,
                label=f"1st percentile, {lab}")
    ax.set_xlim(0, 10); ax.set_ylim(0, 1.25 * max(np.array(K["nominal|25|hold"]["h2_median"])[yr]))
    ax.set_xlabel("Years of operation"); ax.set_ylabel("$H_2$ of one read at 25 °C (bits)")
    h, l = ax.get_legend_handles_labels()                 # two columns: median | 1st percentile
    ax.legend([h[0], h[2], h[1], h[3]], [l[0], l[2], l[1], l[3]], fontsize=7.5, loc="upper center", ncol=2, columnspacing=1.2)
    panel(ax, "b", "Randomness of one read")
    fig.tight_layout(); show()

_system_ageing()
````

**Key insights.**

* *Held value.* The key generator has little margin: fresh, the 99th-percentile chip needs t = 39 of the 45 that the 25 °C enrolment provides and t = 12 of 17 with the enrolment at −40 and 85 °C. With the array powered and holding its power-up values, the nominal law consumes that margin in 26 days (t = 45) and 23 days (t = 17); after one year the 99th-percentile chip fails $3.0 \times 10^{-4}$ of its power-ups at t = 45, after ten years $3.4 \times 10^{-2}$ (Fig. 55(a)). With the pessimistic law the target is lost after 7 hours and 8 hours. **As predicted, the held value gives both codes a finite lifetime, and a short one.**
* *Inverted write-back.* The same shifts drive every cell away from the threshold, and the key-failure probability falls with age instead: at t = 45 from $6.2 \times 10^{-14}$ fresh to $2.6 \times 10^{-50}$ after ten years (nominal law), and below its fresh value at every age for both codes and both laws. **Writing the inverse back after each read is the remedy that ageing requires.**
* *Power gating.* The model counts only the time a pull-up is on, so an array powered only while it is read, a fraction d (its duty) of the time, reaches each state 1/d times later: at a duty d of 1 % the held value keeps t = 45 for 7.0 years, at 0.1 % for 70.0 years. Power gating alone is therefore enough only at a small duty, which a sensor that reads its key at power-up and then powers the array down can reach.
* *The chip nonce.* Ageing moves the randomness of a read the other way: the collision entropy of one 600-cell read at 25 °C of the median chip (11.8 bits fresh with this anchoring, against 9.3 bits in part 2, whose model is anchored to the 512-cell PVT record) rises to 20.8 bits after ten years with the held value but falls to 4.4 bits with the write-back, and that of the 1st-percentile chip to 1.0 bits. The write-back that keeps the key reliable therefore starves a nonce drawn from the same cells: **a design with the write-back needs a separate array for the nonce, held or powered down**, consistent with part 2, which found that the key and the nonce need opposite cells.
* *Limits.* The law is assumed (Section VII-G); recovery is not modelled, and it would lengthen the held-value lifetimes; the array is taken as powered all the time; the study covers tt only.

# **IX. Security evaluation**
---
Section VIII showed that no analog quantity protects the original design against the replica attacker and that the proposed design's security rests on its key, so both designs must be judged against the whole threat model, not the replica attack alone. This section answers the first research question in full: **what security does each design provide?** It runs the 15 attacks A1 to A15 of Section IV-A (Table 3), with the attacker's capabilities C1 to C5, on the bit-exact model of both designs (Section V, consistent with [1], Section VI) with the SKY130-derived parameters (Table 12, Section VII). The threat model is wider than that of [1], an eavesdropper with a replica chip: it also admits an attacker who holds the chip or transmits on the link. The verdicts of A4 and of the single-frame A6 rest on how strongly the two chips' bit streams decorrelate under the model's simplified dither of [1] (Section VI); that of A5 follows from the repeated keystream.

Subsection A **runs every attack on both designs at the current settings** and judges each run by a decision rule fixed in advance, one verdict per attack and design; it then shows where in the link three of the attacks succeed or are stopped, offers any attack interactively with switches for the defences, draws every verdict in one still figure that needs no running kernel, and determines whether an attacker who modifies the helper data and observes whether the receiver accepts can learn the fingerprint bit by bit (A12). Subsection B computes the probability that the PUF fails to regenerate the key at a power-up, exactly over temperature for one modelled chip and 16 others, checks it by Monte Carlo and counts the secret bits the code leaves. Subsection C measures how far the operating point of Subsection A lies from the boundary at which a verdict changes, over a grid of the two parameters that govern each of three experiments. Subsection D measures how much the two keystream generators expose their key to power analysis, which the threat model leaves out of scope (Section IV-A), so that this limitation is quantified rather than only stated.

### A. Attack evaluation: 15 attacks on both designs

*Notation* (Section V). The chip transmits $C = D \oplus  K$, with $D$ the modulator bit and $K$ the keystream bit of the 16-bit LFSR (original) or Grain-128a (proposal); the attacker's estimates are $\hat{K}$ and $\hat{D} = C \oplus  \hat{K}$. The Berlekamp-Massey algorithm (BM) [3] determines an $L$-bit register from $2L$ consecutive correct bits, 32 for the 16-bit LFSR. In the proposed design, $w$ is the enrolled SRAM fingerprint and $h$ its public BCH syndrome ($10t$ bits, with $t$ the number of flipped cells the code corrects), the key $k$ is the hash of the corrected fingerprint, and the IV (session nonce, chip nonce $N_\mathrm{c}$, counter) gives every frame its own keystream.

*Capabilities and decision tests.* The capabilities are those of Section IV-A; every run is judged by the test fixed for its attack in Section V-G (Table 5), applied by `run_attack`: the keystream test (A2 to A4, A6, A7; A15 on the recovered reading), the receiver test (A8 to A10, A13, A14) or a leakage test (A5, A11, A12). **A1 is not run: its verdict holds by design**, from the encryption and the key length (Table 21). NOT APPLICABLE marks the helper-data attacks (A10 to A12) on the original design, which has no helper data.

*The 15 attacks and their predictions.* Table 5 lists what the attacker does in each; four come from [1] (A4 and A5 evaluated there, A1 and A8 named as threats). **Each prediction is stated before the run**, with its reasoning and assumptions:

* **A1. Passive eavesdropping** *[1]* (C1; keystream test). Decryption with a wrong key (the original design's key with 0x5A5A added bit-wise, or a random 128-bit key): safe for both designs.
* **A2. Brute-force key search** (C1; keystream, over the whole frame). The chip ID is not secret, so the encrypted ID reveals the first 16 keystream bits, and exactly one of the 65,535 non-zero seeds matches them: the original design fails (in this frame model the first 16 LFSR outputs are the seed itself; without such a known field the search runs against any 16 known keystream bits, A3, in well under a millisecond, Section II-B). The proposed design's 2¹²⁸ ≈ 3.4 × 10³⁸ keys cannot be searched; the model represents any feasible search by one random guess, so this verdict holds by design.
* **A3. Berlekamp-Massey keystream recovery** (C3; keystream, after the 64 known bits). $L = 16$ and the 64 known bits exceed $2L$, so the original design fails; for Grain-128a, BM returns a register of about 32 bits that predicts no better than chance.
* **A4. Zero/DC input with a replica chip** *[1]* (C2; keystream). The replica differs from the victim by an input-referred offset of 0.1 % of full scale (FS), its own dither and noise, and, with the replica's bits $D_\mathrm{r}$, $\hat{K} = C \oplus D_\mathrm{r} = K \oplus (D \oplus D_\mathrm{r})$. After BM on 80 random windows (`window_attack`), the attacker screens every position of $\hat{K}$ (`lfsr16_scan`): the LFSR polynomial is public, so any 16 correct bits are its state; a candidate must predict the following 1,024 bits better than chance (at least 62 %), and, for up to 400 candidates, its seed must decrypt the chip ID, a test that a wrong seed passes with probability 2⁻¹⁶; the scan is repeated with the replica's input trimmed (C2) until a seed is confirmed. The two chips' bits agree in runs (Section VIII-B), so **the original design fails**. Against the proposed design, even a $\hat{K}$ right on more than 97 % of the bits reads only the forced frame, which carries nothing new, and the per-frame IV protects every other frame: DATA SAFE.
* **A5. Same key, same message** *[1]* (C1; leakage). Two independent bit streams with the same input $u$ differ with probability $(1 - u^2)/2$, so with a repeated keystream the fraction of ones in $C_1 \oplus C_2 = D_1 \oplus D_2$ (subscripts: the two frames) is $(1 - \langle u^2\rangle)/2$ and reveals the mean square of the reading; for a sine of amplitude $A$ the attacker estimates $\hat{A} = \sqrt{2\,(1 - 2\langle C_1 \oplus C_2\rangle)}$. **The original design leaks the amplitude**, however well $D_1$ and $D_2$ decorrelate (its LFSR restarts in every frame, as assumed in A6); in the proposed design the second frame carries the next counter value and $C_1 \oplus C_2$ is uniformly random.
* **A6. Forced-input saturation** (C2; keystream). At 10 µA the S2D buffer gain of 0.5 limits the input to half of full scale, so the bits keep toggling (three of four are ones) and the single-frame guess fails against both designs. The majority of $C$ over 21 frames at about 1.7 µA (input about 0.45 FS), stepped by 5 nA, recovers $K$ if the LFSR restarts from the stored key in every frame (assumed; [1] does not state it), and the scan of A4 confirms the seed: **the multi-frame attack breaks the original design**; under the proposed design's per-frame IV the majority carries no information. Subsection C sweeps the buffer gain.
* **A7. Key extraction from memory** (C4; keystream). The original design fails if its 16-bit key is held in non-volatile on-chip memory, which [1] does not describe (Section II-E); a key loaded from a host would move the attack to that host, and the key still falls to A2 and A3. The proposed design's key exists only in volatile registers while powered, so the attacker is reduced to a guess, by design, provided the enrolment read-out is disabled after test (Section IV-A); probing, laser fault injection and the cloning of an SRAM fingerprint by photonic emission analysis [21] are outside the threat model.
* **A8. Spoofing** *[1]* (C5 after A2; receiver). The receiver of the original design accepts a frame encrypted with the brute-forced key; the proposed design rejects a guessed key, which passes the 32-bit tag with probability 2⁻³².
* **A9. Cloned or counterfeit sensor** (C4 + C5; receiver). A counterfeit of the original design loaded with the extracted key is accepted; a clone of the proposed design has a different fingerprint and, with the genuine helper data, a different key, so the ID and MAC checks fail (assuming the fingerprint cannot be copied).
* **A10. Helper-data tampering** (C5). One inverted bit of $h$ makes the hash yield a different key, which fails the ID and MAC checks: detected.
* **A11. Helper-data leakage** (C1; leakage). With $H_\infty$ the min-entropy of one cell in bits, at least $600\,H_\infty - 10t$ fingerprint bits remain secret (`puf_residual`), against 192 required at the 95 % lower bound of the min-entropy (Section VII-E); at t = 45 the budget lies below it: KEY WEAKENED (analytic, resting on the bias estimate).
* **A12. Helper-data manipulation oracle** (C5; leakage). The answers to 200 chosen changes of $h$ do not depend on the fingerprint (analysed in detail below; analytic).
* **A13. Replay** (C1 + C5; receiver). A frame resent after the reading has fallen to 0.4 times its value is accepted by the receiver of the original design, which has no counter, and rejected by the proposed design for its stale counter.
* **A14. Bit-flipping in transit** (C5; receiver). Payload bits inverted with probability 0.1 pass the receiver of the original design, which checks only the ID; the proposed design's MAC detects them.
* **A15. Keystream reuse through a replayed session nonce** (C1 + C2 + C5; keystream, on the recovered reading). The session nonce travels in clear on the downlink; replaying it in 11 later sessions while forcing about 1.7 µA, the attacker takes the majority of $C$ as $\hat{K}$ and decrypts the recorded frame. The original design fails (keystream restarted in every frame, as in A6), and so would a proposal whose IV held only the session nonce and the counter, which restarts at each session; the chip nonce $N_\mathrm{c}$, hashed from 64 power-up reads, gives every session its own keystream (shared with probability about 2⁻³¹; Section VIII-F), so **the proposed design is safe**. The proposed design is also run without $N_\mathrm{c}$ (`chip_nonce=False`) to expose the flaw it closes.

Overall, the analysis above predicts that **eleven of the 15 attacks succeed against the original design and, at the working setting t = 45, one (A11) against the proposed design.**

The runs use the following settings:

* *Level.* Behavioural, bit-exact model (Section V) with the post-layout values of Table 12 at the nominal corner; Section VIII covers PVT.
* *Frame.* One frame per design; input the test sine of Section V-A unless the attack sets it.
* *Design settings* (`DEFAULTS`). The dither of [1]; S2D buffer gain 0.5; for the proposed design, $t = 45$, counter and MAC on, and the PUF read at the operating temperature of 37 °C after enrolment at 25 °C.
* *Attacker settings:* the values stated with each attack above (`ATK_DEF`, Section V-A; the single-frame 10 µA of A6 is `force_uA` of `DEFAULTS`).
* *Seeds and repetitions.* All streams derive from `seed` = 2026, so the verdicts of A4, A5, A6 and A15 are single samples; Subsection C and Section VIII vary the replica and single-frame saturation attacks, while A5, A15 and the multi-frame A6 are not repeated.
* *Metric.* The verdicts and the number of successful attacks per design.

> **Next cell.** 1. Runs the 15 attacks on each design (`verdict_table()`). 2. Prints Table 21 and a result panel with the details of A4, A5, A6 and A15, including the proposed design without its chip nonce. 3. Defines `attack_figure` for the figures below.

````python
#@title Attack evaluation: verdict table
# Attack lab, part 1: the verdict table of all 15 attacks on both designs, the detail card of the attacks A4, A5, A6 and
# A15, and attack_figure(), which the next cell (the interactive figure) and later cells call.
# Inputs: simulate, verdict_table, ATTACKS, VERDICT, EVIDENCE, PROPERTY (dev/model/secsensor.py); plot_chain,
# chain_items, chain_ratios, card, COL, W2, ASSETS (dev/model/analysis.py). Live: every verdict is a fresh simulation.
# Defines for later cells: T (the default verdict table), verdict_html, attack_figure, GOOD, STOPS, ICON.
import ipywidgets as W
if IN_COLAB:                                  # Colab renders third-party widgets only with its custom manager
    from google.colab import output as _colab_output
    _colab_output.enable_custom_widget_manager()

GOOD = {"safe", "detected"}                  # verdicts in which the design holds
# the mechanism that defeats each attack in the proposal (last column of the verdict table)
STOPS = dict(eaves="encryption (both)", brute="128-bit PUF key", bm="Grain-128a's nonlinearity", knownin="Grain-128a; per-frame IV",
             samekey="per-frame IV", saturate="per-frame IV", extract="no stored key (PUF); read-out disabled",
             spoof="unrecoverable key + MAC", clone="unclonable PUF", htamper="key = hash(w) + MAC", hleak="n·H∞,lo − 10t ≥ 192",
             hmanip="syndrome construction + hash", replay="frame counter + MAC", bitflip="MAC over the payload",
             reuse="chip nonce N<sub>c</sub> in the IV")

# face icons from the figures repo: happy = design holds, sad = attack succeeds, neutral = not applicable
ICON = {k: f'<img src="{ASSETS}/icons/{f}.svg" width="18" height="18" style="vertical-align:-4px;margin-right:6px" alt="{k}">'
        for k, f in (("good", "happy"), ("bad", "sad"), ("na", "neutral"))}

def verdict_html(T):
    """The verdict table: booktabs rules, a face per verdict (happy: safe or detected, sad: the attack succeeds)."""
    def cell(v):
        kind = "good" if v in GOOD else "na" if v == "na" else "bad"
        return f'<td style="padding:4px 14px;white-space:nowrap">{ICON[kind]}{"<b>" if kind == "bad" else ""}{VERDICT[v]}{"</b>" if kind == "bad" else ""}</td>'
    ev = lambda a: f'<td style="padding:4px 10px;opacity:.8;white-space:nowrap">{EVIDENCE[a][0]} / {EVIDENCE[a][1]}</td>'
    rows = "".join(f'<tr><td style="padding:4px 14px">A{k + 1}. {name}{" <i>[1]</i>" if paper else ""}</td>'
                   f'<td style="padding:4px 10px;opacity:.8">{PROPERTY[a]}</td>{cell(T[a][0])}{cell(T[a][1])}{ev(a)}'
                   f'<td style="padding:4px 14px;opacity:.8">{STOPS[a]}</td></tr>' for k, (a, name, paper) in enumerate(ATTACKS))
    # number of successful attacks for design k (0 original, 1 proposal): neither held nor not applicable
    bad = lambda k: sum(T[a][k] not in GOOD | {"na"} for a in T)
    th = lambda s: f'<th style="text-align:left;padding:5px 14px;border-bottom:1px solid currentColor">{s}</th>'
    return (f'<div style="font-family:\'STIX Two Text\',\'Times New Roman\',Times,serif;font-size:15px;max-width:980px;margin:8px 0">'
            f'<div style="text-align:center;font-variant:small-caps;letter-spacing:.04em;margin-bottom:4px">Table 21: verdict of every attack on both designs '
            f'(<i>[1]</i>: from [1]; A4, A5 evaluated there, A1, A8 named)</div>'
            f'<table style="border-collapse:collapse;border-top:1.6px solid currentColor;border-bottom:1.6px solid currentColor">'
            f'<thead><tr>{th("Attack")}{th("Property")}{th("Original S-DSM [1]")}{th("Proposed")}{th("Evidence (original / proposed)")}{th("What stops it")}</tr></thead><tbody>{rows}</tbody></table>'
            f'<div style="margin-top:6px">Attacks that succeed: <b>original {bad(0)} of {len(T)}</b>, <b>proposed {bad(1)} of {len(T)}</b> (t = 45; the proposal resists '
            f'{sum(T[a][1] == "safe" for a in T)} and detects {sum(T[a][1] == "detected" for a in T)}). '
            f'Evidence: <i>simulated</i>, the attack is run on the model; <i>analytic</i>, a formula, an identity or a design argument (A4, A6 of the proposal: the estimate is simulated, the verdict follows from the per-frame IV); '
            f'<i>by design</i>, the verdict follows from the design without an attack simulation (A1, A2, A7, A9 of the proposal: the link is encrypted, '
            f'the key is 128 bits long, no key is stored, the key comes from the chip\'s own PUF; physical attacks are outside the threat model); '
            f'<i>assumed</i>, a property of [1] taken as given (A1 and A7 of the original). By property, the original fails ' +
            # per security property (in order of first appearance): failures of the original out of the applicable attacks
            ", ".join(f"{p} {sum(T[a][0] not in GOOD | {'na'} for a in T if PROPERTY[a] == p)} of {sum(1 for a in T if PROPERTY[a] == p and T[a][0] != 'na')}"
                      for p in dict.fromkeys(PROPERTY[a] for a, _, _ in ATTACKS) if any(PROPERTY[a] == p and T[a][0] != 'na' for a in T)) +
            '.</div></div>')

# default settings: proposal with frame counter + MAC, dither on, S2D buffer gain 0.5
T = verdict_table()
VERDICTS = T                                # kept under its own name for the headline check at the end
display(HTML(verdict_html(T)))
# the attacks that the competent attacker changes, and the proposal without its chip nonce
# A4: known (zero / DC) input with a replica chip; A5: two frames under the same key; A6: forced saturation;
# A15: replayed session nonce. The proposal is run for A15 with and without the chip nonce N_c in the IV, to show that
# N_c is what keeps the keystream from repeating.
_a4, _a5 = simulate(scheme="baseline", attack="knownin")["atk"], simulate(scheme="baseline", attack="samekey")["atk"]
_a6, _a15 = simulate(scheme="baseline", attack="saturate")["atk"], simulate(scheme="baseline", attack="reuse")["atk"]
_p15, _p15n = simulate(attack="reuse")["atk"], simulate(attack="reuse", chip_nonce=False)["atk"]
card("Attacks A4, A5, A6 and A15 in detail",
     [("A4, original: 80 random windows / full scan", f"{_a4['win']['broken']} of 80 windows; full scan: seed confirmed "
       f"at a replica trim of {_a4.get('scan_trim', 0):+.2f} % FS ({_a4.get('scan_cand', 0)} candidate states)" if "scan_seed" in _a4 else
       f"{_a4['win']['broken']} of 80 windows; full scan: no seed confirmed"),
      ("A5, original: fraction of ones in C₁ ⊕ C₂, amplitude estimate", f"{_a5['xfrac']:.4f}; Â = {_a5['a_hat']:.3f} against A = {_a5['a_true']:.3f} FS"),
      ("A6, original: majority over frames", f"{_a6.get('frames', 0)} frames, keystream estimate {100 * _a6.get('mf_kacc', 0):.1f} % correct; "
       + ("seed confirmed" if "scan_seed" in _a6 else "no seed confirmed")),
      ("A15, original: reading recovered from replayed sessions", f"correlation {_a15['corr']:.2f} with the true reading ({_a15['sessions']} sessions)"),
      ("A15, proposal with the chip nonce", f"correlation {_p15['corr']:+.2f}; {_p15['cn_distinct']} distinct chip nonces in {_p15['sessions']} sessions"),
      ("A15, proposal without the chip nonce", f"correlation {_p15n['corr']:.2f}: <b>{VERDICT[_p15n['v']]}</b>")],
     tone="bad" if _p15["v"] != "safe" else "ok")    # orange only if the proposal with its chip nonce fails A15

from matplotlib.lines import Line2D

def attack_figure(attack, mac=True, dither=True, buffer=0.5):
    """Both designs under one attack, side by side, or under several attacks, one row of panels each: the bits that show
    what the attack does over 200 clocks, grouped by where they are observed, each group headed by what happens there;
    then the readings. Returns the simulated frames, (original, proposed) per attack."""
    # attack: one attack id (str) or a sequence of ids; mac, dither, buffer override the model defaults for both designs
    attacks = (attack,) if isinstance(attack, str) else tuple(attack)
    over = dict(mac=mac, dither=dither, buffer=buffer)
    rows = []
    for a in attacks:
        # both designs under the same attack and settings; ratios = height ratios of the rows each frame needs
        sims = [simulate(scheme=sc, attack=a, **over) for sc in ("baseline", "proposed")]
        ratios = [chain_ratios(S) for S in sims]
        n = max(len(r) for r in ratios)
        ratios = [r[:-1] + [0.0] * (n - len(r)) + r[-1:] for r in ratios]      # same rows in both panels
        # common row heights, so that the rows of the two panels line up
        hr = [max(a_, b_) for a_, b_ in zip(*ratios)]
        rows.append((a, sims, n, hr, 0.27 * sum(hr) + 0.97))                    # inches: rows + titles and the x axis
    leg_h = 0.42
    fig = plt.figure(figsize=(W2, sum(r[-1] for r in rows) + leg_h))
    subs = fig.subfigures(len(rows) + 1, 1, height_ratios=[r[-1] for r in rows] + [leg_h], hspace=0.0)
    names = dict((x, nm) for x, nm, _ in ATTACKS)
    for k, ((a, sims, n, hr, h), sub_row) in enumerate(zip(rows, subs)):
        # thin rules between the two designs and between the attacks (journal style: no tinted panels)
        sub_row.add_artist(Line2D([0.5, 0.5], [0.03, 0.97], transform=sub_row.transSubfigure, color=COL["grid"], lw=0.8))
        if k:
            sub_row.add_artist(Line2D([0.02, 0.98], [0.998, 0.998], transform=sub_row.transSubfigure, color=COL["grid"], lw=0.8))
        for c, (sub, S, name) in enumerate(zip(sub_row.subfigures(1, 2, wspace=0.0), sims, ("original S-DSM", "proposed"))):
            # n rows were allocated (the larger of the two designs); the unused rows above the reading are removed
            items = len(chain_items(S))
            axs = sub.subplots(n, 1, gridspec_kw=dict(height_ratios=hr, hspace=0.25))
            for ax in axs[items:-1]:
                ax.remove()
            sub.subplots_adjust(left=0.34, right=0.95, top=1 - 0.58 / h, bottom=0.42 / h)
            plot_chain(S, axs=list(axs[:items]) + [axs[-1]], legend=False)
            v = S["atk"]["v"]
            good, na = v in ("safe", "detected"), v == "na"
            col = COL["faint"] if na else COL["ok"] if good else COL["bad"]
            # panel title: (a), (b), ... across the row; with several attacks it also names the attack (short form). The
            # outcome is the line under it, in the outcome colour (orange: the attacker gains, blue: the design holds)
            title = (f"({'abcdefgh'[2 * k + c]}) " + (f"{names[a].split(':')[0].split(' of ')[0]}: {name}" if len(rows) > 1 else name[0].upper() + name[1:]))
            sub.text(0.645, 1 - 0.08 / h, title, fontsize=10, va="top", ha="center")          # centred over the rows
            sub.text(0.645, 1 - 0.32 / h, VERDICT[v], fontsize=8.5, va="top", ha="center", color=col)
    hl = {}                                    # one legend for all panels, under the figure
    for ax in fig.axes:
        for hd, l in zip(*ax.get_legend_handles_labels()):
            hl.setdefault(l, hd)
    if "attacker's forged reading" in hl and "old reading replayed" in hl:     # same line style: one entry
        hl["attacker's reading: forged (a, b) or replayed (c, d)"] = hl.pop("attacker's forged reading"); del hl["old reading replayed"]
    subs[-1].legend(hl.values(), hl.keys(), loc="center", ncol=min(4, len(hl)), frameon=False, fontsize=9,
                    columnspacing=1.2, handlelength=1.8)
    plt.show()
    return [r[1] for r in rows] if len(rows) > 1 else rows[0][1]
````

**Key insights.** Every simulated verdict of Table 21 agrees with its prediction; the others agree by construction.

* **Against the original design, 11 of 15 attacks succeed.** Six breach the data (A2, A3, A4 through the full scan at the trim given in the result panel, and, under the assumptions stated with them, A6, A7 and A15), A5 leaks the amplitude, and four place fake data on the server (A8, A9, A13, A14). Only A1 fails; A10 to A12 are not applicable. The failures have three architectural causes: a short, linear and stored key (A2, A3, A7); a keystream without nonce, so any repetition of message, forced input or session exposes it (A5, A6, A15); and frames without authentication or freshness (A8, A9, A13, A14). The analog defences against A4 to A6 fail once the attacker scans every window, trims the replica's input and repeats frames. Six of these 11 failures (A7, A8, A9, A13, A14 and A15) need an attacker who holds the chip or transmits on the link, so they show which protections the proposed design adds, not failures against the attacker of [1], who breaks the original design only with A2 to A6.
* **Against the proposed design, 1 of 15 attacks succeeds at the working setting t = 45: the helper-data leakage (A11, KEY WEAKENED), with 94 secret bits at the 95 % lower bound against the 192 required.** Five are detected (A8, A9, A13, A14 by the receiver; A10 through a different key); the other nine are DATA SAFE. With the one-time enrolment at −40 and 85 °C of Section VIII-F, A11 is also DATA SAFE (analytic), so in the model the proposed design stops all 15, resisting ten and detecting the other five; for A10 to A12 this rests on the cell mask of that enrolment being hashed into the key, which the model does not implement (Section IV-B). Without the chip nonce, the proposed design falls to A15 (result panel below the table: correlation 0.95 with the true reading, against −0.03 with the chip nonce), as predicted.
* The proposed design's verdicts rest on three kinds of evidence (column "Evidence"). Seven are simulated. Four are analytic: A4 and A6 (keystream estimate simulated, verdict from the per-frame IV), A11 (the secret-bit budget) and A12 (an identity of the syndrome construction). Four are by design: eavesdropping and brute force rest on the 128-bit key, extraction on the absence of a key in non-volatile memory and of a usable read-out path, and the clone on the unclonability of the fingerprint. The assumption on the 128-bit key also covers the published cryptanalysis of Grain-128a: its fast correlation attacks need $2^{113.8}$ [18] and $2^{106.3}$ [19] keystream bits, far beyond one frame, and the attack of [18] excludes the authenticated mode used here. The verdicts hold for the attacks as implemented, not as a proof of security; physical attacks (Section IV-A) are outside the threat model, and Subsection D measures how much each keystream generator exposes its key to power analysis.

#### A closer analysis: the helper-data manipulation attack

In the helper-data manipulation attack (A12), the attacker modifies the helper data and observes whether the receiver accepts, trying to learn $w$ bit by bit. **Such attacks have recovered keys from PUF key generators whose helper data is not a linear-code syndrome**, such as pattern-matching [23] and ring-oscillator pairing schemes (Delvaux and Verbauwhede, DATE 2014). The analysis covers the syndrome, the helper data of the simulations' 25 °C enrolment; the cell mask of the one-time enrolment at −40 and 85 °C is pointer-type helper data of the kind these attacks exploit, and must be hashed with $w$ into $k$ [24] (Section IV-B; not modelled).

With the syndrome construction [4], $h = Hw$, where $H$ is the parity-check matrix of the BCH code. At power-up the chip reads $w' = w \oplus \epsilon$, where $\epsilon$ marks the cells that flipped. The decoder receives the syndrome $Hw' \oplus  h = H\epsilon$ and corrects $\epsilon$. If the attacker sends $h \oplus  He$ instead, with an error pattern $e$ of the attacker's choice, the decoder receives $H(\epsilon \oplus e)$, so whether decoding succeeds depends only on $\epsilon$ and $e$, never on $w$. When it succeeds, it returns $w \oplus  e$, whose hash is a wrong key, so the frame is rejected. The analysis above therefore predicts that (1) **no tampered power-up is accepted**; (2) the decoder fails when $e$ and the noise together flip more than $t$ cells; and (3) a chip with a different fingerprint and the same noise gives the same answer to every query, so the answers carry no information about $w$.

The test uses `manip_oracle` (Section V-E), read at 37 °C, $t = 45$, with $Q = 200$ queries (`manip_q`), query $q$ adding the syndrome of a random pattern of weight $1 + (q \bmod 55)$, up to $t + 10 = 55$. Each query is a new power-up; a twin chip with a random fingerprint $w_2$ and its own helper data receives the same noise and added syndrome. Seeds from `seed` + 57. The metrics are the accepted power-ups, the decoder failures and the agreeing answers; **the verdict is DATA SAFE only if all 200 agree**.

> **Next cell.** 1. Runs A12 on the proposed design (`simulate(attack="hmanip")`) with the twin chip. 2. Reports the counts and the verdict.

````python
#@title Helper-data manipulation
# helper-data manipulation: does tampering with h and watching accept/reject reveal the fingerprint?
# One frame of the proposal under attack A12: the attacker adds the syndrome of an error pattern of its choice to the
# helper data h and records whether the chip still reconstructs its key (accept or reject), Q times.
Sm = simulate(attack="hmanip")
# M: the number of queries Q, accepts, decoder failures, and 'same', the number of answers that a twin chip with a
# different fingerprint (but the same power-up noise) gives identically; if every answer is the same, the answers carry
# no information about the fingerprint
M = Sm["manip"]
card("Helper-data manipulation attack",
     [("tampered power-ups", f"{M['Q']} (chosen error patterns added to <i>h</i>)"),
      ("receiver accepted / decoder failed", f"{M['acc']} / {M['fails']}"),
      ("same answer from a chip with a different fingerprint", f"{M['same']} of {M['Q']}"),
      ("verdict", "<b>no information about the fingerprint leaks</b>" if M["same"] == M["Q"] else "<b>some information leaks</b>")],
     tone="ok" if M["same"] == M["Q"] else "bad")
````

The oracle thus measures only the noise and the attacker's own pattern: **by the linearity of the syndrome, the syndrome construction and the hashed key defeat this attack in the model without any additional circuit**; the twin-chip comparison illustrates this identity rather than testing it, and assumes that which cells flip is independent of their enrolled values.

### B. PUF reliability: probability of key failure per power-up

**The proposed design's one new failure mode is a power-up at which the PUF does not regenerate the key**, so the chip cannot communicate until the next power-up. With the cell model of Section V-D (mismatch $m_i$, noise-to-mismatch ratio $s(T)$), cell $i$ differs at temperature $T$ from the bit $w_i$ enrolled at 25 °C with probability

$$p_i(T) = \Phi\!\left(-\frac{m_i}{s(T)}\right) \text{ if } w_i = 1, \qquad p_i(T) = \Phi\!\left(\frac{m_i}{s(T)}\right) \text{ if } w_i = 0,$$

where Φ is the standard normal distribution function. The number of flipped cells $N$ is Poisson-binomial, and the key fails when more than $t$ cells flip:

$$P_{\text{fail}}(T) = P(N > t) = \sum_{k > t} P(N = k).$$

`fail_tail` and `err_dist` evaluate it exactly. Only cells with $\lvert m_i\rvert$ comparable to $s(T)$ can flip; away from 25 °C, $s(T)$ grows and the tail beyond $t$ grows by many orders of magnitude. Secrecy sets the opposing constraint: the helper data leave at least $600\,H_\infty - 10t$ secret bits, against 192 required at the 95 % lower bound of $H_\infty$ (A11), so **reliability and secrecy set opposite limits on $t$**.

The model assumes independent cells and Gaussian, thermal-only noise, so the noise-to-mismatch ratio at 25 °C, $s_{25}$, likely underestimates silicon (Table 12); the PDK mismatch is not verified in silicon, and ageing is excluded (`puf_years` = 0; Sections VII-G and VIII-F estimate it with an assumed BTI law). Temperature acts only through $s(T)$; the reproducible flips are not modelled (Sections VII-E and VIII-F size the code with the two-component model). **Probabilities far below 10⁻¹⁰ describe the model's Gaussian tail, not silicon**, and are given to one significant figure.

The analysis above predicts that (1) $P_\mathrm{fail}$ **is lowest at the enrolment temperature of 25 °C and increases on both sides**; (2) with $t = 45$, the linear law keeps $P_\mathrm{fail}$ far below 10⁻⁶ over 0–85 °C, which is optimistic, because the law underestimates the simulated error rate at 85 °C (1.75 % against 3.14 %, Table 12); (3) the Monte Carlo histogram of $N$ follows the exact distribution; and (4) the secret-bit budget is that of Section VII-E, short of the 192 required. Fig. 62 tests predictions (1) and (2) in panel (a), by the key-failure probability against temperature for this chip and 16 others, and prediction (3) in panel (b), by the Monte Carlo histogram of the flipped cells against the exact distribution.

The computation uses the following settings:

* *Level and model.* Behavioural `SRAMPUF` model fitted to the post-layout Monte Carlo of the extracted 6T cell (the values of Table 12), nominal corner, linear law.
* *Chips.* "This chip" has seed 7 (`chip_seed`); sixteen other chips, seeds 107 to 1,607, are enrolled in the same way at 25 °C.
* *Codes.* $t = 45$ (Section VII-E) for every chip; this chip also with $t$ − 10 = 35 and $t$ + 10 = 55.
* *Temperature.* −40 to 85 °C in steps of 5 °C, the range over which the SRAM-PUF cell is simulated; the specified range is 0–85 °C.
* *Monte Carlo check.* 4,000 power-ups of this chip at the operating temperature of 37 °C (`temp`, the bioreactor), each compared bit by bit with the enrolled fingerprint, seeded from `seed` + 909; the exact distribution of $N$ at 37 °C is computed for comparison. With no failure among $M$ power-ups, $P_\mathrm{fail} < 1 - 0.05^{1/M} \approx 3/M$ at 95 % confidence: the sample tests the body of the distribution, not its tail.
* *Metrics.* $P_\mathrm{fail}$ against 10⁻⁶; the mean and maximum flipped cells; the secret bits.

> **Next cell.** 1. Defines `puf_reliability()` (exact key-failure probability over temperature, Monte Carlo check, flipped-cell distribution). No simulation is run.

````python
#@title PUF reliability: definitions
def puf_reliability(P=None, temps=np.arange(-40, 86, 5), n_chips=16, n_mc=4000):
    """Exact key-failure probability per power-up against temperature (this chip and others), plus a Monte
    Carlo check of the error-count distribution at the current temperature."""
    P = {**DEFAULTS, **(P or {})}; age = puf_age(P)
    def chip(seed):
        puf = SRAMPUF(seed, PUF_BITS, P["puf_sigma"], P["puf_bias"], P["puf_tc"])
        return puf, enroll(puf, P["bch_t"], rng_for(seed * 7919 + 1))["w"]
    # this chip at t - 10, t and t + 10; 16 other chips at t (their spread is the band); a Monte Carlo of the error count
    puf, w = chip(P["chip_seed"])
    ts = [t for t in (P["bch_t"] - 10, P["bch_t"], P["bch_t"] + 10) if t >= 1]
    curves = {t: np.array([fail_tail(puf.p_err(w, T, age), t) for T in temps]) for t in ts}
    others = np.array([[fail_tail(o.p_err(wo, T, age), P["bch_t"]) for T in temps]
                       for o, wo in (chip(P["chip_seed"] + 100 * c) for c in range(1, n_chips + 1))])
    r = rng_for(P["seed"] + 909); errs = np.array([int((puf.read(P["temp"], r, age) != w).sum()) for _ in range(n_mc)])
    p_now = fail_tail(puf.p_err(w, P["temp"], age), P["bch_t"])
    return dict(temps=temps, curves=curves, band=(others.min(0), others.max(0)), errs=errs,
                dist=err_dist(puf.p_err(w, P["temp"], age)), p_now=p_now, t=P["bch_t"], temp=P["temp"])
````

> **Next cell.** 1. Runs `puf_reliability()`. 2. Prints the result panel that Fig. 62 plots.

````python
#@title PUF reliability: key failure probability and flipped cells
# Key reliability of one modelled chip (the cell model with the SKY130-fitted parameters): puf_reliability computes,
# exactly, the probability that more than t cells flip (Poisson-binomial tail) over temperature for t = 35, 45, 55, the
# full distribution of flipped cells at the operating temperature, and a Monte Carlo of power-ups as a check
pr = puf_reliability()
res = puf_residual(DEFAULTS)                              # secret bits left after the helper data is published (point estimate)
# the temperatures at which the key-failure probability meets the 10^-6 target with the working t
ok_T = [T for T, p in zip(pr["temps"], pr["curves"][pr["t"]]) if p < 1e-6]
spec = (pr["temps"] >= 0) & (pr["temps"] <= 85)                       # the specified 0-85 °C
# mean number of flipped cells from the exact distribution, to compare with the Monte Carlo mean
mean_exact = float((np.arange(len(pr["dist"])) * pr["dist"]).sum())
mc_ub = 1 - 0.05 ** (1 / len(pr["errs"]))                # 95 % upper bound on P(fail) when no power-up of the sample failed
def sci1(v, d=0):                                         # 1 + d significant figures: 9.79e-40 -> 1 × 10^-39
    m, e = f"{v:.{d}e}".split("e")
    return f"{m} × 10<sup>{int(e)}</sup>".replace("-", "−")
card(f"Key reliability of this chip, BCH t = {pr['t']} (linear temperature law of the cell model)",
     [(f"P(key fails at one power-up), {pr['temp']} °C", f"<b>{sci1(pr['p_now'])}</b> (exact within the cell model; a figure this small "
       "states only that the model's Gaussian tail lies far below the target, not a property of silicon)"),
      ("Monte Carlo power-ups", f"{(pr['errs'] > pr['t']).sum()} of {len(pr['errs']):,} failed: P(key fails) &lt; {sci1(mc_ub, 1)} at 95 % confidence; "
       "the sample tests the body of the distribution, not its tail"),
      ("temperatures below 10<sup>−6</sup>", f"{ok_T[0]} to {ok_T[-1]} °C" if ok_T else "none"),
      (f"flipped cells per power-up, {pr['temp']} °C", f"mean {pr['errs'].mean():.2f} (Monte Carlo), {mean_exact:.2f} (exact); "
       f"largest seen {pr['errs'].max()}"),
      ("16 other chips, worst over 0–85 °C", f"{sci1(pr['band'][1][spec].max())}"),
      ("secret bits left after publishing <i>h</i>: 95 % lower bound – point estimate", f"<b>{budget(H_TT[0], pr['t'])}–{res}</b> "
       f"(need ≥ {SECRET_REQ}, Section VII-E)")],
     tone="ok" if ok_T and budget(H_TT[0], pr["t"]) >= SECRET_REQ else "bad")
````

The margin is smallest at the ends of the temperature range, where the linear law is optimistic. With the two-component model, Section VII-E finds every modelled chip within the target at $t = 45$ at the point estimate, and Section VIII-F finds the corners tt and ss within it and sf, ff and fs far from it.
* *Other chips.* The worst key-failure probability of the 16 other chips over 0–85 °C, with the same law, is 3 × 10⁻²³; this chip lies above their band wherever both are plotted, so a code sized on it is conservative for them.
* *Secrecy.* At $t = 45$ the helper data leave 94–137 secret bits (lower bound to point estimate), as predicted (4): **the binding limit is secrecy** (Section VII-E). Subsection C maps both limits over temperature and $t$.

### C. Parameter sweeps: security boundaries of both designs

Each of three experiments sweeps a grid of the two parameters that govern it:

1. **Replica attack: dither amplitude $a_\mathrm{d}$ against replica offset $\delta$.** As in A4, $\hat{K} = K \oplus (D \oplus D_\mathrm{r})$, and a 64-bit window succeeds when $D$ and $D_\mathrm{r}$ agree throughout it or disagree only within its first $L + 1 = 17$ bits (Section VIII).
   * The offset $\delta$ (% FS) between the two chips' inputs acts ahead of the quantiser and decorrelates the bit streams once large enough (Section VI).
   * The dither multiplies the comparator input by $1 + d$ and adds $d$, where $d \in \{-a_\mathrm{d}, 0, +a_\mathrm{d}\}$ is set by a pair of LFSRs with the chip's own seeds (Section V-C). A large $a_\mathrm{d}$ decorrelates the chips even at $\delta = 0$.

   The analysis above predicts that the attack succeeds against the original design only where both $\delta$ and $a_\mathrm{d}$ are small, and nowhere against the proposed design: BM cannot extend Grain-128a, and the per-frame IV limits even an exact estimate to the forced frame.
2. **Saturation attack: buffer gain $g$ against forced current $I$.**
   * The TIA output is $v = \mathrm{clip}(V_\mathrm{CM} + I R_\mathrm{TIA},\ 0,\ V_\mathrm{DD})$, and the modulator input, normalised to full scale, is $u = g\,(v - V_\mathrm{CM})/(V_\mathrm{DD}/2)$, with the common-mode voltage $V_\mathrm{CM}$, the TIA gain $R_\mathrm{TIA}$ and the supply voltage $V_\mathrm{DD}$ of Section V-A.
   * For $I \ge  V_\mathrm{CM} / R_\mathrm{TIA}$ ≈ 1.86 µA the TIA output is at the supply rail, and $u = g$.
   * A 1-bit loop tracks only $\lvert u\rvert < 1$; for $u \ge 1$ its bits become constant, the attacker's constant estimate of $D$ is correct, and BM recovers the LFSR.

   The analysis above predicts that the attack succeeds against the original design for $I \gtrsim$ 1.86 µA and $g \gtrsim 1$; at the design value $g = 0.5$, $u$ does not exceed 0.5. It succeeds nowhere against the proposed design (experiment 1).
3. **PUF key: temperature $T$ against code strength $t$.** $P_\mathrm{fail}(T, t)$ follows from Subsection B. Rows with $600\,H_\infty - 10t < 192$, with $H_{\infty}$ at its 95 % lower bound (0.908 bit), that is $t \ge 36$, leave too few secret bits (A11). The *design window* is the set of grid points with $P_\mathrm{fail}$ < 10⁻⁶ and at least 192 secret bits. The analysis above predicts a V-shaped window around 25 °C, widening with $t$, capped by the secrecy limit below the current $t = 45$; the linear law makes it too wide at high temperature (Subsection B).

The sweeps use the level and settings of Subsection A; only the swept parameters change:

* *Replica and saturation grids.* 30,000 clocks (`LAB_N`) of DC input per point (0.5 µA, or $I$).
  * Replica grid: $a_\mathrm{d} \in$ {off, 0.01, 0.022, 0.05, 0.1, 0.2, 0.4}, where 0.022 is the dither of [1]; $\delta \in \{0, 0.001, 0.003, 0.01, 0.03, 0.1, 0.3, 1\}$ % FS, where 0.1 % FS is the replica of Subsection A. The victim and the replica have their own dither seeds and noise.
  * Saturation grid: $g$ ∈ {0.25, 0.35, 0.5, 0.65, 0.8, 0.9, 1.0, 1.1, 1.2}; $I$ ∈ {0.5, 1, 2, 3, 5, 8, 12, 16, 20} µA. The attacker's estimate of $D$ is the more frequent bit value.
* *Decision rules.* The attack succeeds against the original design when one of the 80 windows of A4 succeeds (the attacker of [1], criterion of Section VIII), and against the proposed design only if the estimate agrees with the Grain-128a keystream on more than 97 % of the bits while the counter and MAC are off; here they are on.
* *Seeds and repetitions.* One run per grid point (victim from `seed` = 2026), so a point near a boundary may change with the seed.
* *PUF grid.* This chip, $T$ from −40 to 80 °C in steps of 10 °C and at 85 °C, and $t$ from 10 to 58 in steps of 4, evaluated with the exact distribution of Subsection B.
* *Metrics.* The clean-window fraction $f_{64}$, the fraction of 64-bit windows in which victim and replica agree on every bit (clocks 200 to 30,000), for the replica attack, the longest run of equal bits for the saturation attack, $\log_{10} P_\mathrm{fail}$ for the PUF key, and each design's verdict; the result panels give the successful grid points and, for the PUF grid, the largest $t$ with 192 secret bits and the window in the ring's row.

> **Next cell.** 1. Defines `replica_experiment` and `sweep(kind)` for the three grids. No simulation is run.

````python
#@title Sweeps: definitions
def replica_experiment(P, mm_pct, ki_input, Kb, Kp, mac):
    """The attack lab's zero/DC-input replica attack on LAB_N clocks, for both designs' keystreams."""
    u = np.full(LAB_N, s2d(tia(0.0 if ki_input == "zero" else 0.5e-6, P["rtia"]), P["buffer"]))
    Dv = dsm2(u, P, noise_seed=P["seed"])["D"]
    Dr = dsm2(u + mm_pct / 100, P, seeds=(0x2C4D, 0x51E7), noise_seed=P["seed"] + 2)["D"]
    # differing bits after the first 200 clocks; a 64-bit window is clean when its running-sum difference is 0
    x = (Dv ^ Dr)[200:].astype(np.int64); c = np.concatenate([[0], np.cumsum(x)])
    out = dict(flip=float(np.mean(Dv != Dr)), f64=float(np.mean(c[64:] - c[:-64] == 0)))   # f64: clean 64-bit windows (replica_pairs)
    # each design's keystream guess C xor D_replica; the original falls when a BM window breaks, the proposal only when
    # the guess is near-perfect and no MAC (with its per-frame IV) protects the link
    for name, K in (("base", Kb), ("prop", Kp)):
        kg = Dv ^ K ^ Dr
        w = window_attack(kg, K, P["seed"] + 3); raw = float(np.mean(kg == K))
        out[name] = "breached" if (w["broken"] > 0 if name == "base" else (raw > 0.97 and not mac)) else "safe"
    return out

def sweep(kind, P=None):
    """kind: 'replica' (dither x replica mismatch), 'saturate' (buffer gain x forced current), 'puf' (T x BCH t)."""
    P = {**DEFAULTS, **(P or {})}
    if kind == "puf":
        temps, ts = np.append(np.arange(-40, 81, 10), 85), np.arange(10, 61, 4)
        puf = SRAMPUF(P["chip_seed"], PUF_BITS, P["puf_sigma"], P["puf_bias"], P["puf_tc"])
        w = enroll(puf, P["bch_t"], rng_for(P["chip_seed"] * 7919 + 1))["w"]
        # key-failure probability P(more than t errors) from the exact error-count distribution at each temperature;
        # weak: t leaves fewer than 128 secret bits after the 10 t helper bits
        Z = np.zeros((len(ts), len(temps)))
        for i, T in enumerate(temps):
            d = err_dist(puf.p_err(w, T, puf_age(P)))
            for j, t in enumerate(ts):
                Z[j, i] = d[t + 1:].sum()
        weak = np.array([PUF_BITS * min_ent_bit(P["puf_bias"]) - 10 * t < 128 for t in ts])
        return dict(kind=kind, x=temps, y=ts, z=Z, weak=weak, cur=(P["temp"], P["bch_t"]))
    # the two designs' keystreams over LAB_N clocks, shared by every point of the sweep
    Kb = lfsr16_bits(P["key16"], ID_BITS + LAB_N)[ID_BITS:]
    Kp = simulate(**{**P, "scheme": "proposed", "attack": "none"})["K"][:LAB_N]
    if kind == "replica":
        amps, mms = [0, 0.01, 0.022, 0.05, 0.1, 0.2, 0.4], [0, 0.001, 0.003, 0.01, 0.03, 0.1, 0.3, 1.0]
        # grid of dither amplitude (0: off) x replica input mismatch (% FS): bit-flip fraction, clean 64-bit windows and
        # the verdict of each design
        Z = np.zeros((len(mms), len(amps))); F = np.zeros(Z.shape); VB = np.empty(Z.shape, object); VP = np.empty(Z.shape, object)
        for j, m in enumerate(mms):
            for i, a in enumerate(amps):
                X = replica_experiment({**P, "dither": a > 0, "dither_amp": a or P["dither_amp"]}, m, "dc", Kb, Kp, P["mac"])
                Z[j, i], F[j, i], VB[j, i], VP[j, i] = X["flip"], X["f64"], X["base"], X["prop"]
        return dict(kind=kind, x=amps, y=mms, z=Z, f64=F, vb=VB, vp=VP, cur=(P["dither_amp"] if P["dither"] else 0, ATK_DEF["mismatch"]))
    if kind == "saturate":
        gains, cur = [0.25, 0.35, 0.5, 0.65, 0.8, 0.9, 1.0, 1.1, 1.2], [0.5, 1, 2, 3, 5, 8, 12, 16, 20]
        # grid of buffer gain x forced current: the longest run of equal bits, and each design's verdict when the
        # attacker guesses every bit as the saturated level
        Z = np.zeros((len(cur), len(gains))); VB = np.empty(Z.shape, object); VP = np.empty(Z.shape, object)
        for j, c in enumerate(cur):
            for i, g in enumerate(gains):
                Q = {**P, "buffer": g}
                D = dsm2(np.full(LAB_N, s2d(tia(c * 1e-6, Q["rtia"]), g)), Q, noise_seed=Q["seed"])["D"]
                guess = 1 if D.mean() >= 0.5 else 0
                Z[j, i] = longest_run(D)
                kb = D ^ Kb ^ guess; VB[j, i] = "breached" if window_attack(kb, Kb, Q["seed"] + 23)["broken"] > 0 else "safe"
                kp = D ^ Kp ^ guess; VP[j, i] = "breached" if (np.mean(kp == Kp) > 0.97 and not P["mac"]) else "safe"
        return dict(kind=kind, x=gains, y=cur, z=Z, vb=VB, vp=VP, cur=(P["buffer"], P["force_uA"]))
    raise ValueError(kind)
````

> **Next cell.** 1. Defines `show_sweep`, which prints a sweep's result panel and draws its map. No simulation is run.

````python
#@title Sweeps: plotting helper
# Shared drawing code of the three parameter sweeps of Section IX-C (each sweep cell calls show_sweep with its result).
# A sweep result o holds: kind ('replica', 'saturate' or 'puf'); the grid axes x (columns) and y (rows); z, the swept
# quantity on the grid (bits that differ, longest run of equal bits, or the key-failure probability); vb and vp, the
# verdict of the original and of the proposed design at every grid point ('breached' or 'safe'); f64, the fraction of
# clean 64-bit windows (replica only); and cur, the current settings of the notebook, marked by a ring.
from matplotlib.colors import ListedColormap

def show_sweep(o, report=True, draw=True):
    """The card of one sweep (report) and its map (draw). Replica attack: (a) the clean-window fraction f64, (b) the
    original design's verdict. Saturation attack: one map, the run length with the original design's breached points
    hatched and the predicted boundary dashed. PUF key: the key-failure probability with the secrecy rows and the window."""
    kind = o["kind"]
    if kind == "puf":                                  # secrecy rows from the bounded budget of Section VII-E
        o["weak"] = np.array([budget(H_TT[0], int(t)) < SECRET_REQ for t in o["y"]])
    # grid size and the ring's cell (ci, cj): the grid point nearest to the current settings
    nx, ny = len(o["x"]), len(o["y"])
    ci = int(np.argmin(np.abs(np.array(o["x"], float) - o["cur"][0]))); cj = int(np.argmin(np.abs(np.array(o["y"], float) - o["cur"][1])))
    if report and kind != "puf":                       # count the breached grid points of each design
        nb_, np_ = int((o["vb"] == "breached").sum()), int((o["vp"] == "breached").sum())
        rows = [("original S-DSM", f"<b>{nb_} of {o['vb'].size}</b>"), ("proposed", f"<b>{np_} of {o['vp'].size}</b>"),
                ("at the ring (nearest grid point to the current settings)", f"original {o['vb'][cj, ci]}, proposed {o['vp'][cj, ci]}")]
        if kind == "replica":
            F, br = o["f64"], o["vb"] == "breached"
            rows += [("clean 64-bit windows f<sub>64</sub> at the ring", f"{100 * F[cj, ci]:.3f} %"),
                     ("grid points with f<sub>64</sub> = 0 (no clean window in 30,000 clocks)", f"{int((F == 0).sum())} of {F.size}"),
                     ("f<sub>64</sub> at the breached points", f"{100 * F[br].min():.3f} to {100 * F[br].max():.2f} %; {int((F[br] == 0).sum())} of "
                      f"{int(br.sum())} breached points have f<sub>64</sub> = 0" if br.any() else "none"),
                     ("f<sub>64</sub> with an identical replica (δ = 0), a<sub>d</sub> = " + ", ".join("off" if v == 0 else f"{v:g}" for v in o["x"]),
                      ", ".join(f"{100 * v:.2f}" for v in F[0]) + " %"),
                     ("largest f<sub>64</sub> at δ > 0", f"{100 * F[1:].max():.2f} %")]
        card(f"{'Replica' if kind == 'replica' else 'Saturation'} attack: grid points at which the attack succeeds", rows,
             tone="bad" if nb_ else "ok")
    elif report:                                       # the design window at the ring's row, and the secrecy limit on the grid
        win_T = np.array(o["x"])[(~o["weak"][cj]) & (o["z"][cj] < 1e-6)]
        card("PUF key: design window (linear temperature law)",
             [(f"largest t on the grid that leaves ≥ {SECRET_REQ} secret bits (95 % lower bound of the min-entropy)",
               f"<b>{int(np.array(o['y'])[~o['weak']].max())}</b>"),
              (f"grid temperatures in the window at t = {o['y'][cj]} (the ring's row)",
               f"<b>{win_T.min()} to {win_T.max()} °C</b>" if len(win_T) else "<b>none</b>")])
    if not draw:
        return
    # drawing helpers: the ring (white under black, visible on any colour), hatching of the breached cells, the axes with
    # thin white cell borders and tick labels of the swept values, and one design's verdict map
    ring = lambda ax: (ax.plot(ci, cj, "o", ms=12, mfc="none", mec="white", mew=3.2),
                       ax.plot(ci, cj, "o", ms=12, mfc="none", mec=COL["ink"], mew=1.4))
    def hatch(ax, V, ec):
        for j in range(ny):
            for i in range(nx):
                if V[j, i]:
                    ax.add_patch(plt.Rectangle((i - 0.5, j - 0.5), 1, 1, fill=False, hatch="///", ec=ec, lw=0))
    def axes_style(ax, xlabel, ylabel, yticks=True):
        ax.grid(False); ax.tick_params(top=False, right=False, direction="out")
        ax.set_xticks(np.arange(-0.5, nx, 1), minor=True); ax.set_yticks(np.arange(-0.5, ny, 1), minor=True)
        ax.grid(which="minor", color="white", linewidth=0.7, linestyle="-"); ax.tick_params(which="minor", length=0)   # thin cell borders
        sx = max(1, int(np.ceil(nx / 6))); sy = max(1, int(np.ceil(ny / 9)))     # thin the ticks so labels stay >= 9 pt
        ax.set_xticks(range(0, nx, sx), [("off" if (kind == "replica" and v == 0) else f"{v:g}".replace("-", "−")) for v in o["x"]][::sx], fontsize=9)
        ax.set_yticks(range(0, ny, sy), [f"{v:g}" for v in o["y"]][::sy] if yticks else [], fontsize=9)
        ax.set_xlabel(xlabel, fontsize=9.5)
        if ylabel:
            ax.set_ylabel(ylabel, fontsize=9.5)
    breached = lambda k: np.vectorize(lambda v: v == "breached")(o[k]).astype(float)
    ring_h = plt.Line2D([], [], marker="o", ls="none", ms=9, mfc="none", mec=COL["ink"], mew=1.4, label="current settings")
    def verdict(ax, key, lab, title, xlabel, ylabel):               # one design's verdict at every grid point
        V = breached(key)
        ax.imshow(V, origin="lower", aspect="auto", cmap=ListedColormap(["#dbe8fb", COL["bad"]]), vmin=0, vmax=1)
        hatch(ax, V, "white")
        ring(ax); axes_style(ax, xlabel, ylabel, yticks=ylabel is not None)
        panel(ax, lab, title + f": {int(V.sum())} of {V.size}")
    verdict_legend = [Patch(fc="#dbe8fb", ec=COL["ink"], lw=0.5, label="safe"),
                      Patch(fc=COL["bad"], ec=COL["ink"], lw=0.5, hatch="///", label="breached: the attack succeeds")]
    # replica sweep: (a) replica bits that differ, (b) clean-window fraction with the breached points hatched and the
    # 80-window protection level on the colour bar, (c), (d) the verdict of each design
    if kind == "replica":
        XL, YL = "Dither amplitude $a_{\\rm d}$", "Replica offset $\\delta$ (% FS)"
        fig, axs = plt.subplots(2, 2, figsize=(W2, 6.4))
        a0, a1, a2, a3 = axs.ravel()
        im = a0.imshow(o["z"], origin="lower", aspect="auto", cmap="Blues", vmin=0)        # (a) replica bits that differ
        cb = fig.colorbar(im, ax=a0, fraction=0.06, pad=0.03); cb.ax.tick_params(labelsize=9, direction="out")
        cb.set_label("Replica bits wrong", fontsize=9); cb.outline.set_linewidth(0.6)
        ring(a0); axes_style(a0, XL, YL)
        panel(a0, "a", "Replica bits that differ from the victim")
        FL = np.log10(np.maximum(o["f64"], 1e-5))                          # (b) f64 = 0 is drawn at 10^-5
        im = a1.imshow(FL, origin="lower", aspect="auto", cmap="Blues", vmin=-5, vmax=-1)
        cb = fig.colorbar(im, ax=a1, fraction=0.06, pad=0.03); cb.ax.tick_params(labelsize=9, direction="out")
        cb.set_ticks([-5, -4, -3, -2, -1], labels=["0", "0.01", "0.1", "1", "10"]); cb.set_label("$f_{64}$ (%)", fontsize=9)
        cb.outline.set_linewidth(0.6)
        cb.ax.axhline(np.log10(0.0013), color=COL["bad"], lw=2.0, ls="--", zorder=10)     # the theory's protection level (replica_pairs)
        hatch(a1, breached("vb"), COL["bad"])
        ring(a1); axes_style(a1, XL, None, yticks=False)                 # the y axis of (a), labelled once per row
        panel(a1, "b", "Clean 64-bit windows $f_{64}$")
        verdict(a2, "vb", "c", "Original S-DSM, breached", XL, YL)
        verdict(a3, "vp", "d", "Proposed, breached", XL, None)
        fig.legend(handles=verdict_legend + [plt.Line2D([], [], color=COL["bad"], ls="--", lw=1.6, label="$f_{64}$ = 0.13 %: 80-window protection level (colour bar)"), ring_h],
                   loc="lower center", ncol=4, frameon=False, fontsize=9, columnspacing=1.2)
        fig.tight_layout(w_pad=0.6, h_pad=1.0, rect=(0, 0.05, 1, 1))
    # saturation sweep: (a) the longest run of equal bits (a constant run is what the attacker exploits) with the predicted
    # boundary g = 1 and I = V_CM/R_TIA, (b), (c) the verdict of each design
    elif kind == "saturate":
        XL, YL = "S2D buffer gain $g$", "Forced current $I$ (µA)"
        fig, (ax, a2, a3) = plt.subplots(1, 3, figsize=(W2, 3.4), gridspec_kw=dict(width_ratios=[1.35, 1, 1]))
        im = ax.imshow(np.log10(1 + o["z"]), origin="lower", aspect="auto", cmap="Blues", vmin=0)
        cb = fig.colorbar(im, ax=ax, fraction=0.06, pad=0.03); cb.ax.tick_params(labelsize=9, direction="out")
        cb.set_label("$\\log_{10}$(1 + longest run of equal bits)", fontsize=9); cb.outline.set_linewidth(0.6)
        hatch(ax, breached("vb"), COL["bad"])
        gx = float(np.interp(1.0, o["x"], np.arange(nx))) - 0.5            # predicted boundary: g = 1 (between grid columns)
        iy = float(np.interp(VCM / DEFAULTS["rtia"] * 1e6, o["y"], np.arange(ny)))   # and I = V_CM / R_TIA
        ax.plot([gx, gx, nx - 0.5], [ny - 0.5, iy, iy], color=COL["bad"], ls="--", lw=1.6)
        ring(ax); axes_style(ax, XL, YL)
        panel(ax, "a", "Longest run, original")
        verdict(a2, "vb", "b", "Original", XL, None)
        verdict(a3, "vp", "c", "Proposed", XL, None)
        fig.legend(handles=verdict_legend + [plt.Line2D([], [], color=COL["bad"], ls="--", lw=1.6, label="predicted boundary"), ring_h],
                   loc="lower center", ncol=4, frameon=False, fontsize=9, columnspacing=1.2)
        fig.tight_layout(w_pad=0.6, rect=(0, 0.09, 1, 1))
    # PUF key sweep: key-failure probability over temperature and BCH t; rows that leave fewer than SECRET_REQ secret bits
    # hatched; the design window (reliable and secret) outlined
    else:
        fig, ax = plt.subplots(figsize=(4.8, 3.6))
        im = ax.imshow(np.log10(np.maximum(o["z"], 1e-20)), origin="lower", aspect="auto", cmap="Blues_r", vmin=-20)
        cb = fig.colorbar(im, ax=ax, fraction=0.06, pad=0.03); cb.ax.tick_params(labelsize=9, direction="out")
        cb.set_label("$\\log_{10}$ P(key fails)", fontsize=9); cb.outline.set_linewidth(0.6)
        for j in np.flatnonzero(o["weak"]):
            ax.add_patch(plt.Rectangle((-0.5, j - 0.5), nx, 1, fill=False, hatch="///", ec=COL["ink"], lw=0))
        win = (~o["weak"][:, None]) & (o["z"] < 1e-6)
        ax.contour(np.arange(nx), np.arange(ny), win.astype(float), levels=[0.5], colors=COL["ink"], linewidths=2)
        ring(ax); axes_style(ax, "Temperature (°C)", "BCH correction $t$")
        ax.set_title("Key-failure probability per power-up", loc="center", fontsize=10)
        fig.legend(handles=[Patch(fc="white", ec=COL["ink"], lw=0.5, hatch="///", label=f"< {SECRET_REQ} secret bits left"),
                            plt.Line2D([], [], color=COL["ink"], lw=2, label="design window"),
                            plt.Line2D([], [], marker="o", ls="none", ms=9, mfc="none", mec=COL["ink"], mew=1.4, label="current")],
                   loc="lower center", ncol=3, frameon=False, fontsize=9)
        fig.tight_layout(rect=(0, 0.07, 1, 1))
    plt.show()
    iplot(lambda: _isweep(o, ci, cj), {"replica": "Fig.: replica attack sweep (interactive)", "saturate": "Fig.: saturation attack sweep "
          "(interactive)", "puf": "Fig.: PUF key sweep (interactive)"}[kind])

def _isweep(o, ci, cj):
    """The interactive companion of a sweep map: the same grids as heatmaps on grid-index axes labelled with the swept
    values; hover gives both axis values and the exact value of the cell (and the verdict); the ring is the current setting."""
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    kind = o["kind"]; X, Y = np.array(o["x"], float), np.array(o["y"], float); nx, ny = len(X), len(Y)
    xt = [("off" if (kind == "replica" and v == 0) else f"{v:g}".replace("-", "−")) for v in X]
    yt = [f"{v:g}" for v in Y]
    xl, yl, xu, yu = {"replica": ("Dither amplitude a<sub>d</sub>", "Replica offset δ (% FS)", "a<sub>d</sub>", "δ"),
                      "saturate": ("S2D buffer gain g", "Forced current I (µA)", "g", "I"),
                      "puf": ("Temperature (°C)", "BCH correction t", "T", "t")}[kind]
    yunit = {"replica": " % FS", "saturate": " µA", "puf": ""}[kind]; xunit = " °C" if kind == "puf" else ""
    # hover text per cell: the two axis values plus the cell's own fields (customdata)
    XX, YY = np.meshgrid(xt, yt)
    def cd(*extra):                                                   # per-cell hover data: x label, y label, extra fields
        return np.dstack([XX, YY] + [np.asarray(e, dtype=object) for e in extra])
    head = xu + " = %{customdata[0]}" + xunit + ", " + yu + " = %{customdata[1]}" + yunit + "<br>"
    verd_cs = [[0, "#dbe8fb"], [0.5, "#dbe8fb"], [0.5, COL["bad"]], [1, COL["bad"]]]
    isb = lambda k: np.vectorize(lambda v: v == "breached")(o[k]).astype(float)
    def cells(mask):                                                  # closed outlines of the masked cells, one polygon each
        xs, ys = [], []
        for j, i in zip(*np.nonzero(mask)):
            xs += [i - 0.5, i + 0.5, i + 0.5, i - 0.5, i - 0.5, None]; ys += [j - 0.5, j - 0.5, j + 0.5, j + 0.5, j - 0.5, None]
        return xs, ys
    def hatched(f, mask, color, name, show, r, c):                    # hatched cells (as the static map), one legend entry
        xs, ys = cells(mask)
        if xs:
            f.add_trace(go.Scatter(x=xs, y=ys, mode="lines", fill="toself", line=dict(width=0, color=color), fillcolor="rgba(0,0,0,0)",
                                   fillpattern=dict(shape="/", fgcolor=color, size=7, solidity=0.25), name=name, legendgroup=name,
                                   showlegend=show, hoverinfo="skip"), row=r, col=c)
    def ring(f, show, r, c):                                          # white and black, as in the static map: seen on any cell
        f.add_trace(go.Scatter(x=[ci], y=[cj], mode="markers", marker=dict(size=8, color="rgba(0,0,0,0)", line=dict(color="white", width=2.6)),
                               legendgroup="ring", showlegend=False, hoverinfo="skip"), row=r, col=c)
        f.add_trace(go.Scatter(x=[ci], y=[cj], mode="markers", marker=dict(size=8, color="rgba(0,0,0,0)", line=dict(color=COL["ink"], width=1.0)),
                               name="current settings", legendgroup="ring", showlegend=show,
                               hovertemplate=f"current settings (nearest grid point): {xu} = {xt[ci]}{xunit}, {yu} = {yt[cj]}{yunit}<extra></extra>"),
                    row=r, col=c)
    def verdict(f, key, r, c, show):
        V = isb(key)
        f.add_trace(go.Heatmap(z=V, x=np.arange(nx), y=np.arange(ny), colorscale=verd_cs, zmin=0, zmax=1, showscale=False,
                               customdata=cd(o[key]), xgap=1, ygap=1, hovertemplate=head + "%{customdata[2]}<extra></extra>"), row=r, col=c)
        hatched(f, V > 0, "white", "breached: the attack succeeds", False, r, c)
        ring(f, show, r, c)
    def cbar(f, r, c, **kw):                                          # colour bar to the right of subplot (r, c)
        xa, ya = f.get_subplot(r, c).xaxis.domain, f.get_subplot(r, c).yaxis.domain
        return dict(x=xa[1] + 0.008, y=(ya[0] + ya[1]) / 2 - 0.03 * (ya[1] - ya[0]), len=0.82 * (ya[1] - ya[0]), thickness=12,
                    outlinewidth=0.6 * _PT, tickfont=dict(size=round(9 * _PT)), **kw)
    if kind == "replica":
        f = make_subplots(rows=2, cols=2, horizontal_spacing=0.14, vertical_spacing=0.14,
                          subplot_titles=("(a) Replica bits that differ from the victim", "(b) Clean 64-bit windows f<sub>64</sub>",
                                          f"(c) Original S-DSM, breached: {int(isb('vb').sum())} of {isb('vb').size}",
                                          f"(d) Proposed, breached: {int(isb('vp').sum())} of {isb('vp').size}"))
        f.add_trace(go.Heatmap(z=o["z"], x=np.arange(nx), y=np.arange(ny), colorscale="Blues", zmin=0, zmax=float(np.nanmax(o["z"])), xgap=1, ygap=1,
                               customdata=cd(100 * o["z"]),
                               colorbar=cbar(f, 1, 1, title=dict(text="Bits<br>wrong", side="top")),
                               hovertemplate=head + "replica bits that differ: %{customdata[2]:.3f} %<extra></extra>"), row=1, col=1)
        ring(f, True, 1, 1)
        F = np.asarray(o["f64"], float)
        f.add_trace(go.Heatmap(z=np.log10(np.maximum(F, 1e-5)), x=np.arange(nx), y=np.arange(ny), colorscale="Blues", zmin=-5, zmax=-1, xgap=1, ygap=1,
                               customdata=cd(100 * F, o["vb"]),
                               colorbar=cbar(f, 1, 2, title=dict(text="<i>f</i><sub>64</sub> (%)", side="top"), tickvals=[-5, -4, -3, -2, -1],
                                             ticktext=["0", "0.01", "0.1", "1", "10"]),
                               hovertemplate=head + "f<sub>64</sub> = %{customdata[2]:.4f} %<br>original S-DSM: %{customdata[3]}<extra></extra>"),
                    row=1, col=2)
        hatched(f, isb("vb") > 0, COL["bad"], "breached: the attack succeeds", False, 1, 2)
        ring(f, False, 1, 2)
        verdict(f, "vb", 2, 1, False); verdict(f, "vp", 2, 2, False)
        f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", marker=dict(symbol="square", size=5, color=COL["bad"]),
                               name="breached: the attack succeeds", hoverinfo="skip"), row=2, col=1)
        f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", marker=dict(symbol="square", size=5, color="#dbe8fb"),
                               name="safe", hoverinfo="skip"), row=2, col=1)
        grid = [(1, 1), (1, 2), (2, 1), (2, 2)]
    elif kind == "saturate":
        f = make_subplots(rows=1, cols=3, column_widths=[1.35, 1, 1], horizontal_spacing=0.15,
                          subplot_titles=("(a) Longest run, original", f"(b) Original: {int(isb('vb').sum())} of {isb('vb').size}",
                                          f"(c) Proposed: {int(isb('vp').sum())} of {isb('vp').size}"))
        Z = np.asarray(o["z"], float)
        f.add_trace(go.Heatmap(z=np.log10(1 + Z), x=np.arange(nx), y=np.arange(ny), colorscale="Blues", zmin=0, zmax=float(np.nanmax(np.log10(1 + Z))),
                               xgap=1, ygap=1, customdata=cd(Z, o["vb"]),
                               colorbar=cbar(f, 1, 1, title=dict(text="log<sub>10</sub>(1 + longest run)", side="right")),
                               hovertemplate=head + "longest run of equal bits: %{customdata[2]:,.0f}<br>original S-DSM: %{customdata[3]}<extra></extra>"),
                    row=1, col=1)
        hatched(f, isb("vb") > 0, COL["bad"], "breached: the attack succeeds", False, 1, 1)
        gx = float(np.interp(1.0, X, np.arange(nx))) - 0.5; iy = float(np.interp(VCM / DEFAULTS["rtia"] * 1e6, Y, np.arange(ny)))
        f.add_trace(go.Scatter(x=[gx, gx, nx - 0.5], y=[ny - 0.5, iy, iy], mode="lines", line=dict(color=COL["bad"], width=1.6, dash="dash"),
                               name="predicted boundary", hovertemplate="predicted boundary: <i>g</i> = 1, <i>I</i> = V<sub>CM</sub>/R<sub>TIA</sub><extra></extra>"), row=1, col=1)
        ring(f, True, 1, 1)
        verdict(f, "vb", 1, 2, False); verdict(f, "vp", 1, 3, False)
        f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", marker=dict(symbol="square", size=5, color=COL["bad"]),
                               name="breached: the attack succeeds", hoverinfo="skip"), row=1, col=2)
        f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", marker=dict(symbol="square", size=5, color="#dbe8fb"),
                               name="safe", hoverinfo="skip"), row=1, col=2)
        grid = [(1, 1), (1, 2), (1, 3)]
    else:
        f = make_subplots(rows=1, cols=1, subplot_titles=("Key-failure probability per power-up",))
        Z = np.asarray(o["z"], float); weak = np.asarray(o["weak"], bool)
        win = (~weak[:, None]) & (Z < 1e-6)
        sec = np.where(np.repeat(weak[:, None], nx, axis=1), f"fewer than {SECRET_REQ} secret bits", f"at least {SECRET_REQ} secret bits")
        f.add_trace(go.Heatmap(z=np.log10(np.maximum(Z, 1e-20)), x=np.arange(nx), y=np.arange(ny), colorscale="Blues", reversescale=True,
                               zmin=-20, zmax=float(np.nanmax(np.log10(np.maximum(Z, 1e-20)))), xgap=1, ygap=1, customdata=cd(Z, sec, np.where(win, "in the design window", "outside the design window")),
                               colorbar=cbar(f, 1, 1, title=dict(text="log<sub>10</sub> <i>P</i>", side="top")),
                               hovertemplate=head + "P(key fails) = %{customdata[2]:.2e}<br>%{customdata[3]}; %{customdata[4]}<extra></extra>"),
                    row=1, col=1)
        nm_w = f"fewer than {SECRET_REQ} secret bits left"           # light hatching on the dark cells, a dark key in the legend
        hatched(f, np.repeat(weak[:, None], nx, axis=1), "rgba(255,255,255,0.85)", nm_w, False, 1, 1)
        f.add_trace(go.Scatter(x=[None], y=[None], mode="lines", fill="toself", line=dict(width=0), fillcolor="rgba(0,0,0,0)",
                               fillpattern=dict(shape="/", fgcolor=COL["ink"], size=7, solidity=0.25), name=nm_w, legendgroup=nm_w,
                               hoverinfo="skip"), row=1, col=1)
        xs, ys = [], []                                               # outline of the design window: cell edges on its border
        for j, i in zip(*np.nonzero(win)):
            for di, dj, seg in ((-1, 0, ([i - .5, i - .5], [j - .5, j + .5])), (1, 0, ([i + .5, i + .5], [j - .5, j + .5])),
                                (0, -1, ([i - .5, i + .5], [j - .5, j - .5])), (0, 1, ([i - .5, i + .5], [j + .5, j + .5]))):
                if not (0 <= i + di < nx and 0 <= j + dj < ny and win[j + dj, i + di]):
                    xs += seg[0] + [None]; ys += seg[1] + [None]
        if xs:
            f.add_trace(go.Scatter(x=xs, y=ys, mode="lines", line=dict(color=COL["ink"], width=1.6), name="design window", hoverinfo="skip"),
                        row=1, col=1)
        ring(f, True, 1, 1)
        grid = [(1, 1)]
    sx, sy = max(1, int(np.ceil(nx / 6))), max(1, int(np.ceil(ny / 9)))     # the static figure's tick labels
    for r, c in grid:
        f.update_xaxes(tickvals=list(range(0, nx, sx)), ticktext=xt[::sx], range=[-0.5, nx - 0.5], title_text=xl, ticks="outside",
                       mirror=True, row=r, col=c)
        f.update_yaxes(tickvals=list(range(0, ny, sy)), ticktext=yt[::sy], range=[-0.5, ny - 0.5], title_text=yl if c == 1 else None,
                       showticklabels=c == 1 or kind != "saturate", ticks="outside", mirror=True, row=r, col=c)
    f.update_layout(legend=dict(orientation="h", x=0.5, xanchor="center", y=-0.12 if kind == "replica" else -0.22, yanchor="top"))
    return f
````

### D. Side-channel exposure: a leakage test on the register-level power model

The dynamic power of a CMOS register follows the number of flip-flops that toggle at each clock, so the Hamming distance between consecutive states is the first-order power model of a register-dominated block. A non-specific leakage test compares, clock by clock, traces with one fixed key against traces with random keys by Welch's $t$ statistic; $|t|$ above 4.5 marks a key-dependent difference that an attacker could exploit (the threshold of the test vector leakage assessment, TVLA, used to validate side-channel resistance). A control comparing two groups of random keys must stay below it. The analysis above predicts that the LFSR of [1], which restarts from its 16-bit key in every frame without a nonce, leaks at almost every clock; Grain-128a holds the key in its NFSR during the 256 initialisation clocks of every frame and mixes it there with the random IV, **so it should leak during the initialisation and not after it**.

**The test is behavioural, at register level, and live**. It uses 2,000 traces per group, each the number of state bits that toggle per clock (256 for Grain-128a, its LFSR and NFSR; 16 for the LFSR of [1]) over 512 clocks, with Gaussian noise of 4 toggles rms for the measurement; Grain-128a runs in authentication mode with a random 96-bit IV per trace, as in every frame of the proposed design, and the fixed and random keys are drawn with a seed. The vectorised Grain-128a reproduces the model's pre-output bit for bit. Combinational logic, the MAC registers, glitches and the layout are not modelled, and neither design is masked or hidden. Fig. 66 draws $|t|$ against the clock for both generators, with the 256 initialisation clocks of Grain-128a shaded, so that a leak can be attributed to the initialisation or to the keystream phase.

> **Next cell.** 1. Simulates the state registers of both generators for the three groups of traces. 2. Computes Welch's $t$ at every clock for the fixed-versus-random test and the control. 3. Prints the result panel and draws Fig. 66.

````python
#@title Fig.: side-channel exposure, leakage test
# Side-channel exposure of the two keystream generators: a non-specific fixed-versus-random leakage test (TVLA) on a
# register-level power model. Each trace is the Hamming distance of the generator's state register between consecutive
# clocks (the number of flip-flops that toggle), plus Gaussian measurement noise; Welch's t compares, clock by clock, traces
# with one fixed key against traces with random keys, and two groups of random keys (the negative control). Grain-128a in
# authentication mode with a random IV per trace (as every session of the proposal), its 128-bit LFSR and 128-bit NFSR;
# the 16-bit LFSR of [1], restarted from the key in every frame (Section IX-A, A6). The vectorised Grain-128a is
# checked bit for bit against the model's grain_preoutput. Outputs: TV.
N_TR, T_KS, SIG_TV, TH_TV = 2000, 256, 4.0, 4.5     # traces per group, keystream clocks after the 256 of initialisation,
                                                    # noise rms (toggles), threshold on |t|
t0 = time.time()
rng_tv = np.random.default_rng(2027)

def grain_hd(keys, ivs, T):
    """Toggles of the 256-bit Grain-128a state per clock for N traces at once, and the pre-output after initialisation.
    The registers are unrolled as in the model: the LFSR at clock t is s[t:t+128], the NFSR b[t:t+128]."""
    N, L = keys.shape[0], 256 + T
    s = np.zeros((N, 128 + L), np.uint8); b = np.zeros((N, 128 + L), np.uint8)
    b[:, :128] = keys; s[:, :96] = ivs; s[:, 96:127] = 1
    y_out = np.zeros((N, T), np.uint8)
    for t in range(L):
        b12, b95 = b[:, t + 12], b[:, t + 95]
        y = ((b12 & s[:, t + 8]) ^ (s[:, t + 13] & s[:, t + 20]) ^ (b95 & s[:, t + 42]) ^ (s[:, t + 60] & s[:, t + 79])
             ^ (b12 & b95 & s[:, t + 94]) ^ s[:, t + 93] ^ b[:, t + 2] ^ b[:, t + 15] ^ b[:, t + 36] ^ b[:, t + 45]
             ^ b[:, t + 64] ^ b[:, t + 73] ^ b[:, t + 89])
        st = s[:, t]
        fs = st ^ s[:, t + 7] ^ s[:, t + 38] ^ s[:, t + 70] ^ s[:, t + 81] ^ s[:, t + 96]
        fb = (st ^ b[:, t] ^ b[:, t + 26] ^ b[:, t + 56] ^ b[:, t + 91] ^ b[:, t + 96] ^ (b[:, t + 3] & b[:, t + 67])
              ^ (b[:, t + 11] & b[:, t + 13]) ^ (b[:, t + 17] & b[:, t + 18]) ^ (b[:, t + 27] & b[:, t + 59])
              ^ (b[:, t + 40] & b[:, t + 48]) ^ (b[:, t + 61] & b[:, t + 65]) ^ (b[:, t + 68] & b[:, t + 84])
              ^ (b[:, t + 88] & b[:, t + 92] & b[:, t + 93] & b95) ^ (b[:, t + 22] & b[:, t + 24] & b[:, t + 25])
              ^ (b[:, t + 70] & b[:, t + 78] & b[:, t + 82]))
        if t < 256:                                 # initialisation: the pre-output is fed back into both registers
            fs ^= y; fb ^= y
        else:
            y_out[:, t - 256] = y
        s[:, t + 128] = fs; b[:, t + 128] = fb
    return toggles(s, 128, L) + toggles(b, 128, L), y_out

def toggles(r, w, L):
    """Bits that change in a w-bit shift register between clocks t and t + 1, t = 0 .. L-1, for unrolled registers r."""
    c = np.concatenate([np.zeros((r.shape[0], 1), np.int32), np.cumsum(r[:, 1:] ^ r[:, :-1], axis=1, dtype=np.int32)], axis=1)
    return c[:, w:w + L] - c[:, :L]

def welch_t(a, b):
    return (a.mean(0) - b.mean(0)) / np.sqrt(a.var(0, ddof=1) / len(a) + b.var(0, ddof=1) / len(b))

noisy = lambda x: x + SIG_TV * rng_tv.standard_normal(x.shape)
ivs = [rng_tv.integers(0, 2, (N_TR, 96), dtype=np.uint8) for _ in range(3)]
for iv in ivs:
    iv[:, 0] = 1                                    # authentication mode, as in the proposal
k_fix = rng_tv.integers(0, 2, 128, dtype=np.uint8)
g_fix, y_fix = grain_hd(np.tile(k_fix, (N_TR, 1)), ivs[0], T_KS)
g_r1, _ = grain_hd(rng_tv.integers(0, 2, (N_TR, 128), dtype=np.uint8), ivs[1], T_KS)
g_r2, _ = grain_hd(rng_tv.integers(0, 2, (N_TR, 128), dtype=np.uint8), ivs[2], T_KS)
assert all(np.array_equal(y_fix[i], grain_preoutput(k_fix, ivs[0][i], T_KS)) for i in range(3)), "vectorised Grain-128a differs from the model"
t_g, t_g0 = welch_t(noisy(g_fix), noisy(g_r1)), welch_t(noisy(g_r1), noisy(g_r2))

def lfsr_hd(seeds, n):
    """Toggles of the 16-bit LFSR of [1] per clock: its register at clock t is the window of its next 16 output bits."""
    out = np.array([lfsr16_bits(int(k), n + 16) for k in seeds], np.uint8)
    return toggles(out, 16, n)
L_LF = 256 + T_KS                                   # the same number of clocks as Grain-128a
l_fix = lfsr_hd(np.full(N_TR, int(rng_tv.integers(1, 1 << 16))), L_LF)
l_r1, l_r2 = lfsr_hd(rng_tv.integers(1, 1 << 16, N_TR), L_LF), lfsr_hd(rng_tv.integers(1, 1 << 16, N_TR), L_LF)
t_l, t_l0 = welch_t(noisy(l_fix), noisy(l_r1)), welch_t(noisy(l_r1), noisy(l_r2))
above = lambda t: np.flatnonzero(np.abs(t) > TH_TV)
TV = dict(g_max=float(np.abs(t_g).max()), g_init=float(np.abs(t_g[:256]).max()), g_ks=float(np.abs(t_g[256:]).max()),
          g_first=int(above(t_g)[0]) if above(t_g).size else None, g_last=int(above(t_g)[-1]) if above(t_g).size else None,
          g_ctrl=float(np.abs(t_g0).max()), l_max=float(np.abs(t_l).max()), l_n=int(above(t_l).size), l_ctrl=float(np.abs(t_l0).max()),
          runtime=time.time() - t0)
card("Side-channel exposure: fixed-versus-random leakage test on the register-level power model",
     [("Grain-128a (proposal), fixed key vs random keys", f"max |t| = <b>{TV['g_max']:.1f}</b> during the initialisation (clocks "
       f"{TV['g_first']}–{TV['g_last']} above {TH_TV}); <b>{TV['g_ks']:.1f}</b> in the {T_KS} keystream clocks after it"),
      ("LFSR of [1], fixed key vs random keys", f"max |t| = <b>{TV['l_max']:.1f}</b>; {TV['l_n']} of {L_LF} clocks above {TH_TV}"),
      ("negative controls, random vs random keys", f"max |t| = {TV['g_ctrl']:.1f} (Grain-128a), {TV['l_ctrl']:.1f} (LFSR), both below {TH_TV}"),
      ("traces and model", f"{N_TR:,} per group; toggles of the state register per clock, noise {SIG_TV:.0f} toggles rms; "
                           "the vectorised Grain-128a equals the model bit for bit"),
      ("run time", f"{TV['runtime']:.0f} s")], tone="info")
fig, (a1, a2) = plt.subplots(1, 2, figsize=(W2, 2.9), sharey=True)
clk = np.arange(L_LF)
for ax, tf, t0_, lab in ((a1, t_g, t_g0, "a"), (a2, t_l, t_l0, "b")):
    ax.plot(clk, np.abs(tf), color=COL["bad"], lw=1.0, label="fixed vs random key")
    ax.plot(clk, np.abs(t0_), color=COL["faint"], lw=0.8, label="random vs random (control)")
    ax.axhline(TH_TV, color=COL["ink"], ls="--", lw=0.8)
    if lab == "a":
        ax.axvspan(0, 256, color=COL["faint"], alpha=0.12, lw=0)      # Grain-128a's initialisation
    ax.set_xlim(0, L_LF - 1); ax.set_xlabel("Clock cycle"); ax.set_yscale("log"); ax.set_ylim(0.01, 100)
    panel(ax, lab, "Grain-128a, proposal" if lab == "a" else "16-bit LFSR of [1]")
a1.set_ylabel("|t| (Welch)")
a1.text(128, 0.02, "initialisation", ha="center", fontsize=8, color=COL["ink"])
a2.legend(loc="lower right", fontsize=8)
fig.tight_layout(); plt.show()
````

Absolute value of Welch's $t$ against the clock cycle, 2,000 traces per group, toggle count of the state register with 4 toggles rms of noise; orange: one fixed key against random keys; grey: two groups of random keys (control); dashed: $|t|$ = 4.5; shaded: the 256 initialisation clocks of Grain-128a. **Key insights.** Grain-128a exceeds the threshold only during its initialisation (Fig. 66(a)), between clocks 0 and 125, while the NFSR still holds bits of the key, with $\lvert t\rvert$ up to 14.1; it stays at or below 2.6 in the 256 keystream clocks after it. The LFSR of [1] exceeds it at 387 of its 512 clocks, with $\lvert t\rvert$ up to 43.7 (Fig. 66(b)); at the other clocks the fixed key happens to toggle as many bits as random keys do on average. Both controls stay below the threshold (at most 3.0 and 3.3), so the test reports no leakage where none exists. **Both predictions hold, the LFSR leaking at three quarters of its clocks**.

**Neither design resists power analysis as built**, and the exclusion of side channels from the threat model is a substantive limitation. The two differ in where the key is exposed: in [1], at every clock of every frame, and an attacker can average many frames of the same sequence; in the proposed design, only while the key is mixed with the IV at the start of each frame, where masking or a protected key load would have to act. The model is first order: a chip adds combinational and glitch leakage, so the test shows where the key is exposed, not a security level.

**The proposed design's security rests on no analog margin of the signal path**, but on the PUF key generator, Grain-128a, the MAC and the counter. Its case against the S-DSM of [1], which spends 24 µW on security, is complete only with the price of these blocks: Section X determines what they cost in SKY130.

# **X. Implementation cost on SKY130**
---
Section IX established what the proposed design's added blocks buy: 14 of the 15 attacks stopped at the working setting and all 15 with the two-temperature enrolment, where the S-DSM fails 11. This section quantifies **what they cost in SKY130 standard-cell area and gate-level power, and how many bits they must send**, compared with the S-DSM of [1] and with the conventional alternative, an on-chip Trivium. The comparison is decisive, because [1] argues for the S-DSM on efficiency: its security adds 24 µW, an on-chip Trivium 220 µW (Fig. 4(a) of [1]). The proposed design replaces the 16-bit keystream LFSR by Grain-128a with its message authentication code (MAC), a frame counter, a session nonce and a 128-bit key register, so its cost must rest on a design checked against published test vectors, laid out and simulated at gate level in SKY130, **not on a gate-count estimate**.

The section uses the block composition of Section IV, the block powers and energy model of [1] (Fig. 4 of [1], reproduced in Section VI) and the PDK, tools and layout viewer of Section VII, and follows a design flow. Subsection A describes every security block at register-transfer level (RTL), in synthesisable Verilog and in two forms, **so that all areas and powers come from one flow and library**; B checks that every design reproduces the published test vectors bit for bit, as written (RTL) and as synthesised (gate netlist), and measures its area, flip-flops and switching activity; C shows how far the proposed design's block can be carried, through place and route, to a signed-off layout with open-source tools, and which checks remain open. D computes the gate-level power of the S-DSM, Trivium and the proposed design's block with the method of C under identical conditions, in both forms, and measures how much clock gating alone, inserted automatically by Yosys, removes from the block written for one bit per clock. E ranks the blocks of this flow by what their energy buys, with a figure of merit and without scaling any number between processes, counts the bits each design sends for the same information at an oversampling ratio (OSR) of 50, and estimates, as an example, the average power at one frame per minute with the radio of [1]. F shows the chip's power breakdown in the form of Fig. 4(a) of [1], beside the two designs measured there; G compares the complete system power of the proposed design with the chip of [1] block by block, every number's source stated; and H repeats the leakage test of Section IX-D on the netlists that would be built, so that the logic between the registers, the clock gating and the loads of the layout enter the power model, and finds in which phase of its protocol each design exposes its key.

### A. The designs and their RTL

Four designs are written as five Verilog top-level modules, all clocked at the modulator clock of [1], 10 MHz, and loading their key and initialisation vector (IV) in parallel in one clock:

* **S-DSM** [1] (`sdsm_ks`): three 16-bit linear-feedback shift registers (LFSRs) with the polynomial $x^{16} + x^{14} + x^{13} + x^{11} + 1$, bit-exact with the model of Section V: one generates the keystream, XORed with the modulator bit, and two the dither. A zero seed is replaced by 1 to avoid the all-zero lock-up state.
* **Trivium**, radix 1 (`trivium_r1`): the 288-bit state of the specification, loaded with the 80-bit key and the 80-bit IV, followed by 4 × 288 = 1,152 initialisation clocks without output, then one keystream bit per clock; the conventional reference of [1].
* **Grain-128a** [6], built on one core (`grain_core`): a 128-bit nonlinear feedback shift register (NFSR) and a 128-bit LFSR, with the output function $h$ and the feedback functions $f$ and $g$ of the specification. A parameter W sets the pre-output bits per clock; for W = 2 the core evaluates two consecutive steps in one clock, possible because every tap of the second step lies in the current state. Two forms are synthesised:
  * *encryption only* (`grain128a_enc1x`, W = 1): 256 initialisation clocks, then one keystream bit per clock, for reference;
  * *authentication mode* (`grain128a_auth2x`, "Grain-128a + MAC, 2×"): W = 2: each modulator clock yields one keystream bit (the even pre-output bit) and one MAC bit (the odd bit). The MAC (`grain_mac`) is a 32-bit accumulator and a 32-bit shift register, preloaded with the first 64 pre-output bits in 32 clocks after the 128 initialisation clocks; for each message bit equal to 1 the shift register is added to the accumulator, which holds the 32-bit tag after the padding bit. A frame therefore starts 1 + 128 + 32 = 161 clocks after the load.
* **The complete keystream block of the proposed design** (`secure_ks_top`): Grain-128a + MAC 2×; the 128-bit key register, shifted in serially from the fuzzy extractor once per power-up (128 clocks); the 32-bit session nonce, shifted in once per session (32 clocks), which also presets the 32-bit frame counter so that the first frame of a session uses counter 0; the 31-bit chip nonce $N_\mathrm{c}$, shifted in once per session (31 clocks) from the hash of 64 raw power-up reads; the counter, incremented at the start of every frame; the IV, formed as 1 ‖ nonce ‖ counter ‖ $N_\mathrm{c}$ as in the model; and the two dither LFSRs.

Written for one bit per clock, every design clocks all its flip-flops at 10 MHz, and **the power of a standard-cell block is mostly that of its clocked flip-flops** (Subsection D). Each design is therefore also written in a low-power form (`lowpower.v`), bit-exact and with the same ports, every design receiving the same two techniques:

* *Parallel state update (radix 16).* The cipher computes 16 keystream bits per state update, once every 16 clocks; an output multiplexer delivers one bit per clock. The S-DSM's LFSRs advance 16 steps per update (`sdsm_lp`); Trivium computes 16 bits per update (`trivium_lp`, 72 initialisation clocks instead of 1,152); Grain-128a computes 32 pre-output bits, 16 keystream and 16 MAC bits, per update (`grain128a_lp`), **so that a frame starts 11 clocks after the load instead of 161**. The MAC buffers the 16 message bits of each update and adds them in one cycle, its logic held still in the other 15 (operand isolation).
* *Clock gating.* Each register receives a clock edge only when it is written, through an integrated clock-gating cell (`sky130_fd_sc_hd__dlclkp_1`): the state registers once every 16 clocks, and in the proposed design (`secure_ks_lp`) the key, the two nonces and the counter only when loaded or incremented.

The gating cells are instantiated for the implementation (`define ICG_CELLS`) and written as enabled flip-flops for the single-clock simulator of Subsection B; the gated netlist is re-simulated after layout (Subsection C). A loaded next-state value is masked with the load signal on a kept net, **so that a gate-level simulation from undefined flip-flops loads exactly as a fabricated circuit would**.

**Four parts of the proposed design are not in these blocks**. Three run once at power-up: the SRAM-PUF array, the BCH decoder and the hash, which Section XI generates or synthesises and adds to the complete area (XI-E). The fourth is the per-frame MAC alignment: lock-step hardware would need a 32-bit delay line, estimated at about 170 gate equivalents (GE, defined in Subsection B), not included in the area.

> **Next cell.** Writes the RTL and verification flow to `rtl_flow.py`: 1. the Verilog of every design; 2. the Python reference models and the published test vectors; 3. the AIGER simulator; 4. the Yosys scripts.

````python
%%writefile rtl_flow.py
"""rtl_flow.py - RTL, verification and SKY130 synthesis helpers for the Code-a-Chip notebook.

Hardware cost of three ways to encrypt a 1-bit delta-sigma modulator stream:
  (a) S-DSM (ESSERC'25): modulator bit XOR a 16-bit LFSR keystream, plus two 16-bit dither LFSRs
  (b) Trivium, radix 1 (1 keystream bit per clock)
  (c) proposal: Grain-128a keyed from a PUF, authentication mode, 2 pre-output bits per modulator
      clock (even bits encrypt, odd bits feed the 32-bit MAC), with a 32-bit frame counter and a
      32-bit nonce forming the IV.

Standard library + numpy only. Every function that runs Yosys takes the binary name (`yosys` or
`yowasp-yosys`) and the liberty path as arguments; nothing here hard-codes a machine path.

Verification: the default engine ('aig') has Yosys write the design (RTL, or the synthesized sky130
netlist read with the liberty cell functions) as an AIGER file, which AigSim runs cycle by cycle in
numpy, every test job in its own bit lane and flip-flops starting from random values. It is needed
because yowasp-yosys (the WASM build used in Colab) has no `sim` command. Where `sim` exists,
run_all() also replays the same jobs as Verilog testbenches under Yosys `sim` ('yosys-sim').

Typical use (notebook):
    import rtl_flow as R
    res = R.run_all('yowasp-yosys', LIB, 'rtl_build')   # verify + synthesize everything
    print(R.area_table(res))

Bit conventions
  Grain-128a: as in the paper's test vectors (and model.js). Ports are declared [0:127] / [0:95], so
    a hex literal puts k_0 / IV_0 in the MSB (index 0) and port bit i is k_i / IV_i. The tag port is
    [0:31] with t_0 in the MSB, i.e. it prints as the paper's 8 hex digits.
  Trivium: the RTL follows the specification, port bit key[i] / iv[i] loads s_i / s_(93+i). The
    eSTREAM test vectors were produced by the reference code, which reverses the key/IV and maps bit
    sequences to bytes LSB first (see trivium.c's header comment); trivium_estream_bits() and
    trivium_bits_to_bytes() convert between the two.
  LFSR16: model.js class LFSR16, out = s & 1, s <- (s >> 1) | ((s ^ s>>2 ^ s>>3 ^ s>>5) & 1) << 15,
    seed 0 replaced by 1.
"""
import json
import os
import random
import re
import subprocess
import time

import numpy as np

# =====================================================================================
# 1. Verilog sources
#    VERILOG maps a file name to the Verilog text of that file. The strings are written to disk
#    unchanged by write_verilog_files() and are the only design source: the simulations, the
#    synthesis runs and the place-and-route flow all read these files. Each string starts with its
#    own comment header (what the module computes, its port conventions and its bit numbering).
# =====================================================================================
VERILOG = {}

# lfsr16.v: the 16-bit dither / keystream LFSR of the S-DSM baseline [ESSERC'25], bit-exact with the
# Python class LFSR16 below (and with model.js). Used by designs (a) and (c3).
VERILOG['lfsr16.v'] = r'''
// 16-bit LFSR, bit-exact with model.js class LFSR16 (x^16+x^14+x^13+x^11+1, right-shift form).
// out is the current bit (the model's bit() returns s & 1 and then shifts).
module lfsr16 (
  input         clk,
  input         load,              // synchronous seed load
  input  [15:0] seed,
  output        out
);
  reg [15:0] s;
  assign out = s[0];
  always @(posedge clk)
    if (load) s <= (seed == 16'd0) ? 16'd1 : seed;          // zero seed -> 1 (no lock-up)
    else      s <= {s[0] ^ s[2] ^ s[3] ^ s[5], s[15:1]};
endmodule
'''

# sdsm_ks.v: design (a), the S-DSM baseline: encrypted bit c = d XOR one LFSR16 output, plus two
# dither LFSR16s. Verified by jobs_sdsm() against three LFSR16 Python models.
VERILOG['sdsm_ks.v'] = r'''
// (a) ESSERC'25 S-DSM: c = d XOR LFSR16(key16); two more LFSR16s drive the dither
// (model.js dsm2: dither = amp * (l1.bit() - l2.bit())).
module sdsm_ks (
  input         clk,
  input         load,              // load the three seeds
  input  [15:0] key16, seed1, seed2,
  input         d,                 // modulator output bit
  output        c,                 // encrypted bit
  output        dith1, dith2       // dither bits
);
  wire k;
  lfsr16 u_key (.clk(clk), .load(load), .seed(key16), .out(k));
  lfsr16 u_d1  (.clk(clk), .load(load), .seed(seed1), .out(dith1));
  lfsr16 u_d2  (.clk(clk), .load(load), .seed(seed2), .out(dith2));
  assign c = d ^ k;
endmodule
'''

# trivium_r1.v: design (b), Trivium with one keystream bit per clock (288-bit state, 1152 init
# clocks). Verified by jobs_trivium() against the published eSTREAM test vectors (TRIVIUM_VECTORS).
VERILOG['trivium_r1.v'] = r'''
// (b) Trivium (De Canniere & Preneel, eSTREAM), radix 1: one keystream bit per clock.
// s[1:288] uses the specification's numbering. load copies key/iv into the state (parallel load),
// then 4 x 288 = 1152 initialisation clocks run without output; ready marks z = z_1, z_2, ...
module trivium_r1 (
  input         clk,
  input         load,
  input  [1:80] key,               // key[i] -> s_i
  input  [1:80] iv,                // iv[i]  -> s_(93+i)
  output        z,
  output        ready
);
  reg [1:288] s;
  reg [10:0]  cnt;
  reg         busy;
  wire t1 = s[66]  ^ s[93];
  wire t2 = s[162] ^ s[177];
  wire t3 = s[243] ^ s[288];
  assign z     = t1 ^ t2 ^ t3;
  assign ready = ~busy;
  wire n1 = t1 ^ (s[91]  & s[92])  ^ s[171];
  wire n2 = t2 ^ (s[175] & s[176]) ^ s[264];
  wire n3 = t3 ^ (s[286] & s[287]) ^ s[69];
  always @(posedge clk)
    if (load) begin
      s    <= {key, 13'd0, iv, 4'd0, 108'd0, 3'b111};
      cnt  <= 11'd0;
      busy <= 1'b1;
    end else begin
      s <= {n3, s[1:92], n1, s[94:176], n2, s[178:287]};
      if (busy) begin
        cnt <= cnt + 11'd1;
        if (cnt == 11'd1151) busy <= 1'b0;
      end
    end
endmodule
'''

# grain128a.v: the Grain-128a modules. grain_core (NFSR + LFSR + pre-output function, W = 1 or 2
# steps per clock), grain_mac (32-bit accumulator A and shift register R of the authentication
# mode), grain128a_enc1x (design c1, encryption only, 1 bit per clock; verified against the
# paper's pre-output vectors) and grain128a_auth2x (design c2, authentication mode, 2 pre-output
# bits per clock; verified against the paper's keystream, macstream and tag vectors).
VERILOG['grain128a.v'] = r'''
// Grain-128a (Agren, Hell, Johansson, Meier 2011). Index i of b / s is b_(t+i) / s_(t+i);
// [0:N] ranges put index 0 in the MSB, so hex literals follow the paper (k_0 = MSB).

// Pre-output generator: NFSR b, LFSR s, W = 1 or 2 pre-output bits per clock (y[0] = y_t,
// y[1] = y_(t+1)). load copies key / IV in parallel (s_96..s_126 = 1, s_127 = 0); while init is
// high the pre-output is fed back into both registers.
module grain_core #(parameter W = 1) (
  input           clk,
  input           load,
  input           init,
  input  [0:127]  key,
  input  [0:95]   iv,
  output [0:W-1]  y
);
  reg [0:127] b, s;
  // One step at the current offset: {y, s_t+128 (LFSR f), b_t+128 (NFSR g + s_t)}.
  function [0:2] step(input [0:127] bb, input [0:127] ss);
    reg h;
    begin
      h = (bb[12] & ss[8]) ^ (ss[13] & ss[20]) ^ (bb[95] & ss[42]) ^ (ss[60] & ss[79])
        ^ (bb[12] & bb[95] & ss[94]);
      step[0] = h ^ ss[93] ^ bb[2] ^ bb[15] ^ bb[36] ^ bb[45] ^ bb[64] ^ bb[73] ^ bb[89];
      step[1] = ss[0] ^ ss[7] ^ ss[38] ^ ss[70] ^ ss[81] ^ ss[96];
      step[2] = ss[0] ^ bb[0] ^ bb[26] ^ bb[56] ^ bb[91] ^ bb[96]
              ^ (bb[3] & bb[67]) ^ (bb[11] & bb[13]) ^ (bb[17] & bb[18]) ^ (bb[27] & bb[59])
              ^ (bb[40] & bb[48]) ^ (bb[61] & bb[65]) ^ (bb[68] & bb[84])
              ^ (bb[88] & bb[92] & bb[93] & bb[95])
              ^ (bb[22] & bb[24] & bb[25]) ^ (bb[70] & bb[78] & bb[82]);
    end
  endfunction
  wire [0:2] r0 = step(b, s);              // step t
  wire [0:2] r1 = step(b << 1, s << 1);    // step t+1 (taps <= 97, all in the current state)
  wire fs0 = r0[1] ^ (init & r0[0]), fb0 = r0[2] ^ (init & r0[0]);
  wire fs1 = r1[1] ^ (init & r1[0]), fb1 = r1[2] ^ (init & r1[0]);
  generate
    if (W == 1) assign y = r0[0];
    else        assign y = {r0[0], r1[0]};
  endgenerate
  always @(posedge clk)
    if (load) begin
      b <= key;
      s <= {iv, {31{1'b1}}, 1'b0};
    end else if (W == 1) begin
      b <= {b[1:127], fb0};
      s <= {s[1:127], fs0};
    end else begin
      b <= {b[2:127], fb0, fb1};
      s <= {s[2:127], fs0, fs1};
    end
endmodule

// MAC (Section 2.4): accumulator A and shift register R. pre: A||R shifts in the pair
// (ya, yb) for 32 clocks, so A = y_0..y_31 and R = y_32..y_63. Afterwards R shifts in the
// macstream bit yb every clock and acc applies A ^= R (message bit m_i = 1, or the padding bit).
module grain_mac (
  input         clk,
  input         pre,
  input         ya, yb,
  input         acc,
  output [0:31] tag                 // A, t_0 in the MSB
);
  reg [0:31] A, R;
  assign tag = A;
  always @(posedge clk)
    if (pre) begin
      A <= {A[2:31], R[0:1]};
      R <= {R[2:31], ya, yb};
    end else begin
      if (acc) A <= A ^ R;
      R <= {R[1:31], yb};
    end
endmodule

// (c1) Grain-128a, encryption only, 1 bit per clock: z = y_i (IV_0 = 0). 256 init clocks.
module grain128a_enc1x (
  input          clk,
  input          load,
  input  [0:127] key,
  input  [0:95]  iv,
  output         z,
  output         ready
);
  reg [7:0] cnt;
  reg       busy;
  grain_core #(.W(1)) u_core (.clk(clk), .load(load), .init(busy), .key(key), .iv(iv), .y(z));
  assign ready = ~busy;
  always @(posedge clk)
    if (load) begin cnt <= 8'd0; busy <= 1'b1; end
    else if (busy) begin
      cnt <= cnt + 8'd1;
      if (cnt == 8'd255) busy <= 1'b0;
    end
endmodule

// (c2) Grain-128a, authentication mode, 2 pre-output bits per clock = one keystream bit
// z_i = y_(64+2i) and one macstream bit a = y_(65+2i) per modulator clock.
// After load: 128 init clocks (256 pre-output bits), 32 MAC preload clocks, then RUN:
// m is message bit m_i at keystream index i; pad (in RUN) adds the padding bit m_L = 1,
// after which tag_ok is high and tag holds the 32-bit tag.
module grain128a_auth2x (
  input          clk,
  input          load,
  input  [0:127] key,
  input  [0:95]  iv,              // IV_0 must be 1 (authentication mode)
  input          m,
  input          pad,
  output         z,
  output         a,
  output         run,
  output         tag_ok,
  output [0:31]  tag
);
  localparam INIT = 2'd0, PRE = 2'd1, RUN = 2'd2, DONE = 2'd3;
  reg  [1:0] ph;
  reg  [6:0] cnt;
  wire [0:1] y;
  grain_core #(.W(2)) u_core (.clk(clk), .load(load), .init(ph == INIT), .key(key), .iv(iv), .y(y));
  grain_mac u_mac (.clk(clk), .pre(ph == PRE), .ya(y[0]), .yb(y[1]),
                   .acc((ph == RUN) & (m | pad)), .tag(tag));
  assign z = y[0];
  assign a = y[1];
  assign run    = (ph == RUN);
  assign tag_ok = (ph == DONE);
  always @(posedge clk)
    if (load) begin ph <= INIT; cnt <= 7'd0; end
    else case (ph)
      INIT: begin cnt <= cnt + 7'd1; if (cnt == 7'd127) ph <= PRE; end   // cnt wraps to 0
      PRE:  begin cnt <= cnt + 7'd1; if (cnt == 7'd31)  ph <= RUN; end
      RUN:  if (pad) ph <= DONE;
      default: ;
    endcase
endmodule
'''

# secure_ks_top.v: design (c3), the proposal's complete keystream block: serial key, session-nonce
# and chip-nonce registers, the 32-bit frame counter, grain128a_auth2x and two dither LFSR16s.
# Verified by jobs_proposal() (multi-frame, multi-session sequences against the Python model, and
# one sequence that must reproduce the paper's test vector TV3).
VERILOG['secure_ks_top.v'] = r'''
// (c3) Proposal keystream block: PUF key register + session nonce + chip nonce + frame counter feeding
// Grain-128a (auth mode, 2x), plus the two dither LFSR16s.
//   key_sh : shift the 128-bit key in from the fuzzy extractor, k_0 first (128 clocks)
//   nonce_sh: shift the receiver's 32-bit session nonce in, MSB first (32 clocks); this also
//            presets the counter to all-ones so the first frame of a session uses ctr = 0
//   cn_sh  : shift the 31-bit chip nonce N_c in, MSB first (31 clocks, once per session, from the hash of the raw
//            power-up read)
//   start  : new frame: ctr <= ctr + 1 and Grain loads key and IV = 1 | nonce | ctr | N_c
//            (the model's iv_bits); z / run / tag as in grain128a_auth2x
module secure_ks_top (
  input         clk,
  input         key_sh, key_in,
  input         nonce_sh, nonce_in,
  input         cn_sh, cn_in,
  input         start,
  input         m, pad,
  input         dith_load,
  input  [15:0] seed1, seed2,
  output        z,
  output        run, tag_ok,
  output [0:31] tag,
  output [31:0] ctr,               // counter value of the current frame (sent and MACed)
  output        dith1, dith2
);
  wire [0:127] key_r;
  wire [0:31]  nonce_r;
  wire [0:30]  cn_r;
  reg  [31:0]  ctr_r;
  wire [31:0]  ctr_n = ctr_r + 32'd1;
  assign ctr = ctr_r;
  shreg #(.N(128)) u_key   (.clk(clk), .sh(key_sh),   .din(key_in),   .q(key_r));
  shreg #(.N(32))  u_nonce (.clk(clk), .sh(nonce_sh), .din(nonce_in), .q(nonce_r));
  shreg #(.N(31))  u_cn    (.clk(clk), .sh(cn_sh),    .din(cn_in),    .q(cn_r));
  always @(posedge clk)
    if (nonce_sh)   ctr_r <= 32'hFFFF_FFFF;
    else if (start) ctr_r <= ctr_n;
  grain128a_auth2x u_grain (.clk(clk), .load(start), .key(key_r), .iv({1'b1, nonce_r, ctr_n, cn_r}),
                            .m(m), .pad(pad), .z(z), .a(), .run(run), .tag_ok(tag_ok), .tag(tag));
  lfsr16 u_d1 (.clk(clk), .load(dith_load), .seed(seed1), .out(dith1));
  lfsr16 u_d2 (.clk(clk), .load(dith_load), .seed(seed2), .out(dith2));
endmodule

// Serial-in shift register with shift enable; the first bit shifted in ends at index 0 (MSB).
module shreg #(parameter N = 8) (
  input            clk,
  input            sh,
  input            din,
  output reg [0:N-1] q
);
  always @(posedge clk)
    if (sh) q <= {q[1:N-1], din};
endmodule
'''

# Probes for the load-mux cost: the same logic with load tied low (the state can never be
# written, so the per-bit load multiplexers disappear). Synthesised only, never simulated.
# The area difference (design with load) - (probe without load) is the cost of the parallel
# key/IV load multiplexers. grain_core1 / grain_core2 are the bare cores with load, so that the
# core-only difference can be formed without the counter and MAC logic.
VERILOG['probes.v'] = r'''
module sdsm_noload (input clk, input d, output c, output dith1, output dith2);
  sdsm_ks u (.clk(clk), .load(1'b0), .key16(16'd0), .seed1(16'd0), .seed2(16'd0), .d(d), .c(c),
             .dith1(dith1), .dith2(dith2));
endmodule
module trivium_noload (input clk, output z);
  trivium_r1 u (.clk(clk), .load(1'b0), .key(80'd0), .iv(80'd0), .z(z), .ready());
endmodule
module grain_core1_noload (input clk, input init, output y);
  grain_core #(.W(1)) u (.clk(clk), .load(1'b0), .init(init), .key(128'd0), .iv(96'd0), .y(y));
endmodule
module grain_core2_noload (input clk, input init, output [0:1] y);
  grain_core #(.W(2)) u (.clk(clk), .load(1'b0), .init(init), .key(128'd0), .iv(96'd0), .y(y));
endmodule
module grain_core1 (input clk, input load, input init, input [0:127] key, input [0:95] iv, output y);
  grain_core #(.W(1)) u (.clk(clk), .load(load), .init(init), .key(key), .iv(iv), .y(y));
endmodule
module grain_core2 (input clk, input load, input init, input [0:127] key, input [0:95] iv, output [0:1] y);
  grain_core #(.W(2)) u (.clk(clk), .load(load), .init(init), .key(key), .iv(iv), .y(y));
endmodule
'''

# lowpower.v: radix-16, clock-gated versions of designs (a), (b), (c2) and (c3) (sdsm_lp,
# trivium_lp, grain128a_lp, secure_ks_lp). They are verified with the same job builders as the
# 1-bit-per-clock designs (JOB_BUILDERS; only the init latency differs), so the low-power versions
# must produce the same keystreams, macstreams and tags bit for bit.
VERILOG['lowpower.v'] = r'''
// Low-power (radix-K, clock-gated) versions of the three designs, bit-exact with the 1-bit-per-clock RTL above.
// Every register that does not change in a cycle receives no clock edge in it: with `define ICG_CELLS (implementation)
// its clock passes through an integrated clock-gating cell (sky130_fd_sc_hd__dlclkp_1); without it (verification with the
// single-clock AIGER simulator) the same register is written as an enable flip-flop. The state of each cipher advances K
// keystream bits per update, once every K clocks, and an output multiplexer delivers one bit per 10 MHz clock.
module icg (input clk, input en, output gclk);
`ifdef ICG_CELLS
  sky130_fd_sc_hd__dlclkp_1 u (.CLK(clk), .GATE(en), .GCLK(gclk));
`else
  assign gclk = clk;                                   // verification build: the enable is applied at the register
`endif
endmodule

// 16-bit LFSR of lfsr16.v, K steps per update: its next K outputs are s[0..K-1] (right-shifting register).
module lfsr16_lp #(parameter K = 16) (input clk, input load, input en, input [15:0] seed, input [3:0] c, output out);
  reg [15:0] s, t; integer i;
  always @* begin t = s; for (i = 0; i < K; i = i + 1) t = {t[0] ^ t[2] ^ t[3] ^ t[5], t[15:1]}; end
  reg o; integer j;                                    // out = s[c], written with constant indices
  always @* begin o = 1'b0; for (j = 0; j < 16; j = j + 1) if (c == j) o = s[j]; end
  assign out = o;
  // next state = (t AND NOT load) OR (seed AND load), the masked term kept as a net: during the load the old (possibly
  // still undefined) state reaches the register only through an AND with 0, so a gate-level simulation starting from
  // undefined flip-flops loads exactly as the silicon does
  (* keep *) wire [15:0] tm = t & {16{~load}};
  wire [15:0] nx = tm | ({16{load}} & ((seed == 16'd0) ? 16'd1 : seed));
`ifdef ICG_CELLS
  wire g; icg u_g (.clk(clk), .en(load | en), .gclk(g));
  always @(posedge g) s <= nx;
`else
  always @(posedge clk) if (load | en) s <= nx;
`endif
endmodule

// (a') S-DSM of [1], low-power: the three LFSR16s advance 16 steps once every 16 clocks.
module sdsm_lp (input clk, input load, input [15:0] key16, seed1, seed2, input d, output c, output dith1, dith2);
  reg [3:0] cc; wire tick = (cc == 4'd15);
  always @(posedge clk) if (load) cc <= 4'd0; else cc <= cc + 4'd1;
  wire k;
  lfsr16_lp #(16) u_key (.clk(clk), .load(load), .en(tick), .seed(key16), .c(cc), .out(k));
  lfsr16_lp #(16) u_d1  (.clk(clk), .load(load), .en(tick), .seed(seed1), .c(cc), .out(dith1));
  lfsr16_lp #(16) u_d2  (.clk(clk), .load(load), .en(tick), .seed(seed2), .c(cc), .out(dith2));
  assign c = d ^ k;
endmodule

// (b') Trivium, low-power: 16 keystream bits per state update (Trivium allows up to 64), init at 16 steps per clock.
module trivium_lp (input clk, input load, input [1:80] key, input [1:80] iv, output z, output ready);
  reg [1:288] s, t; reg [0:15] zz; reg t1, t2, t3; integer i;
  always @* begin
    t = s;
    for (i = 0; i < 16; i = i + 1) begin
      t1 = t[66] ^ t[93]; t2 = t[162] ^ t[177]; t3 = t[243] ^ t[288];
      zz[i] = t1 ^ t2 ^ t3;
      t1 = t1 ^ (t[91] & t[92]) ^ t[171]; t2 = t2 ^ (t[175] & t[176]) ^ t[264]; t3 = t3 ^ (t[286] & t[287]) ^ t[69];
      t = {t3, t[1:92], t1, t[94:176], t2, t[178:287]};
    end
  end
  reg [6:0] cnt; reg busy; reg [3:0] cc; wire tick = (cc == 4'd15);
  reg zo; integer j;
  always @* begin zo = 1'b0; for (j = 0; j < 16; j = j + 1) if (cc == j) zo = zz[j]; end
  assign ready = ~busy; assign z = zo;
  always @(posedge clk)
    if (load) begin cnt <= 7'd0; busy <= 1'b1; cc <= 4'd0; end
    else if (busy) begin cnt <= cnt + 7'd1; if (cnt == 7'd71) busy <= 1'b0; end     // 72 x 16 = 1152 init steps
    else cc <= cc + 4'd1;
  (* keep *) wire [1:288] tm = t & {288{~load}};        // load masking as in lfsr16_lp
  wire [1:288] nx = tm | ({288{load}} & {key, 13'd0, iv, 4'd0, 108'd0, 3'b111});
`ifdef ICG_CELLS
  wire g; icg u_g (.clk(clk), .en(load | busy | tick), .gclk(g));
  always @(posedge g) s <= nx;
`else
  always @(posedge clk) if (load | busy | tick) s <= nx;
`endif
endmodule

// Grain-128a core, W pre-output bits per update (W <= 32: every tap of the W steps lies in the current state).
module grain_core_lp #(parameter W = 32) (input clk, input load, input en, input init, input [0:127] key, input [0:95] iv,
                                          output [0:W-1] y);
  reg [0:127] b, s;
  function [0:2] step(input [0:127] bb, input [0:127] ss);
    reg h;
    begin
      h = (bb[12] & ss[8]) ^ (ss[13] & ss[20]) ^ (bb[95] & ss[42]) ^ (ss[60] & ss[79]) ^ (bb[12] & bb[95] & ss[94]);
      step[0] = h ^ ss[93] ^ bb[2] ^ bb[15] ^ bb[36] ^ bb[45] ^ bb[64] ^ bb[73] ^ bb[89];
      step[1] = ss[0] ^ ss[7] ^ ss[38] ^ ss[70] ^ ss[81] ^ ss[96];
      step[2] = ss[0] ^ bb[0] ^ bb[26] ^ bb[56] ^ bb[91] ^ bb[96]
              ^ (bb[3] & bb[67]) ^ (bb[11] & bb[13]) ^ (bb[17] & bb[18]) ^ (bb[27] & bb[59])
              ^ (bb[40] & bb[48]) ^ (bb[61] & bb[65]) ^ (bb[68] & bb[84])
              ^ (bb[88] & bb[92] & bb[93] & bb[95]) ^ (bb[22] & bb[24] & bb[25]) ^ (bb[70] & bb[78] & bb[82]);
    end
  endfunction
  wire [0:W-1] fs, fb; genvar k;
  generate for (k = 0; k < W; k = k + 1) begin : st
    wire [0:2] r = step(b << k, s << k);
    assign y[k] = r[0]; assign fs[k] = r[1] ^ (init & r[0]); assign fb[k] = r[2] ^ (init & r[0]);
  end endgenerate
  (* keep *) wire [0:127] bm = {b[W:127], fb} & {128{~load}}, sm = {s[W:127], fs} & {128{~load}};   // load masking
  wire [0:127] bn = bm | ({128{load}} & key), sn = sm | ({128{load}} & {iv, {31{1'b1}}, 1'b0});
`ifdef ICG_CELLS
  wire g; icg u_g (.clk(clk), .en(load | en), .gclk(g));
  always @(posedge g) begin b <= bn; s <= sn; end
`else
  always @(posedge clk) if (load | en) begin b <= bn; s <= sn; end
`endif
endmodule

// (c2') Grain-128a + MAC, low-power: 32 pre-output bits = 16 keystream + 16 macstream bits per update, once every 16
// clocks; same ports and behaviour as grain128a_auth2x except the latencies (RUN 11 clocks after load instead of 161).
module grain128a_lp (input clk, input load, input [0:127] key, input [0:95] iv, input m, input pad,
                     output z, output a, output run, output tag_ok, output [0:31] tag);
  localparam INIT = 2'd0, PRE = 2'd1, RUN = 2'd2, DONE = 2'd3;
  reg [1:0] ph; reg [3:0] cnt, cc;
  wire tick = (cc == 4'd15);
  wire [0:31] y;
  grain_core_lp #(32) u_core (.clk(clk), .load(load), .en((ph == INIT) | (ph == PRE) | (((ph == RUN) | (ph == DONE)) & tick)),
                              .init(ph == INIT), .key(key), .iv(iv), .y(y));
  reg [0:31] A, R; reg [0:15] mb;
  assign tag = A; assign run = (ph == RUN); assign tag_ok = (ph == DONE);
  reg zo, ao; integer j;
  always @* begin zo = 1'b0; ao = 1'b0; for (j = 0; j < 16; j = j + 1) if (cc == j) begin zo = y[2 * j]; ao = y[2 * j + 1]; end end
  assign z = zo; assign a = ao;
  // MAC update for the 16 message bits of the block: bit i (i < cc from the buffer, i = cc the current m | pad); operand
  // isolation keeps the XOR network still except in the update cycle (tick, or pad for the last, partial block)
  integer i; reg [0:31] An, Rn; reg mi; wire upd = (ph == RUN) & (tick | pad);
  always @* begin
    An = A; Rn = R;
    for (i = 0; i < 16; i = i + 1) begin
      mi = upd & ((i < cc) ? mb[i] : (i == cc) ? (m | pad) : 1'b0);
      if (mi) An = An ^ Rn;
      Rn = {Rn[1:31], y[2 * i + 1]};
    end
  end
  wire mac_en = ~load & ((ph == PRE) | upd);
`ifdef ICG_CELLS
  wire gm; icg u_gm (.clk(clk), .en(mac_en), .gclk(gm));
  always @(posedge gm) if (ph == PRE) {A, R} <= {A[0:31], R[0:31]} << 32 | {32'd0, y}; else begin A <= An; R <= Rn; end
`else
  always @(posedge clk) if (mac_en) begin if (ph == PRE) {A, R} <= {A[0:31], R[0:31]} << 32 | {32'd0, y}; else begin A <= An; R <= Rn; end end
`endif
  integer q;
  always @(posedge clk) for (q = 0; q < 15; q = q + 1) if ((ph == RUN) & (cc == q)) mb[q] <= m;   // mb[15] unused
  always @(posedge clk)
    if (load) begin ph <= INIT; cnt <= 4'd0; cc <= 4'd0; end
    else case (ph)
      INIT: begin cnt <= cnt + 4'd1; if (cnt == 4'd7) begin ph <= PRE; cnt <= 4'd0; end end    // 8 x 32 = 256 init steps
      PRE:  begin cnt <= cnt + 4'd1; if (cnt == 4'd1) ph <= RUN; end                            // 2 x 32 = 64 preload bits
      RUN:  begin cc <= cc + 4'd1; if (pad) ph <= DONE; end          // the keystream continues after the tag
      DONE: cc <= cc + 4'd1;
    endcase
endmodule

// (c3') Proposal keystream block, low-power: same ports and behaviour as secure_ks_top (latency aside); the key, nonce,
// chip-nonce and counter registers are clocked only when written.
module secure_ks_lp (
  input clk, input key_sh, key_in, nonce_sh, nonce_in, cn_sh, cn_in, start, m, pad, dith_load,
  input [15:0] seed1, seed2, output z, run, tag_ok, output [0:31] tag, output [31:0] ctr, output dith1, dith2);
  reg [0:127] key_r; reg [0:31] nonce_r; reg [0:30] cn_r; reg [31:0] ctr_r;
  wire [31:0] ctr_n = ctr_r + 32'd1;
  assign ctr = ctr_r;
`ifdef ICG_CELLS
  wire gk; icg u_gk (.clk(clk), .en(key_sh), .gclk(gk)); wire gn; icg u_gn (.clk(clk), .en(nonce_sh), .gclk(gn)); wire gc; icg u_gc (.clk(clk), .en(cn_sh), .gclk(gc)); wire gt; icg u_gt (.clk(clk), .en(nonce_sh | start), .gclk(gt));
  always @(posedge gk) key_r   <= {key_r[1:127], key_in};
  always @(posedge gn) nonce_r <= {nonce_r[1:31], nonce_in};
  always @(posedge gc) cn_r    <= {cn_r[1:30], cn_in};
  always @(posedge gt) ctr_r   <= nonce_sh ? 32'hFFFF_FFFF : ctr_n;
`else
  always @(posedge clk) begin
    if (key_sh)   key_r   <= {key_r[1:127], key_in};
    if (nonce_sh) nonce_r <= {nonce_r[1:31], nonce_in};
    if (cn_sh)    cn_r    <= {cn_r[1:30], cn_in};
    if (nonce_sh) ctr_r <= 32'hFFFF_FFFF; else if (start) ctr_r <= ctr_n;
  end
`endif
  grain128a_lp u_grain (.clk(clk), .load(start), .key(key_r), .iv({1'b1, nonce_r, ctr_n, cn_r}), .m(m), .pad(pad),
                        .z(z), .a(), .run(run), .tag_ok(tag_ok), .tag(tag));
  reg [3:0] dc; wire dt = (dc == 4'd15);
  always @(posedge clk) if (dith_load) dc <= 4'd0; else dc <= dc + 4'd1;
  lfsr16_lp #(16) u_d1 (.clk(clk), .load(dith_load), .en(dt), .seed(seed1), .c(dc), .out(dith1));
  lfsr16_lp #(16) u_d2 (.clk(clk), .load(dith_load), .en(dt), .seed(seed2), .c(dc), .out(dith2));
endmodule
'''

# name -> top module, source files, short description. The key selects the job builder in
# JOB_BUILDERS and the row in DESIGN_INFO; the file list is what Yosys reads for that design.
DESIGNS = {
    'sdsm':         ('sdsm_ks',          ['lfsr16.v', 'sdsm_ks.v'],
                     'S-DSM: 3 x LFSR16 + XOR'),
    'trivium':      ('trivium_r1',       ['trivium_r1.v'],
                     'Trivium, radix 1'),
    'grain_enc1x':  ('grain128a_enc1x',  ['grain128a.v'],
                     'Grain-128a, encryption only, 1 bit/clk'),
    'grain_auth2x': ('grain128a_auth2x', ['grain128a.v'],
                     'Grain-128a + MAC, 2x (1 keystream + 1 MAC bit/clk)'),
    'proposal':     ('secure_ks_top',    ['lfsr16.v', 'grain128a.v', 'secure_ks_top.v'],
                     'Proposal: Grain-128a+MAC 2x + ctr + nonce + chip nonce + key reg + 2 dither LFSR16'),
    # low-power versions (radix 16, clock-gated), bit-exact with the designs above: lowpower.v
    'sdsm_lp':      ('sdsm_lp',          ['lowpower.v'],
                     'S-DSM, low power: 3 x LFSR16 at 16 steps per update'),
    'trivium_lp':   ('trivium_lp',       ['lowpower.v'],
                     'Trivium, low power: 16 bits per update'),
    'grain_lp':     ('grain128a_lp',     ['lowpower.v'],
                     'Grain-128a + MAC, low power: 16 keystream + 16 MAC bits per update'),
    'proposal_lp':  ('secure_ks_lp',     ['lowpower.v'],
                     'Proposal, low power: Grain-128a+MAC radix 16, gated key/nonce/ctr registers, 2 dither LFSR16'),
}
# Designs that are additionally synthesised with explicit clock-gating cells (`define ICG_CELLS)
LOW_POWER = ('sdsm_lp', 'trivium_lp', 'grain_lp', 'proposal_lp')
PROBES = {  # load-mux cost probes: (top, files)
    'sdsm_noload':        ('sdsm_noload',        ['lfsr16.v', 'sdsm_ks.v', 'probes.v']),
    'trivium_noload':     ('trivium_noload',     ['trivium_r1.v', 'probes.v']),
    'grain_core1':        ('grain_core1',        ['grain128a.v', 'probes.v']),
    'grain_core1_noload': ('grain_core1_noload', ['grain128a.v', 'probes.v']),
    'grain_core2':        ('grain_core2',        ['grain128a.v', 'probes.v']),
    'grain_core2_noload': ('grain_core2_noload', ['grain128a.v', 'probes.v']),
}


def write_verilog_files(outdir):
    """Write every Verilog source of VERILOG to outdir (leading newline stripped).

    outdir: directory, created if missing. Returns the list of written paths."""
    os.makedirs(outdir, exist_ok=True)
    paths = []
    for name, text in VERILOG.items():
        p = os.path.join(outdir, name)
        with open(p, 'w') as f:
            f.write(text.lstrip('\n'))
        paths.append(p)
    return paths


# =====================================================================================
# 2. Reference models (Python) and test vectors
#    Bit-level Python models of the three ciphers and the published test vectors they are checked
#    against (reference_selftest). The RTL is then checked against both: published vectors where
#    they exist, the Python models for random keys, IVs, messages and multi-frame sequences.
#    Bit lists are 0/1 Python ints; index 0 is the first bit of the cipher's numbering.
# =====================================================================================
def hex_to_bits(h, n=None):
    """Hex -> bit list, bit 0 = MSB of the first digit (the Grain paper's convention)."""
    h = re.sub(r'\s+', '', h)
    bits = [(int(c, 16) >> (3 - j)) & 1 for c in h for j in range(4)]
    return bits if n is None else bits[:n]


def bits_to_hex(bits):
    """Bit list -> lower-case hex string, bit 0 = MSB of the first digit (zero-padded to a
    multiple of 4 bits); inverse of hex_to_bits."""
    bits = list(bits) + [0] * (-len(bits) % 4)
    return ''.join('%x' % (bits[i] << 3 | bits[i + 1] << 2 | bits[i + 2] << 1 | bits[i + 3])
                   for i in range(0, len(bits), 4))


def bits_to_int(bits):
    """Bit list -> integer, bit 0 is the MSB (so a port declared [0:N-1] gets bit i at index i)."""
    v = 0
    for b in bits:
        v = (v << 1) | (b & 1)
    return v


def int_to_bits(v, n):
    """Integer -> n-bit list, MSB first; inverse of bits_to_int."""
    return [(v >> (n - 1 - i)) & 1 for i in range(n)]


class LFSR16:
    """Port of model.js class LFSR16 (x^16 + x^14 + x^13 + x^11 + 1, right-shift Fibonacci form).

    The output is the current LSB; the feedback s0 ^ s2 ^ s3 ^ s5 enters at bit 15. A zero seed is
    replaced by 1 (the all-zero state is a fixed point), exactly as in lfsr16.v."""
    def __init__(self, seed):
        self.s = (seed & 0xFFFF) or 1

    def bit(self):
        """Return the current output bit s & 1, then advance the register by one step."""
        s = self.s
        fb = (s ^ (s >> 2) ^ (s >> 3) ^ (s >> 5)) & 1
        self.s = (s >> 1) | (fb << 15)
        return s & 1

    def bits(self, n):
        """The next n output bits."""
        return [self.bit() for _ in range(n)]


def grain_pre(key_bits, iv_bits, n):
    """Grain-128a pre-output y_0..y_(n-1) (port of model.js _grainRun / Grain128a).

    key_bits: 128 bits k_0..k_127; iv_bits: 96 bits IV_0..IV_95 (paper numbering). The registers
    are kept as growing lists: b[t + i] / s[t + i] is b_(t+i) / s_(t+i), so the taps are written
    exactly as in the paper with offset q = t, and each step appends b_(t+128) and s_(t+128).
    The first 256 steps are the initialisation (pre-output fed back, nothing emitted)."""
    # Initial state: NFSR = key, LFSR = IV || 1^31 || 0 (s_96..s_126 = 1, s_127 = 0)
    b = list(key_bits)
    s = list(iv_bits) + [1] * 31 + [0]
    out = []
    for t in range(256 + n):
        q = t
        b12, b95 = b[q + 12], b[q + 95]
        # Pre-output y_t = h(b, s) + s_(t+93) + sum of the seven linear NFSR taps
        y = ((b12 & s[q + 8]) ^ (s[q + 13] & s[q + 20]) ^ (b95 & s[q + 42]) ^
             (s[q + 60] & s[q + 79]) ^ (b12 & b95 & s[q + 94]) ^ s[q + 93] ^
             b[q + 2] ^ b[q + 15] ^ b[q + 36] ^ b[q + 45] ^ b[q + 64] ^ b[q + 73] ^ b[q + 89])
        st = s[q]
        # LFSR feedback f(s) and NFSR feedback g(b) + s_t (the LFSR output masks the NFSR input)
        fs = st ^ s[q + 7] ^ s[q + 38] ^ s[q + 70] ^ s[q + 81] ^ s[q + 96]
        fb = (st ^ b[q] ^ b[q + 26] ^ b[q + 56] ^ b[q + 91] ^ b[q + 96] ^
              (b[q + 3] & b[q + 67]) ^ (b[q + 11] & b[q + 13]) ^ (b[q + 17] & b[q + 18]) ^
              (b[q + 27] & b[q + 59]) ^ (b[q + 40] & b[q + 48]) ^ (b[q + 61] & b[q + 65]) ^
              (b[q + 68] & b[q + 84]) ^ (b[q + 88] & b[q + 92] & b[q + 93] & b95) ^
              (b[q + 22] & b[q + 24] & b[q + 25]) ^ (b[q + 70] & b[q + 78] & b[q + 82]))
        if t < 256:                            # initialisation: y is XORed into both feedbacks
            fs ^= y
            fb ^= y
        else:
            out.append(y)
        s.append(fs)
        b.append(fb)
    return out


def grain_keystream(key_bits, iv_bits, n_enc):
    """model.js grainKeystream: auth mode (IV_0 = 1) -> dict(enc, auth, A0, R0); else dict(enc).

    Encryption-only mode: the keystream is the pre-output itself, z_i = y_i. Authentication mode
    (paper Section 2.4): y_0..y_63 preload the MAC (A0 = y_0..y_31, R0 = y_32..y_63, as integers
    with the first bit in the MSB), then the even bits y_(64+2i) form the keystream (n_enc bits)
    and the odd bits y_(65+2i) the macstream. The macstream is 64 bits longer than the keystream
    so grain_tag() can process messages up to n_enc + 63 bits plus the padding bit."""
    if iv_bits[0] != 1:
        return {'enc': grain_pre(key_bits, iv_bits, n_enc)}
    n_auth = n_enc + 64
    y = grain_pre(key_bits, iv_bits, 64 + 2 * n_auth)
    return {'enc': y[64::2][:n_enc], 'auth': y[65::2][:n_auth],
            'A0': bits_to_int(y[0:32]), 'R0': bits_to_int(y[32:64])}


def grain_tag(msg_bits, ks):
    """model.js grainTag: 32-bit tag (t_0 in the MSB) of msg_bits under the keystream dict ks
    from grain_keystream (authentication mode)."""
    # For each message bit m_i: A ^= R if m_i = 1, then R shifts in the next macstream bit
    A, R = ks['A0'], ks['R0']
    for i, m in enumerate(msg_bits):
        if m:
            A ^= R
        R = ((R << 1) | ks['auth'][i]) & 0xFFFFFFFF
    # Padding bit m_L = 1 is always accumulated: the tag is the final A ^ R
    return A ^ R


def iv_bits(nonce, ctr, cn=0):
    """The model's iv_bits: IV_0 = 1, nonce (32 bits, MSB first), counter (32 bits), chip nonce N_c (31 bits, MSB first)."""
    return [1] + int_to_bits(nonce, 32) + int_to_bits(ctr, 32) + int_to_bits(cn, 31)


# Table 3 / Appendix A of Agren, Hell, Johansson, Meier, "Grain-128a: a new version of Grain-128
# with optional authentication", IJWMC 5(1):48-59, 2011 (copied from model.js GRAIN_TEST_VECTORS).
GRAIN_TEST_VECTORS = [
    {'key': '0000000000000000 0000000000000000', 'iv': '0000000000000000 00000000',
     'pre': 'c0207f221660650b 6a952ae26586136f a0904140c8621cfe 8660c0dec0969e94 36f4ace92cf1ebb7'},
    {'key': '0123456789abcdef 123456789abcdef0', 'iv': '0123456789abcdef 12345678',
     'pre': 'f88720c13f46e6a4 3c07eeed89161a4d d73bd6b8be8b6b11 6879714ebb630e0a 4c12f0399412982c'},
    {'key': '0000000000000000 0000000000000000', 'iv': '8000000000000000 00000000',
     'pre': '564b362219bd90e3 01f259cf52bf5da9 deb1845be6993abd 2d3c77c4acb90e42 2640fbd6e8ae642a',
     'accumulator': '564b3622', 'register': '19bd90e3',
     'keystream': '0d2b1f2ebc83da7e 6658ee3150f9ef47', 'macstream': '1cdbc7f1e52da547 36fa252828de82a0',
     'tags': ['4ff6a6c1', '653017e4', '7c8d8707', '522ab34f', '4b7821c9']},
    {'key': '0123456789abcdef 123456789abcdef0', 'iv': '8123456789abcdef 12345678',
     'pre': '7f2acdb7adfb701f 8d2083b3c32b43f1 962b3dcabf679378 db3536bfc25bed48 3008e6bcb395a156',
     'accumulator': '7f2acdb7', 'register': 'adfb701f',
     'keystream': 'a49d971c976bf596 b45f93e242ded8c1', 'macstream': '3015919d61787b5c d7678db840a6571e',
     'tags': ['d2d1bda8', '24dc2d89', '89275d96', '379d2899', '9226b196']},
]
# Appendix A messages: m0 empty, m1 = 0, m2 = 1, m3 = 12340 (20 bits), m4 = 123456789e8 (41 bits)
# Each entry is (hex digits, number of bits); the message is the first n bits of the hex, MSB first.
GRAIN_TEST_MESSAGES = [('', 0), ('0', 1), ('8', 1), ('12340', 20), ('123456789e8', 41)]


def grain_msg_bits(j):
    """Bit list of the paper's test message m_j (j = 0..4)."""
    h, n = GRAIN_TEST_MESSAGES[j]
    return hex_to_bits(h, n) if n else []


# eSTREAM Trivium test vectors, quoted verbatim from the submission package file
# ecrypt/trunk/submissions/trivium/unverified.test-vectors (profile ___H3, key 80 bits,
# IV 80 bits section), eSTREAM SVN: https://www.ecrypt.eu.org/stream/svn/viewcvs.cgi/ecrypt/trunk/
# submissions/trivium/ (archived tarball trivium.tar.gz?view=tar, Wayback 2011-05-27).
# "stream is generated by encrypting 512 zero bytes"; xor-digest = XOR of the 64-byte blocks.
TRIVIUM_VECTORS_TEXT = r'''
Set 1, vector#  0:
                         key = 80000000000000000000
                          IV = 00000000000000000000
               stream[0..63] = 38EB86FF730D7A9CAF8DF13A4420540D
                               BB7B651464C87501552041C249F29A64
                               D2FBF515610921EBE06C8F92CECF7F80
                               98FF20CCCC6A62B97BE8EF7454FC80F9
            stream[192..255] = EAF2625D411F61E41F6BAEEDDD5FE202
                               600BD472F6C9CD1E9134A745D900EF6C
                               023E4486538F09930CFD37157C0EB57C
                               3EF6C954C42E707D52B743AD83CFF297
            stream[256..319] = 9A203CF7B2F3F09C43D188AA13A5A202
                               1EE998C42F777E9B67C3FA221A0AA1B0
                               41AA9E86BC2F5C52AFF11F7D9EE480CB
                               1187B20EB46D582743A52D7CD080A24A
            stream[448..511] = EBF14772061C210843C18CEA2D2A275A
                               E02FCB18E5D7942455FF77524E8A4CA5
                               1E369A847D1AEEFB9002FCD02342983C
                               EAFA9D487CC2032B10192CD416310FA4
                  xor-digest = 7AE3A4B53355061766122E04391EA1E6
                               699B51C21A1F8058D3CF74A209D7E4CB
                               571ED771525CA492552565C10A05E81B
                               945DE28AAC043DEB349FD438784904D2

Set 3, vector#  0:
                         key = 00010203040506070809
                          IV = 00000000000000000000
               stream[0..63] = D2A8740BBA6FD9067077F9AFC0C27D40
                               32B6AEAE50C42ECEFF255C584C0143E7
                               8CFA4E3EBE03074F23D762D0A7563521
                               BE755B2166CD920EECBB5DB84737FA01
            stream[192..255] = 3F6A4CDDA613CE64B1F9C9AC662E4AB2
                               EF2751400CD6A0A119CF0BE7B287E727
                               536D18D953327B2D971EF9F34EA28762
                               CD062B7AEA83C1AC4363333219F767F8
            stream[256..319] = 44D06CB5157B2A8EE1CEEBC6DD5B500D
                               E7FBF83F189DFBE822042F85D814427F
                               F03F108FDB0989E7693257C863947712
                               8BF371CAA422D3306F6CDC1E03645BFE
            stream[448..511] = 30CD0B54E741F4CDD6E9B5CCAB184D7A
                               3453C03D4158FE7CB8BC92ECB66811C6
                               E560C62CF1ADE69BAE308ADC0602667C
                               CADEE71244968844376FBEB113E73345
                  xor-digest = 44DFE4D6F43708EC245C7EACA3B1B20F
                               BFA9436C7B2DC676457C932CC11F3960
                               E5D9852D2F9FDC77AEA2DDEC91CC2E1B
                               DE326CF4E21ED9380983C897CDC005A6

Set 5, vector#  0:
                         key = 00000000000000000000
                          IV = 80000000000000000000
               stream[0..63] = F8901736640549E3BA7D42EA2D07B9F4
                               9233C18D773008BD755585B1A8CBAB86
                               C1E9A9B91F1AD33483FD6EE3696D659C
                               9374260456A36AAE11F033A519CBD5D7
            stream[192..255] = 87423582AF64475C3A9C092E32A53C5F
                               E07D35B4C9CA288A89A43DEF3913EA92
                               37CA43342F3F8E83AD3A5C38D463516F
                               94E3724455656A36279E3E924D442F06
            stream[256..319] = D94389A90E6F3BF2BB4C8B057339AAD8
                               AA2FEA238C29FCAC0D1FF1CB2535A070
                               58BA995DD44CFC54CCEC54A5405B944C
                               532D74E50EA370CDF1BA1CBAE93FC0B5
            stream[448..511] = 4844151714E56A3A2BBFBA426A1D60F9
                               A4F265210A91EC29259AE2035234091C
                               49FFB1893FA102D425C57C39EB4916F6
                               D148DC83EBF7DE51EEB9ABFE045FB282
                  xor-digest = 76772EBDE1D3A73DBF3BB7E1A5BCC049
                               1419FF354D32F42E4D17F999E3B19DA1
                               989D6A1051EB0BBB9F880252F71E16B3
                               15324198AB34162DFEA981CF566F25AD

Set 6, vector#  0:
                         key = 0053A6F94C9FF24598EB
                          IV = 0D74DB42A91077DE45AC
               stream[0..63] = F4CD954A717F26A7D6930830C4E7CF08
                               19F80E03F25F342C64ADC66ABA7F8A8E
                               6EAA49F23632AE3CD41A7BD290A0132F
                               81C6D4043B6E397D7388F3A03B5FE358
'''


def parse_estream_vectors(text):
    """Parse eSTREAM test-vector blocks -> list of dict(name, key, iv, ranges{(a,b): hex}, digest)."""
    out = []
    # One block per "Set s, vector# v:" header; within a block, "name = HEX" starts a field and
    # indented hex-only lines continue the previous field (long streams span several lines)
    for blk in re.split(r'\n(?=Set \d+, vector#)', text.strip()):
        head = re.match(r'(Set \d+, vector#\s*\d+):', blk)
        if not head:
            continue
        fields, cur = [], None
        for line in blk.split('\n')[1:]:
            m = re.match(r'\s*([\w\-\[\]\.]+)\s*=\s*([0-9A-Fa-f]+)\s*$', line)
            if m:
                cur = [m.group(1), m.group(2)]
                fields.append(cur)
            elif cur and re.match(r'\s+[0-9A-Fa-f]+\s*$', line):
                cur[1] += line.strip()
        # ranges: (first byte, last byte) -> expected hex of stream[a..b]
        d = {'name': re.sub(r'\s+', ' ', head.group(1)), 'ranges': {}, 'digest': None}
        for k, v in fields:
            r = re.match(r'stream\[(\d+)\.\.(\d+)\]', k)
            if k == 'key':
                d['key'] = v
            elif k == 'IV':
                d['iv'] = v
            elif k == 'xor-digest':
                d['digest'] = v.upper()
            elif r:
                d['ranges'][(int(r.group(1)), int(r.group(2)))] = v.upper()
        out.append(d)
    return out


TRIVIUM_VECTORS = parse_estream_vectors(TRIVIUM_VECTORS_TEXT)


def trivium_estream_bits(hexstr):
    """eSTREAM key/IV hex -> spec load bits (s_1..s_80 for the key, s_94..s_173 for the IV).
    The reference code maps the bit sequence K_1..K_80 to bytes LSB first and loads
    (s_1..s_80) = (K_80..K_1)."""
    by = bytes.fromhex(hexstr)
    # K_1..K_80: byte by byte, each byte LSB first; the spec state takes them in reverse order
    seq = [(b >> j) & 1 for b in by for j in range(8)]
    return seq[::-1]


def trivium_bits_to_bytes(z):
    """Keystream z_1, z_2, ... -> bytes, first bit in the LSB (eSTREAM reference convention)."""
    z = list(z)
    return bytes(sum(z[i + j] << j for j in range(8)) for i in range(0, len(z) - 7, 8))


def trivium_ref(key_load, iv_load, n):
    """Trivium per the specification: key_load / iv_load are the 80 bits loaded into s_1.. / s_94..;
    returns z_1..z_n."""
    # s[1..288] with the specification's 1-based numbering (s[0] unused); s_286..s_288 = 1
    s = [0] * 289
    s[1:81] = key_load
    s[94:174] = iv_load
    s[286] = s[287] = s[288] = 1
    out = []
    # 4 x 288 = 1152 initialisation rounds without output, then one keystream bit per round
    for r in range(1152 + n):
        t1, t2, t3 = s[66] ^ s[93], s[162] ^ s[177], s[243] ^ s[288]
        if r >= 1152:
            out.append(t1 ^ t2 ^ t3)
        t1 ^= (s[91] & s[92]) ^ s[171]
        t2 ^= (s[175] & s[176]) ^ s[264]
        t3 ^= (s[286] & s[287]) ^ s[69]
        # Rotate the three registers (93, 84, 111 bits): each takes the feedback of the previous one
        s[1:94] = [t3] + s[1:93]
        s[94:178] = [t1] + s[94:177]
        s[178:289] = [t2] + s[178:288]
    return out


def estream_check(stream_bytes, vec):
    """Compare a keystream (bytes) with the vector's ranges and xor-digest -> list of (label, ok)."""
    res = []
    # Only the ranges covered by the generated stream are compared
    for (a, b), hx in sorted(vec['ranges'].items()):
        if b < len(stream_bytes):
            res.append(('stream[%d..%d]' % (a, b), stream_bytes[a:b + 1].hex().upper() == hx))
    # eSTREAM xor-digest: XOR of the eight 64-byte blocks of the first 512 stream bytes
    if vec['digest'] and len(stream_bytes) >= 512:
        dig = bytearray(64)
        for i in range(512):
            dig[i % 64] ^= stream_bytes[i]
        res.append(('xor-digest', bytes(dig).hex().upper() == vec['digest']))
    return res


def reference_selftest():
    """Check the Python reference models against the published vectors -> list of check dicts."""
    checks = []
    # Grain-128a: all four vectors give 320 pre-output bits; TV3 / TV4 (IV_0 = 1, authentication
    # mode) also give the MAC preload A0 / R0, 128 keystream and 128 macstream bits and the tags of
    # the five Appendix A messages
    for n, tv in enumerate(GRAIN_TEST_VECTORS):
        key, iv = hex_to_bits(tv['key']), hex_to_bits(tv['iv'])
        pre = re.sub(r'\s+', '', tv['pre'])
        checks.append(dict(design='python-ref', check='Grain TV%d pre-output y_0..y_319' % (n + 1),
                           ok=bits_to_hex(grain_pre(key, iv, 320)) == pre))
        if 'tags' in tv:
            ks = grain_keystream(key, iv, 128)
            ok = ('%08x' % ks['A0'] == tv['accumulator'] and '%08x' % ks['R0'] == tv['register'] and
                  bits_to_hex(ks['enc']) == re.sub(r'\s+', '', tv['keystream']) and
                  bits_to_hex(ks['auth'][:128]) == re.sub(r'\s+', '', tv['macstream']) and
                  all('%08x' % grain_tag(grain_msg_bits(j), ks) == tv['tags'][j] for j in range(5)))
            checks.append(dict(design='python-ref', check='Grain TV%d A0/R0/keystream/macstream/tags'
                               % (n + 1), ok=ok))
    # Trivium: 4096 keystream bits = 512 bytes, enough for every quoted range and the xor-digest
    for v in TRIVIUM_VECTORS:
        z = trivium_ref(trivium_estream_bits(v['key']), trivium_estream_bits(v['iv']), 4096)
        res = estream_check(trivium_bits_to_bytes(z), v)
        checks.append(dict(design='python-ref', check='Trivium %s (%s)' % (v['name'], ', '.join(r[0] for r in res)),
                           ok=all(r[1] for r in res)))
    return checks


# =====================================================================================
# 3. Running Yosys, VCD parsing
# =====================================================================================
def run_yosys(yosys, script, cwd, log_name='yosys.log', timeout=3600):
    """Run `yosys -s script` (script text is written to cwd). Returns (log text, seconds)."""
    os.makedirs(cwd, exist_ok=True)
    # The script file is named after the log (sim_rtl.log -> sim_rtl.ys) so both stay together in cwd
    sname = log_name.replace('.log', '.ys')
    with open(os.path.join(cwd, sname), 'w') as f:
        f.write(script)
    t0 = time.time()
    p = subprocess.run([yosys, '-q', '-l', log_name, '-s', sname], cwd=cwd, capture_output=True,
                       text=True, timeout=timeout)
    dt = time.time() - t0
    log = open(os.path.join(cwd, log_name)).read() if os.path.exists(os.path.join(cwd, log_name)) else ''
    if p.returncode != 0:                      # report the tails of stderr and of the log
        raise RuntimeError('%s failed (%d):\n%s\n%s' % (yosys, p.returncode, p.stderr[-3000:], log[-3000:]))
    return log, dt


def has_command(yosys, cmd):
    """True if this Yosys build has `cmd` (yowasp-yosys has no `sim`, for example)."""
    p = subprocess.run([yosys, '-p', 'help %s' % cmd], capture_output=True, text=True)
    return p.returncode == 0 and 'No such command' not in p.stdout + p.stderr


def yosys_version(yosys):
    """Version string of the Yosys binary (last line starting with 'Yosys' of `yosys -V`)."""
    p = subprocess.run([yosys, '-V'], capture_output=True, text=True)
    lines = [l for l in (p.stdout + p.stderr).splitlines() if l.startswith('Yosys')]
    return lines[-1].strip() if lines else (p.stdout + p.stderr).strip()


def parse_vcd(path):
    """Minimal VCD reader -> (vars, changes). vars: list of (fullname, kind, width, code);
    changes: code -> list of (time, value string)."""
    vars_, changes, scope, t, header = [], {}, [], 0, True
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            # Header: track the $scope nesting so each $var gets its hierarchical name
            if header:
                tok = line.split()
                if tok[0] == '$scope':
                    scope.append(tok[2])
                elif tok[0] == '$upscope':
                    scope.pop()
                elif tok[0] == '$var':
                    vars_.append(('.'.join(scope + [tok[4]]), tok[1], int(tok[2]), tok[3]))
                elif tok[0] == '$enddefinitions':
                    header = False
                continue
            # Body: '#t' sets the time; 'b<value> <code>' (vector) or '<value><code>' (scalar)
            c = line[0]
            if c == '#':
                t = int(line[1:])
            elif c in 'br':
                v, code = line[1:].split()
                changes.setdefault(code, []).append((t, v))
            elif c in '01xzXZ':
                changes.setdefault(line[1:], []).append((t, c))
    return vars_, changes


def sample_cycles(changes_list, period, ncycles):
    """Value string of a signal in each cycle k (= value at time k*period, after the k-th edge)."""
    # Walk the time-sorted change list once; cur holds the last value at or before k*period
    out, j, cur = [], 0, None
    for k in range(ncycles):
        tk = k * period
        while j < len(changes_list) and changes_list[j][0] <= tk:
            cur = changes_list[j][1]
            j += 1
        out.append(cur)
    return out


def vstr_to_int(v):
    """VCD binary value string -> int, or None when unknown or any bit is x / z."""
    return None if v is None or re.search(r'[xzXZ]', v) else int(v, 2)


class SimResult:
    """Per-cycle samples of a VCD written by `sim` (cycle k = state after the k-th clock edge)."""
    def __init__(self, path, ncycles):
        self.vars, self.changes = parse_vcd(path)
        self.byname = {v[0]: v for v in self.vars}
        # Clock period from the testbench-level clk (tb.clk): distance between its first two rising edges
        clk = [v for v in self.vars if v[0].split('.')[-1] == 'clk' and v[0].count('.') == 1][0]
        rises = [t for t, v in self.changes[clk[3]] if v == '1']
        self.period = rises[1] - rises[0] if len(rises) > 1 else 10
        self.ncycles = ncycles

    def sig(self, name):
        """Integer per cycle (None if x) of the signal tb.<name>."""
        top = self.vars[0][0].split('.')[0]
        code = self.byname[top + '.' + name][3]
        return [vstr_to_int(v) for v in sample_cycles(self.changes.get(code, []), self.period, self.ncycles)]

    def ff_toggles(self, scope_prefix, c0, c1):
        """Total bit toggles of all VCD 'reg' variables (flip-flop outputs in Yosys sim) under
        scope_prefix between cycles c0 and c1 -> (toggles per cycle, number of FF bits)."""
        tog, nbits = 0, 0
        # Toggles = Hamming distance between consecutive cycles, summed over all register bits
        for name, kind, width, code in self.vars:
            if kind != 'reg' or not name.startswith(scope_prefix):
                continue
            vals = [vstr_to_int(v) for v in sample_cycles(self.changes.get(code, []), self.period, c1 + 1)]
            nbits += width
            for k in range(c0 + 1, c1 + 1):
                if vals[k] is not None and vals[k - 1] is not None:
                    tog += bin(vals[k] ^ vals[k - 1]).count('1')
        return tog / max(1, c1 - c0), nbits


# =====================================================================================
# 4. Testbenches (synthesizable, cycle-indexed stimulus; run with Yosys `sim`)
# =====================================================================================
def make_tb(name, dut, ncycles, consts, streams, outputs):
    """Testbench module `name` around module `dut`.
    consts : port -> (width, value) constant inputs (MSB of value -> MSB/index-0 of the port)
    streams: port -> list of 0/1 per cycle (1-bit inputs, value in cycle k)
    outputs: port -> width, exposed as tb outputs o_<port>
    """
    # The testbench is itself synthesisable (Yosys `sim` has no initial blocks / delays): a 16-bit
    # cycle counter cyc indexes constant bit vectors that hold each input stream
    n = ncycles + 2
    L = ['module %s (input clk%s);' % (name, ''.join(
        ', output %so_%s' % ('[%d:0] ' % (w - 1) if w > 1 else '', p) for p, w in outputs.items()))]
    L.append('  reg [15:0] cyc = 16\'d0;')
    L.append('  always @(posedge clk) cyc <= cyc + 16\'d1;')
    conns = ['.clk(clk)']
    for p, (w, v) in consts.items():
        L.append('  wire [%d:0] i_%s = %d\'h%x;' % (w - 1, p, w, v))
        conns.append('.%s(i_%s)' % (p, p))
    # Stream of port p: localparam [0:n-1] with bit k = value in cycle k (bits_to_int puts bit 0 in
    # the MSB = index 0); the stream is zero-padded or truncated to n cycles
    for p, bits in streams.items():
        bits = (list(bits) + [0] * n)[:n]
        L.append('  localparam [0:%d] S_%s = %d\'h%x;' % (n - 1, p, n, bits_to_int(bits)))
        L.append('  wire i_%s = S_%s[cyc];' % (p, p))
        conns.append('.%s(i_%s)' % (p, p))
    for p in outputs:
        conns.append('.%s(o_%s)' % (p, p))
    L.append('  %s u (%s);' % (dut, ', '.join(conns)))
    L.append('endmodule')
    return '\n'.join(L) + '\n'


def run_sims(yosys, workdir, design_files, jobs, gate_netlist=None, liberty=None, tag='rtl'):
    """Simulate every job (dict name, tb, ncycles) in one Yosys run. RTL by default; with
    gate_netlist + liberty the synthesized netlist is simulated on the liberty cell functions.
    Returns (name -> SimResult, seconds)."""
    os.makedirs(workdir, exist_ok=True)
    lines = []
    # Gate level: the liberty file supplies the cell functions so the mapped netlist can be simulated
    if gate_netlist:
        lines.append('read_liberty -ignore_miss_func %s' % liberty)
        lines.append('read_verilog %s' % gate_netlist)
    else:
        lines.append('read_verilog %s' % ' '.join(design_files))
    for j in jobs:
        fn = '%s_%s.v' % (j['name'], tag)
        with open(os.path.join(workdir, fn), 'w') as f:
            f.write(make_tb(j['name'], j['dut'], j['ncycles'], j['consts'], j['streams'], j['outputs']))
        lines.append('read_verilog %s' % fn)
    # Read everything once, then restore that design for each job and simulate its testbench as top
    lines.append('design -save base')
    for j in jobs:
        lines.append('design -load base')
        lines.append('hierarchy -top %s' % j['name'])
        lines.append('proc')
        if gate_netlist:
            lines.append('flatten')
            lines.append('rename -hide w:u.*')      # keep the VCD small: only tb-level nets
        lines.append('sim -clock clk -n %d -vcd %s_%s.vcd %s' % (j['ncycles'], j['name'], tag, j['name']))
    _, dt = run_yosys(yosys, '\n'.join(lines) + '\n', workdir, log_name='sim_%s.log' % tag)
    # ncycles + 1 samples: the initial state (cycle 0) plus one sample after each clock edge
    res = {j['name']: SimResult(os.path.join(workdir, '%s_%s.vcd' % (j['name'], tag)), j['ncycles'] + 1)
           for j in jobs}
    return res, dt


# =====================================================================================
# 4b. AIG simulation: Yosys writes the DUT as an AIGER file (RTL or mapped netlist), numpy runs it.
#     An AIG (and-inverter graph) has primary inputs, latches (flip-flops with a next-state literal)
#     and 2-input AND gates; a literal is 2*variable + inversion bit. One clock cycle is: set the
#     inputs and the latch outputs, evaluate the AND gates in topological (level) order, read the
#     outputs, then load each latch with its next-state literal.
#     Bit-parallel: each test job is one bit lane of a uint64 word (up to 64 jobs per run).
#     Uninitialised flip-flops start from random values (independent per lane), so a passing
#     check also shows that the result does not depend on the power-up state.
# =====================================================================================
def export_aig(yosys, workdir, top, files=None, netlist=None, liberty=None, stem='dut'):
    """RTL (files) or gate-level (netlist + liberty cell functions) -> <stem>.aag + <stem>.map."""
    os.makedirs(workdir, exist_ok=True)
    # Gate level: read the liberty cell functions so every standard cell becomes logic, not a black box
    if netlist:
        s = ['read_liberty -ignore_miss_func %s' % os.path.abspath(liberty),
             'read_verilog %s' % os.path.abspath(netlist)]
    else:
        s = ['read_verilog %s' % ' '.join(os.path.abspath(f) for f in files)]
    # Flatten to one module, map to generic gates (techmap), then to AND/NOT only (aigmap); the .map
    # file records which AIGER input / output / latch index is which port bit
    s += ['hierarchy -top %s' % top, 'proc', 'flatten', 'techmap', 'opt_clean', 'aigmap', 'opt_clean',
          'write_aiger -ascii -map %s.map %s.aag' % (stem, stem)]
    _, dt = run_yosys(yosys, '\n'.join(s) + '\n', workdir, log_name=stem + '_aig.log')
    return os.path.join(workdir, stem + '.aag'), os.path.join(workdir, stem + '.map'), dt


class AigSim:
    """Cycle simulator for an ASCII AIGER file written by Yosys write_aiger (with its -map file).

    Every signal is a numpy uint64 word: bit `lane` of the word is the value in test job `lane`, so
    one pass over the gates simulates up to 64 independent jobs (bit-parallel simulation)."""
    def __init__(self, aag, mapfile):
        lines = open(aag).read().split('\n')
        # Header 'aag M I L O A': max variable index, numbers of inputs, latches, outputs, AND gates.
        # Then I input literals, L latch lines (lit next [init]), O output literals, A AND lines
        # (lhs rhs0 rhs1); the sections follow each other, hence the 1 + I + L + ... offsets.
        _, M, I, L, O, A = lines[0].split()[:6]
        M, I, L, O, A = int(M), int(I), int(L), int(O), int(A)
        self.inputs = [int(lines[1 + i]) >> 1 for i in range(I)]
        lat = [list(map(int, lines[1 + I + k].split())) for k in range(L)]
        self.latch_var = np.array([x[0] >> 1 for x in lat], dtype=np.int64)
        self.latch_next = np.array([x[1] for x in lat], dtype=np.int64)
        self.latch_init = [x[2] if len(x) > 2 else 0 for x in lat]    # 0, 1, or own literal = undefined
        self.out_lit = np.array([int(lines[1 + I + L + k]) for k in range(O)], dtype=np.int64)
        ands = [list(map(int, lines[1 + I + L + O + k].split())) for k in range(A)]
        self.M = M
        self.in_map, self.out_map = {}, {}          # index -> (port, bit offset)
        # Map file lines: 'input|output <index> <bit offset> <port name>' (latch lines are ignored)
        for line in open(mapfile):
            f = line.split()
            if len(f) >= 4 and f[0] in ('input', 'output'):
                (self.in_map if f[0] == 'input' else self.out_map)[int(f[1])] = (f[3], int(f[2]))
        for m in (self.in_map, self.out_map):      # the map prints offset + start index ([1:80] -> 1..80)
            low = {}
            for port, bit in m.values():
                low[port] = min(bit, low.get(port, bit))
            for k, (port, bit) in list(m.items()):
                m[k] = (port, bit - low[port])
        # levelise the AND gates (inputs and latches are level 0): level(g) = 1 + max(level of its
        # two operands). All gates of one level are independent and are evaluated as one vector op.
        level = np.zeros(M + 1, dtype=np.int64)
        defs = {a[0] >> 1: a for a in ands}
        done = np.zeros(M + 1, dtype=bool)
        done[0] = True
        done[self.inputs] = True
        done[self.latch_var] = True
        for a in ands:                               # iterative DFS, robust to any gate order
            stack = [a[0] >> 1]
            while stack:
                v = stack[-1]
                if done[v]:
                    stack.pop()
                    continue
                r0, r1 = defs[v][1] >> 1, defs[v][2] >> 1
                pend = [r for r in (r0, r1) if not done[r]]
                if pend:
                    stack.extend(pend)
                    continue
                level[v] = 1 + max(level[r0], level[r1])
                done[v] = True
                stack.pop()
        # Per level: (output var, operand-0 var, operand-0 inverted, operand-1 var, operand-1 inverted)
        self.levels = []
        if ands:
            arr = np.array(ands, dtype=np.int64)
            lv = level[arr[:, 0] >> 1]
            for k in range(1, int(lv.max()) + 1):
                g = arr[lv == k]
                self.levels.append((g[:, 0] >> 1, g[:, 1] >> 1, (g[:, 1] & 1).astype(bool),
                                    g[:, 2] >> 1, (g[:, 2] & 1).astype(bool)))

    def run(self, in_words, seed=0):
        """in_words: (ncycles, n_inputs) uint64 lane-packed input values per cycle.
        Returns (outputs (ncycles, n_outputs) uint64, latch states (ncycles, n_latches) uint64)."""
        ones = np.uint64(0xFFFFFFFFFFFFFFFF)
        rng = np.random.default_rng(seed)
        # Random 64-bit power-up state per latch, one independent bit per lane (63 random bits shifted
        # up + 1 random LSB); latches with a defined init value get 0 / all ones in every lane
        st = rng.integers(0, 2 ** 63, size=len(self.latch_var), dtype=np.uint64) * np.uint64(2) + \
            rng.integers(0, 2, size=len(self.latch_var), dtype=np.uint64)
        for k, ini in enumerate(self.latch_init):
            if ini in (0, 1):
                st[k] = ones if ini else np.uint64(0)
        # Inversion flags become XOR masks (0 or all ones), so a literal's value is v[var] ^ mask
        lv = [(o, a, np.where(an, ones, np.uint64(0)), b, np.where(bn, ones, np.uint64(0)))
              for o, a, an, b, bn in self.levels]
        nxt_v, nxt_m = self.latch_next >> 1, np.where(self.latch_next & 1, ones, np.uint64(0))
        out_v, out_m = self.out_lit >> 1, np.where(self.out_lit & 1, ones, np.uint64(0))
        inp = np.array(self.inputs, dtype=np.int64)
        v = np.zeros(self.M + 1, dtype=np.uint64)   # value word of every variable; v[0] = constant 0
        nc = in_words.shape[0]
        outs = np.zeros((nc, len(out_v)), dtype=np.uint64)
        states = np.zeros((nc, len(st)), dtype=np.uint64)
        # One iteration = one clock cycle: outputs and states are sampled before the clock edge,
        # then all latches load their next-state literal at once
        for c in range(nc):
            v[inp] = in_words[c]
            v[self.latch_var] = st
            for o, a, am, b, bm in lv:
                v[o] = (v[a] ^ am) & (v[b] ^ bm)
            outs[c] = v[out_v] ^ out_m
            states[c] = st
            st = v[nxt_v] ^ nxt_m
        return outs, states


class AigLane:
    """Per-job view of an AigSim run (same interface as SimResult)."""
    def __init__(self, sim, outs, states, lane, ncycles):
        self.sim, self.lane, self.ncycles = sim, lane, ncycles
        self.outs, self.states = outs, states
        self._ports = {}
        for idx, (port, bit) in sim.out_map.items():
            self._ports.setdefault(port, []).append((idx, bit))

    def sig(self, name):
        """Integer per cycle of output port `name` ('o_' prefix of the testbench naming accepted)
        in this job's lane, assembled from the port's output bits."""
        port = name[2:] if name.startswith('o_') else name
        val = np.zeros(self.ncycles, dtype=object)
        for idx, bit in self._ports[port]:
            b = ((self.outs[:self.ncycles, idx] >> np.uint64(self.lane)) & np.uint64(1)).astype(object)
            val = val + (b << bit)
        return [int(x) for x in val]

    def ff_toggles(self, scope_prefix, c0, c1):
        """Flip-flop toggles per cycle between cycles c0 and c1, counted over all latches of the
        flattened DUT (scope_prefix is accepted for interface compatibility and not used)."""
        s = (self.states[c0:c1 + 1] >> np.uint64(self.lane)) & np.uint64(1)
        tog = int(np.count_nonzero(s[1:] != s[:-1]))
        return tog / max(1, c1 - c0), s.shape[1]


def run_aig_sims(yosys, workdir, top, files, jobs, netlist=None, liberty=None, tag='rtl', seed=0):
    """Export the DUT once, then run all jobs as bit lanes. Returns (name -> AigLane, seconds)."""
    t0 = time.time()
    aag, mp, _ = export_aig(yosys, workdir, top, files=files, netlist=netlist, liberty=liberty,
                            stem='%s_%s' % (top, tag))
    sim = AigSim(aag, mp)
    res = {}
    # Groups of up to 64 jobs share one simulation; job number `lane` within the group drives bit
    # `lane` of every input word
    for g in range(0, len(jobs), 64):
        grp = jobs[g:g + 64]
        nc = max(j['ncycles'] for j in grp) + 1
        words = np.zeros((nc, len(sim.inputs)), dtype=np.uint64)
        for lane, j in enumerate(grp):
            lb = np.uint64(1) << np.uint64(lane)
            # Constant inputs set the lane bit in every cycle; 1-bit streams only in the cycles where
            # the stream is 1. Inputs that are neither (e.g. clk, absorbed by the AIG) stay 0.
            for idx, (port, bit) in sim.in_map.items():
                if port in j['consts']:
                    if (j['consts'][port][1] >> bit) & 1:
                        words[:, idx] |= lb
                elif port in j['streams']:
                    s = np.array((list(j['streams'][port]) + [0] * nc)[:nc], dtype=bool)
                    words[s, idx] |= lb
        outs, states = sim.run(words, seed=seed + g)  # a different power-up state per group
        for lane, j in enumerate(grp):
            res[j['name']] = AigLane(sim, outs, states, lane, j['ncycles'] + 1)
    return res, time.time() - t0


# =====================================================================================
# 5. Verification jobs: each job = one testbench + a check(sim) -> [(label, ok, detail)]
#    A job holds the stimulus (constant inputs, per-cycle input streams), the outputs to record
#    and a closure check(sim) that compares the recorded outputs with the expected values: a
#    published test vector or the Python reference model. The same job runs on the RTL and on the
#    synthesized netlist, with either engine. Cycle k of a trace is the state after clock edge k,
#    so an input pulse in cycle L takes effect in cycle L + 1.
# =====================================================================================
def _pulses(n, cycles):
    """n-cycle 0/1 stream that is 1 only in the listed cycles."""
    v = [0] * n
    for c in cycles:
        v[c] = 1
    return v


def _job(name, dut, ncycles, consts, streams, outputs, check, window=None):
    """window = (c0, c1): steady-state cycles used for the flip-flop toggle count."""
    return dict(name=name, dut=dut, ncycles=ncycles, check=check, window=window,
                consts=consts, streams=streams, outputs=outputs)


def _clean(bits):
    """Per-cycle samples -> 0/1 list (an x sample becomes 2 so it can never match)."""
    return [b if b in (0, 1) else 2 for b in bits]


def _first(lst, val, start=0):
    """Index of the first lst[k] == val with k >= start, or None."""
    for k in range(start, len(lst)):
        if lst[k] == val:
            return k
    return None


def jobs_sdsm(rng, dut='sdsm_ks'):
    """S-DSM jobs (also used for sdsm_lp): two seed sets, one with key16 = 0 to exercise the
    zero-seed replacement. A random modulator stream d is applied; the check compares c with
    d XOR LFSR16(key16) and both dither outputs with LFSR16(seed1) / LFSR16(seed2), bit by bit."""
    jobs = []
    for j, (k16, s1, s2) in enumerate([(0xACE1, 0x1234, 0xBEEF), (0x0000, 0xFFFF, 0x0000)]):
        nc, L = 260, 2                         # load in cycle L -> first LFSR bit in cycle L+1
        n = nc - L - 2
        d = [rng.getrandbits(1) for _ in range(nc + 2)]
        ks, e1, e2 = LFSR16(k16).bits(n), LFSR16(s1).bits(n), LFSR16(s2).bits(n)

        def check(sim, d=d, ks=ks, e1=e1, e2=e2, n=n, lab='seeds %04x/%04x/%04x' % (k16, s1, s2)):
            r = slice(L + 1, L + 1 + n)
            c, d1, d2 = _clean(sim.sig('o_c'))[r], _clean(sim.sig('o_dith1'))[r], _clean(sim.sig('o_dith2'))[r]
            exp_c = [d[L + 1 + k] ^ ks[k] for k in range(n)]
            return [('c = d ^ LFSR16(key16), %d bits, %s' % (n, lab), c == exp_c, ''),
                    ('dither LFSR16 x 2, %d bits each, %s' % (n, lab), d1 == e1 and d2 == e2, '')]
        # The first seed set also provides the steady-state window for the flip-flop toggle count
        jobs.append(_job('tb_sdsm%d' % j, dut, nc,
                         {'key16': (16, k16), 'seed1': (16, s1), 'seed2': (16, s2)},
                         {'load': _pulses(nc + 2, [L]), 'd': d}, {'c': 1, 'dith1': 1, 'dith2': 1},
                         check, window=(L + 10, nc - 2) if j == 0 else None))
    return jobs


def jobs_trivium(rng=None, dut='trivium_r1', init=1152):
    """Trivium jobs, one per eSTREAM vector (Sets 1, 3, 5, 6). init: initialisation clocks after
    the load (1152 for trivium_r1, 72 for trivium_lp at 16 steps per clock). The check requires
    ready to rise exactly init clocks after the load edge, then compares the keystream from that
    cycle on with every quoted stream range and, for 4096-bit runs, the xor-digest."""
    jobs = []
    for j, v in enumerate(TRIVIUM_VECTORS):
        nbits = 4096 if v['digest'] else 512  # 512 bytes for the digest; Set 6 quotes only stream[0..63]
        L = 1
        c0 = L + 1 + init                      # load edge + init clocks (1152 at 1 step per clock)
        nc = c0 + nbits + 1
        # Convert the eSTREAM key / IV to spec order: port bit key[i] then loads s_i (port [1:80])
        key = bits_to_int(trivium_estream_bits(v['key']))
        iv = bits_to_int(trivium_estream_bits(v['iv']))

        def check(sim, v=v, nbits=nbits, c0=c0, L=L):
            rdy, z = sim.sig('o_ready'), _clean(sim.sig('o_z'))
            first = _first(rdy, 1, L + 1)
            res = [('%s: ready at cycle %d (load + %d init clocks)' % (v['name'], c0, c0 - L - 1),
                    first == c0 and all(r == 0 for r in rdy[L + 1:c0]), 'got %s' % first)]
            if first is not None:
                zs = z[first:first + nbits]
                # An x sample (2 in the cleaned list) fails every range check
                for lab, ok in estream_check(trivium_bits_to_bytes(zs), v):
                    res.append(('%s: %s' % (v['name'], lab), ok and 2 not in zs, ''))
            return res
        jobs.append(_job('tb_triv%d' % j, dut, nc, {'key': (80, key), 'iv': (80, iv)},
                         {'load': _pulses(nc + 2, [L])}, {'z': 1, 'ready': 1}, check,
                         window=(c0 + 10, c0 + 2000) if j == 1 else None))
    return jobs


def jobs_grain_enc1x(rng=None):
    """Grain-128a encryption-only jobs, one per paper test vector TV1..TV4: ready must rise 256
    init clocks after the load, and the next 320 output bits must equal the published
    pre-output y_0..y_319 (with IV_0 = 1 in TV3 / TV4 this core still outputs the pre-output)."""
    jobs = []
    for n, tv in enumerate(GRAIN_TEST_VECTORS):
        L, nbits = 1, 320
        c0 = L + 257                           # load edge + 256 init clocks
        nc = c0 + nbits + 1
        pre = re.sub(r'\s+', '', tv['pre'])

        # The key / IV hex literals map directly to the [0:127] / [0:95] ports (k_0 = MSB)
        def check(sim, pre=pre, c0=c0, n=n, L=L):
            rdy, z = sim.sig('o_ready'), _clean(sim.sig('o_z'))
            first = _first(rdy, 1, L + 1)
            return [('TV%d: ready at cycle %d (load + 256 init clocks)' % (n + 1, c0),
                     first == c0 and all(r == 0 for r in rdy[L + 1:c0]), 'got %s' % first),
                     ('TV%d: pre-output y_0..y_319 = %s...' % (n + 1, pre[:16]),
                      first is not None and bits_to_hex(z[first:first + 320]) == pre, '')]
        jobs.append(_job('tb_genc%d' % n, 'grain128a_enc1x', nc,
                         {'key': (128, int(re.sub(r'\s+', '', tv['key']), 16)),
                          'iv': (96, int(re.sub(r'\s+', '', tv['iv']), 16))},
                         {'load': _pulses(nc + 2, [L])}, {'z': 1, 'ready': 1}, check,
                         window=(c0 + 10, nc - 1) if n == 1 else None))
    return jobs


RUN_DELAY = 161          # auth2x: load edge + 128 init clocks + 32 MAC preload clocks


def jobs_grain_auth2x(rng, dut='grain128a_auth2x', run_delay=RUN_DELAY):
    """Grain-128a authentication-mode jobs (grain128a_auth2x, and grain128a_lp with run_delay =
    RUN_DELAY_LP). Ten published cases (TV3 / TV4 x messages m0..m4: keystream, macstream and tag)
    plus one random key / IV / 400-bit message checked against the Python model. Each check: RUN
    rises exactly run_delay cycles after the load, the keystream z and macstream a from that cycle
    match, and the tag is valid one cycle after the pad pulse and stays held to the end."""
    cases = []
    for t in (2, 3):                           # TV3, TV4 (authentication mode) x messages m0..m4
        tv = GRAIN_TEST_VECTORS[t]
        for j in range(5):
            cases.append(dict(lab='TV%d m%d' % (t + 1, j), key=hex_to_bits(tv['key']), iv=hex_to_bits(tv['iv']),
                              msg=grain_msg_bits(j), nks=128,
                              ks=re.sub(r'\s+', '', tv['keystream']), ms=re.sub(r'\s+', '', tv['macstream']),
                              tag=tv['tags'][j]))
    key = [rng.getrandbits(1) for _ in range(128)]
    iv = [1] + [rng.getrandbits(1) for _ in range(95)]
    msg = [rng.getrandbits(1) for _ in range(400)]
    ks = grain_keystream(key, iv, 400)
    cases.append(dict(lab='random key/IV, 400-bit message', key=key, iv=iv, msg=msg, nks=400,
                      ks=bits_to_hex(ks['enc']), ms=bits_to_hex(ks['auth'][:400]), tag='%08x' % grain_tag(msg, ks)))
    jobs = []
    for i, cs in enumerate(cases):
        # Load pulse in cycle L; message bit m_k is applied in RUN cycle r0 + k and pad in r0 + Lm
        L = 1
        r0 = L + run_delay
        Lm = len(cs['msg'])
        nc = r0 + max(cs['nks'], Lm + 2) + 2
        m = [0] * (nc + 2)
        for k, b in enumerate(cs['msg']):
            m[r0 + k] = b

        def check(sim, cs=cs, r0=r0, Lm=Lm, L=L):
            run, z, a = sim.sig('o_run'), _clean(sim.sig('o_z')), _clean(sim.sig('o_a'))
            tag, ok_ = sim.sig('o_tag'), sim.sig('o_tag_ok')
            nks = cs['nks']
            sub = {'RUN starts at cycle %d' % r0: _first(run, 1, L + 1) == r0 and all(r == 0 for r in run[L + 1:r0]),
                   'keystream z_0..z_%d' % (nks - 1): bits_to_hex(z[r0:r0 + nks]) == cs['ks'],
                   'macstream (%d bits)' % nks: bits_to_hex(a[r0:r0 + nks]) == cs['ms'],
                   'tag = %s' % cs['tag']: ok_[r0 + Lm + 1] == 1 and tag[r0 + Lm + 1] == int(cs['tag'], 16),
                   'tag held': ok_[-1] == 1 and tag[-1] == int(cs['tag'], 16)}
            bad = [k for k, v in sub.items() if not v]
            return [('%s: %s' % (cs['lab'], ', '.join(sub)), not bad,
                     'FAILED: ' + '; '.join(bad) if bad else 'tag %08x' % (tag[r0 + Lm + 1] or 0))]
        jobs.append(_job('tb_gauth%d' % i, dut, nc,
                         {'key': (128, bits_to_int(cs['key'])), 'iv': (96, bits_to_int(cs['iv']))},
                         {'load': _pulses(nc + 2, [L]), 'm': m, 'pad': _pulses(nc + 2, [r0 + Lm])},
                         {'z': 1, 'a': 1, 'run': 1, 'tag_ok': 1, 'tag': 32}, check,
                         window=(r0 + 5, r0 + Lm) if cs['lab'].startswith('random') else None))
    return jobs


def _proposal_job(name, key, frames, seeds, nks, tv=None, window_frame=None, dut='secure_ks_top', run_delay=RUN_DELAY):
    """frames: list of (session or None to keep the session, message bits); a session is a nonce N or a pair (N, N_c)
    (the chip nonce defaults to 0). tv: expected keystream / tags from a paper test vector (key = 0, N = 0, ctr = 0,
    N_c = 0).

    Builds the whole serial protocol of secure_ks_top as input streams: dither seed load (cycle 1),
    128-clock key shift-in, then per session the 32-bit nonce and 31-bit chip-nonce shift-ins, and
    per frame a start pulse, the message bits and the pad pulse. The expected keystream and tag of
    every frame come from grain_keystream / grain_tag with IV = iv_bits(nonce, ctr, cn), where ctr
    counts frames from 0 within a session (the hardware presets ctr to -1 on the nonce shift)."""
    st = {p: [0] * 20000 for p in ('key_sh', 'key_in', 'nonce_sh', 'nonce_in', 'cn_sh', 'cn_in', 'start', 'm', 'pad',
                                   'dith_load')}
    st['dith_load'][1] = 1
    for j in range(128):                       # key from the fuzzy extractor, k_0 first
        st['key_sh'][2 + j], st['key_in'][2 + j] = 1, key[j]
    c, exp, nonce, cn, ctr = 130, [], None, 0, None   # c: next free cycle (after the key shift-in)
    for f, (nn, msg) in enumerate(frames):
        if nn is not None:                     # new session: nonce MSB first (ctr preset to -1), then the chip nonce
            nonce, cn = nn if isinstance(nn, tuple) else (nn, 0)
            ctr = -1
            for j in range(32):
                st['nonce_sh'][c + j], st['nonce_in'][c + j] = 1, (nonce >> (31 - j)) & 1
            c += 32
            for j in range(31):
                st['cn_sh'][c + j], st['cn_in'][c + j] = 1, (cn >> (30 - j)) & 1
            c += 31
        ctr += 1                               # frame counter of this frame (0 for the first of a session)
        st['start'][c] = 1
        r0 = c + run_delay
        for k, b in enumerate(msg):
            st['m'][r0 + k] = b
        st['pad'][r0 + len(msg)] = 1
        ks = grain_keystream(key, iv_bits(nonce, ctr, cn), max(nks, len(msg)))
        e = dict(frame=f, nonce=nonce, cn=cn, ctr=ctr, r0=r0, L=len(msg), ks=bits_to_hex(ks['enc'][:nks]),
                 tag='%08x' % grain_tag(msg, ks))
        if tv is not None:
            e['tv_ks'], e['tv_tag'] = re.sub(r'\s+', '', tv['keystream']), tv['tags'][f]
        exp.append(e)
        # Next frame starts after nks keystream bits (and after the tag) have been observed
        c = r0 + max(nks, len(msg) + 2) + 1
    nc = c + 2
    streams = {p: v[:nc + 2] for p, v in st.items()}

    def check(sim):
        run, z, tag, ok_, cv = sim.sig('o_run'), _clean(sim.sig('o_z')), sim.sig('o_tag'), sim.sig('o_tag_ok'), sim.sig('o_ctr')
        res = []
        # Per frame: RUN rises exactly run_delay cycles after start, the counter output equals the
        # frame's ctr, keystream and tag match the model (and, with tv, the paper's TV3 values)
        for e in exp:
            r0, L = e['r0'], e['L']
            sub = {'RUN at cycle %d' % r0: run[r0] == 1 and all(r == 0 for r in run[r0 - run_delay + 1:r0]),
                   'ctr = %d' % e['ctr']: cv[r0] == e['ctr'],
                   'keystream %d bits' % nks: bits_to_hex(z[r0:r0 + nks]) == e['ks'],
                   'tag = %s' % e['tag']: ok_[r0 + L + 1] == 1 and tag[r0 + L + 1] == int(e['tag'], 16)}
            if 'tv_ks' in e:
                sub['= paper TV3 keystream and tag(m%d)' % e['frame']] = (
                    bits_to_hex(z[r0:r0 + 128]) == e['tv_ks'] and tag[r0 + L + 1] == int(e['tv_tag'], 16))
            bad = [k for k, v in sub.items() if not v]
            res.append(('frame %d (nonce %08x, chip nonce %08x, ctr %d, %d-bit msg): %s' % (
                e['frame'], e['nonce'], e['cn'], e['ctr'], L, ', '.join(sub)), not bad, 'FAILED: ' + '; '.join(bad) if bad else ''))
        # Dither LFSRs: loaded in cycle 1, so their first output bit is in cycle 2
        n = nc - 3
        d1, d2 = _clean(sim.sig('o_dith1'))[2:2 + n], _clean(sim.sig('o_dith2'))[2:2 + n]
        res.append(('dither LFSR16 x 2 (seeds %04x/%04x), %d bits each' % (seeds[0], seeds[1], n),
                    d1 == LFSR16(seeds[0]).bits(n) and d2 == LFSR16(seeds[1]).bits(n), ''))
        return res
    win = None
    if window_frame is not None:
        e = exp[window_frame]
        win = (e['r0'] + 5, e['r0'] + e['L'])
    return _job(name, dut, nc, {'seed1': (16, seeds[0]), 'seed2': (16, seeds[1])}, streams,
                {'z': 1, 'run': 1, 'tag_ok': 1, 'tag': 32, 'ctr': 32, 'dith1': 1, 'dith2': 1}, check, window=win)


def jobs_proposal(rng, dut='secure_ks_top', run_delay=RUN_DELAY):
    """Two proposal jobs. tb_prop0: key 0, five frames carrying the paper messages m0..m4, each
    opened as a new session with nonce 0 and chip nonce 0, so every frame has ctr = 0 and
    IV = 1 | 0^95 = TV3's IV; the keystream and tag(m_f) of frame f are compared with the model
    and with the paper's TV3 values. tb_prop1: random key, two sessions with random nonces
    and chip nonces; session 1 carries three frames (one 400-bit message for the toggle window),
    session 2 a 1-bit message; all checked against the Python model."""
    tv3 = GRAIN_TEST_VECTORS[2]                # key 0, IV 8000.. = (IV_0 = 1, nonce 0, ctr 0, chip nonce 0)
    p1 = _proposal_job('tb_prop0', [0] * 128, [(0, grain_msg_bits(j)) for j in range(5)],
                       (0xACE1, 0x0000), 128, tv=tv3, dut=dut, run_delay=run_delay)
    key = [rng.getrandbits(1) for _ in range(128)]
    msgs = [[rng.getrandbits(1) for _ in range(n)] for n in (rng.randint(60, 90), 400, rng.randint(60, 90), 1)]
    n1, n2 = rng.getrandbits(32), rng.getrandbits(32)
    c1, c2 = rng.getrandbits(31), rng.getrandbits(31)
    p2 = _proposal_job('tb_prop1', key, [((n1, c1), msgs[0]), (None, msgs[1]), (None, msgs[2]), ((n2, c2), msgs[3])],
                       (0x1234, 0xBEEF), 100, window_frame=1, dut=dut, run_delay=run_delay)
    return [p1, p2]


RUN_DELAY_LP = 11       # low-power Grain-128a: load edge + 8 init updates (32 bits each) + 2 MAC preload updates
# design key -> job builder(rng); the low-power designs reuse the builders of their 1-bit-per-clock
# counterparts with the low-power DUT name and latency, so both must match the same expected data
JOB_BUILDERS = {'sdsm': jobs_sdsm, 'trivium': jobs_trivium, 'grain_enc1x': jobs_grain_enc1x,
                'grain_auth2x': jobs_grain_auth2x, 'proposal': jobs_proposal,
                'sdsm_lp': lambda rng: jobs_sdsm(rng, dut='sdsm_lp'),
                'trivium_lp': lambda rng: jobs_trivium(rng, dut='trivium_lp', init=72),
                'grain_lp': lambda rng: jobs_grain_auth2x(rng, dut='grain128a_lp', run_delay=RUN_DELAY_LP),
                'proposal_lp': lambda rng: jobs_proposal(rng, dut='secure_ks_lp', run_delay=RUN_DELAY_LP)}


def verify_design(yosys, key, src_dir, workdir, netlist=None, liberty=None, seed=2027, engine='aig'):
    """Run the test jobs of design `key` on the RTL (or on a synthesized netlist when netlist and
    liberty are given). engine='aig': Yosys exports an AIG, simulated here in numpy (works with
    yowasp-yosys, which has no `sim`); engine='yosys-sim': Verilog testbenches run by Yosys `sim`.
    Returns dict(checks=[...], seconds, toggles=(per clock, FF bits) or None)."""
    # Fixed seed: the random keys, IVs and messages (and the AIG power-up states) are reproducible
    jobs = JOB_BUILDERS[key](random.Random(seed))
    files = [os.path.abspath(os.path.join(src_dir, f)) for f in DESIGNS[key][1]]
    tag = 'gate' if netlist else 'rtl'
    if engine == 'aig':
        sims, dt = run_aig_sims(yosys, workdir, DESIGNS[key][0], files, jobs, netlist=netlist,
                                liberty=liberty, tag=tag, seed=seed)
    else:
        sims, dt = run_sims(yosys, workdir, files, jobs, gate_netlist=netlist and os.path.abspath(netlist),
                            liberty=liberty, tag=tag)
    checks, toggles = [], None
    for j in jobs:
        try:
            for lab, ok, det in j['check'](sims[j['name']]):
                checks.append(dict(design=key, level=tag, engine=engine, check=lab, ok=bool(ok), detail=det))
        except Exception as ex:                # a malformed / x-filled trace counts as a failure
            checks.append(dict(design=key, level=tag, engine=engine, check=j['name'], ok=False,
                               detail='error: %r' % ex))
        # Flip-flop activity (switching-power proxy) from the RTL run of the job that defines a window
        if j['window'] and not netlist:
            toggles = sims[j['name']].ff_toggles(j['name'] + '.u.', *j['window'])
    return dict(checks=checks, seconds=dt, toggles=toggles)


# =====================================================================================
# 6. Synthesis (Yosys -> sky130_fd_sc_hd) and statistics
# =====================================================================================
def liberty_cells(liberty):
    """cell name -> dict(area, ff, latch) from a liberty file (area in um^2; ff / latch: the cell
    has an ff / latch group). A text scan, not a full liberty parser: each cell body runs from its
    `cell (...) {` header to the next cell header."""
    txt = open(liberty).read()
    heads = list(re.finditer(r'\n\s*cell\s*\(\s*"?(\w+)"?\s*\)\s*\{', txt))
    out = {}
    for i, m in enumerate(heads):
        body = txt[m.end(): heads[i + 1].start() if i + 1 < len(heads) else len(txt)]
        a = re.search(r'\barea\s*:\s*([\d.]+)', body)
        out[m.group(1)] = dict(area=float(a.group(1)) if a else 0.0,
                               ff=bool(re.search(r'\n\s*ff\s*\(', body)),
                               latch=bool(re.search(r'\n\s*latch\s*\(', body)))
    return out


# Special-purpose cells kept out of mapping, as in OpenLane / OpenROAD flows: the lpflow_* power-
# isolation cells (ABC otherwise uses lpflow_isobufsrc / lpflow_inputiso1p as AND/OR gates; same
# 6.256 um2 as and2_0 / or2_0, so the area effect is ~0). Pass dont_use=() for the plain script.
DONT_USE = ('sky130_fd_sc_hd__lpflow_*',)


# Clock gating (option of synth_script): every group of at least CG_MIN flip-flops that share a clock enable gets one
# integrated clock-gating cell (latch + AND, sky130_fd_sc_hd__dlclkp_1) and loses its enable multiplexer
CG_CELL, CG_MIN = 'sky130_fd_sc_hd__dlclkp_1 GATE:CLK:GCLK', 4


def synth_script(files, top, liberty, flatten=True, netlist=None, stem='synth', dont_use=DONT_USE, clock_gating=False, defines=()):
    """Yosys script text: generic `synth`, optional clock gating, flip-flop mapping (dfflibmap) and
    combinational mapping (abc) onto the liberty cells, then area statistics as text and JSON
    (<stem>_stat.txt / .json) and, if netlist is given, the mapped Verilog netlist.

    defines: Verilog macros (e.g. ICG_CELLS); with any define the liberty is also read as a cell
    library first so instantiated standard cells (dlclkp_1) are known as black boxes."""
    lib = liberty
    du = ''.join('-dont_use %s ' % c for c in dont_use)
    s = (['read_liberty -lib %s' % lib] if defines else []) + ['read_verilog %s%s' % (''.join('-D%s ' % d for d in defines), ' '.join(files)),
         'synth %s-top %s' % ('-flatten ' if flatten else '', top)]
    if clock_gating:
        s.append('clockgate -pos %s -min_net_size %d' % (CG_CELL, CG_MIN))
    s += ['dfflibmap %s-liberty %s' % (du, lib),
         'abc %s-liberty %s' % (du, lib),
         'opt_clean',
         'tee -o %s_stat.txt stat -liberty %s' % (stem, lib),
         'tee -q -o %s_stat.json stat -json -liberty %s' % (stem, lib)]
    if netlist:
        s.append('write_verilog -noattr %s' % os.path.abspath(netlist))
    return '\n'.join(s) + '\n'


def _modname(name):
    """'$paramod\\grain_core\\W=s32'00..10' -> 'grain_core(W=2)'."""
    name = name.lstrip('\\')
    m = re.match(r'\$paramod\\(\w+)\\(.*)$', name)
    if not m:
        return name
    ps = []
    for kv in m.group(2).split('\\'):
        k, v = kv.split('=', 1)
        b = re.match(r"s?\d+'([01]+)$", v)
        ps.append('%s=%d' % (k, int(b.group(1), 2)) if b else kv)
    return '%s(%s)' % (m.group(1), ','.join(ps))


def _read_stat_json(path):
    """Parse a `stat -json` file, skipping any text Yosys printed before the JSON object."""
    txt = open(path).read()
    return json.loads(txt[txt.index('{'):])


def synthesize(yosys, top, files, liberty, workdir, flatten=True, netlist=None, cells=None, stem=None,
               dont_use=DONT_USE, clock_gating=False, defines=()):
    """Synthesize `top` and return its statistics. flatten=True: one netlist, totals.
    flatten=False: also per-module local areas (for the breakdown)."""
    cells = cells or liberty_cells(liberty)
    nand2 = cells['sky130_fd_sc_hd__nand2_1']['area']
    stem = stem or ('%s_%s' % (top, 'flat' if flatten else 'hier'))
    files = [os.path.abspath(f) for f in files]
    _, dt = run_yosys(yosys, synth_script(files, top, os.path.abspath(liberty), flatten, netlist, stem, dont_use,
                                          clock_gating, defines),
                      workdir, log_name=stem + '.log')
    st = _read_stat_json(os.path.join(workdir, stem + '_stat.json'))

    # Area from the liberty cell areas (um^2) and in gate equivalents (GE = area / nand2_1 area);
    # entries of num_cells_by_type that are not liberty cells are submodule instances
    def summarize(by_type):
        lib = {c: n for c, n in by_type.items() if c in cells}
        area = sum(cells[c]['area'] * n for c, n in lib.items())
        return dict(area_um2=round(area, 4), ge=round(area / nand2, 1),
                    cells=sum(lib.values()), ffs=sum(n for c, n in lib.items() if cells[c]['ff']),
                    latches=sum(n for c, n in lib.items() if cells[c]['latch']),
                    seq_area_um2=round(sum(cells[c]['area'] * n for c, n in lib.items() if cells[c]['ff'] or cells[c]['latch']), 4),
                    by_type=dict(sorted(lib.items())),
                    submodules={_modname(c): n for c, n in by_type.items()
                                if c not in cells and not c.startswith('$scopeinfo')})
    d = st['design']
    out = dict(top=top, flatten=flatten, seconds=round(dt, 2), nand2_um2=nand2, dont_use=list(dont_use),
               yosys_area_um2=d.get('area'), **summarize(d.get('num_cells_by_type', {})))
    if not flatten:
        mods = {_modname(name): summarize(m.get('num_cells_by_type', {})) for name, m in st['modules'].items()}
        # Hierarchical total = sum over modules of (local area x number of instances under the top)
        inst = {top: 1}                       # instances of each module under the top

        def walk(mn, k):
            for sub, n in mods[mn]['submodules'].items():
                inst[sub] = inst.get(sub, 0) + k * n
                walk(sub, k * n)
        walk(top, 1)
        for mn, s in mods.items():
            s['instances'] = inst.get(mn, 0)
        out['modules'] = mods
        out['hier_total_um2'] = round(sum(s['area_um2'] * s['instances'] for s in mods.values()), 4)
    return out


# What each design does per modulator clock (for the table). init_clocks counts the load edge plus
# the clocks until the first keystream bit (the RUN / ready latency checked by the jobs).
DESIGN_INFO = {
    'sdsm':         dict(ks_bits_per_clk=1, mac_bits_per_clk=0, init_clocks=1,
                         note='1-clock seed load; no IV, no MAC'),
    'trivium':      dict(ks_bits_per_clk=1, mac_bits_per_clk=0, init_clocks=1 + 1152,
                         note='load + 1152 init clocks; no MAC'),
    'grain_enc1x':  dict(ks_bits_per_clk=1, mac_bits_per_clk=0, init_clocks=1 + 256,
                         note='load + 256 init clocks; no MAC'),
    'grain_auth2x': dict(ks_bits_per_clk=1, mac_bits_per_clk=1, init_clocks=1 + 128 + 32,
                         note='load + 128 init clocks (2 bits/clk) + 32 MAC preload clocks'),
    'proposal':     dict(ks_bits_per_clk=1, mac_bits_per_clk=1, init_clocks=1 + 128 + 32,
                         note='per frame as grain_auth2x; key shift-in 128 clocks per power-up, '
                              'nonce shift-in 32 + chip-nonce shift-in 31 clocks per session'),
    'sdsm_lp':      dict(ks_bits_per_clk=1, mac_bits_per_clk=0, init_clocks=1,
                         note='as sdsm; state updated once every 16 clocks'),
    'trivium_lp':   dict(ks_bits_per_clk=1, mac_bits_per_clk=0, init_clocks=1 + 72,
                         note='load + 72 init clocks (16 steps each); state updated once every 16 clocks'),
    'grain_lp':     dict(ks_bits_per_clk=1, mac_bits_per_clk=1, init_clocks=1 + 8 + 2,
                         note='load + 8 init + 2 MAC preload clocks (32 bits each); state and MAC once every 16 clocks'),
    'proposal_lp':  dict(ks_bits_per_clk=1, mac_bits_per_clk=1, init_clocks=1 + 8 + 2,
                         note='per frame as grain_lp; registers clocked only when written'),
}


def run_all(yosys, liberty, workdir, gate_level=True, designs=None, plain_variant=True, cross_check=True,
            log=print):
    """Write the sources, check the Python references, then for every design: RTL simulation
    against the test vectors, flattened synthesis, gate-level simulation of the netlist,
    hierarchical synthesis (breakdown), plus the load-mux probes. Simulation uses the AIG engine;
    with cross_check and a Yosys that has `sim`, the Yosys simulator runs the same jobs as well.
    Returns a results dict."""
    t_all = time.time()
    # Live run: every number in the result comes from this Yosys run and these simulations
    src = os.path.join(workdir, 'src')
    write_verilog_files(src)
    cells = liberty_cells(liberty)
    engines = ['aig'] + (['yosys-sim'] if cross_check and has_command(yosys, 'sim') else [])
    res = dict(yosys=yosys, yosys_version=yosys_version(yosys), liberty=os.path.abspath(liberty),
               sim_engines=engines,
               nand2_um2=cells['sky130_fd_sc_hd__nand2_1']['area'], reference_selftest=reference_selftest(),
               designs={}, probes={})
    log('%s | nand2_1 = %.4f um2' % (res['yosys_version'], res['nand2_um2']))
    # Per design: RTL verification -> flat synthesis (netlist) -> gate-level verification of that
    # netlist -> [ICG netlist for low-power designs] -> hierarchical synthesis -> plain-script variant
    for key in designs or DESIGNS:
        top, files, desc = DESIGNS[key]
        wd = os.path.join(workdir, key)
        os.makedirs(wd, exist_ok=True)
        r = dict(top=top, description=desc, **DESIGN_INFO[key])
        r['rtl_checks'], r['rtl_sim_seconds'] = [], {}
        for eng in engines:
            rtl = verify_design(yosys, key, src, os.path.join(wd, eng), engine=eng)
            r['rtl_checks'] += rtl['checks']
            r['rtl_sim_seconds'][eng] = round(rtl['seconds'], 2)
            if rtl['toggles']:
                r.setdefault('ff_toggles_per_clk_by_engine', {})[eng] = round(rtl['toggles'][0], 3)
                if eng == 'aig':
                    r['ff_toggles_per_clk'], r['ff_bits_in_sim'] = round(rtl['toggles'][0], 2), rtl['toggles'][1]
        netlist = os.path.join(wd, top + '_netlist.v')
        r['synth'] = synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd,
                                flatten=True, netlist=netlist, cells=cells)
        if gate_level:
            r['gate_checks'], r['gate_sim_seconds'] = [], {}
            for eng in engines:
                g = verify_design(yosys, key, src, os.path.join(wd, eng), netlist=netlist,
                                  liberty=os.path.abspath(liberty), engine=eng)
                r['gate_checks'] += g['checks']
                r['gate_sim_seconds'][eng] = round(g['seconds'], 2)
        if key in LOW_POWER:                   # the implementation: explicit clock-gating cells (`define ICG_CELLS);
            # verified after place-and-route by the gate-level simulation of the same test jobs (pnr_flow.gate_sim)
            r['synth_icg'] = synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd, flatten=True,
                                        netlist=os.path.join(wd, top + '_icg_netlist.v'), cells=cells,
                                        stem=top + '_flat_icg', defines=('ICG_CELLS',))
        r['hier'] = synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd,
                               flatten=False, cells=cells)
        if plain_variant:                      # the script exactly as specified (no -dont_use)
            pv = synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd,
                            flatten=True, cells=cells, stem=top + '_flat_plain', dont_use=())
            r['synth_plain'] = {k: pv[k] for k in ('area_um2', 'ge', 'ffs', 'cells', 'by_type')}
        allc = r['rtl_checks'] + r.get('gate_checks', [])
        r['verified'] = all(c['ok'] for c in allc)
        log('%-13s %9.1f um2 %7.1f GE %4d FF %5d cells | %d/%d checks pass (%s, %s)' % (
            key, r['synth']['area_um2'], r['synth']['ge'], r['synth']['ffs'], r['synth']['cells'],
            sum(c['ok'] for c in allc), len(allc), 'RTL + gate' if gate_level else 'RTL', '+'.join(engines)))
        res['designs'][key] = r
    # Load-mux probes: synthesis only (areas for the load-multiplexer cost)
    for key, (top, files) in PROBES.items():
        wd = os.path.join(workdir, 'probes')
        os.makedirs(wd, exist_ok=True)
        p = synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd, flatten=True, cells=cells)
        res['probes'][key] = {k: p[k] for k in ('area_um2', 'ge', 'ffs', 'cells')}
    res['seconds'] = round(time.time() - t_all, 1)
    return res


def area_table(res):
    """Markdown table of the flattened synthesis results."""
    L = ['| Design | Area (um^2) | GE | Flip-flops | Cells | Keystream + MAC bits / clk | Init clocks | FF toggles / clk | Verified |',
         '|---|---:|---:|---:|---:|---|---:|---:|---|']
    for key, r in res['designs'].items():
        s = r['synth']
        L.append('| %s | %.1f | %.0f | %d | %d | %d + %d | %d | %s | %s |' % (
            r['description'], s['area_um2'], s['ge'], s['ffs'], s['cells'], r['ks_bits_per_clk'],
            r['mac_bits_per_clk'], r['init_clocks'], r.get('ff_toggles_per_clk', '-'),
            'pass' if r['verified'] else 'FAIL'))
    return '\n'.join(L)


# Single-point RTL mutations: each must make at least one check of its design fail.
# Entry: (label, file, original text, mutated text, design whose jobs must catch it). They test the
# test bench itself: an LFSR tap, a Trivium tap and its init length, a Grain h / g tap, the second
# step of the 2x core, the MAC shift input, the counter value in the IV, and the chip-nonce input.
MUTATIONS = [
    ('lfsr16',  'lfsr16.v',      's[0] ^ s[2] ^ s[3] ^ s[5]', 's[0] ^ s[2] ^ s[4] ^ s[5]', 'sdsm'),
    ('trivium', 'trivium_r1.v',  "^ s[171]",                  "^ s[170]",                  'trivium'),
    ('trivium', 'trivium_r1.v',  "11'd1151",                  "11'd1150",                  'trivium'),
    ('grain-h', 'grain128a.v',   'bb[73] ^ bb[89]',           'bb[73] ^ bb[88]',           'grain_enc1x'),
    ('grain-g', 'grain128a.v',   '(bb[22] & bb[24] & bb[25])', '(bb[22] & bb[24] & bb[26])', 'grain_enc1x'),
    ('grain-2x', 'grain128a.v',  'step(b << 1, s << 1)',      'step(b << 1, s << 2)',      'grain_auth2x'),
    ('mac',     'grain128a.v',   'R <= {R[1:31], yb};',       'R <= {R[1:31], ya};',       'grain_auth2x'),
    ('iv-ctr',  'secure_ks_top.v', 'nonce_r, ctr_n, cn_r',    'nonce_r, ctr_r, cn_r',      'proposal'),
    ('iv-cn',   'secure_ks_top.v', '.din(cn_in),    .q(cn_r)', '.din(1\'b0),    .q(cn_r)', 'proposal'),
]


def mutation_test(yosys, workdir):
    """Apply each mutation to a copy of the sources and check that verification fails."""
    out = []
    # Each mutation gets a fresh copy of all sources with exactly one text replacement (RTL, AIG engine)
    for name, fn, old, new, key in MUTATIONS:
        src = os.path.join(workdir, 'mut_%s_%s' % (name, key))
        write_verilog_files(src)
        p = os.path.join(src, fn)
        txt = open(p).read()
        assert old in txt, (name, old)
        open(p, 'w').write(txt.replace(old, new, 1))
        r = verify_design(yosys, key, src, os.path.join(src, 'sim'))
        nfail = sum(not c['ok'] for c in r['checks'])
        out.append(dict(mutation='%s: %s -> %s' % (fn, old, new), design=key, failing_checks=nfail,
                        total_checks=len(r['checks']), caught=nfail > 0))
    return out
````

### B. Verification and synthesis

The area is quoted in µm² and in gate equivalents, GE = $A / A_{\text{NAND2}}$, where $A$ is the synthesised cell area and $A_{\text{NAND2}}$ = 3.7536 µm² the area of the library's two-input NAND (`sky130_fd_sc_hd__nand2_1`, from its Liberty file). The flip-flop toggle rate (FF toggles/clk) is the average number of flip-flop outputs that change per clock while the block encrypts, **the basis of a first estimate of dynamic power**; Subsection D tests whether that estimate suffices.

Every design must reproduce the published test vectors bit for bit at both levels; the checks, the synthesis and the toggle count use these settings:

* *Levels and synthesis.* Every test runs on the RTL and on the synthesised gate netlist, its cells read with their logic functions from the Liberty file (zero-delay functional simulation). Yosys (`yowasp-yosys`, version on the result panel) maps each design onto `sky130_fd_sc_hd` at tt, 25 °C, 1.8 V with `synth -flatten`, `dfflibmap`, `abc`, `opt_clean` and `stat` against the Liberty file, excluding the `lpflow_*` power-isolation cells as the OpenLane flows do; a second, hierarchical synthesis gives the area of each module (Fig. 69(a)).
* *Simulator.* Colab's Yosys has no `sim` command, so Yosys writes each design as an AIGER graph (`proc`, `flatten`, `techmap`, `aigmap`, `write_aiger`), and a bit-parallel numpy simulator evaluates its AND gates level by level, once per clock, with one test job per bit lane of a 64-bit word. Every flip-flop starts from a random value, independent in each lane, so a passing check also shows independence from the power-up state. In an earlier run with a native Yosys 0.64, `sim` agreed with this simulator down to the toggle counts (`dev/rtl/RESULTS.md`).
* *Test vectors* (54 checks per level for the five designs written for one bit per clock, 35 for the three low-power forms; **89 per level, 178 in all**): the S-DSM (4: the encrypted bit and both dither streams against the reference LFSRs, 256 bits each, two seed sets); Trivium (20: four eSTREAM vectors, byte ranges and the 512-byte XOR digest, and the ready signal exactly 1,152 clocks after the load); Grain-128a (8: the 320 pre-output bits of TV1–TV4 of [6], ready 256 clocks after the load); Grain-128a + MAC 2× (11: TV3 and TV4 of [6] with messages m0–m4, starting 161 clocks after the load, keystream, MAC stream, tag and tag held after the padding, and a random key and IV with a 400-bit message against the Python reference); the proposed design (11: session A, key, nonce and chip nonce 0, five frames whose IV equals that of TV3, so that keystream and tags are checked against TV3 itself, and session B, random key and non-zero chip nonces, four frames with a new nonce and chip nonce in the fourth; each frame checks the start time, counter, keystream and tag, and both dither streams are checked over the whole run); and the low-power forms on their own timing (S-DSM 4; Trivium 20, ready 72 clocks after the load; the proposed design 11, each frame starting 11 clocks after the load, which verifies `grain128a_lp`). The Python reference models of Grain-128a (Table 3 and Appendix A of [6]) and Trivium (four eSTREAM vectors) are self-tested on 10 checks against the published values.
* *Toggle rate.* Counted on the RTL simulation over a steady-state window after initialisation: a random key and IV (seed 2027) and a 400-bit random message for Grain-128a + MAC and the proposed design, 2,000 clocks of an eSTREAM vector for Trivium, the keystream of TV2 for Grain-128a, and the first seed set with random modulator bits for the S-DSM; the low-power forms are counted on their enabled-flip-flop form, so that an unwritten register does not toggle.
* *Metrics:* checks passed; area (µm², GE), flip-flops, cells, keystream and MAC bits per clock, initialisation clocks and FF toggles per clock.

> **Next cell.** 1. Self-tests the reference models. 2. Simulates the RTL of every design on its 89 checks and counts the toggles. 3. Synthesises each design, flattened and hierarchically. 4. Repeats the checks on the gate netlists (178 in all). 5. Prints the result panel, the area table and the RTL. Run time: about 2 min on two cores.

### C. Place and route and sign-off of the low-power block of the proposed design

**Five sign-off checks decide whether a digital layout is ready for fabrication**: the *design-rule check* (DRC: every shape obeys the foundry's geometric rules) and the *antenna check* (no long wire connected only to a gate, which collects charge during plasma etching and can damage the oxide; repaired by diodes), both with zero violations; *layout versus schematic* (LVS: the extracted transistors and connections form the simulated netlist); the *setup slack* (clock period $T_{\text{clk}}$ less the slowest path's arrival and the setup time), positive at the slow corner (slow transistors, low supply, high temperature, maximum wire RC); and the *hold slack* (fastest arrival less the hold time), positive at the fast corner (the opposite extremes). Before routing, the wire RC is estimated from the placement or the global routes; after routing it is extracted, coupling included. With $T_{\text{clk}}$ = 100 ns and paths of a few tens of SKY130 gates, which settle within a few nanoseconds even at the slow corner, the analysis above predicts a setup slack close to the full period at every corner, a small hold slack, repaired by the flow and smallest at the fast corner, a power slightly above the estimate once the wires are extracted, and an unchanged function after routing.

All steps are at gate level, on the clock-gated netlist of the low-power block (`secure_ks_lp` with its `dlclkp` cells), with the clock tree built and timed through the gating cells. **Steps 1–3, 7 and 8 run live**. Detailed routing takes about 7 minutes on two cores (9 minutes for the whole recorded flow), so steps 4–6 and the power with extracted wires are shown from the recorded run of the same flow (`dev/rtl/block_lp_routed.json`, written by `dev/rtl/record_block_lp.py`) unless `ROUTE_BLOCK = True`; steps 7 and 8 then use the placed netlist and the wire RC estimated on the placement. Placement, routing and power use tt, 25 °C, 1.8 V; timing is checked at three corners (step 6). The flow has eight steps:

1. **Netlist preparation** (Yosys): constant drivers become tie cells (`conb_1`), multi-bit nets are split and assignments replaced by buffers (`buf_2`), the form OpenROAD reads.
2. **Floorplan and power grid** (OpenROAD): a square core at 25 % target utilisation with a 6 µm margin, pins on met3 (horizontal) and met2 (vertical), tap cells every 13 µm and end caps, met1 rails along the rows and 1.6 µm met4 stripes at a 40 µm pitch.
3. **Placement and clock tree:** global placement at a target density of 0.35, electrical repair (`repair_design`), detailed placement, clock-tree synthesis (`clkbuf_16` root, `clkbuf_4`/`clkbuf_8` buffers) and hold repair (`repair_timing -hold`) with the propagated clock at the typical corner, aiming at a hold slack of 0.30 ns as a margin for the fast corner and the extracted wires (with 0.15 ns the fast-corner hold slack ended at -0.004 ns). Every input port first gets a `buf_2` buffer at its pin (`buffer_ports -inputs`), so that the long wire from the pin drives one buffer rather than gates spread over the core.
4. **Routing:** global routing on met2–met5 (met1 left to the cells and the power rails; 30 congestion iterations), antenna repair with `diode_2` cells (three iterations on the global routes; a net counts as violating above 20 % of the allowed antenna ratio), filler cells, and detailed routing (TritonRoute) with its routing-DRC report and the antenna check. This OpenROAD version repairs antennas only on the global routes, and the detailed routes of some nets end longer. Four settings were recorded: met1–met5 at an 80 % margin (28 violations), met2–met5 (24), met2–met5 with input buffering (16), and the same with a 95 % margin and ten repair iterations (18, with 564 diodes); the third is kept. Input and output delays are 5 % of the 100 ns period.
5. **GDS, DRC and LVS.** KLayout merges the routed Design Exchange Format (DEF) file with the standard-cell layouts into a GDS. The router can leave small met3 and met4 islands where two vias meet without a wire (about 0.585 µm × 0.33 µm against the 0.24 µm² minimum area of rule met3.6); each is extended along its layer's preferred direction to the minimum area, keeping the 0.3 µm minimum spacing to all other shapes. Magic checks the GDS against the full SKY130 rule deck (`drc(full)`); for LVS, Magic extracts it hierarchically (fill and tap cells flattened) and netgen compares it with OpenROAD's routed netlist, expanded into transistors by the library's SPICE netlists: each cell type transistor by transistor, the block cell by cell and net by net.
6. **Extraction and timing at three corners.** OpenRCX extracts the resistance, ground and coupling capacitance of every routed net with the PDK's rule sets (`rules.openrcx.sky130A.{min,nom,max}`), and OpenSTA checks setup and hold with the propagated clock at slow (ss, 100 °C, 1.6 V, maximum RC), typical (tt, 25 °C, 1.8 V, nominal) and fast (ff, −40 °C, 1.95 V, minimum).
7. **Re-simulation of the routed netlist** (Icarus Verilog, the PDK's functional cell models: no delay except 1 ns clock-to-output for flip-flops and latches) on test job B of the proposed design, whose four frames' keystream, tag and counter and the dither streams must be correct, and a value-change dump (VCD) of every net over the steady-state window of the second frame (a 400-bit message).
8. **Power** (OpenSTA within OpenROAD, tt) from that VCD, the library's internal and leakage power tables and the wire capacitance extracted in step 6 (nominal rule set), and, for comparison with Subsection D, with the wire capacitance estimated on the placement as there (per-layer RC of met2 for signals and met3 for the clock); reported per group (flip-flops, clock tree, logic) and per component (internal, switching, leakage).

The metrics are the rows of the result panel. Fig. 67 shows the result of the flow: in (a) the placed core, with the standard-cell rows of step 2 running under its vertical met4 stripes, in (b) a detail of the cells at the centre, and in (c) the power of step 8 per group and component. The interactive viewer of Fig. 68 **shows the same block with each layer on or off**, which separates the cells' poly and n-wells from the power grid.

DEF to GDS (KLayout); 4. parasitic extraction (OpenRCX) and timing at three corners (OpenSTA); 5. gate-level simulation (Icarus Verilog); 6. power (OpenSTA); 7. LVS (Magic, netgen); 8. the clock-gating study.

````python
%%writefile pnr_flow.py
"""pnr_flow.py - place-and-route, sign-off checks and gate-level power of a synthesised SKY130 netlist.

Open-source tools only, each called as a program:
  Yosys     the synthesised netlist made ready for place-and-route: constants become tie cells, assigns buffers
  OpenROAD  floorplan, I/O pins, tap and end-cap cells, power grid, global and detailed placement, clock tree,
            hold repair, global routing, antenna repair, fill, detailed routing (TritonRoute); parasitic extraction of
            the routed wires (OpenRCX, with the PDK's rule files libs.tech/librelane/rules.openrcx.sky130A.{min,nom,max}),
            multi-corner timing and power (OpenSTA)
  KLayout   DEF -> GDS, with the standard cells' real layouts and the PDK's layer map
  Magic     DRC of the GDS (full SKY130 deck, sky130_layout.drc) and extraction of the GDS for LVS
  netgen    LVS: the transistor netlist extracted from the GDS against the routed netlist expanded with the cells'
            transistor netlists (sky130_fd_sc_hd.spice)
  Icarus    gate-level simulation of the routed netlist on the PDK's cell models, with the RTL flow's own checks;
            its switching activity (VCD) is what OpenSTA's power analysis uses
Where only a placement exists, wire parasitics are OpenROAD's per-layer wire RC estimate on the placed design;
after detailed routing they are extracted from the routed wires (extract()).

Typical use (notebook):
    import pnr_flow as PR
    net = PR.prep(YOSYS, netlist, LIBERTY, top, wd)
    run = PR.place_and_route(OPENROAD, PDK_ROOT, net, top, wd)
    gds = PR.to_gds(PDK_ROOT, run["def"], wd + "/core.gds")
    spef = PR.extract(OPENROAD, PDK_ROOT, run, wd)                  # routed designs only
    sta = PR.sta(OPENROAD, PDK_ROOT, run, spef, wd)                 # setup and hold at ss, tt, ff
    sim = PR.gate_sim(TOOLS_BIN, PDK_ROOT, run["verilog"], job, top, wd)
    pw  = PR.power(OPENROAD, PDK_ROOT, run, sim["activity_vcd"], wd, spef=spef["nom"])
    lvs = PR.lvs(MAGIC, NETGEN, PDK_ROOT, gds, run, top, wd)
"""
import json
import os
import random
import re
import subprocess
import time

import numpy as np

import rtl_flow as R

LIB = "sky130_fd_sc_hd"
CORNER = "tt_025C_1v80"
# Sign-off corners: library corner and the OpenRCX rule set that goes with it. Setup is checked at the slow corner with
# maximum wire RC, hold at the fast corner with minimum wire RC, and both at the typical corner.
CORNERS = {"ss": ("ss_100C_1v60", "max"), "tt": ("tt_025C_1v80", "nom"), "ff": ("ff_n40C_1v95", "min")}


def _hd(pdk):
    """Root directory of the high-density standard-cell library in the PDK (lef, lib, gds, spice, verilog)."""
    return os.path.join(str(pdk), "sky130A", "libs.ref", LIB)


def _run(cmd, cwd, log, timeout=3600):
    """Run one tool in cwd, write its stdout + stderr to the file `log` there, raise with the log's tail on a non-zero
    exit code. Returns (output text, wall-clock seconds)."""
    t0 = time.time()
    p = subprocess.run(cmd, cwd=cwd, capture_output=True, text=True, timeout=timeout)
    open(os.path.join(cwd, log), "w").write(p.stdout + p.stderr)
    if p.returncode != 0:
        raise RuntimeError(f"{os.path.basename(cmd[0])} failed ({p.returncode}); see {os.path.join(cwd, log)}:\n{(p.stdout + p.stderr)[-2500:]}")
    return p.stdout + p.stderr, time.time() - t0


def prep(yosys, netlist, liberty, top, workdir):
    """The synthesised netlist as OpenROAD wants it: constant drivers as tie cells (conb_1) and no assign statements
    (buffers instead). Returns the path of <top>_pnr.v."""
    os.makedirs(workdir, exist_ok=True)
    out = f"{top}_pnr.v"
    # hilomap: every constant 1/0 becomes the HI/LO output of one shared conb_1 tie cell; splitnets: buses split into
    # single-bit nets; insbuf: every assign between two nets becomes a buf_2, so that each net has one cell driver;
    # write_verilog without expressions or numeric literals, as OpenROAD's Verilog reader requires
    R.run_yosys(yosys, "\n".join([
        f"read_liberty -lib {os.path.abspath(liberty)}", f"read_verilog {os.path.abspath(netlist)}", f"hierarchy -top {top}",
        f"hilomap -singleton -hicell {LIB}__conb_1 HI -locell {LIB}__conb_1 LO", "splitnets",
        f"insbuf -buf {LIB}__buf_2 A X", "opt_clean", "check",
        f"write_verilog -noattr -noexpr -nohex -nodec {out}"]) + "\n", workdir, log_name="prep.log")
    return os.path.join(workdir, out)


def flow_tcl(pdk, netlist, top, period_ns, utilization, route, layers="met1-met5", density=0.35, hold_margin=0.0,
             antenna_margin=20, buffer_inputs=False, antenna_iterations=3):
    """The OpenROAD Tcl script of the whole place-and-route flow (arguments as place_and_route()). Returns the script
    text; the flow writes core.{odb,def,v,sdc}, core_pwr.v and, if route, route_drc.rpt into the working directory."""
    hd, L = _hd(pdk), LIB + "__"
    # routing tracks of each metal layer from the PDK's LibreLane tracks file: lines "<layer> <X|Y> <offset> <pitch>"
    # (um), collected as tr[layer][axis] = [offset, pitch]
    tracks = os.path.join(str(pdk), "sky130A", "libs.tech", "librelane", LIB, "tracks.info")
    tr = {}
    for f in (l.split() for l in open(tracks)):
        if len(f) == 4:
            tr.setdefault(f[0], {})[f[1]] = f[2:]
    # design and constraints: typical-corner library, one clock on port clk, input and output delays of 5 % of the period
    # (all inputs except the clock itself); square core at the given utilization (%) with a 6 um core-to-die margin
    T = [f"read_lef {hd}/techlef/{L}nom.tlef", f"read_lef {hd}/lef/{LIB}.lef", f"read_liberty {hd}/lib/{L}{CORNER}.lib",
         f"read_verilog {netlist}", f"link_design {top}",
         f"create_clock -name clk -period {period_ns} [get_ports clk]",
         f"set_input_delay {0.05 * period_ns} -clock clk [delete_from_list [all_inputs] [get_ports clk]]",
         f"set_output_delay {0.05 * period_ns} -clock clk [all_outputs]",
         f"initialize_floorplan -utilization {utilization} -aspect_ratio 1 -core_space 6 -site unithd"]
    T += [f"make_tracks {lay} -x_offset {v['X'][0]} -x_pitch {v['X'][1]} -y_offset {v['Y'][0]} -y_pitch {v['Y'][1]}" for lay, v in tr.items()]
    # pins on met3 (horizontal edges) and met2 (vertical edges); well taps every 13 um with end caps (decap_3) at the row
    # ends; power grid: met1 rails along the cell rows, met4 straps 1.6 um wide every 40 um, connected met1 to met4.
    # Placement-stage wire RC: signals on met2, the clock on met3 (estimate used until extraction).
    T += ["place_pins -hor_layers met3 -ver_layers met2",
          f"tapcell -distance 13 -tapcell_master {L}tapvpwrvgnd_1 -endcap_master {L}decap_3",
          "add_global_connection -net VPWR -pin_pattern {^VPWR$} -power", "add_global_connection -net VPWR -pin_pattern {^VPB$}",
          "add_global_connection -net VGND -pin_pattern {^VGND$} -ground", "add_global_connection -net VGND -pin_pattern {^VNB$}",
          "global_connect", "set_voltage_domain -power VPWR -ground VGND", "define_pdn_grid -name grid",
          "add_pdn_stripe -grid grid -layer met1 -width 0.48 -followpins",
          "add_pdn_stripe -grid grid -layer met4 -width 1.6 -pitch 40 -offset 8",
          "add_pdn_connect -grid grid -layers {met1 met4}", "pdngen",
          "set_wire_rc -signal -layer met2", "set_wire_rc -clock -layer met3",
          f"global_placement -density {density}", "estimate_parasitics -placement"]
    if buffer_inputs:      # a buffer at each input pin drives the long route to the loads (its diffusion protects their gates)
        T += [f"buffer_ports -inputs -buffer_cell {L}buf_2"]
    # repair_design: buffering and sizing for slew, capacitance and fanout limits; legal placement; clock tree with a
    # clkbuf_16 root and clkbuf_4/8 buffers; hold repair on the propagated clock (with hold_margin ns of extra slack);
    # global routing with a guide file for the detailed router; antenna repair with diode_2 cells; the worst setup and
    # hold slack on the global routes (printed as SLACK_MAX / SLACK_MIN for place_and_route() to parse); filler cells
    T += ["repair_design", "detailed_placement",
          f'clock_tree_synthesis -root_buf {L}clkbuf_16 -buf_list "{L}clkbuf_4 {L}clkbuf_8" -sink_clustering_enable',
          "set_propagated_clock [all_clocks]", "estimate_parasitics -placement",
          "repair_timing -hold" + (f" -hold_margin {hold_margin}" if hold_margin else ""), "detailed_placement",
          "check_placement", f"set_routing_layers -signal {layers} -clock {layers}",
          "global_route -guide_file route.guide -congestion_iterations 30", f"repair_antennas {L}diode_2 -iterations {antenna_iterations} -ratio_margin {antenna_margin}",
          "estimate_parasitics -global_routing",
          'puts "SLACK_MAX [sta::worst_slack -max]"', 'puts "SLACK_MIN [sta::worst_slack -min]"',
          f'filler_placement "{L}fill_1 {L}fill_2 {L}fill_4 {L}fill_8"']
    if route:                                              # TritonRoute; its DRC report counts the routing violations
        T += ["detailed_route -output_drc route_drc.rpt -verbose 0", "check_antennas"]
    # global_connect again: the cells inserted after the first one (clock buffers, hold buffers, diodes) get their power
    # pins connected in the database, so that the netlist with power pins (core_pwr.v, for LVS) is complete
    T += ["global_connect", "report_design_area", "write_db core.odb", "write_def core.def", "write_verilog core.v",
          "write_verilog -include_pwr_gnd core_pwr.v", "write_sdc core.sdc"]
    stamped = ["set T0 [clock milliseconds]"]
    for line in T:                                         # time stamps around the expensive steps
        if line.split()[0] in ("global_placement", "clock_tree_synthesis", "global_route", "detailed_route", "report_design_area"):
            stamped.append(f'puts "STAGE {line.split()[0]} [expr {{([clock milliseconds] - $T0) / 1000.0}}]"')
        stamped.append(line)
    return "\n".join(stamped + ["exit"]) + "\n"


def place_and_route(openroad, pdk, netlist, top, workdir, period_ns=100, utilization=25, route=True, threads=None, layers="met1-met5",
                    density=0.35, hold_margin=0.0, antenna_margin=20, buffer_inputs=False, antenna_iterations=3):
    """OpenROAD from netlist to routed layout. Returns a dict of results and the paths of core.{odb,def,v,sdc}.
    layers: the signal and clock routing layers. The detailed router can leave via2/via3 landing pads on met3 below the
    0.24 um2 minimum area (met3.6); to_gds() grows them (fix_min_area), so the default met1-met5 is DRC clean.
    utilization, density: core utilization (%) and global-placement target density. The low defaults (25 %, 0.35) leave the
    detailed router fewer conflicts to repair: on two cores the keystream block routes in 105 s instead of 175 s at 45 % and
    0.6, and the compiled key unit in about half the time.
    hold_margin (ns): hold repair (at tt, with wire RC estimated on the placement) aims at this slack instead of zero, so
    that hold still holds at the fast corner with the extracted wires (checked by sta()).
    antenna_margin (%): the antenna repair on the global routes treats a net as violating above (100 - margin) % of the
    allowed antenna ratio; a larger margin covers nets whose detailed routes end longer than the global-routing estimate."""
    os.makedirs(workdir, exist_ok=True)
    open(os.path.join(workdir, "pnr.tcl"), "w").write(flow_tcl(pdk, os.path.abspath(netlist), top, period_ns, utilization, route, layers, density,
                                                                       hold_margin, antenna_margin, buffer_inputs, antenna_iterations))
    # threads: by default all cores this process may run on (taskset), so a run pinned to two cores uses two threads.
    # OpenROAD's multi-threaded timing analysis can crash (SIGSEGV in its arrival search) on a machine with more cores
    # (seen with four threads, never with Colab's two): the same flow is then run once more, on one thread
    n_thr = threads or len(os.sched_getaffinity(0)) or 1
    try:
        log, dt = _run([openroad, "-no_init", "-threads", str(n_thr), "-exit", "pnr.tcl"], workdir, "pnr.log")
    except RuntimeError as e:
        if n_thr == 1 or "failed (-11)" not in str(e):
            raise
        log, dt = _run([openroad, "-no_init", "-threads", "1", "-exit", "pnr.tcl"], workdir, "pnr.log")
    # results parsed from the OpenROAD log: f() takes the first match of a pattern (None if absent)
    f = lambda pat, cast=float: cast(re.search(pat, log).group(1)) if re.search(pat, log) else None
    last = lambda pat: int(re.findall(pat, log)[-1]) if re.findall(pat, log) else 0     # the final check of a repeated one
    # STAGE time stamps (s since start); a stage's duration is the time to the next stamp, so it includes the cheaper
    # commands that follow it in the script
    stages = [(m.group(1), float(m.group(2))) for m in re.finditer(r"STAGE (\w+) ([\d.]+)", log)]
    drc_rpt = os.path.join(workdir, "route_drc.rpt")
    out = dict(seconds=round(dt, 1), area_um2=f(r"Design area ([\d.]+) u\^2"), utilization=f(r"Design area [\d.]+ u\^2 ([\d.]+)%"),
               slack_setup_ns=f(r"SLACK_MAX (-?[\d.eE+-]+)"), slack_hold_ns=f(r"SLACK_MIN (-?[\d.eE+-]+)"),
               taps=f(r"Inserted (\d+) tapcells", int), endcaps=f(r"Inserted (\d+) endcaps", int),
               diodes=sum(int(x) for x in re.findall(r"Inserted (\d+) diodes", log)), fillers=f(r"Placed (\d+) filler", int),
               clock_buffers=f(r"Inserted (\d+) clock buffers", int), period_ns=period_ns, routed=route,
               antenna_violations=(last(r"Found (\d+) net violations") + last(r"Found (\d+) pin violations")) if route else None,
               route_drc=(sum(1 for l in open(drc_rpt) if l.startswith("violation type")) if os.path.exists(drc_rpt) else None) if route else None,
               stage_seconds={a: round(b - a0, 1) for (a, a0), (_, b) in zip(stages, stages[1:])})
    # instance count and die size from the DEF (DIEAREA in database units of 1 nm -> um)
    comp = re.search(r"COMPONENTS (\d+)", open(os.path.join(workdir, "core.def")).read())
    out["instances"] = int(comp.group(1)) if comp else None
    die = re.search(r"DIEAREA \( (\d+) (\d+) \) \( (\d+) (\d+) \)", open(os.path.join(workdir, "core.def")).read())
    out["die_um"] = ((int(die.group(3)) - int(die.group(1))) / 1000, (int(die.group(4)) - int(die.group(2))) / 1000) if die else None
    out.update({k: os.path.join(workdir, f"core.{k2}") for k, k2 in (("odb", "odb"), ("def", "def"), ("verilog", "v"), ("sdc", "sdc"))})
    return out


def to_gds(pdk, def_path, gds_path):
    """DEF -> GDS with KLayout: the routing from the DEF (through the PDK's LEF/DEF layer map) and every standard cell's
    real layout from the library GDS in place of its LEF abstract."""
    import klayout.db as kdb
    hd = _hd(pdk)
    opt = kdb.LoadLayoutOptions()
    cfg = opt.lefdef_config
    cfg.lef_files = [f"{hd}/techlef/{LIB}__nom.tlef", f"{hd}/lef/{LIB}.lef"]
    cfg.macro_layout_files = [f"{hd}/gds/{LIB}.gds"]
    cfg.map_file = os.path.join(str(pdk), "sky130A", "libs.tech", "klayout", "tech", "sky130A.map")
    ly = kdb.Layout()
    ly.read(def_path, opt)
    # patch the sub-minimum-area via landing pads left by the detailed router (met3.6) before writing
    fix_min_area(ly, cfg.lef_files[0])
    ly.write(gds_path)
    return gds_path


# GDS drawing layers of the metals whose via landing pads can fall below the minimum area, their preferred directions
# and minimum spacings (sky130A.map; SKY130 rules met3.2 and met4.2)
METALS = {"met3": ((70, 20), "h", 0.3), "met4": ((71, 20), "v", 0.3)}

def fix_min_area(ly, tlef):
    """Grow every met3/met4 island smaller than its layer's minimum area (the AREA rule of the tech LEF) until it meets it.
    The detailed router leaves such islands where a via2 and a via3 meet on met3 with no wire between them (landing pads
    of about 0.585 x 0.33 um against a 0.24 um2 minimum: met3.6). An island is extended along the layer's preferred
    direction (both ends, then either end), else across it, by the first extension that keeps the minimum spacing to all
    other shapes of the layer; an island with no such room is left as it is (Magic's DRC then reports it). Only existing
    metal is extended, so no net is shorted or opened. Returns the number of islands patched and left."""
    import klayout.db as kdb
    text = open(tlef).read()
    top, dbu, patched, left = ly.top_cell(), ly.dbu, 0, 0
    for name, ((lay, dt), direction, space) in METALS.items():
        # minimum area (um2) of this layer from the tech LEF
        m = re.search(r"LAYER %s\b.*?\n\s*AREA ([0-9.]+)" % name, text, re.S)
        li = ly.find_layer(lay, dt)
        if m is None or li is None:
            continue
        # target area in dbu^2 with a 2 % margin over the rule (grid rounding); minimum spacing in dbu
        a_min, s_dbu = float(m.group(1)) / dbu ** 2 * 1.02, int(round(space / dbu))
        # all metal of the layer, flattened and merged: each polygon is one connected island
        region = kdb.Region(top.begin_shapes_rec(li)).merged()
        for poly in region.each():
            if poly.area() >= a_min:
                continue
            b, others = poly.bbox(), region - kdb.Region(poly)
            # extension (dbu) that makes the bounding box reach a_min: ceil(a_min / height - width) + 1 along x, and the
            # same along y; the -(-a // 1) form is a ceiling division
            need_h = int(-(-(a_min / max(b.height(), 1) - b.width()) // 1)) + 1          # extra length along x
            need_v = int(-(-(a_min / max(b.width(), 1) - b.height()) // 1)) + 1          # extra length along y
            # candidate patches, in order: grow both ends symmetrically, then only the right/top end, then only the
            # left/bottom end; preferred direction first
            along_x = [kdb.Box(b.left - need_h // 2 - 1, b.bottom, b.right + need_h // 2 + 1, b.top),
                       kdb.Box(b.left, b.bottom, b.right + need_h, b.top), kdb.Box(b.left - need_h, b.bottom, b.right, b.top)]
            along_y = [kdb.Box(b.left, b.bottom - need_v // 2 - 1, b.right, b.top + need_v // 2 + 1),
                       kdb.Box(b.left, b.bottom, b.right, b.top + need_v), kdb.Box(b.left, b.bottom - need_v, b.right, b.top)]
            for box in (along_x + along_y) if direction == "h" else (along_y + along_x):
                # accept the first patch whose spacing halo (minimum spacing less 1 dbu) touches no other island
                if (kdb.Region(box).sized(s_dbu - 1) & others).is_empty():
                    top.shapes(li).insert(box); patched += 1
                    break
            else:
                left += 1
    return patched, left


class _GateSim(R.SimResult):
    """The VCD of the testbench instance top.tb, read as the RTL flow reads Yosys `sim` traces (so its checks apply)."""
    def __init__(self, path, ncycles):
        v, ch = R.parse_vcd(path)
        # keep only the testbench's own signals (top.tb.<name>, one level below), renamed to tb.<name> as in the RTL flow
        self.vars = [(n.split(".", 1)[1], k, w, c) for n, k, w, c in v if n.startswith("top.tb.") and n.count(".") == 2]
        self.changes, self.byname, self.ncycles = ch, {x[0]: x for x in self.vars}, ncycles
        # clock period in VCD time units, from the first two rising edges of tb.clk
        rises = [t for t, val in ch[self.byname["tb.clk"][3]] if val == "1"]
        self.period = rises[1] - rises[0]


def gate_sim(bindir, pdk, netlist, job, top, workdir, period_ns=100):
    """Icarus Verilog on the routed netlist with the PDK's functional cell models, driven by an RTL-flow test job.
    Two runs: the testbench outputs over the whole job (checked with the job's own checks) and, for power, every net of
    the design over the job's steady-state window. The cell models are functional: combinational cells and clock buffers
    switch without delay, flip-flops and latches 1 ns after their clock (UNIT_DELAY), so that every flip-flop samples its
    input before any output changes, also behind the clock buffers and clock-gating cells of a routed clock tree (the real
    delays are checked by sta()). Returns dict(checks, activity_vcd, window, seconds)."""
    os.makedirs(workdir, exist_ok=True)
    models = os.path.join(str(pdk), "sky130A", "libs.ref", LIB, "verilog")
    # the RTL flow's testbench module for this job (constant inputs, input streams, monitored outputs)
    open(os.path.join(workdir, "tb.v"), "w").write("`timescale 1ns/1ps\n" + R.make_tb(job["name"], top, job["ncycles"], job["consts"],
                                                                                         job["streams"], job["outputs"]))
    c0, c1 = job["window"] or (0, job["ncycles"])         # power window in clock cycles (whole job if none)
    t0, out = time.time(), {}
    # "check": the testbench's own signals over the whole job (+1 cycle); "act": every net of the design instance
    # top.tb.u, dumped only from cycle c0 to c1, for read_power_activities. A small wrapper module `top` drives the clock.
    for tag, dump in (("check", f'$dumpfile("check.vcd"); $dumpvars(1, top.tb); #({period_ns} * {job["ncycles"] + 1});'),
                      ("act", f'#({period_ns} * {c0}); $dumpfile("act.vcd"); $dumpvars(0, top.tb.u); #({period_ns} * {c1 - c0});')):
        open(os.path.join(workdir, f"top_{tag}.v"), "w").write(
            f"`timescale 1ns/1ps\nmodule top;\n  reg clk = 0;\n  always #{period_ns / 2:g} clk = ~clk;\n"
            f"  {job['name']} tb (.clk(clk){''.join(f', .o_{p}()' for p in job['outputs'])});\n"
            f"  initial begin {dump} $finish; end\nendmodule\n")
        # PDK cell models: primitives.v (user-defined primitives) and the library's cell models, functional with unit delay
        _run([os.path.join(bindir, "iverilog"), "-g2012", "-DFUNCTIONAL", "-DUNIT_DELAY=#1", "-o", f"{tag}.vvp", "-s", "top",
              f"{models}/primitives.v", f"{models}/{LIB}.v", os.path.abspath(netlist), "tb.v", f"top_{tag}.v"], workdir, f"iverilog_{tag}.log")
        _run([os.path.join(bindir, "vvp"), "-n", f"{tag}.vvp"], workdir, f"vvp_{tag}.log")
    # the job's checks (the same functions the RTL simulation is judged with) applied to the gate-level waveforms
    checks = [dict(check=lab, ok=bool(ok), detail=det) for lab, ok, det in job["check"](_GateSim(os.path.join(workdir, "check.vcd"), job["ncycles"] + 1))]
    return dict(checks=checks, activity_vcd=os.path.join(workdir, "act.vcd"), window=(c0, c1), seconds=round(time.time() - t0, 1))


def extract(openroad, pdk, run, workdir, rc=("min", "nom", "max")):
    """OpenRCX extraction of the routed wires (resistance, ground and coupling capacitance of every net) with the PDK's
    rule files, one SPEF per rule set. Returns dict(rule set -> SPEF path, seconds)."""
    rules = os.path.join(str(pdk), "sky130A", "libs.tech", "librelane", "rules.openrcx.sky130A.%s.calibre")
    out, t0 = {}, time.time()
    for c in rc:                                          # min / nom / max: the RC corners paired with ff / tt / ss
        # -lef_res: wire resistance from the LEF's per-layer resistance values
        open(os.path.join(workdir, f"rcx_{c}.tcl"), "w").write("\n".join([
            f"read_db {run['odb']}", "define_process_corner -ext_model_index 0 X",
            f"extract_parasitics -ext_model_file {rules % c} -lef_res", f"write_spef core_{c}.spef", "exit"]) + "\n")
        _run([openroad, "-no_init", "-exit", f"rcx_{c}.tcl"], workdir, f"rcx_{c}.log")
        out[c] = os.path.join(workdir, f"core_{c}.spef")
    m = re.search(r"Extract (\d+) nets", open(os.path.join(workdir, f"rcx_{rc[0]}.log")).read())
    out["nets"] = int(m.group(1)) if m else None
    out["seconds"] = round(time.time() - t0, 1)
    return out


def sta(openroad, pdk, run, spef, workdir, corners=None):
    """Static timing analysis of the routed design at several corners at once (OpenSTA): each corner's library with the
    SPEF of its RC rule set (CORNERS), propagated clock. Returns per corner the worst setup slack (WNS, ns), the total
    negative setup slack (TNS, ns) and the worst hold slack (ns); sta.log also holds the clock skew."""
    hd, corners = _hd(pdk), corners or CORNERS
    # one OpenSTA session with all corners: each corner's Liberty and SPEF, the SDC written by place-and-route
    T = [f"define_corners {' '.join(corners)}"]
    T += [f"read_liberty -corner {c} {hd}/lib/{LIB}__{lib}.lib" for c, (lib, _) in corners.items()]
    T += [f"read_db {run['odb']}", f"read_sdc {run['sdc']}", "set_propagated_clock [all_clocks]"]
    T += [f"read_spef -corner {c} {spef[rc]}" for c, (_, rc) in corners.items()]
    for c in corners:                                     # max = setup paths, min = hold paths
        T += [f'puts "STA {c} wns [sta::worst_slack_corner [sta::find_corner {c}] max]"',
              f'puts "STA {c} tns [sta::total_negative_slack_corner_cmd [sta::find_corner {c}] max]"',
              f'puts "STA {c} hold [sta::worst_slack_corner [sta::find_corner {c}] min]"']
    T += ["report_parasitic_annotation", "report_clock_skew -setup", "report_clock_skew -hold", "exit"]
    open(os.path.join(workdir, "sta.tcl"), "w").write("\n".join(T) + "\n")
    log, dt = _run([openroad, "-no_init", "-exit", "sta.tcl"], workdir, "sta.log")
    res = {c: {} for c in corners}
    for c, k, v in re.findall(r"^STA (\w+) (\w+) (\S+)", log, re.M):
        res[c][k] = float(v) * 1e9                                            # STA's Tcl API works in seconds
    # drivers without extracted parasitics (from report_parasitic_annotation); 0 means the SPEF covers every net
    m = re.search(r"Found (\d+) unannotated drivers", log)
    res["unannotated_drivers"] = int(m.group(1)) if m else None
    res["seconds"] = round(dt, 1)
    return res


def power(openroad, pdk, run, vcd, workdir, scope="top/tb/u", spef=None, tag="power"):
    """OpenSTA power of the design (tt, 25 °C, 1.8 V) with the simulated switching activity (every net of the window's
    VCD). Wire parasitics: the extracted SPEF of a routed design if `spef` is given, else the per-layer wire RC estimate
    on the placement. Returns µW per group (sequential, combinational, clock) and in total, split into internal,
    switching and leakage."""
    hd = _hd(pdk)
    # wire parasitics: the extracted SPEF if given, else the same per-layer placement estimate as in flow_tcl()
    rc = ([f"read_spef {os.path.abspath(spef)}"] if spef else
          ["set_wire_rc -signal -layer met2", "set_wire_rc -clock -layer met3", "estimate_parasitics -placement"])
    open(os.path.join(workdir, f"{tag}.tcl"), "w").write("\n".join([
        f"read_db {run['odb']}", f"read_liberty {hd}/lib/{LIB}__{CORNER}.lib", f"read_sdc {run['sdc']}",
        "set_propagated_clock [all_clocks]"] + rc +
        [f"read_power_activities -scope {scope} -vcd {os.path.abspath(vcd)}", "report_power", "exit"]) + "\n")
    log, dt = _run([openroad, "-no_init", "-exit", f"{tag}.tcl"], workdir, f"{tag}.log")
    # report_power table rows: group, internal, switching, leakage, total (W); converted to uW
    groups = {}
    for name in ("Sequential", "Combinational", "Clock", "Total"):
        m = re.search(rf"^{name}\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)", log, re.M)
        if m:
            groups[name.lower()] = dict(zip(("internal", "switching", "leakage", "total"), (1e6 * float(x) for x in m.groups())))
    ann = re.search(r"Annotated (\d+) pin activities", log)
    return dict(uw=groups, annotated_pins=int(ann.group(1)) if ann else 0, seconds=round(dt, 1))


def lvs(magic, netgen, pdk, gds, run, top, workdir):
    """Layout versus schematic of the routed block. Magic reads the GDS (standard cells with their real layouts) and
    extracts a hierarchical transistor netlist; netgen compares it with the routed netlist written by OpenROAD with its
    power pins (core_pwr.v), whose cells are expanded into transistors by the library's SPICE netlists, cell by cell and
    at the top level. Fill and tap cells hold no devices that connect to a signal and are ignored, as the PDK's netgen
    setup does. Returns dict(match, devices, nets, report, seconds) for the top cell."""
    tech = os.path.join(str(pdk), "sky130A", "libs.tech")
    wd = os.path.join(workdir, "lvs")
    os.makedirs(wd, exist_ok=True)
    t0 = time.time()
    # Magic: fill and tap cells are flattened into their parents (flatglob) so they do not appear as subcircuits;
    # devices only (no parasitic R or C), hierarchical extraction, SPICE netlist in LVS format.
    # MAGIC_EXT_USE_GDS=1 in the environment: the cells read from the GDS are extracted from that layout
    open(os.path.join(wd, "ext.tcl"), "w").write("\n".join([
        "gds flatglob *__fill_*", "gds flatglob *__tapvpwrvgnd_*",
        f"gds read {os.path.abspath(gds)}", f"load {top}", "select top cell", "extract do local", "extract no capacitance",
        "extract no coupling", "extract no resistance", "extract all", "ext2spice lvs", f"ext2spice -o {top}_layout.spice",
        "quit -noprompt"]) + "\n")
    env = dict(os.environ, PDK_ROOT=os.path.abspath(str(pdk)), MAGIC_EXT_USE_GDS="1")
    p = subprocess.run([str(magic), "-dnull", "-noconsole", "-rcfile", os.path.join(tech, "magic", "sky130A.magicrc"), "ext.tcl"],
                       cwd=wd, capture_output=True, text=True, timeout=3600, env=env)
    open(os.path.join(wd, "ext.log"), "w").write(p.stdout + p.stderr)
    if p.returncode != 0 or not os.path.exists(os.path.join(wd, f"{top}_layout.spice")):
        raise RuntimeError(f"magic extraction failed; see {wd}/ext.log")
    t_ext = time.time() - t0
    lib_spice = os.path.join(str(pdk), "sky130A", "libs.ref", LIB, "spice", f"{LIB}.spice")
    rpt = os.path.join(wd, "lvs.rpt")
    # netgen: the layout netlist against the routed Verilog netlist (with power pins), whose cell instances are resolved
    # by the library's SPICE subcircuits; the PDK's setup file sets device equivalences and the cells to ignore
    open(os.path.join(wd, "lvs.tcl"), "w").write("\n".join([
        f"set layout [readnet spice {top}_layout.spice]", f"set source [readnet spice {lib_spice}]",
        f"readnet verilog {os.path.abspath(run['verilog'].replace('core.v', 'core_pwr.v'))} $source",
        f'lvs "$layout {top}" "$source {top}" {os.path.join(tech, "netgen", "sky130A_setup.tcl")} {rpt}', "quit"]) + "\n")
    p = subprocess.run([str(netgen), "-batch", "source", "lvs.tcl"], cwd=wd, capture_output=True, text=True, timeout=3600, env=env)
    log = p.stdout + p.stderr
    open(os.path.join(wd, "lvs.log"), "w").write(log)
    text = open(rpt).read() if os.path.exists(rpt) else log
    # device and net counts (layout | schematic) from the last subcircuit summary, i.e. the top cell, which netgen
    # compares last; the verdict from its "Final result" line; the cell classes that were matched one by one
    tail = text[text.rfind(f"Subcircuit summary"):] if "Subcircuit summary" in text else text[-4000:]
    final = text[text.rfind("Final result"):] if "Final result" in text else ""
    dev = re.findall(r"Number of devices:\s*(\d+)\s*\|\s*Number of devices:\s*(\d+)", tail)
    nets = re.findall(r"Number of nets:\s*(\d+)\s*\|\s*Number of nets:\s*(\d+)", tail)
    cells = sorted(set(re.findall(r"Device classes (\S+) and \S+ are equivalent", text)) - {top})
    return dict(match="Circuits match uniquely" in final and "do not match" not in text, final=" ".join(final.split())[:400],
                cell_types=len(cells), devices=tuple(map(int, dev[-1])) if dev else None,
                nets=tuple(map(int, nets[-1])) if nets else None, report=rpt,
                seconds_extract=round(t_ext, 1), seconds=round(time.time() - t0, 1))


def signoff(yosys, openroad, bindir, magic, netgen, pdk, netlist, liberty, top, job, workdir, drc=None, threads=None,
            hold_margin=0.15, antenna_margin=60, layers="met1-met5", buffer_inputs=False, antenna_iterations=3):
    """The complete flow of one routed block, as the notebook runs it: netlist preparation, place and route with a hold
    margin of hold_margin ns and an antenna margin of antenna_margin %, GDS, Magic DRC (drc: a callable gds -> dict(count),
    optional), OpenRCX extraction, multi-corner STA, gate-level simulation of the routed netlist on `job`, power from that activity (extracted SPEF at tt, and the
    placement estimate for comparison), and LVS. Returns a dict of results; file paths under the key 'files'."""
    t0 = time.time()
    net = prep(yosys, netlist, liberty, top, workdir)
    run = place_and_route(openroad, pdk, net, top, workdir, route=True, threads=threads, hold_margin=hold_margin,
                          antenna_margin=antenna_margin, layers=layers, buffer_inputs=buffer_inputs,
                          antenna_iterations=antenna_iterations)
    # each step consumes the previous step's files: routed DEF -> GDS -> DRC and LVS; ODB -> SPEF -> STA and power;
    # routed netlist -> gate-level simulation -> VCD activity -> power
    gds = to_gds(pdk, run["def"], os.path.join(workdir, "core.gds"))
    d = drc(gds) if drc else None
    spef = extract(openroad, pdk, run, workdir)
    timing = sta(openroad, pdk, run, spef, workdir)
    sim = gate_sim(bindir, pdk, run["verilog"], job, top, os.path.join(workdir, "gls"))
    pw_est = power(openroad, pdk, run, sim["activity_vcd"], workdir, tag="power_est")
    pw = power(openroad, pdk, run, sim["activity_vcd"], workdir, spef=spef["nom"], tag="power_rcx")
    lv = lvs(magic, netgen, pdk, gds, run, top, workdir)
    files = dict({k: run[k] for k in ("odb", "def", "verilog", "sdc")}, gds=gds, spef=spef["nom"], vcd=sim["activity_vcd"])
    return dict(run={k: v for k, v in run.items() if k not in files}, magic_drc=d["count"] if d else None,
                rcx=dict(seconds=spef["seconds"], nets=spef.get("nets")), sta=timing, checks=sim["checks"],
                gls_seconds=sim["seconds"], power=pw, power_estimated=pw_est,
                lvs={k: v for k, v in lv.items() if k != "report"},
                seconds=round(time.time() - t0, 1), files=files)


def clock_gating_study(yosys, openroad, bindir, magic, netgen, pdk, liberty, work, drc_for=None, log=print):
    """The proposal's block without and with clock gating, from the same RTL, each through signoff(): the gated netlist is
    first verified with Icarus on every test job of the proposal. drc_for(workdir) returns the DRC callable of signoff().
    Returns a dict of both variants' results (no file paths), as recorded in dev/rtl/clock_gating.json."""
    t0 = time.time()
    src = os.path.join(work, "rtl", "src")
    R.write_verilog_files(src)
    top, files, _ = R.DESIGNS["proposal"]
    jobs = R.JOB_BUILDERS["proposal"](random.Random(2027))   # fixed seed: the same test stimuli in every run
    job = [j for j in jobs if j["window"]][-1]             # the last job with a power window drives the power analysis
    out = dict(top=top, period_ns=100, icg_cell=R.CG_CELL.split()[0], icg_min_group=R.CG_MIN, variants={})
    for name, cg in (("ungated", False), ("gated", True)):
        wd = os.path.join(work, "cg", name)
        os.makedirs(wd, exist_ok=True)
        net = os.path.join(wd, f"{top}_netlist.v")
        syn = R.synthesize(yosys, top, [os.path.join(src, f) for f in files], liberty, wd, flatten=True, netlist=net,
                           stem="synth", clock_gating=cg)
        text = open(net).read()
        # count flip-flops (sky130 df* cells) per clock net: those not clocked by the primary clk sit behind an ICG cell
        clocks = {}
        for m in re.finditer(r"sky130_fd_sc_hd__df\w+\s+\S+\s*\(([^;]*)\);", text, re.S):
            c = re.search(r"\.CLK\(([^)]+)\)", m.group(1)).group(1).strip()
            clocks[c] = clocks.get(c, 0) + 1
        gated_ffs = sum(n for c, n in clocks.items() if c != "clk")
        checks = []
        if cg:                                     # the gated netlist on every test job (the ungated one is checked by the RTL flow)
            for j in jobs:
                checks += gate_sim(bindir, pdk, net, j, top, os.path.join(wd, "gls_" + j["name"]))["checks"]
        s = signoff(yosys, openroad, bindir, magic, netgen, pdk, net, liberty, top, job, os.path.join(wd, "pnr"),
                    drc=drc_for(os.path.join(wd, "pnr")) if drc_for else None)
        s.pop("files")
        out["variants"][name] = dict(synth={k: syn[k] for k in ("area_um2", "ge", "ffs", "cells")},
                                     icgs=syn["by_type"].get(R.CG_CELL.split()[0], 0), gated_ffs=gated_ffs,
                                     netlist_checks=checks, **s)
        log(name, json.dumps({k: out["variants"][name][k] for k in ("synth", "icgs", "gated_ffs", "magic_drc", "sta")}),
            out["variants"][name]["power"]["uw"]["total"], out["variants"][name]["lvs"]["match"])
    out["seconds"] = round(time.time() - t0, 1)
    return out


# ------------------------------------------------------------------------------------------------------------------------
# Gate-level switching energy of a placed netlist, many traces at once (the leakage test of Section X-H)
# ------------------------------------------------------------------------------------------------------------------------
def net_caps(openroad, pdk, run, workdir, spef=None, tag="netcap"):
    """Capacitance of every net in F (pins and wire): the wire RC estimated on the placement, as in power(), or the
    extracted SPEF of a routed design."""
    hd = _hd(pdk)
    rc = ([f"read_spef {os.path.abspath(spef)}"] if spef else
          ["set_wire_rc -signal -layer met2", "set_wire_rc -clock -layer met3", "estimate_parasitics -placement"])
    out = os.path.join(os.path.abspath(workdir), f"{tag}.txt")
    open(os.path.join(workdir, f"{tag}.tcl"), "w").write("\n".join([
        f"read_db {run['odb']}", f"read_liberty {hd}/lib/{LIB}__{CORNER}.lib", f"read_sdc {run['sdc']}"] + rc + [
        f'set fh [open "{out}" w]',
        'foreach net [get_nets *] { puts $fh "[get_full_name $net] [$net capacitance [sta::cmd_corner] max]" }',
        "close $fh", "exit"]) + "\n")
    _run([openroad, "-no_init", "-exit", f"{tag}.tcl"], workdir, f"{tag}.log")
    caps = {}
    for line in open(out):
        name, c = line.rsplit(None, 1)
        caps[name.replace("\\", "")] = float(c)
    return caps


def lib_logic(liberty, names):
    """The logic of the standard cells `names` from the liberty text: per cell, its pins (direction, function) and, for a
    flip-flop, the clock pin and the next-state function of its state IQ; integrated clock-gating cells are marked."""
    txt = open(liberty).read()
    heads = [(m.group(1), m.start()) for m in re.finditer(r'\n\s*cell\s*\(\s*"?(\w+)"?\s*\)\s*\{', txt)]
    start = dict(heads)
    end = {n: (heads[i + 1][1] if i + 1 < len(heads) else len(txt)) for i, (n, _) in enumerate(heads)}
    out = {}
    for n in names:
        body = txt[start[n]:end[n]]
        pins, pm = {}, list(re.finditer(r'\n\s*pin\s*\(\s*"?(\w+)"?\s*\)\s*\{', body))
        for i, m in enumerate(pm):
            pb = body[m.end():pm[i + 1].start() if i + 1 < len(pm) else len(body)]
            d = re.search(r'\bdirection\s*:\s*"?(\w+)"?', pb)
            f = re.search(r'(?<![\w])function\s*:\s*"([^"]*)"', pb)       # not power_down_function
            pins[m.group(1)] = dict(dir=d.group(1) if d else None, func=f.group(1) if f else None)
        cell = dict(pins=pins, icg="clock_gating_integrated_cell" in body)
        ff = re.search(r'\bff\s*\(\s*"(\w+)"\s*,\s*"(\w+)"\s*\)\s*\{([^}]*)\}', body)
        if ff:
            cell["ff"] = dict(state=ff.group(1), clock=re.search(r'clocked_on\s*:\s*"([^"]*)"', ff.group(3)).group(1),
                              next=re.search(r'next_state\s*:\s*"([^"]*)"', ff.group(3)).group(1))
        out[n] = cell
    return out


def _bool_fn(expr, args):
    """A liberty function (!, &, *, |, +, ^, parentheses, 0, 1; every library function used is fully parenthesised) as
    a Python function of its pins on uint64 words."""
    py = expr.replace("!", "~").replace("*", "&").replace("+", "|")
    py = re.sub(r"(?<![\w])1(?![\w])", "ONES", re.sub(r"(?<![\w])0(?![\w])", "ZERO", py))
    return eval(f"lambda {', '.join(args)}: ({py})", {"ONES": np.uint64(0xFFFFFFFFFFFFFFFF), "ZERO": np.uint64(0)})


class GateSim:
    """Zero-delay, cycle-based simulation of a placed sky130_fd_sc_hd netlist for many traces at once. Every net is a row
    of uint64 words, bit j of word w its value in trace 64 w + j (bit-parallel, as rtl_flow.AigSim). The combinational
    cells are evaluated level by level, one vector operation per cell type and level; the flip-flops load at the rising
    edge of their clock: the root clock, or the gated clock of an integrated clock-gating cell, which passes the edge when
    its enable is 1 before the edge (its latch holds the enable while the clock is high). Each clock, every data net that
    changes dissipates 1/2 C V^2 with its capacitance C (pins and wire); the clock tree, the same in every trace, is left
    out. Physical-only cells (fill, tap, decap, diode) are skipped."""
    def __init__(self, netlist, liberty, caps, vdd=1.8, clock="clk"):
        text = open(netlist).read()
        self.vdd = vdd
        self.ports = {}                                  # port -> (direction, its bit nets, most significant first)
        for kind, rng_, names in re.findall(r"^\s*(input|output)\s*(\[\s*\d+\s*:\s*\d+\s*\])?\s*([^;]+);", text, re.M):
            for nm in [x.strip() for x in names.split(",") if x.strip()]:
                if rng_:                                 # [msb:lsb], in either direction ([0:31]: bit 0 is the MSB)
                    a, b = map(int, re.findall(r"\d+", rng_))
                    self.ports[nm] = (kind, [f"{nm}[{i}]" for i in range(a, b + (1 if b >= a else -1), 1 if b >= a else -1)])
                else:
                    self.ports[nm] = (kind, [nm])
        inst = []
        for typ, name, body in re.findall(r"(sky130_fd_sc_hd__\w+)\s+(\S+)\s*\((.*?)\);", text, re.S):
            if any(s in typ for s in ("fill", "tap", "decap", "diode")):
                continue
            pins = {p: n.strip().replace("\\", "").strip() for p, n in re.findall(r"\.(\w+)\(([^)]*)\)", body)}
            inst.append((typ, name, {p: n for p, n in pins.items() if n}))
        lib = lib_logic(liberty, sorted({t for t, _, _ in inst}))
        nets = {}
        idx = lambda n: nets.setdefault(n, len(nets))
        idx("1'b0"), idx("1'b1"), idx("__nc__")       # constants, and the sink of unconnected outputs
        # port bits
        self.inputs, self.outputs = {}, {}
        for nm, (kind, bits) in self.ports.items():
            for bn in bits:
                (self.inputs if kind == "input" else self.outputs)[bn] = idx(bn)
        # classify the cells: clock-gating cells, flip-flops, combinational
        drv, icg, ffs, comb = {}, [], [], []
        for typ, name, pins in inst:
            cell = lib[typ]
            outs = [p for p in pins if cell["pins"].get(p, {}).get("dir") == "output"]
            for p in outs:
                drv[pins[p]] = (typ, name, pins)
            if cell["icg"]:
                icg.append((name, pins))
            elif "ff" in cell:
                ffs.append((typ, name, pins))
            else:
                comb.append((typ, name, pins))
        # the clock tree: nets reached from the root clock through buffers, and from each gating cell's GCLK
        dom = {clock: -1}
        for k, (name, pins) in enumerate(icg):
            dom[pins["GCLK"]] = k
        fan = {}
        for typ, name, pins in comb:
            for p, n in pins.items():
                if lib[typ]["pins"].get(p, {}).get("dir") == "input":
                    fan.setdefault(n, []).append((typ, name, pins))
        stack, clk_cells = list(dom), set()
        while stack:
            n = stack.pop()
            for typ, name, pins in fan.get(n, []):
                outs = [p for p in pins if lib[typ]["pins"][p]["dir"] == "output"]
                assert len(outs) == 1 and lib[typ]["pins"][outs[0]]["func"].strip("() ") in pins and \
                    len([p for p in pins if lib[typ]["pins"][p]["dir"] == "input"]) == 1, \
                    f"{name} ({typ}): only non-inverting buffers may carry a clock"
                clk_cells.add(name)
                o = pins[outs[0]]
                if o not in dom:
                    dom[o] = dom[n]
                    stack.append(o)
        for _, pins in icg:
            assert pins["CLK"] in dom and dom[pins["CLK"]] == -1, "a clock-gating cell must be fed by the root clock"
        self.clock_nets = set(dom)
        comb = [c for c in comb if c[1] not in clk_cells]
        # levelise the combinational cells (sources: inputs, flip-flop outputs, constants)
        lvl = {n: 0 for n in list(self.inputs) + ["1'b0", "1'b1"]}
        for typ, name, pins in ffs:
            for p in pins:
                if lib[typ]["pins"].get(p, {}).get("dir") == "output":
                    lvl[pins[p]] = 0
        cdrv = {}
        for c in comb:
            for p, n in c[2].items():
                if lib[c[0]]["pins"][p]["dir"] == "output":
                    cdrv[n] = c
        clvl = {}
        def level(n):
            if n in lvl:
                return lvl[n]
            todo = [n]
            while todo:
                m = todo[-1]
                if m in lvl:
                    todo.pop(); continue
                assert m in cdrv, f"net {m} has no driver"
                assert m not in self.clock_nets, f"net {m} is a clock but feeds data logic"
                typ, name, pins = cdrv[m]
                ins = [pins[p] for p in pins if lib[typ]["pins"][p]["dir"] == "input"]
                pend = [x for x in ins if x not in lvl]
                if pend:
                    todo.extend(pend); continue
                L = 1 + max([lvl[x] for x in ins] + [0])
                for p in pins:
                    if lib[typ]["pins"][p]["dir"] == "output":
                        lvl[pins[p]] = L
                clvl[name] = L
                todo.pop()
            return lvl[n]
        for c in comb:
            for p, n in c[2].items():
                if lib[c[0]]["pins"][p]["dir"] == "output":
                    level(n)
        # groups of one cell type at one level: index arrays per pin, one function per output pin
        groups = {}
        for typ, name, pins in comb:
            groups.setdefault((clvl[name], typ), []).append(pins)
        self.groups = []
        for (L, typ), members in sorted(groups.items()):
            cp = lib[typ]["pins"]
            ins = sorted(p for p in cp if cp[p]["dir"] == "input")
            outs = sorted(p for p in cp if cp[p]["dir"] == "output")
            fins = [np.array([idx(m.get(p, "1'b0")) for m in members]) for p in ins]
            fouts = [(np.array([idx(m.get(o, "__nc__")) for m in members]), _bool_fn(cp[o]["func"], ins)) for o in outs]
            self.groups.append((fins, fouts))
        # flip-flops, grouped by type and clock domain: next-state function of their input pins and the state IQ
        fgroups = {}
        for typ, name, pins in ffs:
            ff = lib[typ]["ff"]
            fgroups.setdefault((typ, dom[pins[ff["clock"]]]), []).append(pins)
        self.ffs = []
        for (typ, d), members in sorted(fgroups.items()):
            cp, ff = lib[typ]["pins"], lib[typ]["ff"]
            ins = sorted(p for p in cp if cp[p]["dir"] == "input" and p != ff["clock"])
            nxt = _bool_fn(ff["next"], ins + [ff["state"]])
            qs = [(np.array([idx(m.get(o, "__nc__")) for m in members]), cp[o]["func"].strip()) for o in cp
                  if cp[o]["dir"] == "output"]
            self.ffs.append((d, [np.array([idx(m.get(p, "1'b0")) for m in members]) for p in ins], nxt, qs, len(members)))
        self.icg_gate = np.array([idx(pins["GATE"]) for _, pins in icg], dtype=np.int64)
        self.n = len(nets)
        self.names = {v: k for k, v in nets.items()}
        # capacitance of every data net (OpenSTA's names; a net it does not list, an unconnected output, counts 0)
        data = [i for i in range(3, self.n) if self.names[i] not in self.clock_nets]
        self.data = np.array(data)
        self.cap = np.array([caps.get(self.names[i], 0.0) for i in data])
        self.missing = sum(self.names[i] not in caps for i in data)
        self.n_cells, self.n_ff, self.n_icg = len(comb), sum(g[4] for g in self.ffs), len(icg)
        qn = {i for g in self.ffs for qi, _ in g[3] for i in qi}
        self.q_cap = float(np.mean([caps.get(self.names[i], 0.0) for i in qn if self.names[i] != "__nc__"]))   # mean flip-flop output

    def bits(self, port, value):
        """{bit net: 0/1} of an integer value on a port."""
        b = self.ports[port][1]
        return {n: (value >> (len(b) - 1 - k)) & 1 for k, n in enumerate(b)}

    def value(self, port, rec, trace):
        """The integer on a recorded port in every cycle, in one trace."""
        b, w, s = self.ports[port][1], trace // 64, np.uint64(trace % 64)
        return sum(((rec[n][:, w] >> s) & np.uint64(1)).astype(object) << (len(b) - 1 - k) for k, n in enumerate(b))

    def run(self, inputs, n_cycles, n_traces, seed=0, record=()):
        """inputs: {port bit: (n_cycles, W) uint64 words, or 0/1 for every trace and cycle}; the flip-flops start from one
        power-up state, the same in every trace. Returns (switching energy (n_cycles, n_traces) in J, {net: (n_cycles, W)}
        of the bits of the ports in `record`)."""
        W = (n_traces + 63) // 64
        ones = np.uint64(0xFFFFFFFFFFFFFFFF)
        V = np.zeros((self.n, W), dtype=np.uint64)
        V[1] = ones
        rng = np.random.default_rng(seed)
        state = [rng.integers(0, 2, size=(g[4], 1)).astype(np.uint64) * ones * np.ones((1, W), np.uint64) for g in self.ffs]
        # energy: nets binned by capacitance (5 % bins, each at its mean), each bin's toggles counted per trace
        order = np.argsort(self.cap)
        lc = np.log(np.maximum(self.cap[order], 1e-18))
        bins = np.floor((lc - lc[0]) / np.log(1.05)).astype(int)
        edges = np.flatnonzero(np.diff(np.concatenate([[-1], bins, [bins[-1] + 1]])))
        rows = [order[a:b] for a, b in zip(edges[:-1], edges[1:])]
        cbin = np.array([self.cap[r].mean() for r in rows])
        E = np.zeros((n_cycles, W * 64))
        rid = {n: self.inputs.get(n, self.outputs.get(n)) for p in record for n in self.ports[p][1]}
        rec = {n: np.zeros((n_cycles, W), np.uint64) for n in rid}
        prev = None
        for c in range(n_cycles):
            for nm, i in self.inputs.items():
                v = inputs.get(nm, 0)
                V[i] = v[c] if isinstance(v, np.ndarray) else (ones if v else np.uint64(0))
            for g, (d, fin, nxt, qs, _) in zip(state, self.ffs):
                for qi, fn in qs:
                    V[qi] = g if fn == "IQ" else ~g
            for fins, fouts in self.groups:
                a = [V[i] for i in fins]
                for oi, fn in fouts:
                    V[oi] = fn(*a)
            cur = V[self.data]
            if prev is not None:
                t = cur ^ prev
                for r, cb in zip(rows, cbin):
                    bits = np.unpackbits(t[r].view(np.uint8), axis=1, bitorder="little")
                    E[c] += cb * bits.sum(axis=0, dtype=np.int64)
            prev = cur
            for r, i in rid.items():
                rec[r][c] = V[i]
            # the clock edge: every flip-flop whose clock passes the edge loads its next state
            en = V[self.icg_gate] if len(self.icg_gate) else None
            for k, (d, fin, nxt, qs, _) in enumerate(self.ffs):
                nx = nxt(*[V[i] for i in fin], state[k])
                if d < 0:
                    state[k] = nx
                else:
                    e = en[d][None, :]
                    state[k] = (nx & e) | (state[k] & ~e)
        return 0.5 * self.vdd ** 2 * E[:, :n_traces], rec
````

### D. Gate-level power of every design

The power of a CMOS block at clock frequency $f$ and supply $V_{DD}$ has three parts,

$$P = P_{\text{int}} + P_{\text{sw}} + P_{\text{leak}},\qquad P_{\text{sw}} = \tfrac{1}{2}\sum_i \alpha_i\, C_i\, V_{DD}^2\, f ,$$

where $P_{\text{sw}}$ is the switching power of the nets, $\alpha_i$ is the number of transitions of net $i$ per clock and $C_i$ its load capacitance (pins and wire); $P_{\text{int}}$ is the power inside the cells at their transitions (internal nodes and short-circuit current), from the library's power tables; and $P_{\text{leak}}$ the static leakage. A flip-flop costs power even when its output does not change: its clock pin switches twice every cycle, and so does the clock tree that drives it; in a stream cipher every state bit also shifts every cycle. **Power is therefore set by the number of flip-flops times the rate at which they are clocked**, $P \approx N_{\text{FF}}\,E_{\text{FF}}\,f_{\text{clk,FF}} + P_{\text{tree}} + P_{\text{logic}}$, with $N_{\text{FF}}$ the number of flip-flops, $E_{\text{FF}}$ the energy per clocked flip-flop and cycle, $f_{\text{clk,FF}}$ the rate at which they are clocked, and $P_{\text{tree}}$ and $P_{\text{logic}}$ the power of the clock tree and of the combinational logic.

The low-power form of Subsection A lowers $f_{\text{clk,FF}}$ **without changing the function**. Its parallel update of K = 16 bits (Trivium allows up to 64 [5], Grain-128a up to 32 pre-output bits [6]) clocks the state at 10 MHz / 16 = 625 kHz, with K-fold logic that switches K times less often; clock gating removes the clock from the key, the two nonces and the counter, which change once per power-up, session or frame.

The prediction is that the state power falls by about K = 16, but what must still run at 10 MHz remains: the clock trunk up to the gating cells, the output-bit counter and, for the proposed design, the capture of one message bit per clock for the MAC. **The remaining ratio between two designs is then set by their state and this interface**: 16 flip-flops of LFSR state against 256 of Grain-128a state and 64 of MAC state.

The power of every design is computed under identical conditions:

* *Level:* gate level, with the PDK's cell models and power tables, at tt, 25 °C, 1.8 V and 10 MHz.
* *Designs:* the S-DSM, Trivium and the proposed design's block in both forms (`sdsm_lp`, `trivium_lp`, `secure_ks_lp`, verified in Subsection B); Grain-128a + MAC as written, for the information-equivalent baseline of Subsection E.
* *Layout and wires, the same for every entry:* steps 1–4 of the flow of Subsection C (placement and clock tree) without detailed routing, wire capacitance estimated on the placement; for the low-power proposal, the placement of its routed run, with its extracted-wire power beside it.
* *Activity:* each placed netlist is simulated with Icarus Verilog (the functional cell models of Subsection C) on the test job of Subsection B that contains its toggle window, whose checks must pass; OpenSTA reads the activity of every net over that window.
* *Metrics:* power and clock-tree share in both forms, reductions, ratios, and the values measured in [1].

> **Next cell.** 1. Places every design described above that is not yet placed, with its clock tree. 2. Simulates each placed netlist on its test job and checks it. 3. Computes its power with OpenSTA. 4. Tabulates both forms. 5. Assigns the power of every cell of the low-power proposal to its part of the block. Run time: about 1 min on two cores.

````python
#@title Gate-level power of every design, in both forms
# The same gate-level power for every design: placed with its clock tree, wire RC estimated on the placement, simulated with
# its own steady-state test job, power from that activity (tt, 25 °C, 1.8 V, 10 MHz). Each design is measured in two forms:
# as written for one keystream bit per clock (Subsection B) and in its low-power form (radix 16, clock-gated, `lowpower.v`).
# The low-power proposal is the routed block above (its placement estimate here, so that every entry has the same wires).
t0 = time.time()
P130, PW130, RUN130 = {"proposal_lp": P_CORE_EST}, {"proposal_lp": PW_EST_C}, {"proposal_lp": run_c}   # RUN130: the placed runs (Subsection H)
# place each design (no routing), simulate its gate-level netlist with its own test job (the assert stops the run if the
# netlist does not reproduce the reference), and keep the total power and its breakdown
for key in ("sdsm", "trivium", "grain_auth2x", "proposal", "sdsm_lp", "trivium_lp"):
    run_k, sim_k, pw_k = gate_level(key, route=False)
    assert all(c["ok"] for c in sim_k["checks"]), f"{key}: gate-level simulation failed"
    P130[key], PW130[key], RUN130[key] = pw_k["uw"]["total"]["total"], pw_k, run_k
D = RTL["designs"]
# share of a design's power spent in its clock network (clock tree and the clock pins of the flip-flops)
clk_pct = lambda k: f"{100 * PW130[k]['uw']['clock']['total'] / P130[k]:.0f} %"
names = {"sdsm": "S-DSM [1]", "trivium": "Trivium (conventional, [1])", "proposal": "Proposal, full block"}
table(["Design", "Flip-flops", "1 bit/clock: power (µW)", "of which clock", "Low power: power (µW)", "of which clock",
       "Reduction", "Measured in [1] (65 nm)"],
      [(names[k], SYN[k]["ffs"], f"{P130[k]:.0f}", clk_pct(k), f"<b>{P130[k + '_lp']:.1f}</b>", clk_pct(k + "_lp"),
        f"{P130[k] / P130[k + '_lp']:.1f}×", {"sdsm": "24 µW", "trivium": "220 µW"}.get(k, "–")) for k in ("sdsm", "trivium", "proposal")]
      + [("Proposal ÷ S-DSM", "", f"{P130['proposal'] / P130['sdsm']:.1f}×", "", f"<b>{P130['proposal_lp'] / P130['sdsm_lp']:.1f}×</b>", "", "", ""),
         ("Proposal ÷ Trivium", "", f"{P130['proposal'] / P130['trivium']:.1f}×", "", f"{P130['proposal_lp'] / P130['trivium_lp']:.1f}×", "", "", "")],
      caption=f"Table 23: gate-level power of each security block in SKY130 under the same conditions (sky130_fd_sc_hd, tt, "
              f"25 °C, 1.8 V, 10 MHz, one keystream bit per clock): each placed with its clock tree, wire RC estimated on the "
              f"placement, activity from an Icarus simulation of its own test job; as written for one bit per clock and in its "
              f"low-power form (radix 16, clock-gated). The routed low-power proposal with OpenRCX-extracted wires: {P_CORE:.1f} µW "
              f"({time.time() - t0:.0f} s)",
      align=["left"] + ["right"] * 7)

# Where the low-power proposal's power goes: OpenSTA's power of every cell instance of its placed netlist (the activity
# above), each flip-flop assigned to the register it implements (its output net keeps the RTL name), each logic cell to
# the registers it feeds, and the clock buffers and gating cells to the clock tree.
import re, pathlib
def power_by_block(vlog, wd, group, tag):
    Path = pathlib.Path
    # rerun OpenSTA's power analysis of the placed block with a per-instance report added to its script
    tcl = Path(wd) / "power_inst.tcl"
    tcl.write_text((Path(wd) / f"{tag}.tcl").read_text().replace("report_power\n", "report_power\nreport_power -instances [get_cells *]\n"))
    log = subprocess.run([str(OPENROAD), "-no_init", "-exit", str(tcl)], cwd=wd, capture_output=True, text=True).stdout
    # instance name -> total power in µW, from the report's columns (internal, switching, leakage, total, instance)
    pw = {m[4].replace("\\", ""): 1e6 * float(m[3]) for m in re.findall(r"^\s*([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+(\S+)\s*$", log, re.M)}
    # the netlist's cell instances: type, name and pin connections; physical-only cells (fill, tap, decap, diode) skipped;
    # loads[net] lists the instances whose inputs the net drives
    cells = re.findall(r"(sky130_fd_sc_hd__\w+)\s+(\S+)\s*\((.*?)\);", Path(vlog).read_text(), re.S)
    OUT = {"X", "Y", "Q", "Q_N", "GCLK"}
    info, loads = {}, {}
    for typ, name, body in cells:
        if any(s in typ for s in ("fill", "tap", "decap", "diode")):
            continue
        pins = {p: n.strip().replace("\\", "").strip() for p, n in re.findall(r"\.(\w+)\(([^)]*)\)", body)}
        info[name.replace("\\", "")] = (typ, pins)
        for p, n in pins.items():
            if p not in OUT:
                loads.setdefault(n, []).append(name.replace("\\", ""))
    # each flip-flop belongs to the register named by its output net (the RTL name survives synthesis)
    ff = {n: group(pins.get("Q", "")) for n, (typ, pins) in info.items() if "df" in typ}
    memo = {}
    def feeds(n, depth=0):                       # registers reached through combinational logic
        if n not in memo:
            memo[n], out = {}, {}
            for p, net in info[n][1].items():
                if p in OUT:
                    for ld in loads.get(net, []):
                        sub = {ff[ld]: 1} if ld in ff else (feeds(ld, depth + 1) if depth < 40 else {})
                        for g, c in sub.items():
                            out[g] = out.get(g, 0) + c
            memo[n] = out
        return memo[n]
    # assign every instance's power: flip-flops to their register, clock buffers and gating cells to the clock tree, and
    # logic cells to the register they feed most (traced forward through the combinational logic)
    res = {}
    for n, (typ, pins) in info.items():
        if "df" in typ:
            g = ff[n]
        elif "dlclkp" in typ or "clkbuf" in typ or "clkinv" in typ:
            g = "clock tree (buffers and gating cells)"
        else:
            f = feeds(n); g = max(f, key=f.get) if f else "output multiplexers"
        res[g] = res.get(g, 0.0) + pw.get(n, 0.0)
    return res, {g: sum(1 for v in ff.values() if v == g) for g in set(ff.values())}

# map an RTL register name of the proposal (hierarchical, possibly indexed) to the part of the block it belongs to
def group_proposal(q):
    s = q.split("[")[0].strip().split(".")
    if s[0] == "u_grain":
        return ("Grain-128a state (NFSR + LFSR)" if len(s) > 2 else
                {"A": "MAC (accumulator, shift register)", "R": "MAC (accumulator, shift register)",
                 "mb": "MAC message capture (10 MHz)"}.get(s[1], "cipher control and counters"))
    return {"key_r": "key, nonce and counter registers (gated)", "nonce_r": "key, nonce and counter registers (gated)",
            "cn_r": "key, nonce and counter registers (gated)", "ctr_r": "key, nonce and counter registers (gated)",
            "u_d1": "dither LFSRs", "u_d2": "dither LFSRs", "dc": "dither bit counter"}.get(s[0], "other")
BLK, BLK_FF = power_by_block(VLOG_C, str(WD_C), group_proposal, PW_TAG_C)
tot_blk = sum(BLK.values())
table(["Part of the low-power proposal", "Flip-flops", "Power (µW)", "Share"],
      [(g, BLK_FF.get(g, "–"), f"{w:.1f}", f"{100 * w / tot_blk:.0f} %") for g, w in sorted(BLK.items(), key=lambda x: -x[1])]
      + [("<b>total</b>", sum(BLK_FF.values()), f"<b>{tot_blk:.1f}</b>", "100 %")],
      caption="Table 24: power of the low-power proposal by part (the block of Subsection C, " + ("routed, extracted wires" if ROUTE_BLOCK else "placed, wire RC estimated on the placement") + ", the activity of its "
              "gate-level simulation; every cell instance assigned to the register it implements or feeds)", align=["left", "right", "right", "right"])
````

As written for one bit per clock, the S-DSM draws 43 µW, Trivium 223 µW and the proposed design's full block 415 µW, the clock tree taking 30–32 % of each. In their low-power forms, the S-DSM draws 18.1 µW (2.4× less), Trivium 32.3 µW (6.9× less) and the proposed design **64.3 µW (6.5× less)**. Trivium's 223 µW is close to the 220 µW measured in 65 nm in [1], whereas the S-DSM's 43 µW is about 1.8× the 24 µW measured.
* **The parallel update and clock gating remove most of the proposed design's power, but not its distance from the S-DSM**: the ratio of the proposed design to the S-DSM falls from 9.7× to 3.6×; to Trivium it stays at 1.9–2.0×. The S-DSM gains least because its state is small: the output counter, multiplexers and clock trunk kept at 10 MHz are a large part of what remains, and its clock-tree share rises from 32 to 48 %.
* By part (Table 24), the clock tree draws 20.5 µW (32 %) of the low-power proposal, the Grain-128a state 14.2 µW (22 %), the MAC 11.6 µW (18 %) and the message capture for the MAC, still at 10 MHz, 7.4 µW (12 %); the cipher control and counters, the dither bit counter, the dither LFSRs and the output multiplexers draw 9.7 µW together, and the gated key, nonce and counter registers less than 0.05 µW. As predicted, what remains of the proposed design is its 10 MHz part (clock tree, message capture, counters: about 56 % of the block) and its 320 bits of cipher and MAC state; the factor of 3.6 is the cost of a 128-bit authenticated cipher over a 16-bit LFSR.
* The zero-delay simulation counts no glitches, so these values are lower bounds on the dynamic power, looser for the deeper logic of the low-power forms.

#### Clock gating alone, on the block written for one bit per clock

The block written for one bit per clock clocks its key, nonce and counter registers on every cycle, although they hold their value for a whole frame. An enabled flip-flop has an input multiplexer that selects the new value or its own output; its clock pin and clock tree switch on every cycle. An integrated clock-gating (ICG) cell instead passes the clock to a group of flip-flops only when their enable is set; its latch keeps the gated clock free of glitches. With the enable off, the gated flip-flops and the clock branch behind the ICG cell draw no dynamic power, and the multiplexers are no longer needed. If $N_g$ of the $N$ flip-flops are gated and their enable is almost never set during a frame, **the prediction is that the flip-flop and clock-tree power fall roughly in the ratio** $N_g/N$, less the power of the ICG cells, which see every clock edge.

The gating is inserted automatically by Yosys, with these settings:

* *Design:* the RTL of Subsection B, unchanged; **Yosys inserts the gating during synthesis** (`clockgate`): every group of at least four flip-flops sharing a clock enable gets one `sky130_fd_sc_hd__dlclkp_1` cell, and its flip-flops lose their enable. The rest of the script is that of Subsection B.
* *Verification:* the gated netlist is simulated with Icarus Verilog on both test jobs of the proposed design (seed 2027), with their own checks, because the AIGER simulator handles a single clock only.
* *Implementation and power:* the gated and, as reference, the ungated netlist go through the complete flow of Subsection C, the clock tree built through the ICG cells, with power from the routed netlist's simulation on the 400-bit frame and the extracted wires.
* *Recorded run:* the two routings take about 6–9 minutes on two cores, so the study is shown from its recorded run (`dev/rtl/clock_gating.json`, written by `dev/rtl/record_signoff.py` with the same code); `CLOCK_GATING_LIVE = True` runs it live.
* *Metrics:* the rows of Table 25.

### E. A figure of merit, transmitted bits and average power

The powers of Subsections C and D are SKY130 numbers, those of [1] 65 nm numbers; the process acts on digital blocks through $P = \alpha\,C\,V_{DD}^2\,f$, with $\alpha$ the switching activity and $C$ the switched capacitance. Between two designs in the same flow, $C$, $V_{DD}$ and $f$ cancel: **the ratios of Table 23 are independent of the process to first order.** Across processes, **no number is scaled in this section**: a law such as $L_{\min} V_{\text{nom}}^2$ (minimum channel length $L_{\min}$, nominal supply $V_{\text{nom}}$) assumes ideal scaling, which favours old, high-voltage nodes such as SKY130 (150 nm, 1.8 V devices), while no normalisation favours new nodes.

**Energy per bit alone favours weak security**. The proposed figure of merit divides it by the energy of one switching of a reference gate of the same library and by the security delivered,

$$\text{FOM} = \frac{E_{\text{bit}}}{E_{\text{NAND2}}\, S_\Sigma}, \qquad E_{\text{bit}} = \frac{P_{\text{sec}}}{f_{\text{bit}}},$$

**dimensionless, lower is better**, where $E_{\text{bit}}$ is the energy per keystream bit, $f_{\text{bit}}$ the keystream bit rate and $P_{\text{sec}}$ is the power of the security logic alone: the two dither LFSRs of [1]'s modulator, with their counter, output selection and share of the clock tree, are taken out of the S-DSM's and the proposed design's blocks; Trivium's block is the cipher alone. In the denominator:

* $E_{\text{NAND2}}$, the energy of one output transition of the library's minimum NAND2 (`sky130_fd_sc_hd__nand2_1`) at fan-out 1, at tt, 25 °C, 1.8 V: the driving gate's internal energy, the supply energy $\tfrac{1}{2} C_{\text{in}} V^2$ of charging the driven input (input capacitance $C_{\text{in}}$, supply $V$), and the driven input's internal energy. Both $E_{\text{bit}}$ and $E_{\text{NAND2}}$ are proportional to $C V^2$, so $E_{\text{bit}}/E_{\text{NAND2}}$ counts NAND2-equivalent switchings per bit, the energy analogue of gate equivalents, and the process enters through measured library data rather than an assumed scaling law.
* $S_\Sigma$, the security delivered, in bits, summed over the goals a design meets. **Cipher strength,** by one rule for every design: $\log_2$ of the cost of the best attack applicable as the cipher is used, at most the key length. The S-DSM's 16-bit LFSR falls to Berlekamp-Massey from 32 keystream bits in about $2^{10}$ steps (attack A3), so 10; Trivium has no attack below brute force, 80; Grain-128a 128, since the fast correlation attacks of [18] and [19] need $2^{113.8}$ and $2^{106.3}$ keystream bits from one initialisation, while every frame of 107,400 bits re-initialises the cipher (and [18] excludes the authenticated mode). **Message authentication:** the tag length, 32 bits for Grain-128a's MAC, 0 without a tag. **Freshness:** the chip-nonce bits, 31, 0 without a nonce. **Key protection on a captured chip:** the cipher strength for a key that exists only while the chip is powered, 0 for a key stored on the chip (attack A7). The S-DSM has $S_\Sigma$ = 10 and Trivium 80; the proposed design 128 + 32 + 31 + 128 = 319.

The FOM is computed only for blocks of this flow with simulated power: the published blocks' libraries are not public ([1]'s blocks and the DTLS engine of [9] keep their reported energy per bit), and the information-equivalent baseline's power is estimated. The weight of the tag, nonce and key-protection bits is a choice, **so the result panel gives the weight above which the proposed design leads**. The FOM credits strength and goals met, not the attacks of Section IX withstood, linearly in $S_\Sigma$ (conservative, since the attacker's work grows exponentially with each goal's bits); wires and leakage do not scale exactly with gate energy, so the process cancels to first order only.

$E_{\text{bit}}$ is the power over the keystream rate, one bit per 10 MHz clock: the powers of Table 23; for reference, [1]'s measured 24 µW and 220 µW at 1.2 V and the DTLS engine's 0.89 nJ per byte [9]. $E_{\text{NAND2}}$ is computed from the Liberty file of `sky130_fd_sc_hd`, with the input slew set to the gate's own output slew at fan-out 1. The share of the chip's power is $P_{\text{sec}}/(P_{\text{off}} + P_{\text{sec}})$, each block at its own power, with $P_{\text{off}}$, the chip's power with security off, = 321 + 271 + 250 + 510 = 1,352 µW (front end, TIA, S-to-D buffer and ΔΣ modulator, Fig. 4(a) of [1]); Subsection F gives the overhead $P_{\text{sec}}/P_{\text{off}}$, the measure of [1].

The bits sent for the same information are counted per reading (Nyquist sample); **they are set by the architecture independently of any radio**, and with equal energy per bit, transmission energies are in their ratio:

* *Modulator-level* (the S-DSM and the proposed design): every modulator bit is sent, OSR bits per reading, at OSR = 50, the ratio of every accuracy result of this notebook (Section VIII). The proposed design adds the 32-bit counter, the 32-bit chip nonce and the 32-bit MAC tag sent in clear with every frame, $96\cdot\text{OSR}/107{,}400 \approx 0.045$ bits per reading.
* *The conventional design of [1]*: a decimating ADC with Trivium sends 16-bit words at SR/NR = 4.5 to 5 times the Nyquist rate (sample rate SR over Nyquist rate NR) (the two OSR-50 rows of Fig. 4(b) of [1]), **72 to 80 bits per reading**.
* *An information-equivalent conventional design*: the receiver's third-order CIC decimator (Section V) moves onto the sensor, and Grain-128a with its MAC encrypts and authenticates 16-bit words, the word length of [1], at the Nyquist rate, with the same frame fields. A shorter word matched to the sensor's resolution would rest on the behavioural model's SNDR, which no comparison uses.

**As no radio is designed in this work**, the average power is an example with the radio of [1]: the radio energy of [1]'s energy table (3.1 mW at 200 kb/s, 15.5 nJ per bit) is used, one frame (10.74 ms, 2,148 readings at OSR 50) is sent once per minute, and the average power is the radio energy of the frame's bits plus the energy of [1]'s front end and the security block during the frame, over the minute. Key generation at power-up and the sleep power, neither designed in this work, are excluded; the decimator of the information-equivalent design enters with its estimated power.

The prediction is that the proposed design sends about 50 bits per reading, like the S-DSM, 1.4–1.6× fewer than [1]'s conventional design, not the 2.7× of [1]'s OSR-30 comparison, and that the information-equivalent design sends about 16, about 3× **fewer** than any modulator-level scheme. The radio's energy per bit, more than 300× the keystream energy per bit of every block here, should dominate the frame energy, so the average power should follow the bits sent, in the tens of microwatts.

**The study is a calculation on the results of Subsection D and on published values**, plus one block in RTL: the CIC decimator (decimation by 50, three 19-bit integrators at 10 MHz and three combs at the output rate), simulated with Icarus Verilog on a 20,000-bit modulator-like stream and compared output for output with the model's CIC, then synthesised. Its power is estimated from its area at the gate-level power per GE of Grain-128a + MAC, also a 10 MHz datapath; its 31-tap compensation filter (at 200 kS/s) and the proposed design's 32-bit delay line of Subsection A are not included. The metrics are the columns of Table 26 and the rows of the result panel.

Computes the bits per reading at OSR 50 and the average power with the radio of [1]. 4. Prints the table and the result panel. 5. Plots area, power, FOM and transmitted bits.

````python
#@title Fig.: area, power and energy
# Cost of each security block, without scaling any number from one process to another: [1]'s two blocks keep their measured
# 65 nm, 1.2 V powers, this work's blocks their simulated SKY130 (1.8 V) gate-level powers. The blocks synthesised in this
# flow are ranked by a figure of merit that combines energy, process and security, with no modelled quantity in it:
#   FOM = E_bit / (E_NAND2 S_sum), dimensionless, lower is better;
#   E_bit = P_sec / f_bit, the energy per keystream bit of the security logic alone, simulated at gate level: the S-DSM's
#            and the proposal's blocks also carry the two dither LFSRs of [1]'s modulator, whose power (with their counter,
#            output selection and share of the clock tree) is taken out of both; Trivium's block is the cipher alone;
#   E_NAND2 = the energy of one output transition of the library's minimum NAND2 (sky130_fd_sc_hd__nand2_1) driving one
#            identical gate, at the library's typical corner and nominal supply (tt, 25 °C, 1.8 V), from its Liberty data.
#            E_bit and E_NAND2 both scale with C V^2, so E_bit / E_NAND2 counts NAND2-equivalent switchings per bit (the
#            energy analogue of gate equivalents) and is to first order independent of the process;
#   S_sum = the security bits of the goals a design meets, summed: cipher strength (log2 of the cost of the best attack
#            applicable as the cipher is used, at most the key length), the MAC tag bits, the chip-nonce bits, and the protection
#            of the key on a captured chip (the cipher strength for a key present only while the chip is powered, 0 for a key
#            stored on the chip: attack A7).
# The published blocks get no FOM: the E_NAND2 of their libraries is not public.
D = RTL["designs"]
ge = {k: SYN[k]["ge"] for k in D}                # low-power designs: the netlist with clock-gating cells
base_uw = sum(PAPER_FIG4A_UW[k] for k in ("fe", "tia", "s2d", "dsm"))
F_BIT = 10e6                                     # one keystream bit per 10 MHz clock in every design
share = lambda p: 100 * p / (base_uw + p)        # security power / total chip power with security on

def e_nand2(lib=LIBERTY, cell="sky130_fd_sc_hd__nand2_1"):
    """(E_NAND2 in J, its three parts in J, nominal supply in V, input capacitance in pF, input slew in ns): one output
    transition of `cell` driving one input of an identical gate (fan-out 1), from the Liberty tables: the driving gate's
    internal energy (mean of its rise and fall tables over both input arcs), 1/2 C_in V^2 (the supply energy of charging
    the driven input, per transition) and the driven input's own internal energy (mean of rise and fall). The input slew is
    the gate's own output slew at fan-out 1, iterated to a fixed point. The sky130 output tables carry +-1/2 C_L V^2 in the
    rise and fall energies, which the mean cancels, so the load term is added explicitly. Liberty units: pF, ns, pJ."""
    txt = pathlib.Path(lib).read_text()
    v_nom = float(re.search(r"nom_voltage\s*:\s*([\d.]+)", txt).group(1))
    c = txt[txt.index(f'cell ("{cell}")'):]
    c = c[:c.find('cell ("', 10)] if c.find('cell ("', 10) > 0 else c          # this cell only
    nums = lambda t: np.array([float(x) for x in re.findall(r"-?[\d.]+(?:e-?\d+)?", t)])
    def pin(name):
        i = c.index(f'pin ("{name}")'); j = c.find('pin ("', i + 5)
        return c[i:j if j > 0 else len(c)]
    def lut(block, kind):                        # a 2-D table: rows input slew (ns), columns output load (pF)
        m = re.search(kind + r'\s*\("[^"]*"\)\s*\{\s*index_1\("([^"]*)"\);\s*index_2\("([^"]*)"\);\s*values\((.*?)\);', block, re.S)
        i1, i2 = nums(m.group(1)), nums(m.group(2))
        return i1, i2, nums(m.group(3)).reshape(len(i1), len(i2))
    def at(tab, slew, load):
        i1, i2, v = tab
        return float(np.interp(slew, i1, [np.interp(load, i2, r) for r in v]))
    def lut1(block, kind):                       # a 1-D table of an input pin: input slew (ns)
        m = re.search(kind + r'\s*\("[^"]*"\)\s*\{\s*index_1\("([^"]*)"\);\s*values\("([^"]*)"\)', block)
        return nums(m.group(1)), nums(m.group(2))
    c_in = float(re.search(r"\bcapacitance\s*:\s*([\d.]+)", pin("A")).group(1))   # fan-out 1: input A of an identical gate
    out = pin("Y")
    power_arcs, timing_arcs = re.split(r"internal_power \(\)", out)[1:], re.split(r"\btiming \(\)", out)[1:]
    slew = 0.05
    for _ in range(20):
        slew = float(np.mean([0.5 * (at(lut(a, "rise_transition"), slew, c_in) + at(lut(a, "fall_transition"), slew, c_in))
                              for a in timing_arcs]))
    e_int = float(np.mean([0.5 * (at(lut(a, "rise_power"), slew, c_in) + at(lut(a, "fall_power"), slew, c_in)) for a in power_arcs]))
    e_in = 0.5 * float(np.interp(slew, *lut1(pin("A"), "rise_power")) + np.interp(slew, *lut1(pin("A"), "fall_power")))
    e_load = 0.5 * c_in * v_nom ** 2
    return 1e-12 * (e_int + e_load + e_in), dict(internal=1e-12 * e_int, load=1e-12 * e_load, input=1e-12 * e_in), v_nom, c_in, slew
E_NAND2, E_NAND2_PARTS, V_LIB, C_IN, SLEW_FO1 = e_nand2()
E_NAND2 = round(E_NAND2 * 1e15, 2) * 1e-15                 # the printed value (7.93 fJ) enters the FOM, as in the ISCAS paper

# Security logic only: OpenSTA's power of every cell instance of each placed netlist (the activity of its gate-level simulation
# above), each flip-flop assigned to the register it implements, each logic cell to the registers it feeds, each clock cell
# shared among the flip-flops it clocks; the dither LFSRs of [1] (and their counter and output selection) are removed.
CLOCK_CELLS, OUT_PINS = ("dlclkp", "clkbuf", "clkinv"), {"X", "Y", "Q", "Q_N", "GCLK"}
def _grp(table):
    return lambda q: table(q.split("[")[0].strip().split("."))
GROUPS = {
    "sdsm": _grp(lambda s: {"u_key": "key LFSR", "u_d1": "dither LFSRs", "u_d2": "dither LFSRs", "cc": "bit position counter"}.get(s[0], "other")),
    "trivium": _grp(lambda s: {"s": "cipher state", "cnt": "cipher control", "busy": "cipher control", "cc": "bit position counter"}.get(s[0], "other")),
    "proposal": _grp(lambda s: ({"u_core": "cipher state", "u_mac": "MAC"}.get(s[1], "cipher control") if s[0] == "u_grain" else
                                {"u_key": "key, nonce, counter registers", "u_nonce": "key, nonce, counter registers", "u_cn": "key, nonce, counter registers",
                                 "ctr_r": "key, nonce, counter registers", "u_d1": "dither LFSRs", "u_d2": "dither LFSRs"}.get(s[0], "other"))),
    "proposal_lp": _grp(lambda s: (("cipher state" if len(s) > 2 else {"A": "MAC", "R": "MAC", "mb": "MAC message capture"}.get(s[1], "cipher control"))
                                   if s[0] == "u_grain" else
                                   {"key_r": "key, nonce, counter registers", "nonce_r": "key, nonce, counter registers", "cn_r": "key, nonce, counter registers",
                                    "ctr_r": "key, nonce, counter registers", "u_d1": "dither LFSRs", "u_d2": "dither LFSRs",
                                    "dc": "dither bit counter"}.get(s[0], "other")))}
GROUPS["sdsm_lp"], GROUPS["trivium_lp"] = GROUPS["sdsm"], GROUPS["trivium"]
def power_parts(key, tag="power"):
    """{part: µW} of the placed design `key` (its power run `tag`), and the dither's and the security logic's total."""
    wd = WORK / "pnr" / key; group = GROUPS[key]
    tcl = wd / "power_parts.tcl"
    tcl.write_text((wd / f"{tag}.tcl").read_text().replace("report_power\n", "report_power\nreport_power -instances [get_cells *]\n"))
    log = subprocess.run([str(OPENROAD), "-no_init", "-exit", str(tcl)], cwd=wd, capture_output=True, text=True).stdout
    pw = {m[4].replace("\\", ""): 1e6 * float(m[3]) for m in
          re.findall(r"^\s*([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+([\d.eE+-]+)\s+(\S+)\s*$", log, re.M)}
    text = (wd / "core.v").read_text()
    ports = set(re.findall(r"^\s*output\s+(?:\[[^\]]*\]\s*)?(\w+)\s*;", text, re.M))
    info, loads = {}, {}
    for typ, name, body in re.findall(r"(sky130_fd_sc_hd__\w+)\s+(\S+)\s*\((.*?)\);", text, re.S):
        if any(x in typ for x in ("fill", "tap", "decap", "diode")):
            continue
        pins = {pp: n.strip().replace("\\", "").strip() for pp, n in re.findall(r"\.(\w+)\(([^)]*)\)", body)}
        nm = name.replace("\\", ""); info[nm] = (typ, pins)
        for pp, n in pins.items():
            if pp not in OUT_PINS:
                loads.setdefault(n, []).append(nm)
    ff = {n: group(pp.get("Q", "")) for n, (t, pp) in info.items() if "df" in t}
    memo, pmemo = {}, {}
    def feeds(n, clk, depth=0):                  # register groups reached through logic, or through the clock tree
        if (n, clk) not in memo:
            memo[(n, clk)], acc = {}, {}
            for pp, net in info[n][1].items():
                if pp in OUT_PINS:
                    for ld in loads.get(net, []):
                        if ld in ff:
                            if clk and info[ld][1].get("CLK") != net:
                                continue
                            sub = {ff[ld]: 1}
                        else:
                            is_clk = any(c in info[ld][0] for c in CLOCK_CELLS)
                            sub = feeds(ld, clk, depth + 1) if depth < 60 and (is_clk or not clk) else {}
                        for g, c in sub.items():
                            acc[g] = acc.get(g, 0) + c
            memo[(n, clk)] = acc
        return memo[(n, clk)]
    def reach_ports(n, depth=0):                 # output ports that a register-free logic cell drives
        if n not in pmemo:
            pmemo[n], r = set(), set()
            for pp, net in info[n][1].items():
                if pp in OUT_PINS:
                    if net.split("[")[0].strip() in ports:
                        r.add(net.split("[")[0].strip())
                    for ld in loads.get(net, []):
                        if ld not in ff and depth < 60:
                            r |= reach_ports(ld, depth + 1)
            pmemo[n] = r
        return pmemo[n]
    res = {}
    for n, (typ, pins) in info.items():
        pn = pw.get(n, 0.0)
        if "df" in typ:
            sh = {ff[n]: 1.0}
        elif any(c in typ for c in CLOCK_CELLS):
            f = feeds(n, True); tot = sum(f.values())
            sh = {("clock: " + g): c / tot for g, c in f.items()} if tot else {"clock: unassigned": 1.0}
        else:
            f = feeds(n, False)
            if f:
                sh = {max(f, key=f.get): 1.0}
            else:
                r = reach_ports(n)
                sh = {("dither output selection" if r and all(q.startswith("dith") for q in r) else "output selection"): 1.0}
        for g, w in sh.items():
            res[g] = res.get(g, 0.0) + pn * w
    dither = sum(w for g, w in res.items() if g.replace("clock: ", "").startswith("dither"))
    return res, dither, sum(res.values()) - dither
PARTS = {k: power_parts(k, "power_est" if k == "proposal_lp" else "power") for k in ("sdsm", "trivium", "proposal", "sdsm_lp", "trivium_lp", "proposal_lp")}
for _k, (_r, _d, _s) in PARTS.items():                   # the parts add up to the block's power of Subsection D
    assert abs(sum(_r.values()) - P130[_k]) < 0.02 * P130[_k], (_k, sum(_r.values()), P130[_k])
P_SEC = {k: v[2] for k, v in PARTS.items()}            # µW of the security logic alone
P_DITH = {k: v[1] for k, v in PARTS.items()}

# Security bits per goal met (S_sum is their sum): cipher strength by the best attack applicable as the cipher is used, MAC
# tag, chip nonce, key protection on a captured chip
TAG_BITS, NC_BITS = 32, 31                       # Grain-128a's MAC tag; the chip nonce N_c in the IV (Section IV)
FRAME_BITS = N_DSM                               # keystream bits per initialisation (one frame; every frame re-initialises)
S_FCA = {"todo": 113.8, "zhang": 106.3}          # log2 of the keystream data from ONE initialisation that the fast correlation
                                                 # attacks on Grain-128a need [18], [19]: far beyond a frame, so they do not apply
S_BM = np.log2(16 ** 2 * 4)                      # Berlekamp-Massey: about 2^10 steps from 32 keystream bits (attack A3)
SEC = {"sdsm": dict(cipher=round(S_BM), tag=0, nonce=0, key=0),   # 16-bit LFSR, broken by Berlekamp-Massey; stored key (A7)
       "trivium": dict(cipher=80, tag=0, nonce=0, key=0),         # 80-bit key, no attack below brute force; no tag or nonce in [1]
       "proposal": dict(cipher=128, tag=TAG_BITS, nonce=NC_BITS, key=128)}   # 128-bit key, present only while powered
assert 2 ** min(S_FCA.values()) > 1e6 * FRAME_BITS
S_SUM = {k: sum(v.values()) for k, v in SEC.items()}
base_of = lambda k: k.split("_")[0]
EBIT = {k: P_SEC[k] * 1e-6 / F_BIT for k in P_SEC}
EBIT_BLOCK = {k: P130[k] * 1e-6 / F_BIT for k in P_SEC}          # the whole block, dither included (Subsection D)
EBIT_MEAS = dict(sdsm=PAPER_FIG4A_UW["sdsm"] * 1e-6 / F_BIT, trivium=PAPER_FIG4A_UW["trivium"] * 1e-6 / F_BIT,
                 dtls=0.89e-9 / 8)               # [9]: 0.89 nJ per byte encrypted, 65 nm (operated at 0.8 V)
N_SW = {k: e / E_NAND2 for k, e in EBIT.items()}                 # NAND2-equivalent switchings per keystream bit
FOM = {k: n / S_SUM[base_of(k)] for k, n in N_SW.items()}
FOM_CIPHER = {k: N_SW[k] / SEC[base_of(k)]["cipher"] for k in ("sdsm_lp", "trivium_lp", "proposal_lp")}   # cipher strength only
# robustness: weight w on the tag, nonce and key-protection bits; the proposal leads Trivium for w above W_LEAD
_extra = S_SUM["proposal"] - SEC["proposal"]["cipher"]
W_LEAD = max(0.0, (N_SW["proposal_lp"] / FOM["trivium_lp"] - SEC["proposal"]["cipher"]) / _extra)

# Information-equivalent conventional baseline: the receiver's third-order CIC decimator (R = OSR = 50) moved onto the sensor
# in RTL, checked against the model's CIC and synthesised; then Grain-128a + MAC encrypts B-bit words at the Nyquist rate.
CIC_V = r"""// third-order CIC decimator by 50 for the 1-bit modulator output (0/1 -> -1/+1): three integrators at the modulator
// clock, three combs at the output rate, 19-bit registers (1 + 3 log2 50, rounded up); phase aligned with the receiver model
module cic3(input wire clk, input wire rst, input wire d, output reg [18:0] y, output reg out_valid);
  reg [18:0] i1, i2, i3, c1, c2, c3; reg [5:0] cnt;
  wire [18:0] x = d ? 19'd1 : {19{1'b1}};
  wire [18:0] d1 = i3 - c1, d2 = d1 - c2;
  always @(posedge clk) begin
    if (rst) begin i1 <= 0; i2 <= 0; i3 <= 0; c1 <= 0; c2 <= 0; c3 <= 0; cnt <= 47; y <= 0; out_valid <= 0; end
    else begin
      i1 <= i1 + x; i2 <= i2 + i1; i3 <= i3 + i2; out_valid <= 0;
      if (cnt == 49) begin cnt <= 0; c1 <= i3; c2 <= d1; c3 <= d2; y <= d2 - c3; out_valid <= 1; end
      else cnt <= cnt + 1;
    end
  end
endmodule
"""
t0 = time.time(); wd_b = WORK / "baseline"; wd_b.mkdir(parents=True, exist_ok=True)
(wd_b / "cic3.v").write_text(CIC_V)
r_b = rng_for(5150); n_b = 50 * 400
D_b = (r_b.random(n_b) < 0.5 + 0.45 * np.sin(2 * np.pi * np.arange(n_b) / 2000)).astype(np.uint8)   # a busy bit stream
(wd_b / "d.mem").write_text("\n".join(map(str, D_b)) + "\n")
(wd_b / "tb.v").write_text(f"""module tb; reg clk = 0, rst = 1, d = 0; reg dm [0:{n_b - 1}]; integer c, fo; wire [18:0] y; wire ov;
  cic3 u(.clk(clk), .rst(rst), .d(d), .y(y), .out_valid(ov)); always #50 clk = ~clk;
  initial begin $readmemb("d.mem", dm); fo = $fopen("y.txt"); @(posedge clk); #1 rst = 0;
    for (c = 0; c < {n_b} + 3; c = c + 1) begin d = (c < {n_b}) ? dm[c] : 0; @(posedge clk); #1 if (ov) $fwrite(fo, "%0d\\n", $signed(y)); end
    $finish; end
endmodule
""")
subprocess.run([str(PNR_BIN / "iverilog"), "-g2012", "-o", "cic.vvp", "tb.v", "cic3.v"], cwd=wd_b, check=True, capture_output=True)
subprocess.run([str(PNR_BIN / "vvp"), "-n", "cic.vvp"], cwd=wd_b, check=True, capture_output=True)
y_hw = np.array([int(v) for v in (wd_b / "y.txt").read_text().split()])[1:]     # the first output precedes any input
y_ref = np.rint(cic(D_b) * 50 ** 3).astype(int)
cic_ok = len(y_hw) >= len(y_ref) and np.array_equal(y_hw[:len(y_ref)], y_ref)
S_CIC = R.synthesize(YOSYS, "cic3", [str(wd_b / "cic3.v")], str(LIBERTY), str(wd_b))
# its power: the SKY130 gate-level power per GE of Grain-128a + MAC (a datapath clocked at 10 MHz, like the integrators); an estimate
P130_CIC = P130["grain_auth2x"] / ge["grain_auth2x"] * S_CIC["ge"]
# the baseline sends 16-bit words, the word length of [1]: a shorter word would rest on the modelled SNDR, which is not used here

OSR_E = 50                                               # every accuracy result of this notebook is at OSR 50
frame_extra = lambda osr: 96 * (osr * NYQ) / N_DSM       # counter + chip nonce + tag (96 bits) per frame, per 20 samples
P_BASE = P130["grain_auth2x"] + P130_CIC                 # baseline security power, SKY130 (decimator estimated per GE)
# Bits the sensor must transmit per reading (Nyquist sample) for the same information: set by the architecture alone, no radio
# assumed (no radio is designed here). Modulator-level designs send every modulator bit (OSR per sample, plus the frame fields
# of the proposal); the conventional design of [1] sends 16-bit words at 4.5-5x the Nyquist rate (Fig. 4(b) of [1]); the
# information-equivalent design sends B-bit words at the Nyquist rate with the same frame fields.
BITS = {"sdsm": OSR_E, "proposal": OSR_E + frame_extra(OSR_E) / NYQ,
        "trivium": (16 * 4.5, 16 * 5.0),
        "base16": 16 + frame_extra(OSR_E) / NYQ}
BITS30 = dict(sdsm=30, proposal=30 + frame_extra(30) / NYQ, trivium=16 * 5.0)
for _k in ("sdsm_lp", "trivium_lp", "proposal_lp"):
    BITS[_k] = BITS[_k.replace("_lp", "")]
fmt_bits = lambda b: f"{b[0]:.0f} – {b[1]:.0f}" if isinstance(b, tuple) else f"{b:.2f}" if b % 1 else f"{b:.0f}"
# the information-equivalent baseline has no FOM: its decimator power is an estimate, not a simulated or measured value

# Average power of a duty-cycled sensor, as an example with the radio of [1] (no radio is designed in this work): one frame of
# N_DSM modulator clocks per minute; per frame, the radio energy of the bits sent (the radio energy per bit of [1]'s energy
# table, 3.1 mW at 200 kb/s) plus the front end of [1] and the security block running for the frame; key generation at
# power-up and the sleep power are excluded (neither is designed here).
FRAME_S, PERIOD_S = N_DSM / FS, 60.0
READINGS = N_DSM / OSR_E                                 # Nyquist samples (readings) per frame at OSR 50
def avg_power(bits_per_reading, p_sec_uw):
    """(average power in µW at one frame per PERIOD_S, radio share of the frame energy)."""
    e_radio = bits_per_reading * READINGS * RADIO_J_PER_BIT
    e_chip = (base_uw + p_sec_uw) * 1e-6 * FRAME_S
    return 1e6 * (e_radio + e_chip) / PERIOD_S, e_radio / (e_radio + e_chip)
AVG = {"sdsm": avg_power(BITS["sdsm"], P130["sdsm_lp"]), "proposal": avg_power(BITS["proposal"], P130["proposal_lp"]),
       "base16": avg_power(BITS["base16"], P_BASE)}

fmt_p = lambda p: f"{p:.1f}" if p < 100 else f"{p:.0f}"
fmt_f = lambda f: f"{f:.2f}" if f < 10 else f"{f:.1f}"
def sec_txt(k):
    parts = [f"{v:g}" for v in SEC[k].values() if v]
    return " + ".join(parts) + (f" = {S_SUM[k]:g}" if len(parts) > 1 else "")
COST = []
for name, k in [("S-DSM [1]", "sdsm"), ("Trivium (conventional, [1])", "trivium")]:
    p = PAPER_FIG4A_UW[k]
    COST.append((name, "–", "measured [1], 65 nm, 1.2 V", fmt_p(p), "–", f"{p / 10:.2f}", "–", "–", "– (no E<sub>NAND2</sub>)",
                 fmt_bits(BITS[k])))
for name, k in [("S-DSM [1], as written, this flow", "sdsm"), ("Trivium, as written, this flow", "trivium"),
                ("Proposal, 1 bit/clock (Subsection B)", "proposal"), ("S-DSM [1], low power, this flow", "sdsm_lp"),
                ("Trivium, low power, this flow", "trivium_lp"), ("Proposal, low power (full block)", "proposal_lp")]:
    best = k == "proposal_lp"
    COST.append((f"<b>{name}</b>" if best else name, f"{ge[k]:,.0f}", "simulated, SKY130, 1.8 V", fmt_p(P130[k]), fmt_p(P_SEC[k]),
                 f"{P_SEC[k] / 10:.2f}", f"{N_SW[k]:,.0f}", sec_txt(base_of(k)), f"<b>{FOM[k]:.2f}</b>" if best else fmt_f(FOM[k]),
                 fmt_bits(BITS[k])))
COST.append(("Decimate + Grain-128a/MAC, 16-bit words at Nyquist", f"{ge['grain_auth2x'] + S_CIC['ge']:,.0f}*",
             "simulated*, SKY130, 1.8 V", f"{P_BASE:.0f}*", "–", f"{P_BASE / 10:.2f}*", "–", "–", "– (estimated power)",
             fmt_bits(BITS["base16"])))
table(["Design", "Area (GE)", "Source", "Power at 10 MHz (µW)", "Security logic (µW)", "Energy per bit (pJ/b)",
       "E<sub>bit</sub>/E<sub>NAND2</sub> (switchings per bit)", "S<sub>Σ</sub> (bits)", "FOM", f"Bits sent per reading, OSR {OSR_E}"], COST,
      caption=f"Table 26: cost of each security block, every number as measured or simulated in its own process, none scaled: "
              f"[1]'s two blocks measured in 65 nm at 1.2 V; this work's blocks simulated at gate level in SKY130 "
              f"(sky130_fd_sc_hd, tt, 25 °C, 1.8 V, the same flow for every design). Energy per bit: power over the 10 Mb/s "
              f"keystream, of the security logic alone for this flow's blocks (the dither LFSRs of [1] removed). E<sub>NAND2</sub> = {E_NAND2 * 1e15:.2f} fJ: one output transition of the library's minimum NAND2 at "
              f"fan-out 1 (tt, 25 °C, 1.8 V, from its Liberty data). S<sub>Σ</sub>: the security bits of the goals met (cipher "
              f"strength by the best attack applicable as used + MAC tag + chip nonce + key protection on a captured chip). FOM = E<sub>bit</sub>/(E<sub>NAND2</sub> "
              f"S<sub>Σ</sub>), dimensionless, lower is better, computed for the blocks of this flow only (the E<sub>NAND2</sub> of "
              f"the published blocks' libraries is not public). Bits sent per reading: transmitted bits per Nyquist sample for the "
              f"same information, set by the architecture. *: the decimator's area is synthesised and its power estimated per GE; "
              f"its 31-tap compensation filter is not included; the baseline's cipher is the one written for one bit per clock, "
              f"which disadvantages it")
card("Cost, figure of merit, transmitted bits and average power",
     [("figure of merit (no number scaled, nothing modelled)", "FOM = E<sub>bit</sub> / (E<sub>NAND2</sub> S<sub>Σ</sub>): the energy "
       "per keystream bit of the security logic (gate-level simulation, the dither of [1] removed) over the energy of one switching of the library's minimum NAND2 gate, so "
       "E<sub>bit</sub>/E<sub>NAND2</sub> counts NAND2-equivalent switchings per bit (C V<sup>2</sup> cancels: the energy analogue "
       "of gate equivalents), per bit of security delivered, summed over the goals met; lower is better"),
      ("E<sub>NAND2</sub>, SKY130 (sky130_fd_sc_hd__nand2_1, fan-out 1, tt, 25 °C, 1.8 V)",
       f"<b>{E_NAND2 * 1e15:.2f} fJ</b> = internal {E_NAND2_PARTS['internal'] * 1e15:.2f} + ½C<sub>in</sub>V<sup>2</sup> "
       f"{E_NAND2_PARTS['load'] * 1e15:.2f} + driven input {E_NAND2_PARTS['input'] * 1e15:.2f} fJ (C<sub>in</sub> = "
       f"{C_IN * 1e3:.2f} fF, input slew {SLEW_FO1 * 1e3:.0f} ps, from the Liberty tables)"),
      ("security bits S<sub>Σ</sub> (cipher strength + MAC tag + chip nonce + key protection)",
       f"S-DSM {S_SUM['sdsm']:g} (16-bit LFSR: Berlekamp-Massey in about 2<sup>{S_BM:.0f}</sup> steps, A3; key stored), Trivium "
       f"{S_SUM['trivium']:g} (80-bit key, stored), proposal {sec_txt('proposal')} (128-bit key: the fast correlation attacks "
       f"[18], [19] need 2<sup>{S_FCA['zhang']:g}</sup> or more bits from one initialisation, a frame {FRAME_BITS:,}; "
       f"{TAG_BITS}-bit tag, {NC_BITS}-bit chip nonce, a key present only while powered)"),
      ("security logic: the dither LFSRs of [1] removed (S-DSM and proposal)",
       f"S-DSM {P_SEC['sdsm_lp']:.1f} of {P130['sdsm_lp']:.1f} µW ({100 * P_DITH['sdsm_lp'] / P130['sdsm_lp']:.1f} % dither), "
       f"Trivium {P_SEC['trivium_lp']:.1f} µW (no dither), proposal {P_SEC['proposal_lp']:.1f} of {P130['proposal_lp']:.1f} µW "
       f"({100 * P_DITH['proposal_lp'] / P130['proposal_lp']:.1f} % dither), low-power form"),
      ("FOM, this flow, 1 bit/clock", f"S-DSM {fmt_f(FOM['sdsm'])}, Trivium {fmt_f(FOM['trivium'])}, proposal {fmt_f(FOM['proposal'])}"),
      ("FOM, this flow, low power", f"S-DSM {fmt_f(FOM['sdsm_lp'])}, Trivium {fmt_f(FOM['trivium_lp'])}, proposal "
       f"<b>{FOM['proposal_lp']:.2f}</b>"),
      ("proposal against Trivium and the S-DSM (low power, same flow)",
       f"<b>{FOM['trivium_lp'] / FOM['proposal_lp']:.1f}× lower FOM than Trivium</b> ({EBIT['proposal_lp'] / EBIT['trivium_lp']:.1f}× "
       f"the energy per bit for {S_SUM['proposal'] / S_SUM['trivium']:.1f}× the security bits) and "
       f"{FOM['sdsm_lp'] / FOM['proposal_lp']:.1f}× lower than the S-DSM ({EBIT['proposal_lp'] / EBIT['sdsm_lp']:.1f}× the energy "
       f"per bit for {S_SUM['proposal'] / S_SUM['sdsm']:.1f}× the security bits)"),
      ("with cipher strength alone (no credit for tag, nonce or key protection)",
       f"S-DSM {fmt_f(FOM_CIPHER['sdsm_lp'])}, Trivium {fmt_f(FOM_CIPHER['trivium_lp'])}, proposal {fmt_f(FOM_CIPHER['proposal_lp'])}; "
       f"with weight w on the tag, nonce and key-protection bits the proposal leads Trivium for w above {100 * W_LEAD:.1f} %"),
      ("energy per bit as measured in [1] and [9] (no FOM: E<sub>NAND2</sub> of their libraries not public)",
       f"S-DSM {EBIT_MEAS['sdsm'] * 1e12:.2f} pJ/b (24 µW), Trivium {EBIT_MEAS['trivium'] * 1e12:.1f} pJ/b (220 µW), DTLS engine [9] "
       f"{EBIT_MEAS['dtls'] * 1e12:.0f} pJ/b (0.89 nJ/byte), all 65 nm; this flow, low power: S-DSM {EBIT['sdsm_lp'] * 1e12:.2f}, "
       f"Trivium {EBIT['trivium_lp'] * 1e12:.2f}, proposal {EBIT['proposal_lp'] * 1e12:.2f} pJ/b (security logic); whole blocks "
       f"{EBIT_BLOCK['sdsm_lp'] * 1e12:.2f}, {EBIT_BLOCK['trivium_lp'] * 1e12:.2f}, {EBIT_BLOCK['proposal_lp'] * 1e12:.2f} pJ/b"),
      (f"bits sent per reading for the same information, OSR {OSR_E}", f"S-DSM {fmt_bits(BITS['sdsm'])}, proposal "
       f"{fmt_bits(BITS['proposal'])} (with its counter, chip nonce and tag), decimate + Trivium of [1] {fmt_bits(BITS['trivium'])} "
       f"(16-bit words at 4.5–5× Nyquist): the proposal sends <b>{BITS['trivium'][0] / BITS['proposal']:.2f}–"
       f"{BITS['trivium'][1] / BITS['proposal']:.2f}× fewer bits</b> than the conventional design of [1]"),
      ("information-equivalent baseline (decimate on the sensor, Grain-128a + MAC, words at the Nyquist rate)",
       f"CIC decimator {S_CIC['ge']:,.0f} GE, {S_CIC['ffs']} flip-flops, {'bit-exact with the model' if cic_ok else '<b>MISMATCH</b>'} "
       f"on {len(y_ref)} outputs; 16-bit words at the Nyquist rate, the word length of [1]: <b>{fmt_bits(BITS['base16'])} bits per reading</b>, "
       f"i.e. the proposal sends <b>{BITS['proposal'] / BITS['base16']:.1f}× more</b>"),
      ("for reference: OSR 30, the setting of the 2.67× of [1]", f"bits per reading: proposal {fmt_bits(BITS30['proposal'])}, S-DSM "
       f"{BITS30['sdsm']}, decimate + Trivium {BITS30['trivium']:.0f}: {BITS30['trivium'] / BITS30['proposal']:.2f}× fewer"),
      (f"average power at one frame per minute, with the radio of [1] (an example; no radio is designed here)",
       f"radio {RADIO_J_PER_BIT * 1e9:.1f} nJ/bit ([1]'s energy table), one frame of {N_DSM:,} clocks ({1e3 * FRAME_S:.2f} ms, "
       f"{READINGS:,.0f} readings), front end of [1] ({base_uw:,} µW) during the frame: proposal <b>{AVG['proposal'][0]:.1f} µW</b>, "
       f"S-DSM {AVG['sdsm'][0]:.1f} µW, decimate + Grain-128a/MAC {AVG['base16'][0]:.1f} µW (decimator power estimated); the radio "
       f"takes {100 * AVG['proposal'][1]:.1f} % of the proposal's frame energy; key generation at power-up and sleep power excluded"),
      ("run time", f"{time.time() - t0:.0f} s (decimator simulation and synthesis)")],
     note="No power is scaled from one process to the other. E<sub>NAND2</sub> takes the energy scale of the process out of the "
          "energy per bit through the library's own characterisation data, not through an assumed scaling law; wires and leakage "
          "do not scale exactly with gate energy, so the cancellation holds to first order. The FOM credits security strength and "
          "the goals met, not the attacks of Section IX that a design withstands (the S-DSM 4 of 15; the proposal 14 at "
          "the working setting and all 15 with the two-temperature enrolment). The average powers use the radio energy per bit of "
          "[1] only as an example: with any radio whose energy per bit is the same for every design and dominates, the energy "
          "ratio equals the bit ratio.", tone="ok" if cic_ok else "bad")

# (a)-(c) as Fig. 6 of the ISCAS paper: area, power at 10 MHz and the proposed FOM of the three blocks, each as written for
# one bit per clock (light, hatched) and in the low-power form (dark, solid), one hue per metric; (d) the bits sent
from matplotlib.patches import Patch
HUE = {"area": ("#ddd0ef", "#7b5ea7"), "power": ("#fbe2b5", "#d98a00"), "fom": ("#c9dbf8", COL["ok"])}
PAIR = [("S-DSM\n[1]", "sdsm"), ("Tri-\nvium", "trivium"), ("Pro-\nposed", "proposal")]
xs, wb = np.arange(len(PAIR)) * 1.25, 0.46
fig, (a1, a2, a3, a4) = plt.subplots(1, 4, figsize=(W2, 3.1), gridspec_kw=dict(width_ratios=[1, 1, 1, 1.1], wspace=0.5))
def bars(ax, vals, hue, fmt, log=False):
    light, dark = HUE[hue]
    ax.set_axisbelow(True); ax.yaxis.grid(True, color=COL["grid"], lw=0.5, ls="-")
    for j, (col, off) in enumerate(((light, -wb / 2), (dark, wb / 2))):
        v = [vals[k + ("_lp" if j else "")] for _, k in PAIR]
        ax.bar(xs + off, v, wb, color=col, ec=COL["ink"], lw=0.5, hatch=None if j else "////")
        for xi, vi in zip(xs + off, v):
            bold = j == 1 and xi > 1.5
            ax.text(xi, vi * 1.1 if log else vi + 0.015 * ax.get_ylim()[1], fmt(vi), ha="center",     # above its own bar
                    va="bottom", fontsize=6.3, fontweight="bold" if bold else "normal")
    ax.set_xlim(-0.95, xs[-1] + 0.8); ax.set_xticks(xs, [n for n, _ in PAIR], fontsize=7.5, rotation=0)
    ax.tick_params(axis="x", length=0)
a1.set_ylim(0, 1.2 * max(ge[k + "_lp"] for _, k in PAIR) / 1e3)
bars(a1, {k: v / 1e3 for k, v in ge.items()}, "area", lambda v: f"{v:.1f}")
a1.set_ylabel("Area (kGE)"); panel(a1, "a", "area")
a2.set_ylim(0, 1.2 * max(P130[k] for _, k in PAIR))
bars(a2, P130, "power", lambda v: f"{v:.1f}" if v < 100 else f"{v:.0f}")
a2.set_ylabel("Power (µW)"); panel(a2, "b", "power at 10 MHz")
a3.set_yscale("log"); a3.set_ylim(1, 100)
bars(a3, FOM, "fom", fmt_f, log=True)
a3.set_yticks([1, 10, 100], ["1", "10", "100"]); a3.set_ylabel("FOM (lower is better)"); panel(a3, "c", "proposed FOM")
fig.legend([Patch(fc="#e6e8eb", ec=COL["ink"], lw=0.5, hatch="////"), Patch(fc="#6b7280", ec=COL["ink"], lw=0.5)],
           ["one bit per clock (light, hatched)", "low-power form (dark, solid)"], loc="upper left", ncol=2, fontsize=7.5,
           frameon=False, handlelength=1.2, columnspacing=1.6, bbox_to_anchor=(0.04, 0.985))
rows_b = [("S-DSM [1]", BITS["sdsm"]), ("Proposal", BITS["proposal"]), ("Dec. +\nTrivium [1]", BITS["trivium"][1]),
          ("Dec. + Grain/\nMAC, 16 bit", BITS["base16"])]
for i, (lab, b) in enumerate(rows_b):
    a4.barh(i, BITS["trivium"][0] if i == 2 else b, color=COL["ok"] if lab == "Proposal" else "#c9ced6", ec=COL["ink"], lw=0.5, height=0.55)
    a4.text(b + 1.5, i, f"{BITS['trivium'][0]:.0f}–{b:.0f}" if i == 2 else fmt_bits(b), va="center", fontsize=7.5)
a4.barh(2, BITS["trivium"][1] - BITS["trivium"][0], left=BITS["trivium"][0], color="white", hatch="////", ec=COL["ink"], lw=0.5,
        height=0.55, label="range of [1]")
a4.axhline(2.5, color=COL["faint"], lw=0.6, ls=":")
a4.text(0.97 * 1.5 * BITS["trivium"][1], 0.5, "undecimated", ha="right", va="center", fontsize=7, color=COL["faint"])
a4.text(0.97 * 1.5 * BITS["trivium"][1], 3.0, "decimated\non the sensor", ha="right", va="center", fontsize=7, color=COL["faint"])
a4.set_yticks(range(len(rows_b)), [r[0] for r in rows_b]); a4.tick_params(axis="y", left=False, labelsize=7)
a4.set_ylim(len(rows_b) - 0.5, -1.1); a4.set_xlim(0, 1.5 * BITS["trivium"][1]); a4.set_xlabel("Bits per reading")
a4.legend(fontsize=7, loc="upper right", handlelength=1.0)
panel(a4, "d", "bits sent")
fig.subplots_adjust(left=0.065, right=0.995, bottom=0.15, top=0.83)
plt.show()
````

### G. Complete system power: the proposed design against [1] under the same conditions

The chip's power is $P_{\text{chip}} = P_{\text{FE}} + P_{\text{TIA}} + P_{\text{S2D}} + P_{\Delta\Sigma} + P_{\text{sec}}$, where $P_{\text{FE}}$, $P_{\text{TIA}}$, $P_{\text{S2D}}$ and $P_{\Delta\Sigma}$ are the powers of the front end, the TIA, the S-to-D buffer and the ΔΣ modulator, and its current at the 1.2 V supply of [1] is $I = P_{\text{chip}}/1.2\ \text{V}$. The proposed design leaves the four analog blocks unchanged (Section IV), so only $P_{\text{sec}}$ differs. This work does not simulate the analog blocks' power: they are behavioural in the model (Section V), except the comparator, simulated at transistor level at about 2 µW (Section VIII); their powers are the values measured in [1], marked as such, and the radio is left out. The two fair comparisons answer different questions. *Unscaled:* the proposed design's SKY130 gate-level block on the chip of [1], beside the S-DSM's 24 µW as measured; a SKY130 gate switches more energy than a 65 nm gate, which disadvantages the proposed design, but [1]'s measured blocks may contain more than their logic (interfaces, key loading, a supply that feeds more than the block), which disadvantages [1], so the net direction of this bias is not established. *Strictly equal treatment:* [1]'s design re-implemented in the same flow with the same low-power techniques, a ratio independent of the process, and the FOM of Subsection E. **The prediction from Subsection D is that the equal-treatment ratio stays above one**.

The analog blocks are those of Subsection E; the security blocks take [1]'s measured values and this work's powers of Table 23 in both forms and routed. The metrics are each block's power and current, the chip's totals, the security share and the bits for the same information at OSR 50.

### H. Side-channel exposure of the placed netlists

The leakage test of Section IX-D is repeated on the netlists intended for fabrication. In one clock, a CMOS netlist dissipates $\tfrac{1}{2} C V_\mathrm{DD}^2$ for every net that changes, with $C$ the net's capacitance (the gate inputs it drives and its wire). Against the register-level model of Section IX-D, this adds the combinational logic (for Grain-128a, the 32 rounds that the low-power form evaluates per clock), the clock gating that keeps idle registers still, and the unequal loads of a placed design; it omits glitches (zero delay) and the cells' internal power. The test and its threshold on the magnitude of Welch's $t$ statistic, $\lvert t\rvert$ = 4.5, are those of Section IX-D. The analysis above predicts that the proposed design leaks while the key is shifted into its register, since every shift toggles the register according to the key bits, and during the first updates of the initialisation, but not in the keystream and MAC phase, and that **the S-DSM leaks whenever its key LFSR is updated**, once every 16 clocks in its low-power form. In Fig. 71, a leak appears as a clock at which the solid $\lvert t\rvert$ curve crosses the dashed threshold: the prediction places these clocks in the shaded key shift-in and initialisation of panel (a) and at the LFSR updates of panel (b).

**The test runs live at gate level on the placed netlists** of Subsections C and D: the proposed design's low-power block (the netlist that the recorded run of Subsection C routes) and the S-DSM's low-power block. The capacitance of every net is read from OpenSTA with the wire RC estimated on the placement, as for the power of Subsection D (the extracted wires with `ROUTE_BLOCK`). Each trace of the proposed design runs its whole protocol (Subsection A): the key shift-in in clocks 2–129, the session and chip nonces in 130–192, the start in 193, the initialisation and the MAC preload, then a 64-bit message with its tag, 272 clocks in all, with a random nonce, chip nonce and message per trace; the S-DSM loads its 16-bit key in clock 2 and encrypts a random modulator stream for as many clocks. 2,000 traces per group are simulated in one pass, bit-parallel and with zero delay, every flip-flop starting from the same power-up state, with Gaussian measurement noise of 4 changes of an average flip-flop output per clock, as the 4 toggles of Section IX-D. Before the test, the keystream and the tag of five traces, from all three groups, are compared bit for bit with the RTL flow's reference model (Subsection B), and three traces of the S-DSM with the LFSR of [1].

> **Next cell.** 1. Reads the capacitance of every net of both placed netlists from OpenSTA. 2. Simulates the three groups of traces through each netlist and checks some of them against the model. 3. Computes Welch's $t$ at every clock for the fixed-versus-random test and the control. 4. Prints the result panel and draws Fig. 71.

````python
#@title Fig.: side-channel exposure of the placed netlists
# The fixed-versus-random leakage test of Section IX-D, repeated on the placed gate-level netlists of this
# section (the proposal's low-power block as placed in Subsection C, to which the recorded routing adds only buffers,
# antenna diodes and fill cells, none of which changes its logic, and the S-DSM's low-power block): each
# trace is the switching energy per clock of every data net of the netlist, 1/2 C V^2 per change with the net's
# capacitance from the placement's wire estimate (the extracted wires with ROUTE_BLOCK), simulated bit-parallel for
# thousands of traces at once (pnr_flow.GateSim, zero delay); measurement noise is added and Welch's t compares, clock by
# clock, one fixed key against random keys, and two groups of random keys (the control). Every trace runs the block's
# whole protocol: key shift-in, session nonce and chip nonce, start, initialisation and a 64-bit message with its tag;
# random nonces, chip nonces and messages per trace. Before the test, the keystream and the tag of traces of every group
# are checked against the model of the RTL flow. Inputs: R, PR, OPENROAD, PDK_ROOT, LIBERTY, WORK, run_c, VLOG_C, SO_C,
# RUN130, TH_TV. Outputs: TVG.
import random
t0_g = time.time()
N_G, L_MSG = 2000, 64                               # traces per group; message bits of the frame
rng_g = random.Random(2041)
def _words(bits):                                   # one bit per trace -> uint64 words (bit j of word w: trace 64 w + j)
    b = np.zeros(((len(bits) + 63) // 64) * 64, np.uint8); b[:len(bits)] = bits
    return np.packbits(b.reshape(-1, 64), axis=1, bitorder="little").view("<u8").astype(np.uint64).ravel()
def _stream(n_cycles, per_trace):                   # per_trace: {cycle: bit list over the traces} -> (n_cycles, W) words
    W = (len(next(iter(per_trace.values()))) + 63) // 64
    a = np.zeros((n_cycles, W), np.uint64)
    for c, bits in per_trace.items():
        a[c] = _words(bits)
    return a
N_T = 3 * N_G                                       # fixed key, random keys A, random keys B
# the proposal: key k_0 first in cycles 2-129, nonce MSB first in 130-161, chip nonce in 162-192, start in 193, the
# message from RUN (cycle 204) on, the pad after it; dither seeds fixed
k_fix = [rng_g.getrandbits(1) for _ in range(128)]
keys = [k_fix] * N_G + [[rng_g.getrandbits(1) for _ in range(128)] for _ in range(2 * N_G)]
nonces, cns = [rng_g.getrandbits(32) for _ in range(N_T)], [rng_g.getrandbits(31) for _ in range(N_T)]
msgs = [[rng_g.getrandbits(1) for _ in range(L_MSG)] for _ in range(N_T)]
R0 = 193 + R.RUN_DELAY_LP
NC_P = R0 + L_MSG + 4
sim_p = PR.GateSim(VLOG_C, str(LIBERTY), PR.net_caps(str(OPENROAD), PDK_ROOT, run_c, str(WD_C),
                                                    spef=SO_C["files"]["spef"] if ROUTE_BLOCK else None))
ins_p = dict(sim_p.bits("seed1", 0x1234), **sim_p.bits("seed2", 0xBEEF))
ins_p["dith_load"] = _stream(NC_P, {1: [1] * N_T})
ins_p["key_sh"] = _stream(NC_P, {2 + j: [1] * N_T for j in range(128)})
ins_p["key_in"] = _stream(NC_P, {2 + j: [k[j] for k in keys] for j in range(128)})
ins_p["nonce_sh"] = _stream(NC_P, {130 + j: [1] * N_T for j in range(32)})
ins_p["nonce_in"] = _stream(NC_P, {130 + j: [(n >> (31 - j)) & 1 for n in nonces] for j in range(32)})
ins_p["cn_sh"] = _stream(NC_P, {162 + j: [1] * N_T for j in range(31)})
ins_p["cn_in"] = _stream(NC_P, {162 + j: [(n >> (30 - j)) & 1 for n in cns] for j in range(31)})
ins_p["start"] = _stream(NC_P, {193: [1] * N_T})
ins_p["m"] = _stream(NC_P, {R0 + j: [m[j] for m in msgs] for j in range(L_MSG)})
ins_p["pad"] = _stream(NC_P, {R0 + L_MSG: [1] * N_T})
E_p, rec_p = sim_p.run(ins_p, NC_P, N_T, seed=7, record=("z", "tag", "tag_ok"))
# check: keystream and tag of two traces of every group against the RTL flow's model (IV = 1 | nonce | ctr 0 | chip nonce)
for i in (0, N_G - 1, N_G, 2 * N_G, N_T - 1):
    ks = R.grain_keystream(keys[i], R.iv_bits(nonces[i], 0, cns[i]), L_MSG + 2)
    z = [int(x) for x in sim_p.value("z", rec_p, i)[R0:R0 + L_MSG]]
    tg = int(sim_p.value("tag", rec_p, i)[R0 + L_MSG + 1])
    assert z == ks["enc"][:L_MSG] and tg == R.grain_tag(msgs[i], ks), f"gate-level trace {i} differs from the model"
# the S-DSM: key16 loaded in cycle 2 (fixed or random), a random modulator stream d per trace, the same number of clocks
k16_fix = rng_g.randrange(1, 1 << 16)
k16 = [k16_fix] * N_G + [rng_g.randrange(1, 1 << 16) for _ in range(2 * N_G)]
sim_s = PR.GateSim(str(RUN130["sdsm_lp"]["verilog"]), str(LIBERTY),
                   PR.net_caps(str(OPENROAD), PDK_ROOT, RUN130["sdsm_lp"], str(WORK / "pnr" / "sdsm_lp")))
ins_s = dict(sim_s.bits("seed1", 0x1234), **sim_s.bits("seed2", 0xBEEF))
for b, net in enumerate(sim_s.ports["key16"][1]):
    ins_s[net] = _stream(NC_P, {c: [(k >> (15 - b)) & 1 for k in k16] for c in range(NC_P)})
d_s = [[rng_g.getrandbits(1) for _ in range(N_T)] for _ in range(NC_P)]
ins_s["load"] = _stream(NC_P, {2: [1] * N_T})
ins_s["d"] = _stream(NC_P, {c: d_s[c] for c in range(NC_P)})
E_s, rec_s = sim_s.run(ins_s, NC_P, N_T, seed=7, record=("c",))
for i in (0, N_G, N_T - 1):                         # c = d XOR LFSR16(key16), from the first output bit on
    ks16 = R.LFSR16(k16[i]).bits(NC_P - 4)
    c_ = [int(x) for x in sim_s.value("c", rec_s, i)[3:NC_P - 1]]
    assert c_ == [d_s[3 + k][i] ^ ks16[k] for k in range(NC_P - 4)], f"gate-level S-DSM trace {i} differs from the model"
# noise: 4 changes of an average flip-flop output per clock, as the register-level test's 4 toggles
sig_g = 4 * 0.5 * sim_p.vdd ** 2 * sim_p.q_cap
rng_n = np.random.default_rng(2041)
noisy_g = lambda e: e + sig_g * rng_n.standard_normal(e.shape)
def tests(E):
    Ef, Ea, Eb = noisy_g(E[:, :N_G].T), noisy_g(E[:, N_G:2 * N_G].T), noisy_g(E[:, 2 * N_G:].T)
    return welch_t(Ef, Ea), welch_t(Ea, Eb)
t_p, t_p0 = tests(E_p)
t_s, t_s0 = tests(E_s)
PH = dict(key=(2, 130), nonce=(130, 193), init=(193, R0), run=(R0, NC_P))    # protocol phases of the proposal (clocks)
mx = lambda t, a, b: float(np.nanmax(np.abs(t[a:b])))
TVG = dict(p={k: mx(t_p, *v) for k, v in PH.items()}, p_n={k: int((np.abs(t_p[slice(*v)]) > TH_TV).sum()) for k, v in PH.items()},
           p_ctrl=mx(t_p0, 1, NC_P), s_max=mx(t_s, 3, NC_P), s_n=int((np.abs(t_s[3:]) > TH_TV).sum()), s_ctrl=mx(t_s0, 3, NC_P),
           n_clk=NC_P, e_p=float(E_p[1:].mean()), e_s=float(E_s[1:].mean()), sig=sig_g, cells=sim_p.n_cells, ffs=sim_p.n_ff,
           runtime=time.time() - t0_g)
card("Leakage test on the placed gate-level netlists: fixed key against random keys",
     [("proposal, low-power block (" + ("routed, extracted wires" if ROUTE_BLOCK else "placed, wires estimated on the placement") + ")",
       f"{sim_p.n_cells:,} logic cells, {sim_p.n_ff} flip-flops, {sim_p.n_icg} clock-gating cells; mean {1e12 * TVG['e_p']:.2f} pJ per clock "
       f"in its data nets; keystream and tag of five traces equal the model"),
      ("key shift-in, clocks 2–129", f"max |t| = <b>{TVG['p']['key']:.1f}</b> ({TVG['p_n']['key']} of 128 clocks above {TH_TV})"),
      ("session and chip nonce, clocks 130–192", f"max |t| = {TVG['p']['nonce']:.1f} ({TVG['p_n']['nonce']} of 63 above)"),
      (f"start and initialisation, clocks 193–{R0 - 1}", f"max |t| = <b>{TVG['p']['init']:.1f}</b> ({TVG['p_n']['init']} of {R0 - 193} above)"),
      (f"keystream and MAC, clocks {R0}–{NC_P - 1}", f"max |t| = <b>{TVG['p']['run']:.1f}</b> ({TVG['p_n']['run']} of {NC_P - R0} above)"),
      ("S-DSM, LFSR of [1] restarted from the key (placed)", f"max |t| = <b>{TVG['s_max']:.1f}</b>; {TVG['s_n']} of {NC_P - 3} clocks above {TH_TV}"),
      ("controls, random against random keys", f"max |t| = {TVG['p_ctrl']:.1f} (proposal), {TVG['s_ctrl']:.1f} (S-DSM)"),
      ("traces and noise", f"{N_G:,} per group, {NC_P} clocks each; noise {1e15 * sig_g:.0f} fJ rms per clock (4 changes of an average "
                           "flip-flop output)"),
      ("run time", f"{TVG['runtime']:.0f} s")], tone="info")
fig, (a1, a2) = plt.subplots(1, 2, figsize=(W2, 2.9), sharey=True)
clk_g = np.arange(NC_P)
for ax, tf, t0_, lab in ((a1, t_p, t_p0, "a"), (a2, t_s, t_s0, "b")):
    ax.plot(clk_g, np.abs(tf), color=COL["bad"], lw=1.0, label="fixed vs random key")
    ax.plot(clk_g, np.abs(t0_), color=COL["faint"], lw=0.8, label="random vs random (control)")
    ax.axhline(TH_TV, color=COL["ink"], ls="--", lw=0.8)
    ax.set_xlim(0, NC_P - 1); ax.set_xlabel("Clock cycle"); ax.set_yscale("log"); ax.set_ylim(0.01, 300)
for k, (a, b) in PH.items():                         # the protocol phases of the proposal: key and initialisation shaded
    if k in ("key", "init"):
        a1.axvspan(a, b, color=COL["faint"], alpha=0.12, lw=0)
for x, y, txt, ha in ((66, 120, "key shift-in", "center"), (161, 120, "nonces", "center"), (190, 50, "init.", "right"),
                      (238, 120, "keystream, MAC", "center")):
    a1.text(x, y, txt, ha=ha, fontsize=8, color=COL["ink"])
panel(a1, "a", "proposal, placed low-power block"); panel(a2, "b", "S-DSM [1], placed")
a1.set_ylabel("|t| (Welch)")
a2.legend(loc="upper right", fontsize=8)
fig.tight_layout(); plt.show()
````

**Key insights.**

* **The proposed design exceeds the threshold in two phases** (Fig. 71(a)): during the key shift-in, at 49 of its 128 clocks with $\lvert t\rvert$ up to 22.3, and in 3 of the 11 clocks of the start and initialisation, with $\lvert t\rvert$ up to 31.7, the highest value of the test. It stays below in the nonce loads (at most 4.2) and in the 68 clocks of keystream and MAC (at most 2.4). The S-DSM exceeds it at 15 of 269 clocks, with $\lvert t\rvert$ up to 20.5, at the updates of its key LFSR (Fig. 71(b)). Both controls stay below (3.2 and 3.1).
* Both predictions hold: at gate level the initialisation leak of the proposed design is compressed into the first updates of its 32-round form, and the key shift-in, which the register-level model does not contain, is its longest exposure. The proposed design thus exposes its key where the register-level model places it, at the start of every frame, and in one place more, the serial key transfer from the fuzzy extractor; both lie in the first clocks of a session or a frame, where a countermeasure would act (a masked or parallel key load, and masking of the first initialisation updates).
* The keystream and MAC phase, which carries the data, **shows no first-order leakage at 2,000 traces**, whereas **the S-DSM exposes its key at every LFSR update of every frame**. Since the model omits glitches and internal power, a chip leaks at least this much: the test locates the exposure, it does not give a security level.

This section costs the keystream block; the key generator that feeds it, sized by hand in Section VIII-F, **is still a specification**. Section XI turns that specification into hardware: it sizes the key generator, generates, verifies, synthesises and lays out its decoder with the flow of this section, and completes the area with the hash and the helper data.

# **XI. From specification to hardware: a key-generator compiler**
---
Section X costed the keystream block but not the key generator that feeds it. For a given reliability and secrecy specification of the key, this section determines which SRAM-PUF array and which error-correcting code meet it at every process corner and temperature, and what the resulting key-reconstruction hardware costs. The question matters because the proposed design stores no key: its security depends on the 128-bit key being regenerated correctly at every power-up on every chip, with a failure probability below $10^{-6}$ over the specified temperatures at every corner. **Sizing the key generator by hand cannot establish that claim.** The hand design fails at some corners: with 600 cells and a 25 °C enrolment, reliability needs more correction than the 192-bit secrecy budget allows (Section VII-E), and the working setting t = 45, the number of errors the BCH code corrects, fails at sf and at ff and fs above room temperature (Section VIII-F). Whether any code meets the specification at every corner, temperature and enrolment, given reproducible flips that a stronger code absorbs only by publishing more helper data than the cells contain secrecy, is a search result; the trade-off between more cells (an inner repetition code) and a smaller decoder (a smaller t) is settled only by synthesis; and a reliability figure must describe the decoder that is built, bit for bit, including when it reports failure. A compiler, a program that turns a specification into a verified and laid-out decoder, answers all four, again whenever the specification, the process or the cell changes: a *code-to-chip* flow in the literal sense.

The section uses the model's fuzzy extractor (Section V: a BCH code over $\mathrm{GF}(2^{10})$, the finite field of $2^{10}$ elements, whose syndromes are the helper data), the SRAM cell's corner models anchored to SPICE Monte Carlo (Section VII-F) and the two-temperature enrolment (Section VIII-F), the drawn cell (Section VII-D) and the flow of Section X. **The compiler runs in six steps, which the subsections follow**. Subsection A states the specification (step 1: key-failure probability per power-up, temperatures, corners, enrolment and secret bits) and searches a concatenated code at every corner and temperature (step 2), to find, for each of three specifications and every repetition factor, the smallest PUF array and the weakest BCH code that meet the reliability target within the secrecy budget of the hand design (Section VII-E). Subsection B turns the feasible designs into hardware by Verilog generation and synthesis, and selects the smallest total area (decoder plus array) per specification (step 3); A and B share one code cell, after which an explorer lets the reader run steps 1 to 3 on another specification, searching its design space and synthesising the selected design. Subsection C verifies the generated decoder against the model's fuzzy extractor (step 4): it shows that the generated Verilog reconstructs exactly what the model does, bit for bit and beyond the code's capacity too, and measures how often such a read is miscorrected with success reported; a mutation test then shows that this comparison detects a wrong decoder, not only that it accepts a correct one, and a proof covers, for every input rather than for the tested ones, the block on which every other step depends, the decoder's multiplier in $\mathrm{GF}(2^{10})$. Subsection D places and routes the key generator of specification D, on which the security claims rest, from its synthesised netlist to a GDS with the flow of Section X-C (step 5). Subsection E gives the area of everything the proposed design adds to the chip of [1], with the hash (step 6), each part marked as synthesised, drawn, estimated or not costed, and places the decoder among published ones. All steps run live except the SAT proof of step 4 and, in step 5, the detailed routing (with its routing-DRC and antenna counts) and the Magic DRC, which are shown from recorded runs of the same code; `SAT_PROOF = True` and `ROUTE_COMPILED = True` (977 s on two cores) run them live.

The compiler is the module `keygen.py`, written by the first cell and **independent of the notebook**: its inputs are the specification and the per-cell error probabilities, its Verilog is not tied to a cell library (synthesised in this work only onto `sky130_fd_sc_hd`), and codes beyond repetition with BCH over $\mathrm{GF}(2^{10})$ would need extensions.

> **Next cell.** Writes the compiler to `keygen.py`: 1. the design-space search; 2. the Verilog generator; 3. the Icarus test bench.

````python
%%writefile keygen.py
"""keygen.py - a compiler for the proposal's PUF key-reconstruction hardware: from a specification to verified Verilog.

    cfg = design_keygen(spec, cell_err)   # design-space search: repetition factor r, BCH length n and strength t
    files = emit_verilog(cfg, outdir)     # synthesisable Verilog: repetition voter + serial BCH(n, t) decoder
    out = simulate(cfg, outdir, tests, iverilog, vvp)   # Icarus simulation of the generated RTL on given test streams

The fuzzy extractor is the model's (Section V): the syndrome construction over GF(2^10) with the primitive polynomial
x^10 + x^3 + 1, the t odd syndromes of the fingerprint as helper data, Berlekamp-Massey and a Chien search. The compiler
adds an optional inner repetition code (concatenated construction): the cells form n groups of r; the enrolled value of a
group is its first cell, the helper data holds the other cells xor the first, and reconstruction takes the majority. The
BCH code then corrects the n group values. The hardware decodes with the inversionless Berlekamp-Massey algorithm, whose
error locator is a non-zero multiple of the model's, so the corrected bits and the success flag are identical.

Inputs: a specification dict and a per-cell error-probability model cell_err(corner, N, T) supplied by the notebook
(from the PUF cell's simulated or recorded statistics). Outputs: the chosen configuration, the Verilog file
keygen_rec.v, and the bits, success flags and cycle counts of a live Icarus simulation. Nothing here is recorded.
"""
import math, os, subprocess
import numpy as np

# GF(2^10) arithmetic tables. Field elements are 10-bit integers (bit k = coefficient of x^k); alpha = x = 2 is a
# primitive element of x^10 + x^3 + 1 (0x409), so GF_N = 1023 is the multiplicative order and the longest BCH length.
# EXP[i] = alpha^i, LOG[alpha^i] = i; EXP is stored twice over (length 2 GF_N) so that EXP[LOG[a] + LOG[b]] needs no
# modulo reduction.
M, PRIM = 10, 0x409                      # GF(2^10), x^10 + x^3 + 1 (as the model)
GF_N = (1 << M) - 1
EXP = [0] * (2 * GF_N); LOG = [0] * (GF_N + 1)
_x = 1
for _i in range(GF_N):                   # successive powers of alpha: multiply by x, reduce by PRIM on overflow of bit 10
    EXP[_i] = _x; LOG[_x] = _i
    _x <<= 1
    if _x & (1 << M):
        _x ^= PRIM
for _i in range(GF_N, 2 * GF_N):
    EXP[_i] = EXP[_i - GF_N]


# ------------------------------------------------------------------------------------------- 1. design-space search
def _dist(p):
    """Poisson-binomial distribution of the number of errors among independent events with probabilities p.
    Returns d with d[k] = P(exactly k events), k = 0..len(p), computed exactly by convolution, one event at a time."""
    d = np.zeros(len(p) + 1); d[0] = 1.0
    for q in p:
        # adding one event: k errors either were k already and this one is correct, or were k-1 and this one fails
        d[1:] = d[1:] * (1 - q) + d[:-1] * q
        d[0] *= 1 - q
    return d


def group_err(p, r):
    """Probability that the majority of each group of r consecutive cells is wrong (r odd); p: per-cell error probabilities."""
    if r == 1:
        return np.asarray(p, float)             # no repetition code: a group is one cell
    # one row per group of r consecutive cells (a trailing incomplete group is dropped)
    g = np.asarray(p, float)[: len(p) // r * r].reshape(-1, r)
    # per-group Poisson-binomial distribution of the number of wrong cells, vectorised over the groups (as _dist)
    d = np.zeros((g.shape[0], r + 1)); d[:, 0] = 1.0
    for c in range(r):
        q = g[:, c:c + 1]
        d[:, 1:] = d[:, 1:] * (1 - q) + d[:, :-1] * q
        d[:, 0] *= 1 - q[:, 0]
    # the majority vote is wrong when (r + 1) / 2 or more of the r cells are wrong
    return d[:, (r + 1) // 2:].sum(axis=1)


def secret_bits(n, r, t, bias):
    """Fingerprint bits that stay secret: min-entropy of n groups of r cells, less the (r-1) repetition and m = 10 BCH
    helper bits per group and per unit of t (for r = 1 this is the model's puf_residual).
    bias: probability of the more likely cell value (or of the less likely one; the larger of bias and 1 - bias is used).
    Returns floor(n (r h - (r - 1)) - 10 t) bits, h = -log2 max(bias, 1 - bias) the per-cell min-entropy."""
    h = -math.log2(max(bias, 1 - bias))              # min-entropy per cell (bits)
    # each group of r cells leaks r - 1 bits through its repetition helper data; the BCH helper data (t odd syndromes
    # of M bits) leaks at most M t bits of the n group values
    return int(math.floor(n * (r * h - (r - 1)) - M * t))


def evaluate(n, r, cell_err, corners, temps, t_max):
    """Worst-case (over corners and temperatures) probability of more than t group errors, for t = 0..t_max.
    A BCH(n, t) decoder fails exactly when more than t of the n group values are wrong, so entry t is the key-failure
    probability of a decoder of strength t. Returns an array of length t_max + 1."""
    worst = np.zeros(t_max + 1)
    for c in corners:
        for T in temps:
            # per-cell error probabilities of the n r cells -> per-group majority error -> distribution of the
            # number of erroneous groups (Poisson-binomial over the groups' error probabilities)
            d = _dist(group_err(cell_err(c, n * r, T), r))
            # reversed cumulative sum: entry k is P(at least k errors); from index 1 on this is P(more than t)
            tail = np.cumsum(d[::-1])[::-1][1: t_max + 2]          # P(more than t), summed exactly
            worst = np.maximum(worst, tail)
    return worst


def design_keygen(spec, cell_err, log=None):
    """The smallest configuration (fewest PUF cells; then the smallest t) meeting spec at every corner and temperature.
    spec: p_fail, temps (list, deg C), corners, secret (bits), bias, r (tuple of odd repetition factors), n_step, and keep:
    the fraction of the array that the enrolment keeps (1 for a single 25 C enrolment; the worst corner's fraction for a
    masked enrolment), which sets the physical array size.
    cell_err(corner, N, T): per-cell error probabilities of N enrolled (kept) cells at temperature T.
    Returns dict(choice=..., per_r={r: best or None})."""
    per_r = {}
    for r in spec.get("r", (1, 3, 5, 7)):
        # net secret bits per group: r cells of min-entropy each, less the r - 1 repetition helper bits. With a strong
        # bias the repetition code leaks more than the cells hold, and that r is infeasible.
        h = r * -math.log2(max(spec["bias"], 1 - spec["bias"])) - (r - 1)
        if h <= 0:
            per_r[r] = None; continue
        # smallest n that could hold the secret with t = 1 (one syndrome of M bits); at least 16 groups
        n = max(16, int(math.ceil((spec["secret"] + M) / h)))
        best, step = None, spec.get("n_step", 32)
        # coarse search: grow n in steps of n_step up to the field's maximum length GF_N = 1023; at each n take the
        # smallest t that meets both the failure probability and the secrecy requirement
        while n <= GF_N and best is None:
            # t is capped so that the M t helper bits stay below the n received bits
            t_max = min(spec.get("t_max", 120), n // M - 1)
            w = evaluate(n, r, cell_err, spec["corners"], spec["temps"], t_max)
            ok_t = [t for t in range(1, t_max + 1) if w[t] <= spec["p_fail"] and secret_bits(n, r, t, spec["bias"]) >= spec["secret"]]
            if ok_t:
                best = dict(r=r, n=n, t=ok_t[0], p_fail=float(w[ok_t[0]]), secret=secret_bits(n, r, ok_t[0], spec["bias"]))
            else:
                n += step
        # fine search: walk n back down in steps of 4 from the coarse solution and keep the last n that still meets
        # the specification (the first failing n stops the walk)
        if best and step > 4:                                # refine downwards in steps of 4
            for n2 in range(best["n"] - 4, best["n"] - step, -4):
                t_max = min(spec.get("t_max", 120), n2 // M - 1)
                w = evaluate(n2, r, cell_err, spec["corners"], spec["temps"], t_max)
                ok_t = [t for t in range(1, t_max + 1) if w[t] <= spec["p_fail"] and secret_bits(n2, r, t, spec["bias"]) >= spec["secret"]]
                if not ok_t:
                    break
                best = dict(r=r, n=n2, t=ok_t[0], p_fail=float(w[ok_t[0]]), secret=secret_bits(n2, r, ok_t[0], spec["bias"]))
        if best:
            # physical array size: n r kept cells, scaled up by the fraction the enrolment keeps
            best["cells"] = int(math.ceil(best["n"] * r / spec.get("keep", 1.0)))
        per_r[r] = best
        if log:
            log(f"r = {r}: " + (f"n = {best['n']}, t = {best['t']}, {best['cells']} cells" if best else "infeasible"))
    # across repetition factors: fewest physical cells first, then the weakest (smallest, cheapest) BCH decoder
    feas = [b for b in per_r.values() if b]
    choice = min(feas, key=lambda b: (b["cells"], b["t"])) if feas else None
    return dict(choice=choice, per_r=per_r, spec=spec)


# ------------------------------------------------------------------------------------------- 2. Verilog generation
def latency(cfg):
    """Clock cycles of one reconstruction (load, syndromes, Berlekamp-Massey, two Chien passes, output).
    Terms, in the order of the state machine of bch_rec: n r cell reads (S_LOADW, through the voter), M t helper bits
    (S_LOADH), t odd syndromes of n cycles each (S_SYN), t even syndromes by squaring (S_EVEN), 2t Berlekamp-Massey
    iterations of (t + 1) discrepancy plus (t + 1) update cycles (S_DISC, S_UPD), two Chien passes of n positions with a
    (t + 1)-cycle Horner evaluation each (S_CH1, S_CH2), n output cycles (S_OUT), and 8 cycles of control overhead.
    Cells discarded at enrolment (cmask = 0) are not counted."""
    n, t, r = cfg["n"], cfg["t"], cfg["r"]
    return n * r + M * t + t * n + t + 2 * t * (2 * (t + 1)) + 2 * n * (t + 1) + n + 8


def _w(x):
    """Bit width needed to hold the non-negative integer x (at least 1)."""
    return max(1, int(x).bit_length())


def emit_verilog(cfg, outdir):
    """Write the reconstruction unit for cfg = dict(r, n, t): returns the list of files (top module 'keygen_rec').
    The design is fully serial (one received bit, one helper bit or one GF(2^10) multiply-accumulate step per clock), so
    its area grows with t (syndrome and locator registers) and n (the word register) but not with the product n t."""
    os.makedirs(outdir, exist_ok=True)
    n, t, r = cfg["n"], cfg["t"], cfg["r"]
    a_inv = EXP[GF_N - 1]                       # alpha^-1 = alpha^1022: the Chien search steps beta = alpha^-pos
    # register widths: position counters (0..n), BCH loop indices (0..2t+1), voter counters (0..r)
    W_POS, W_T, W_R = _w(n), _w(2 * t + 1), _w(max(r, 1))
    # bch_rec: the serial BCH decoder (Verilog text; its own // comments document each state)
    bch = f"""// BCH({n}, t = {t}) reconstruction over GF(2^{M}), x^10 + x^3 + 1: generated by keygen.py. Serial: odd syndromes of the
// received word xor the helper data, the even ones by squaring, inversionless Berlekamp-Massey, then two Chien passes
// (count the roots; if they match the locator degree, correct). Bit-compatible with the model's reconstruct().
module bch_rec (
    input  wire clk, input wire rst, input wire start,
    input  wire din, input wire din_valid,        // received word, position 0 first
    input  wire hin, input wire hin_valid,        // helper data: syndromes S1, S3, ..., MSB first
    output reg  dout, output reg dout_valid,      // corrected word, position 0 first
    output reg  done, output reg ok);
  localparam N = {n}, T = {t}, M = {M};
  localparam [M-1:0] ALPHA = 10'd2, ALPHA2 = 10'd4, ALPHA_INV = 10'd{a_inv};
  function [M-1:0] gmul(input [M-1:0] a, input [M-1:0] b);
    integer q; reg [M-1:0] p, x;
    begin
      p = 0; x = a;
      for (q = 0; q < M; q = q + 1) begin
        if (b[q]) p = p ^ x;
        x = x[M-1] ? ((x << 1) ^ 10'h009) : (x << 1);
      end
      gmul = p;
    end
  endfunction
  localparam S_IDLE = 4'd0, S_LOADW = 4'd1, S_LOADH = 4'd2, S_SYN = 4'd3, S_EVEN = 4'd4, S_DISC = 4'd5, S_UPD = 4'd6,
             S_CH1 = 4'd7, S_CH2 = 4'd8, S_OUT = 4'd9, S_DONE = 4'd10;
  reg [3:0] st;
  reg [N-1:0] w;                                  // the word, rotated one position per clock
  reg [M-1:0] S [1:2*T];                          // syndromes
  reg [M-1:0] L [0:T];                            // error locator
  reg [M-1:0] B [0:T];                            // correction polynomial
  reg [M-1:0] acc, pw, as_, gam, del, v, beta, hsh;
  reg [{W_POS}:0] pos, cnt;
  reg [{W_T}:0] j, i, rr, k;
  reg cond;
  integer z;
  wire [M-1:0] m1 = (st == S_SYN) ? gmul(pw, as_) : (st == S_DISC) ? gmul(L[i], ((rr + 1 >= i) && (rr + 1 - i >= 1) && (rr + 1 - i <= 2*T)) ? S[rr + 1 - i] : {{M{{1'b0}}}})
                    : (st == S_UPD) ? gmul(gam, L[i]) : gmul(v, beta);
  wire [M-1:0] m2 = gmul(del, (i == 0) ? {{M{{1'b0}}}} : B[i - 1]);
  wire [M-1:0] sq = gmul(S[j], S[j]);
  always @(posedge clk) begin
    if (rst) begin
      st <= S_IDLE; done <= 0; ok <= 0; dout <= 0; dout_valid <= 0;
    end else begin
      dout_valid <= 0;
      case (st)
        S_IDLE: if (start) begin
          st <= S_LOADW; pos <= 0; j <= 0; hsh <= 0; done <= 0; ok <= 0;
        end
        S_LOADW: if (din_valid) begin
          w <= {{din, w[N-1:1]}}; pos <= pos + 1;
          if (pos == N - 1) begin st <= S_LOADH; pos <= 0; end
        end
        S_LOADH: if (hin_valid) begin
          hsh <= {{hsh[M-2:0], hin}}; pos <= pos + 1;
          if (pos == M - 1) begin
            S[2*j + 1] <= {{hsh[M-2:0], hin}}; pos <= 0; j <= j + 1;
            if (j == T - 1) begin st <= S_SYN; j <= 0; acc <= 0; pw <= 1; as_ <= ALPHA; end
          end
        end
        S_SYN: begin                              // S[2j+1] ^= sum_pos w[pos] alpha^(pos (2j+1))
          if (w[0]) acc <= acc ^ pw;
          pw <= m1; w <= {{w[0], w[N-1:1]}}; pos <= pos + 1;
          if (pos == N - 1) begin
            S[2*j + 1] <= S[2*j + 1] ^ acc ^ (w[0] ? pw : {{M{{1'b0}}}});
            pos <= 0; acc <= 0; pw <= 1; as_ <= gmul(as_, ALPHA2); j <= j + 1;
            if (j == T - 1) begin st <= S_EVEN; j <= 1; end
          end
        end
        S_EVEN: begin                             // S[2j] = S[j]^2
          S[2*j] <= sq; j <= j + 1;
          if (j == T) begin
            st <= S_DISC; rr <= 0; i <= 0; del <= 0; k <= 0; gam <= 1;
            for (z = 0; z <= T; z = z + 1) begin L[z] <= (z == 0); B[z] <= (z == 0); end
          end
        end
        S_DISC: begin                             // discrepancy: sum_i L[i] S[rr+1-i]
          del <= del ^ m1; i <= i + 1;
          if (i == T) begin st <= S_UPD; i <= T; cond <= 0; end
        end
        S_UPD: begin                              // L = gam L - del x B; B = cond ? L : x B (i from T down to 0)
          if (i == T) cond <= (del != 0) && (2 * k <= rr);
          L[i] <= m1 ^ m2;
          if (((i == T) ? ((del != 0) && (2 * k <= rr)) : cond)) B[i] <= L[i]; else B[i] <= (i == 0) ? {{M{{1'b0}}}} : B[i - 1];
          if (i == 0) begin
            if (cond || ((T == 0) && (del != 0) && (2 * k <= rr))) begin k <= rr + 1 - k; gam <= del; end
            if (rr == 2*T - 1) begin st <= S_CH1; pos <= 0; i <= T; cnt <= 0; beta <= 1; v <= L[T]; end
            else begin st <= S_DISC; rr <= rr + 1; i <= 0; del <= 0; end
          end else i <= i - 1;
        end
        S_CH1, S_CH2: begin                       // Horner: v = L(beta), beta = alpha^-pos
          if (i != 0) begin
            v <= m1 ^ L[i - 1]; i <= i - 1;
          end else begin
            if (st == S_CH1) cnt <= cnt + (v == 0);
            else w <= {{w[0] ^ (v == 0), w[N-1:1]}};
            beta <= gmul(beta, ALPHA_INV); i <= T; v <= L[T]; pos <= pos + 1;
            if (pos == N - 1) begin
              pos <= 0; beta <= 1;
              if (st == S_CH2) st <= S_OUT;
              else if ((k <= T) && (cnt + (v == 0) == k)) begin st <= S_CH2; ok <= 1; end
              else st <= S_OUT;
            end
          end
        end
        S_OUT: begin
          dout <= w[0]; dout_valid <= 1; w <= {{w[0], w[N-1:1]}}; pos <= pos + 1;
          if (pos == N - 1) st <= S_DONE;
        end
        S_DONE: begin done <= 1; st <= S_IDLE; end
        default: st <= S_IDLE;
      endcase
    end
  end
endmodule
"""
    # with a repetition code (r > 1) a majority voter sits in front of the decoder and turns r cell bits into one group
    # value; the top module gates cin_valid with cmask so that cells discarded at enrolment are skipped
    if r > 1:
        vote = f"""
// Repetition voter (r = {r}): the cells arrive with their helper bit (0 for the first cell of a group); the majority of
// cell xor helper over each group is the group's value.
module rep_vote (input wire clk, input wire rst, input wire cin, input wire rin, input wire cin_valid,
                 output reg gout, output reg gout_valid);
  localparam R = {r};
  reg [{W_R}:0] c, ones;
  always @(posedge clk) begin
    if (rst) begin c <= 0; ones <= 0; gout_valid <= 0; end
    else begin
      gout_valid <= 0;
      if (cin_valid) begin
        if (c == R - 1) begin
          gout <= (ones + (cin ^ rin)) > (R / 2); gout_valid <= 1; c <= 0; ones <= 0;
        end else begin c <= c + 1; ones <= ones + (cin ^ rin); end
      end
    end
  end
endmodule
"""
        top = """
module keygen_rec (input wire clk, input wire rst, input wire start, input wire cin, input wire rin, input wire cmask,
                   input wire cin_valid, input wire hin, input wire hin_valid, output wire dout, output wire dout_valid,
                   output wire done, output wire ok);
  wire g, gv;                                   // cmask = 0: a cell discarded at enrolment, skipped
  rep_vote u_vote (.clk(clk), .rst(rst), .cin(cin), .rin(rin), .cin_valid(cin_valid & cmask), .gout(g), .gout_valid(gv));
  bch_rec u_bch (.clk(clk), .rst(rst), .start(start), .din(g), .din_valid(gv), .hin(hin), .hin_valid(hin_valid),
                 .dout(dout), .dout_valid(dout_valid), .done(done), .ok(ok));
endmodule
"""
    else:
        # r = 1: no voter; each kept cell is one received bit of the BCH word (rin, the repetition helper, is unused)
        vote = ""
        top = """
module keygen_rec (input wire clk, input wire rst, input wire start, input wire cin, input wire rin, input wire cmask,
                   input wire cin_valid, input wire hin, input wire hin_valid, output wire dout, output wire dout_valid,
                   output wire done, output wire ok);
  bch_rec u_bch (.clk(clk), .rst(rst), .start(start), .din(cin), .din_valid(cin_valid & cmask), .hin(hin), .hin_valid(hin_valid),
                 .dout(dout), .dout_valid(dout_valid), .done(done), .ok(ok));
endmodule
"""
    path = os.path.join(outdir, "keygen_rec.v")
    with open(path, "w") as f:
        f.write(bch + vote + top)
    return [path]


# ------------------------------------------------------------------------------------------- 3. simulation
def _tb(cfg, ntests, nc):
    """Icarus testbench for keygen_rec: ntests reconstructions of nc array cells each, read from the .mem files written
    by simulate(). Per test it pulses start, streams the cells (with repetition helper and mask bits), then the M t BCH
    helper bits MSB first, and writes to out.txt the corrected bits, the ok flag and the cycles from the end of the helper
    stream to done. Clock period 100 ns (only the cycle count matters)."""
    n, t, r = cfg["n"], cfg["t"], cfg["r"]
    nh = M * t                                  # BCH helper bits: t odd syndromes of M = 10 bits
    return f"""`timescale 1ns/1ps
module tb;
  reg clk = 0, rst = 1, start = 0, cin = 0, rin = 0, cmask = 1, cin_valid = 0, hin = 0, hin_valid = 0;
  wire dout, dout_valid, done, ok;
  keygen_rec dut (.clk(clk), .rst(rst), .start(start), .cin(cin), .rin(rin), .cmask(cmask), .cin_valid(cin_valid), .hin(hin),
                  .hin_valid(hin_valid), .dout(dout), .dout_valid(dout_valid), .done(done), .ok(ok));
  always #50 clk = ~clk;
  reg [{nc - 1}:0] C [0:{ntests - 1}];
  reg [{nc - 1}:0] RH [0:{ntests - 1}];
  reg [{nc - 1}:0] MK [0:{ntests - 1}];
  reg [{nh - 1}:0] H [0:{ntests - 1}];
  integer k, q, fo, cyc;
  initial begin
    $readmemb("cells.mem", C); $readmemb("rhelp.mem", RH); $readmemb("mask.mem", MK); $readmemb("bhelp.mem", H);
    fo = $fopen("out.txt", "w");
    repeat (3) @(posedge clk); rst = 0;
    for (k = 0; k < {ntests}; k = k + 1) begin
      @(negedge clk) start = 1; @(negedge clk) start = 0;
      for (q = 0; q < {nc}; q = q + 1) begin
        @(negedge clk) begin cin = C[k][q]; rin = RH[k][q]; cmask = MK[k][q]; cin_valid = 1; end
      end
      @(negedge clk) cin_valid = 0;
      repeat (3) @(negedge clk);
      for (q = 0; q < {nh}; q = q + 1) begin
        @(negedge clk) begin hin = H[k][{nh - 1} - q]; hin_valid = 1; end
      end
      @(negedge clk) hin_valid = 0;
      cyc = 0;
      while (!done) begin @(posedge clk); cyc = cyc + 1; if (dout_valid) $fwrite(fo, "%b", dout); end
      $fwrite(fo, " %0d %0d\\n", ok, cyc);
    end
    $fclose(fo); $finish;
  end
endmodule
"""


def simulate(cfg, outdir, tests, iverilog, vvp, timeout=1800):
    """tests: list of dict(cells=0/1 array of the array's read cells, mask=0/1 array (1: kept at enrolment; default all 1),
    rhelp=0/1 array (0 at each group's first kept cell), bhelp=0/1 array of 10t BCH helper bits); all tests use arrays of
    the same length. Returns a list of dict(bits=n corrected group values, ok, cycles)."""
    files = emit_verilog(cfg, outdir)
    nc = len(tests[0]["cells"])
    with open(os.path.join(outdir, "tb.v"), "w") as f:
        f.write(_tb(cfg, len(tests), nc))
    # one binary line per test for $readmemb, which puts the first character at the MSB. The testbench streams the
    # per-cell vectors from bit 0 upwards, so they are written reversed (cell 0 at the LSB); the BCH helper is streamed
    # from the MSB down, so it is written in array order (first helper bit first). A missing key defaults to all ones
    # (used for the mask: every cell kept).
    for name, key in (("cells.mem", "cells"), ("rhelp.mem", "rhelp"), ("mask.mem", "mask"), ("bhelp.mem", "bhelp")):
        with open(os.path.join(outdir, name), "w") as f:
            for tst in tests:
                b = np.asarray(tst.get(key, np.ones(nc, np.uint8)), np.uint8)
                f.write("".join("1" if x else "0" for x in (b[::-1] if key != "bhelp" else b)) + "\n")
    # compile the generated RTL with the testbench and run it (live, Icarus Verilog)
    subprocess.run([iverilog, "-g2012", "-o", "sim.vvp", "tb.v"] + [os.path.abspath(p) for p in files], cwd=outdir,
                   check=True, capture_output=True, text=True, timeout=timeout)
    subprocess.run([vvp, "-n", "sim.vvp"], cwd=outdir, check=True, capture_output=True, text=True, timeout=timeout)
    # out.txt: per test "<n corrected bits, position 0 first> <ok> <cycles>"
    out = []
    for line in open(os.path.join(outdir, "out.txt")):
        bits, ok, cyc = line.split()
        out.append(dict(bits=np.array([int(c) for c in bits], np.uint8), ok=bool(int(ok)), cycles=int(cyc)))
    return out


# ------------------------------------------------------------------------------------------- 5. proofs of the field arithmetic
def gmul_modules(rtl):
    """Two Verilog modules over the field of the generated RTL `rtl`: gmul_gen, the decoder's own multiplier function
    copied verbatim into a wrapper, and gmul_ref, an independent reference (the carry-less product of the operands, then
    reduced by the field polynomial PRIM from the top bit down)."""
    import re
    fn = re.search(r"\n(\s*function \[M-1:0\] gmul\(.*?endfunction)", rtl, re.S)
    assert fn, "no gmul function in the generated RTL"
    m = int(re.search(r"localparam N = \d+, T = \d+, M = (\d+);", rtl).group(1))
    assert m == M, "the generated RTL is over another field"
    gen = (f"module gmul_gen (input [{M - 1}:0] a, input [{M - 1}:0] b, output [{M - 1}:0] p);\n"
           f"  localparam M = {M};\n{fn.group(1)}\n  assign p = gmul(a, b);\nendmodule\n")
    ref = (f"module gmul_ref (input [{M - 1}:0] a, input [{M - 1}:0] b, output [{M - 1}:0] p);\n"
           f"  localparam [{2 * M - 2}:0] PRIM = {2 * M - 1}'h{PRIM:x};\n"
           f"  reg [{2 * M - 2}:0] c; integer i, j;\n"
           f"  always @* begin\n    c = 0;\n"
           f"    for (i = 0; i < {M}; i = i + 1) for (j = 0; j < {M}; j = j + 1) c[i + j] = c[i + j] ^ (a[i] & b[j]);\n"
           f"    for (i = {2 * M - 2}; i >= {M}; i = i - 1) if (c[i]) c = c ^ (PRIM << (i - {M}));\n"
           f"  end\n  assign p = c[{M - 1}:0];\nendmodule\n")
    return gen, ref


def gmul_sat_proof(yosys, rtl, workdir):
    """SAT proof (Yosys `miter -equiv` and `sat -prove`, MiniSAT) that the decoder's multiplier equals the reference for
    every pair of operands. Returns dict(proved, cells, seconds, line): the solver's verdict line. Several minutes."""
    import re, time
    os.makedirs(workdir, exist_ok=True)
    gen, ref = gmul_modules(rtl)
    open(os.path.join(workdir, "gmul_gen.v"), "w").write(gen)
    open(os.path.join(workdir, "gmul_ref.v"), "w").write(ref)
    open(os.path.join(workdir, "gmul_sat.ys"), "w").write(
        "read_verilog gmul_gen.v gmul_ref.v\nproc; opt_clean\n"
        "miter -equiv -flatten -make_assert gmul_gen gmul_ref miter\nhierarchy -top miter\nflatten; opt\n"
        "sat -verify -prove-asserts miter\n")
    t0 = time.time()
    r = subprocess.run([yosys, "-q", "-l", "gmul_sat.log", "-s", "gmul_sat.ys"], cwd=workdir, capture_output=True, text=True)
    log = open(os.path.join(workdir, "gmul_sat.log")).read()
    line = next((l.strip() for l in log.splitlines() if "SAT proof finished" in l), "no verdict")
    cells = re.search(r"Imported (\d+) cells to SAT database", log)
    return dict(proved=r.returncode == 0 and "SUCCESS" in line, cells=int(cells.group(1)) if cells else 0,
                seconds=round(time.time() - t0, 1), line=line)
````

### A. Steps 1 and 2: specification and design-space search

*Concatenated code.* The compiler **searches a concatenated code**, as in the helper-data key extractor of [26]: the $N = n\,r$ enrolled cells form $n$ groups of $r$ cells ($r$ odd), each decided by majority (a repetition code whose helper data are the other $r - 1$ cells XOR the first), and a BCH code of length $n \le 1023$ over $\mathrm{GF}(2^{10})$ corrects up to $t$ wrong groups with $t$ syndromes of 10 bits each; $r = 1$ is the model's fuzzy extractor.

*Cell errors and key failure.* Cell $i$ of a chip, with mismatch $\delta_i$ (in units of the mismatch spread of the cell model, Section VIII-F) and enrolled value $w_i$, powers up to the other value at temperature $T$ with probability

$$p_i(T) = \Phi\big(-\delta_i / s(T)\big)\ \text{if}\ w_i = 1,\qquad p_i(T) = \Phi\big(\delta_i / s(T)\big)\ \text{if}\ w_i = 0,$$

with $\Phi$ the standard normal cumulative distribution function and $s(T)$ the effective noise-to-mismatch ratio of the corner, anchored to its SPICE Monte Carlo (Section VII-F; one fit for the 25 °C enrolment, another for the cells kept by the two-temperature enrolment). A group is wrong when more than $r/2$ of its cells are, with a probability $q_g$ that follows exactly from its cells' $p_i$, and the key fails, with probability $P_{\text{fail}}$, when more than $t$ groups are wrong:

$$P_{\text{fail}}(n, r, t) = \max_{\text{corner},\ T}\ \Pr\Big[\sum_g E_g > t\Big],$$

with $E_g$ independent Bernoulli($q_g$) events, summed exactly (Poisson-binomial). $s(T)$ is interpolated linearly between −40, 25 and 85 °C and the failure probability grows with $s$, so **the worst case over 0–85 °C lies at 0, 25 or 85 °C**, the three temperatures evaluated.

*Secrecy.* Each cell carries a min-entropy $h = -\log_2 \max(b, 1-b)$ ($b$: the probability of powering up to 1), taken at its 95 % lower bound from the 1,024-cell post-layout Monte Carlo (Section VII-E), and the helper data publish $r - 1$ bits per group and $10t$ syndrome bits, so the secret bits left must satisfy

$$H = \big\lfloor n\,\big(r\,h - (r - 1)\big) - 10\,t \big\rfloor \ \ge 192 ,$$

the 128-bit key plus the heuristic 64-bit leftover-hash margin of Section VII-E.

*Search.* For each $r \in \{1, 3, 5, 7\}$, $n$ starts at the smallest value that could hold 192 secret bits and grows in steps of 32 until some $t \le \lfloor n/10 \rfloor - 1$ meets both conditions, then is refined downwards in steps of 4; the smallest such $t$ is kept. The physical array is $\lceil n r / \kappa \rceil$ cells, with $\kappa$ the fraction of cells the enrolment keeps (1 for a single 25 °C enrolment; the worst corner's fraction for the two-temperature enrolment).

A majority of $r$ cells fails far less often than one cell, so the analysis above predicts that a larger $r$ allows a much smaller $t$, at the cost of $r$ times more cells and $r - 1$ published bits per group. Reproducible flips are different: at fs, 27.8 % of the bits are wrong at 85 °C, 25.6 % of them through reproducible flips (Section VII-F), so many that even a majority of seven cells errs often. **With a 25 °C enrolment no code should therefore meet the target at every corner**, whereas the two-temperature enrolment, which discards the flipping cells, should, with a weaker code (Section VIII-F). The search assumes, with the noise-only corner models of Section VII-F and unlike the two-component model of Section VIII-F, that the reproducible flips are represented by a larger random noise that reproduces each corner's simulated error rate, not by the same cells being wrong at every hot power-up; it further assumes that the cells err independently, that the per-cell probabilities are those of one simulated chip per corner (the model's chip seed) with one enrolment read, and that the corners cover process and temperature, not supply.

The search runs at the behavioural level, **on exact probabilities from the cell models above (no further sampling)**. The specifications differ only in the corners and the enrolment:

**Specification D, on which the security claims rest**, is not searched but compiled as Section VIII-F sizes it for the 99th-percentile chip of a population of 1,000 (BCH alone, $t = 17$ on 600 kept cells, 935 physical cells). The compiler evaluates its failure probability with the search's corner models (those of the −40 and 85 °C enrolment) and its secret at the lower bound of the min-entropy of the kept cells at the worst corner (0.758 bit at fs), stricter than the typical-corner bias used for A to C. All four specifications share the key-failure target, below $10^{-6}$ per power-up from 0 to 85 °C, the secrecy condition $H \ge 192$, and the cell bias $b$ of the post-layout Monte Carlo (Section VII-D).

### B. Step 3: Verilog generation, synthesis and selection

For a configuration $(r, n, t)$ **the compiler writes synthesisable Verilog (RTL)**: a serial repetition voter (`rep_vote`), which adds over each group of $r$ cells the cell XOR its helper bit and outputs the majority, and a serial BCH($n$, $t$) reconstruction unit (`bch_rec`) over $\mathrm{GF}(2^{10})$ with the model's primitive polynomial $x^{10} + x^3 + 1$ and a combinational shift-and-add multiplier, one operation per clock. It loads the $n$ group values and the $10t$ helper bits; computes the $t$ odd syndromes of the received word, one pass of $n$ clocks each, and adds the helper data to them; obtains the even syndromes by squaring; runs $2t$ iterations of the inversionless Berlekamp-Massey algorithm over the $t + 1$ coefficients; and evaluates the error locator by Horner's rule at every position in two Chien-search passes, the first counting the roots and the second correcting the word only if the count equals the locator's degree (otherwise `ok` = 0). The top level (`keygen_rec`) adds a mask input that skips the cells discarded at enrolment. One reconstruction takes $n r + 10t + t n + t + 4t(t + 1) + 2n(t + 1) + n + 8$ clocks (loading the cells and the helper data, syndromes, Berlekamp-Massey, two Chien passes, output), at 10 MHz.

Synthesis is the slow step, so an estimate of the total area, $A_{\text{est}}$, ranks the feasible designs first: $A_{\text{est}} = 50\ \mu\text{m}^2 \times F_{\text{est}} + N_{\text{cells}}\,A_{\text{cell}}$, with $F_{\text{est}} = n + 20\,t + 20\,(t + 1) + 50$ the decoder's flip-flops (the word, $2t$ syndromes and two locator polynomials of $t + 1$ coefficients of 10 bits, plus 50 for control), $N_{\text{cells}}$ the number of physical PUF cells, 50 µm² per flip-flop with its share of the logic (an assumption read from the synthesised designs) and $A_{\text{cell}}$ the SRAM cell drawn in Section VII-D. The two candidates with the smallest estimate per specification are synthesised with Yosys onto `sky130_fd_sc_hd` (as in Section X-B), and the smallest *synthesised* total area, decoder plus array, is selected; the others keep their estimate. The metrics are the columns of Table 29. Fig. 72 draws the trade-off that the selection resolves: panel (a) gives the PUF cells each specification needs at every repetition factor r, and panel (b) splits the total area of specifications A and C into decoder and PUF array, **so a larger r can be seen to shrink the decoder while it enlarges the array**.

> **Next cell.** 1. Builds each corner's per-cell error model. 2. Searches every r for A to C. 3. Ranks by the estimate. 4. Synthesises the two best of each, and D as sized. 5. Selects the smallest synthesised total. 6. Prints the table and the result panel and draws the design space. Run time: about 40 s on two cores.

````python
#@title Fig.: key-generator compiler
# Steps 1 to 3 of the compiler: three specifications searched and one (D, the key generator of Section VIII-F) compiled as sized, the design-space search with the corner models of Section
# VII-F, and the generation and synthesis of the most promising designs; the smallest total area (decoder + PUF array) wins
# Inputs: PUF_SYS / PUF_CELL (per-corner anchored cell models, Section F cells), SECRET_REQ and H_TT (model parameters),
# SRAMPUF, DEFAULTS, rng_for (dev/model/secsensor.py), R (rtl_flow), ps (puf_spice), YOSYS, LIBERTY, WORK.
# dev/rtl/keygen.py does the search (design_keygen), the cycle count (latency) and the Verilog generation (emit_verilog).
# Live: the search and the Yosys syntheses run here. Outputs for later cells: DSE, SYN, SPECS, BASE, CELL_UM2, cell_err_for.
import importlib, keygen as KG, puf_layout as pl
importlib.reload(KG)
t0 = time.time()
_ce = {}                     # cache: (corner, N, enrolment) -> (PUF instance, enrolled bits, s(T))
def cell_err_for(enrol):
    """Per-cell error probabilities of N enrolled (kept) cells at temperature T, from each corner's anchored cell model."""
    def f(corner, N, T):
        # One chip (DEFAULTS['chip_seed']) of N cells at this corner: mismatch m_i ~ N(mu, 1), noise scale s(T) from the
        # anchored model (s1: 25 C enrolment, s2: 25 + 85 C enrolment, kept cells); tc = 0 because s(T) is used directly.
        # The enrolled bits w are one read at 25 C.
        if (corner, N, enrol) not in _ce:
            q = PUF_SYS[corner]; s = q["s%d" % enrol]
            puf = SRAMPUF(DEFAULTS["chip_seed"], N, s(25), float(ndtr(q["mu"])), 0.0)
            _ce[corner, N, enrol] = (puf, puf.read(25, rng_for(DEFAULTS["chip_seed"] * 7919 + 1)), s)
        puf, w, s = _ce[corner, N, enrol]
        # P(cell reads != w at T): a cell enrolled as 1 flips when m_i + s(T) n < 0, i.e. Phi(-m_i / s(T)); enrolled 0: Phi(m_i / s(T))
        return np.where(w == 1, ndtr(-puf.mismatch / s(T)), ndtr(puf.mismatch / s(T)))
    return f
# KEEP: the smallest fraction of cells a masked (two-temperature) enrolment keeps over the corners; it sizes the array
ALL, KEEP = list(PUF_SYS), min(q["keep"] for q in PUF_SYS.values())
# Secrecy: the bounded budget of Section VII-E, the same as for the hand design: at least SECRET_REQ = 192 secret bits
# (128 + 2 log2(1/eps), eps = 2^-32, leftover-hash lemma) at the 95 % lower bound of the min-entropy per cell, given to the
# search as the equivalent bias max(P(1), 1 - P(1)) = 2^-h
BASE = dict(p_fail=1e-6, secret=SECRET_REQ, bias=2.0 ** -H_TT[0], temps=[0, 25, 85], r=(1, 3, 5, 7))   # s(T) is piecewise linear in T,
SPECS = {"A": ("typical corner, 25 °C enrolment", dict(BASE, corners=["tt"]), 1),                # so the worst case over 0-85 °C
         "B": ("all five corners, 25 °C enrolment", dict(BASE, corners=ALL), 1),                   # lies at 0, 25 or 85 °C
         "C": ("all five corners, enrolment at 25 and 85 °C", dict(BASE, corners=ALL, keep=KEEP), 2)}
CELL_UM2 = (lambda z: z["width_um"] * z["height_um"])(pl.cell_report(ps.CELL))       # the SRAM cell drawn in Section VII-D
# Synthesis is the slow step, so an estimate ranks the candidates first and only the two most promising of each specification
# are synthesised. The estimate counts the decoder's storage (the word, 2t syndromes and two locator polynomials of t + 1
# coefficients, 10 bits each) at about 50 µm² per flip-flop including its share of the logic, as the synthesised designs show.
EST_UM2_PER_FF = 50
# flip-flops: word n + 2t syndromes x 10 bits + two locators x (t + 1) x 10 bits + about 50 for control and counters
est_ffs = lambda b: b["n"] + 20 * b["t"] + 20 * (b["t"] + 1) + 50
DSE, SYN = {}, {}
for key, (label, spec, enrol) in SPECS.items():
    # per repetition factor r: the smallest feasible (n, t, cells) meeting P(key fails) < 1e-6 and the secret budget
    # at every corner and temperature of the specification, or None
    DSE[key] = KG.design_keygen(spec, cell_err_for(enrol))
    feas = [b for b in DSE[key]["per_r"].values() if b]
    for b in feas:
        # array area = physical cells x drawn SRAM cell; cycles = one reconstruction (KG.latency); est_um2 = ranking estimate
        b.update(arr_um2=b["cells"] * CELL_UM2, cycles=KG.latency(b), est_um2=EST_UM2_PER_FF * est_ffs(b) + b["cells"] * CELL_UM2)
    for b in sorted(feas, key=lambda b: b["est_um2"])[:2]:                  # the two most promising: emit the Verilog, synthesise
        tag = (b["r"], b["n"], b["t"])
        if tag not in SYN:
            wd = WORK / "keygen" / ("r%d_n%d_t%d" % tag)
            SYN[tag] = R.synthesize(YOSYS, "keygen_rec", KG.emit_verilog(b, str(wd)), str(LIBERTY), str(wd),
                                    netlist=str(wd / "keygen_rec_netlist.v"))
        s = SYN[tag]
        b.update(dec_ge=s["ge"], dec_um2=s["area_um2"], ffs=s["ffs"], total_um2=s["area_um2"] + b["arr_um2"])
    synth = [b for b in feas if "total_um2" in b]
    # the selected design: smallest decoder + array area among the synthesised candidates
    DSE[key]["best"] = min(synth, key=lambda b: b["total_um2"]) if synth else None

# D: the key generator that Section VIII-F sizes for the population of chips (enrolment at −40 and 85 °C, BCH alone on
# PUF_BITS kept cells, t = t_fin, n_cells_fin physical cells), compiled as specified there rather than searched: its failure
# probability with this search's corner models (enrolment 3: −40 + 85 °C) and its secret at the lower bound of the min-entropy
# of the kept cells at the worst corner (h_keep[k_h], Section VIII-F), stricter than the typical-corner bias of A to C
KEEP3 = min(q["keep3"] for q in PUF_SYS.values())
SPECS["D"] = ("all five corners, enrolment at −40 and 85 °C (key generator of Section VIII-F)",
              dict(BASE, corners=ALL, keep=KEEP3, r=(1,)), 3)
bD = dict(r=1, n=PUF_BITS, t=t_fin, cells=n_cells_fin, secret=int(np.floor(PUF_BITS * h_keep[k_h][0] - 10 * t_fin)))
bD.update(p_fail=float(KG.evaluate(bD["n"], 1, cell_err_for(3), ALL, BASE["temps"], t_fin)[t_fin]),
          arr_um2=bD["cells"] * CELL_UM2, cycles=KG.latency(bD))
tagD = (1, bD["n"], bD["t"]); wdD = WORK / "keygen" / ("r%d_n%d_t%d" % tagD)
SYN[tagD] = R.synthesize(YOSYS, "keygen_rec", KG.emit_verilog(bD, str(wdD)), str(LIBERTY), str(wdD), netlist=str(wdD / "keygen_rec_netlist.v"))
bD.update(dec_ge=SYN[tagD]["ge"], dec_um2=SYN[tagD]["area_um2"], ffs=SYN[tagD]["ffs"], total_um2=SYN[tagD]["area_um2"] + bD["arr_um2"])
DSE["D"] = dict(per_r={1: bD}, best=bD)

# Table: one row per (specification, r); the selected design in bold; latency = cycles at 10 MHz, in ms
rows_d = []
for key, (label, spec, enrol) in SPECS.items():
    for r, b in DSE[key]["per_r"].items():
        sel = b is not None and b is DSE[key]["best"]
        f = (lambda x: f"<b>{x}</b>") if sel else str
        syn = b is not None and "total_um2" in b
        rows_d.append((f"{key}. {label}" if r == spec["r"][0] else "", r) + (
            (f(b["n"]), f(b["t"]), f(b["cells"]), b["secret"], f"{b['p_fail']:.1e}",
             f"{b['dec_ge']:,.0f} / {b['ffs']:,}" if syn else "not synthesised",
             f(f"{b['total_um2'] / 1e3:.1f}") if syn else f"≈ {b['est_um2'] / 1e3:.0f} (estimate)", f"{b['cycles'] / 10e6 * 1e3:.1f}")
            if b else ("–", "–", "infeasible", "–", "–", "–", "–", "–")))
table(["Specification", "r", "Groups n", "t", "PUF cells", "Secret bits (95 % lower bound)", "P(key fails)", "Decoder (GE / FF)",
       "Total area (10<sup>3</sup> µm²)", "Latency (ms)"], rows_d,
      caption="Table 29: the compiler's design-space search for three specifications, and the key generator of Section VIII-F (D), compiled as sized there (key failure below 10<sup>−6</sup> from 0 to "
              f"85 °C, at least {SECRET_REQ} secret bits at the 95 % lower bound of the min-entropy): for each repetition factor r, the smallest array that meets the specification, "
              "its decoder synthesised on sky130_fd_sc_hd (the two most promising candidates of each specification, ranked by an "
              "estimate), and the total area with the SRAM cell of Section VII-D; bold: the design selected (smallest total area). PUF cells: the physical array, including the cells that a masked enrolment "
              "discards. Latency: one reconstruction at 10 MHz")
card("Key-generator compiler: design-space search and synthesis",
     [(f"{key}. {SPECS[key][0]}", (f"<b>r = {d['best']['r']}, {d['best']['n']} groups, t = {d['best']['t']}</b>: {d['best']['cells']} PUF cells, "
       f"decoder {d['best']['dec_ge']:,.0f} GE, {d['best']['cycles'] / 10e6 * 1e3:.1f} ms per reconstruction" if d["best"] else
       "<b>no design</b> on the grid (r ≤ 7, n in steps of 32 up to 1023) meets it")) for key, d in DSE.items()]
     + [("designs synthesised", f"{len(SYN)} (Yosys onto sky130_fd_sc_hd), {time.time() - t0:.0f} s in all")],
     tone="ok" if DSE["C"]["best"] and DSE["D"]["best"]["p_fail"] <= BASE["p_fail"] and DSE["D"]["best"]["secret"] >= SECRET_REQ else "bad")

# Figure: (a) array size per r and specification (three bars per r, x: no feasible design, star: selected);
# (b) for A and C, the total area stacked as decoder (solid; faded when it is the estimate) + PUF array (hatched)
fig, axs = plt.subplots(1, 2, figsize=(W2, 3.0))
rs = BASE["r"]; xs = np.arange(len(rs)); wdt = 0.26
STAR = dict(marker="*", ms=10, mfc=COL["yellow"], mec=COL["ink"], mew=0.6, ls="none")
for k, (key, c) in enumerate(zip(SPECS, (COL["ink"], COL["bad"], COL["ok"]))):
    v = [DSE[key]["per_r"][r]["cells"] if DSE[key]["per_r"][r] else 0 for r in rs]
    axs[0].bar(xs + (k - 1) * wdt, v, wdt, color=c, ec=COL["ink"], lw=0.4, label=key)
    for x, b in zip(xs, (DSE[key]["per_r"][r] for r in rs)):
        if not b:
            axs[0].text(x + (k - 1) * wdt, 20, "×", ha="center", va="bottom", color=c, fontsize=10, fontweight="bold")
        elif b is DSE[key]["best"]:                                     # the design the compiler selects
            axs[0].plot(x + (k - 1) * wdt, b["cells"] + 70, **STAR)
axs[0].plot([], [], label="selected", **STAR)
axs[0].set_xticks(xs, [f"$r$ = {r}" for r in rs]); axs[0].set_ylabel("PUF cells")
axs[0].set_ylim(0, 1.12 * max(b["cells"] for d in DSE.values() for b in d["per_r"].values() if b))
axs[0].legend(fontsize=7.5, loc="upper left", title="specification", title_fontsize=7.5)
axs[0].grid(axis="x", visible=False); panel(axs[0], "a", "array size (×: no design)")
for k, (key, c) in enumerate(zip(("A", "C"), (COL["ink"], COL["ok"]))):
    for x, r in zip(xs, rs):
        b = DSE[key]["per_r"][r]
        if not b:
            continue
        syn = "dec_um2" in b
        # decoder area: synthesised if available, else the estimate minus the array share
        dec = (b["dec_um2"] if syn else b["est_um2"] - b["arr_um2"]) / 1e3
        xb = x + (k - 0.5) * 0.36
        axs[1].bar(xb, dec, 0.34, color=c, ec=COL["ink"], lw=0.4, alpha=1 if syn else 0.4)
        axs[1].bar(xb, b["arr_um2"] / 1e3, 0.34, bottom=dec, color="white", ec=c, hatch="////", lw=0.6)
        if b is DSE[key]["best"]:
            axs[1].plot(xb, dec + b["arr_um2"] / 1e3 + 4, **STAR)
from matplotlib.patches import Patch
axs[1].legend(handles=[h for key, c in (("A", COL["ink"]), ("C", COL["ok"])) for h in
                       (Patch(fc=c, ec=COL["ink"], lw=0.4, label=f"{key}: decoder"),
                        Patch(fc="white", ec=c, hatch="////", lw=0.6, label=f"{key}: PUF array"))],
              fontsize=7, ncol=2, loc="upper right")
axs[1].set_xticks(xs, [f"$r$ = {r}" for r in rs]); axs[1].set_ylabel("Area (10$^3$ µm²)")
axs[1].set_ylim(0, 1.12 * axs[1].get_ylim()[1])
axs[1].grid(axis="x", visible=False); panel(axs[1], "b", "total area (faded decoder: estimate)")
fig.tight_layout(); plt.show()

def _idse():
    """Interactive companion: panels (a) and (b) above from DSE, hover with r, n, t, cells, decoder GE, secret bits and areas."""
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    f = make_subplots(1, 2, horizontal_spacing=0.1,
                      subplot_titles=("(a) Array size (×: no design)", "(b) Total area (faded decoder: estimate)"))
    def info(key, b):
        dec = f"{b['dec_ge']:,.0f} GE, {b['ffs']:,} FF (synthesised)" if "dec_ge" in b else "not synthesised"
        tot = (f"{b['total_um2'] / 1e3:.1f}" if "total_um2" in b else f"about {b.get('est_um2', 0) / 1e3:.0f} (estimate)")
        return (f"specification {key}, r = {b['r']}<br>n = {b['n']} groups, t = {b['t']}<br>PUF cells {b['cells']:,}"
                f"<br>secret bits {b['secret']}<br>P(key fails) {b['p_fail']:.1e}<br>decoder {dec}"
                f"<br>total area {tot} ×10³ µm²" + ("<br><b>selected</b>" if b is DSE[key]["best"] else ""))
    sx, sy, st = [], [], []
    for k, (key, c) in enumerate(zip(("A", "B", "C"), (COL["ink"], COL["bad"], COL["ok"]))):
        if key not in DSE:
            continue
        bs = [DSE[key]["per_r"].get(r) for r in rs]
        xk = [float(x + (k - 1) * wdt) for x in xs]
        f.add_trace(go.Bar(x=xk, y=[b["cells"] if b else 0 for b in bs], width=wdt, name=key, legendgroup=key, legendrank=2 + k,
                           marker=dict(color=c, line=dict(color=COL["ink"], width=0.5)),
                           hovertext=[info(key, b) if b else f"specification {key}, r = {r}: no feasible design" for b, r in zip(bs, rs)],
                           hovertemplate="%{hovertext}<extra></extra>"), 1, 1)
        nx = [x for x, b in zip(xk, bs) if not b]
        if nx:
            f.add_trace(go.Scatter(x=nx, y=[20] * len(nx), mode="text", text=["<b>×</b>"] * len(nx), textposition="top center",
                                   textfont=dict(size=round(10 * _PT), color=c),
                                   legendgroup=key, showlegend=False, hovertemplate=f"specification {key}: no feasible design<extra></extra>"), 1, 1)
        for x, b in zip(xk, bs):
            if b and b is DSE[key]["best"]:
                sx.append(x); sy.append(b["cells"]); st.append(info(key, b))
    f.add_trace(go.Scatter(x=sx, y=[y + 70 for y in sy], mode="markers", name="selected", legendrank=1, hovertext=st,
                           hovertemplate="%{hovertext}<extra></extra>",
                           marker=dict(symbol="star", size=9, color=COL["yellow"], line=dict(color=COL["ink"], width=0.6))), 1, 1)
    sx, sy, st = [], [], []
    for k, (key, c) in enumerate(zip(("A", "C"), (COL["ink"], COL["ok"]))):
        pts = [(float(x + (k - 0.5) * 0.36), DSE[key]["per_r"].get(r)) for x, r in zip(xs, rs)]
        pts = [(x, b) for x, b in pts if b]
        dec = [(b["dec_um2"] if "dec_um2" in b else b["est_um2"] - b["arr_um2"]) / 1e3 for _, b in pts]
        txt = [info(key, b) for _, b in pts]
        f.add_trace(go.Bar(x=[x for x, _ in pts], y=dec, width=0.34, name=f"{key}: decoder", legendgroup=f"{key}b", legend=f"legend{2 + k}",
                           marker=dict(color=c, opacity=[1 if "dec_um2" in b else 0.4 for _, b in pts], line=dict(color=COL["ink"], width=0.5)),
                           hovertext=txt, hovertemplate="decoder %{y:.1f} ×10³ µm²<br>%{hovertext}<extra></extra>"), 1, 2)
        f.add_trace(go.Bar(x=[x for x, _ in pts], y=[b["arr_um2"] / 1e3 for _, b in pts], base=dec, width=0.34,
                           name=f"{key}: PUF array", legendgroup=f"{key}b", legend=f"legend{2 + k}",
                           marker=dict(color="white", line=dict(color=c, width=0.8), pattern=dict(shape="/", fgcolor=c, size=5, solidity=0.25)),
                           hovertext=txt, hovertemplate="PUF array %{y:.1f} ×10³ µm²<br>%{hovertext}<extra></extra>"), 1, 2)
        for (x, b), d in zip(pts, dec):
            if b is DSE[key]["best"]:
                sx.append(x); sy.append(d + b["arr_um2"] / 1e3); st.append(info(key, b))
    f.add_trace(go.Scatter(x=sx, y=[y + 4 for y in sy], mode="markers", name="selected", showlegend=False, hovertext=st,
                           hovertemplate="%{hovertext}<extra></extra>",
                           marker=dict(symbol="star", size=9, color=COL["yellow"], line=dict(color=COL["ink"], width=0.6))), 1, 2)
    for c in (1, 2):
        f.update_xaxes(tickvals=[float(x) for x in xs], ticktext=[f"<i>r</i> = {r}" for r in rs], ticks="", row=1, col=c)
    top_a = 1.12 * max(b["cells"] for d in DSE.values() for b in d["per_r"].values() if b)
    top_b = 1.25 * max((b.get("dec_um2", b.get("est_um2", 0) - b["arr_um2"]) + b["arr_um2"]) / 1e3
                       for key in ("A", "C") for b in DSE[key]["per_r"].values() if b)
    f.update_yaxes(title_text="PUF cells", range=[0, top_a], row=1, col=1)
    f.update_yaxes(title_text="Area (10<sup>3</sup> µm<sup>2</sup>)", range=[0, top_b], row=1, col=2)
    leg = lambda **kw: dict(bgcolor="rgba(255,255,255,0.92)", borderwidth=0, font=dict(size=round(7 * _PT)), **kw)
    f.update_layout(barmode="overlay",
                    legend=leg(x=0.012, y=0.99, xanchor="left", yanchor="top", title=dict(text="specification", font=dict(size=round(7.5 * _PT)))),
                    legend2=leg(x=0.862, y=0.99, xanchor="right", yanchor="top"),
                    legend3=leg(x=0.995, y=0.99, xanchor="right", yanchor="top"))
    return f
iplot(_idse, "Fig.: key-generator design space (interactive)")
````

(a) PUF cells required for each repetition factor r, one bar per specification of Table 29 (×: no design meets it). **Key insights.**

* *Search (step 2; Table 29, Fig. 72(a)).* Specification A is feasible at every $r$: 611 cells at $r = 1$ ($n = 611$, $t = 36$, 194 secret bits at the lower bound, $P_\mathrm{fail} = 8.4 \times 10^{-7}$), 1,092 at $r = 3$ ($t = 7$), 2,075 at $r = 5$ ($t = 3$) and 3,990 at $r = 7$ ($t = 1$). Specification C is feasible at every $r$ as well: 636 cells at $r = 1$ ($n = 443$, $t = 21$), 1,395 at $r = 3$ ($n = 324$, $t = 4$), 2,690 at $r = 5$ ($t = 1$) and 5,723 at $r = 7$ ($t = 1$). As predicted, C needs a weaker code than A for $r \le 5$ although it covers five corners, but more cells at the same $r$, because its enrolment discards the cells that change between 25 and 85 °C. Compared with 128 secret bits at the point estimate of the min-entropy, the bounded budget enlarges the arrays (C at $r = 3$: 1,395 cells instead of 741; A at $r = 1$: 611 instead of 441): each group keeps only $r h - (r - 1)$ secret bits, small for $h$ near its lower bound, so repetition costs more cells.
* *Specification B* is infeasible, as predicted: no design on the searched grid meets it. The search reports only the worst case over the corners, not the failing corner; the error rates of Section VII-F point to fs. As no searched code meets B, **the two-temperature enrolment is what makes a specification over all corners achievable**.
* *Synthesis and selection (step 3; Table 29, Fig. 72(b)).* Four designs were synthesised: for A, $r = 3$ (9,980 GE, 807 flip-flops, $50.7 \times 10^3$ µm² in total) and $r = 5$ (8,567 GE, 696 flip-flops, $57.4 \times 10^3$ µm²); for C, $r = 3$ (8,984 GE, 707 flip-flops, $50.7 \times 10^3$ µm²) and $r = 5$ (7,467 GE, 572 flip-flops, $60.7 \times 10^3$ µm²). Both selected designs have $r = 3$: 364 groups and $t = 7$ for A (1.0 ms per reconstruction), 324 groups and $t = 4$ for C (0.6 ms). The candidates not synthesised are all estimated above the selected totals (about 113 and $82 \times 10^3$ µm² for A at $r = 1$ and 7, about 75 and $104 \times 10^3$ µm² for C), so the selection does not depend on the estimate's ranking. The selected decoders are 0.91–1.01× the proposed design's low-power keystream block (Section X) and 2.9–3.3× the bare Grain-128a + MAC core: the inner repetition code keeps them that small by lowering $t$, while repetition beyond $r = 3$ costs more in cells than it saves in decoder.
* *Specification D*, with the search's corner models, has a failure probability of $3.4 \times 10^{-10}$, far below the target (it is sized for the 99th-percentile chip, not for this one), with 284 secret bits at the worst corner's lower bound. Synthesised as sized, it takes 19,128 GE with 1,725 flip-flops, $83.2 \times 10^3$ µm² with its 935 cells, and 3.4 ms per reconstruction. It shows the other side of the trade-off: without repetition its code corrects 17 errors in a 600-bit word, and its decoder is about twice C's, 1.9× the keystream block, while its array is the smallest of the five-corner designs; it covers 0–85 °C with the margin of the population analysis, whereas C's enrolment covers 25–85 °C.

#### Explore: compile a key generator for another specification

The explorer runs steps 1 to 3 for a specification set by its menus: the corners (tt alone, or all five), the enrolment (one read at 25 °C; or masked, at 25 and 85 °C or at −40 and 85 °C, which sets the fraction of the array kept from the corner models of Section VII-F), the key-failure target and the secret, at the 95 % lower bound of the min-entropy as for A to C. The search is `design_keygen` of step 2, with the per-cell error models and the grid of the cell above. The area is the synthesised decoder where the cell above synthesised that design, else the estimate of step 3, plus the array of SRAM cells of Section VII-D. **Every combination of the menus was searched beforehand**, so a choice shows its figure and result panel at once, without running the cell. The explorer's view, Fig. 73, has the two panels of Fig. 72 for the chosen specification, so the effect of each control on the cells and the area at every r can be read directly.

````python
#@title Explore: key-generator compiler
# Explore: compile a key generator for another specification: the corners, the enrolment, the key-failure target and
# the secret. Each view runs the compiler's design-space search (KG.design_keygen with the corner models and the rules of
# specifications A to C) in compile_search below and draws, for each repetition factor r, the smallest design: its array
# and its area (the decoder synthesised in Section XI-B where that design was synthesised, else the estimate).
# Every setting of the menus was computed beforehand with these functions (dev/explore_precompute.py) and is stored in
# the figures repository, so the menus respond at once, without a kernel; the default view, specification C, is also
# embedded as the plot of record.
# enrolment options: label, the error model's enrolment index (cell_err_for), and the fraction of cells the masking keeps
EN_CK = {"25": ("one read at 25 °C", 1, 1.0), "25+85": ("25 and 85 °C, masked", 2, KEEP), "−40+85": ("−40 and 85 °C, masked", 3, KEEP3)}
def compile_search(corners, enrol, p_fail, secret):
    """The smallest design per r for this specification; areas from the synthesis where it exists (SYN), else estimated."""
    lab, e, keep = EN_CK[enrol]
    # the specification in the compiler's format, then its design-space search over r (and n, t for each r)
    spec = dict(BASE, corners=["tt"] if corners == "tt" else ALL, keep=keep, p_fail=p_fail, secret=secret)
    d = KG.design_keygen(spec, cell_err_for(e))
    # per feasible design: array area, reconstruction latency and the estimated total; the synthesised area if the cell
    # above (or the button) synthesised this exact (r, n, t)
    for b in (b for b in d["per_r"].values() if b):
        b.update(arr_um2=b["cells"] * CELL_UM2, cycles=KG.latency(b), est_um2=EST_UM2_PER_FF * est_ffs(b) + b["cells"] * CELL_UM2)
        if (b["r"], b["n"], b["t"]) in SYN:
            s = SYN[b["r"], b["n"], b["t"]]
            b.update(dec_ge=s["ge"], dec_um2=s["area_um2"], ffs=s["ffs"], total_um2=s["area_um2"] + b["arr_um2"])
    feas = [b for b in d["per_r"].values() if b]
    # the selected design: the smallest total area (synthesised where available, else estimated)
    d["best"] = min(feas, key=lambda b: b.get("total_um2", b["est_um2"])) if feas else None
    return d
def compile_view(d):
    """Draws (a) the array of each r's design and (b) its area, decoder (solid: synthesised, faded: estimate) and array
    (hatched); returns the card's rows."""
    rs = BASE["r"]; xs = np.arange(len(rs)); bs = [d["per_r"].get(r) for r in rs]
    STAR = dict(marker="*", ms=10, mfc=COL["yellow"], mec=COL["ink"], mew=0.6, ls="none")
    fig, axs = plt.subplots(1, 2, figsize=(W2, 2.9))
    top = max([b["cells"] for b in bs if b] or [1])
    for x, b in zip(xs, bs):
        if not b:
            axs[0].text(x, 0.02 * top, "×", ha="center", va="bottom", color=COL["bad"], fontsize=12, fontweight="bold"); continue
        axs[0].bar(x, b["cells"], 0.5, color=COL["ok"], ec=COL["ink"], lw=0.4)
        dec = (b["dec_um2"] if "dec_um2" in b else b["est_um2"] - b["arr_um2"]) / 1e3
        axs[1].bar(x, dec, 0.5, color=COL["ok"], ec=COL["ink"], lw=0.4, alpha=1 if "dec_um2" in b else 0.4)
        axs[1].bar(x, b["arr_um2"] / 1e3, 0.5, bottom=dec, color="white", ec=COL["ok"], hatch="////", lw=0.6)
        if b is d["best"]:
            axs[0].plot(x, b["cells"] + 0.06 * top, **STAR); axs[1].plot(x, dec + b["arr_um2"] / 1e3 + 0.05 * axs[1].get_ylim()[1], **STAR)
    from matplotlib.patches import Patch
    axs[1].legend(handles=[Patch(fc=COL["ok"], ec=COL["ink"], lw=0.4, label="decoder, synthesised"),
                           Patch(fc=COL["ok"], ec=COL["ink"], lw=0.4, alpha=0.4, label="decoder, estimate"),
                           Patch(fc="white", ec=COL["ok"], hatch="////", lw=0.6, label="PUF array")], fontsize=7, loc="upper right")
    for ax, yl in zip(axs, ("PUF cells", "Area (10$^3$ µm²)")):
        ax.set_xticks(xs, [f"$r$ = {r}" for r in rs]); ax.set_ylabel(yl); ax.grid(axis="x", visible=False)
        ax.set_ylim(0, 1.3 * ax.get_ylim()[1])
    panel(axs[0], "a", "Array of each design (×: none)"); panel(axs[1], "b", "Area of each design")
    fig.tight_layout(); show()
    rows = []
    for r, b in zip(rs, bs):
        if not b:
            rows.append((f"r = {r}", "no design on the grid (n in steps of 32 up to 1023) meets the specification")); continue
        area = (f"{b['total_um2'] / 1e3:.1f} ×10³ µm² (decoder {b['dec_ge']:,.0f} GE, synthesised)" if "total_um2" in b
                else f"≈ {b['est_um2'] / 1e3:.0f} ×10³ µm² (estimate)")
        txt = (f"n = {b['n']} groups, t = {b['t']}, {b['cells']:,} PUF cells, P(key fails) {b['p_fail']:.1e}, {b['secret']} secret bits; "
               f"{area}; {b['cycles'] / 10e6 * 1e3:.1f} ms per reconstruction at 10 MHz")
        rows.append((f"r = {r}" + (" (selected)" if b is d["best"] else ""), f"<b>{txt}</b>" if b is d["best"] else txt))
    return rows
explore("compiler", "Key-generator compiler explorer")
````

### C. Step 4: RTL verification of the generated decoder

*Equivalence.* The hardware's inversionless Berlekamp-Massey algorithm produces a non-zero multiple of the model's error locator, with the same roots, and both decoders accept a read exactly when the locator's degree is at most $t$ and equals the number of its roots among the $n$ positions of the shortened code. **The corrected bits and the success flag must therefore be identical in every test**, whatever the number of errors.

*Miscorrection beyond t.* A read with more than $t$ errors cannot be corrected, but the decoder may still find a locator of degree $L \le t$ with all $L$ roots inside the shortened code and **flag success with a wrong word**. Far beyond $t$, the syndrome behaves like a random element of the $2^{mt}$ possible ones ($m$ = 10, the bits per syndrome), and the decoder accepts it when it equals the syndrome of one of the $\sum_{j \le t}\binom{n}{j}$ correctable patterns:

$$P_{\text{mis}} \approx \frac{\sum_{j=0}^{t}\binom{n}{j}}{2^{mt}},$$

where $P_{\text{mis}}$ is the probability that such a read is miscorrected with success reported.

The analysis above predicts about 37 % for $t$ = 1 and $n$ = 375 (specification C at $r$ = 5: every single-error syndrome points to one of 1,023 positions, 375 of them in the code), about 4 × 10⁻⁴ for $t$ = 4 and $n$ = 324, about 10⁻⁷ for $t$ = 7 and $n$ = 364, and negligible for $t$ = 45 and $n$ = 600. A miscorrection does not change the key-failure probability of step 2, which counts every read beyond $t$ as a failure; it only means that **the wrong key is first detected by the receiver, as a MAC failure of the first frame**.

**The verification runs at RTL**: the generated `keygen_rec` is simulated with Icarus Verilog for the key generator D and the selected designs of specifications C and A, against the model's `reconstruct` (Section V). A generated test bench (100 ns clock) reads all tests from memory files; for each it pulses `start`, streams the physical cells with their repetition helper and mask bits, then the $10t$ BCH helper bits, and records the corrected word, the success flag `ok` and the cycles from the end of the helper data to `done`. The tests and the metrics are:

* *RTL tests:* per design, 12 directed and 60 random tests (seeds 2027 and 77), each with a random enrolment mask (the fraction the specification's enrolment keeps), a random enrolled word and its helper data, and a read in which $k$ groups are corrupted (the majority of their cells flipped): $k$ cycles through 0, 1, about $t/2$, $t$, $t + 1$ and $t + 3$ in the directed tests and is uniform in 0 to $t + 4$ in the random ones; other groups receive single-cell errors below the majority, which the vote must absorb.
* *Random error patterns:* 4,000 per BCH code on the model's decoder, which the RTL matches (thousands of reads in Icarus would cost minutes): a random enrolled word and an error pattern of weight $w$ at random positions, $w$ from 0 to $t$ and up to six weights beyond ($t + 1$, $t + 2$, $t + 3$, $t + 5$, $2t + 1$ and $2t + 5$; five for $t$ = 1, where $2t + 1 = t + 2$), equally many each, for the codes of D, of the selected designs of C and A, and of the hand design of Section VII-E.
* *Metrics:* identical reconstructions, corrections, flagged failures and miscorrections against the prediction, clock cycles.

> **Next cell.** 1. Generates the Verilog of D, C and A. 2. Builds the 72 test reads of each. 3. Simulates them with Icarus Verilog. 4. Compares with the model. 5. Runs the random error patterns. 6. Prints the interface, the table and the result panel.

````python
#@title Key-generator compiler: verification
# Step 4: the generated Verilog of three designs (D, C, A), simulated with Icarus against the model's fuzzy extractor
# (reconstruct of Section V) on the same cell streams: 12 directed tests (k = 0, 1, t/2, t, t+1, t+3 corrupted
# groups) and N_RTL random ones per design; then a Monte Carlo of random error patterns at the BCH level, on the model's
# decoder (which the RTL matches decision for decision), that counts miscorrections beyond t flagged as success
N_RTL, N_MC = 60, 4000                   # random RTL tests per design; random error patterns per BCH code
def kg_tests(cfg, keep, ks, seed=2027):
    """Random enrolments and reads; for each k in ks, k groups corrupted (majority of their cells flipped), plus single-cell
    errors in other groups, which the majority must absorb. Returns the streams and the model's expectation."""
    r, n, t = cfg["r"], cfg["n"], cfg["t"]; rg = np.random.default_rng(seed)
    nk = n * r; nphys = int(np.ceil(nk / keep)); tests, expect = [], []
    for kk in ks:
        # enrolment: the mask selects the nk kept cells among nphys physical cells; wk is the enrolled value of the kept
        # cells; the repetition code takes each group's first cell as its value g and stores the others XOR g as helper
        # data rh; the BCH syndrome h of g is the outer helper data
        mask = np.zeros(nphys, np.uint8); mask[rg.choice(nphys, nk, replace=False)] = 1
        wk = rg.integers(0, 2, nk).astype(np.uint8); g = wk[0::r].copy(); rh = wk ^ np.repeat(g, r)
        h = bch_syndrome(g, t); rd = wk.copy()
        # the field read rd: in kk randomly chosen groups a majority of the cells flips (the group value is wrong), in the
        # other groups, half of the time, a minority flips (the majority vote must absorb it)
        bad = rg.choice(n, min(kk, n), replace=False)
        for gi in range(n):
            cells_g = gi * r + rg.permutation(r)
            rd[cells_g[: (r + 1) // 2 if gi in bad else (r - 1) // 2 * int(rg.random() < 0.5)]] ^= 1
        # the physical array (unkept cells random), the helper data as the RTL receives it, and the model's answer: majority
        # vote per group, then BCH reconstruction; expect = (reconstructed word, success flag, equal to the enrolled word)
        cells = rg.integers(0, 2, nphys).astype(np.uint8); cells[mask == 1] = rd
        rhp = np.zeros(nphys, np.uint8); rhp[mask == 1] = rh
        gv = ((rd ^ rh).reshape(n, r).sum(axis=1) > r // 2).astype(np.uint8)
        res = reconstruct(gv, h, t)
        tests.append(dict(cells=cells, mask=mask, rhelp=rhp, bhelp=syn_to_bits(h)))
        expect.append((res["w"].astype(np.uint8), bool(res["ok"]), bool(np.array_equal(res["w"], g))))
    return tests, expect
t0, VER = time.time(), {}
# per design: 12 directed tests (k = 0, 1, t/2, t, t+1, t+3 corrupted groups, twice) and N_RTL random k; simulate the
# generated Verilog with Icarus and compare bit for bit and in the success flag with the model
for key in ("D", "C", "A"):
    cfg = DSE[key]["best"]; t = cfg["t"]
    keep = SPECS[key][1].get("keep", 1.0)
    ks = [[0, 1, max(1, t // 2), t, t + 1, t + 3][j % 6] for j in range(12)] + list(np.random.default_rng(77).integers(0, t + 5, N_RTL))
    tests, expect = kg_tests(cfg, keep, ks)
    out = KG.simulate(cfg, str(WORK / "keygen" / f"verify_{key}"), tests, str(PNR_BIN / "iverilog"), str(PNR_BIN / "vvp"))
    same = [np.array_equal(o["bits"], e[0]) and o["ok"] == e[1] for o, e in zip(out, expect)]
    VER[key] = dict(match=sum(same[:12]), n=12, fails=sum(not e[1] for e in expect[:12]), cycles=max(o["cycles"] for o in out),
                    rmatch=sum(same[12:]), rn=len(same) - 12, rbeyond=sum(k > t for k in ks[12:]),
                    rmis=sum(o["ok"] and not e[2] for o, e in zip(out[12:], expect[12:])))
t_rtl = time.time() - t0

# Monte Carlo at the BCH level: a random enrolled word, a random error pattern of weight w at uniformly random positions
def bch_mc(n, t, weights, n_mc, seed):
    rg = np.random.default_rng(seed); res = {w: [0, 0, 0] for w in weights}      # [correct, failure flagged, miscorrected]
    for j in range(n_mc):
        w = weights[j % len(weights)]
        g = rg.integers(0, 2, n).astype(np.uint8); h = bch_syndrome(g, t); rd = g.copy()
        rd[rg.choice(n, w, replace=False)] ^= 1
        o = reconstruct(rd, h, t)
        res[w][0 if (o["ok"] and np.array_equal(o["w"], g)) else 1 if not o["ok"] else 2] += 1
    return res
from math import comb, log2
def p_mis_theory(n, t, m=10):          # a random syndrome beyond t: decodable patterns of the shortened code / all syndromes
    return sum(comb(n, j) for j in range(t + 1)) / 2.0 ** (m * t)
t1 = time.time()
MC_CODES = [(f"D (r = {DSE['D']['best']['r']})", DSE["D"]["best"]["n"], DSE["D"]["best"]["t"]),
            (f"C (r = {DSE['C']['best']['r']})", DSE["C"]["best"]["n"], DSE["C"]["best"]["t"]),
            (f"A (r = {DSE['A']['best']['r']})", DSE["A"]["best"]["n"], DSE["A"]["best"]["t"]),
            ("hand design (Section VII-E)", PUF_BITS, DEFAULTS["bch_t"])]
MC = {}
for lab, n, t in MC_CODES:
    wts = list(range(0, t + 1)) + [t + 1, t + 2, t + 3, t + 5, 2 * t + 1, 2 * t + 5]
    wts = sorted(set(w for w in wts if w <= n))
    MC[lab] = (n, t, bch_mc(n, t, wts, N_MC, 4040 + n + t))
t_mc = time.time() - t1

# print the interface of the generated module, then tabulate the Monte Carlo: patterns of weight <= t must be corrected;
# beyond t the decoder either flags a failure or miscorrects to another codeword with the success flag set
src = open(WORK / "keygen" / "verify_D" / "keygen_rec.v").read().splitlines()
print(f"keygen_rec.v generated for specification D ({len(src)} lines of Verilog); its interface:")
print("\n".join(l for l in src if l.startswith("module") or l.lstrip().startswith(("input", "output")))[:900])
rows_mc = []
for lab, (n, t, res) in MC.items():
    le = [w for w in res if w <= t]; be = [w for w in res if w > t]
    ok_le = sum(res[w][0] for w in le); n_le = sum(sum(res[w]) for w in le)
    n_be = sum(sum(res[w]) for w in be); mis = sum(res[w][2] for w in be); det = sum(res[w][1] for w in be)
    rows_mc.append((lab, n, t, f"{ok_le:,}/{n_le:,}", f"{det:,}/{n_be:,}", f"<b>{mis:,}/{n_be:,}</b> ({100 * mis / n_be:.2g} %)" if mis else f"0/{n_be:,}",
                    f"{p_mis_theory(n, t):.1e}" if p_mis_theory(n, t) > 1e-30 else f"2<sup>{log2(p_mis_theory(n, t)):.0f}</sup>"))
table(["Code", "Length n", "t", "Weight ≤ t: corrected", "Weight > t: failure flagged", "Weight > t: miscorrected, success flagged",
       "Theory, random syndrome"], rows_mc,
      caption=f"Table 30: random error patterns against the BCH decoder ({N_MC:,} per code, weights 0 to t and up to six weights beyond t, "
              "equally many each; model decoder, which the generated RTL matches decision for decision). Theory: the fraction of "
              "syndromes that the shortened code decodes, Σ<sub>j≤t</sub> C(n, j) / 2<sup>10t</sup>")
card("Generated key-reconstruction hardware: verification against the model",
     [(f"specification {key}: r = {DSE[key]['best']['r']}, n = {DSE[key]['best']['n']}, t = {DSE[key]['best']['t']}",
       f"directed: <b>{v['match']}/{v['n']}</b> identical to the model, bit for bit and in the success flag ({v['fails']} beyond the "
       f"code's capacity, correctly reported as failures); random: <b>{v['rmatch']}/{v['rn']}</b> identical ({v['rbeyond']} with more "
       f"than t corrupted groups, {v['rmis']} of them miscorrected with the success flag set, as in the model); "
       f"longest {v['cycles']:,} clock cycles after the helper data")
      for key, v in VER.items()]
     + [("random error patterns, model decoder (table above)", "; ".join(f"{r[0]}: {r[5]} miscorrected beyond t" for r in rows_mc).replace("<b>", "").replace("</b>", "")),
        ("run time", f"Icarus Verilog {t_rtl:.0f} s ({len(VER) * (12 + N_RTL)} tests), Monte Carlo {t_mc:.0f} s")],
     tone="ok" if all(v["match"] == v["n"] and v["rmatch"] == v["rn"] for v in VER.values()) else "bad")
````

**Key insights.** The generated `keygen_rec.v` for specification D has 120 lines of Verilog; its interface is printed above.

* *Equivalence.* For each of the three designs, D ($r = 1$, $n = 600$, $t = 17$), C ($r = 3$, $n = 324$, $t = 4$) and A ($r = 3$, $n = 364$, $t = 7$), **12/12 directed and 60/60 random reconstructions are identical to the model, bit for bit and in the success flag**, as predicted. In each, the 4 directed tests beyond the code's capacity are correctly reported as failures, and none of the random reads with more than $t$ corrupted groups (9 for D, 23 for C, 20 for A) is miscorrected. The generated RTL thus makes the same decisions as the model on 216 reads (72 per design), so the key-failure probabilities of step 2 are taken to apply to it, under the cell models.
* *Latency.* The longest reconstructions take 33,643, 4,946 and 8,969 cycles after the helper data; without its loading terms ($n r + 10t$), the latency formula of Subsection B gives 33,649, 4,952 and 8,975 cycles, within 6 cycles of the simulation. The latencies of Table 29 include the loading, in which a masked enrolment streams the whole physical array instead of the $n r$ kept cells.
* *Miscorrection.* The random error patterns (Table 30) confirm the prediction. Every pattern of weight up to $t$ is corrected (3,004/3,004 for D, 2,000/2,000 for C, 2,288/2,288 for A, 3,542/3,542 for the hand design). Beyond $t$, none of 996 reads for D ($2.6 \times 10^{-19}$ predicted per read), none of 2,000 for C (4.2 × 10⁻⁴ predicted, about one expected), none of 1,712 for A (1.4 × 10⁻⁷ predicted) and none of 458 for the hand design ($t = 45$, $n = 600$) is miscorrected with the success flag set. With $t = 17$, 4 and 7 the success flag is therefore a useful failure detector (a retry of the power-up read on a failed flag would be sound); for $t = 1$ and $n = 375$ a random syndrome is decoded with probability 376/1024, so it would not be.

#### Mutation test of the generated decoder

Mutation testing plants single faults in the design (a changed constant, operator or condition) and counts the faults the tests detect; a fault they miss marks a gap in them. Two control mutants change only code that never executes (an unreachable default state, and a branch taken only for $t = 0$): they must survive, so a harness that reported every mutant as caught would fail them. The analysis above predicts that **every functional fault changes at least one corrected word or success flag** in the directed tests, which cover 0, 1, $t/2$, $t$ and $t + 1$ corrupted groups and $t + 3$ beyond the strength of the code.

The test runs at RTL, live. The Verilog of specification C, the smallest generated design (repetition 3, BCH(324, $t = 4$)), carries one fault per mutant: 19 faults in six blocks (GF($2^{10}$) arithmetic, syndromes, Berlekamp-Massey, the Chien search and the success rule, the data path, the repetition voter and the cell mask) and the two controls. Each mutant is compiled and simulated with Icarus Verilog on the 12 directed tests of Step 4, one mutant per core, with a time limit of 60 s. **A mutant is caught when its output differs from that of the unmutated RTL**, which equals the model on these tests, or when it does not finish.

> **Next cell.** 1. Plants each fault in a copy of the generated Verilog. 2. Simulates the mutants in parallel with Icarus. 3. Compares their corrected words and success flags with those of the unmutated RTL and prints the result panel.

````python
#@title Key-generator compiler: mutation test
# Step 4, mutation test of the generated decoder: single-site faults planted in the Verilog of specification C (the
# smallest generated design, with its repetition voter and cell mask), each mutant simulated with Icarus on the 12 directed
# tests of the verification above. A mutant counts as caught when any corrected word or success flag differs from the
# unmutated RTL, which equals the model on these tests (card above), or when it does not finish in the time limit. Two
# controls change only code that never executes and must survive: a harness that reported every mutant as caught would
# fail them. Inputs: VER, WORK, PNR_BIN and the files of the verification run of C. Outputs: MUT (one row per mutant).
import re, shutil, subprocess
from concurrent.futures import ThreadPoolExecutor
# (block, fault, original text, mutated text): each original text occurs exactly once in the generated keygen_rec.v
MUTANTS = [
    ("GF(2^10) arithmetic", "reduction polynomial x^10 + x^3 + x + 1", "((x << 1) ^ 10'h009)", "((x << 1) ^ 10'h00b)"),
    ("GF(2^10) arithmetic", "partial products ORed instead of XORed", "if (b[q]) p = p ^ x;", "if (b[q]) p = p | x;"),
    ("GF(2^10) arithmetic", "wrong inverse of alpha in the Chien search", "ALPHA_INV = 10'd516", "ALPHA_INV = 10'd517"),
    ("syndromes", "received bit inverted", "if (w[0]) acc <= acc ^ pw;", "if (!w[0]) acc <= acc ^ pw;"),
    ("syndromes", "power step alpha instead of alpha^2", "as_ <= gmul(as_, ALPHA2);", "as_ <= gmul(as_, ALPHA);"),
    ("syndromes", "even syndromes not squared", "S[2*j] <= sq;", "S[2*j] <= S[j];"),
    ("Berlekamp-Massey", "discrepancy not accumulated", "del <= del ^ m1; i <= i + 1;", "del <= m1; i <= i + 1;"),
    ("Berlekamp-Massey", "length condition 2k < r instead of 2k <= r",
     "if (i == T) cond <= (del != 0) && (2 * k <= rr);", "if (i == T) cond <= (del != 0) && (2 * k < rr);"),
    ("Berlekamp-Massey", "locator update without the correction term", "L[i] <= m1 ^ m2;", "L[i] <= m1;"),
    ("Berlekamp-Massey", "new length r - k instead of r + 1 - k", "k <= rr + 1 - k; gam <= del;", "k <= rr - k; gam <= del;"),
    ("Berlekamp-Massey", "gamma not updated", "k <= rr + 1 - k; gam <= del;", "k <= rr + 1 - k; gam <= 1;"),
    ("Chien search and success rule", "non-roots counted", "cnt <= cnt + (v == 0);", "cnt <= cnt + (v != 0);"),
    ("Chien search and success rule", "roots found but not corrected", "w <= {w[0] ^ (v == 0), w[N-1:1]};", "w <= {w[0], w[N-1:1]};"),
    ("Chien search and success rule", "success without the root count", "(k <= T) && (cnt + (v == 0) == k)", "(k <= T)"),
    ("data path", "output inverted", "dout <= w[0]; dout_valid <= 1;", "dout <= ~w[0]; dout_valid <= 1;"),
    ("data path", "word loaded one bit short", "if (pos == N - 1) begin st <= S_LOADH; pos <= 0; end",
     "if (pos == N - 2) begin st <= S_LOADH; pos <= 0; end"),
    ("repetition voter and cell mask", "majority threshold >= instead of >", "> (R / 2)", ">= (R / 2)"),
    ("repetition voter and cell mask", "repetition helper data ignored", "(ones + (cin ^ rin)) >", "(ones + cin) >"),
    ("repetition voter and cell mask", "cell mask ignored", ".cin_valid(cin_valid & cmask)", ".cin_valid(cin_valid)"),
    ("control", "unreachable default state", "default: st <= S_IDLE;", "default: st <= S_DONE;"),
    ("control", "branch taken only for t = 0", "(T == 0) && (del != 0)", "(T == 0) && (del == 0)"),
]
N_DIR, T_LIMIT = 12, 60                  # the directed tests of Step 4; time limit per mutant [s] (about 1 s each here)
t0 = time.time()
src_dir, mdir = WORK / "keygen" / "verify_C", WORK / "keygen" / "mutants"
rtl = (src_dir / "keygen_rec.v").read_text()
tb = re.sub(r"k < \d+;", f"k < {N_DIR};", (src_dir / "tb.v").read_text(), count=1)      # the first 12 tests: the directed ones
golden = (src_dir / "out.txt").read_text().splitlines()[:N_DIR]
assert VER["C"]["match"] == VER["C"]["n"] == N_DIR, "the unmutated RTL must equal the model on the directed tests"
def run_mutant(j, block, fault, old, new):
    """Compile and simulate one mutant; returns the number of directed tests whose output differs (-1: no completion)."""
    assert rtl.count(old) == 1, f"mutation site not unique in keygen_rec.v: {fault}"
    d = mdir / f"m{j:02d}"; d.mkdir(parents=True, exist_ok=True)
    for f in ("cells.mem", "rhelp.mem", "mask.mem", "bhelp.mem"):
        shutil.copy(src_dir / f, d)
    (d / "keygen_rec.v").write_text(rtl.replace(old, new)); (d / "tb.v").write_text(tb)
    (d / "out.txt").unlink(missing_ok=True)
    subprocess.run([str(PNR_BIN / "iverilog"), "-g2012", "-o", "sim.vvp", "tb.v", "keygen_rec.v"], cwd=d, check=True,
                   capture_output=True, text=True)                 # every mutant must compile: a failure stops the cell
    try:
        r = subprocess.run([str(PNR_BIN / "vvp"), "-n", "sim.vvp"], cwd=d, capture_output=True, timeout=T_LIMIT)
    except subprocess.TimeoutExpired:
        return -1                                                  # never finishes: caught
    if r.returncode != 0 or not (d / "out.txt").exists():
        return N_DIR                                               # the simulation fails: caught on every test
    out = (d / "out.txt").read_text().splitlines()
    return sum(a != b for a, b in zip(out, golden)) + abs(len(out) - len(golden))
with ThreadPoolExecutor(max(1, len(os.sched_getaffinity(0)))) as ex:     # one Icarus process per core
    diffs = list(ex.map(lambda a: run_mutant(*a), [(j, *m) for j, m in enumerate(MUTANTS)]))
MUT = [dict(block=m[0], fault=m[1], differ=d, caught=d != 0) for m, d in zip(MUTANTS, diffs)]
t_mut = time.time() - t0
func = [m for m in MUT if m["block"] != "control"]; ctrl = [m for m in MUT if m["block"] == "control"]
blocks = list(dict.fromkeys(m["block"] for m in func))
cfg_c = DSE["C"]["best"]
card(f"Mutation test of the generated decoder (specification C: repetition {cfg_c['r']}, BCH({cfg_c['n']}, t = {cfg_c['t']}); "
     f"{N_DIR} directed tests)",
     [(f"{b}", f"{sum(m['caught'] for m in func if m['block'] == b)} of {sum(m['block'] == b for m in func)} faults caught")
      for b in blocks] +
     [("all planted faults", f"<b>{sum(m['caught'] for m in func)} of {len(func)} caught</b>; each by "
                             f"{min(m['differ'] for m in func if m['differ'] > 0)} to {max(m['differ'] for m in func)} of "
                             f"{N_DIR} tests" + (f", {sum(m['differ'] < 0 for m in func)} by not finishing"
                                                 if any(m["differ"] < 0 for m in func) else "")),
      ("controls (code that never executes)", f"{sum(not m['caught'] for m in ctrl)} of {len(ctrl)} survive, as they must"),
      ("survivors among the faults", ", ".join(m["fault"] for m in func if not m["caught"]) or "none"),
      ("run time (Icarus, one mutant per core)", f"{t_mut:.0f} s")],
     tone="ok" if all(m["caught"] for m in func) and not any(m["caught"] for m in ctrl) else "bad")
````

**Key insights.** **All 19 faults are caught**, each by 2 to 12 of the 12 directed tests, and both controls survive, as predicted: the directed tests of Step 4 are sensitive to every block of the decoder, including the error-locator update, which a test with few errors would exercise only partly. With the bit-for-bit match of the 60 random tests and the Monte Carlo beyond $t$, the verification constrains the generated hardware, not only its successful cases. D and A, not mutated, share C's decoder template and are verified as in Step 4.

#### Proof of the generated field arithmetic

The syndromes, the discrepancies and locator updates of Berlekamp-Massey and every step of the Chien search are products in GF($2^{10}$). A multiplier wrong for a few operand pairs would corrupt only the reconstructions that meet those pairs, which random tests can miss. The multiplier has two 10-bit operands, so all $2^{20}$ = 1,048,576 pairs can be evaluated, and **comparing every product with a trusted reference is a proof by exhaustion**. The reference is the model's field (Section V-D), whose tables computed the helper data, so agreement ties the hardware to the helper data it has to decode. A SAT solver proves the same equivalence symbolically, against a second reference written independently. The analysis above predicts that no product differs.

The evaluation runs at RTL, live. For each of the specifications D, C and A, the function `gmul` of the generated Verilog (a shift-and-add multiplier with reduction by $x^{10} + x^3 + 1$, the same template in all three) is copied verbatim into a wrapper, converted to an and-inverter graph by Yosys and evaluated on all $2^{20}$ operand pairs at once, bit-parallel; each product is compared with the model's $\mathrm{EXP}[\mathrm{LOG}[a] + \mathrm{LOG}[b]]$ for operands $a$ and $b$, with EXP and LOG the model's antilogarithm and logarithm tables, and the decoder's constants, the field's primitive element $\alpha$, $\alpha^2$ and $\alpha^{-1}$ with the model's powers of $\alpha$. The SAT proof builds a miter of the same function and a reference written differently (the 19-bit carry-less product of the operands, then reduction from the top bit down) and **proves with MiniSAT, in Yosys, that their outputs never differ**; it takes about 5 minutes and is shown from its recorded run unless `SAT_PROOF = True`.

> **Next cell.** 1. Copies the multiplier of each generated decoder into a wrapper and converts it to an and-inverter graph. 2. Evaluates all $2^{20}$ operand pairs and compares every product and the field constants with the model. 3. Takes the SAT proof from its recorded run, or runs it with `SAT_PROOF = True`, and prints the result panel. *Recorded by default, so that the whole notebook stays within 20 minutes on free Colab; to repeat it live, tick `SAT_PROOF` in the form beside the code and run the cell (about 5 minutes); the result panel then says "this session" instead of "recorded run".*

````python
#@title Key-generator compiler: proof of the field arithmetic
# Step 4, proof of the generated field arithmetic. Every syndrome, discrepancy, locator update and Chien step of the
# generated decoder goes through one function, its GF(2^10) multiplier gmul. For each compiled specification, that
# function, copied verbatim from the generated Verilog into a wrapper, is synthesised to an and-inverter graph (Yosys) and
# evaluated on all 2^20 pairs of operands at once (bit-parallel); every product is compared with the model's GF(2^10)
# tables of Section V, which computed the helper data, and the decoder's field constants alpha, alpha^2 and
# alpha^-1 with the model's powers of alpha. A SAT proof that the same function equals an independent reference (the
# carry-less product reduced by x^10 + x^3 + 1) for every operand pair, Yosys `miter` and `sat` with MiniSAT, takes about
# 5 minutes and is shown from its recorded run unless SAT_PROOF = True. Inputs: R, KG, YOSYS, WORK, EXP, LOG. Outputs: MULP.
import re
SAT_PROOF = False                                                  #@param {type:"boolean"}
GMUL_SAT_REC = recorded("gmul_sat")
t0 = time.time()
K_ = np.arange(1 << KG.M, dtype=np.int64)
A_, B_ = np.repeat(K_, 1 << KG.M), np.tile(K_, 1 << KG.M)          # all 2^20 operand pairs
PROD = np.where((A_ == 0) | (B_ == 0), 0, EXP[LOG[A_] + LOG[B_]])   # the model's products
ONES = np.uint64(0xFFFFFFFFFFFFFFFF)
def eval_comb(sim, ins):
    """A combinational and-inverter graph (rtl_flow.AigSim) on many input vectors at once: ins {port: integer array};
    returns {port: integer array}. Bit j of word w of a signal is its value for vector 64 w + j."""
    n = len(next(iter(ins.values()))); W = (n + 63) // 64
    V = np.zeros((sim.M + 1, W), np.uint64)                         # V[0]: the constant 0 of the AIGER format
    for idx, (port, bit) in sim.in_map.items():
        b = np.zeros(64 * W, np.uint8); b[:n] = (ins[port] >> bit) & 1
        V[sim.inputs[idx]] = np.packbits(b.reshape(-1, 64), axis=1, bitorder="little").view("<u8").ravel()
    for o, a, an, b, bn in sim.levels:                              # one vector operation per level of AND gates
        V[o] = (V[a] ^ np.where(an, ONES, np.uint64(0))[:, None]) & (V[b] ^ np.where(bn, ONES, np.uint64(0))[:, None])
    out = {}
    for k, (port, bit) in sim.out_map.items():
        lit = int(sim.out_lit[k]); w = V[lit >> 1] ^ (ONES if lit & 1 else np.uint64(0))
        out[port] = out.get(port, 0) + (np.unpackbits(w.view(np.uint8), bitorder="little")[:n].astype(np.int64) << bit)
    return out
MULP = {}
for key in ("D", "C", "A"):
    rtl_k = (WORK / "keygen" / f"verify_{key}" / "keygen_rec.v").read_text()
    wd_k = WORK / "keygen" / f"proof_{key}"; wd_k.mkdir(parents=True, exist_ok=True)
    gen_k, _ = KG.gmul_modules(rtl_k)
    (wd_k / "gmul_gen.v").write_text(gen_k)
    aag, mp, _ = R.export_aig(YOSYS, str(wd_k), "gmul_gen", files=[str(wd_k / "gmul_gen.v")], stem="gmul_gen")
    sim_k = R.AigSim(aag, mp)
    p_k = eval_comb(sim_k, {"a": A_, "b": B_})["p"]
    consts = dict(re.findall(r"(ALPHA|ALPHA2|ALPHA_INV) = \d+'d(\d+)", rtl_k))
    MULP[key] = dict(pairs=len(A_), wrong=int(np.count_nonzero(p_k != PROD)), ands=sum(len(l[0]) for l in sim_k.levels),
                     consts=[int(consts["ALPHA"]) == EXP[1], int(consts["ALPHA2"]) == EXP[2], int(consts["ALPHA_INV"]) == EXP[KG.GF_N - 1]])
    assert MULP[key]["wrong"] == 0 and all(MULP[key]["consts"]), f"specification {key}: the generated field arithmetic differs from the model"
if SAT_PROOF:
    MULP["sat"] = dict(KG.gmul_sat_proof(YOSYS, (WORK / "keygen" / "verify_C" / "keygen_rec.v").read_text(), str(WORK / "keygen" / "proof_sat")),
                       src="this session")
else:
    MULP["sat"] = dict(GMUL_SAT_REC, src="recorded run (<code>python dev/rtl/record_gmul_sat.py</code>; <code>SAT_PROOF = True</code> re-runs it)")
assert MULP["sat"]["proved"], "SAT proof of the multiplier failed"
card("Proof of the generated decoder's field arithmetic",
     [(f"multiplier of specification {k}, all {MULP[k]['pairs']:,} operand pairs", f"<b>{MULP[k]['wrong']} products differ</b> from the model's "
       f"GF(2<sup>10</sup>) tables ({MULP[k]['ands']} AND gates); α, α², α<sup>−1</sup> equal the model's") for k in ("D", "C", "A")]
     + [("SAT proof against an independent reference (Yosys miter, MiniSAT)", f"<b>{'proved' if MULP['sat']['proved'] else 'NOT PROVED'}</b> for every "
         f"operand pair ({MULP['sat']['cells']} cells, {MULP['sat']['seconds']:.0f} s; {MULP['sat']['src']})"),
        ("run time", f"{time.time() - t0:.0f} s")], tone="ok")
````

**Key insights.** **No product differs** in any of the three specifications, as predicted: for all 1,048,576 operand pairs the multiplier, an and-inverter graph of 840 AND gates, gives the model's product, and the constants $\alpha$, $\alpha^2$ and $\alpha^{-1}$ equal the model's. The SAT proof finds no operand pair on which the generated multiplier and the independent reference differ (241 cells; recorded run, 266 s). The field arithmetic of the generated decoder is thus proven, not sampled: a fault of the decoder can lie only in its control (the state machine, the counters and the success rule), which the directed tests and the mutation test above exercise. The proof covers the multiplier as generated for this field; a change of the field or of the template needs it again, which the cell repeats in seconds.

````python
#@title Fig.: compiled key-reconstruction unit
# Step 5 for specification D (the key generator of Section VIII-F): from the synthesised netlist to a placed (by default) or routed and
# DRC-checked layout (the flow of Section X), so that the specification ends in GDS. Placement, clock tree, global
# routing and timing run live in seconds; the detailed routing of this ~38,600-instance design takes about 16 minutes on Colab's
# two cores, so by default its sign-off is shown from the recorded run of the same flow (dev/rtl/compiled_unit_routed.json)
ROUTE_COMPILED = False                  #@param {type:"boolean"}
KG_ROUTED = recorded("compiled_unit_routed")
t0 = time.time()
cfg = DSE["D"]["best"]; tag = "r%d_n%d_t%d" % (cfg["r"], cfg["n"], cfg["t"]); wd = WORK / "keygen" / tag
net_k = PR.prep(YOSYS, wd / "keygen_rec_netlist.v", LIBERTY, "keygen_rec", str(wd / "pnr"))
run_k = PR.place_and_route(str(OPENROAD), PDK_ROOT, net_k, "keygen_rec", str(wd / "pnr"), route=ROUTE_COMPILED,
                           antenna_margin=KG_ROUTED.get("antenna_margin", 20),   # the settings of the recorded run
                           buffer_inputs=KG_ROUTED.get("buffer_inputs", False))
GDS_K = PR.to_gds(PDK_ROOT, run_k["def"], str(wd / "pnr" / "keygen.gds"))
if ROUTE_COMPILED:
    drc_k = sl.drc(sl.tool_config(str(PDK_ROOT), magic=str(MAGIC), netgen=str(NETGEN), workdir=str(wd / "pnr")), GDS_K, "keygen_rec")
    sign = dict(route_drc=run_k["route_drc"], antenna_violations=run_k["antenna_violations"], magic_drc=drc_k["count"], src="this run")
else:
    same = KG_ROUTED["design"] == tag
    sign = dict(KG_ROUTED, src=(f"recorded run of the same design, {KG_ROUTED['total_seconds'] / 60:.0f} min on two cores; "
                                "<code>ROUTE_COMPILED = True</code> re-runs it" if same else
                                f"recorded run of {KG_ROUTED['design']} (this session selected {tag}): set <code>ROUTE_COMPILED = True</code>"))
ok_k = sign["magic_drc"] == 0 and sign["route_drc"] == 0 and sign["antenna_violations"] == 0
card("Compiled key-reconstruction unit (specification D) in SKY130: " + ("placed, routed and checked" if ROUTE_COMPILED else
     "placed and globally routed live; detailed routing and DRC recorded"),
     [("design", f"r = {cfg['r']}, {cfg['n']} groups, t = {cfg['t']}: {SYN[cfg['r'], cfg['n'], cfg['t']]['ge']:,.0f} GE, "
                 f"{SYN[cfg['r'], cfg['n'], cfg['t']]['ffs']:,} flip-flops"),
      ("die / standard cells", f"{run_k['die_um'][0]:.0f} × {run_k['die_um'][1]:.0f} µm, {run_k['instances']:,} instances"),
      ("timing at 10 MHz (tt, 25 °C, 1.8 V; global-routing parasitics)", f"setup slack <b>{run_k['slack_setup_ns']:+.1f} ns</b>, "
       f"hold slack <b>{run_k['slack_hold_ns']:+.2f} ns</b>"),
      (f"detailed routing ({sign['src']})", f"<b>{sign['route_drc']} DRC violations</b>, <b>{sign['antenna_violations']} antenna violations</b>"),
      ("DRC of the routed GDS (Magic, full SKY130 deck)", f"<b>{sign['magic_drc']} violations</b>"),
      ("GDS of this session (KLayout)", f"{'routed' if ROUTE_COMPILED else 'placed'} design, <code>{str(GDS_K).rsplit('/', 1)[-1]}</code> (<code>GDS_K</code>), drawn below"),
      ("run time", f"{time.time() - t0:.0f} s")], tone="ok" if ok_k else "bad")

# drawn like the keystream block's placement: standard cells and power stripes only (no signal routing), with a scale bar
V_K = kl_view(GDS_K); kl_size(V_K, 760, 760)
V_K.set_config("grid-show-ruler", "false")                                     # the figure draws its own 50 µm bar
for name, _, it in kl_layers(V_K):
    it.current().visible = name in {"nwell.drawing", "diff.drawing", "tap.drawing", "poly.drawing", "met4.drawing"}
img = plt.imread(io.BytesIO(kl_png(V_K)), format="png"); vb = V_K.box()          # vb: the visible box, in µm
fig, ax = plt.subplots(figsize=(W1, W1 + 0.15))
ax.imshow(img); ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
px_um, hgt = img.shape[1] / vb.width(), img.shape[0]
x0, y0 = 0.06 * img.shape[1], 0.95 * hgt
ax.add_patch(plt.Rectangle((x0 - 0.02 * hgt, y0 - 0.085 * hgt), 50 * px_um + 0.04 * hgt, 0.11 * hgt, color="white", alpha=0.9, lw=0))
ax.plot([x0, x0 + 50 * px_um], [y0, y0], color=COL["ink"], lw=2.5, solid_capstyle="butt")
ax.text(x0 + 25 * px_um, y0 - 0.02 * hgt, "50 µm", ha="center", va="bottom", fontsize=8.5, color=COL["ink"])
ax.set_title(("Routed" if ROUTE_COMPILED else "Placement") + f", {run_k['die_um'][0]:.0f} × {run_k['die_um'][1]:.0f} µm", fontsize=10)
fig.tight_layout(); plt.show()
````

### E. Step 6: the complete area

The key generator runs once per power-up: it reads the array, corrects the fingerprint and hashes the corrected word into the 128-bit key and the chip nonce (SHA-256 in the model, Section V), so the hash must be on the chip. A round-per-clock SHA-256 core holds the eight chaining words, the eight working words and a 16-word message schedule, 1,024 bits of state, so it needs more than 1,000 flip-flops (about 5,500 GE at 5.33 GE each) before its adders and round constants. The analysis above therefore predicts that **the hash and the decoder are the largest parts of the key generator**. The helper data of specification D are the enrolment mask (one bit per physical cell, 935) and the BCH syndromes ($10t$, 170): 1,105 bits (C: 1,395 mask, 648 repetition and 40 BCH bits, 2,083 in all). They are public and sent on the downlink at the start of a session (Section IV), so they need no storage if the decoder consumes them as they arrive.

The area of everything the proposed design adds to the chip of [1] is collected part by part, each part marked as synthesised, drawn, estimated or not costed:

* *Hash:* a SHA-256 compression core in Verilog (one round per clock, 65 clocks per 512-bit block, message words entering serially), simulated with Icarus Verilog on five messages of one and two blocks, among them a 516-bit message that, like D's 600-bit corrected word, needs two, compared with Python's `hashlib`, and synthesised on `sky130_fd_sc_hd`.
* *Decoder:* specification D from step 3. *Keystream block:* Table 22.
* *Array:* the physical cells of D times the area of the SRAM cell of Section VII-D, drawn with logic design rules, not as a foundry bit cell, **so it overstates the array**.
* *Helper data:* zero on the chip if streamed; otherwise estimated as flip-flops. The delay line takes its estimate from Section X-A.
* *Not costed:* the nonce sequencer (the array's power switching and the control of the 64 power-up reads hashed into the chip nonce), the array periphery (word-line drivers, read-out, power switching), the downlink receiver, the hash's padding and control, and the power and energy of the decoder and the hash.
* *Metrics:* area in GE and µm² per part, with its source; the total of the designed parts and each part's share.

> **Next cell.** 1. Writes the SHA-256 core. 2. Checks it against hashlib. 3. Synthesises it. 4. Collects the area of every part of D with its source and prints the table and the result panel.

````python
#@title Complete area of the proposal
# Step 6: the complete area of the proposal's digital security, specification C. The hash that turns the corrected fingerprint
# into the key (SHA-256 in the model) is written in RTL, checked against Python's hashlib and synthesised; the array area is
# its cells times the SRAM cell drawn in Section VII-D; the parts that are not designed are listed, not guessed
# Inputs: DSE, SYN, CELL_UM2 (compiler search cell), ge (cost cell: synthesised GE per design), RTL (nand2 area), R
# (rtl_flow), PNR_BIN (Icarus), YOSYS, LIBERTY, WORK. Live: the SHA-256 core is simulated with Icarus and synthesised with
# Yosys here. Outputs: the area table, a card, and AREA, S_SHA, n_help for later cells.
import hashlib, struct
SHA_V = r"""// SHA-256 compression, one round per clock: the 16 words of a block enter serially during rounds 0-15 (win, read at the
// round given by rnd); the message schedule is a 16-word shift register; init loads the initial hash value; after round 63
// the chaining value H is updated (feed-forward) and done pulses. Padding and block sequencing are left to the controller.
module sha256_core(input wire clk, input wire rst, input wire init, input wire start, input wire [31:0] win,
                   output reg done, output reg [6:0] rnd, output wire [255:0] digest);
  reg [31:0] H [0:7]; reg [31:0] a, b, c, d, e, f, g, h; reg [31:0] W [0:15]; reg busy; integer z;
  function [31:0] rotr(input [31:0] x, input [4:0] n); rotr = (x >> n) | (x << (32 - n)); endfunction
  function [31:0] K(input [5:0] i);
    case (i)
""" + "".join(f"      {i}: K = 32'h{k:08x};\n" for i, k in enumerate([
    0x428a2f98, 0x71374491, 0xb5c0fbcf, 0xe9b5dba5, 0x3956c25b, 0x59f111f1, 0x923f82a4, 0xab1c5ed5, 0xd807aa98, 0x12835b01,
    0x243185be, 0x550c7dc3, 0x72be5d74, 0x80deb1fe, 0x9bdc06a7, 0xc19bf174, 0xe49b69c1, 0xefbe4786, 0x0fc19dc6, 0x240ca1cc,
    0x2de92c6f, 0x4a7484aa, 0x5cb0a9dc, 0x76f988da, 0x983e5152, 0xa831c66d, 0xb00327c8, 0xbf597fc7, 0xc6e00bf3, 0xd5a79147,
    0x06ca6351, 0x14292967, 0x27b70a85, 0x2e1b2138, 0x4d2c6dfc, 0x53380d13, 0x650a7354, 0x766a0abb, 0x81c2c92e, 0x92722c85,
    0xa2bfe8a1, 0xa81a664b, 0xc24b8b70, 0xc76c51a3, 0xd192e819, 0xd6990624, 0xf40e3585, 0x106aa070, 0x19a4c116, 0x1e376c08,
    0x2748774c, 0x34b0bcb5, 0x391c0cb3, 0x4ed8aa4a, 0x5b9cca4f, 0x682e6ff3, 0x748f82ee, 0x78a5636f, 0x84c87814, 0x8cc70208,
    0x90befffa, 0xa4506ceb, 0xbef9a3f7, 0xc67178f2])) + r"""    endcase
  endfunction
  wire [31:0] s0 = rotr(W[1], 7) ^ rotr(W[1], 18) ^ (W[1] >> 3), s1 = rotr(W[14], 17) ^ rotr(W[14], 19) ^ (W[14] >> 10);
  wire [31:0] wt = (rnd < 16) ? win : W[0] + s0 + W[9] + s1;          // W[t] from W[t-16], W[t-15], W[t-7], W[t-2]
  wire [31:0] t1 = h + (rotr(e, 6) ^ rotr(e, 11) ^ rotr(e, 25)) + ((e & f) ^ (~e & g)) + K(rnd[5:0]) + wt;
  wire [31:0] t2 = (rotr(a, 2) ^ rotr(a, 13) ^ rotr(a, 22)) + ((a & b) ^ (a & c) ^ (b & c));
  assign digest = {H[0], H[1], H[2], H[3], H[4], H[5], H[6], H[7]};
  always @(posedge clk) begin
    if (rst) begin busy <= 0; done <= 0; rnd <= 0; end
    else begin
      done <= 0;
      if (init) begin
        H[0] <= 32'h6a09e667; H[1] <= 32'hbb67ae85; H[2] <= 32'h3c6ef372; H[3] <= 32'ha54ff53a;
        H[4] <= 32'h510e527f; H[5] <= 32'h9b05688c; H[6] <= 32'h1f83d9ab; H[7] <= 32'h5be0cd19;
      end else if (start && !busy) begin
        a <= H[0]; b <= H[1]; c <= H[2]; d <= H[3]; e <= H[4]; f <= H[5]; g <= H[6]; h <= H[7]; rnd <= 0; busy <= 1;
      end else if (busy) begin
        if (rnd == 64) begin
          H[0] <= H[0] + a; H[1] <= H[1] + b; H[2] <= H[2] + c; H[3] <= H[3] + d;
          H[4] <= H[4] + e; H[5] <= H[5] + f; H[6] <= H[6] + g; H[7] <= H[7] + h; busy <= 0; done <= 1;
        end else begin
          h <= g; g <= f; f <= e; e <= d + t1; d <= c; c <= b; b <= a; a <= t1 + t2;
          for (z = 0; z < 15; z = z + 1) W[z] <= W[z + 1];
          W[15] <= wt; rnd <= rnd + 1;
        end
      end
    end
  end
endmodule
"""
# Verification of the core against hashlib (FIPS 180-4 padding in Python, digests compared word for word).
t0 = time.time(); wd_h = WORK / "hash"; wd_h.mkdir(parents=True, exist_ok=True); (wd_h / "sha256_core.v").write_text(SHA_V)
# SHA-256 padding: 0x80, zeros up to 56 mod 64 bytes, then the 64-bit big-endian message length in bits
pad = lambda m: m + b"\x80" + b"\x00" * ((55 - len(m)) % 64) + struct.pack(">Q", 8 * len(m))
r_h = np.random.default_rng(256)
# test messages: the FIPS 'abc' vector, the empty message, 55 bytes (largest one-block message), 100 bytes (two blocks)
# and a 516-bit two-block message (65 bytes after packbits); C's corrected word, 324 bits, fits in one block
msgs = [b"abc", b"", r_h.bytes(55), r_h.bytes(100), np.packbits(r_h.integers(0, 2, 516).astype(np.uint8)).tobytes()]
blocks, tests = [], []
for m in msgs:
    # tests: (index of the first block, number of blocks, expected digest); blocks: all 64-byte blocks in order
    p = pad(m); tests.append((len(blocks), len(p) // 64, hashlib.sha256(m).hexdigest())); blocks += [p[i:i + 64] for i in range(0, len(p), 64)]
# blocks as 32-bit big-endian words, one hex word per line, for $readmemh
(wd_h / "blk.mem").write_text("".join("%08x\n" % w for b in blocks for w in struct.unpack(">16I", b)))
# Testbench: the core reads word rnd[3:0] of the current block during rounds 0-15; per message: init (load IV),
# then per block a start pulse and wait for done; the chained digest is written to out.txt after the last block
tb = (f"module tb; reg clk = 0, rst = 1, init = 0, start = 0; wire done; wire [6:0] rnd; wire [255:0] dg; reg [31:0] mem [0:{16 * len(blocks) - 1}];\n"
      "  integer blk, fo; wire [31:0] win = mem[blk * 16 + rnd[3:0]];\n"
      "  sha256_core u(.clk(clk), .rst(rst), .init(init), .start(start), .win(win), .done(done), .rnd(rnd), .digest(dg));\n"
      "  always #50 clk = ~clk;\n  initial begin $readmemh(\"blk.mem\", mem); fo = $fopen(\"out.txt\"); blk = 0; @(posedge clk); rst <= 0;\n")
for b0, nb_, _ in tests:
    tb += "    @(posedge clk) init <= 1; @(posedge clk) init <= 0;\n"
    tb += "".join(f"    blk = {b0 + j}; @(posedge clk) start <= 1; @(posedge clk) start <= 0; @(posedge done); @(posedge clk);\n" for j in range(nb_))
    tb += '    $fwrite(fo, "%h\\n", dg);\n'
(wd_h / "tb.v").write_text(tb + "    $finish;\n  end\nendmodule\n")
subprocess.run([str(PNR_BIN / "iverilog"), "-g2012", "-o", "sha.vvp", "tb.v", "sha256_core.v"], cwd=wd_h, check=True, capture_output=True)
subprocess.run([str(PNR_BIN / "vvp"), "-n", "sha.vvp"], cwd=wd_h, check=True, capture_output=True)
sha_ok = [o == e for o, (_, _, e) in zip((wd_h / "out.txt").read_text().split(), tests)]
# Yosys synthesis onto sky130_fd_sc_hd (tt, 25 C, 1.8 V); returns GE (NAND2 equivalents), flip-flops, area
S_SHA = R.synthesize(YOSYS, "sha256_core", [str(wd_h / "sha256_core.v")], str(LIBERTY), str(wd_h))
t_sha = time.time() - t0

# the complete area, specification D (the key generator of Section VIII-F, which the security claims use)
# ff_ge: area of one D flip-flop (dfxtp_1) in GE, for helper data held on chip
cfg = DSE["D"]["best"]; dec = SYN[cfg["r"], cfg["n"], cfg["t"]]; nand2 = RTL["nand2_um2"]
lib = R.liberty_cells(str(LIBERTY)); ff_ge = lib["sky130_fd_sc_hd__dfxtp_1"]["area"] / nand2
# helper data: the enrolment mask (one bit per physical cell), r - 1 repetition bits per group, 10 t BCH syndrome bits
# (GF(2^10))
help_bits = dict(mask=cfg["cells"], rep=cfg["n"] * (cfg["r"] - 1), bch=10 * cfg["t"]); n_help = sum(help_bits.values())
# (part, area in GE, source); the array converts cells x drawn cell area (um^2) into GE at the NAND2 rate; the MAC
# delay line is an estimate from the cost section
AREA = [("keystream block (Grain-128a + MAC, key, nonces, counter, IV, dither)", ge["proposal_lp"], "low-power form; synthesised, placed and routed (Section X)"),
        (f"key decoder (repetition vote, r = {cfg['r']}; BCH, n = {cfg['n']}, t = {cfg['t']})", dec["ge"], "synthesised, placed; routed in the recorded run"),
        ("hash (SHA-256 compression core)", S_SHA["ge"], "synthesised; padding and block control not included"),
        (f"SRAM-PUF array, {cfg['cells']} bit cells", cfg["cells"] * CELL_UM2 / nand2, f"drawn cell ({CELL_UM2:.2f} µm², Section VII-D) × cells; no periphery"),
        ("32-bit MAC delay line", 170, "estimate (Section X-A)")]
tot = sum(a for _, a, _ in AREA)
# table rows: GE and 10^3 um^2 (GE x NAND2 area / 1000)
rows_a = [(n_, f"{a:,.0f}", f"{a * nand2 / 1e3:.1f}", src) for n_, a, src in AREA]
rows_a.append(("<b>total, designed parts</b>", f"<b>{tot:,.0f}</b>", f"<b>{tot * nand2 / 1e3:.1f}</b>", ""))
rows_a += [(f"helper data, {n_help:,} bits ({help_bits['mask']} mask, {help_bits['rep']} repetition, {help_bits['bch']} BCH)",
            "0 or " + f"{n_help * ff_ge:,.0f}", "0 or " + f"{n_help * ff_ge * nand2 / 1e3:.1f}",
            f"none if streamed from the receiver at power-up; if held on chip, as flip-flops ({ff_ge:.2f} GE each), an estimate"),
           ("array periphery (word lines, read-out, power switching)", "–", "–", "not designed"),
           ("downlink receiver (session nonce and helper data)", "–", "–", "not designed")]
table(["Part", "Area (GE)", "Area (10<sup>3</sup> µm²)", "Source"], rows_a,
      caption="Table 32: complete area of the proposal's digital security for specification D (the key generator of Section VIII-F), in GE of sky130_fd_sc_hd "
              "(NAND2 = %.2f µm²; the array's area converted at the same rate)" % nand2, align=["left", "right", "right", "left"])
card("Complete area of the proposal (specification D)",
     [("SHA-256 core in RTL", f"<b>{sum(sha_ok)}/{len(sha_ok)}</b> digests equal to Python's hashlib (1 and 2 blocks, incl. a 516-bit "
                             f"two-block message); {S_SHA['ge']:,.0f} GE, {S_SHA['ffs']} flip-flops; 65 clocks per block ({t_sha:.0f} s)"),
      ("designed parts", f"<b>{tot:,.0f} GE</b> ({tot * nand2 / 1e3:.1f} × 10<sup>3</sup> µm²): keystream "
                         f"{100 * ge['proposal_lp'] / tot:.0f} %, decoder {100 * dec['ge'] / tot:.0f} %, hash {100 * S_SHA['ge'] / tot:.0f} %, "
                         f"array {100 * cfg['cells'] * CELL_UM2 / nand2 / tot:.0f} %"),
      ("helper data held on chip instead of streamed", f"+{n_help * ff_ge:,.0f} GE as flip-flops ({n_help:,} bits)"),
      ("not costed", "array periphery, downlink receiver, padding and control of the hash, and the power of the decoder and hash")],
     tone="ok" if all(sha_ok) else "bad")
````

**Key insights.**

* The SHA-256 core reproduces Python's `hashlib` on 5/5 messages and synthesises to 15,097 GE, about 5,700 GE of it in its 1,065 flip-flops. With it, the designed parts of the proposed design's digital security for specification D total **47,292 GE (177.5 × 10³ µm²)** (Table 32), about 4.8× the low-power keystream block and 54× the S-DSM's low-power block in the same flow: the low-power keystream block 21 %, the key decoder 40 %, the hash 32 % and the SRAM-PUF array (935 drawn cells) 6 %, with the 32-bit delay line estimated at 170 GE. As predicted, the decoder (19,128 GE) and the hash are the largest parts. Held on the chip as flip-flops instead of streamed, the 1,105 helper bits of D would add an estimated 5,893 GE.
* *Comparison with published decoders.* Published FPGA key generators have smaller decoders: in PUFKY [14], the BCH decoder is a microcoded processor of 112 Spartan-6 slices, and the whole key generator, ring-oscillator PUF, decoder and a SPONGENT-128 hash, occupies 1,162 slices, 82 % of them for the PUF. Slices and GE cannot be converted exactly, but a Spartan-6 slice holds eight flip-flops, so the 707 flip-flops of C's decoder alone would fill at least 89 slices, most of PUFKY's decoder, and the 1,725 of D at least 216.
* **Two choices of this work, not of the scheme, make the key generator large:** a round-per-clock SHA-256, where a serialised SHA-256 or a lightweight hash such as SPONGENT-128 [14] would be several times smaller, and a decoder that holds all its data in flip-flops, where a microcoded decoder with its word in a small memory, as in [14], would be smaller. Both blocks run only at power-up, so their area matters more than their power.

Every block of the proposed design now exists at the level at which it would be built, but each was checked on its own, and **every system result still rests on the behavioural model with SPICE-derived parameters**. Section XII joins them: the decoder generated here and the keystream block of Section X run at RTL, with transistor-level SRAM cells and comparators, in one simulation of the complete sensor.

# **XII. Full-chain mixed-level simulation of the complete sensor**
---
With the key generator compiled (Section XI), every block of the proposed design exists at the level at which it would be built. This section tests whether they work together: whether the complete sensor still works, and whether the security results of Section VIII still hold, when the circuits that set its analog behaviour are simulated at transistor level and the digital blocks run as RTL. Every system-level result so far comes from the bit-exact behavioural model of Section V, with circuit parameters from the SPICE simulations of Section VII. That model makes two abstractions: it replaces the StrongARM comparator, the modulator's 1-bit quantiser, by a threshold with a constant offset and Gaussian noise, and it reconstructs the key in Python, not in hardware. Both abstractions are tested where they could fail: inside the closed modulator loop, and in the hardware that regenerates the key from transistor-level simulations of the extracted SRAM cells. It supports the abstract's claim that the full chain is validated by a mixed-level simulation in SKY130 with open-source tools only (cells drawn with KLayout and extracted with Magic, Section VII-D; ngspice, Icarus Verilog and Python).

The section takes from earlier sections the extracted 6T SRAM cell and StrongARM comparator, their PVT data and the BCH correction capability t (Section VII); the replica attack with its behavioural prediction (Section VIII-B) and the SRAM-PUF error law (Section VII-F); the keystream block's RTL (Section X); and the compiler's BCH decoder (Section XI). With them, **one simulation of the complete sensor runs each block at the level at which its behaviour matters**, the blocks exchanging the real signals:

Two chips are simulated, the victim and an attacker's replica, each with its own transistor-level comparator (own Monte Carlo mismatch and transient noise), so that the replica attack of Section VIII-B **runs on transistor-level bits against both designs**.

The theory and the simulation setup below explain why the long transient can be cut into segments without error, how accurately the solver must solve the behavioural integrators and how the comparator model is tested decision by decision, and list the run's parameters. A fixes the operating condition of the run and looks up the comparator noise $\sigma_\mathrm{n}$ that the decision test of step 3 needs at that condition. B (step 1, transistor level) produces the chip's fingerprint from transistor-level SRAM cells, at enrolment and in the field, and counts the cells that power up differently: the errors that step 2 must correct. C (step 2, RTL) reconstructs the key from the field read of step 1 in the hardware that would be built and checks it against the key the server stored at enrolment. D (step 3, transistor level in the loop) runs the victim's and the replica's second-order modulators with each chip's post-layout StrongARM inside the loop and tests whether the behavioural comparator model of Section VIII predicts every decision the circuit makes. E (step 4, RTL) encrypts and authenticates the victim's transistor-level bits of step 3 in the RTL of the proposed design's keystream block, keyed by the key of step 2, and checks the frame against the model. F (step 5, behavioural) receives the frame of step 4 as the server does and attacks it three ways: a forger who changes the reading, an eavesdropper with a wrong key, and the replica attacker of Section VIII-B with its own chip's transistor-level bits. G (step 6, transistor level, recorded) runs the replica attack with transistor-level chips over a meaningful capture and compares its statistics with the behavioural model **on the same pair of chips**, so that any difference can be attributed to the comparator abstraction and not to different chips, seeds or settings. H shows the run in signal-chain order and tests the security of the frame it actually produces (transistor-level bits, RTL ciphertext and tag), and I collects, block by block, the level, simulator, run time and pass check of every step in one table, Table 35.

#### Theory: segments, solver accuracy and the decision-by-decision test

Symbols: $x_1, x_2$ are the two integrator outputs and $u$ the modulator input, in model units (full scale ±1); $v = \pm 1$ is the DAC output; $d$ is the dither; $\Delta V_k = (V_{DD}/4)\,x_2'$ is the comparator's differential input at clock $k$, with $V_{DD}$ the supply, with the dithered state $x_2' = x_2(1+d)+d$; $V_{os}$ is the comparator's input-referred offset in volts (in the model units of Section VIII, the offset $o = -V_{os}/(V_{DD}/4)$) and $\sigma_n$ its input-referred rms noise.

*1. Segments without error.* The modulator's only state is $(x_1, x_2)$, and the StrongARM is reset every clock and keeps no state beyond it. **A long transient can therefore be cut into segments**, each started from the previous integrator state (an `.ic` statement) with one warm-up decision to bring the comparator into its normal cycle; the cut is exact up to the comparator's reset, which the circuit performs anyway, and keeps ngspice's cost per clock constant. *Assumption:* the reset is complete; an incomplete one would show up as memory in the decision test of point 3.

*2. The behavioural integrators must be solved exactly.* ngspice solves every node to a relative tolerance `reltol`, $10^{-3}$ by default. During development (numbers recorded in the docstring of `mixedsim.py`) the default left effective integrator gains of 0.4993 and 0.4972 instead of 0.5 (errors of 0.1 to 0.6 %) and per-clock errors of $10^{-3}$ to $10^{-2}$: far above the integrators' thermal noise ($0.9 \times 10^{-5}$ of full scale per clock at the first integrator's output, Section V-A), and different from chip to chip, so two chips' loops would drift apart for a reason no real chip has. With `reltol` = $10^{-6}$ the gains are exact to $10^{-6}$ and the per-clock error, $8 \times 10^{-6}$, is below the thermal noise, at about twice the simulation time. The analysis above therefore predicts that **the transistor-level loop decodes to the same reading as the behavioural model of the same chip**.

*3. Omissions of the behavioural comparator and their test.* The model of Section VIII decides
$$v_k = \operatorname{sgn}(\Delta V_k - V_{os} + n_k), \qquad n_k \sim \mathcal{N}(0, \sigma_n^2),$$
where $v_k$ is the decision at clock $k$ and $n_k$ the comparator noise at that clock.
A real comparator can deviate in three ways: (i) an offset that depends on the input history (memory from an incomplete reset, kickback); (ii) metastability: the regeneration time grows as $\tau \ln(V_{DD}/|\Delta V|)$, with $\tau$ the regeneration time constant, so an input below about $V_{DD}\,e^{-t_{avail}/\tau}$ ($t_{avail}$: the time before the latch samples) may not resolve; (iii) non-Gaussian noise. **The decisive test is decision by decision**: given the input $\Delta V_k$ the transistor-level comparator actually saw, the threshold model predicts $[\Delta V_k > V_{os}]$ and can be wrong only through noise, with probability $\Phi(-|\Delta V_k - V_{os}|/\sigma_n)$ ($\Phi$: the standard normal cumulative distribution). The expected number of disagreements is
$$\mathbb{E}[N_{\neq}] = \sum_k \Phi\!\left(-\frac{|\Delta V_k - V_{os}|}{\sigma_n}\right).$$
The analysis above predicts that the observed count $N_{\neq}$ is consistent with $\mathbb{E}[N_{\neq}]$ (approximately Poisson) and that every disagreement lies within a few $\sigma_n$ of the threshold; more disagreements, or disagreements at large inputs, would expose an effect the model lacks.

*4. Decision by decision, not bit for bit.* One differing decision changes the integrator state; from then on the two loops produce different but statistically equivalent bitstreams, so **a bit-for-bit comparison would measure only the time to the first noise-driven disagreement**. The loops are therefore compared decision by decision on the same input (point 3) and by the decoded reading. The same holds for two chips: victim and replica start from the same state and input, but their offsets, dither and noise differ, so they do not stay in step, which the replica attack must account for (steps 5 and 6).

Nothing is simulated yet.

````python
%%writefile mixedsim.py
"""mixedsim.py - one simulation of the complete secure sensor across abstraction levels.

Each block is simulated at the level at which its behaviour matters, and the blocks exchange real signals:

  SRAM-PUF power-up (640 extracted 6T cells: ten 64-cell Monte
  Carlo chunks; the first 600 form the fingerprint), mismatch +
  transient noise                                                    transistor level   ngspice   puf_power_up()
  TIA, buffer, two integrators, 1-bit DAC, dither                    behavioural        ngspice   modulator()
  StrongARM comparator, the modulator's quantiser (extracted,
  with this chip's mismatch and its transient noise), in the loop    transistor level   ngspice   modulator()
  BCH fuzzy-extractor decoder (the compiler's Verilog)               RTL                Icarus    (keygen.simulate)
  Grain-128a keystream, MAC, XOR with the modulator bits             RTL                Icarus    keystream_rtl()
  radio frame, receiver, eavesdropper, replica attacker              behavioural        Python    (the notebook's model)

The modulator netlist follows the difference equations of the notebook's bit-exact model: x1 += g1 (u - v) + n1,
x2 += g2 (x1 - v), with g1 = g2 = 0.5, clipping at +-2.5 and, optionally, an OTA leak; the comparator sees
dVin = (VDD/4) x2' with the dither x2' = x2 (1 + d) + d, and its latched decision drives the DAC v = +-1.
The behavioural integrators update in the comparator's reset phase, from gate pulses of unit area, so the latch holds
the decision they use.
"""
import os, subprocess, time
import numpy as np


# ------------------------------------------------------------------------------- modulator: behavioural loop + SPICE quantiser
def _pwl(name, node, times, values, rise=1e-9):
    """A piecewise-constant source that steps to values[k] at times[k] (linear over `rise`)."""
    pts = ["0", f"{values[0]:.8g}"]
    for t, a, b in zip(times[1:], values[:-1], values[1:]):
        pts += [f"{t - rise:.10g}", f"{a:.8g}", f"{t:.10g}", f"{b:.8g}"]
    return f"V{name} {node} 0 PWL(" + " ".join(pts) + ")"


def modulator(cs, cfg, u, d, n1, seed, S, seg=250, ota_gain=None, waves=4, log=None, scaled=True, options=(".option reltol=1e-6",)):
    """Simulate len(u) modulator clocks with the transistor-level comparator in the loop.

    cs: the comparator module; cfg: its config with layout, corner, vdd, vcm and temp_c set; u, d, n1: per-clock input,
    dither and first-integrator noise (model units); seed: the Monte Carlo mismatch seed (the chip); S: the comparator's
    channel-noise PSD at this PVT (cs.channel_noise_psd), or None for a noiseless comparator.
    Runs in segments of `seg` clocks: each segment starts from the previous one's integrator state (.ic) with one
    warm-up decision, so that ngspice's cost per clock stays constant. scaled: unit-amplitude gate pulses into 12 nF
    integrating capacitors (the same unit-area update; the original 1/12 ns pulses into 1 F left the circuit matrix
    badly conditioned next to the comparator's currents); options: extra '.option' lines. The default reltol = 1e-6 makes
    the integrators exact: with ngspice's default 1e-3 their effective gains were 0.4993 and 0.4972 with per-clock errors
    of 1e-3 to 1e-2, which made two chips' loops drift apart over long runs; at 1e-6 the gains are 0.5000 to 1e-6 and the
    per-clock error, 8e-6, is below the integrator's own thermal noise (about twice the run time). Returns dict(D, x1,
    x2, waves, seconds)."""
    # n clocks of period T; vu = VDD/4 is the comparator input that corresponds to x2 = 1; leak = 1/(1 + A) models an OTA
    # of finite DC gain A (0: ideal integrators)
    n, T, vu = len(u), 1.0 / cfg["f_clk"], cfg["vdd"] / 4
    leak = 0.0 if not ota_gain or not np.isfinite(ota_gain) else 1.0 / (1.0 + ota_gain)
    D, X1, X2, W = np.zeros(n, np.uint8), np.zeros(n), np.zeros(n), None
    x1, x2, t0 = 0.0, 0.0, time.time()
    for s0 in range(0, n, seg):
        m = min(seg, n - s0)
        te = cs.edge_times(cfg, m + 1)                     # edge 0 warms the comparator up; edges 1..m are clocks s0..s0+m-1
        # gate pulse 1 + 11 + 1 ns of amplitude amp into Cint: the update is amp * 12 ns / Cint = 1 times the bracket
        amp, cint = (1.0, 12e-9) if scaled else (1.0 / 12e-9, 1.0)
        gk = cint / 1e-9                                   # clamp conductance: a 1 ns time constant beyond +-2.5
        # the segment's input, dither and noise samples, the first repeated for the warm-up decision
        uu, dd, nn = [np.concatenate([[a[s0]], a[s0:s0 + m]]) for a in (u, d, n1)]
        # The behavioural part of the loop, as SPICE elements around the transistor-level comparator:
        #   su, sd, sn  piecewise-constant input u, dither d and integrator noise n1, stepped at each clock edge (the dither
        #               10 ns early so that it is settled when the comparator samples, the noise 30 ns late)
        #   sg1, sg2    gate pulses that let integrator 1 (55 ns after the edge) and then 2 (70 ns) update once per clock,
        #               in the comparator's reset phase, while its SR latch holds the decision
        #   sv          the 1-bit DAC: +-1 from the SR latch outputs q, qb (tanh: a smooth, steep sign)
        #   x1, x2      integrating capacitors charged by B-source currents: x1 += 0.5 (u - v) + n1, x2 += 0.5 (x1 - v),
        #               minus the OTA leak; Bk1, Bk2 clamp both states at +-2.5 (the model's clipping)
        lines = [
            _pwl("sys_u", "su", te, uu), _pwl("sys_d", "sd", te - 10e-9, dd), _pwl("sys_n", "sn", te + 30e-9, nn),
            f"Vsys_g1 sg1 0 PULSE(0 {amp:.8g} {te[1] + 55e-9:.10g} 1n 1n 11n {T:.6g})",
            f"Vsys_g2 sg2 0 PULSE(0 {amp:.8g} {te[1] + 70e-9:.10g} 1n 1n 11n {T:.6g})",
            "Bsys_v sv 0 V = tanh(20*(V(c0_q)-V(c0_qb)))",      # the 1-bit DAC, driven by the comparator's SR latch
            f"Cx1 x1 0 {cint:.6g}", "Rx1 x1 0 1e15", f"Bx1 0 x1 I = V(sg1)*(0.5*(V(su)-V(sv)) + V(sn) - {leak:.6g}*V(x1))",
            f"Cx2 x2 0 {cint:.6g}", "Rx2 x2 0 1e15", f"Bx2 0 x2 I = V(sg2)*(0.5*(V(x1)-V(sv)) - {leak:.6g}*V(x2))",
            f"Bk1 x1 0 I = {gk:.6g}*(max(V(x1)-2.5,0)+min(V(x1)+2.5,0))", f"Bk2 x2 0 I = {gk:.6g}*(max(V(x2)-2.5,0)+min(V(x2)+2.5,0))",
            *options,
            f".ic v(x1)={x1:.10g} v(x2)={x2:.10g}"]
        # the comparator's transient channel noise (sources per device, a new realisation per segment) and the breakpoints
        # that keep ngspice's time step fine around each decision
        if S is not None:
            src, tstr = cs.noise_sources(cfg, "c0", S, te[1:], np.random.default_rng(seed * 1000 + s0 // seg))
            lines += src + ["Vbrk brk 0 PWL(0 0 " + " ".join(f"{x} 0" for x in tstr) + ")", "Rbrk brk 0 1e3"]
        # the comparator's differential input: the dithered x2 scaled by VDD/4, split around the common mode
        x2in = "(V(x2)*(1+V(sd))+V(sd))"
        inputs = {"c0": [f"Bc0p c0_inp 0 V = {cfg['vcm']:.6g} + {vu / 2:.6g}*{x2in}",
                         f"Bc0n c0_inn 0 V = {cfg['vcm']:.6g} - {vu / 2:.6g}*{x2in}"]}
        saves = ["v(c0_q)", "v(x1)", "v(x2)"] + (["v(clk)", "v(c0_outp)", "v(c0_outn)", "v(c0_qb)"] if s0 == 0 and waves else [])
        name = f"seg{s0 // seg:03d}"
        net = cs.make_netlist(cfg, ["c0"], te[-1] + 0.9 * T, "secure sensor modulator, mixed level", mismatch=True, seed=seed,
                              saves=saves, wrfile=name + ".dat", tmax=5e-9, inputs=inputs, extra=lines, with_latch=True, fine_edges=te)
        cs.run_ngspice(cfg["ngspice_bin"], net, cfg["workdir"], name, 3600)
        w = cs.read_wrdata(os.path.join(cfg["workdir"], name + ".dat"))
        # read the result: each decision from the latch output 0.45 T after its edge, the integrator states at the edges,
        # and the final state of the segment (0.88 T after its last edge, after the last update) to start the next one
        D[s0:s0 + m] = (cs.sample_at(w["time"], w["v(c0_q)"], te[1:] + 0.45 * T) > cfg["vdd"] / 2)
        X1[s0:s0 + m] = cs.sample_at(w["time"], w["v(x1)"], te[1:])
        X2[s0:s0 + m] = cs.sample_at(w["time"], w["v(x2)"], te[1:])
        x1 = float(cs.sample_at(w["time"], w["v(x1)"], [te[-1] + 0.88 * T])[0])
        x2 = float(cs.sample_at(w["time"], w["v(x2)"], [te[-1] + 0.88 * T])[0])
        if s0 == 0 and waves:
            k = (w["time"] >= te[1] - 0.2 * T) & (w["time"] <= te[1 + waves] - 0.2 * T)
            W = {key: np.asarray(w[key])[k] for key in w} | dict(te=te[1:1 + waves])
        if log:
            log(f"  modulator: {s0 + m}/{n} clocks, {time.time() - t0:.0f} s")
    return dict(D=D, x1=X1, x2=X2, waves=W, seconds=time.time() - t0)


def loop_reference(u, d, n1, offset, noise=None, ota_gain=None):
    """The same loop, behavioural, in the netlist's order (decide on x2, then update with that decision): the reference
    for a bit-by-bit comparison. offset, noise: the comparator's, in x2 units (noise: per-clock samples or None)."""
    leak = 0.0 if not ota_gain or not np.isfinite(ota_gain) else 1.0 / (1.0 + ota_gain)
    x1 = x2 = 0.0; D = np.zeros(len(u), np.uint8)
    for k in range(len(u)):
        q = x2 * (1 + d[k]) + d[k] + offset + (0.0 if noise is None else noise[k])
        D[k] = q >= 0; v = 1.0 if D[k] else -1.0
        x1 = min(max(x1 + 0.5 * (u[k] - v) + n1[k] - leak * x1, -2.5), 2.5)
        x2 = min(max(x2 + 0.5 * (x1 - v) - leak * x2, -2.5), 2.5)
    return D


# ------------------------------------------------------------------------------- SRAM PUF: transistor-level power-ups
def puf_power_up(ps, pcfg, cell, n_cells, temps, repeats, chip, noise_seed, pu=None, name="sys_puf", log=None):
    """Power the first n_cells cells of chip `chip` (mismatch seeds chip, chip + 1, ... per 64-cell chunk, as the recorded
    runs) up `repeats` times at each temperature in `temps`. Returns bits[T, R, N] (uint8)."""
    env = ps.noise_envelope(pcfg, sorted(set(temps) | {25}), cell=cell, name=name + "_env")
    mc = ps.run_montecarlo(pcfg, n_cells, temps, repeats, env, chunk=64, mismatch_seed=chip, noise_seed=noise_seed,
                           cell=cell, pu=pu or ps.POWERUP, name=name, log=log or (lambda *a: None))
    return mc["bits"].astype(np.uint8)


# ------------------------------------------------------------------------------- Grain-128a keystream block: RTL in Icarus
TB_SYSTEM = r"""`timescale 1ns/1ps
// The proposal's keystream block (secure_ks_top) driven by the modulator bitstream, with IV = 1 | N | ctr | N_c:
// enc_id = ID ^ z_0..z_15,
// C[j] = D[j] ^ z_(16+j), tag = Grain-128a MAC over ctr(32) | enc_id(16) | C(N), as the model's build_frame.
module tb_system;
  reg clk = 1'b0;
  always #50 clk = ~clk;
  reg key_sh = 0, key_in = 0, nonce_sh = 0, nonce_in = 0, cn_sh = 0, cn_in = 0, start = 0, m = 0, pad = 0, dith_load = 0;
  reg [15:0] seed1 = 16'h1D2B, seed2 = 16'h7A31;
  wire z, run, tag_ok, dith1, dith2;
  wire [0:31] tag;
  wire [31:0] ctr;
  secure_ks_top dut (.clk(clk), .key_sh(key_sh), .key_in(key_in), .nonce_sh(nonce_sh), .nonce_in(nonce_in),
                     .cn_sh(cn_sh), .cn_in(cn_in), .start(start), .m(m), .pad(pad), .dith_load(dith_load), .seed1(seed1), .seed2(seed2),
                     .z(z), .run(run), .tag_ok(tag_ok), .tag(tag), .ctr(ctr), .dith1(dith1), .dith2(dith2));
  localparam integer NMAX = 1 << 18;
  reg dmem [0:NMAX-1];
  reg [127:0] KEY; reg [31:0] NONCE, CTR, CN; reg [15:0] CHIP_ID;
  integer N, k, i, fo; reg [31:0] dly; reg e; reg [1023:0] dfile, ofile;
  task tick; begin @(posedge clk); #1; end endtask
  initial begin
    if (!$value$plusargs("KEY=%h", KEY) || !$value$plusargs("NONCE=%h", NONCE) || !$value$plusargs("CTR=%d", CTR) ||
        !$value$plusargs("ID=%h", CHIP_ID) || !$value$plusargs("N=%d", N) || !$value$plusargs("D=%s", dfile) ||
        !$value$plusargs("OUT=%s", ofile) || !$value$plusargs("CN=%h", CN)) begin $display("missing plusarg"); $finish; end
    $readmemb(dfile, dmem, 0, N - 1);
    fo = $fopen(ofile, "w"); #1;
    dith_load = 1; tick; dith_load = 0;
    key_sh = 1; for (i = 0; i < 128; i = i + 1) begin key_in = KEY[127 - i]; tick; end key_sh = 0;       // key, k_0 first
    nonce_sh = 1; for (i = 0; i < 32; i = i + 1) begin nonce_in = NONCE[31 - i]; tick; end nonce_sh = 0;  // nonce, MSB first
    cn_sh = 1; for (i = 0; i < 31; i = i + 1) begin cn_in = CN[30 - i]; tick; end cn_sh = 0;               // chip nonce, MSB first
    start = 1; for (i = 0; i <= CTR; i = i + 1) tick; start = 0;                                        // counter = CTR
    while (run !== 1'b1) tick;                                                                          // 128 + 32 init clocks
    dly = 32'd0;
    for (k = 0; k < 48 + N; k = k + 1) begin
      if (k < 16) e = CHIP_ID[15 - k] ^ z; else if (k < 16 + N) e = dmem[k - 16] ^ z; else e = 1'b0;
      m = (k < 32) ? ctr[31 - k] : dly[31];                       // MAC message ctr | enc_id | C (32-clock delay line)
      if (k < 16 + N) $fdisplay(fo, "%s %b %b", (k < 16) ? "I" : "C", e, z);
      tick; dly = {dly[30:0], e};
    end
    m = 0; pad = 1; tick; pad = 0;
    $fdisplay(fo, "TAG_OK %b", tag_ok); $fdisplay(fo, "TAG %h", tag); $fclose(fo); $finish;
  end
endmodule
"""


def keystream_rtl(R, key_bits, nonce, ctr, chip_id, D, wd, iverilog, vvp, cn=0):
    """Run the RTL keystream block on the bitstream D (Icarus), with session nonce `nonce` and chip nonce `cn` (31 bits) in
    the IV. The counter is reached by ctr + 1 start pulses, as on the chip. Returns dict(enc_id, C, z (the keystream bits used), tag, tag_ok, seconds)."""
    # write the proposal's Verilog sources, the test bench and the bitstream (one bit per line, for $readmemb)
    os.makedirs(wd, exist_ok=True)
    src = os.path.join(wd, "src"); R.write_verilog_files(src)
    srcs = [os.path.join(src, f) for f in R.DESIGNS["proposal"][1]]
    with open(os.path.join(wd, "tb_system.v"), "w") as f:
        f.write(TB_SYSTEM)
    with open(os.path.join(wd, "D.mem"), "w") as f:
        f.write("\n".join(str(int(b)) for b in D) + "\n")
    t0 = time.time()
    subprocess.run([iverilog, "-g2012", "-o", "sys.vvp", "-s", "tb_system", *srcs, "tb_system.v"], cwd=wd, check=True,
                   capture_output=True, text=True)
    # compile, then simulate with the key (128 bits as hex), nonces, counter, chip ID and bitstream as plusargs
    key_hex = "".join(f"{int(''.join(str(int(b)) for b in key_bits[i:i + 4]), 2):x}" for i in range(0, 128, 4))
    subprocess.run([vvp, "-n", "sys.vvp", f"+KEY={key_hex}", f"+NONCE={nonce:08x}", f"+CN={cn:08x}", f"+CTR={ctr}", f"+ID={chip_id:04x}",
                    f"+N={len(D)}", "+D=D.mem", "+OUT=out.txt"], cwd=wd, check=True, capture_output=True, text=True)
    # parse the bench's output: one line per encrypted ID bit (I) or payload bit (C) with its keystream bit, then the tag
    out = dict(I=[], C=[], z=[])
    for line in open(os.path.join(wd, "out.txt")):
        f = line.split()
        if f[0] in ("I", "C"):
            out[f[0]].append(int(f[1])); out["z"].append(int(f[2]))
        elif f[0] == "TAG":
            out["tag"] = int(f[1], 16)
        elif f[0] == "TAG_OK":
            out["tag_ok"] = f[1] == "1"
    return dict(enc_id=np.array(out["I"], np.uint8), C=np.array(out["C"], np.uint8), z=np.array(out["z"], np.uint8),
                tag=out["tag"], tag_ok=out["tag_ok"], seconds=time.time() - t0)
````

### B. Step 1, transistor level: the SRAM-PUF array powers up

The fingerprint $w$ is read at enrolment (25 °C) and its syndrome published (Section V-D); the field power-up at the operating temperature gives $w'$, in which a cell whose mismatch is small against the noise, or changes with temperature, can flip. The bit error rate $\mathrm{BER} = |w \oplus w'|/600$ follows the error law of Section VII-F, set mainly by noise at tt and moderate temperatures. The analysis above predicts, as the step's pass criterion, that the number of differing cells is close to 600 × BER, with BER from the recorded transistor-level Monte Carlo of Section VII-F at this corner and temperature, within the binomial spread $\sqrt{600\,\mathrm{BER}(1-\mathrm{BER})}$; **the key is recoverable only if that number is at most t**.

The cells are simulated as ngspice transients of the extracted 6T cell with chip 1's mismatch and transient-noise seed 7 (Simulation setup). Fig. 75 draws the enrolled fingerprint $w$ as a 20 × 30 grid and marks in red the cells of $w \oplus w'$, **so the red cells are the differing cells that the pass criterion counts**.

> **Next cell.** 1. Powers the 640 cells up in ngspice at 25 °C (enrolment) and at the chosen temperature (field read) and keeps the first 600; 2. counts the cells that differ; 3. compares the error rate with the recorded Monte Carlo at this corner and with t; 4. draws the fingerprint and the differing cells. About two and a half minutes on two cores (162 s in the published run).

````python
#@title Fig.: mixed level, step 1: SRAM-PUF power-ups
# Step 1, transistor level: this chip's SRAM-PUF array powered up in ngspice: once at enrolment (25 °C, the factory) and
# once in the field at TEMP_C. 640 cells of the extracted layout (the first 600 form the fingerprint), mismatch and noise
# Live (ngspice). Inputs: ps (puf_spice), MS (mixedsim), LAY_PUF (the extracted cell), CORNER, TEMP_C, PUF_CHIP, PVT
# (recorded PUF Monte Carlo per corner), TEMPS, CAL["bch_t"]. Outputs: W_EN (enrolled fingerprint), W_RD (field read),
# n_err (number of differing bits); a card and the fingerprint map.
t0 = time.time()
# one ngspice job per available core (at least two)
pcfg_m = ps.make_config(str(SKY130_LIB), ngspice=str(NGSPICE), workdir=str(MIX / "puf"), corner=CORNER, jobs=max(2, len(os.sched_getaffinity(0)) or 2))
# the post-layout cell: puf_spice's sram6t subcircuit with the extracted parasitics
POST_M = dict(ps.CELL, layout=LAY_PUF)
# puf_power_up returns bits[T, R, N] (temperature, repeat, cell). The mismatch seed PUF_CHIP fixes the chip; the noise
# seed 7 fixes the transient-noise draws. At 25 °C both reads are two repeats at one temperature; otherwise one power-up
# at 25 °C (enrolment) and one at TEMP_C (field)
if TEMP_C == 25:
    _b = MS.puf_power_up(ps, pcfg_m, POST_M, 640, [25], 2, PUF_CHIP, 7, name="mixed_puf")
    W_EN, W_RD = _b[0, 0, :PUF_BITS], _b[0, 1, :PUF_BITS]
else:
    _b = MS.puf_power_up(ps, pcfg_m, POST_M, 640, [25, TEMP_C], 1, PUF_CHIP, 7, name="mixed_puf")
    W_EN, W_RD = _b[0, 0, :PUF_BITS], _b[1, 0, :PUF_BITS]
TIM_M["puf"] = time.time() - t0
# expected BER: the recorded Monte Carlo's BER against a 25 °C enrolment, interpolated linearly in temperature
_pc = next(c for c in PVT["puf"]["corners"] if c["corner"] == CORNER)       # the recorded Monte Carlo at this corner
ber_exp = float(np.interp(TEMP_C, TEMPS, [r["ber"] for r in _pc["rows"]]))
# the number of bit errors the BCH decoder of step 2 must correct (it succeeds if n_err <= t)
n_err = int((W_EN != W_RD).sum())
card("Step 1, transistor level: SRAM-PUF power-ups",
     [("fingerprint", f"{PUF_BITS} cells, P(1) = {W_EN.mean():.3f}"),
      (f"bits that differ from the enrolment at {TEMP_C} °C", f"<b>{n_err}</b> (BER {100 * n_err / PUF_BITS:.2f} %); the "
       f"recorded Monte Carlo at this corner gives {100 * ber_exp:.2f} %" + (f" (interpolated to {TEMP_C} °C)" if TEMP_C not in TEMPS else "")
       + f", that is {PUF_BITS * ber_exp:.1f} ± {np.sqrt(PUF_BITS * ber_exp * (1 - ber_exp)):.1f} expected (binomial)"),
      ("BCH correction t", f"{CAL['bch_t']}: the key is recovered if at most {CAL['bch_t']} bits differ"),
      ("run time (ngspice)", f"{TIM_M['puf']:.0f} s")], tone="ok" if n_err <= CAL["bch_t"] else "bad")
# the fingerprint as a 20 x 30 map, cell k at row k // 30, column k % 30; grey = 1 at enrolment
from matplotlib.patches import Patch, Rectangle
fig, ax = plt.subplots(figsize=(W1 + 0.9, 2.75))
from matplotlib.colors import ListedColormap
ax.imshow(W_EN.reshape(20, 30), cmap=ListedColormap(["white", "#a3a3a3"]), vmin=0, vmax=1, interpolation="nearest")
for k in np.flatnonzero(W_EN != W_RD):                  # the cells that power up to the other value in the field read
    ax.add_patch(Rectangle((k % 30 - 0.5, k // 30 - 0.5), 1, 1, fc=COL["key"], ec=COL["ink"], lw=0.8))
ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
ax.legend(handles=[Patch(fc="#a3a3a3", ec=COL["ink"], lw=0.4, label="1 at enrolment"), Patch(fc="white", ec=COL["ink"], lw=0.4, label="0 at enrolment"),
                   Patch(fc=COL["key"], ec=COL["ink"], lw=0.8, label=f"differs at {TEMP_C} °C ({n_err})")],
          fontsize=7.5, loc="upper center", bbox_to_anchor=(0.5, -0.02), ncol=3, frameon=False, handlelength=1.1, columnspacing=1.0)
ax.set_title(f"{n_err} of {PUF_BITS} cells differ; the code corrects up to {CAL['bch_t']}", loc="center", fontsize=9.5, pad=4)
fig.tight_layout(); plt.show()
# interactive companion: the same fingerprint, each cell with its index, enrolled value and field read (hover)
def _imixpuf():
    import plotly.graph_objects as go
    Z, Rd = W_EN.reshape(20, 30), W_RD.reshape(20, 30)
    txt = [[f"cell {30 * r + c}: enrolment {Z[r, c]}, field read at {TEMP_C} °C {Rd[r, c]}" + (", differs" if Z[r, c] != Rd[r, c] else "")
            for c in range(30)] for r in range(20)]
    f = go.Figure(go.Heatmap(z=Z, colorscale=[[0, "white"], [1, "#a3a3a3"]], zmin=0, zmax=1, showscale=False, text=txt,
                             hovertemplate="%{text}<extra></extra>"))
    for k in np.flatnonzero(W_EN != W_RD):
        f.add_shape(type="rect", x0=k % 30 - 0.5, x1=k % 30 + 0.5, y0=k // 30 - 0.5, y1=k // 30 + 0.5, fillcolor=COL["key"],
                    line=dict(color=COL["ink"], width=0.8 * _PT))
    for nm, fc in (("1 at enrolment", "#a3a3a3"), ("0 at enrolment", "white"), (f"differs at {TEMP_C} °C ({n_err})", COL["key"])):
        f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", name=nm, hoverinfo="skip",
                               marker=dict(symbol="square", size=5.5, color=fc, line=dict(color=COL["ink"], width=0.6))))
    f.update_xaxes(range=[-0.5, 29.5], showticklabels=False, ticks="", constrain="domain")
    f.update_yaxes(range=[19.5, -0.5], showticklabels=False, ticks="", scaleanchor="x", scaleratio=1)
    f.add_annotation(text=f"{n_err} of {PUF_BITS} cells differ; the code corrects up to {CAL['bch_t']}", x=0.5, y=1, xref="paper",
                     yref="paper", yanchor="bottom", yshift=round(4 * _PT), showarrow=False, font=dict(size=round(9.5 * _PT)))
    f.update_layout(legend=dict(orientation="h", x=0.5, xanchor="center", y=-0.04, yanchor="top", font=dict(size=round(7.5 * _PT))))
    return f
iplot(_imixpuf, "Fig.: mixed level, step 1, the enrolled fingerprint and the cells that differ in the field read (interactive)")
````

**Key insights.** The fingerprint has 600 cells with a fraction of ones P(1) = 0.472. In the field read at 37 °C, **7 of 600 cells** power up differently (Fig. 75, red), a BER of 1.17 %, scattered over the array. The recorded Monte Carlo at tt, interpolated to 37 °C, gives 1.32 %, that is 7.9 ± 2.8 expected errors: the observed 7 lies within one standard deviation, as predicted, and the code corrects up to 45, a margin of 38 errors. The transistor-level cells are thus consistent with the PUF error model of Section VII-F. The result holds for one Monte Carlo chip and one power-up: the key-failure probability over chips and corners, and the −40 and 85 °C enrolment on which the security claims rest, come from Sections VII-E and VIII-F.

### C. Step 2, RTL: the key is reconstructed in hardware

The decoder corrects the field read of step 1 with the helper data as in Section V-D: if the error pattern has at most t ones, the corrected word is exactly $w$, and its hash equals the key the server derived at enrolment. The decoder is deterministic, so its RTL and the model's `reconstruct()` must agree bit for bit; **a difference would be a bug, not noise**. The analysis above predicts, as the step's pass criterion, that with the errors of step 1 and t = 45 the decoder succeeds, that the corrected word equals the enrolled fingerprint and the model's output, and that the hashed key equals the server's.

The decoder is the one that the compiler of Section XI generates for n = 600 and t of Section VII-E with no repetition stage, **simulated cycle by cycle in Icarus Verilog**; the hash is behavioural (Python).

> **Next cell.** 1. Computes the helper data (BCH syndrome) of the enrolled fingerprint; 2. simulates the generated decoder in Icarus Verilog on the field read and the helper data; 3. runs the model's decoder on the same inputs; 4. hashes the corrected word into the key and compares it with the server's enrolled key. A few seconds.

````python
#@title Mixed level, step 2: key reconstruction in RTL
# Step 2, RTL: the key-reconstruction hardware generated by the compiler of Section XI (BCH over GF(2^10),
# n = 600, t of Section VII-E), simulated cycle by cycle in Icarus Verilog on the field read and the helper data
# published at enrolment, and checked against the model's fuzzy extractor
# Live (Icarus Verilog). Inputs: W_EN, W_RD (step 1), KG (the key-generator compiler, dev/rtl/keygen.py), CAL["bch_t"].
# Outputs: H_EN (helper data), KG_M (the RTL decoder's corrected word, success flag, cycle count), REC_M (the model's
# decoder), KEY_M (the chip's key) and KEY_SRV (the key the server stored at enrolment).
t0 = time.time(); t_m = CAL["bch_t"]
# syndrome construction: h = the t odd syndromes S_1, S_3, ..., S_(2t-1) of w over GF(2^10), 10 bits each
H_EN = bch_syndrome(W_EN, t_m)                                  # helper data stored at enrolment: 10 t bits
# the compiled decoder with no repetition code (r = 1, n = 600 cells, strength t); the field read enters as the cells,
# the repetition helper is all zero (unused at r = 1) and the BCH helper is h serialised MSB first. The decoder computes the
# syndromes of w', adds h (giving the syndromes of the error pattern), runs Berlekamp-Massey and a Chien search, and
# outputs w' xor e
KG_M = KG.simulate(dict(r=1, n=PUF_BITS, t=t_m), str(MIX / "keygen"),
                   [dict(cells=W_RD, rhelp=np.zeros(PUF_BITS, np.uint8), bhelp=syn_to_bits(H_EN))],
                   str(PNR_BIN / "iverilog"), str(PNR_BIN / "vvp"))[0]
# the same reconstruction in the behavioural model (Section V), for the bit-for-bit comparison
REC_M = reconstruct(W_RD, H_EN, t_m)
# key = the first 128 bits of SHA-256 of the 600-bit word: equal keys iff the corrected word equals the enrolled one
KEY_M, KEY_SRV = hash_key(KG_M["bits"]), hash_key(W_EN)         # the chip's key; the key the server stored at enrolment
TIM_M["bch"] = time.time() - t0
card("Step 2, RTL: key reconstruction in the generated hardware",
     [("errors in the field read", f"{n_err} of {PUF_BITS} (t = {t_m})"),
      ("RTL decoder", f"success flag {KG_M['ok']}; corrected word equals the enrolled fingerprint: <b>{np.array_equal(KG_M['bits'], W_EN)}</b>"),
      ("RTL against the model's reconstruct()", f"identical bit for bit: <b>{np.array_equal(KG_M['bits'], REC_M['w'])}</b>"),
      ("key (SHA-256 of the corrected fingerprint, 128 bits)", f"equals the server's enrolled key: <b>{np.array_equal(KEY_M, KEY_SRV)}</b>"),
      # cycles / 10 = µs at the 10 MHz clock
      ("latency", f"{KG_M['cycles']:,} clock cycles = {KG_M['cycles'] / 10:.0f} µs at 10 MHz, once per power-up"),
      ("run time (Icarus Verilog)", f"{TIM_M['bch']:.0f} s")], tone="ok" if np.array_equal(KEY_M, KEY_SRV) else "bad")
````

**Key insights.** The RTL decoder reports success; its corrected word equals the enrolled fingerprint and the model's `reconstruct()` bit for bit, and the 128-bit key equals the server's enrolled key, as predicted: **the key generator works on transistor-level cells**, and the generated RTL agrees with the model behind Sections VIII and IX. Decoding takes 91,127 clock cycles, 9.1 ms at 10 MHz, once per power-up, a latency negligible for a sensor that then runs continuously.

### D. Step 3, transistor level in the loop: two chips' modulators

The step applies the decision-by-decision test of Theory, point 3, to each chip with its own $V_{os}$. The analysis above predicts, as the step's pass criterion, that for each chip the number of decisions that differ from the threshold model is consistent with $\mathbb{E}[N_{\neq}]$ from that chip's $\Delta V_k$, $V_{os}$ and $\sigma_n$, that **every differing decision lies within a few** $\sigma_n$ of the threshold (the cell marks the run as failed if any lies beyond 6 $\sigma_n$), and that the decoded reading equals the behavioural model's and the input, within the decimator's resolution. Panel (c) of Fig. 77 shows this test: it plots each decision's distance from the threshold in units of $\sigma_n$ and marks in grey the decisions within 3 $\sigma_n$, where noise can change a decision, so a decision that differs from the threshold model is expected among them; panels (a), (b) and (d) show the comparator's waveforms, the integrator states and the decoded reading.

Each chip's extracted StrongARM (seeds 1000 and 1004) runs inside the behavioural loop of the Simulation setup (Fig. 76), the two chips in parallel on two cores. **Each offset is measured in SPICE before the loop** (Section VII-B). The behavioural reference is the model of Section V-C with the same offset and $\sigma_n$. The reading uses the CIC decimator only: the 31-tap compensation FIR needs 31 output samples (1,550 clocks) to fill, more than this run.

<p align="center" style="text-align:center"><img src="https://raw.githubusercontent.com/chennakeshavadasa/secure-sensor-figures/dbb22aab44d606226b956ce604be2fc14db322b0/img/xschem_mixed.png" width="820" alt="xschem testbench of the mixed-level modulator" style="display:inline-block"></p>

<p align="center" style="text-align:center"><b>Fig. 76.</b> Mixed-level modulator testbench in xschem (SKY130 symbols): the comparator x1 with SR latch and noise currents, its loads, the behavioural second-order loop (input, dither and noise sources; gate pulses $\mathrm{sg}_1$, $\mathrm{sg}_2$; integrators $x_1$, $x_2$ with clamps; DAC sv; input sources $B_\mathrm{c0p}$, $B_\mathrm{c0n}$), the breakpoint helpers and the NGSPICE block.</p>

> **Next cell.** 1. Measures each chip's comparator offset in SPICE; 2. runs both modulators in ngspice in parallel, with each chip's StrongARM in the loop (96 s for 1,000 clocks on two cores in the published run, in proportion to the number of clocks); 3. tests every decision against the threshold model and counts the disagreements against $\mathbb{E}[N_{\neq}]$; 4. runs the behavioural model of the same chip and decodes both bitstreams; 5. draws the figure.

````python
#@title Fig.: mixed level, step 3: modulators with the transistor-level comparator
# Step 3, transistor level in the loop: each chip's second-order modulator in ngspice, with behavioural integrators, DAC
# and dither around its post-layout StrongARM comparator (its own Monte Carlo mismatch and transient noise). The two chips
# run in parallel. Then the decision-by-decision test of the behavioural comparator model of Section VIII
t0 = time.time()
# one SPICE configuration per chip: the extracted comparator at the chosen corner, supply and temperature, own work dir
def ccfg_m(tag):
    c = cs.default_config(str(SKY130_LIB), str(NGSPICE), str(MIX / f"cmp_{tag}"), quick=True)
    c = dict(c, layout=LAY_CMP, corner=CORNER, vdd=VDD_M, vcm=VDD_M / 2, temp_c=float(TEMP_C))
    os.makedirs(c["workdir"], exist_ok=True); return c
CM = [ccfg_m("victim"), ccfg_m("replica")]
S_M = cs.channel_noise_psd(CM[0], cs.reference_trajectory(CM[0]))     # the devices' noise PSD at this corner and T
# the modulator input (constant 0.5 µA, or a sine), each chip's dither sequence (its own LFSR seeds) and each chip's
# first-integrator noise realisation; the comparator's noise is the transient noise injected in ngspice
u_m = (np.full(N_CLOCKS, s2d(tia(0.5e-6, P_sys["rtia"]), P_sys["buffer"])) if INPUT == "constant" else
       s2d(tia(sensor_current({**P_sys, "sig": "sine", "bin": 100}, N_CLOCKS), P_sys["rtia"]), P_sys["buffer"]))
d_m = [dsm2(u_m, P_sys)["DI"].astype(float), dsm2(u_m, P_sys, seeds=REPLICA_DITHER)["DI"].astype(float)]   # each chip's dither
n1_m = [P_sys["int_noise"] * np.random.default_rng(31 + i).standard_normal(N_CLOCKS) for i in range(2)]
def vos_of(c, seed):                                   # this chip's comparator offset, by the SAR search of Section VII-B
    net, te = cs.run_offset_job(c, 1, seed, "vos"); cs.run_ngspice(c["ngspice_bin"], net, c["workdir"], "vos")
    return cs.read_offsets(c, "vos", 1, te)[0][0]
# both chips at once (one ngspice process per core): first each comparator's offset, then its modulator: behavioural
# integrators, DAC and dither around the transistor-level comparator, run in segments of 250 clocks with the integrator
# state handed over (the victim also keeps the waveforms of 4 clocks for panel (a))
with cf.ThreadPoolExecutor(2) as ex:
    VOS_M = [f.result() for f in [ex.submit(vos_of, CM[i], CHIPS[i]) for i in range(2)]]
    MOD = [f.result() for f in [ex.submit(MS.modulator, cs, CM[i], u_m, d_m[i], n1_m[i], CHIPS[i], S_M, 250, None, 4 if i == 0 else 0)
                                for i in range(2)]]
TIM_M["mod"] = time.time() - t0
VU_M = VDD_M / 4
chk = []
for m, d, vo in zip(MOD, d_m, VOS_M):                   # decision by decision: the threshold model on the input the circuit saw
    # dv: the comparator input in volts on every clock (x2 with the dither applied); the threshold model decides 1 when
    # dv >= the chip's offset; exp: the decisions that noise alone should flip, sum of Phi(-|dv - vos| / sigma_n)
    dv = VU_M * (m["x2"] * (1 + d) + d)
    dis = m["D"] != (dv >= vo).astype(np.uint8)
    chk.append(dict(dv=dv, dis=dis, exp=float(ndtr(-np.abs(dv - vo) / SIG_N_M).sum()),
                    worst=float(np.abs(dv - vo)[dis].max() / SIG_N_M) if dis.any() else 0.0))
D_A, D_B = MOD[0]["D"], MOD[1]["D"]
y_true = float(u_m.mean())
# the receiver's CIC stage alone: its 31-tap compensation FIR runs at the output rate and needs 31 output samples (1,550
# clocks) to fill, most of a short run, so the full decoder would still be settling here
dec_m = lambda D: cic(D)
reading = lambda D: float(dec_m(D)[3:].mean())
D_beh = dsm2(u_m, {**P_sys, "comp_noise": SIG_N_M / VU_M}, offset=-VOS_M[0] / VU_M)["D"]       # the behavioural model, same chip
card("Step 3, transistor level in the loop: the two chips' modulators",
     [("comparator offsets (SPICE, this corner and T)", f"victim {1e3 * VOS_M[0]:+.3f} mV, replica {1e3 * VOS_M[1]:+.3f} mV"),
      ("decisions equal to the threshold model [ΔV > V<sub>os</sub>]", "; ".join(
          f"{nm}: <b>{100 * (1 - c['dis'].mean()):.2f} %</b> ({int(c['dis'].sum())} differ, theory expects {c['exp']:.1f} from noise; "
          + (f"largest at |ΔV − V<sub>os</sub>| = {c['worst']:.1f} σ<sub>n</sub>)" if c["dis"].any() else "none differs)")
          for nm, c in zip(("victim", "replica"), chk))),
      ("both chips: observed against expected", (lambda n, lam: f"{n} differ, {lam:.1f} expected; a Poisson count of mean {lam:.1f} is "
       + (f"≤ {n} with probability {sum(np.exp(-lam) * lam**k / math.factorial(k) for k in range(n + 1)):.0%}" if n <= lam else
          f"≥ {n} with probability {1 - sum(np.exp(-lam) * lam**k / math.factorial(k) for k in range(n)):.0%}"))(
       int(sum(c['dis'].sum() for c in chk)), sum(c['exp'] for c in chk))),
      ("decoded reading of the victim's bits", f"transistor-level loop {reading(D_A):+.4f}, behavioural model {reading(D_beh):+.4f}, "
       f"input {y_true:+.4f} (full scale ±1)" if INPUT == "constant" else "see the figure"),
      ("run time (ngspice, two chips in parallel)", f"{TIM_M['mod']:.0f} s for {N_CLOCKS:,} clocks per chip "
       f"({1e3 * TIM_M['mod'] / N_CLOCKS:.0f} ms per clock)")],
     tone="ok" if all(c["worst"] < 6 for c in chk) else "bad")
# (a) the victim's StrongARM in the loop, (b) the integrator states from ngspice, (c) the decision-by-decision test: where
# every decision's input lay relative to the threshold and which ones differ from the model, (d) what the receiver decodes
n_dis, n_exp = int(sum(c["dis"].sum() for c in chk)), sum(c["exp"] for c in chk)
fig, axs = plt.subplots(2, 2, figsize=(W2, 5.8))
w = MOD[0]["waves"]; tw = 1e9 * (w["time"] - w["te"][0])
ax = axs[0, 0]                                                                  # a timing diagram: one row per node (names as in
ROWS_A = [(r"clock $\phi$", "v(clk)", COL["faint"]), (r"$\mathrm{out}_p$", "v(c0_outp)", COL["ok"]),   # the schematic), each row
          (r"$\mathrm{out}_n$", "v(c0_outn)", COL["bad"]), (r"SR latch $q$", "v(c0_q)", COL["ink"])]   # from 0 to 1.8 V
for k, (lab, node, col) in enumerate(ROWS_A):
    y0 = 1.3 * (len(ROWS_A) - 1 - k)
    ax.axhline(y0, color=COL["faint"], lw=0.4, alpha=0.6)
    ax.plot(tw, y0 + w[node] / 1.8, color=col, lw=1.2)
ax.set_yticks([1.3 * (len(ROWS_A) - 1 - k) + 0.5 for k in range(len(ROWS_A))], [r[0] for r in ROWS_A])
ax.tick_params(axis="y", left=False, right=False); ax.set_ylim(-0.25, 1.3 * (len(ROWS_A) - 1) + 1.25); ax.grid(axis="y", visible=False)
ax.set_xlabel("Time (ns)")
panel(ax, "a", "the victim's StrongARM in the loop (4 clocks)")
ax = axs[0, 1]; kk = np.arange(min(200, N_CLOCKS))
ax.step(kk, MOD[0]["x2"][kk], where="post", color=COL["ink"], lw=1.2, label="$x_2$ (ngspice)")
ax.step(kk, MOD[0]["x1"][kk], where="post", color=COL["aqua"], lw=1.0, label="$x_1$ (ngspice)")
ax.set_xlabel("Clock"); ax.set_ylabel("Integrator output (model units)"); ax.legend(fontsize=7, loc="upper right")
YB = ax.get_ylim(); YB = (YB[0], YB[1] + 0.3 * (YB[1] - YB[0])); ax.set_ylim(*YB)         # headroom: the key above the states
panel(ax, "b", "integrator states")
ax = axs[1, 0]
xs_m = [np.log10(np.maximum(np.abs(c["dv"] - vo) / SIG_N_M, 1e-3)) for c, vo in zip(chk, VOS_M)]
lo_m = min(-1.25, min(x.min() for x in xs_m) - 0.15)
ax.axvspan(lo_m, np.log10(3), color=COL["faint"], alpha=0.15, lw=0, label=r"within $3\sigma_\mathrm{n}$")
for x, c, col, nm in zip(xs_m, chk, (COL["ok"], COL["bad"]), ("victim", "replica")):
    ax.hist(x, bins=60, color=col, alpha=0.35, label=f"{nm}: all decisions")
top_m = 1.45 * ax.get_ylim()[1]; ax.set_ylim(0, top_m)
for k, (x, c, col, nm) in enumerate(zip(xs_m, chk, (COL["ok"], COL["bad"]), ("victim", "replica"))):
    ax.plot(x[c["dis"]], np.full(int(c["dis"].sum()), top_m * (0.93 - 0.06 * k)), "v", color=col, ms=7, mec=COL["ink"], mew=0.5,
            label=f"{nm}: differs ({int(c['dis'].sum())})")
ax.set_xlim(lo_m, None)
ax.set_xlabel(r"$\log_{10}(|\Delta V - V_\mathrm{os}|/\sigma_\mathrm{n})$"); ax.set_ylabel("Decisions")
ax.legend(fontsize=6.3, loc="center left", bbox_to_anchor=(0.0, 0.5), handlelength=1.2)
panel(ax, "c", f"decision test: {n_dis} of {2 * N_CLOCKS:,} differ, noise predicts {n_exp:.1f}")
ax = axs[1, 1]; yA, yB = dec_m(D_A), dec_m(D_beh)
ax.plot(yA, "o-", color=COL["ok"], ms=3, lw=1.2, label="transistor-level loop")
ax.plot(yB, "s--", color=COL["key"], ms=3, lw=1.0, label="behavioural model")
ax.plot(u_m[::OSR][:len(yA)], ":", color=COL["ink"], lw=1.2, label="input")
ax.set_xlabel("Output sample (OSR = 50)"); ax.set_ylabel("CIC output (full scale ±1)"); ax.legend(fontsize=7, loc="best")
panel(ax, "d", "what the receiver's filters make of the bits")
fig.tight_layout(); plt.show()

def _imod():
    """Interactive companion of the figure above, all four panels: (a) the victim's comparator nodes in the loop (hover in
    volts), (b) the integrator states (every clock, at most 5,000 points; opens on the first 200 as the static panel), (c)
    the decision test binned as above with each differing decision marked (clock, distance, ΔV in hover), (d) the decoded
    output."""
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    f = make_subplots(2, 2, horizontal_spacing=0.11, vertical_spacing=0.17,
                      subplot_titles=("(a) The victim's StrongARM in the loop (4 clocks)", "(b) Integrator states",
                                      f"(c) Decision test: {n_dis} of {2 * N_CLOCKS:,} differ, noise predicts {n_exp:.1f}",
                                      "(d) What the receiver's filters make of the bits"))
    rows_i = [("clock ϕ", "v(clk)", COL["faint"]), ("out<sub>p</sub>", "v(c0_outp)", COL["ok"]),
              ("out<sub>n</sub>", "v(c0_outn)", COL["bad"]), ("SR latch <i>q</i>", "v(c0_q)", COL["ink"])]
    for k, (lab, node, col) in enumerate(rows_i):                   # (a) one row per node, each from 0 to 1.8 V, as the static
        y0 = 1.3 * (len(rows_i) - 1 - k)
        f.add_trace(go.Scatter(x=[tw[0], tw[-1]], y=[y0, y0], mode="lines", line=dict(color=COL["faint"], width=0.4), showlegend=False,
                               hoverinfo="skip"), 1, 1)
        v_ = np.asarray(w[node]); i_ = _ithin(tw, v_, 0.002)          # the transient within 2 mV, without its flat stretches
        f.add_trace(go.Scatter(x=tw[i_], y=y0 + v_[i_] / 1.8, mode="lines", line=dict(color=col, width=1.2), showlegend=False,
                               customdata=v_[i_], hovertemplate=f"{lab}<br>" + "%{x:.1f} ns: %{customdata:.3f} V<extra></extra>"), 1, 1)
    f.update_yaxes(tickvals=[1.3 * (len(rows_i) - 1 - k) + 0.5 for k in range(len(rows_i))], ticktext=[r[0] for r in rows_i], ticks="",
                   range=[-0.25, 1.3 * (len(rows_i) - 1) + 1.25], row=1, col=1)
    f.update_xaxes(title_text="Time (ns)", row=1, col=1)
    n = len(MOD[0]["x2"]); st = max(1, int(np.ceil(n / 5000))); k = np.arange(0, n, st)
    for nm, key, col, wd in (("<i>x</i><sub>2</sub> (ngspice)", "x2", COL["ink"], 1.2), ("<i>x</i><sub>1</sub> (ngspice)", "x1", COL["aqua"], 1.0)):
        f.add_trace(go.Scatter(x=k, y=np.asarray(MOD[0][key])[k], mode="lines", line=dict(color=col, width=wd, shape="hv"), name=nm,
                               hovertemplate="clock %{x}<br>%{y:.4f}<extra>" + key + "</extra>"), 1, 2)
    f.update_xaxes(title_text="Clock", range=[-2, min(200, n)], row=1, col=2)
    f.update_yaxes(title_text="Integrator output (model units)", dtick=0.25, range=list(YB), row=1, col=2)
    edges = np.linspace(lo_m, max(float(x.max()) for x in xs_m) + 1e-9, 61)
    ctr = 0.5 * (edges[1:] + edges[:-1]); top = 0
    f.add_vrect(x0=lo_m, x1=float(np.log10(3)), fillcolor=COL["faint"], opacity=0.15, line_width=0, layer="below",
                exclude_empty_subplots=False, row=2, col=1)
    f.add_trace(go.Scatter(x=[None], y=[None], mode="markers", name="within 3σ<sub>n</sub>", legend="legend2", hoverinfo="skip",
                           marker=dict(symbol="square", size=5, color=COL["faint"], opacity=0.3)), 2, 1)
    for x, c, col, nm in zip(xs_m, chk, (COL["ok"], COL["bad"]), ("victim", "replica")):
        h, _ = np.histogram(x, edges); top = max(top, int(h.max()))
        f.add_trace(go.Bar(x=ctr, y=h, width=float(edges[1] - edges[0]), name=f"{nm}: all decisions", legend="legend2",
                           marker=dict(color=col, opacity=0.35, line=dict(width=0)), customdata=np.c_[10 ** edges[:-1], 10 ** edges[1:]],
                           hovertemplate="|ΔV − V<sub>os</sub>|/σ<sub>n</sub> from %{customdata[0]:.3g} to %{customdata[1]:.3g}"
                                         "<br>%{y} decisions<extra>" + nm + "</extra>"), 2, 1)
    for j, (x, c, col, nm) in enumerate(zip(xs_m, chk, (COL["ok"], COL["bad"]), ("victim", "replica"))):
        idx = np.flatnonzero(c["dis"])
        yd_ = np.full(len(idx), 1.45 * top * (0.93 - 0.06 * j))   # an empty marker set still gets its legend entry, as the static
        f.add_trace(go.Scatter(x=x[idx] if len(idx) else [None], y=yd_ if len(idx) else [None], mode="markers",
                               legend="legend2", name=f"{nm}: differs ({len(idx)})",
                               marker=dict(symbol="triangle-down", size=7, color=col, line=dict(color=COL["ink"], width=0.5)),
                               customdata=np.c_[idx, 10 ** x[idx], 1e3 * c["dv"][idx]],
                               hovertemplate="clock %{customdata[0]:.0f}<br>|ΔV − V<sub>os</sub>|/σ<sub>n</sub> = %{customdata[1]:.3g}"
                                             "<br>ΔV = %{customdata[2]:.3f} mV<extra>" + nm + "</extra>"), 2, 1)
    f.update_xaxes(title_text="log<sub>10</sub>(|Δ<i>V</i> − <i>V</i><sub>os</sub>|/σ<sub>n</sub>)", range=[lo_m, None], row=2, col=1)
    f.update_yaxes(title_text="Decisions", range=[0, 1.45 * top], row=2, col=1)
    yA_, yB_ = dec_m(D_A), dec_m(D_beh)
    for y, nm, col, dash, sym, wd in ((yA_, "transistor-level loop", COL["ok"], "solid", "circle", 1.2),
                                      (yB_, "behavioural model", COL["key"], "dash", "square", 1.0)):
        f.add_trace(go.Scatter(y=y, mode="lines+markers", name=nm, legend="legend3", line=dict(color=col, dash=dash, width=wd),
                               marker=dict(symbol=sym, size=3), hovertemplate="sample %{x}<br>%{y:+.4f}<extra>" + nm + "</extra>"), 2, 2)
    f.add_trace(go.Scatter(y=u_m[::OSR][:len(yA_)], mode="lines", name="input", legend="legend3", line=dict(color=COL["ink"], dash="dot", width=1.2),
                           hovertemplate="sample %{x}<br>%{y:+.4f}<extra>input</extra>"), 2, 2)
    f.update_xaxes(title_text="Output sample (OSR = 50)", row=2, col=2)
    f.update_yaxes(title_text="CIC output (full scale ±1)", row=2, col=2)
    leg = lambda fs=7, **kw: dict(bgcolor="rgba(255,255,255,0.92)", borderwidth=0, font=dict(size=round(fs * _PT)), **kw)
    xb, yb = f.layout.xaxis2.domain, f.layout.yaxis2.domain
    xc, yc = f.layout.xaxis3.domain, f.layout.yaxis3.domain
    xd, yd = f.layout.xaxis4.domain, f.layout.yaxis4.domain
    f.update_layout(barmode="overlay",
                    legend=leg(x=xb[1] - 0.005, y=yb[1] - 0.005, xanchor="right", yanchor="top"),
                    legend2=leg(x=xc[0] + 0.005, y=(yc[0] + yc[1]) / 2, xanchor="left", yanchor="middle", fs=6.5),
                    legend3=leg(x=xd[1] - 0.005, y=yd[0] + 0.02, xanchor="right", yanchor="bottom"))
    return f
iplot(_imod, "Fig.: step 3, transistor-level modulators and the decision test (interactive)")
````

### E. Step 4, RTL: encryption and MAC

From the 128-bit key and the initialisation vector IV = 1 ‖ N ‖ ctr ‖ $N_\mathrm{c}$ (session nonce, frame counter, chip nonce), Grain-128a produces a keystream $z_0, z_1, \dots$. The frame carries the chip ID encrypted with the first 16 keystream bits, the ciphertext $C_j = D_j \oplus z_{16+j}$ of the modulator bits $D_j$, and a 32-bit tag from Grain-128a's authentication mode over the counter, the encrypted ID and $C$. The model's Grain-128a matches the published test vectors (Section V-B); since every output is deterministic, any difference would be an error in the hardware's bit order, counter or MAC timing. The analysis above therefore predicts, as the step's pass criterion, that **the RTL's ciphertext, encrypted chip ID and tag equal the model's bit for bit**.

The RTL is the block's one-bit-per-clock form (`secure_ks_top`), simulated in Icarus Verilog with the key of step 2, the frame settings of the Simulation setup, the victim's 1,000 bits of step 3 and the 31-bit chip nonce $N_\mathrm{c}$ that the model derives from the single transistor-level read of step 1 (printed on the result panel; the 63 further reads of a session are not simulated at transistor level, and $N_\mathrm{c}$ enters the block as an input, **so its value does not affect the checks**); the model computes the same frame in Python.

> **Next cell.** 1. Simulates the keystream block in Icarus Verilog with the key of step 2 and the victim's bits of step 3; 2. computes the same frame with the model's Grain-128a; 3. compares the ciphertext, the encrypted chip ID and the MAC tag. Under a second.

````python
#@title Mixed level, step 4: Grain-128a and MAC in RTL
# Step 4, RTL: the proposal's keystream block (Grain-128a, MAC, frame counter, chip nonce; the block placed and routed in
# Section X) driven by the victim's transistor-level bitstream and the key of step 2, in Icarus Verilog; checked
# against the model's Grain-128a, which matches the published test vectors (Section V-B). The chip nonce N_c is
# the model's, from the raw field read of step 1 (IV = 1 | N | ctr | N_c)
# Live (Icarus Verilog). Inputs: R (rtl_flow, the block's Verilog), KEY_M (step 2), W_RD (step 1), D_A (the victim's
# transistor-level bits of step 3), NONCE, CTR, P_sys["chip_id"]. Outputs: CN_M (chip nonce), RTL_M (the RTL's encrypted ID,
# ciphertext C, keystream and tag), KS_M (the model's keystream), C_ref, EID_ref, TAG_ref (the model's frame).
t0 = time.time()
# N_c = 31 bits of SHA-256("Nc" | w' | session index 0), from the raw read before BCH correction; the chip hashes 64
# reads (nonce_reads), of which only this one is simulated at transistor level; N_c is an input of the block
CN_M = chip_nonce(W_RD, 0)
# the testbench shifts in the key (k_0 first), N and N_c (MSB first), pulses start ctr + 1 times to reach the counter,
# waits for the 128 + 32 initialisation clocks, then XORs the 16-bit chip ID and D with the keystream and feeds the MAC
RTL_M = MS.keystream_rtl(R, KEY_M, NONCE, CTR, P_sys["chip_id"], D_A, str(MIX / "rtl"), str(PNR_BIN / "iverilog"), str(PNR_BIN / "vvp"),
                         cn=CN_M)
# the reference: Grain-128a in authentication mode (IV_0 = 1): the encryption stream is every second pre-output bit
# from y_64 on, the MAC stream the bits in between; ID_BITS + N_CLOCKS encryption bits (16 for the ID, then one per clock)
KS_M = keystream_nl(KEY_M, iv_bits(NONCE, CTR, CN_M), ID_BITS + N_CLOCKS)
ID_M = int_to_bits(P_sys["chip_id"], ID_BITS)
# the model's frame: encrypted ID = ID xor z_0..z_15, C = D xor z_16..; MAC over counter (32) | encrypted ID (16) | C
C_ref = D_A ^ KS_M["enc"][ID_BITS:]; EID_ref = ID_M ^ KS_M["enc"][:ID_BITS]
TAG_ref = grain_tag(mac_msg(int_to_bits(CTR, 32), EID_ref, C_ref), KS_M)
TIM_M["rtl"] = time.time() - t0
card("Step 4, RTL: encryption and MAC in the generated hardware",
     [("IV = 1 ‖ N ‖ ctr ‖ N<sub>c</sub>", f"N = {NONCE:08x}, ctr = {CTR}, chip nonce N<sub>c</sub> = {CN_M:08x} (31 bits, from the step-1 read)"),
      ("ciphertext C = D ⊕ keystream", f"{N_CLOCKS:,} bits; equal to the model bit for bit: <b>{np.array_equal(RTL_M['C'], C_ref)}</b>"),
      ("encrypted chip ID", f"equal to the model: <b>{np.array_equal(RTL_M['enc_id'], EID_ref)}</b>"),
      ("MAC tag over counter | encrypted ID | C", f"RTL {RTL_M['tag']:08x}, model {TAG_ref:08x}: <b>{RTL_M['tag'] == TAG_ref}</b>"),
      ("run time (Icarus Verilog)", f"{TIM_M['rtl']:.1f} s")], tone="ok" if RTL_M["tag"] == TAG_ref else "bad")
````

**Key insights.** The 1,000-bit ciphertext and the encrypted chip ID equal the model's bit for bit, and the two tags agree (the result panel prints both; the tag changes from run to run with the bits it authenticates), as predicted. **The RTL of the block that Section X prices computes the same frame as the model that Section IX attacks**, for one 1,000-bit frame, one key and one counter value; the security results of that section carry over to the hardware as far as this test and the test vectors of Section X cover its behaviour.

### F. Step 5: the server, an eavesdropper and the replica attacker

The expected behaviour of each party, and the limit of a short run, are as follows:

* *Server.* Since step 2 regenerated the enrolled key, the tag verifies and $D_j = C_j \oplus z_{16+j}$ recovers the victim's bits exactly.
* *Forger.* A frame with one ciphertext bit changed carries the old tag, which matches only with probability $2^{-32}$.
* *Eavesdropper.* Its bit is right exactly where its keystream bit equals the true one, whatever the modulator bits; a wrong key gives an unrelated keystream, so this happens with probability 1/2, with a standard deviation of $\sqrt{0.25/1000} \approx 1.6$ % over 1,000 bits.
* *Replica attacker* (A4 of Section IX-A). With the victim's ciphertext $C_A = D_A \oplus K$ (victim's modulator bits $D_A$, keystream $K$) and the replica's bits $D_B$, $C_A \oplus D_B = K \oplus (D_A \oplus D_B)$ holds 64 true keystream bits in any *clean window*, 64 consecutive clocks on which the chips decide alike: **one clean window yields the original design's key, and none yields anything against Grain-128a**.
* *Limit of a short run.* For N = 1,000 the window starts (clock 200 to N − 777, Simulation setup) take only 24 positions, 200 to 223: the 80 attempts lie on at most 24 heavily overlapping windows within clocks 200 to 286, one stretch of the run. Below 2,000 clocks the result panel labels the result illustrative; step 6 is the actual test.

The analysis above predicts, as the step's pass criterion, that the server accepts the frame and recovers the bits exactly, that the forgery is rejected, that the eavesdropper fails the MAC check and agrees with the true bits in about 50 % (±1.6 %), and that **the replica attack on the proposed design recovers nothing**. On the original design it succeeds if and only if the stretch near clock 200 is clean, which one short run cannot predict statistically.

**The step is behavioural (Python)**: the model's receiver and MAC check on the RTL frame, the eavesdropper and forger of the Simulation setup, and the replica's bits of step 3 against both designs (the original design keyed by its 16-bit LFSR); the clean-window fraction is measured from clock 200 to the end of the run. Fig. 78 shows the outcomes that the criterion distinguishes: in (a) the CIC output decoded with the enrolled key and with a wrong key, against the input, and in (b) the tries of the replica attack that recover each design's key.

Verifies, decrypts and decodes the frame with the server's key and with an eavesdropper's random key; 2. checks a frame with one ciphertext bit changed; 3. runs the replica attack (80 windows) on both designs and counts the distinct windows; 4. draws both readings and the attack's result.

````python
#@title Fig.: mixed level, step 5: receiver and attackers
# Step 5, behavioural: the server receives the frame, verifies the MAC with the key stored at enrolment, decrypts and
# decodes; an eavesdropper without the key; and the replica attack of Section VIII-B with the replica's
# transistor-level bits, against the original design (LFSR-16) and against the proposal (Grain-128a)
ok_rx, _, ks_rx = mac_verify(KEY_SRV, NONCE, CTR, RTL_M["enc_id"], RTL_M["C"], RTL_M["tag"], cn=CN_M)   # N_c read from the frame
D_rx = RTL_M["C"] ^ ks_rx["enc"][ID_BITS:]
key_eve = np.random.default_rng(99).integers(0, 2, 128).astype(np.uint8)
ok_eve, _, ks_eve = mac_verify(key_eve, NONCE, CTR, RTL_M["enc_id"], RTL_M["C"], RTL_M["tag"], cn=CN_M)
D_eve = RTL_M["C"] ^ ks_eve["enc"][ID_BITS:]
C_forged = RTL_M["C"].copy(); C_forged[N_CLOCKS // 2] ^= 1                     # one bit of a forged reading
ok_forged = mac_verify(KEY_SRV, NONCE, CTR, RTL_M["enc_id"], C_forged, RTL_M["tag"], cn=CN_M)[0]
K16_M = lfsr16_bits(DEFAULTS["key16"], ID_BITS + N_CLOCKS)[ID_BITS:]
# window_attack draws its 80 window starts from clocks 200 ... 199 + N_POS (it keeps 976 clocks for the 64-bit window, its
# 512-bit check and a margin): a short run offers few, overlapping windows, so the attack below is only illustrative there
N_POS = max(0, min(N_CLOCKS, 30000) - 976)
_rw = rng_for(100); WIN_M = len({int(_rw.integers(N_POS)) for _ in range(80)}) if N_POS else 0   # distinct windows of the 80
SHORT_M = N_CLOCKS < 2000
if N_POS:
    atk_orig = window_attack(D_A ^ K16_M ^ D_B, K16_M, 100)                    # original: C = D_A ^ K16; the attacker has D_B
    atk_prop = window_attack(RTL_M["C"] ^ D_B, KS_M["enc"][ID_BITS:], 100)    # proposal: C = D_A ^ Grain keystream
else:                                                                           # too short for one window and its check
    atk_orig = atk_prop = dict(broken=0, tries=0, first=-1)
atk_note = ("" if not SHORT_M else f" (illustrative: the {atk_orig['tries']} tries fall on {WIN_M} distinct, overlapping windows "
            f"starting at clocks 200 to {199 + N_POS}; step 6 is the test)" if N_POS else " (skipped: the run is shorter than 977 clocks)")
x_ = (D_A ^ D_B)[200:].astype(int); c_ = np.concatenate([[0], np.cumsum(x_)]); f64_m = float(np.mean(c_[64:] - c_[:-64] == 0))
card("Step 5: the server, the eavesdropper and the replica attacker",
     [("server: MAC check with the enrolled key", f"<b>{'accepted' if ok_rx else 'REJECTED'}</b>; bits recovered exactly: {np.array_equal(D_rx, D_A)}; "
       f"decoded reading {reading(D_rx):+.4f} (input {y_true:+.4f})" if INPUT == "constant" else f"<b>{'accepted' if ok_rx else 'REJECTED'}</b>"),
      ("a forged frame (one ciphertext bit changed)", f"MAC {'accepted' if ok_forged else '<b>rejected</b>'}"),
      ("eavesdropper with a wrong key", f"MAC {'accepted' if ok_eve else 'rejected'}; the eavesdropper's bits agree with the true ones in {100 * np.mean(D_eve == D_A):.1f} %"),
      ("the two chips' transistor-level bitstreams", f"disagree on {100 * x_.mean():.1f} % of bits; clean 64-bit windows {100 * f64_m:.1f} %"),
      ("replica attack on the original design (LFSR-16)", f"<b>{atk_orig['broken']} of {atk_orig['tries']}</b> windows recover the key" + atk_note),
      ("replica attack on the proposal (Grain-128a)", f"<b>{atk_prop['broken']} of {atk_prop['tries']}</b> windows" + atk_note)],
     tone="ok" if ok_rx and not ok_forged and atk_prop["broken"] == 0 else "bad")
# (a) what the server and an eavesdropper read; (b) the replica attack with the transistor-level chips
fig, axs = plt.subplots(1, 2, figsize=(W2, 3.0), gridspec_kw=dict(width_ratios=[1.35, 1]))
ax = axs[0]
ax.plot(dec_m(D_rx), "o-", color=COL["ok"], ms=3, lw=1.3, label=f"server, enrolled key: MAC {'accepted' if ok_rx else 'rejected'}")
ax.plot(dec_m(D_eve), "x-", color=COL["bad"], ms=3.5, lw=1.0,
        label=f"eavesdropper, wrong key: MAC {'accepted' if ok_eve else 'rejected'}, {100 * np.mean(D_eve == D_A):.1f} % of bits right")
ax.plot(u_m[::OSR][:len(dec_m(D_rx))], ":", color=COL["ink"], lw=1.3, label="sensor input")
ax.set_xlabel("Output sample (OSR = 50)"); ax.set_ylabel("CIC output (full scale ±1)")
lo_r, hi_r = ax.get_ylim(); ax.set_ylim(lo_r, hi_r + 0.55 * (hi_r - lo_r))
ax.legend(fontsize=7, loc="upper left", handlelength=1.6)
panel(ax, "a", "what the server and an eavesdropper read")
ax = axs[1]
ax.bar([0, 1], [atk_orig["broken"], atk_prop["broken"]], color=[COL["bad"], COL["ok"]], width=0.55)
for x, a in ((0, atk_orig), (1, atk_prop)):
    ax.text(x, a["broken"] + 0.02 * max(1, atk_orig["tries"]), f"{a['broken']} of {a['tries']}", ha="center", va="bottom", fontsize=8.5)
ax.set_xticks([0, 1], ["original (LFSR-16)", "proposal (Grain-128a)"]); ax.set_ylim(0, 1.15 * max(1, atk_orig["tries"]))
ax.set_ylabel(f"Windows that recover the key (of {atk_orig['tries']})"); ax.grid(axis="x", visible=False)
panel(ax, "b", "replica attack" + (f" ({WIN_M} distinct windows)" if SHORT_M else ""))
fig.tight_layout(); plt.show()
````

### G. Step 6, transistor level: the replica attack over 30,000 clocks

Once the chips have drifted apart (Theory, point 4) they disagree on a fraction $p$ of their bits, but in bursts, so a fraction $f_{64}$ of the 64-bit windows remains clean, far more than the $(1-p)^{64}$ of independent errors. With 80 windows drawn at random, at least one is clean with probability $1 - (1 - f_{64})^{80}$, and about $80 f_{64}$ are clean. This is a lower bound on the windows that yield the original design's key, because Berlekamp-Massey also succeeds when a window's only wrong bits lie within its first $L = 16$ bits (the LFSR length) (Section VIII-B). The behavioural model differs from the circuit only in the comparator, so on the same pair it should give the same $p$ and $f_{64}$ within the spread of the comparator's noise realisation, which no two simulators can share. The analysis above predicts, as the step's pass criterion, that the transistor-level chips disagree on about a quarter of their bits with a clean-window fraction of a few per cent, within or near the behavioural spread on the same pair, **that the original design's key is recovered from at least one of the 80 windows, and that the proposed design's never is**. Fig. 79 shows this mechanism: panel (b) plots the running disagreement between the two chips' bits with the clean 64-bit windows marked, and panel (a) the start clocks of the 80 windows tried, so the windows that yield the original design's key can be set against the clean stretches between the bursts.

The step runs the chips of step 3 (own dither and noise) at tt, 37 °C and constant 0.5 µA, 30,000 clocks per chip, with the same segments and `reltol`; the 80 window starts, drawn from clock 200 to 29,223, are distinct and spread over the capture. As the run exceeds a Colab session, the cell shows the recorded run of `dev/system/long_replica.py` (same code and seeds, embedded in the notebook), and `LONG_RUN = True` re-runs it at the corner and temperature of the form. **The recorded run was made before the correction of the first-integrator noise (Section VII-E): its chips carry twice the corrected integrator noise, and it is not re-recorded.** Every 30,000-clock number in this section refers to this setting.

*Behavioural run on the same pair (live).* The loop of `mixedsim.loop_reference`, in the netlist's order, with the same input, each chip's dither, the same two first-integrator noise realisations at the recorded setting, each chip's offset measured in step 3 and the $\sigma_\mathrm{n}$ of this corner, repeated with 20 comparator-noise seeds. Metrics for both levels: the disagreement p, $f_{64}$ (from clock 200), and the windows that yield the key among the same 80 and over the whole capture. **The comparison holds only at the recorded run's tt, 37 °C**; the result panel states when the form selects another condition.

> **Next cell.** 1. Loads the two chips' 30,000-clock bitstreams from the recorded run (or simulates them with `LONG_RUN = True`); 2. measures disagreement and clean windows; 3. runs the replica attack (80 windows) on both designs; 4. runs the behavioural model on the same pair (20 comparator-noise seeds) and compares the two levels; 5. draws the windows and the agreement map. *Recorded by default, so that the whole notebook stays within 20 minutes on free Colab; to repeat it live, tick `LONG_RUN` in the form beside the code and run the cell (about 70 minutes on two cores); the result panel then says "this session" instead of "recorded run".*

````python
#@title Fig.: mixed level, step 6: replica attack, 30,000 clocks
# Step 6, transistor level: the replica attack over 30,000 clocks, the length of the behavioural attack runs of Section
# VIII-B. The same two chips as step 3 (tt, 37 °C, constant input). Over an hour on two cores (72 min recorded), so the run of
# dev/system/long_replica.py (same code, same seeds) is shown; LONG_RUN re-runs it in this session
LONG_RUN = False                        #@param {type:"boolean"}
if LONG_RUN:
    t0 = time.time(); NL = 30000; CL = [ccfg_m("long_victim"), ccfg_m("long_replica")]
    S_L = cs.channel_noise_psd(CL[0], cs.reference_trajectory(CL[0]))
    u_l = np.full(NL, s2d(tia(0.5e-6, P_sys["rtia"]), P_sys["buffer"]))
    d_l = [dsm2(u_l, P_sys)["DI"].astype(float), dsm2(u_l, P_sys, seeds=REPLICA_DITHER)["DI"].astype(float)]
    n1_l = [P_sys["int_noise"] * np.random.default_rng(31 + i).standard_normal(NL) for i in range(2)]
    with cf.ThreadPoolExecutor(2) as ex:
        ML = [f.result() for f in [ex.submit(MS.modulator, cs, CL[i], u_l, d_l[i], n1_l[i], CHIPS[i], S_L, 250, None, 0) for i in range(2)]]
    TIM_M["long"] = time.time() - t0
    DLA, DLB, src_l = ML[0]["D"], ML[1]["D"], f"simulated in this session ({TIM_M['long']:.0f} s)"
else:
    LR = recorded("long_replica")
    DLA, DLB = (np.frombuffer(LR[k].encode(), np.uint8) - 48 for k in ("D_victim", "D_replica"))
    TIM_M["long"] = LR["seconds"]
    src_l = f"recorded run of dev/system/long_replica.py ({LR['seconds'] / 60:.0f} min on two cores)"
    LR_NOISE = LR.get("int_noise", 1.798e-05)          # records before the noise correction carry no field: 2x the value of [1]
    if abs(LR_NOISE - P_sys["int_noise"]) > 0.01 * P_sys["int_noise"]:
        src_l += (f"; recorded with the earlier first-integrator noise setting ({LR_NOISE:.2e} at the integrator output, "
                  f"{LR_NOISE / P_sys['int_noise']:.1f}× the corrected {P_sys['int_noise']:.2e}); LONG_RUN = True re-runs it "
                  "with the corrected value")
NL = len(DLA)
K16_L = lfsr16_bits(DEFAULTS["key16"], ID_BITS + NL)[ID_BITS:]
ks_l = keystream_nl(KEY_M, iv_bits(NONCE, CTR), ID_BITS + NL)["enc"][ID_BITS:]
# the replica attack on the 30,000-clock transistor-level bits: the attacker's keystream guess is C xor D_replica =
# D_victim xor K xor D_replica, attacked with 80 random windows; xl marks the disagreements, clean_l the clean windows
atk_lo = window_attack(DLA ^ K16_L ^ DLB, K16_L, 100)                          # original design
atk_lp = window_attack(DLA ^ ks_l ^ DLB, ks_l, 100)                            # proposal
xl = (DLA ^ DLB)[200:].astype(int); cl_ = np.concatenate([[0], np.cumsum(xl)]); clean_l = (cl_[64:] - cl_[:-64]) == 0
# the behavioural model on the same pair: the same input, each chip's dither sequence, the same first-integrator noise
# realisations and setting as the transistor-level run (from LR_NOISE: the recorded run's own value), each chip's comparator
# offset from step 3 and the comparator noise of this corner. Only the comparator's noise realisation cannot be shared
# with ngspice, so the behavioural run is repeated with N_BEH comparator-noise seeds, which gives its spread
def key_windows(D1, D2, K):                                    # windows (start >= 200) whose Berlekamp-Massey register yields the key
    e = (D1 ^ D2).astype(np.int64); kg = D1 ^ K[:len(D1)] ^ D2; c = np.concatenate([[0], np.cumsum(e)])
    s_ = np.arange(200, len(e) - 63 - 512); cand = s_[c[s_ + 64] == c[s_ + 24]]
    return int(sum(bm_attack(kg[s0:s0 + 64], K[s0 + 64:s0 + 576])["acc"] > 0.99 for s0 in cand))
t0_b, N_BEH = time.time(), 20
NI_L = P_sys["int_noise"] if LONG_RUN else LR_NOISE
same_l = LONG_RUN or (CORNER == LR["corner"] and float(TEMP_C) == float(LR["temp"]) and tuple(CHIPS) == tuple(LR["chips"]))
u_b = np.full(NL, s2d(tia(0.5e-6, P_sys["rtia"]), P_sys["buffer"]))
d_b = [dsm2(u_b, P_sys)["DI"].astype(float), dsm2(u_b, P_sys, seeds=REPLICA_DITHER)["DI"].astype(float)]
n1_b = [NI_L * np.random.default_rng(31 + i).standard_normal(NL) for i in range(2)]
BEH = []
for r_ in range(N_BEH):
    Db = [MS.loop_reference(u_b, d_b[i], n1_b[i], -VOS_M[i] / VU_M,
                            noise=SIG_N_M / VU_M * np.random.default_rng((900 + r_, i)).standard_normal(NL)) for i in range(2)]
    xb = (Db[0] ^ Db[1])[200:].astype(np.int64); cb = np.concatenate([[0], np.cumsum(xb)])
    BEH.append(dict(p=float(xb.mean()), f64=float(np.mean(cb[64:] == cb[:-64])), b80=window_attack(Db[0] ^ K16_L ^ Db[1], K16_L, 100)["broken"],
                    kw=key_windows(Db[0], Db[1], K16X)))
BEH = {k: np.array([b[k] for b in BEH]) for k in BEH[0]}
KW_T = key_windows(DLA, DLB, K16X)                             # the same count for the transistor-level bits
TIM_M["long_beh"] = time.time() - t0_b
rank = lambda v, a: f"{int(np.sum(a < v))} below, {int(np.sum(a == v))} equal, {int(np.sum(a > v))} above"
card("Step 6, transistor level: the replica attack over 30,000 clocks",
     [("bitstreams", f"{NL:,} clocks per chip, {src_l}"),
      ("the two chips' bits", f"disagree on {100 * xl.mean():.1f} %; clean 64-bit windows {100 * clean_l.mean():.1f} % "
       f"(behavioural model of Section VIII-B at PVT, other pairs: {100 * SYS['f64_pvt']:.1f} %)"),
      ("replica attack on the original design (LFSR-16)", f"<b>{atk_lo['broken']} of {atk_lo['tries']}</b> windows recover the key"
       + (": the key is recovered" if atk_lo["broken"] else "")),
      ("replica attack on the proposal (Grain-128a)", f"<b>{atk_lp['broken']} of {atk_lp['tries']}</b> windows"),
      (f"behavioural model on the same pair ({N_BEH} comparator-noise seeds; " + ("same chips, corner, temperature, dither and "
       f"integrator noise, {NI_L:.2e} at the integrator output" if same_l else "<b>not the same pair</b>: the settings form differs from "
       "the recorded run") + ")",
       f"bits disagree on {100 * BEH['p'].mean():.1f} % ({100 * BEH['p'].min():.1f}–{100 * BEH['p'].max():.1f} %); clean 64-bit windows "
       f"{100 * BEH['f64'].mean():.1f} % ({100 * BEH['f64'].min():.1f}–{100 * BEH['f64'].max():.1f} %); the 80-window attack recovers the key "
       f"from {BEH['b80'].mean():.1f} windows ({BEH['b80'].min()}–{BEH['b80'].max()}); windows that yield the key over the whole capture "
       f"{BEH['kw'].mean():.0f} ({BEH['kw'].min()}–{BEH['kw'].max()})"),
      ("transistor level against the behavioural spread (seeds below / equal / above)",
       f"disagreement {rank(xl.mean(), BEH['p'])}; clean windows {rank(clean_l.mean(), BEH['f64'])}; 80-window successes "
       f"{atk_lo['broken']}: {rank(atk_lo['broken'], BEH['b80'])}; key windows {KW_T}: {rank(KW_T, BEH['kw'])}"),
      ("run time of the behavioural comparison", f"{TIM_M['long_beh']:.1f} s")],
     tone="ok" if atk_lp["broken"] == 0 else "bad")
# the 80 windows the attack tried on the original design (the same draws as window_attack), and which of them yield the key
_rl = rng_for(100); starts_l = [200 + int(_rl.integers(min(NL, 30000) - 64 - 512 - 400)) for _ in range(atk_lo["tries"])]
hit_l = np.array([bm_attack((DLA ^ K16_L ^ DLB)[s0:s0 + 64], K16_L[s0 + 64:s0 + 576])["acc"] > 0.99 for s0 in starts_l])
fig, (ax0, ax) = plt.subplots(2, 1, figsize=(W2, 2.9), sharex=True, gridspec_kw=dict(height_ratios=[1, 3.2], hspace=0.38))
st_l = np.array(starts_l)
ax0.plot(st_l[~hit_l], np.zeros((~hit_l).sum()), "|", color=COL["faint"], ms=9, mew=1.0, label="no key")
ax0.plot(st_l[hit_l], np.zeros(hit_l.sum()), "v", color=COL["bad"], ms=7, mec=COL["ink"], mew=0.5, label="key recovered")
ax0.set_yticks([]); ax0.set_ylim(-1, 1); ax0.grid(False); ax0.tick_params(axis="x", bottom=False)
ax0.legend(fontsize=7, loc="lower right", bbox_to_anchor=(1.0, 0.98), ncol=2, handlelength=1.0, borderpad=0.1, frameon=False)
ax0.set_title(f"(a) replica attack, {atk_lo['tries']} windows: the original design's key from {int(hit_l.sum())}, the proposal's from "
              f"{atk_lp['broken']}", loc="left", fontsize=9, pad=3)
cov = np.zeros(len(xl), bool)
for s0 in np.flatnonzero(clean_l):
    cov[s0:s0 + 64] = True
ax.fill_between(np.arange(len(xl)) + 200, 0, 1, where=cov, color=COL["ok"], alpha=0.4, lw=0, step="mid", label="clean 64-bit windows")
ax.plot(np.arange(len(xl)) + 200, np.convolve(xl, np.ones(64) / 64, "same"), color=COL["ink"], lw=0.8, label="disagreement (64-clock mean)")
ax.set_xlim(0, NL); ax.set_ylim(0, 1); ax.set_xlabel("Clock"); ax.set_ylabel("Fraction"); ax.legend(fontsize=7, loc="upper right", ncol=2)
ax.set_title(f"(b) victim and replica, transistor level: bits disagree on {100 * xl.mean():.1f} %, {100 * clean_l.mean():.1f} % of 64-bit "
             "windows clean", loc="left", fontsize=9, pad=3)
fig.subplots_adjust(left=0.07, right=0.97, top=0.9, bottom=0.17); plt.show()
````

### H. The complete sensor in one figure

The figure reuses steps 1 to 6 without re-simulating at transistor level and **adds four security tests on the frame the run actually produces**. (1) Key search on the original design: each of the 65,535 non-zero 16-bit LFSR keys decrypts the original design's ciphertext of the victim's step-3 bits and is scored by its agreement with the replica's bits, the attacker's plaintext guess under the known input. (2) The same search against the proposed design, the attacker knowing the session nonce, counter and chip nonce (sent in clear): 1,000 random 128-bit keys and the 128 keys one bit from the true key, which show whether a partly correct key is rewarded (a bit-by-bit search). (3) Forgery: every single-bit change of the RTL frame (32 counter, 16 encrypted-ID, every payload and 32 tag bits) goes to the receiver, which recomputes the MAC with the enrolled key (the original design checks only the decrypted chip ID), plus 100,000 random tags with the genuine frame. (4) Four NIST SP 800-22 tests (frequency, block frequency with 128-bit blocks, runs, cumulative sums) on the 30,000 bits of step 6 before (*D*) and after (*C*) encryption with the proposed design's keystream; pass at a p-value *p* ≥ 0.01. The spectrum of panel (e) is a Welch average of 4,096-point Blackman-Harris segments of the step-6 victim bitstream, against one behavioural run of the same chip (one comparator-noise seed). Fig. 80 places these tests beside the simulated signal chain: panels (a) to (h) follow the signal from the enrolled fingerprint to the bits of the link, panels (i) to (k) show the replica attack of steps 5 and 6, and panels (l) to (n) the four tests above, so each security result can be read next to the signals it was computed on.

> **Next cell.** 1. Collects the results of steps 1 to 6. 2. Runs the four security tests listed above. 3. Draws the figure (14 panels in signal-chain order) and prints the result panel. About 5 s.

````python
#@title Fig.: the complete sensor in one figure
# The complete sensor in one figure: every result of steps 1-6 in the order of the signal chain, from the SRAM cells to
# the attacker. Nothing is re-simulated except one behavioural loop of 30,000 clocks (about a second), the reference for
# the spectrum of the transistor-level bitstream of step 6. Names stay local to the function (the step cells' globals are
# only read)
def _mixed_overview():
    from scipy.signal import welch
    from matplotlib.patches import Rectangle
    from matplotlib.colors import ListedColormap
    t0 = time.time()
    GREY1 = "#a3a3a3"
    from scipy.special import erfc, gammaincc
    # ---- security tests on the mixed-level frame (step 3's transistor-level bits, step 4's RTL ciphertext and tag)
    # 1. brute force of the original design's 16-bit key: every key scored by how well its decryption of the victim's frame
    #    agrees with the replica's transistor-level bits (the replica attacker's prediction of D); all 65,535 keys at once
    C_o = D_A ^ K16_M                                            # the original design's ciphertext of the same bits
    # all 65,535 non-zero 16-bit LFSR states advance together, one array element per candidate key: at each clock the
    # output bit (s & 1) decrypts the ciphertext bit, a match with the replica's bit scores a point, and the state shifts
    # right with the feedback of taps 0, 2, 3 and 5 entering at bit 15 (the first ID_BITS clocks encrypt the chip ID)
    s_ = np.arange(1, 65536, dtype=np.uint32); agree_o = np.zeros(len(s_))
    for i in range(ID_BITS + N_CLOCKS):
        if i >= ID_BITS:
            agree_o += ((s_ & 1) ^ C_o[i - ID_BITS]) == D_B[i - ID_BITS]
        fb = (s_ ^ (s_ >> 2) ^ (s_ >> 3) ^ (s_ >> 5)) & 1
        s_ = (s_ >> 1) | (fb << 15)
    agree_o /= N_CLOCKS
    # the true key's score, its rank among all keys (1 = the attacker's best guess) and the scores of the wrong keys
    k_true = DEFAULTS["key16"] & 0xFFFF or 1
    rank_o = int(np.sum(agree_o > agree_o[k_true - 1])) + 1
    others = np.delete(agree_o, k_true - 1)
    # 2. the same search against the proposal: the 128 keys one bit from the true key and 1,000 random keys (the attacker
    #    knows N, ctr and N_c, which travel in clear); a correct key would stand out as for the original design
    iv_p = iv_bits(NONCE, CTR, CN_M)
    score = lambda k: float(np.mean((RTL_M["C"] ^ keystream_nl(k, iv_p, ID_BITS + N_CLOCKS)["enc"][ID_BITS:]) == D_B))
    near = np.array([score(KEY_M ^ np.eye(128, dtype=np.uint8)[j]) for j in range(128)])
    rk = np.random.default_rng(2027).integers(0, 2, (1000, 128)).astype(np.uint8)
    rand_ = np.array([score(k) for k in rk]); true_p = score(KEY_M)
    right_eve = np.array([np.mean((RTL_M["C"] ^ keystream_nl(k, iv_p, ID_BITS + N_CLOCKS)["enc"][ID_BITS:]) == D_A) for k in rk[:200]])
    # 3. forgery: every single-bit change of the frame (counter, encrypted ID, payload, tag) presented to the receiver, and
    #    100,000 random tags for the genuine frame; the original design checks only the decrypted chip ID
    ctr_b = int_to_bits(CTR, 32); eid, C0, tag0 = RTL_M["enc_id"], RTL_M["C"], RTL_M["tag"]
    acc_p = dict(counter=0, ID=0, payload=0, tag=0)
    for j in range(32):
        cb = ctr_b.copy(); cb[j] ^= 1
        acc_p["counter"] += mac_verify(KEY_SRV, NONCE, bits_to_int(cb), eid, C0, tag0, cn=CN_M)[0]
    for j in range(ID_BITS):
        e2 = eid.copy(); e2[j] ^= 1; acc_p["ID"] += grain_tag(mac_msg(ctr_b, e2, C0), ks_rx) == tag0
    for j in range(N_CLOCKS):
        c2 = C0.copy(); c2[j] ^= 1; acc_p["payload"] += grain_tag(mac_msg(ctr_b, eid, c2), ks_rx) == tag0
    acc_p["tag"] = sum(((tag0 ^ (1 << j)) == tag0) for j in range(32))
    n_fields = dict(counter=32, ID=ID_BITS, payload=N_CLOCKS, tag=32)
    acc_o = dict(counter=None, ID=0, payload=N_CLOCKS, tag=None)  # original: no counter, no tag; a changed ID fails its check
    rand_tags = int(np.sum(np.random.default_rng(7).integers(0, 2**32, 100000, dtype=np.uint64) == tag0))
    # 4. randomness of the bits on the air: NIST SP 800-22 frequency, block frequency (M = 128), runs and cumulative sums,
    #    on the 30,000-clock transistor-level bits before (D) and after (C) encryption
    def nist(x):
        x = np.asarray(x, np.int64); n = len(x); S = np.cumsum(2 * x - 1); p_ = {}
        p_["frequency"] = erfc(abs(S[-1]) / np.sqrt(2 * n))
        M = 128; Nb = n // M; pib = x[:Nb * M].reshape(Nb, M).mean(1)
        p_["block frequency"] = gammaincc(Nb / 2, 4 * M * np.sum((pib - 0.5) ** 2) / 2)
        pi = x.mean(); V = 1 + np.sum(x[1:] != x[:-1])
        p_["runs"] = 0.0 if abs(pi - 0.5) >= 2 / np.sqrt(n) else erfc(abs(V - 2 * n * pi * (1 - pi)) / (2 * np.sqrt(2 * n) * pi * (1 - pi)))
        z = np.max(np.abs(S)); from scipy.special import ndtr as _N
        k1 = np.arange(int((-n / z + 1) // 4), int((n / z - 1) // 4) + 1)
        k2 = np.arange(int((-n / z - 3) // 4), int((n / z - 1) // 4) + 1)
        p_["cumulative sums"] = float(1 - np.sum(_N((4 * k1 + 1) * z / np.sqrt(n)) - _N((4 * k1 - 1) * z / np.sqrt(n)))
                                      + np.sum(_N((4 * k2 + 3) * z / np.sqrt(n)) - _N((4 * k2 + 1) * z / np.sqrt(n))))
        return {k: float(np.clip(v, 0, 1)) for k, v in p_.items()}
    P_D, P_C = nist(DLA), nist(DLA ^ ks_l)
    t_sec = time.time() - t0
    Dbeh_l = MS.loop_reference(u_b, d_b[0], n1_b[0], -VOS_M[0] / VU_M,         # behavioural reference of the spectrum
                               noise=SIG_N_M / VU_M * np.random.default_rng((900, 0)).standard_normal(NL))

    # ---- the figure: one journal page (7.16 in wide), 3 columns x 5 rows of equal panels, the 30,000-clock trace over two
    # columns; 8 pt labels, 7 pt ticks; one meaning per colour: black the transistor level, grey dashed the behavioural model,
    # blue the proposal and the server, orange the original design and what the attacker gains. Outcomes are drawn as parts
    # of the whole (accepted and rejected, key and no key), so that a result of zero is visible
    from matplotlib.transforms import blended_transform_factory as _bt
    RC = {"font.size": 8, "axes.labelsize": 8, "axes.titlesize": 8.6, "xtick.labelsize": 7, "ytick.labelsize": 7,
          "legend.fontsize": 6.5, "legend.title_fontsize": 6.5, "axes.linewidth": 0.6, "lines.linewidth": 0.9,
          "xtick.major.size": 2.5, "ytick.major.size": 2.5, "xtick.minor.size": 1.4, "ytick.minor.size": 1.4,
          "xtick.major.width": 0.6, "ytick.major.width": 0.6, "legend.handlelength": 1.3, "legend.borderpad": 0.3,
          "legend.labelspacing": 0.2, "legend.handletextpad": 0.4, "legend.borderaxespad": 0.35, "legend.columnspacing": 0.8}
    TR, MO, PR, OR, LG = COL["ink"], COL["faint"], COL["ok"], COL["bad"], "#c3c8cf"
    with plt.rc_context(RC):
        FH, CW, AH = 9.05, W2 / 3, 1.12
        TOP = [0.24 + 1.8 * r for r in range(5)]
        fig = plt.figure(figsize=(W2, FH))
        def AX(col, row, span=1):
            x0, w = col * CW + 0.5, span * CW - 0.6
            return fig.add_axes([x0 / W2, 1 - (TOP[row] + AH) / FH, w / W2, AH / FH])
        def ttl(ax, lab, txt):
            ax.set_title(f"({lab}) {txt}", loc="center", pad=3.5)
        def headroom(ax, f):
            lo, hi = ax.get_ylim(); ax.set_ylim(lo, hi + f * (hi - lo))
        def two_level(ax, groups, xs):                     # O/P under each bar, the group's name below
            ax.set_xticks(xs, ["O", "P"] * len(groups), fontsize=6.5)
            for g, nm in enumerate(groups):
                ax.text(g, -0.2, nm, transform=_bt(ax.transData, ax.transAxes), ha="center", va="top", fontsize=7)

        # (a) key generation: fingerprint, the cells that flip in the field read, and the key
        ax = AX(0, 0)
        ax.imshow(W_EN.reshape(20, 30), cmap=ListedColormap(["white", LG]), vmin=0, vmax=1, interpolation="nearest")
        for k in np.flatnonzero(W_EN != W_RD):
            ax.add_patch(Rectangle((k % 30 - 0.5, k // 30 - 0.5), 1, 1, fc=OR, ec=TR, lw=0.5))
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        key_ok = np.array_equal(KEY_M, KEY_SRV) and np.array_equal(KG_M["bits"], REC_M["w"])
        ax.set_xlabel(f"{n_err} of {PUF_BITS} cells flip (BCH t = {t_m})\n"
                      + ("128-bit key equals the server's" if key_ok else "key NOT recovered"), fontsize=7, labelpad=3)
        ttl(ax, "a", f"SRAM-PUF fingerprint, {TEMP_C} °C")

        # (b) the victim's comparator in the loop, (c) the decision test
        ax = AX(1, 0); w = MOD[0]["waves"]; tw = 1e9 * (w["time"] - w["te"][0])
        rows = [(r"$\phi$", "v(clk)", MO), (r"$\mathrm{out}_p$", "v(c0_outp)", TR), (r"$\mathrm{out}_n$", "v(c0_outn)", TR),
                (r"$q$", "v(c0_q)", TR)]
        for k, (lab, node, c) in enumerate(rows):
            y0 = 1.3 * (len(rows) - 1 - k)
            ax.axhline(y0, color=MO, lw=0.35, alpha=0.5); ax.plot(tw, y0 + w[node] / VDD_M, color=c, lw=0.9)
        ax.set_yticks([1.3 * (len(rows) - 1 - k) + 0.5 for k in range(len(rows))], [r[0] for r in rows])
        ax.tick_params(axis="y", left=False, right=False); ax.set_ylim(-0.2, 1.3 * (len(rows) - 1) + 1.2)
        ax.set_xlim(tw[0], tw[-1]); ax.grid(axis="y", visible=False); ax.set_xlabel("Time (ns)")
        ttl(ax, "b", "Comparator in the loop")
        ax = AX(2, 0)
        xs = [np.log10(np.maximum(np.abs(c["dv"] - vo) / SIG_N_M, 1e-3)) for c, vo in zip(chk, VOS_M)]
        lo_x = min(-1.25, min(x.min() for x in xs) - 0.15); bins_c = np.linspace(lo_x, max(x.max() for x in xs) + 0.05, 45)
        ax.axvspan(lo_x, np.log10(3), color=MO, alpha=0.18, lw=0, label=r"within $3\sigma_\mathrm{n}$")
        for x, c, col, nm in zip(xs, chk, (TR, OR), ("victim", "replica")):
            ax.hist(x, bins=bins_c, histtype="step", color=col, lw=0.9, label=f"{nm}, {int(c['dis'].sum())} differ")
        ax.set_xlim(lo_x, bins_c[-1]); headroom(ax, 0.75)
        ax.set_xlabel(r"$\log_{10}(|\Delta V - V_\mathrm{os}|/\sigma_\mathrm{n})$"); ax.set_ylabel("Decisions")
        ax.legend(loc="upper left", title=f"noise predicts {n_exp:.1f}")
        ttl(ax, "c", f"Decision test, {2 * N_CLOCKS:,} decisions")

        # (d) integrator states, (e) spectrum of the 30,000-clock bitstream, (f) decoded readings
        ax = AX(0, 1)
        for m, c, nm in ((MOD[1], OR, "replica"), (MOD[0], TR, "victim")):
            ax.plot(m["x1"], m["x2"], "o", ms=1.4, mew=0, color=c, alpha=0.6, label=nm)
        ax.set_xlabel(r"$x_1$ (model units)"); ax.set_ylabel(r"$x_2$"); headroom(ax, 0.3)
        ax.legend(loc="upper left", markerscale=2.5, ncol=2)
        ttl(ax, "d", f"Integrator states, {N_CLOCKS:,} clocks")
        ax = AX(1, 1)
        for D_, c, lab, ls, lw in ((DLA, TR, "transistor level", "-", 0.7), (Dbeh_l, MO, "behavioural model", "--", 0.9)):
            f_, p_ = welch(2.0 * np.asarray(D_, float) - 1.0, fs=10e6, window="blackmanharris", nperseg=4096)
            ax.semilogx(f_[1:], 10 * np.log10(p_[1:] * (f_[1] - f_[0]) / 2 + 1e-30), color=c, ls=ls, lw=lw, label=lab)
        ax.axvline(100e3, color=MO, lw=0.6, ls=":")
        ax.set_xlim(f_[1], 5e6); ax.set_xlabel("Frequency (Hz)"); ax.set_ylabel("PSD (dBFS/bin)")
        ax.legend(loc="lower right"); ttl(ax, "e", f"Bitstream spectrum, {NL:,} clocks")
        ax = AX(2, 1); ys = dec_m(D_rx); n_o = len(ys); xo = np.arange(n_o)
        ax.plot(xo, u_m[::OSR][:n_o], ":", color=TR, lw=1.0, label="input")
        ax.plot(xo, ys, "o-", color=PR, ms=2.5, lw=1.0, label=f"server (MAC {'accepted' if ok_rx else 'rejected'})")
        ax.plot(xo, dec_m(D_beh)[:n_o], "s", color=TR, mfc="none", ms=3.6, mew=0.6, label="behavioural model")
        ax.plot(xo, dec_m(D_eve)[:n_o], "x-", color=OR, ms=2.8, lw=0.8, label=f"wrong key (MAC {'accepted' if ok_eve else 'rejected'})")
        headroom(ax, 1.0); ax.legend(loc="upper left")
        ax.set_xlabel("Output sample (OSR = 50)"); ax.set_ylabel("CIC output (FS)"); ttl(ax, "f", "Decoded reading")

        # (g) the bits of the link, (h) autocorrelation before and after encryption, (i) the replica attack's windows
        ax = AX(0, 2); s0, n_ = 200, 40; seg = slice(s0, s0 + n_)
        rows = [(r"$D$", D_A[seg], TR), (r"$K$", (RTL_M["C"] ^ D_A)[seg], COL["key"]), (r"$C$", RTL_M["C"][seg], PR),
                (r"$D_\mathrm{rx}$", D_rx[seg], TR), (r"$\hat{D}$", D_eve[seg], OR)]
        for j in np.flatnonzero(D_eve[seg] != D_A[seg]):
            ax.add_patch(Rectangle((s0 + j, -0.12), 1, 1.07, fc=OR, alpha=0.2, lw=0))
        t_ = np.arange(s0, s0 + n_ + 1)
        for k, (lab, b, c) in enumerate(rows):
            y0 = 1.3 * (len(rows) - 1 - k)
            ax.axhline(y0, color=MO, lw=0.35, alpha=0.5); ax.step(t_, y0 + 0.85 * np.r_[b, b[-1]], where="post", color=c, lw=0.8)
        ax.set_yticks([1.3 * (len(rows) - 1 - k) + 0.42 for k in range(len(rows))], [r[0] for r in rows])
        ax.tick_params(axis="y", left=False, right=False); ax.set_ylim(-0.25, 1.3 * (len(rows) - 1) + 1.1); ax.grid(False)
        ax.set_xlim(s0, s0 + n_); ax.set_xlabel("Clock"); ttl(ax, "g", "Bits of the link (RTL)")
        ax = AX(1, 2); Lg = np.arange(1, 33); C_l = DLA ^ ks_l
        def acf(x):
            x = 2.0 * np.asarray(x, float) - 1.0; x = x - x.mean()
            return np.array([np.dot(x[:-l], x[l:]) / np.dot(x, x) for l in Lg])
        ax.axhspan(-2 / np.sqrt(NL), 2 / np.sqrt(NL), color=MO, alpha=0.3, lw=0)
        ax.plot(Lg, acf(DLA), "o-", color=TR, ms=2.2, lw=0.8, label="$D$, modulator")
        ax.plot(Lg, acf(C_l), "s-", color=PR, ms=2.2, lw=0.8, label="$C$, on the air")
        headroom(ax, 0.45); ax.legend(loc="upper right", ncol=2)
        ax.set_xlabel("Lag (clocks)"); ax.set_ylabel("Autocorrelation"); ttl(ax, "h", "Before and after encryption")
        ax = AX(2, 2); xb = []
        for g, (ao, ap) in enumerate(((atk_orig, atk_prop), (atk_lo, atk_lp))):
            for j, (at, c) in enumerate(((ao, OR), (ap, PR))):
                x = g + (j - 0.5) * 0.44; xb.append(x)
                ax.bar(x, at["broken"], 0.38, color=OR, ec=TR, lw=0.5, label="key recovered" if (g, j) == (0, 0) else None)
                ax.bar(x, at["tries"] - at["broken"], 0.38, bottom=at["broken"], color=LG, ec=TR, lw=0.5,
                       label="no key" if (g, j) == (0, 0) else None)
                ax.text(x, at["tries"] + 2, f"{at['broken']}", ha="center", va="bottom", fontsize=6.5)
        two_level(ax, (f"{N_CLOCKS:,} clocks", f"{NL:,} clocks"), xb)
        ax.set_ylim(0, 1.45 * atk_orig["tries"]); ax.set_xlim(-0.6, 1.6); ax.set_ylabel(f"Windows (of {atk_orig['tries']})")
        ax.legend(loc="upper center", ncol=2); ax.grid(axis="x", visible=False); ttl(ax, "i", "Replica attack")

        # (j) victim against replica over the 30,000-clock capture, (k) transistor level against the behavioural spread
        ax = AX(0, 3, span=2)
        cov = np.zeros(len(xl), bool)
        for s in np.flatnonzero(clean_l):
            cov[s:s + 64] = True
        ax.fill_between(np.arange(len(xl)) + 200, 0, 1, where=cov, color=OR, alpha=0.22, lw=0, step="mid", label="clean 64-bit window")
        ax.plot(np.arange(len(xl)) + 200, np.convolve(xl, np.ones(64) / 64, "same"), color=TR, lw=0.55, label="disagreement, 64-clock mean")
        ax.axhline(1.0, color=MO, lw=0.4)
        ax.plot(st_l[~hit_l], np.full((~hit_l).sum(), 1.07), "|", color=MO, ms=5.5, mew=0.8, label="window tried")
        ax.plot(st_l[hit_l], np.full(hit_l.sum(), 1.07), "v", color=OR, ms=4.5, mec=TR, mew=0.4, label="key recovered")
        ax.set_xlim(0, NL); ax.set_ylim(0, 1.13); ax.set_yticks([0, 0.5, 1.0]); ax.set_xlabel("Clock"); ax.set_ylabel("Fraction")
        ax.legend(loc="upper right", bbox_to_anchor=(1.0, 0.92), ncol=2)
        ttl(ax, "j", f"Victim against replica over {NL:,} clocks (recorded)")
        ax = AX(2, 3); rj = np.random.default_rng(5)
        mets = [("bits differ", BEH["p"], xl.mean()), ("clean win.", BEH["f64"], clean_l.mean()), ("key win.", BEH["kw"], KW_T)]
        for k, (lab, beh, tl) in enumerate(mets):
            m_ = beh.mean()
            ax.plot(k + rj.uniform(-0.14, 0.14, len(beh)), beh / m_, "o", ms=2.3, mew=0, color=MO,
                    label=f"model, {N_BEH} seeds" if k == 0 else None)
            ax.plot(k, tl / m_, "D", ms=4.2, color=TR, label="transistor level" if k == 0 else None)
        ax.axhline(1, color=MO, lw=0.6, ls=":"); ax.set_xticks(range(3), [m[0] for m in mets], fontsize=6.8); ax.set_xlim(-0.5, 2.5)
        headroom(ax, 0.5); ax.set_ylabel("Ratio to model mean"); ax.legend(loc="upper left"); ax.grid(axis="x", visible=False)
        ttl(ax, "k", "Transistor level vs. model")

        # (l) key search on both designs, (m) single-bit forgeries, (n) randomness of the bits on the air
        ax = AX(0, 4); bins = np.linspace(0.40, 1.0, 121); allp = np.r_[rand_, near]
        ax.hist(others, bins=bins, weights=np.full(len(others), 1 / len(others)), color=LG, label=f"original, {len(others):,} keys")
        ax.hist(allp, bins=bins, weights=np.full(len(allp), 1 / len(allp)), histtype="step", color=PR, lw=1.0,
                label=f"proposal, {len(allp):,} keys")
        ax.axvline(agree_o[k_true - 1], color=OR, lw=1.2)
        ax.text(agree_o[k_true - 1] - 0.015, 2e-3, f"original's key,\nrank {rank_o} of 65,535", ha="right", va="center",
                fontsize=6.5, color=OR)
        ax.set_yscale("log"); ax.set_xlim(0.4, 1.0); ax.set_ylim(1e-5, 8)
        ax.set_xlabel("Agreement with the replica's bits"); ax.set_ylabel("Fraction of keys"); ax.legend(loc="upper left")
        ttl(ax, "l", "Key search")
        ax = AX(1, 4); fl = list(n_fields); xb = []
        for k, f in enumerate(fl):
            for j, acc in enumerate((acc_o, acc_p)):
                x = k + (j - 0.5) * 0.44; xb.append(x); v = acc[f]
                if v is None:
                    ax.bar(x, 100, 0.38, color="white", ec=MO, lw=0.5, hatch="////", label="field absent" if k == 0 else None)
                    continue
                a_ = 100 * v / n_fields[f]
                ax.bar(x, a_, 0.38, color=OR, ec=TR, lw=0.5, label="accepted" if (k, j) == (2, 0) else None)
                ax.bar(x, 100 - a_, 0.38, bottom=a_, color=PR, ec=TR, lw=0.5, label="rejected" if (k, j) == (2, 1) else None)
        two_level(ax, fl, xb)
        ax.set_ylim(0, 138); ax.set_yticks([0, 50, 100]); ax.set_xlim(-0.6, len(fl) - 0.4); ax.set_ylabel("Forgeries (%)")
        h_, l_ = ax.get_legend_handles_labels(); o_ = [l_.index(n) for n in ("accepted", "rejected", "field absent")]
        ax.legend([h_[i] for i in o_], [l_[i] for i in o_], loc="upper center", ncol=3); ax.grid(axis="x", visible=False)
        ttl(ax, "m", "Single-bit forgeries")
        ax = AX(2, 4); tests = list(P_D); xq = np.arange(len(tests)); FL = 1e-30
        ax.axhline(0.01, color=OR, lw=0.7, ls=":", label="$p$ = 0.01")
        ax.plot(xq - 0.13, [max(P_D[t], FL) for t in tests], "s", color=TR, ms=4, label="$D$, modulator bits")
        ax.plot(xq + 0.13, [max(P_C[t], FL) for t in tests], "o", color=PR, ms=4, label="$C$, on the air")
        ax.set_yscale("log"); ax.set_ylim(FL / 300, 3e4); ax.set_yticks([1e-30, 1e-20, 1e-10, 1e0])
        ax.set_xticks(xq, ["freq.", "block", "runs", "cusum"]); ax.set_xlim(-0.5, len(tests) - 0.5); ax.set_ylabel(r"$p$-value")
        ax.text(-0.45, 0.01 * 0.2, "fail", ha="left", va="top", fontsize=6.5, color=OR)
        ax.text(-0.45, 0.01 * 5, "pass", ha="left", va="bottom", fontsize=6.5, color=OR)
        ax.legend(loc="center left", bbox_to_anchor=(0.0, 0.45)); ax.grid(axis="x", visible=False)
        ttl(ax, "n", f"NIST SP 800-22, {NL:,} bits")
        plt.show()
    card("The complete sensor in one figure: security of the mixed-level frame",
         [("brute force of the original design's 16-bit key with the replica's transistor-level bits (" f"{N_CLOCKS:,} clocks)",
           f"true key agrees on <b>{100 * agree_o[k_true - 1]:.1f} %</b>, rank <b>{rank_o} of 65,535</b>; wrong keys "
           f"{100 * others.mean():.1f} ± {100 * others.std():.1f} % (highest {100 * others.max():.1f} %)"),
          ("the same search against the proposal (the attacker knows N, ctr and N<sub>c</sub>)",
           f"1,000 random keys {100 * rand_.mean():.1f} ± {100 * rand_.std():.1f} % (highest {100 * rand_.max():.1f} %); the 128 keys "
           f"one bit from the true key {100 * near.mean():.1f} ± {100 * near.std():.1f} % (highest {100 * near.max():.1f} %): "
           f"no partial key is rewarded; the true key itself would give {100 * true_p:.1f} %"),
          ("eavesdropper with a wrong 128-bit key (200 keys)", f"{100 * right_eve.mean():.1f} ± {100 * right_eve.std():.1f} % of the "
           "victim's bits right"),
          ("single-bit forgeries accepted, proposal (counter, ID, payload, tag)",
           f"<b>{sum(acc_p.values())} of {sum(n_fields.values()):,}</b> ({', '.join(f'{f} {acc_p[f]}/{n_fields[f]}' for f in fl)})"),
          ("single-bit forgeries accepted, original (ID, payload; it has no counter or tag)",
           f"<b>{acc_o['payload']} of {N_CLOCKS + ID_BITS:,}</b> (every payload change; an ID change fails its ID check)"),
          ("100,000 random tags for the genuine frame (proposal)", f"{rand_tags} accepted (2<sup>−32</sup> each)"),
          (f"NIST SP 800-22 tests on {NL:,} transistor-level bits: p for D | for C",
           "; ".join(f"{t} {P_D[t]:.1e} | {P_C[t]:.2f}" for t in tests)),
          ("run time of the tests and the figure", f"{t_sec:.1f} s tests, {time.time() - t0:.1f} s in total")],
         tone="ok" if sum(acc_p.values()) == 0 and rand_tags == 0 and rank_o == 1 else "bad")
    return time.time() - t0
TIM_M["overview"] = _mixed_overview()
````

**Key insights.**

* Panels (a) to (k) of Fig. 80 restate steps 1 to 6 in signal-chain order (numbers in their steps); two views are new: the transistor-level bitstream has the noise shape of the behavioural model (e), and encryption removes the structure of the modulator bits (g, h).
* Panels (l) to (n) test the frame itself. Against the original design, the key search ranks the true key first: it agrees with the replica's bits on 95.2 %, against 50.0 ± 1.6 % for the wrong keys (at most 56.4 %) (l). With the comparators as transistors and the cipher and MAC as RTL, the original design thus falls to the simplest attack, a search of its 16-bit key with the replica's bits as the plaintext guess. Against the proposed design, 1,000 random keys agree on 50.1 ± 1.6 % (at most 55.1 %) and the 128 keys one bit from the true key on 49.8 ± 1.7 % (at most 53.9 %) (l): the search gets no guidance, since a key one bit from the true key scores no better than a random key. An eavesdropper with a wrong key gets 49.8 ± 1.5 % of the bits right.
* The receiver rejects all 1,080 single-bit forgeries of the proposed design's RTL frame and all 100,000 random tags, whereas the original design accepts every one of the 1,000 payload changes (m). The ciphertext passes the four NIST tests (p = 0.06 to 0.93), which the modulator bits fail by far (p ≤ 3 × 10⁻²⁶) (n): the NIST tests and the autocorrelation detect no structure of the sensor signal in the proposed design's bits on the air.
* **On the transistor-level and RTL bits, the tests applied here find no weakness of the proposed design**; they cover one chip pair at one condition and a 1,000-clock frame, and the statistics over chips and conditions are those of Sections VIII and IX.

# **XIV. Comparison with prior work**
---
Section III motivated this work by a gap in prior work; with the security and cost of the proposal now measured, this section returns to that review. **Where does the proposal stand among the four lines of work?** Section III reviewed them: compact cryptographic engines, physical-layer and analog-domain security, silicon PUF key generators, and the S-DSM of [1]. Table 36 places the attack verdicts of Section IX and the cost of Section X against what the other approaches provide and cost. Subsection A explains how the table is read, B **why no figure of merit is scaled between processes** and how the designs are compared like for like within one, and C states what the comparison shows for security, power, transmitted bits and area.

### A. How to read the table
1. **Columns.** One per crypto engine (NanoAES [8], the SIMON engines [10] and the DTLS engine [9]) and one per family for the physical-layer and analog-domain designs [11], [12] and the PUF key generators [13]–[17], whose icons give the family's typical behaviour rather than that of every member; then the two designs of [1] (decimate-then-Trivium and the S-DSM) and this work (shaded).
2. **Upper row groups: properties** (cost, confidentiality, key protection, link integrity, evaluation), read across. A positive icon means the approach provides the property, a neutral icon in part or depending on the design, a negative icon not; a dash means not applicable or not reported. This work's security icons are the 15 verdicts of Table 21, its cost icons the results of Table 26; the prior works' icons are this work's reading of their papers.
3. **Lower row groups: figures of merit,** read down a column, in two groups. **The second group is the like-for-like comparison:** Trivium and the S-DSM of [1] re-synthesised in this work's SKY130 flow, with the same cell library, supply, clock and gate-level power simulation as the proposal and in the same two forms (Table 23), so that their differences from the proposal in area, power, energy and the proposed figure of merit, FOM = $E_\mathrm{bit}/(E_\mathrm{NAND2}\,S_\Sigma)$ (the energy per bit of the security logic $E_\mathrm{bit}$, over the library's NAND2 switching energy $E_\mathrm{NAND2}$ and the security delivered, in bits summed over the goals met, $S_\Sigma$; Section X-E), come from the designs, not from the process. The first group places every work in its own node, with its basis (measured chip, FPGA or simulation) and technology in its first two rows; the prior works are quoted as reported, except the energy per bit of [1] (its measured power over the 10 Mb/s keystream, not a figure of [1]), and this group is not ranked.

### B. A note on technology
The prior works were fabricated in 14 to 65 nm CMOS, except PUFKY [14], an FPGA implementation; this work is synthesised in SKY130, whose 1.8 V devices have a minimum channel length $L_\mathrm{min}$ = 150 nm, 2.3× that of 65 nm and 6.8× to 10.7× that of 22 and 14 nm, at a supply 1.5× the 1.2 V of [1]. Their published figures of merit therefore favour the prior works, and no number is scaled between processes, since a law such as $L_\mathrm{min} V_\mathrm{nom}^2$, with $V_\mathrm{nom}$ the nominal supply, assumes ideal scaling and in practice favours old, high-voltage nodes such as SKY130. **The designs are therefore compared like for like within one process.** The S-DSM and Trivium of [1] are re-implemented in this work's SKY130 flow, with the same cell library, supply, clock and gate-level power simulation, and in the same two forms as the proposal (one bit per clock, and radix 16 with clock gating), so that their area, power and energy ratios to the proposal involve no scaling (Table 23); the proposed figure of merit **FOM = $E_\mathrm{bit}/(E_\mathrm{NAND2}\,S_\Sigma)$** (Section X-E; lower is better), the energy per bit in units of the library's NAND2 switching energy per bit of security delivered over the goals met, is given only to these same-flow blocks, since the $E_\mathrm{NAND2}$ of the published libraries is not public. The other prior works are not re-synthesised; their numbers place them in context, with area in gate equivalents (GE, the area of a two-input NAND gate) of each library, which removes most of the node dependence, and a stated basis, measured or simulated, for every number. The open PDK makes every number reproducible; a tape-out in a smaller node is future work (Section XV-G).

### C. What the comparison shows
1. **In this work's reading of the cited papers, the proposal, with the two-temperature enrolment, is the only approach that provides every security property of the table while encrypting the modulator bits.** Crypto engines store their key and encrypt words, so the sensor must decimate first; PUF key generators protect the key but do not secure a link; the S-DSM keeps the cheapest link but fails 11 of the 15 attacks (Table 21) for three reasons: a short, linear, stored key, no nonce and no tag. The proposal resists or detects 14 at the working setting (all but A11) and all 15 with the enrolment at −40 and 85 °C (Section VIII-F).
2. **The proposal has the lowest FOM of the blocks synthesised in this flow, but costs more absolute power than the S-DSM.** Under the same low-power techniques its keystream block dissipates 64.3 µW in gate-level simulation in SKY130 at 1.8 V (73.4 µW routed), 3.6× the S-DSM and 2.0× Trivium, ratios within one process; its FOM is 2.21, against 5.09 for Trivium and 14.1 for the S-DSM, and with the cipher strength alone 5.52, just above Trivium (Section X-E). Placed unscaled on the chip of [1], its SKY130 power takes 4.5 % of the total, against 1.7 % for the S-DSM and 14.0 % for Trivium as measured (Table 27).
3. **The transmitted bits stay at the S-DSM's level, but are not the fewest possible.** At OSR 50 the proposal sends 50.04 bits per reading, the S-DSM 50 and the decimating ADC with Trivium of [1] 72–80; a design that decimates on the sensor and encrypts 16-bit Nyquist-rate words with Grain-128a + MAC sends 16.04, **3.1× fewer than the proposal** (Table 26); with the radio of [1] at one frame per minute, 28.0 µW against 9.2 µW on average (Section X-E).
4. **The area is larger than that of the smallest engines.** The low-power keystream block is 9,869 GE, against 2,090 gates for NanoAES [8] and 548–1,080 for the SIMON engines [10], and the designed parts total 47,292 GE with the decoder and hash, which work only at power-up. Those engines, however, still need a decimator and a stored key and, except for the DTLS engine [9], provide no integrity by themselves.

**The comparison places a simulated design against measured and published ones**; Section XV draws the conclusions of the study from it and from the evaluation, together with the limits of that evidence.

# **XV. Conclusion, limitations and future work**
---
This work evaluated the S-DSM of [1] against the attacker of Table 3 and replaced its stored key and unauthenticated frames with **a key regenerated from SRAM cells at every power-up, a nonce-based Grain-128a keystream and a MAC on every frame**, **without changing its analog front end**. The findings are summarised below for the security of the original design (A), the security and cost of the proposed link (B), the influence of the circuit non-idealities (C), and the specification-driven key generator and the consistency of the models across levels (D), which together answer the four questions of Section I-B with the numbers of the published run. The lessons that apply beyond this design (E), the limitations of the evidence (F) and future work (G) follow.

### A. Security of the original design
The S-DSM withstands the cryptanalyses of [1] only in the form in which [1] carries them out, and **fails 11 of the 15 attacks** of the threat model (Table 21). The model reproduces the two-channel test of [1] (Table 7), but the protection observed there stems from the decorrelation between the bits of the victim and those of the replica: a scan of the whole capture with the public LFSR polynomial, a trimmed replica, multi-frame voting and an amplitude estimate turn the same experiments into successful attacks (A4-A6, Section IX). Without the inter-chip offset, the 80-window attack of [1] already breaks 356 of 360 pairs (Section VIII-B, Fig. 49; Section IX-C, Fig. 63).

A brute-force search of the 16-bit key requires only eavesdropping. **Each of the other ten failures requires one further capability** (a few keystream bits, a chip, or transmission on the link), and six of them (key extraction, keystream reuse, spoofing, cloning, replay and bit-flipping) lie beyond the eavesdropper that [1] evaluates (Section IX). No comparator or integrator property within the accuracy budget alters this outcome (Subsection C). Moreover, since the key LFSR restarts from the key in every frame without a nonce, first-order leakage tests, outside the threat model, find the key exposed to power analysis at every update of that register (Sections IX-D and X-H).

### B. Security and cost of the proposed link
Eliminating the stored key and authenticating every frame are both achieved, at a cost in digital power. At the working setting, the proposed design resists 9 and detects 5 of the 15 attacks; there, helper-data leakage (A11) leaves 94 secret bits at the 95 % lower bound, against 192 required. With the one-time enrolment at −40 and 85 °C, it resists 10 and detects 5 (7 verdicts simulated, 4 analytic and 4 by design; Section IX, Table 21, Section VIII-F). Power analysis, outside the threat model, remains open: the leakage tests locate the proposed design's exposure in the key load and the first initialisation updates of each session or frame, where masking or a protected key load would act, and find none while it encrypts and authenticates the data (Sections IX-D and X-H).

With the same low-power techniques applied to every design, the keystream block dissipates 64.3 µW in gate-level simulation in SKY130 (73.4 µW routed), 3.6× the 18.1 µW of the S-DSM in the same flow and 4.5 % of the chip of [1] unscaled. Its figure of merit (FOM) is **2.21, against 5.09 for Trivium and 14.1 for the S-DSM**, the lowest of the blocks re-synthesised in this flow (Section X, Table 23, Table 27; requirement R1). The link sends 50.04 bits per reading at an oversampling ratio (OSR) of 50 (the S-DSM: 50), 1.44-1.60× fewer than the decimate-then-Trivium design of [1] but 3.1× more than a design that decimates on the sensor (16.04 bits); with the radio of [1] and one frame per minute, the two correspond to 28.0 µW and 9.2 µW (Table 26). The keystream block occupies 9,869 GE; with the decoder of specification D (19,128 GE), a SHA-256 core (15,097 GE) and the array, the designed parts total 47,292 GE (Table 32).

### C. Influence of the circuit non-idealities
**Circuit non-idealities change neither the accuracy of the sensor nor the insecurity of the original design at any PVT condition**, but they determine the size and enrolment of the key generator and, through ageing, how its array must be operated. Around the post-layout comparator, the SNDR stays at 60.3-60.4 dB over its 45 PVT conditions with an 88× margin in comparator noise, and the modulator requires an OTA gain of about 40 dB or more (Table 20). **No analog property protects the S-DSM**: the full-scan replica attack breaks all 360 simulated chip pairs (Fig. 49) and an offset-only replica 24 of 24 (Fig. 48); comparator noise protects only from an input-referred rms noise $\sigma_\mathrm{n}$ = 160 mV, 197× its largest PVT value, at a cost of 7.5 dB of SNDR, and neither offset, integrator noise (1,000× that of [1]) nor capacitor mismatch (standard deviation σ = 50 %) does (Section VIII).

For the key, an array of 600 cells enrolled at 25 °C cannot meet reliability and secrecy together, even at tt: $10^{-6}$ key failures from 0 to 85 °C at the 99th-percentile chip require a BCH code that corrects t = 39 errors, whereas secrecy allows at most t = 35, and at the fast-nMOS corners most errors are reproducible flips (Table 12, Table 15). **A one-time enrolment at −40 and 85 °C resolves both**: t = 17 meets $10^{-6}$ for at least 99.7 % of the modelled chips at every corner and leaves 284-352 secret bits (935 cells, 600 kept; Sections VII-E, VII-F and VIII-F).

Under an assumed NBTI law (a published law scaled to SKY130, since the PDK has no ageing model), an array that is kept powered and holds its power-up values ages its cells towards the threshold: with the nominal law, both codes lose $10^{-6}$ within 26 days (t = 45) and 23 days (t = 17), and within hours with the pessimistic law. **Writing the inverse of each read back into the array reverses this drift**, after which the key-failure probability falls with age; powering the array only while it is read stretches every age by the inverse of the duty cycle (Sections VII-G and VIII-F).

### D. Specification-driven key generation and consistency across levels
**The key generator can be derived from a specification, and the models agree across levels**, for the specifications compiled and for the chip pair simulated at mixed level. The compiler finds no design on its grid with a 25 °C enrolment that meets the target at all five corners; for an enrolment at 25 and 85 °C it selects the repetition factor r = 3 with t = 4 (1,395 cells, 194 secret bits, 8,984 GE), and it compiles specification D as sized (19,128 GE). Both decoders pass 72 of 72 RTL tests bit for bit, and D is placed and routed with 0 DRC violations in a recorded run (Table 29, Section XI). In this serial template, **the key generator, not the cipher, dominates the area**.

In the mixed-level simulation at tt and 37 °C, 7 of 600 transistor-level cells differ and the RTL decoder recovers the key, the RTL keystream block matches the model, and 3 of the 2,000 transistor-level comparator decisions differ from the threshold model, consistent with the comparator noise (2.4 expected) (Table 35). Over the recorded 30,000-clock replica attack at transistor level, **Berlekamp-Massey recovers the key of the original design from 8 of 80 windows and that of the proposed design from none**, with 5.1 % clean windows against 3.9-4.6 % for the behavioural model of the same pair (Section XII-G).

### E. Implications beyond this design
Six results of this study apply to other secure sensors, PUF key generators and open-source circuit evaluations.

*Analog differences between chips are not a key.* The protection of the S-DSM against an attacker with a replica rests on chip-to-chip analog differences; **a full scan of the capture breaks 360 of 360 pairs** over the PVT conditions, and only a comparator noise of 160 mV, 197 times the PVT maximum and at a cost of 7.5 dB of SNDR, would stop it (Section VIII-B). A physical-layer scheme that relies on analog variation for secrecy must therefore be tested against an attacker who owns a replica, trims it and scans the whole capture.

*A PUF key and a PUF nonce require opposite cells.* The key needs cells that power up the same every time, the nonce cells that do not: one read of the same array holds 9.3 bits of nonce entropy at tt and 25 °C but 1.1 bits at fs and −40 °C, so **64 reads are needed for 68 bits** (Section VIII-F). A design that draws both from one array must size each at the corner where it is weakest. Ageing pulls them further apart: the inverted write-back that keeps the key reliable for life lowers the collision entropy of one read of the median chip from 11.8 to 4.4 bits in ten years, whereas the held value raises it to 20.8 bits (Section VIII-F, tt).

*A PUF should be enrolled at both ends of its temperature range rather than at 25 °C.* With a 25 °C enrolment, no 600-cell code meets both $10^{-6}$ key failures and the secrecy budget at every corner (t = 33 needed at tt against t ≤ 27); enrolled at −40 and 85 °C, **t = 17 meets both with 284–352 secret bits** (Sections VIII-F and XI). The fs corner has the lowest error rate of all at 25 °C and the highest at 85 °C, so a room-temperature screen passes the chips that fail hot.

*Baselines should be optimised as far as the proposed design.* Under the same low-power techniques, the proposed design falls from 9.7 to 3.6 times the power of the S-DSM (Section X-D); **comparing an optimised design with baselines as first written would have overstated its cost ratio by a factor of 2.7**.

*SPICE processes should be pinned to one core each.* Single-threaded ngspice processes sharing two cores **ran up to 3.5 times slower than pinned ones, without any error or warning**; one process per core keeps the transistor-level runs of this notebook within its 20-minute budget.

*Every failed measurement should stop the run.* An interpolation that clamps outside a waveform, or a measurement that silently returns no value, **turns a failed simulation into a plausible number**. Every extraction of this notebook raises an error instead (Section VII-A).

### F. Limitations
**The evidence comes from simulation, not silicon**, and is limited in the following respects. The SRAM cell model includes thermal noise only (no 1/f, telegraph or supply noise), ageing only through an assumed BTI law (Section VII-G), and a linear temperature law fitted at −40, 25 and 85 °C (the earlier linear sizing underestimated the 85 °C error rate, 1.75 % against 3.14 %); the two-temperature enrolment is checked on 512 cells per corner. With 2.7× more noise, a 25 °C enrolment would need t = 60 and leave no secret bits (Section VII-E), so **the cell must be characterised in silicon**. The system simulations use the working setting, t = 45 with a 25 °C enrolment, which leaves 94-137 secret bits and fails A11; the security claims rest on the −40 and 85 °C key generator, whose A11 verdict is analytic. Of the 15 verdicts, 4 are analytic and 4 hold by design; three of the failures of the S-DSM (A6 and A15, an LFSR that restarts from the key every frame; A7, a key in non-volatile memory) rest on assumptions that [1] does not state, and none of the verdicts is a formal proof of security. The model of the S-DSM simplifies its dither, and its protection against the attacks of [1] rests on an assumed 0.1 % FS offset between the inputs of the two chips (one seed).

The integrators, the TIA and the buffer are behavioural; the mixed-level simulation covers one chip pair at tt and 37 °C live, and its 30,000-clock replica attack is a recorded run made with twice the corrected first-integrator noise (Section XII-G). Digital power is obtained from gate-level simulation at the typical corner; routing and sign-off are recorded runs, with 16 antenna violations open on the keystream block and 8 on the decoder of D; the BCH-only designs are estimated rather than synthesised; the nonce sequencer, the SRAM periphery, the MAC delay line and the downlink receiver are not designed, and the energy of the key generator is not computed. **No radio is designed or simulated**: the designs are compared in transmitted bits per reading, and the radio of [1] appears once, as a labelled example (Section X-E). Only the two designs of [1] are re-synthesised in this flow, compared like for like and ranked with the proposed design by the figure of merit; the other prior works are quoted as published, in their own 14–65 nm nodes (Section XIV). Finally, side-channel exposure is measured, on the register-level model and on the placed netlists (Sections IX-D and X-H), but not countered; invasive and fault attacks on the chip, PUF cloning [21], remanence read-out [22] and jamming are beyond the scope of this work.

### G. Future work
**Five extensions follow from these limitations**. First, a transistor-level front end, with SKY130 integrators and TIA, would complete the mixed-level simulation of Section XII. Second, a noisier cell model, with 1/f and random telegraph noise in the SRAM-PUF cell and its ageing measured on silicon, would allow the compiler of Section XI to be re-run against it and dark-bit masking to be evaluated as an alternative to the two-temperature enrolment. Third, the compiled key generator, specification D with its array, should be enrolled and reconstructed from transistor-level cells in the mixed-level simulation, at the fast-nMOS corners where the 25 °C design fails. Fourth, the complete security subsystem (the key-reconstruction unit, a synthesised lightweight hash, the MAC delay line and the keystream block) should be placed and routed in one layout, with RC extraction and the power of every block. Finally, **a tape-out in a node comparable to the prior works (65 nm or below)** would allow the power to be measured rather than scaled.

**Source.** The notebook is self-contained. The bit-level model of both designs and the 15 attacks are defined in the cells of Section V; the larger modules are written into the session by `%%writefile` cells, where they can be read: the comparator and SRAM-PUF SPICE testbenches and the layout generators (`comparator_spice.py`, `puf_spice.py`, `sky130_layout.py`, `puf_layout.py`, `comparator_layout.py`; Section VII), the RTL, synthesis and place-and-route flows (`rtl_flow.py`, `pnr_flow.py`; Section X), the key-generator compiler (`keygen.py`; Section XI) and the mixed-level engine (`mixedsim.py`; Section XII). The open-source tools are installed by the setup cell of Section VII-A.

**Recorded runs.** These simulations are too long for a Colab session and are shown from recorded runs of the same flows and seeds; each can be re-run (the 30,000-clock replica run, recorded with twice the corrected first-integrator noise, Section XII, is not reproduced exactly): the 1,024-cell SRAM-PUF Monte Carlo runs (`PUF_MODE = "full"`; the live 128-cell run is checked bit for bit against the record), the post-layout Monte Carlo checks of both cells (`PEX_LIVE = True`), the PVT grids of both cells (`dev/spice_comparator/demo_pvt.py` and `dev/spice_puf/demo_pvt.py` in the source repository), the SRAM-PUF mechanism checks with floating or grounded bit lines and the supply-ramp sweep (`dev/spice_puf/demo_mechanism.py`), the SRAM-PUF ageing study with its assumed BTI law (`dev/spice_puf/demo_ageing.py`, summarised by `dev/spice_puf/ageing_report.py`; it has no switch, since a re-run in the session would only repeat the same assumption), the clock-gating study of the keystream block, two routings of the gated and ungated block (`CLOCK_GATING_LIVE = True`; recorded by `dev/rtl/record_signoff.py`), the detailed routing of the compiled key-reconstruction unit (`ROUTE_COMPILED = True`), the detailed routing, sign-off (Magic DRC, antenna, LVS), OpenRCX extraction, three-corner timing and routed power of the low-power keystream block (`ROUTE_BLOCK = True`) and the 30,000-clock transistor-level replica attack (`LONG_RUN = True`). The data of every recorded run is one JSON file in `data/` of the public figures repository, at a fixed commit, which the cell that shows the run loads (`recorded()`), so that no code cell carries its record inline. The recorded SRAM-PUF runs also contain power-ups at 125 °C, above the specified 0–85 °C range; they are not used, and the cell is characterised at −40, 25 and 85 °C (`dev/spice_puf/refit_0_85.py`). All other results are computed in the session.

**Explorer.** The explorer of Section XIII is the first author's own open-source tool ([nithinpuru.github.io/secure_sensor_with_puf](https://nithinpuru.github.io/secure_sensor_with_puf/)).
