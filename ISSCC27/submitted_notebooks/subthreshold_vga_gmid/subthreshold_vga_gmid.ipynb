## 1. Introduction

*Why we built this.* My teammate Vrinda and I started this as a course project and as a fun challenge: how accurate can
table-based flows like gm/ID and C/ID really be? I expected sizing the transistors to be the hard part, but the flow made
it far easier than I thought. The hardest part was finding out that the 9T core's gain hits a floor, so it can never reach

first analog flow. It also convinced me (Ayush) that analog design is where I want to work. Next, we would like to build
the envelope detector and the gain loop and, if we can, tape the core out on a shuttle run.

### 1.1 Problem
A hearing-aid compressor fits the wide dynamic range of everyday sound into the narrow range an impaired ear can still
use. Its gain element sets the noise floor and much of the battery budget, and subthreshold operation suits it:
$g_m/I_D$ peaks in weak inversion, at $\approx 1/(nU_T)$.

Analog subthreshold compression is old: complete hearing-aid and bionic-ear processors [7], [13], compressor patents
[12], and SKY130 audio front ends [8]. The circuit function is not new, and we do not claim it is.

### 1.2 What this work contributes
The earlier Code-a-Chip lookup-table flows [2], [3] showed that tables can size a circuit from its specification. A
VLSI'26 award-winning entry [14] builds tables of $R_{on}/g_m$ for inverter-based dynamic amplifiers in IHP SG13G2: a
different circuit class and process, and its notebook has no layout. This work asks what it takes for one sizing and
one layout to hold at every corner. In short: device tables made exact by building every
transistor from whole fingers (§5–§6); a flow that says when no sizing exists, and why (§7); a designer's fix sized by
that flow at every corner (§8–§9); and the first layout's wiring fed back into the sizing (§9.3, §13). Every claim is a
test (§10, ledger in §12), the same code runs other specifications (§11), and the defects found along the way stay in
the notebook (§4.3, §9.2, §9.3).

## 3. Circuit and analytical model

### 3.1 Topology
Figure 1. VGA core topology

*Figure 1. Nine-transistor current-steering VGA core. M1–M5: current-mirror OTA; M6–M9: steering stage.*

<details><summary>Node list (4 rows)</summary>

| Node | Pins | | Node | Pins |
|:--|:--|:--|:--|:--|
| `vdd` | M1.S, M4.S, M7.D, M8.D, **PMOS wells** | | `net4` (mirror) | M1.D, M1.G, M6.D, M4.G |
| `net5` (tail) | M2.S, M3.S, M5.D | | `v_out` | M4.D, M9.D |
| `net2` (node A) | M2.D, M6.S, M7.S | | `vctrl_p` / `vctrl_n` | M6.G, M9.G / M7.G, M8.G |
| `net3` (node B) | M3.D, M8.S, M9.S | | NMOS bodies | ground |

</details>

The input pair converts $v_{id}$ to a differential current. Steering pairs split each branch current between an output
path (M6 → mirror, M9 → output) and a dump path to $V_{DD}$ (M7, M8). M1/M4 mirror the A side onto `v_out`, so the
stage is single-ended and non-inverting. A diode-loaded alternative was rejected: in weak inversion its gain is capped
at $g_{mn}/g_{mp}$ whatever the sizing (4.5 dB on an early build).

### 3.2 Steering law
M6 and M7 share a source, so in weak inversion

$$
\alpha(V_{ctrl}) = \frac{I_6}{I_6+I_7} = \frac{1}{1+e^{-(V_{ctrl}-V_{off})/(nU_T)}},
$$

and since $g_m \propto I_D$, the small-signal split equals the DC split ($V_{off}$ is a small offset from the unequal
drain voltages of M6 and M7). For $V_{ctrl} \ge 0$, $\alpha$ only rises from ≈0.5 to 1, so steering alone changes the
gain by at most ≈6 dB.

### 3.3 Gain model
$$
A_v = \alpha \, g_{m2}\, R_{out}, \qquad
R_{out} = \Big[\, r_{o9} + R_B + g_{m9}\, r_{o9}\, R_B \,\Big] \,\Big\|\, r_{o4}, \qquad
R_B = r_{o3} \,\|\, \frac{1}{g_{m8}} .
$$

M9's source is the active node B, so M9 is a cascode on M3. As $V_{ctrl}$ rises, M8 cuts off and $R_B$ rises, and the
cascode boost with it: the control voltage also raises $R_{out}$. That is why the measured range exceeds the 6 dB
steering bound, and why the textbook estimate $\alpha g_{m2}(r_{o4}\|r_{o9})$ fails.

### 3.4 Gain floor
For $V_{ctrl} < 0$, M8 is fully on and $R_{out} \approx r_{o9}\|r_{o4}$. Both output devices carry $\alpha I_{branch}$, so
$r_o \propto 1/\alpha$ while the delivered transconductance $\propto \alpha$. The two cancel and the gain flattens to a
floor that no sizing removes (measured in §4.2).

### 3.5 Target specification
**Figures of merit.** The call names gain, bandwidth, power, noise and linearity. Table 2 sets a target for each (gain as
control range, power as supply current, linearity as the largest swing at 1 % THD) and adds headroom. Table 23 scores
them: range, bandwidth, headroom and current at the worst of the 24 corners, noise and NEF at tt. NEF is the one combined
figure, input noise per unit of supply current [15]:

$$\mathrm{NEF} = v_{n,\mathrm{rms}}\sqrt{\frac{2\,I_{tot}}{\pi\,U_T\cdot 4kT\cdot \mathrm{BW}}}$$

with $v_{n,\mathrm{rms}}$ the integrated input noise, BW the 20 Hz–20 kHz band and $I_{tot}$ the whole supply current.

*Table 2. Targets, fixed before characterisation unless marked revised.*

| Parameter | Target | Rationale |
|:--|:--:|:--|
| Supply $V_{DD}$ | 1.8 V | SKY130 nominal |
| Total current $I_{tot}$ | ≤ 5 µA | Our budget for the core. For scale, published subthreshold AGC loops: < 1.8 µW at 0.6 V [9], < 36 µW at 2.8 V [11] |
| Gain-control range $\Delta A_v$ | ≥ 30 dB | WDRC compression range (single stage); *revised:* monotonic, and counted only over gain steps that meet the bandwidth target (§7) |
| $f_{-3\,dB}$ at **every** gain setting, $C_L = 1$ pF | ≥ 10 kHz | Audio band; $C_L$ = assumed next-stage input |
| Input noise density at 1 kHz | ≤ 50 nV/√Hz | Our target, not taken from a source; it sets the integrated target below |
| Integrated input noise, 20 Hz–20 kHz | ≤ 7 µV$_{rms}$ | 50 nV/√Hz × √20 kHz |
| NEF | ≤ 5 | Our target (defined above). For scale: 4.0 for a micropower neural amplifier [16] |
| Linearity | *revised:* report max output swing at THD ≤ 1 % | The original "THD ≤ 1 % at 10 mV$_{pp}$ input" is unreachable by construction at ≥ 40 dB gain (≥ 1 V$_{pp}$ output) |
| Device headroom $V_{DS}$ (M2, M3, M5, M6, M9) | ≥ 130 mV at **every** control step | *added* after the defect in §4.3 (≈5 $U_T$, weak-inversion saturation); *revised* from "both control endpoints" after the flow found the input pair's minimum mid-span (§9.2) |

### 4.2 Gain-control characteristic and the gain floor

````python
#@ VERIFY #@ FIG
VGRID = np.round(np.arange(-0.5, 0.5001, 0.025), 4)
G = nu.cached("gain_curve_baseline",
              lambda: dict(vctrl=VGRID, av=vf.gain_curve(dict(BASE), VGRID)), RERUN_SPICE)
gv, ga = np.array(G["vctrl"]), 20*np.log10(np.array(G["av"]))

def monotonic_span(v, a_db):
    # largest interval around V_ctrl = 0 on which gain rises monotonically with V_ctrl
    i0 = int(np.argmin(abs(v))); lo = hi = i0
    while lo > 0 and a_db[lo-1] < a_db[lo]: lo -= 1
    while hi < len(v)-1 and a_db[hi+1] > a_db[hi]: hi += 1
    return lo, hi
lo, hi = monotonic_span(gv, ga)
floor_db = float(np.median(ga[(gv >= -0.35) & (gv <= -0.05)]))

# steering-only prediction, anchored at the measured maximum-gain point
UT = vf.K_B*300.15/vf.Q
v_off = -N_REF*UT*math.log(b0["alpha"]/(1-b0["alpha"]))            # offset implied by alpha(0)
alpha = lambda v: 1/(1+np.exp(-(v - v_off)/(N_REF*UT)))
i3 = int(np.argmin(abs(gv-0.3)))
steer_only = ga[i3] + 20*np.log10(alpha(gv)/alpha(gv[i3]))

fig, ax = plt.subplots()
ax.plot(gv, ga, "o-", ms=3, label="SPICE (.ac, 100 Hz)")
ax.plot(gv, steer_only, "--", label=r"steering term only ($\alpha$), anchored at 0.3 V")
ax.axhline(floor_db, color="gray", lw=0.8, ls=":"); ax.text(-0.48, floor_db+0.8, f"floor ≈ {floor_db:.1f} dB", fontsize=8)
ax.axvspan(gv[lo], gv[hi], color="C2", alpha=0.08, label="monotonic control range")
ax.set_ylim(10, 65); ax.set_xlabel(r"$V_{ctrl}$ (V)"); ax.set_ylabel(r"$|A_v|$ (dB)"); ax.legend(fontsize=8)
fig.tight_layout(); fig.savefig(nu.FIGS / "gain_curve_baseline.png")
plt.show()
print(f"monotonic span {gv[lo]:+.3f} .. {gv[hi]:+.3f} V : {ga[lo]:.2f} -> {ga[hi]:.2f} dB  ({ga[hi]-ga[lo]:.2f} dB)")
print(f"gain floor (median over -0.35..-0.05 V): {floor_db:.2f} dB;  steering offset V_off = {1e3*v_off:+.1f} mV")
R = nu.record(base_floor_db=floor_db, base_mono_lo_v=gv[lo], base_mono_hi_v=gv[hi],
              base_mono_range_db=ga[hi]-ga[lo], base_peak_db=float(ga.max()), v_off_mv=1e3*v_off)
````

### 4.3 Development record: two defects, two design rules
1. **Floating PMOS wells.** The PMOS bodies were tied to a net the supply did not drive, so the n-wells floated about
   40 mV low. Gain barely moved, but M9 appeared to have more headroom than it has (Table 4, `float_cm120` vs
   `tied_cm120`). *Rule:* the harness ties wells explicitly.
2. **Input pair out of saturation.** At a steering common mode of 0.95 V the input pair's drain–source voltage fell to a
   few $U_T$ and it lost about half its transconductance at low gain (Table 4, `tied_cm095`), which looked like extra
   control range. *Rule:* $V_{DS} \ge 130$ mV on M2, M3, M5, M6, M9 at every control step, with margins of 20 mV in
   SPICE and 30 mV in the table screen.

## 6. The evaluator, and how far to trust it

The evaluator solves the DC point by damped Newton, with every device current read from the table at its own bias. Table
$g_m$, $g_{ds}$ and $g_{mb}$ fill the small-signal nodal matrix, which gives $A_v$, $R_{out}$ and each device's noise
transfer; thermal and flicker noise add in power; bandwidth is the output pole (the load plus the drain capacitance of
M4, M9 and M11). At the corners the same model runs on the corner tables, without noise. Table 8 checks it on the
baseline, Table 9 against the textbook estimate, and Table 9b on random designs.

````python
#@ DESIGN #@ VERIFY
BASE_F = dict(BASE, nf_in=int(round(BASE["W_in"]/vf.W_F)))        # the baseline's input pair as 5 um fingers
def _sp(p):
    out = {}
    for v in (0.0, 0.3):
        c = vf.characterise(p, v); nz = vf.noise(p, v, *BAND)
        out[f"{v:.1f}"] = dict(c, vrms=nz["vrms"], vn_1k=nz["vn_1k"])
    return out
SB = nu.cached("baseline_eval_ref", lambda: {"drawn": _sp(BASE), "fingers": _sp(BASE_F)}, RERUN_SPICE)
qs = [("A_v (dB)", "av_db", "av_db", 1), ("R_out (MΩ)", "rout", "rout", 1e-6), ("I_tot (µA)", "i_tot", "i_tot", 1e6),
      ("V_tail (V)", "v_tail", "v_tail", 1), ("V_DS M2 (mV)", "vds_m2", "m2_vds", 1e3), ("V_DS M9 (mV)", "vds_m9", "m9_vds", 1e3),
      ("Input noise, 20 Hz–20 kHz (µV)", "vrms", "vrms", 1e6)]
rows, errs = [], {"drawn": {}, "fingers": {}}
for name, ke, ks, sc in qs:
    for v in (0.0, 0.3):
        cells = [name, f"{v:g}"]
        for tag, p in (("drawn", BASE), ("fingers", BASE_F)):
            e = ev.VGAModel(LUT, p).evaluate(v, CL, with_noise=True); s = SB[tag][f"{v:.1f}"]
            a, b = e[ke]*sc, s[ks]*sc
            err = (a-b) if ("dB" in name or "mV" in name) else 100*(a/b-1)
            unit = " dB" if "dB" in name else (" mV" if "mV" in name else " %")
            errs[tag][(name, v)] = err; cells += [f"{a:.4g}", f"{b:.4g}", f"{err:+.2f}{unit}"]
        rows.append(cells)
nu.show_md("*Table 8. LUT evaluator versus SPICE at the hand-sized baseline (tt, 27 °C): as drawn (input pair one "
           "20 µm finger) and with the input pair rebuilt as four 5 µm fingers.*\n\n" +
           nu.md_table(["Quantity", "V_ctrl", "LUT (drawn)", "SPICE (drawn)", "error", "LUT (fingers)", "SPICE (fingers)", "error"], rows))
E = {v: ev.VGAModel(LUT, BASE).evaluate(v, CL) for v in (0.0, 0.3)}; S = {0.0: b0, 0.3: b3}
naive = [[f"{v:g}", f"{E[v]['naive_av_db']:.2f}", f"{E[v]['av_db']:.2f}", f"{S[v]['av_db']:.2f}",
          f"{E[v]['naive_av_db']-S[v]['av_db']:+.2f}", f"{E[v]['av_db']-S[v]['av_db']:+.2f}"] for v in (0.0, 0.3)]
nu.show_md("*Table 9. Textbook estimate $\\alpha g_{m2}(r_{o4}\\|r_{o9})$ versus the cascode-aware evaluator, both from "
           "the same LUT data, against SPICE (baseline as drawn).*\n\n" +
           nu.md_table(["V_ctrl", "textbook (dB)", "evaluator (dB)", "SPICE (dB)", "textbook error (dB)", "evaluator error (dB)"], naive))
mx = lambda tag, key: max(abs(errs[tag][(key, v)]) for v in (0.0, 0.3))
R = nu.record(eval_err_av_db=mx("drawn", "A_v (dB)"), eval_err_rout_pct=mx("drawn", "R_out (MΩ)"),
              eval_err_vtail_drawn_mv=1e3*max(abs(E[v]["v_tail"]-S[v]["v_tail"]) for v in (0.0, 0.3)),
              eval_err_av_db_fingers=mx("fingers", "A_v (dB)"), eval_err_rout_pct_fingers=mx("fingers", "R_out (MΩ)"),
              eval_err_noise_pct_fingers=mx("fingers", "Input noise, 20 Hz–20 kHz (µV)"),
              naive_err_db=max(abs(E[v]['naive_av_db']-S[v]['av_db']) for v in (0.0, 0.3)))
````

The textbook estimate treats M9 as a plain $r_o$ and predicts the wrong direction of control; the evaluator solves

````python
#@ VERIFY #@ FIG
from scipy.stats import spearmanr
def _accuracy(n12=30, n9=12, seed=7):
    """Random designs (rejected by the model = not designs, redrawn), evaluator vs SPICE at tt. The 12T draws come
    from the grid widened to every value the search extension reaches (ev.SPACE_ACC_12T): up to 128 fingers, L to 16 um."""
    rng = np.random.default_rng(seed); out = []
    for topo, space, n in (("12T", ev.SPACE_ACC_12T, n12), ("9T", ev.SPACE_9T, n9)):
        keys = list(space); got = 0
        while got < n:
            pt = {k_: space[k_][rng.integers(len(space[k_]))] for k_ in keys}
            pt = {k_: (int(v) if k_.startswith(("nf_", "m_")) else float(v)) for k_, v in pt.items()}
            p = ev.materialise(BASE, pt)
            try:
                e = ev.evaluate_gate(LUT, p, CL=CL)
            except ValueError:
                continue
            try:
                s = flow.spice_gate(p, "tt", 27, SPEC)
            except RuntimeError:
                continue
            got += 1
            out.append(dict(topology=topo, point=pt,
                            lut=dict(av=e["av_curve_db"], bw=e["bw_curve"], i_tot=e["i_tot"], vds_min=e["vds_min"], vrms=e["vrms"]),
                            spice=dict(av=s["av_steps_db"], bw=s["bw_curve"], i_tot=s["i_tot"], vds_min=s["vds_min"], vrms=s["vrms"])))
    return out
ACC = nu.cached("eval_accuracy", _accuracy, RERUN_SPICE)

def _err(a):
    l, s = a["lut"], a["spice"]
    return dict(gain=max(abs(x-y) for x, y in zip(l["av"], s["av"])), bw=100*max(abs(x/y-1) for x, y in zip(l["bw"], s["bw"])),
                bw_opt=100*max(x/y-1 for x, y in zip(l["bw"], s["bw"])),      # > 0: the LUT promises more bandwidth than SPICE
                cur=100*abs(l["i_tot"]/s["i_tot"]-1), vds=1e3*abs(l["vds_min"]-s["vds_min"]), noise=100*abs(l["vrms"]/s["vrms"]-1))
ERR = pd.DataFrame([dict(topology=a["topology"], **_err(a)) for a in ACC])
old = json.loads((nu.CACHE / "front_spice.json").read_text())      # the previous evaluator, recorded 6 Oct
old_av = max(max(abs(f["lut"]["av_lo_db"]-f["av_lo_db"]), abs(f["lut"]["av_hi_db"]-f["av_hi_db"])) for f in old)
old_ro = max(100*abs(f["lut"]["rout_hi"]/f["rout_hi"]-1) for f in old)
rho_n = spearmanr([a["lut"]["vrms"] for a in ACC], [a["spice"]["vrms"] for a in ACC]).statistic
rho_b = spearmanr([min(a["lut"]["bw"]) for a in ACC], [min(a["spice"]["bw"]) for a in ACC]).statistic
rows = []
for name, sel in (("12T designs (widened space)", ERR["topology"] == "12T"), ("9T designs", ERR["topology"] == "9T")):
    e_ = ERR[sel]
    rows.append([name, len(e_), f"{e_['gain'].max():.2f} dB", f"{e_['bw'].max():.1f} % / {max(e_['bw_opt'].max(), 0):.1f} %", f"{e_['cur'].max():.2f} %",
                 f"{e_['vds'].max():.1f} mV", f"{e_['noise'].max():.1f} %"])
OLD_NOTE = (f"For comparison, the previous evaluator (6 Oct: a 5 µm table applied to single-finger widths "
            f"{min(f['p']['W_st'] for f in old):g}–{max(f['p']['W_in'] for f in old):g} µm, {len(old)} designs) erred by up to "
            f"{old_av:.2f} dB in gain and {old_ro:.0f} % in output resistance, and did not model noise.")
def _time_eval():
    [ev.evaluate_gate(LUT, ev.materialise(BASE, a["point"]), CL=CL) for a in ACC[:20]]     # warm-up: builds each table view once
    t_ = time.time(); [ev.evaluate_gate(LUT, ev.materialise(BASE, a["point"]), CL=CL) for a in ACC[:20]]
    return dict(ms=1e3*(time.time()-t_)/20, platform=platform.platform())
_TM = nu.CACHE / "eval_timing.json"                 # measured once and stored, so the summary does not change run to run
if RERUN_FLOW or not _TM.exists():
    _TM.write_text(json.dumps(_time_eval(), indent=1))
ms = json.loads(_TM.read_text())["ms"]
nu.show_md("*Table 9b. Evaluator against SPICE across random designs (tt, 27 °C), worst case over every control step. "
           f"Rank correlation (Spearman) of the predicted against the simulated noise: {rho_n:.3f}; of the minimum "
           f"bandwidth: {rho_b:.3f}. Evaluation time: {ms:.1f} ms per design (measured once, `cache/eval_timing.json`).*\n\n" +
           nu.md_table(["Designs", "n", "gain", "bandwidth: any error / over-estimate", "supply current", "headroom", "input noise"], rows) +
           "\n\nThe largest bandwidth errors are under-estimates at the lowest gain step, which only make the screen stricter; the "
           "over-estimate column is the one the screen's bandwidth margin has to cover.")
fig, axs = plt.subplots(1, 2, figsize=(10, 3.8))
for topo, mk in (("12T", "o"), ("9T", "s")):
    A = [a for a in ACC if a["topology"] == topo]
    axs[0].plot([1e6*a["spice"]["vrms"] for a in A], [1e6*a["lut"]["vrms"] for a in A], mk, ms=5, label=topo)
    axs[1].plot([min(a["spice"]["bw"])/1e3 for a in A], [min(a["lut"]["bw"])/1e3 for a in A], mk, ms=5, label=topo)
for ax, lab in ((axs[0], "input noise, 20 Hz–20 kHz (µV)"), (axs[1], "minimum bandwidth over the steps (kHz)")):
    lo, hi = ax.get_xlim(); ax.plot([lo, hi], [lo, hi], "k:", lw=0.8)
    ax.set_xlabel(f"SPICE: {lab}"); ax.set_ylabel(f"LUT: {lab}"); ax.legend(fontsize=8)
fig.tight_layout(); fig.savefig(nu.FIGS / "evaluator_accuracy.png"); plt.show()
R = nu.record(acc_n=len(ACC), acc_gain_db=float(ERR["gain"].max()), acc_bw_pct=float(ERR["bw"].max()), acc_bw_opt_pct=float(ERR["bw_opt"].max()),
              acc_cur_pct=float(ERR["cur"].max()), acc_vds_mv=float(ERR["vds"].max()), acc_noise_pct=float(ERR["noise"].max()),
              acc_rho_noise=float(rho_n), acc_rho_bw=float(rho_b), eval_ms=ms, old_err_av_db=old_av, old_err_rout_pct=old_ro)
````

## 7. The flow

`run_flow(spec, topology)` (`scripts/flow.py`) returns a SPICE-verified design, or the evidence that none exists:

1. **Screen** every grid point at tt with the evaluator: gain at seven control steps, bandwidth, monotonicity, range,
   current, headroom at every step, noise.
2. **Filter** on every limit, with the table margins of Table 10b.
3. **Corner screen** the survivors on the corner tables.
4. **Extend** past the grid: screen the unexplored single-axis neighbours of the best designs until two rounds improve
   the noise by less than 0.5 %. An axis stops only at a stated limit: the last tabulated length, 128 fingers, or a
   neighbouring stage's signal range.
5. **Front:** the noise–current Pareto front of everything that passes every corner.
6. **Verify** the front in SPICE at every corner.
7. **Pick:** most noise targets met, then lowest SPICE noise (within 2 % is a tie), then least gate area (within 5 %),
   then most headroom.
8. **Audit** every single-axis neighbour of the pick, and simulate any that would win.

`flow.run_topologies` runs this over topologies, simplest first, and names the limit that stops each (`flow.diagnose`);
the 12T was added to the list by hand (§8). The enforced specification (Table 10b) tightens four definitions of Table 2,
each after the flow exposed a gap: headroom at every step (§9.2), range only over steps that meet the bandwidth limit,
$V_{DD}$ ± 10 % as corners, and the current limit with three sigma of bias mismatch (§10.6). Noise is the objective; the
gate only rejects designs noisier than the hand-sized 9T.

## 8. The fix: a 12-transistor core

The floor exists because the steering fraction also sets the output devices' bias (§3.4). The fix came from a designer
reading that diagnosis, not from the flow: two **keep-alive sinks**, M10 on the mirror node and M11 on the output, draw a
current that does not depend on $\alpha$, so the output devices never starve and the gain follows the steering law across
the span:

$$
\Delta A_v\Big|_{-V}^{+V} \approx 20\log_{10}\frac{\alpha(+V)}{\alpha(-V)} = \frac{20}{\ln 10}\,\frac{V}{nU_T}.
$$

M1/M4 copy M10's current into the output and M11 removes it, so the sinks' common bias and its noise cancel there; only
their mismatch and uncorrelated noise remain. The tail and both sinks mirror an external reference through the
diode-connected M12, all as copies of one unit transistor, so every ratio is an integer in SPICE and in layout. The cost
is a lower maximum gain and the sinks' noise.

### 9.3 Layout in the loop: the first layout's wiring goes back into the flow

part of the load. `wiring.fit_c_wire` finds the lumped capacitor on `v_out` with the same effect, `flow.wire_spec` adds

### 10.1 Gain law over temperature
The steering law depends on $nU_T$, so the dB-per-volt slope scales with $1/T$. A control voltage proportional to
absolute temperature cancels this; here that source is behavioural, not a claimed circuit.

### 10.6 Mismatch: DC match and Monte Carlo with the PDK's statistical model
SKY130's mismatch model varies $V_{TH0}$, $V_{OFF}$, $T_{OX}$ and (pfet) $N_{FACTOR}$ per device. In weak inversion
$V_{OFF}$ moves the current like $V_{TH0}$ (twice as steeply in the nfet), so a threshold-only (Pelgrom [10]) estimate
understates the spread. `vga_flow.mismatch_lib()` turns the PDK's global mismatch parameters into per-device ones.
**DC match** (`verify.dc_match`) combines the SPICE sensitivities of the output level; **Monte Carlo**
(`verify.monte_carlo`) applies every draw at once at tt and at the most critical corners. With this much open-loop gain,
mismatch sets the output level, so each die is measured behind an **ideal output DC servo**, and the current the servo
needs is its specification.

````python
#@ VERIFY #@ FIG
def _corner_of(r):
    parts = r["corner"].split("/")
    return (parts[0], float(parts[1]), float(parts[2][:-1]) if len(parts) > 2 else 1.8), flow.corner_label(r)
MCC = [[("tt", 27.0, 1.8), "tt, 27 °C", ["nominal"]]]       # Monte Carlo corners: nominal, then the pick's critical ones
for r_, why_ in ((max(PICK_V["corners"], key=lambda r: SPEC.i_eff(r["i_tot"], r.get("i_sig", 0.0))), "highest current with its mismatch margin"),
                 (min(PICK_V["corners"], key=lambda r: r["vds_min"]), "least headroom"),           # least-headroom and
                 (min(PICK_V["corners"], key=lambda r: r["bw_min"]), "least bandwidth")):          # least-bandwidth corners
    k_, lab_ = _corner_of(r_)
    hit = [m for m in MCC if m[0] == k_]
    if hit: hit[0][2].append(why_)
    else: MCC.append([k_, lab_, [why_]])
DM = nu.cached("dc_match", lambda: {"pick": vfy.dc_match(PICK), "baseline": vfy.dc_match(BASE)}, RERUN_SPICE)
MC = nu.cached("monte_carlo", lambda: {f"{c}/{t:g}/{v:g}": vfy.monte_carlo(dict(PICK, vdd=v), n=MC_N, corner=c, temp=t, spec=SPEC)
                                       for (c, t, v), *_ in MCC}, RERUN_SPICE)
rows = []
for name, key in (("9T baseline", "baseline"), ("12T pick", "pick")):
    d_ = DM[key]
    for i in (0, 3, 6):
        sh = sorted(((k_, v_[i]) for k_, v_ in d_["share"].items()), key=lambda kv: -kv[1])[:3]
        rows.append([name, f"{d_['steps'][i]:+.1f}", f"{d_['a_dc'][i]:.3g}", f"{1e3*d_['sigma_vout'][i]:.0f}",
                     f"{1e3*d_['sigma_vin'][i]:.2f}" if abs(d_["a_dc"][i]) > 1 else "— (gain < 1)",
                     ", ".join(f"{k_.replace(':', ' ')} {100*s_:.0f} %" for k_, s_ in sh)])
nu.show_md("*Table 21. DC-match analysis with the PDK mismatch model (tt, 27 °C): one-sigma spread of the open-loop output "
           "DC level and of the input-referred offset, and the three largest contributors (device, parameter).* The output "
           "spread is a linear estimate; beyond a few hundred mV it means the open-loop output would saturate.\n\n" +
           nu.md_table(["Design", "V_ctrl (V)", "DC gain", "σ output DC (mV)", "σ input offset (mV)", "largest contributors"], rows))
lim_spec, lim_flow = SPEC.vds_min_v, SPEC.vds_min_v + SPEC.vds_margin_v
def _mc_rows(m_, lab):
    S_ = m_["samples"]; n_ = len(S_)
    gate = lambda s_, vl: (s_["monotonic"] and s_["range_db"] >= SPEC.range_min_db and s_["bw_min"] >= SPEC.bw_min_hz
                           and s_["i_tot"] <= SPEC.i_tot_max and s_["vds_min"] >= vl)
    y1, y2 = sum(gate(s_, lim_spec) for s_ in S_), sum(gate(s_, lim_flow) for s_ in S_)
    rng_ = np.array([s_["range_db"] for s_ in S_]); bw_ = np.array([s_["bw_min"] for s_ in S_])
    vd_ = np.array([s_["vds_min"] for s_ in S_]); nz_ = np.array([s_["vrms"] for s_ in S_])
    sv_ = np.array([max(abs(x) for x in s_["servo_i"]) for s_ in S_]); dv_ = np.array([s_["dv_open"][0] for s_ in S_])
    res_ = max(max(abs(x) for x in s_["resid_final"]) for s_ in S_)
    it_ = np.array([s_["i_tot"] for s_ in S_])
    y0 = sum(s_["vds_open_min"] >= lim_spec for s_ in S_) if "vds_open_min" in S_[0] else None
    return [lab, n_, f"{y1} / {n_}", f"{y2} / {n_}", "—" if y0 is None else f"{y0} / {n_}", f"{rng_.min():.1f} / {rng_.mean():.1f}", f"{bw_.min()/1e3:.1f}",
            f"{1e3*vd_.min():.0f}", f"{1e6*nz_.mean():.2f} / {1e6*nz_.max():.2f}", f"{1e3*dv_.std():.0f}",
            f"{1e9*np.percentile(sv_, 99.7):.0f}", f"{100*it_.std()/it_.mean():.2f} / {100*SIG_I/it_.mean():.2f}", f"{1e6*it_.max():.2f}",
            f"{1e6*res_:.0f}"], dict(y_spec=y1, y_flow=y2, y_open=y0, n=n_, range_min=float(rng_.min()), i_sd_rel=float(it_.std()/it_.mean()),
            i_max=float(it_.max()),
            bw_min=float(bw_.min()), vds_min=float(vd_.min()), noise_max=float(nz_.max()), dv_open_sd=float(dv_.std()),
            servo_i_997=float(np.percentile(sv_, 99.7)))
SIG_I = flow.current_sigma_abs(PICK)                 # the flow's analytic one-sigma supply-current spread (A)
MCR = []
for (c, t, v), lab_, why_ in MCC:
    r_, s_ = _mc_rows(MC[f"{c}/{t:g}/{v:g}"], f"{lab_} ({', '.join(why_)})"); MCR.append((r_, s_, lab_))
s1 = MCR[0][1]
nu.show_md("*Table 21b. Monte Carlo of the full gate with the PDK mismatch model behind an ideal output DC servo (SPICE).* Yield "
           f"against the specification's headroom limit ({1e3*lim_spec:g} mV) and the flow's margin ({1e3*lim_flow:g} mV); "
           f"without the servo, the dies that keep every device (plus M4 and M11) above {1e3*lim_spec:g} mV at every step; "
           "the servo current for 99.7 % of dies; and the supply current's spread, Monte Carlo against the flow's estimate.\n\n" + nu.md_table(
    ["Corner", "dies", "yield (spec)", "yield (flow margin)", "headroom yield, no servo", "range worst / mean (dB)", "min f-3dB (kHz)", "min V_DS (mV)",
     "noise mean / max (µV)", "σ open-loop v_out (mV)", "servo current, 99.7 % (nA)", "σ I_tot, MC / flow (%)",
     "max I_tot (µA)", "servo residual (µV)"], [m[0] for m in MCR]))
fig, axs = plt.subplots(1, 3, figsize=(13, 3.4))
for (c, t, v), lab, _ in MCC:
    S_ = MC[f"{c}/{t:g}/{v:g}"]["samples"]
    axs[0].hist([s_["range_db"] for s_ in S_], bins=25, alpha=0.6, label=lab)
    axs[1].hist([1e3*s_["vds_min"] for s_ in S_], bins=25, alpha=0.6, label=lab)
    axs[2].hist([1e3*s_["dv_open"][0] for s_ in S_], bins=25, alpha=0.6, label=lab)
axs[0].axvline(SPEC.range_min_db, c="k", ls="--", lw=0.8); axs[1].axvline(1e3*lim_spec, c="k", ls="--", lw=0.8)
axs[0].set_xlabel("control range in the bandwidth window (dB)"); axs[1].set_xlabel("min V_DS over every step and gated device (mV)")
axs[2].set_xlabel("open-loop output shift at max gain, no servo (mV)")
for ax in axs: ax.set_ylabel("dies"); ax.legend(fontsize=7)
fig.tight_layout(); fig.savefig(nu.FIGS / "monte_carlo.png"); plt.show()
_y0 = [m[1] for m in MCR if m[1]["y_open"] is not None]
if _y0:
    _w = min(_y0, key=lambda d_: d_["y_open"] / d_["n"])
    print(f"Without a servo, {_w['n'] - _w['y_open']} of {_w['n']} dies at the worst corner leave a device below "
          f"{1e3*lim_spec:g} mV of headroom at some control step; behind the servo, {min(m[1]['y_spec'] for m in MCR)} of "
          f"{MCR[0][1]['n']} pass at every Monte Carlo corner. The servo is a requirement of the core, and Table 21b gives its "
          f"specification: an output current of {1e9*max(m[1]['servo_i_997'] for m in MCR):.0f} nA covers 99.7 % of dies.")
dmp, dmb = DM["pick"], DM["baseline"]
R = nu.record(mm_pick_vin_mv=1e3*dmp["sigma_vin"][0], mm_pick_vout_mv=1e3*dmp["sigma_vout"][0],
              mm_base_vin_mv=1e3*dmb["sigma_vin"][0], mm_base_vout_mv=1e3*dmb["sigma_vout"][0],
              mc_tt=s1, mc_corners=[dict(label=m[2], **m[1]) for m in MCR],
              mc_pred_vs_mc_vout=[1e3*dmp["sigma_vout"][0], s1["dv_open_sd"]*1e3],
              mc_isig_flow_vs_mc=[SIG_I/np.mean([s_["i_tot"] for s_ in MC["tt/27/1.8"]["samples"]]), s1["i_sd_rel"]])
````

## 13. Physical implementation

The flow hands the layout two files (§9.4): the sizes (`layout/vga_core_12t_sizes.json`) and the LVS reference netlist
(`layout/vga_core_12t.spice`). A parametric gLayout generator written by Vrinda Pandey (`layout/generator/gen_vga12t.py`)
builds the layout from them, so no size is typed by hand and every ratio the flow sized holds exactly.

**Placement** (Figure 11, Table 27d): the input pair in a 2-D common-centroid array with dummies; the mirror
interdigitated in an n-well; the steering devices in one matched array; the bias devices in one unit array, with M10 and
M11 on a common centroid because their mismatch moves the output level. **Routing:** met2 to met3 buses inside each
array, one met4 trunk per net between arrays; `net2`–`net5` are labelled, so the post-layout gate reads headroom at the
schematic's nodes.

**Sign-off** (Tables 27–27c) runs twice, natively and in IIC-OSIC-TOOLS with a different PDK commit, and records tool
versions and file hashes: DRC with three rule decks; LVS with netgen including device sizes, plus six deliberately
altered reference netlists that must each fail; Magic's antenna check and the official KLayout deck's off-grid and
floating-metal checks, each with a control that must be flagged. No KLayout antenna deck was run. Magic's messages about
five p-well marker shapes and the 44 dummy transistors are explained in `layout/README.md`.

**Post-layout** (Tables 28–28c): the Magic extraction, every finger and parasitic capacitor included, goes through the
same 24-corner gate as the schematic, and again with every capacitor scaled by 1.5. A second extraction adds every wire's
resistance; at about an hour per corner in ngspice, it runs at the corners that bound the capacitance-only result, with
noise at tt. Monte Carlo and the other checks of §10 are not rerun on the extracted netlist.

## 14. Discussion

### 14.1 Limitations
1. **Noise.** The pick meets the integrated-noise and NEF targets at tt and misses the 1 kHz density target (Table 20c).
   Flicker noise dominates and the input pair and mirror sit at the 128-finger limit. Noise is optimised at tt; Table 20b
   gives every corner.
2. **Post-layout scope.** The extracted netlist runs the 24-corner gate with capacitance, and only the bounding corners
   with resistance (Tables 28, 28c). Monte Carlo, the control port, PSRR/CMRR, noise at every corner and the 21 stress
   corners are schematic-level.
3. **Low-gain dynamic range.** The usable range is the part of the span with at least 20 dB of output dynamic range
   (§10.2), not the whole gain-law span.
4. **DC servo.** Every Monte Carlo die passes behind an ideal servo; Table 21b specifies it, but it is not designed here.
5. **Ideal bias.** The reference current and the common-mode voltages are ideal; Table 17g gives the windows their
   circuits must hold.
6. **Temperature.** The control slope follows $1/T$ (Table 18); the PTAT correction is behavioural.
7. **Model scope.** The worst random-design bandwidth over-estimate exceeds the screen's margin (Table 9b), so SPICE
   decides every pick; on the verified designs the corner tables stay inside it (Table 12b). The margins cost noise in
   the first pass (Table 14) and none in the second (§9.3).
8. **Human steps.** The 12T topology came from a designer reading the diagnosis, and the layout generator was written
   for this topology.
9. **Area.** No area target was set and the pick is large: 17,120 µm² of gate area against 398 µm² for the hand screen.
   With flicker noise dominant, area buys noise (Table 11c).

### 14.2 Scope: a gain core, not a compressor
There is no envelope detector or gain-control loop, so no compression ratio, threshold, knee or attack/release behaviour
is claimed. What a loop designer needs from the core, how fast the gain follows the control voltage and how far the
output level jumps, is in §10.7.

### 14.3 Future work
An output DC servo on M11, bias circuits for the windows of Table 17g, a PTAT control generator, noise tables at the
